### 🥊 3-3-1 max() 與 min() 多參數比大小——誰是擂台冠軍與墊底守門員

想像一場熱血的格鬥錦標賽，擂台上站著三位、四位甚至十位選手。主審裁判拿著大聲公宣布：「今天我們要找出全場攻擊力最高的戰士，以及防禦最低的選手！」
在 Python 中，你不需要自己辦多場一對一的循環淘汰賽（不用寫一大堆 `if-else`），因為我們有兩位專屬裁判：
- **`max(arg1, arg2, arg3, ...)`**：最大值裁判！只要把所有競爭者丟給它，它會一眼挑出「最大的數值」！
- **`min(arg1, arg2, arg3, ...)`**：最小值裁判！不管傳入多少個數字，它專門揪出「最小的數值」！

**運作機制與語法規範：**
在 `max()` 與 `min()` 的小括號裡，你可以傳入兩個以上的數值，各數值之間**以逗號 `,` 隔開**。
例如：`max(10, 25, 8, 33)` ➔ 裁判會逐一檢視，回傳最大的 **`33`**。
`min(10, 25, 8, 33)` ➔ 裁判會回傳最小的 **`8`**。
它支援正整數、負整數與浮點數混合比較（例如 `min(-5, 3.2, 0)` 會正確回傳 `-5`）。

**常見錯誤陷阱（括號引數守則）：**
`max()` 與 `min()` 傳入獨立引數時，**至少要有「2 個」以上的引數**！
如果你只傳一個數字，例如 `max(42)`，Python 會拋出：
`TypeError: 'int' object is not iterable`（因為它會以為你想對一個容器做走訪，但 42 只是一個孤單的整數）。比大小一定要有對手，千萬別讓數字孤單上場喔！


```python
# [2] Code 範例區：展示 max() 與 min() 多參數極值比較

# 範例 1：基本多引數比大小
demo_n1 = 15
demo_n2 = 42
demo_n3 = 28
print(max(demo_n1, demo_n2, demo_n3))     # 印出 42（最大值）
print(min(demo_n1, demo_n2, demo_n3))     # 印出 15（最小值）

# 範例 2：負數與浮點數混合比較
demo_nx = -10
demo_ny = -3.5
demo_nz = 0
print(max(demo_nx, demo_ny, demo_nz))     # 印出 0
print(min(demo_nx, demo_ny, demo_nz))     # 印出 -10

# 範例 3：四個數字比大小
print(max(7, 99, 54, 88))  # 印出 99
```

### 🪆 3-3-2 巢狀合成函數運算——由內而外的同心圓求值順序

在數學課上，我們常常玩「俄羅斯套娃」：大娃娃肚子裡裝著中娃娃，中娃娃肚子裡又藏著小娃娃。打開娃娃時，我們必須一層一層由外往內剝開；但當電腦在**計算答案**時，順序剛好顛倒過來：**電腦永遠先計算最內層的小娃娃，將算出的答案交給上一層，一路往外推演！**
這在數學與電腦科學中，稱為「**合成函數（Composition of Functions）**」或「**巢狀函數（Nested Functions）**」！

**吳邦一老師講義第 7 頁的經典教材範例：**
吳老師在講義中展示了一道極具代表性的巢狀算式：
`x = max(7, 3*5, 5+x*2, y*2)`
以及更深層的套娃運算：
`print( max(1, 3, max(4*y-20, x-3), min(x-8, y-7)) )`

**運作機制拆解：**
面對這麼複雜的一長串括號，初學者千萬不要慌張！請像機器人一樣啟動「同心圓雷達」：
1. **尋找最內層小括號**：先算裡面的 `max(4*y-20, x-3)` 與 `min(x-8, y-7)`。
2. **替換為數值**：把最內層算出的純數字放回原位。
3. **推算外層裁判**：外層的 `max(1, 3, 內層結果1, 內層結果2)` 接收到所有數值後，做出最終裁決！

**APCS 解題建議：**
雖然 Python 支援無限層套娃，但吳邦一老師在講義中貼心建議：「**算式不要寫得太複雜，可以拆開寫！**」
將複雜的內層運算先存入中間變數（例如 `inner_max = max(...)`），能讓程式碼可讀性大幅躍升，除錯時也更容易印出中間數值檢查！


```python
# [2] Code 範例區：展示巢狀合成函數運作過程

# 範例 1：由內而外的雙層巢狀計算
demo_x = 10
demo_y = 6
# 內層先算：demo_x - 3 = 7，demo_y + 2 = 8
# 外層再算：max(7, 8) = 8
demo_res = max(demo_x - 3, demo_y + 2)
print(demo_res)  # 印出 8

# 範例 2：講義第 7 頁原汁原味巢狀極值範例
demo_ans = max(1, 3, max(4*demo_y - 20, demo_x - 3), min(demo_x - 8, demo_y - 7))
# 4*6 - 20 = 4, demo_x - 3 = 7 -> max(4, 7) = 7
# demo_x - 8 = 2, demo_y - 7 = -1 -> min(2, -1) = -1
# 外層: max(1, 3, 7, -1) = 7
print(demo_ans)  # 印出 7

# 範例 3：拆開寫的好習慣示範
demo_p1 = max(4*demo_y - 20, demo_x - 3)
demo_p2 = min(demo_x - 8, demo_y - 7)
demo_clean = max(1, 3, demo_p1, demo_p2)
print(demo_clean)  # 同樣印出 7，清晰好讀！
```

### 🏃‍♂️ 3-3-3 三數排序中間值技巧——總和扣除極值的數學巧思

現在，我們要解鎖數學與演算法思維中最優雅的神技——**「總和扣除極值法」**！
假設我們手上有三個數值 $a, b, c$（例如三位選手的百米賽跑成績、三款商品的售價、或是三次段考成績），我們想要將它們「由小到大」依序排好：
$$low \le mid \le high$$

**剛學程式時的困惑：**
初學者此時尚未學過複雜的排序演算法，如果想靠寫判斷式來兩兩比較，不僅程式碼冗長，還容易把自己繞暈。

**數學家傳授的絕妙巧思：**
既然只有三個數值：
1. **最小值 $low$**：呼叫 `low = min(a, b, c)` 一秒挑出！
2. **最大值 $high$**：呼叫 `high = max(a, b, c)` 一秒挑出！
3. **最關鍵的中間值 $mid$ 該怎麼找？**
   仔細觀察：三個數的總和就是 $a + b + c$。
   如果把三個數的總和，減掉最小的、再減掉最大的，**剩下的那一個，絕對百分之百就是中間值！**
   $$mid = a + b + c - low - high$$

完全不需要排序工具，不需要寫任何多重判斷分支，簡潔的三行算式，瞬間將三數由小到大整整齊齊排列出來！這就是演算法思維最迷人的優雅之處！

```python
# [2] Code 範例區：三數排序中間值技巧示範

# 給定三位選手百米短跑的完賽秒數（秒數越小成績越好）
demo_t1, demo_t2, demo_t3 = 14, 12, 16

# 步驟 1：找出第一名（秒數最小者）
demo_1st = min(demo_t1, demo_t2, demo_t3)

# 步驟 2：找出最後一名（秒數最大者）
demo_3rd = max(demo_t1, demo_t2, demo_t3)

# 步驟 3：總和扣除極值求出第二名（中間值）
demo_2nd = demo_t1 + demo_t2 + demo_t3 - demo_1st - demo_3rd

# 成果：三位選手秒數由快到慢（由小到大）排列輸出
print(demo_1st, demo_2nd, demo_3rd)  # 印出 12 14 16
```

```python
#@title 🧪 點擊此處取得單元 3-3 學習成效自動評分報告 { display-mode: "form" }
# ------------------------------------------------------------------------------
# 官方智慧全環境評分載入器（支援：本地開發庫優先 ➔ 公開正式庫雙重備援）
# ------------------------------------------------------------------------------
import os
import sys
import json
import logging
import urllib.request

# 靜音內部 httplib2 警告
logging.getLogger("google_auth_httplib2").setLevel(logging.ERROR)

# 1. 互動式輸入學員姓名與學號（分開獨立欄位，按 Enter 採用預設值）
print("👤【學員身分登記】（自學測試可直接按 Enter 鍵使用預設值）：")
student_name = input("請輸入您的姓名 [預設: 自學冒險者]：").strip() or "自學冒險者"
student_id = input("請輸入學號或座號 [預設: 無座號]：").strip() or "無座號"

# 2. 自動嘗試獲取 Google 帳號官方身分認證（真實 Email 與 Google 暱稱）
google_email = None
google_name = None
try:
    from google.colab import auth
    print("🔐 正在連結 Google 帳號進行身分認證（若跳出授權視窗，請點選允許）...")
    auth.authenticate_user()

    # 方法 1: 使用 Google Drive API v3 (about:get) 抓取真實 displayName 與 emailAddress
    # Colab ADC 預設包含 drive 權限，能 100% 精準獲取 Google 帳號中文顯示名稱（如：葉明昌）
    try:
        import google.auth
        from googleapiclient.discovery import build
        creds, _ = google.auth.default()
        drive_service = build('drive', 'v3', credentials=creds, cache_discovery=False)
        about = drive_service.about().get(fields="user(displayName,emailAddress)").execute()
        u_obj = about.get("user", {})
        google_name = u_obj.get("displayName")
        google_email = u_obj.get("emailAddress")
    except Exception:
        pass

    # 方法 2: 若 Drive API 失敗，嘗試透過 OAuth2 userinfo / tokeninfo 備援
    if not google_email or not google_name:
        try:
            import google.auth
            from google.auth.transport.requests import Request
            credentials, _ = google.auth.default()
            credentials.refresh(Request())
            token = credentials.token

            for ep in [
                f"https://www.googleapis.com/oauth2/v3/tokeninfo?access_token={token}",
                "https://www.googleapis.com/oauth2/v3/userinfo",
                "https://www.googleapis.com/oauth2/v2/userinfo"
            ]:
                try:
                    req = urllib.request.Request(
                        ep,
                        headers={"Authorization": f"Bearer {token}"}
                    )
                    with urllib.request.urlopen(req, timeout=5) as resp:
                        u_data = json.loads(resp.read().decode('utf-8'))
                        if not google_email:
                            google_email = u_data.get("email") or u_data.get("emailAddress")
                        if not google_name:
                            google_name = (
                                u_data.get("name")
                                or u_data.get("given_name")
                                or u_data.get("displayName")
                            )
                        if google_email and google_name:
                            break
                except Exception:
                    continue
        except Exception:
            pass

    # 方法 3: 若只有 email 但 Google API 沒給 name，以 email 前綴作為暱稱回退
    if google_email and not google_name:
        google_name = google_email.split("@")[0]

    if google_email:
        print(f"✅ 身分認證成功！已綁定 Google 帳號：{google_name} ({google_email})")
except Exception:
    # 非 Colab 環境、本機離線測試或略過授權
    pass

# 3. 身分資料獨立注入環境變數（學號與姓名完全分開）
os.environ["STUDENT_NAME"] = student_name
os.environ["STUDENT_ID"] = student_id
if google_email:
    os.environ["GOOGLE_EMAIL"] = google_email
if google_name:
    os.environ["GOOGLE_NAME"] = google_name

# 4. 評分核心腳本載入（本機優先 ➔ 遠端記憶體載入，絕不產生實體暫存檔）
# 找不到來源就換下一條。內容已拿到但執行失敗時，停止改試其他來源。
# 評分腳本放在獨立命名空間執行，成功後也不把名稱寫回筆記本。答案讀執行前的學生存放格快照。
_loaded = False
_load_error = None

# 4.1 優先嘗試讀取開發庫/本機檔案（本機離線開發測試零延遲）
_local_paths = [
    "web/graders/grade_3_3.py",
    "../graders/grade_3_3.py",
    "../../graders/grade_3_3.py",
    "../web/graders/grade_3_3.py",
    "../../web/graders/grade_3_3.py"
]
for _lp in _local_paths:
    if not os.path.exists(_lp):
        continue
    try:
        with open(_lp, "r", encoding="utf-8") as _f:
            _code = _f.read()
        _grader_ns = {"__builtins__": __builtins__, "__name__": "apcs_grader", "__STUDENT_ENV__": dict(globals())}
        try:
            exec(_code, _grader_ns)
        finally:
            del _grader_ns
        os.environ["GRADER_SOURCE"] = _lp
        _loaded = True
        break
    except Exception as _exc:
        _load_error = _exc
        break

# 4.2 若在雲端環境（Colab），自動透過公開正式庫雙重備援以純記憶體載入
if not _loaded and _load_error is None:
    _grader_urls = [
        "https://raw.githubusercontent.com/johnnyy-lab/APCS1to3/main/graders/grade_3_3.py",
        "https://johnnyy-lab.github.io/APCS1to3/graders/grade_3_3.py",
        "https://raw.githubusercontent.com/johnnyy-lab/APCS1to3-dev/main/web/graders/grade_3_3.py"
    ]
    for _url in _grader_urls:
        try:
            _req = urllib.request.Request(_url, headers={"User-Agent": "Mozilla/5.0"})
            with urllib.request.urlopen(_req, timeout=10) as _resp:
                _code = _resp.read().decode("utf-8")
        except Exception:
            continue
        try:
            _grader_ns = {"__builtins__": __builtins__, "__name__": "apcs_grader", "__STUDENT_ENV__": dict(globals())}
            try:
                exec(_code, _grader_ns)
            finally:
                del _grader_ns
            os.environ["GRADER_SOURCE"] = _url
            _loaded = True
            break
        except Exception as _exc:
            _load_error = _exc
            break

if not _loaded:
    if _load_error is not None:
        print("❌ 評分程式載入失敗，請稍後重新執行。若仍失敗，請把這個畫面交給老師。")
        if os.environ.get("GRADER_DEBUG", "").strip().lower() in ("1", "true", "yes", "on"):
            _reason = str(_load_error).strip().splitlines()
            _line = _reason[0][:180] if _reason else ""
            if _line:
                print("（開發診斷：" + type(_load_error).__name__ + "：" + _line + "）")
            else:
                print("（開發診斷：" + type(_load_error).__name__ + "）")
    else:
        print("❌ 評分系統連線異常，請確認網路連線狀態（或稍後重新執行）")
```
