### 🔍 4-2-1 認識分隔參數 sep——揭開預設半形空格的秘密

想像一下學校朝會時的廣播司儀。當司儀在台上朗讀：「第一名：小明、第二名：小華、第三名：小強」時，每唸完一個名字，司儀都會禮貌性地「停頓半秒、吸一口氣」，這個小小的停頓，就像是項目與項目之間的間隔。在 Python 的世界裡，`print()` 這個大喇叭在播報多個資料時，也自帶了這個「呼吸停頓」，它的名字就叫做 `sep`！

**運作機制與底層原理：**
在 Python 中，`sep` 是英文單字 **separator**（分隔器、間隔物）的縮寫。當我們在 `print()` 括號中丟入多個項目（用逗號隔開）時，例如 `print("蘋果", "香蕉", "橘子")`，Python 直譯器在背後其實悄悄幫我們設定了一個「預設分隔符號」：一個半形的空格字元 `' '`。
換句話說，你平常寫的：
`print("蘋果", "香蕉", "橘子")`
其實完全等價於：
`print("蘋果", "香蕉", "橘子", sep=' ')`
`sep=' '` 屬於 Python 中的「關鍵字引數（Keyword Argument）」。它告訴大喇叭：「請在每兩個輸出的寶物之間，自動塞入等號右邊所指定的字串！」如果你明確寫出 `sep=' '`，大喇叭就會照著你的吩咐，在每樣水果中間放一個空格。

**常見錯誤陷阱（初學者必讀）：**
1. **忘記關鍵字名稱（寫成一般引數）**：有些同學想自訂分隔符號，卻寫成 `print("A", "B", " ")`。這時候 Python 不會知道最後的 `" "` 是分隔符號，而是會把它當成「第三個要印出來的資料」，結果輸出就變成了 `A B  `！請牢記：自訂分隔符號一定要加上 `sep=` 這個標籤！
2. **關鍵字拼字錯誤**：有些初學者會不小心把 `sep` 拼成 `step`（步進值）、`spe` 或 `set`。一旦拼錯，Python 會立刻無情地拋出錯誤：`TypeError: 'step' is an invalid keyword argument for print()`，意思是「print 函數根本聽不懂什麼是 step」！
3. **單一資料項目時 sep 不起作用**：如果小括號內只有「一個」項目，例如 `print("你好", sep='***')`，螢幕上只會單純印出 `你好`，完全看不到 `***`！因為 `sep` 的作用是「分隔兩個項目之間」，如果只有一件物品，自然就沒有任何「中間的縫隙」可以塞東西了！

**APCS 考試實務建議：**
在 APCS 程式檢定與各類競賽題庫中，高達八成以上的題目都要求輸出格式為「多個整數以單一空格隔開」（例如：`10 20 30`）。理解 `print()` 預設就是以單一空格分隔，能讓你在寫題時直接利用 `print(a, b, c)` 輸出，不需要手動在變數之間拼湊空白字串，既快速又絕不失手！


```python
# [2] Code 範例區：認識 sep 參數與預設空格

# 範例 1：不寫 sep（採用 Python 預設的單一空格隔開）
print("小明", "小華", "小強")

# 範例 2：明確寫出 sep=' '（效果與範例 1 完全一模一樣）
print("小明", "小華", "小強", sep=' ')

# 範例 3：搭配數值變數與運算結果，觀察預設空格
score_a = 90
score_b = 85
score_c = 95
print(score_a, score_b, score_c, sep=' ')

# 範例 4：特別注意——只有單一項目時，sep 不會生效！
print("單獨一人", sep='---')  # 只會印出「單獨一人」，不會有 ---
```

### 🏷️ 4-2-2 自訂常用標點符號分隔——逗號、連字號與斜線

想像你正在手工製作一串串珠項鍊。原本每顆彩色珠子之間都只是空著的小縫隙（預設空格），但現在你想要在每顆珠子中間穿插一顆小金珠、一截黑色吸管、或是一顆小銀扣。這時候，你只需要換上不同的「間隔珠材料」，整串項鍊的風格就會截然不同！在 Python 裡，`sep` 允許你放進任何自訂的標點符號，輕鬆打造各種生活常見的標準資料格式！

**運作機制與底層原理：**
當我們把 `sep` 後方的等號設定為某個符號字串時，例如 `sep=','`、`sep='-'` 或 `sep='/'`，`print()` 函數就會在輸出每一對資料項目的交界處，精準插入該符號：
1. **逗號分隔 `sep=','`**：這就是大名鼎鼎的 **CSV 格式（Comma-Separated Values）**！全球各大資料庫、Excel 試算表在交換資料時，最常用的就是逗號分隔。
2. **連字號分隔 `sep='-'`**：廣泛應用於市話與手機號碼（如 `0912-345-678`）、產品零件序號（如 `SN-2026-A1`）、ISO 日期格式（如 `2026-09-15`）。
3. **斜線分隔 `sep='/'`**：常見於公曆日期標註（如 `2026/09/15`）與網路網址路徑。
4. **冒號分隔 `sep=':'`**：標準電子時鐘與運動賽事的時分秒時間格式（如 `14:30:00`）。

**常見錯誤陷阱（初學者必讀）：**
1. **忘記將符號包在引號中**：初學者最常犯的打字失誤就是寫成 `sep=-` 或 `sep=/`。在 Python 中，減號 `-` 與斜線 `/` 是數學運算子！如果你沒加引號，直譯器會以為你要進行減法或除法運算，因為找不到運算的數字，就會立刻報出嚴重的語法錯誤：`SyntaxError: invalid syntax`！請牢記：**符號是文字字串，必須用單引號或雙引號包起來**（如 `sep='-'`）！
2. **全形符號與半形符號搞混**：在 APCS 與自動評判系統（OJ）中，系統要求通常是半形標點（例如英文逗號 `,`）。如果你在中文輸入法狀態下打成了全形逗號 `，`（例如 `sep='，'`），在螢幕上看起來雖然長得很像，但評判系統比對 ASCII 編碼時會判定為完全不相符，直接送你一個心碎的 **WA（Wrong Answer）**！

**APCS 考試實務建議：**
APCS 題目中經常要求學生依照指定格式印出時間、日期或特定符號分隔的整數清單。善用 `sep` 可以在一行之內同時完成資料輸出與符號裝飾，省去用 `str()` 轉換再用 `+` 串接的大量括號與加號，程式碼既乾淨又不易打錯！


```python
# [2] Code 範例區：自訂常用標點符號分隔

# 範例 1：以逗號分隔（CSV 資料格式）
print("蘋果", "香蕉", "芒果", sep=',')

# 範例 2：以連字號 - 串聯電話號碼區碼與號碼
part1 = "0912"
part2 = "345"
part3 = "678"
print(part1, part2, part3, sep='-')

# 範例 3：以斜線 / 輸出標準年月日日期格式
year = 2026
month = 9
day = 15
print(year, month, day, sep='/')

# 範例 4：以冒號 : 輸出電子時鐘時間格式
hour = 14
minute = 30
second = 45
print(hour, minute, second, sep=':')
```

### 🧩 4-2-3 消除間隔的無縫緊密黏合——空字串 sep=''

想像你正在玩樂高積木，兩塊積木原本拼在一起時中間有一個小卡榫縫隙。如果我們在兩塊積木中間塗上一層「超強力透明無痕膠水」，把空氣縫隙完全抽空，兩塊積木就會「天衣無縫」地緊緊黏在一起，合體成一個完整的物件！在 Python 的輸出大喇叭中，這瓶神奇的透明無痕膠水，就是**空字串 `sep=''`**！

**運作機制與底層原理：**
在 Python 程式語言中，一對引號中間什麼都不放（長度為 0 的文字），被稱為**空字串（Empty String）**，寫作 `""` 或 `''`。
當我們在 `print()` 中指定 `sep=''` 時，直譯器在處理多個輸出引數時，項目與項目之間所插入的字元數量為 0！這意味著：
1. **無縫緊密相連**：所有丟給大喇叭的變數、字串與數字，將會首尾緊緊貼合，中間不會產生任何半形空格！
2. **數值與單位完美結合**：例如 `print(98, "分", sep='')`，輸出結果就是乾淨俐落的 `98分`！如果沒有加 `sep=''`，預設輸出就會變成尷尬的 `98 分`（中間多了一個空格）。
3. **免轉型的安全神技**：在過去，如果想把整數與文字黏在一起，初學者可能會想寫 `98 + "分"`，但這會引發嚴重的型態錯誤（整數與字串不能相加），必須繁瑣地寫成 `str(98) + "分"`。而使用 `print(98, "分", sep='')`，`print()` 會自動將所有項目轉換為字串並無縫貼合，完全不需要呼叫 `str()` 函數！

**常見錯誤陷阱（初學者必讀）：**
1. **「空字串」與「空白字串」的巨大盲區**：
   - `sep=''`：引號中間完全沒有任何東西（長度為 0），這是「**空字串**」，效果是消除所有間隔！
   - `sep=' '`：引號中間按了一下空白鍵（長度為 1），這是「**單一空白字串**」，效果是維持預設空格！
   初學同學常常手滑在引號裡誤按了空白鍵，導致明明寫了 `sep` 卻依然留著空格，請務必張大眼睛檢查引號內部！
2. **括號位置放錯**：寫成 `print("得分:", sep='', score)`，這會引發語法錯誤！在 Python 語法中，**關鍵字引數（如 `sep`）必須擺在所有位置引數的最後面**，不能插在要輸出的資料中間！

**APCS 考試實務建議：**
APCS 題目經常要求將多個單一數字拼接成一個大整數，或者輸出帶有特定前後綴單位的答案（例如：`No.12`、`$500`、`100%`）。使用 `print("$", money, sep='')` 或 `print(hundreds, tens, units, sep='')`，可以讓你用最精簡、最不容易出錯的寫法通過檢定測資！


```python
# [2] Code 範例區：消除空格的無縫緊密黏合

# 範例 1：比較有無 sep='' 的巨大排版差異
score = 100
print("考試成績：", score, "分")            # 預設空格：考試成績： 100 分
print("考試成績：", score, "分", sep='')     # 無縫黏合：考試成績：100分

# 範例 2：金額與貨幣符號無縫結合
price = 450
print("$", price, sep='')                   # 印出：$450

# 範例 3：百分比符號緊貼數值
rate = 85
print(rate, "%", sep='')                    # 印出：85%

# 範例 4：把百位數、十位數、個位數拼合成一個大數字
digit1 = 3
digit2 = 8
digit3 = 9
print(digit1, digit2, digit3, sep='')       # 印出：389
```

### ⚔️ 4-2-4 趣味多字元與文字字串分隔——流程箭頭與對戰排版

如果你以為 `sep` 的肚子裡只能放一個小小的標點符號，那可就太小看它了！大喇叭的間隔槽空間無窮大，你不但可以塞進單一符號，還可以塞進一整串「帶有空格的華麗箭頭」、「比賽對決標籤」，甚至是「可愛的顏文字與星星裝飾」！就像在一條長長的隊伍中，每兩個隊員中間都站著一位揮舞隊旗的領隊，讓整個畫面的氣勢瞬間提升！

**運作機制與底層原理：**
在 Python 中，`sep` 可以接受**任意長度的字串**！當 Python 直譯器在每兩個輸出項目之間填充時，它會將 `sep` 括號內的整個字串原汁原味地嵌入進去：
1. **箭頭流程圖 `sep=' -> '`**：在演算法執行過程、關卡闖關路徑或步驟指引時，使用前後各帶一個空格的箭頭，能夠清晰呈現「從 A 到 B 再到 C」的流暢動線（例如：`起點 -> 中繼站 -> 終點`）。
2. **運動競賽對決 `sep=' vs '`**：在球賽比分看板或電競對戰清單中，使用 ` vs ` 作為間隔，能立即營造出緊張熱血的對決氣氛（例如：`勇士隊 vs 湖人隊`）。
3. **垂直管線分隔 `sep=' | '`**：這是現代軟體界面與終端機最流行的「卡片式欄位排版」，能讓各項數值一目了然且維持整潔（例如：`玩家: 小明 | 等級: 15 | 金幣: 800`）。
4. **星號徽章排版 `sep=' ★ '`**：讓輸出的清單瞬間充滿榮譽感與節慶氛圍！

**常見錯誤陷阱（初學者必讀）：**
1. **忽視符號前後的「空格美學」**：
   - 如果你寫 `sep='->'`（箭頭前後沒有空格），輸出會變成 `起點->中繼站->終點`，文字全部擠在一起，閱讀起來非常費力。
   - 如果你寫 `sep=' -> '`（箭頭前後各留一個空格），輸出就會變成呼吸感十足的 `起點 -> 中繼站 -> 終點`！
   在設計多字元分隔符號時，請隨時留意是否有必要在前後保留適當的空白空間。
2. **與一般字串加法混淆**：初學者若不用 `sep`，常常會寫出超長的一串代碼：`print(a + " -> " + b + " -> " + c)`，這樣不僅要手動輸入多次 `" -> "`，而且如果 `a` 或 `b` 是數值，還會引發型態錯誤報錯！使用 `print(a, b, c, sep=' -> ')`，既優雅又安全，程式碼簡短三倍！

**APCS 考試實務建議：**
在 APCS 程式除錯（Debug）的過程中，我們常常需要印出變數在每一步驟的變化情況。利用 `print(step1, step2, step3, sep=' -> ')` 可以讓你一眼看清數值狀態的轉移與推移路徑，是選手必備的高效除錯視覺化技巧！


```python
# [2] Code 範例區：多字元字串與對戰箭頭排版

# 範例 1：流程圖箭頭導覽
step1 = "森林入口"
step2 = "精靈之泉"
step3 = "龍之巢穴"
print(step1, step2, step3, sep=' -> ')

# 範例 2：運動會球賽對決看板
team_red = "紅衫巨龍隊"
team_blue = "藍甲飛鷹隊"
print(team_red, team_blue, sep=' vs ')

# 範例 3：儀表板數據管線分隔
player = "艾爾登小勇士"
level = 42
rank = "SS級"
print(player, level, rank, sep=' | ')

# 範例 4：榮譽星號裝飾分隔
star1 = "金牌"
star2 = "銀牌"
star3 = "銅牌"
print(star1, star2, star3, sep=' ★ ')
```

### 🚀 4-2-5 特殊跳脫分隔符號——換行符號 sep='\n' 與跳格 sep='\t'

想像學校大樓裡有一座「快速降落電梯」。廣播員在頂樓廣播完第一個名字後，電梯「唰！」一聲降下一層樓，接著廣播第二個名字，然後再降一層樓廣播第三個名字！原本横著排成一列的名字，瞬間在空中排成了一支整整齊齊的「直立縱隊」！在 Python 的大喇叭中，只要在 `sep` 裡面放入神奇的「跳脫字元」，你就能施展這種瞬間換行與表格對齊的強大魔法！

**運作機制與底層原理：**
在 Python 中，反斜線 `\` 是「跳脫字元（Escape Character）」，它具備將後方字母轉化為特殊控制訊號的能力：
1. **換行跳脫字元 `\n`（Newline）**：
   - 當我們設定 `sep='\n'` 時，Python 會在每兩個項目之間插入一個「換行訊號」！
   - 這代表：原本需要呼叫 3 次 `print()` 才能印出的 3 行文字，現在只需要在 `print()` 裡放入 3 個項目，搭配 `sep='\n'`，**一行指令就能優雅地垂直分行輸出**！
   - 例如：`print("蘋果", "香蕉", "橘子", sep='\n')`，螢幕上就會分 3 行依序印出蘋果、香蕉與橘子。
2. **跳格定位字元 `\t`（Tab）**：
   - 當我們設定 `sep='\t'` 時，項目之間會插入一個寬度為固定欄位的定位空格（通常為 4 到 8 個字元寬）。
   - 它能讓不同長度的單字和數值自動對齊欄位，非常適合用來印製整齊的成績單或商品價目表。

**常見錯誤陷阱（初學者必讀）：**
1. **斜線方向致命搞反（`/n` vs `\n`）**：
   - 這是所有初學者最容易犯的低級失誤！鍵盤上有兩種斜線：
     - `/` 是正斜線（除法符號，位於問號鍵）。
     - `\` 是反斜線（位於 Enter 鍵上方或旁邊）。
   - 如果你寫成 `sep='/n'`，Python 會把它當成純文字「斜線加字母 n」，螢幕上會一字不差地印出 `蘋果/n香蕉/n橘子`，完全不會換行！請務必認清是**反斜線 `\n`**！
2. **最後一項後面會不會多換一行？**：
   - 請特別注意：`sep` 只負責「項目與項目之間」！因此 `sep='\n'` 只會在第一項與第二項之間、第二項與第三項之間換行。最後一項後面是否換行，是由 `print()` 結尾的 `end` 參數決定的（預設也是換行）。

**APCS 考試實務建議：**
在 APCS 與各類程式競賽中，非常多題目要求「每一筆答案單獨佔據一行」（例如輸出三組座標、三個極值或多行文字）。使用 `print(ans1, ans2, ans3, sep='\n')`，不僅可以大幅縮減程式碼行數，更能避免反覆呼叫 `print()` 造成的打字疲勞與變數名稱填錯，是頂尖選手非常鍾愛的一行化輸出技巧！


```python
# [2] Code 範例區：跳脫分隔字元換行與跳格

# 範例 1：使用 sep='\n' 將多個項目分行垂直輸出
print("【旗美公告】", "1. 週三全校大掃除", "2. 週五社團成果發表", sep='\n')

# 範例 2：將三個數值一行化各自印在獨立的一行
a = 100
b = 200
c = 300
print(a, b, c, sep='\n')

# 範例 3：使用 sep='\t' 製作整齊的欄位對齊報表
print("姓名", "國文", "數學", sep='\t')
print("小明", 95, 88, sep='\t')
print("小美", 100, 92, sep='\t')
```
