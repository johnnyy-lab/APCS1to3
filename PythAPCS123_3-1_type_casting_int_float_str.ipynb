### 🪓 3-1-1 int() 浮點數轉整數——無條件向零截斷（Truncation）

親愛的國一同學們，想像在一家蛋糕工廠裡，師傅手裡拿著一把鋒利的「整數固化雷射刀」。只要傳送帶送來裝著小數點果汁的量杯（例如 `3.99` 杯或是 `5.1` 杯），雷射刀就會「喀嚓」一聲，毫不留情地把小數點後面所有的果汁完全削平切除，只留下結結實實、完完整整的整數蛋糕積木！這把雷射刀，就是 Python 中的內建函數 **`int()`**。

**運作機制與底層原理：**
當我們把一個帶有小數點的浮點數（`float`）放進小括號裡傳給 `int()` 時，例如 `int(3.99)`，電腦執行的動作是「**向零截斷（Truncation towards Zero）**」，也就是**無條件捨去**！
請各位同學在心裡打上三個大驚嘆號：**它絕對不是數學課上學的「四捨五入」**！
- 就算小數部分高達 `0.999`，例如 `int(9.999)`，結果也頑固地是 **`9`**，絕對不會自動進位成 `10`。
- 對於負數也是同樣的道理：`int(-4.8)` 會直接切掉小數點後面的 `.8`，變成 **`-4`**（朝著 0 的方向靠近截斷），這和數學上的高斯符號（向下取整 Floor 會變成 `-5`）是不同的喔！

**APCS 考場實務與避坑指南：**
吳邦一老師在講義第 8 頁與第 22 頁（解一元二次方程式公式解）中特別叮嚀：競技程式評判系統（OJ）對格式有著「潔癖級」的要求！
當題目規定輸出「整數答案」時，如果你直接用 `/` 除法（例如 `4 / 1`）或是 `**0.5` 開根號（例如 `16**0.5`），算出來雖然數值看似正確，但它們骨子裡是浮點數 `4.0`！OJ 比對文字時看到多了一個 `.0`，就會當場判定為 **Wrong Answer（WA）**。
因此，在公式解或幾何求整數解時，及時使用 `int()` 將浮點數無情切除小數點，是奪取滿分 AC 的必備保險！


```python
# [2] Code 範例區：展示 int() 對浮點數的無條件截斷效果

# 範例 1：正浮點數無條件捨去（絕非四捨五入）
score_float = 89.99
score_int = int(score_float)
print(score_int)       # 印出 89，不會進位成 90！

# 範例 2：負浮點數向零截斷
neg_float = -3.75
neg_int = int(neg_float)
print(neg_int)         # 印出 -3，直接削去小數點！

# 範例 3：APCS 公式解開根號轉整數
root_float = 25 ** 0.5 # 開根號結果為浮點數 5.0
clean_root = int(root_float)
print(clean_root)      # 印出乾淨的整數 5，避免 5.0 格式錯誤
```

### 📦 3-1-2 int() 純整數字串轉整數——打破型態隔閡與致命的 ValueError 陷阱

在第一章我們介紹過，變數是貼有名牌的置物櫃，但你有沒有遇過這樣的尷尬情況：郵差送來一個快遞紙箱，紙箱表面用麥克筆寫著大大的文字 `'50'`。這時候的 `'50'` 只是畫在紙箱上的「外觀文字圖案（字串 `str`）」，根本不是箱子裡可以拿去買飲料的「真金白銀積木（整數 `int`）」！
如果你想要把兩張紙條 `'50'` 和 `'30'` 加在一起，Python 的加法機只會把兩張紙條黏成一長串：`'5030'`！若你試著拿紙條去加上實體硬幣（`'50' + 30`），Python 直譯器會立刻憤怒抗議：
`TypeError: can only concatenate str (not "int") to str`（型態錯誤：字串只能與字串拼接，不能跟整數混在一起做加法）！

**核心魔法：使用 int() 拆開包裝紙！**
要讓文字變回真正的整數積木，我們必須呼叫 `int(text)`。例如 `int('123')` 會逐字研讀文字內容，將其鑄造回可以在 CPU 算術單元中奔馳的數值 `123`！

**💣 考場最高頻引爆陷阱：ValueError 到底是怎麼來的？**
吳邦一老師在講義第 10 頁與第 14 頁特別警告初學者：`int()` 是個非常純粹的工匠，它眼裡容不下一粒沙子！被轉換的字串**內部必須完完全全都是十進位數字**（最前面可以有負號或正號）：
- **陷阱 1（超高頻失誤）**：`int('3.14')` ➔ **大爆炸！** 拋出 `ValueError: invalid literal for int() with base 10: '3.14'`！很多初學者會以為 `int('3.14')` 會自動變成 `3`，錯！因為字串裡的那個小數點 `.` 對 `int()` 來說不是合法的整數字元！
- **陷阱 2**：`int('123s')`、`int('abc')` 或 `int('一百')` ➔ **同樣爆炸！** 只要字串夾帶英文字母或非數字符號，就會觸發 `ValueError`。

**APCS 實務起手式：**
在 APCS 考場中，透過標準輸入 `input()` 讀進來的所有使用者輸入，**一律都是字串 `str`**！因此要讀入一個整數，所有選手的第一句咒語必定是：
`n = int(input())`！如果不轉型，後續拿去乘除比大小就會全面潰敗！


```python
# [2] Code 範例區：字串轉整數與避免字串拼接誤區

# 範例 1：字串拼接 vs 整數加法的差別
str_a = "50"
str_b = "30"
print(str_a + str_b)             # 印出 "5030"（文字黏在一起，非數學加法）

# 範例 2：使用 int() 成功轉型為數值進行加法
num_a = int(str_a)
num_b = int(str_b)
print(num_a + num_b)             # 印出 80（數學上的正確加總！）

# 範例 3：APCS 標準輸入轉型模擬
user_input_mock = "1024"         # 模擬 input() 讀入的整數字串
data_size = int(user_input_mock) # 鑄造為整數積木
print(data_size * 2)             # 印出 2048
```

### 🧪 3-1-3 float() 轉型為浮點數——打造小數刻度與兩段式字串小數轉整數法

當我們在進行科學實驗、測量體溫或是計算圓周率時，粗顆粒的整數已經無法滿足精準測量的需求。這時候，我們需要請出 Python 魔法工廠的「浮點稀釋儀」——**`float()`**！
不論你丟給它一顆硬邦邦的整數積木（例如 `float(8)` 變成 `8.0`），還是遞給它一張寫著小數點數字的紙條（例如 `float('3.14159')`），它都能精確地為數值標註上細膩的小數點刻度！

**運作機制與底層原理：**
1. **整數轉浮點數**：`float(7)` 會將整數升級為帶有單一零尾數的浮點數 `7.0`。
2. **文字字串轉浮點數**：`float('24.8')` 會精準辨識字串中的數字與小數點符號，將其轉換為浮點數 `24.8`。它甚至支援科學記號，例如 `float('1.5e-3')` 會被解析為 `0.0015`！

**💡 救星級競賽神技：破解 int('3.14') 崩潰的「兩段式轉型法」！**
在上一個小節中，我們學到直接執行 `int('3.14')` 會當場引發 `ValueError` 導致程式崩潰，因為 `int()` 無法解讀字串中的小數點。
但是，在實際 APCS 題目或真實資料處理中，如果輸入的字串偏偏就是含有小數點（例如感測器讀進來的溫度字串 `'36.7'`），而題目明確要求我們「只取整數部分做後續運算」，該怎麼辦呢？
答案就是這套大師級的**兩段式轉換法（Two-step Casting）**：
1. **第一段（化為果汁）**：先用 `float('36.7')` 將文字字串成功解析為浮點數 `36.7`！
2. **第二段（固化截斷）**：在外面套上 `int()`，即 `int(float('36.7'))`！`int()` 面對浮點數時就能合法發動向零截斷，乾淨俐落地切除小數點，得到整數 `36`！

記住這道優雅的組合咒語：`int(float(str_value))`，它能在考場上拯救你免於無數次 `ValueError` 的慘劇！


```python
# [2] Code 範例區：展示 float() 轉型與兩段式轉型法

# 範例 1：整數升級為浮點數
base_score = 90
float_score = float(base_score)
print(float_score)             # 印出 90.0

# 範例 2：小數字串轉為浮點數
pi_str = "3.1415926"
pi_val = float(pi_str)
print(pi_val)                  # 印出 3.1415926

# 範例 3：APCS 破坑神技——兩段式轉型法 int(float(...))
# 直接 int("99.8") 會引發 ValueError 崩潰！
# 透過兩段式：字串 -> 浮點數 -> 整數截斷
temp_str = "99.8"
temp_int = int(float(temp_str))
print(temp_int)                # 成功印出 99！順利規避報錯！
```

### 📜 3-1-4 str() 轉型為文字字串——安全無縫拼接與格式輸出

想像你想把這學期辛苦考得的段考成績貼到寫給家長的通知信裡。如果直接拿一塊沉甸甸的鐵牌（整數 `int`）硬塞進薄薄的信封（字串 `str`），郵差肯定會退件拒收！
在 Python 也是一樣：加法符號 `+` 面對字串時是「黏合膠水」，但這瓶膠水**只能用在字串與字串之間**。如果你寫：
`"總分是：" + 95`
Python 會當場罷工噴錯：`TypeError: can only concatenate str (not "int") to str`！這時我們需要一把「文字打包膠膜機」——**`str()`**，把數字外層鍍上一層字串膜，變成文字紙條，才能跟其他文字天衣無縫地黏在一起！

**運作機制與底層原理：**
函數 `str(any_object)` 能將任何型態的物件（整數、浮點數、布林值等）轉換為純字串。例如 `str(100)` 變成 `'100'`，`str(3.14)` 變成 `'3.14'`。

**🏆 實務排版格式神技：緊密無縫輸出比分**
在程式排版與各類資訊系統中，經常有嚴格的輸出排版格式要求。
例如體育賽事看板中，要求輸出的兩隊比分格式為 `85:72`（主隊與客隊分數中間只有一個冒號，前後**絕對不能有任何多餘空格**）：
- 如果寫 `print(host, ':', guest)`，`print()` 預設會在逗號參數間自動補上一格空白，結果輸出變成 `85 : 72` ➔ **當場判定 WA（格式錯誤）！**
- 競賽高手的經典解法正是：**使用 `str()` 將數字轉成字串，再以 `+` 緊密黏接**！
  寫法如下：`print(str(host) + ':' + str(guest))`！

### 🏪 3-1-5 數值與文字強制轉型實務綜合應用——建立商品標籤與計費字串輸出

在真實世界的軟體開發與生活應用（例如便利商店 POS 收銀機、自動販賣機、外送平台計算系統）中，電腦每分每秒都在進行**數值與文字的跨界合作**！

**為什麼必須綜合運用三大轉型？**
1. **讀入資料常是「文字」**：
   使用者在螢幕上輸入的數字、或是條碼掃描機讀取的價格代碼，在電腦眼裡一開始都是純文字（如 `'35'`、`'5'`）。文字不能直接拿來算錢，必須使用 **`int()`** 或 **`float()`** 轉成純數值才能進行四則運算。
2. **打折換算常產生「浮點數」**：
   商品打八五折（乘以 `0.85`）或滿額扣除折數時，計算結果往往會帶有小數點（例如 `89.25` 元）。但在現金交易中，通常會使用 **`int()`** 將小數點切除，進行無條件捨去取整。
3. **明細輸出必須封裝成「文字」**：
   算好了最終金額，要列印收據紙條或在螢幕上顯示「`總計：89 元`」時，加號 `+` 面對字串與整數會報錯！我們必須使用 **`str()`** 把金額數字鍍上一層文字外衣，才能與商品品名、貨幣單位安全黏接拼接！

只要靈活串聯 `int()`、`float()` 與 `str()`，我們就能親手打造出穩定、優雅且符合商用規範的收據計算程式！

```python
# [2] Code 範例區：數值與文字強制轉型實務綜合應用

# 情境：便利商店智慧收銀機計算結帳金額與列印收據明細
item_name = "特上奶茶"
unit_price_str = "35"    # 從訂單系統讀入的文字單價
quantity_str = "3"       # 從訂單系統讀入的文字數量
discount_rate = 0.85     # 會員專屬折扣率（浮點數）

# 步驟 1：將文字轉型為整數進行算術計算
unit_price = int(unit_price_str)
quantity = int(quantity_str)
subtotal = unit_price * quantity  # 35 * 3 = 105

# 步驟 2：套用浮點數折扣，並以 int() 轉回無條件捨去的實付整數金額
discounted_price = subtotal * discount_rate  # 105 * 0.85 = 89.25
final_total = int(discounted_price)          # 截斷為整數 89

# 步驟 3：使用 str() 將整數數值轉為文字，拼裝出完整收據商品標籤
receipt_line = item_name + " x " + str(quantity) + "，折後實付：" + str(final_total) + " 元"
print(receipt_line)  # 印出：特上奶茶 x 3，折後實付：89 元
```

```python
# ==========================================
# [4] Code 練習題
# 任務說明：
# 電子遊樂場的代幣兌換機：
# 給定遊客投入的紙鈔金額 cash_input_str = "500"（字串型態）。
# 每一枚代幣售價 coin_price = 15 元。
# 請依序完成以下計算與轉型：
# 1. 將 cash_input_str 轉為整數 cash。
# 2. 計算可以兌換的代幣數量 coins（使用整數除法 //）。
# 3. 計算找零金額 change（使用模除取餘數 %）。
# 4. 將 coins 與 change 分別轉為字串，組合成以下格式並用 print() 印出：
#    "兌換代幣：" + str(coins) + " 枚，找零：" + str(change) + " 元"
#
# 【公開測試資料 1】
# cash_input_str = "500"
# 預期輸出：
# 兌換代幣：33 枚，找零：5 元
#
# 【公開測試資料 2】
# cash_input_str = "120"
# 預期輸出：
# 兌換代幣：8 枚，找零：0 元
# ==========================================
# 請在下方撰寫你的程式碼並執行測試：

cash_input_str = "500"
cash = int(cash_input_str)
coins = cash // 15
change = cash % 15
result = "兌換代幣：" + str(coins) + " 枚，找零：" + str(change) + " 元"
print(result)
```

```python
# ==========================================
# [5] Code 挑戰題
# 任務說明：
# （本題為自由挑戰題，無公開測資，請依題目情境自行思考並撰寫完整程式碼）
# 影城團體訂票系統計算標籤：
# 給定電影單張全票票價 price = 280 元，購買張數 count = 4 張。
# 由於是學生團體訂票，影城提供 88 折優惠（discount_rate = 0.88）。
# 另外每張票需酌收 20 元的線上訂票系統手續費（手續費不打折）。
# 請依序完成：
# 1. 計算票價打折後的金額（price * count * discount_rate），並用 int() 無條件捨去取整。
# 2. 加上全部票券的手續費（手續費 * count），計算出最終實付總金額 final_fee。
# 3. 將 final_fee 轉換為字串，組裝成格式：
#    "訂票 4 張，實付總額：" + str(final_fee) + " 元" 並使用 print() 印出！
# （提示：280 * 4 * 0.88 = 985.6 -> 轉 int 為 985；手續費 20 * 4 = 80；總額為 1065 元）
# ==========================================
# 請在下方撰寫你的程式碼：

price = 280
count = 4
discount_rate = 0.88
service_fee = 20

discounted_tickets = int(price * count * discount_rate)
final_fee = discounted_tickets + (service_fee * count)
msg = "訂票 " + str(count) + " 張，實付總額：" + str(final_fee) + " 元"
print(msg)
```
