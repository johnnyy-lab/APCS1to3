### 👐 1-3-1 多變數同時賦值——雙手齊發的打包與解包

在前面的章節裡，如果我們想給兩個變數箱子裝入寶物，我們必須寫兩行指令：
```python
a = 5
b = 3
```
這就像是你一次只能用一隻手搬一個寶物，先搬好 5 送進 a 箱，再轉身搬 3 送進 b 箱。

但是在 Python 的神奇魔法世界中，小助手有一雙靈巧的雙手！Python 允許我們在等號兩邊用「**逗號（,）**」把好幾個變數與數值串起來，一次完成賦值：
```python
a, b = 5, 3
```

這行神奇指令是怎麼運作的呢？
1. **右側先打包**：直譯器小助手看到等號右邊有 `5, 3`，會先把它們像一顆雙黃蛋一樣打包放在一個隨身托盤上。
2. **左側對應拆開**：接著看到等號左側有 `a, b` 兩個箱子，小助手就按照從左到右的順序，將托盤上的第一個寶物（5）精準放進第一個箱子 `a`，將第二個寶物（3）放進第二個箱子 `b`！

這在程式專業術語中稱為「**多重賦值（Multiple Assignment）**」或「**元組解包（Tuple Unpacking）**」。透過逗號，我們只要一行簡潔俐落的代碼，就能一口氣設定好幾組變數，既好看又省力喔！

````python
# [2] Code 範例區：使用逗號同時給 a 與 b 賦值，並印出內容
a, b = 5, 3
print(a)
print(b)
````

### ⚖️ 1-3-2 等號兩側天平必須對稱——個數不符的 ValueError 警報

在享受多重賦值的便利時，吳邦一老師在講義第 5 頁特別嚴格叮嚀了一句話：
「**等號左右可以有逗號分開的若干個，但個數必須一樣！**」

讓我們來看看如果兩邊個數不一樣，會發生什麼慘劇：

**情況一：右邊寶物太少，左邊箱子分不夠**
```python
a, b, c = 1, 2
```
此時直譯器小助手看到左邊準備了 3 個箱子，托盤上卻只端來 2 個寶物，第 3 個箱子空等不到東西，電腦就會大發脾氣報警：
`ValueError: not enough values to unpack (expected 3, got 2)`
（白話文：寶物不夠分啦！我要 3 個，你只給我 2 個！）

**情況二：右邊寶物太多，左邊箱子裝不下**
```python

```

### 🥛 1-3-3 傳統變數交換的慘痛悲劇——為什麼需要第三個空杯子 temp？

現在我們來看一個經典的程式謎題：
假設你有兩個變數：`a` 裝著 5，`b` 裝著 3。如果我們希望「**把 a 和 b 的內容對調**」，讓 a 變成 3、b 變成 5，初學者通常會怎麼寫？

很多人會天真地寫出以下兩行代碼：
```python
a = b
b = a
```
吳邦一老師在講義第 5 頁犀利地問道：「**這會將 a 與 b 的內容交換嗎？為何？**」

讓我們用「兩杯飲料」來還原這個車禍現場：
想像你的左手拿著一杯鮮奶（`a`），右手拿著一杯紅茶（`b`）。
1. 當你執行第一行 `a = b` 時，相當於你直接把右手那杯紅茶倒進左手的杯子裡！這時候，左手原本的鮮奶被活生生沖洗稀釋、徹底覆蓋掉了！現在兩隻手拿的「全部都是紅茶」！
2. 接著你執行第二行 `b = a`，把左手的飲料倒回右手，結果倒來倒去還是紅茶！原本珍貴的鮮奶永遠消失了，根本沒有完成交換！

在 C 語言、Java 或傳統程式語言中，要解決這個悲劇，我們必須向桌子借「**第三個空杯子**」，在程式中通常叫做 `t` 或 `temp`（暫存變數 Temporary）：
```python
t = a  # 步驟 1：先將鮮奶暫存在空杯子 t
a = b  # 步驟 2：放心地把紅茶倒進 a 杯
b = t  # 步驟 3：把暫存杯裡的鮮奶倒回 b 杯！大功告成！
```
請大家親自執行看看，深刻體會為什麼覆蓋特性會需要第三個變數小幫手！

````python
# [2] Code 範例區：傳統交換三步驟（利用暫存變數 t）
a = 5
b = 3

t = a  # 第一步：把 a 的舊值 5 保存在 t 箱子
a = b  # 第二步：把 b 的值 3 覆蓋進 a（此時 a 變為 3）
b = t  # 第三步：把 t 裡的 5 覆蓋進 b（此時 b 變為 5）

print(a)  # 印出 3
print(b)  # 印出 5
````

### 🎩 1-3-4 Python 專屬的瞬間乾坤大挪移——`a, b = b, a` 變數交換

傳統的暫存變數 `t = a; a = b; b = t` 雖然能解決問題，但要寫三行代碼，還要額外多開一個 `t` 箱子，是不是覺得有點麻煩又累贅呢？

身為最優雅現代的 Python，給了全世界程式設計師一個超級大驚喜！在 Python 裡，要將兩數交換，**只需要一行指令**：
```python
a, b = b, a
```

哇！這到底是什麼神奇的魔法？為什麼這樣寫不會像剛才倒牛奶一樣被覆蓋掉呢？
吳邦一老師在講義第 5 頁為大家揭曉了秘密：
「**記得等號的運作順序：先計算右邊的值，再放入左邊的記憶體！**」

讓我們用慢動作再次分解這奇蹟的一瞬間：
假設一開始 `a = 5, b = 3`：
1. **第一步（先看右邊托盤）**：電腦小助手先看等號右邊的 `b, a`。這時候它迅速去偷看現在箱子裡的值——`b` 目前是 3，`a` 目前是 5。小助手把 `(3, 5)` 這兩個珍貴的舊數值穩穩當當地放在隱形托盤上！
2. **第二步（再存左邊箱子）**：托盤端穩之後，小助手才啟動左邊的等號傳送帶！托盤上的第一個數字 3 倒進左邊的第一個箱子 `a`；托盤上的第二個數字 5 倒進左邊的第二個箱子 `b`！

看懂了嗎？因為右邊的舊值早就在記憶體後台被完整打包保存了，所以在賦值給左邊時，根本不會互相踩踏覆蓋！
這行語法是 Python 最優雅的招牌魔法，不論是兩人交換座位還是數值互換，一行指令就能乾淨俐落完成！

````python
# [2] Code 範例區：Python 一行神技 a, b = b, a 瞬間交換
a = 5
b = 3

# 一行乾坤大挪移！
a, b = b, a

print(a)  # a 成功變成 3
print(b)  # b 成功變成 5
````

### 🤹 1-3-5 三人傳球大輪替——多變數循環交換（Cyclic Swap）

學會了兩個人交換，那如果是三個人、甚至四個人要進行「**大風吹輪流換位置**」呢？

吳邦一老師在講義第 5 頁出了一道非常精彩的思考題：
「**如果要將 a, b, c 三個變數循環交換（a 變成 b, b 變成 c, c 變成 a），應該如何下指令？**」

如果用剛剛學到的「先看右邊打包托盤，再存左邊箱子」的原理，這題對現在的你來說簡直易如反掌！
讓我們來對齊目標：
* 左邊的第一個箱子是 `a` ➔ 題目要求它變成 `b` 的值。
* 左邊的第二個箱子是 `b` ➔ 題目要求它變成 `c` 的值。
* 左邊的第三個箱子是 `c` ➔ 題目要求它變成 `a` 的值。

所以只要把左邊與右邊像鏡子一樣排好：
```python
a, b, c = b, c, a
```

運作過程是：右邊先把 `(b原本值, c原本值, a原本值)` 放在托盤上，然後同一時間倒進左邊的 `a, b, c` 箱子！
完全不需要宣告任何複雜的暫存變數，一個指令就能完成三人的大風吹輪替！這就是 Python 語言令無數程式設計師驚豔讚嘆的魔力所在！

````python
# [2] Code 範例區：三個變數循環交換 a 變 b, b 變 c, c 變 a
a, b, c = 1, 2, 3

# 大風吹循環輪替！
a, b, c = b, c, a

print(a)  # a 得到原本 b 的值：2
print(b)  # b 得到原本 c 的值：3
print(c)  # c 得到原本 a 的值：1
````

4. **瞬間交換神技**：`a, b = b, a` 利用右側先求值特性，一行完成優雅對調。
5. **多變數循環輪替**：`a, b, c = b, c, a` 輕鬆駕馭多變數大風吹。
