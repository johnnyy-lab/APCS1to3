### ✂️ 4-6-1 字串的魔法剪刀——`split()` 空白切割與多變數同時接收

想像你買了一長條尚未切片的法國長棍吐司，長棍吐司上每隔一段距離就有一個壓痕（空白鍵空格）。如果直接把整條長棍吐司塞給朋友，朋友根本無法一口吞下；此時你需要一把鋒利的麵包刀（`split()`），順著那些空格壓痕「喀嚓！喀嚓！」切下去，把長棍吐司分成數塊獨立的吐司片，然後同時遞給坐在餐桌旁的兩位朋友！

在 Python 裡，字串內建的 `.split()` 方法就是這把**「字串的魔法剪刀」**！

**運作機制與底層原理：**
1. **抓取一整行**：`input()` 依然扮演耳朵角色，忠實抓取使用者在鍵盤敲下的「整行文字」（直到按下 Enter）。
2. **施展魔法剪刀**：在 `input()` 後方加上 `.split()`（即 `input().split()`）。當電腦看到 `.split()` 且括號內不給任何參數時，它會自動發動「空白切割術」——**自動以任何連續的空白字元（包含單個空白、多個連續空格、Tab 鍵跳格）作為分界刀口，把整行文字切碎成數個獨立的小字串包裹**！
3. **多變數解包接收（結合 1-3 單元）**：我們在第 1-3 節學過「多變數同時賦值」：`x, y = 10, 20`。魔法剪刀切出的多個文字包裹，正好可以由等號左邊相同數量的變數名牌**同時接住**！例如：
   `word1, word2 = input().split()`
   如果使用者在終端機輸入：`Apple Banana`
   電腦就會將 `"Apple"` 分配給 `word1`，將 `"Banana"` 分配給 `word2`！

**常見錯誤陷阱（初學者必讀三大地雷）：**
1. **變數名牌數量與切出來的塊數不一致（最致命的高頻報錯）**：
   等號左側的變數數量，必須和使用者輸入的單字數量**完全一模一樣**！
   - 如果輸入了 3 個單字（`cat dog pig`），左邊卻只寫了 2 個變數（`a, b = ...`），Python 會立即報錯：`ValueError: too many values to unpack (expected 2)`（資料太多，名牌不夠貼！）。
   - 反過來，如果輸入只有 1 個單字（`cat`），左邊卻寫了 2 個變數，Python 則會報錯：`ValueError: not enough values to unpack (expected 2, got 1)`（名牌有 2 個，包裹卻只有 1 個！）。
2. **切開後的碎片依然是「字串」**：
   `split()` 只是把長字串切成短字串，裡面的內容依舊帶有引號！如果是輸入數字 `5 8`，`word1` 是 `"5"`，`word2` 是 `"8"`。如果直接拿來相加 `word1 + word2`，會得到 `"58"` 而不是 `13`！
3. **忘記在 split 後方加上小括號 `()`**：
   寫成 `input().split`（漏掉括號）是不會執行切割動作的，必須明確寫出函式呼叫小括號 `()`！

**APCS 考試實務建議：**
APCS 考試的輸入測資，空白數量有時不只一個（可能包含不小心多按的空格或製表符）。使用無參數的 `input().split()` 是最安全的做法，因為它能自動忽略所有多餘的空白！

```python
# [2] Code 範例區：展示 input().split() 空白切割與多變數同時接收

# 步驟 1：單行讀入使用者的「姓氏」與「名字」，中間用空白隔開
# 例如輸入：陳 大明  或者  Lin Alex
print("請在下方輸入你的姓氏與名字（中間請用一個空白隔開），並按 Enter：")
last_name, first_name = input().split()

# 步驟 2：印出接收到的獨立變數內容
print("--- 讀取結果確認 ---")
print("成功讀取姓氏：", last_name)
print("成功讀取名字：", first_name)

# 步驟 3：驗證兩者的資料型態
print("last_name 的型態：", type(last_name))
print("first_name 的型態：", type(first_name))
print("組合問候語：哈囉，親愛的", last_name, first_name, "同學！")
```

### 🏭 4-6-2 自動化加工輸送帶——`map(int, ...)` 一次將多個字串變整數

在上一小節中，我們學會了用 `split()` 剪刀把一行字串剪開。但如果剪開的是數字文字（例如 `"10"` 和 `"20"`），我們該怎麼讓它們變成真正的數字整數呢？

難道我們得寫繁瑣的兩行代碼：
`s1, s2 = input().split()`
`a = int(s1)`
`b = int(s2)`
這樣寫不僅冗長，而且如果一行有 5 個數字，就要寫 10 行！這在分秒必爭的 APCS 考場上太浪費時間了。

Python 提供了極度優雅的秘密武器——**自動化加工輸送帶 `map()` 函數**！

**生活比喻：工廠模具自動化輸送帶**
想像一座自動化蛋糕工廠。輸送帶上整齊排隊送過來一個個剛切好的麵團（`input().split()` 切開後的各個字串）。輸送帶上方懸掛著一個高溫壓印烘烤模具（`int` 函數）。
`map(int, ...)` 就像啟動了這條輸送帶開關！每個經過模具的麵團，都會被模具「重重壓一下」，瞬間從生麵團（字串 `str`）徹底變形為香噴噴的烤蛋糕（整數 `int`）！

**運作機制與語法解析：**
標準 APCS 經典一行流整數讀入模板：
`a, b = map(int, input().split())`

讓我們像洋蔥一樣，**由內而外**層層剖析電腦的執行順序：
1. **第一層（最內層）**：`input()` 讀入使用者鍵入的整行文字，例如 `"10 20"`。
2. **第二層**：`.split()` 沿著空白剪開，變成兩個字串包裹 `"10"` 與 `"20"`。
3. **第三層**：`map(int, ...)` 啟動加工輸送帶，將 `"10"` 與 `"20"` 依序丟入 `int()` 函數進行型態轉換，將它們轉成真正具有數學加減能力的整數 `10` 與 `20`！
4. **第四層（最外層）**：等號左側的變數名牌 `a, b` 同步出動，`a` 接住第一個整數 `10`，`b` 接住第二個整數 `20`！

**常見錯誤陷阱：**
1. **妄想寫成 `int(input().split())`（致命型態錯誤）**：
   有些同學會想：「既然單一整數是 `int(input())`，那多個整數是不是外層包 `int()` 就好？」
   **大錯特錯！** `int()` 模具一次只能壓印一個單一物件，根本無法直接吞下整批被切開的資料碎片，這樣寫會噴出 `TypeError: int() argument must be a string, not a 'list'`！
2. **忘記帶引數或寫錯函數名稱**：
   `map` 的第一個參數必須是轉換函數的名牌 `int`（注意：**後面不要加小括號**，寫 `int` 而不是 `int()`），第二個參數才是要加工的資料來源 `input().split()`。

**APCS 考試實務建議：**
`a, b = map(int, input().split())` 是全台 APCS 考生寫過最多次的一行指令，熟練程度必須達到「不用思考，手指就能本能敲出」的境界！

```python
# [2] Code 範例區：展示 map(int, input().split()) 一次將單行多數值轉換為整數

# 步驟 1：同一行讀入兩個整數，中間用空白隔開
print("請在同一行輸入兩個整數（例如：15 7），並按 Enter：")
num1, num2 = map(int, input().split())

# 步驟 2：驗證資料型態是否成功轉變為 int
print("num1 的數值為：", num1, "，型態為：", type(num1))
print("num2 的數值為：", num2, "，型態為：", type(num2))

# 步驟 3：進行真實的算術運算（不再是文字接龍，而是真正的加法與乘法）
total_sum = num1 + num2
product = num1 * num2
print("兩數相加總和：", total_sum)
print("兩數相乘積：", product)
```

### 🏃 4-6-3 單行讀入三個及以上整數——三數運算、極值與全距實戰

如果題目在同一行中給的不是 2 個數字，而是 3 個、4 個甚至更多個數字呢？

**運作原理與擴充性：**
`map(int, input().split())` 具有極高的彈性！只要輸入的同一行中有幾筆空白隔開的資料，等號左側就準備好相應數量的變數名牌，一對一排好隊，就能完美接收！

例如在同一行輸入三個整數：
`a, b, c = map(int, input().split())`
如果輸入為：`7 25 13`
則 `a` 會是 `7`，`b` 會是 `25`，`c` 會是 `13`！

如果要讀入四個整數（例如平面幾何座標點 $(x_1, y_1)$ 與 $(x_2, y_2)$）：
`x1, y1, x2, y2 = map(int, input().split())`
一口氣讀入四個數值，乾淨俐落！

**與第 3 章極值與絕對值函數的完美結合：**
在第三章中，我們學過了強大的內建數學函數：
- `max(a, b, c)`：快速找出三個整數中的最大值。
- `min(a, b, c)`：快速找出三個整數中的最小值。
- **全距（Range）**：統計學上的全距，即「資料組中的最大值減去最小值」，公式為 `max(...) - min(...)`。
- `abs(x)`：計算絕對值，求數線或座標距離。

將單行多數值讀入與這些函數搭配，就能以不到 5 行代碼，解決許多 APCS 基礎題型！

**常見錯誤陷阱：**
1. **手指打結，左邊名牌少寫或多寫一個**：
   如果題目要求輸入 3 個數字，等號左側卻寫了 4 個變數（如 `a, b, c, d = map(...)`），按下 Enter 後 Python 就會立刻拋出 `ValueError: not enough values to unpack`。務必看清楚題目說明的一行內有幾個數字！
2. **忘記變數以逗號隔開**：
   等號左邊的多變數必須使用半形逗號 `,` 隔開，不能使用空白隔開。

```python
# [2] Code 範例區：展示單行讀入三個整數並結合極值函數 max, min 計算

# 步驟 1：單行讀入三個整數
print("請在同一行輸入三個整數（例如：18 42 27），中間以空白隔開：")
a, b, c = map(int, input().split())

# 步驟 2：利用第三章學過的 max 與 min 找出最大值與最小值
biggest = max(a, b, c)
smallest = min(a, b, c)

# 步驟 3：印出分析結果
print("--- 數據統計 ---")
print("輸入的三數分別為：", a, b, c)
print("三數中的最大值：", biggest)
print("三數中的最小值：", smallest)
print("三數的全距（最大值 - 最小值）：", biggest - smallest)
```

### 🔬 4-6-4 單行多浮點數映射——`map(float, ...)` 與精準實數運算

除了整數之外，在自然科學實驗、體育競賽計分、地理量測等領域，輸入的資料往往帶有小數點（浮點數 `float`）。例如選手跳水評分 `8.5 9.2`、身高體重 `1.75 68.5`、直角三角形兩股長 `3.5 4.5` 等。

如果遇到含有小數點的輸入，我們若繼續使用 `map(int, ...)`，Python 會瞬間報錯——因為 `int()` 拒絕直接解析含有小數點的字串（如 `int("3.14")` 會噴出 `ValueError`）！

這時候，只要把輸送帶上的模具更換為 **`float`**，就能輕巧解鎖：
`a, b = map(float, input().split())`！

**運作原理與比較：**
- `map(int, input().split())` ➔ 將切開的文字塊加工成**整數（int）**。
- `map(float, input().split())` ➔ 將切開的文字塊加工成**浮點數（float）**。
- 即使使用者輸入的是整數（如 `3 4`），`map(float, ...)` 也會自動將它們提升為 `3.0` 與 `4.0`，後續可以安全進行除法或小數運算！

**結合第 2 章 2-4 節的次方向量與開根號運算：**
在第二章中我們學過：
- 開根號可以寫成 `x ** 0.5`！
- 直角三角形商高定理（畢氏定理）：斜邊長 $c = \sqrt{a^2 + b^2} = (a^2 + b^2)^{0.5}$。
搭配 `map(float, input().split())`，我們就能單行讀入直角三角形的兩股長度，瞬間算出精確的斜邊長度！

**常見錯誤陷阱：**
1. **輸入有小數卻誤用 `int` 映射**：
   輸入 `3.5 4.2` 時若執行 `map(int, ...)`，會拋出 `ValueError: invalid literal for int() with base 10: '3.5'`。記住：有小數點請務必指名 `float`！
2. **忘記小數除法可能帶來的浮點誤差**：
   回顧第二章 2-5 節，浮點數除法可能存在極微小的精度誤差，但對於常規實數計算與公式求值已完全足夠。

```python
# [2] Code 範例區：展示 map(float, input().split()) 讀入浮點數並計算畢氏定理斜邊長

# 步驟 1：單行讀入直角三角形的兩股長（可能為小數，以空白隔開）
print("請在同一行輸入直角三角形的兩股邊長 a 與 b（例如：3.0 4.0），中間用空白隔開：")
a, b = map(float, input().split())

# 步驟 2：印出變數數值與型態驗證
print("股長 a：", a, "型態：", type(a))
print("股長 b：", b, "型態：", type(b))

# 步驟 3：利用商高定理計算斜邊長：c = (a**2 + b**2) ** 0.5
hypotenuse = (a**2 + b**2) ** 0.5
print("直角三角形的斜邊長度為：", hypotenuse)
```

### 🛡️ 4-6-5 APCS / OJ 競技實戰——一行流解析多數值與易錯陷阱防禦

恭喜你！學會了 `split()` 與 `map()`，你已經掌握了 APCS 與各類競賽程式（如 ZeroJudge、Codeforces、LeetCode）中最核心的測資讀入技術！

然而，許多同學即使知道了這個語法，在考場上仍經常遭遇以下三大致命挫折：
- 一跑測試就出現紅字 `ValueError`？
- 算出來的答案完全不對，數字變成巨大一長串？
- 上傳到自動評判系統卻拿到 `WA`（Wrong Answer）或 `RE`（Runtime Error）？

讓我們在此小節建立最堅固的**「競技防禦心法」**！

**三大高頻地雷深層剖析：**

1. **地雷一：變數名牌與資料個數不對等（ValueError）**
   - **現象**：題目明明說輸入三科成績，程式碼卻寫 `a, b = map(int, input().split())`。
   - **後果**：拋出 `ValueError: too many values to unpack (expected 2)`。
   - **防禦**：審題時務必在草稿紙上圈出「這行測資總共有幾個數值」，左側名牌變數一個都不能少！

2. **地雷二：忘記包 map，發生悲劇的文字接龍**
   - **現象**：寫成 `a, b = input().split()`，接著直接 `print(a + b)`。
   - **後果**：當輸入 `10 20`，輸出的不是 `30`，而是 `"1020"`！
   - **防禦**：只要資料要拿來做加減乘除或比大小，一定要在外面裹上 `map(int, ...)` 或 `map(float, ...)`！

3. **地雷三：括號位置擺錯造成 SyntaxError**
   - **現象**：括號層次太多，把括號包錯地方，例如 `a, b = map(int, input().split)`（漏掉 split 的括號）或 `a, b = map(int(input().split()))`。
   - **防禦**：牢記口訣——「**先耳朵 input()，後剪刀 .split()，模具在外包 map(int, ...)**」！

**競技評判系統（Online Judge）的最高準則：純淨讀入！**
永遠記住上一單元的警惕：在 `input()` 的括號內，**絕對不可以留下任何提示字串**（例如千萬不能寫 `input("請輸入數字:")`）！
評判系統是機器人，任何未規定的輸出字元都會直接判定為答案錯誤！

```python
# [2] Code 範例區：展示標準 APCS 規格純淨單行讀入與三數極值輸出模板

# 題目情境模擬：
# APCS 試題給定一行包含三個整數，要求輸出其中最大的整數。
# （遵守 OJ 規範：無提示字元輸入、無多餘裝飾輸出）

# 步驟 1：標準一行流純淨讀入三個整數
x, y, z = map(int, input().split())

# 步驟 2：利用 max 找出最大值
ans = max(x, y, z)

# 步驟 3：精準單行輸出結果
print(ans)
```
