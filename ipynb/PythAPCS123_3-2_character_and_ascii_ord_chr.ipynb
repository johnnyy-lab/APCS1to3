### 🔍 3-2-1 ord() 字元身分證——探訪 ASCII 編碼與連續性特徵

想像你來到了一個國際機場的海關櫃台。每位旅客（英文字母、數字符號）走上海關前，海關人員都會拿出一台「身分證掃描器」，對著旅客「逼」一聲，螢幕上就會跳出這位旅客的專屬身分證號碼。在 Python 中，這台神奇的掃描器就是內建函數 **`ord()`**（Ordinal 的縮寫，代表序數、編號）！

**運作機制與三大連續性特徵：**
吳邦一老師在講義第 46 頁與第 50 頁特別強調：初學者**完全不需要死記整張 ASCII 表**！你只需要記住三個最核心的「連續性」黃金基準點：
1. **大寫英文字母 `'A'` ～ `'Z'` 是連續的**：`ord('A')` 是 **`65`**，`'B'` 就是 `66`，一直連續排到 `'Z'`（`90`）。
2. **小寫英文字母 `'a'` ～ `'z'` 是連續的**：`ord('a')` 是 **`97`**，`'b'` 就是 `98`，一直連續排到 `'z'`（`122`）。
3. **數字字元 `'0'` ～ `'9'` 也是連續的**：`ord('0')` 是 **`48`**，一直排到 `'9'`（`57`）。注意：字元 `'0'` 的編碼可不是 `0`，而是 `48` 喔！

**常見錯誤陷阱（重要！報錯解析）：**
`ord()` 的肚子非常小，**一次只能吃「長度剛好為 1 的單一字元」**！
如果你不小心傳入兩個以上的字母，例如 `ord('AB')` 或是傳入空字串 `ord('')`，Python 就會立刻噴出：
`TypeError: ord() expected a character, but string of length 2 found`！
請牢記：它專屬為單一字元服務，不能對整個單字或句子直接使用喔！


```python
# [2] Code 範例區：使用 ord() 查驗字元的 ASCII 身分證字號

# 範例 1：大寫字母與小寫字母基準點
print(ord('A'))  # 印出 65（大寫基準）
print(ord('B'))  # 印出 66（連續緊接）
print(ord('a'))  # 印出 97（小寫基準）
print(ord('b'))  # 印出 98

# 範例 2：數字字元的 ASCII 編碼
print(ord('0'))  # 印出 48（注意：不是 0！）
print(ord('1'))  # 印出 49

# 範例 3：查驗特殊標點符號
print(ord('!'))  # 印出 33
print(ord(' '))  # 印出 32（空白鍵也是一個合法字元）
```

### 🧙‍♂️ 3-2-2 chr() 編碼變身術——數字召喚字元

如果說 `ord()` 是把字元還原成數字的身分證掃描器，那麼它的孿生兄弟 **`chr()`**（Character 的縮寫）就是一座神奇的「字元 3D 列印召喚機」！
只要你把一個整數編號放進 `chr()` 肚子裡，它就能像變魔術一樣，精準將對應的字元召喚到螢幕上！

**運作機制與底層原理：**
- 輸入整數 `65` 呼叫 `chr(65)`，召喚出大寫英文字母 **`'A'`**。
- 輸入整數 `97` 呼叫 `chr(97)`，召喚出小寫英文字母 **`'a'`**。
- 輸入整數 `48` 呼叫 `chr(48)`，召喚出數字字元 **`'0'`**。
- `ord()` 與 `chr()` 是一對完美的互逆操作（Inverse Operations）：
  `chr(ord('K'))` 會回到 `'K'`；而 `ord(chr(88))` 算出來剛好就是 `88`！

**常見錯誤陷阱（ValueError 邊界守則）：**
`chr()` 只接受代表 Unicode / ASCII 的合法非負整數（ASCII 範圍通常在 0 到 127 之間）。
如果你傳入負數，例如 `chr(-1)`，或是傳入超過範圍的荒謬數字，Python 會立刻發出警告：
`ValueError: chr() arg not in range(0x110000)`！
另外，傳給 `chr()` 的必須是**整數**，傳入小數（如 `chr(65.0)`）或字串（如 `chr('65')`）會引發 `TypeError`，務必確認傳入的是純整數喔！


```python
# [2] Code 範例區：展示 chr() 召喚字元與互逆操作

# 範例 1：使用數字召喚英文字母
print(chr(65))   # 印出 'A'
print(chr(66))   # 印出 'B'
print(chr(97))   # 印出 'a'

# 範例 2：連續數字召喚字母接龍
demo_code = 70  # 70 是 'F'
print(chr(demo_code), chr(demo_code + 1), chr(demo_code + 2))  # 印出 F G H

# 範例 3：ord 與 chr 互逆驗證
demo_ch = 'X'
demo_recon = chr(ord(demo_ch))
print(demo_recon == demo_ch)  # 印出 True！完全復原！
```

### ⚖️ 3-2-3 經典差值 32（0x20）——純手工大小寫無痛轉換

在英文字母的世界裡，大寫字母與小寫字母就像是住在不同樓層的孿生兄弟。你有沒有算過，他們樓層之間相差幾公尺呢？
讓我們來做個簡單的減法：
小寫 `'a'` 的編碼是 `97`，大寫 `'A'` 的編碼是 `65`：
$$97 - 65 = 32$$
再檢查一組：小寫 `'b'`（`98`）減去大寫 `'B'`（`66`），依然是 **`32`**！
在十六進位制中，`32` 剛好就是 **`0x20`**（$2 \times 16 = 32$）！

**吳邦一老師講義第 10 頁與第 46 頁的經典必殺技：**
所有的英文字母，大寫與小寫之間**永遠嚴格相差 32**！這項特性是計算機歷史上最偉大的精巧設計之一：
- **大寫轉小寫**：只要把大寫編碼「加上 32」，再召喚成字元：
  `chr(ord('C') + 32)` ➔ 就會無痛變成 `'c'`！
- **小寫轉大寫**：只要把小寫編碼「減去 32」，再召喚成字元：
  `chr(ord('g') - 32)` ➔ 就會俐落變身為 `'G'`！

**APCS 考場實務價值：**
很多初學者在寫競賽題目時，若忘記了字串內建的 `.upper()` 或 `.lower()`，只要心中默念「差值 32」，單靠 `ord()` 與 `chr()` 就能純手工打造出 100% 正確的大小寫轉換器，在考場上無往不利！


```python
# [2] Code 範例區：展示大小寫字母差值 32 的手工轉換魔法

# 範例 1：驗證大小寫差值永遠為 32（十六進位 0x20）
print(ord('a') - ord('A'))  # 印出 32
print(ord('z') - ord('Z'))  # 印出 32

# 範例 2：大寫轉小寫（+32）
demo_up1 = 'E'
demo_low1 = chr(ord(demo_up1) + 32)
print(demo_low1)           # 印出 'e'

# 範例 3：小寫轉大寫（-32）
demo_low2 = 'q'
demo_up2 = chr(ord(demo_low2) - 32)
print(demo_up2)         # 印出 'Q'
```

### 📊 3-2-4 字母相對索引 ord(c) - ord('A')——26 個字母計數器起手式

在後續章節與 APCS 考題中，有一種無比高頻的題目：
「輸入一篇文章，統計 26 個英文字母每個字母出現了幾次？」
面對 26 個英文字母，難道我們要宣告 26 個變數 `count_a, count_b, ... count_z` 嗎？
當然不行！那樣寫會讓程式碼冗長到無法收拾！

**吳邦一老師講義第 26 頁與第 46 頁的提示秘笈：**
在後續進階章節中，我們可以用 26 個置物櫃格子，讓編號 `0` 代表 `'A'`，索引 `1` 代表 `'B'`，...，索引 `25` 代表 `'Z'`！
但問題來了：如何把英文字母快速變成 `0` 到 `25` 的整數索引呢？
核心公式正是：**「字母相對索引（Relative Indexing）」**！
$$\text{index} = \text{ord}(c) - \text{ord}('A')$$
- 當 $c = 'A'$ 時：$65 - 65 = \mathbf{0}$（正好對應第一個格子！）
- 當 $c = 'B'$ 時：$66 - 65 = \mathbf{1}$（正好對應第二個格子！）
- 當 $c = 'Z'$ 時：$90 - 65 = \mathbf{25}$（正好對應最後一個格子！）

如果是小寫字母，公式只要改成減去小寫基準 `'a'`：
$$\text{index} = \text{ord}(c) - \text{ord}('a')$$
掌握了這道公式，你就擁有了打通字元統計與陣列計數器之間的終極橋樑！


```python
# [2] Code 範例區：展示字母相對索引計算與 26 字母計數器原理

# 範例 1：大寫字母轉 0~25 相對序號
print(ord('A') - ord('A'))  # 印出 0（A 是第 0 號）
print(ord('B') - ord('A'))  # 印出 1（B 是第 1 號）
print(ord('Z') - ord('A'))  # 印出 25（Z 是第 25 號）

# 範例 2：小寫字母轉 0~25 相對序號
print(ord('a') - ord('a'))  # 印出 0
print(ord('c') - ord('a'))  # 印出 2

# 範例 3：查驗任意英文字母在字母表中的位置
demo_c = 'D'
demo_idx = ord(demo_c) - ord('A')
print("D 的字母編號（0 起算）:", demo_idx)        # 印出 3
print("若從第 1 個起算（自然序號）:", demo_idx + 1)  # 印出 4
```

### 🛡️ 3-2-5 凱撒密碼位移——字元循環輪替加密小魔法

在西元前古羅馬時代，尤利烏斯·凱撒（Julius Caesar）將軍為了傳送軍事機密，發明了一種簡單卻強大的加密方法：把信中的每一個英文字母，在字母表上「往後位移固定步數」！
例如，如果約定位移量為 `3`，那麼字母 `'A'` 就會變成 `'D'`，字母 `'B'` 就會變成 `'E'`！這就是資訊科學中最著名的「**凱撒密碼（Caesar Cipher）**」！

**運作機制與底層算式：**
透過今天學到的 `ord()` 與 `chr()`，我們可以在 Python 中優雅實作這項傳奇密碼：
1. 先用 `ord(c)` 取得字母的 ASCII 身分證。
2. 加上位移步數 `shift`，得到新的編號：`new_code = ord(c) + shift`。
3. 用 `chr(new_code)` 召喚出加密後的密文字元！

**如果位移超過 'Z' 該怎麼辦？（循環求模進階思維）**
如果字母是 `'Y'`，往後移 3 步就會衝出大寫字母範圍。在進階演算法中，我們會結合前一小節的相對索引與模除 `% 26` 達成無縫圓環輪替：
$$\text{cipher} = \text{chr}((\text{ord}(c) - \text{ord}('A') + \text{shift}) \% 26 + \text{ord}('A'))$$
在基礎階段，只要理解「**字元 ➔ 編號 ➔ 數學運算 ➔ 字元**」這套標準處理流，你就已經掌握了高階文字加密與字串處理的核心本質！


```python
# [2] Code 範例區：展示凱撒密碼字元位移加密

# 範例 1：基本字母位移加密
demo_orig = 'A'
demo_step = 3
demo_c_num = ord(demo_orig) + demo_step
demo_c_ch = chr(demo_c_num)
print(demo_c_ch)  # 印出 'D'（A 往後移 3 步）

# 範例 2：解密操作（反向倒退 3 步）
demo_dec = chr(ord(demo_c_ch) - demo_step)
print(demo_dec)  # 成功復原為 'A'！
```

```python
#@title 🧪 點擊此處取得單元 3-2 學習成效自動評分報告 { display-mode: "form" }
# ------------------------------------------------------------------------------
# 官方智慧全環境評分載入器（支援：本地開發庫優先 ➔ 公開正式庫雙重備援）
# ------------------------------------------------------------------------------
import os
import sys
import json
import logging
import urllib.request

# 靜音內部 httplib2 警告
logging.getLogger("google_auth_httplib2").setLevel(logging.ERROR)

# 1. 互動式輸入學員姓名與學號（分開獨立欄位，按 Enter 採用預設值）
print("👤【學員身分登記】（自學測試可直接按 Enter 鍵使用預設值）：")
student_name = input("請輸入您的姓名 [預設: 自學冒險者]：").strip() or "自學冒險者"
student_id = input("請輸入學號或座號 [預設: 無座號]：").strip() or "無座號"

# 2. 自動嘗試獲取 Google 帳號官方身分認證（真實 Email 與 Google 暱稱）
google_email = None
google_name = None
try:
    from google.colab import auth
    print("🔐 正在連結 Google 帳號進行身分認證（若跳出授權視窗，請點選允許）...")
    auth.authenticate_user()

    # 方法 1: 使用 Google Drive API v3 (about:get) 抓取真實 displayName 與 emailAddress
    # Colab ADC 預設包含 drive 權限，能 100% 精準獲取 Google 帳號中文顯示名稱（如：葉明昌）
    try:
        import google.auth
        from googleapiclient.discovery import build
        creds, _ = google.auth.default()
        drive_service = build('drive', 'v3', credentials=creds, cache_discovery=False)
        about = drive_service.about().get(fields="user(displayName,emailAddress)").execute()
        u_obj = about.get("user", {})
        google_name = u_obj.get("displayName")
        google_email = u_obj.get("emailAddress")
    except Exception:
        pass

    # 方法 2: 若 Drive API 失敗，嘗試透過 OAuth2 userinfo / tokeninfo 備援
    if not google_email or not google_name:
        try:
            import google.auth
            from google.auth.transport.requests import Request
            credentials, _ = google.auth.default()
            credentials.refresh(Request())
            token = credentials.token

            for ep in [
                f"https://www.googleapis.com/oauth2/v3/tokeninfo?access_token={token}",
                "https://www.googleapis.com/oauth2/v3/userinfo",
                "https://www.googleapis.com/oauth2/v2/userinfo"
            ]:
                try:
                    req = urllib.request.Request(
                        ep,
                        headers={"Authorization": f"Bearer {token}"}
                    )
                    with urllib.request.urlopen(req, timeout=5) as resp:
                        u_data = json.loads(resp.read().decode('utf-8'))
                        if not google_email:
                            google_email = u_data.get("email") or u_data.get("emailAddress")
                        if not google_name:
                            google_name = (
                                u_data.get("name")
                                or u_data.get("given_name")
                                or u_data.get("displayName")
                            )
                        if google_email and google_name:
                            break
                except Exception:
                    continue
        except Exception:
            pass

    # 方法 3: 若只有 email 但 Google API 沒給 name，以 email 前綴作為暱稱回退
    if google_email and not google_name:
        google_name = google_email.split("@")[0]

    if google_email:
        print(f"✅ 身分認證成功！已綁定 Google 帳號：{google_name} ({google_email})")
except Exception:
    # 非 Colab 環境、本機離線測試或略過授權
    pass

# 3. 身分資料獨立注入環境變數（學號與姓名完全分開）
os.environ["STUDENT_NAME"] = student_name
os.environ["STUDENT_ID"] = student_id
if google_email:
    os.environ["GOOGLE_EMAIL"] = google_email
if google_name:
    os.environ["GOOGLE_NAME"] = google_name

# 4. 評分核心腳本載入（本機優先 ➔ 遠端記憶體載入，絕不產生實體暫存檔）
# 找不到來源就換下一條。內容已拿到但執行失敗時，停止改試其他來源。
# 評分腳本放在獨立命名空間執行，成功後也不把名稱寫回筆記本。答案讀執行前的學生存放格快照。
_loaded = False
_load_error = None

# 4.1 優先嘗試讀取開發庫/本機檔案（本機離線開發測試零延遲）
_local_paths = [
    "web/graders/grade_3_2.py",
    "../graders/grade_3_2.py",
    "../../graders/grade_3_2.py",
    "../web/graders/grade_3_2.py",
    "../../web/graders/grade_3_2.py"
]
for _lp in _local_paths:
    if not os.path.exists(_lp):
        continue
    try:
        with open(_lp, "r", encoding="utf-8") as _f:
            _code = _f.read()
        _grader_ns = {"__builtins__": __builtins__, "__name__": "apcs_grader", "__STUDENT_ENV__": dict(globals())}
        try:
            exec(_code, _grader_ns)
        finally:
            del _grader_ns
        os.environ["GRADER_SOURCE"] = _lp
        _loaded = True
        break
    except Exception as _exc:
        _load_error = _exc
        break

# 4.2 若在雲端環境（Colab），自動透過公開正式庫雙重備援以純記憶體載入
if not _loaded and _load_error is None:
    _grader_urls = [
        "https://raw.githubusercontent.com/johnnyy-lab/APCS1to3/main/graders/grade_3_2.py",
        "https://johnnyy-lab.github.io/APCS1to3/graders/grade_3_2.py",
        "https://raw.githubusercontent.com/johnnyy-lab/APCS1to3-dev/main/web/graders/grade_3_2.py"
    ]
    for _url in _grader_urls:
        try:
            _req = urllib.request.Request(_url, headers={"User-Agent": "Mozilla/5.0"})
            with urllib.request.urlopen(_req, timeout=10) as _resp:
                _code = _resp.read().decode("utf-8")
        except Exception:
            continue
        try:
            _grader_ns = {"__builtins__": __builtins__, "__name__": "apcs_grader", "__STUDENT_ENV__": dict(globals())}
            try:
                exec(_code, _grader_ns)
            finally:
                del _grader_ns
            os.environ["GRADER_SOURCE"] = _url
            _loaded = True
            break
        except Exception as _exc:
            _load_error = _exc
            break

if not _loaded:
    if _load_error is not None:
        print("❌ 評分程式載入失敗，請稍後重新執行。若仍失敗，請把這個畫面交給老師。")
        if os.environ.get("GRADER_DEBUG", "").strip().lower() in ("1", "true", "yes", "on"):
            _reason = str(_load_error).strip().splitlines()
            _line = _reason[0][:180] if _reason else ""
            if _line:
                print("（開發診斷：" + type(_load_error).__name__ + "：" + _line + "）")
            else:
                print("（開發診斷：" + type(_load_error).__name__ + "）")
    else:
        print("❌ 評分系統連線異常，請確認網路連線狀態（或稍後重新執行）")
```
