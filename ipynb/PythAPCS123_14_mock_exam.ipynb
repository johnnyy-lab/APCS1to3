````python
# 開始：抽出 3 題，或沿用這份試卷
import json
import os
import random
import time
import urllib.request


def exam_read_text(local_paths, urls):
    """先讀本機檔，讀不到再依序試遠端網址。"""
    for path in local_paths:
        if os.path.exists(path):
            with open(path, "r", encoding="utf-8") as handle:
                return handle.read()
    last_error = None
    for url in urls:
        try:
            request = urllib.request.Request(url, headers={"User-Agent": "Mozilla/5.0"})
            with urllib.request.urlopen(request, timeout=10) as response:
                return response.read().decode("utf-8")
        except Exception as exc:
            last_error = exc
            continue
    if last_error is None:
        raise RuntimeError("讀不到模擬考題庫。")
    raise RuntimeError("讀不到模擬考題庫：" + str(last_error))


def exam_load_pool():
    """讀入題庫。"""
    text = exam_read_text(
        [
            "web/mock/ch14_pool.json",
            "../mock/ch14_pool.json",
            "../../mock/ch14_pool.json",
            "../web/mock/ch14_pool.json",
            "../../web/mock/ch14_pool.json",
        ],
        [
            "https://raw.githubusercontent.com/johnnyy-lab/APCS1to3/main/mock/ch14_pool.json",
            "https://johnnyy-lab.github.io/APCS1to3/mock/ch14_pool.json",
            "https://cdn.jsdelivr.net/gh/johnnyy-lab/APCS1to3@main/mock/ch14_pool.json",
        ],
    )
    pool = json.loads(text)
    if not isinstance(pool, list) or len(pool) < 3:
        raise RuntimeError("模擬考題庫至少要 3 題。")
    return pool


def exam_signature(item):
    """用該題專用的 answer_name 組出 def 簽名行，避免多題同名互相覆蓋。"""
    sig = item["signature"].strip()
    ans_name = item.get("answer_name") or item["function"]
    if sig.startswith("def ") and "(" in sig:
        return "def " + ans_name + sig[sig.index("("):]
    return sig


def exam_show_paper(paper, note):
    """印出這 3 題的題幹與專用函式簽名。"""
    blocks = [
        note,
        "",
        "本卷題號：" + "、".join(item["id"] + " " + item["code"] for item in paper),
        "",
    ]
    for index, item in enumerate(paper, 1):
        blocks.append("## 第 " + str(index) + " 題　" + item["id"] + " " + item["code"] + " " + item["title"])
        blocks.append("")
        blocks.append(item["statement"].strip())
        blocks.append("")
        blocks.append("### 本題要你完成的函式")
        blocks.append("")
        blocks.append(item["returns"])
        blocks.append("")
        blocks.append("```python")
        blocks.append(exam_signature(item))
        blocks.append("    pass")
        blocks.append("```")
        blocks.append("")
    text = "\n".join(blocks)
    try:
        from IPython.display import Markdown, display
        display(Markdown(text))
    except Exception:
        print(text)
    return text


if not isinstance(globals().get("exam_paper"), list) or len(exam_paper) != 3:
    exam_paper = random.sample(exam_load_pool(), 3)
    exam_started_at = time.time()
    exam_was_redrawn = False
    exam_note = "已抽出 3 題。請把簽名貼到下面三格作答，寫完再執行「交卷」。同一份執行環境再跑這一格，不會換題。"
else:
    exam_note = "沿用這份試卷，沒有重新抽題。"

exam_show_paper(exam_paper, exam_note)
````

````python
# 交卷：只跑這 3 題的對拍
import os
import time
import urllib.request


def exam_read_text(local_paths, urls):
    """先讀本機檔，讀不到再依序試遠端網址。"""
    for path in local_paths:
        if os.path.exists(path):
            with open(path, "r", encoding="utf-8") as handle:
                return handle.read()
    last_error = None
    for url in urls:
        try:
            request = urllib.request.Request(url, headers={"User-Agent": "Mozilla/5.0"})
            with urllib.request.urlopen(request, timeout=10) as response:
                return response.read().decode("utf-8")
        except Exception as exc:
            last_error = exc
            continue
    if last_error is None:
        raise RuntimeError("讀不到評分程式。")
    raise RuntimeError("讀不到評分程式：" + str(last_error))


def exam_load_grader(filename):
    """載入該節評分腳本，但不跑教學用的整張成績單。"""
    text = exam_read_text(
        [
            "web/graders/" + filename,
            "../graders/" + filename,
            "../../graders/" + filename,
            "../web/graders/" + filename,
            "../../web/graders/" + filename,
        ],
        [
            "https://raw.githubusercontent.com/johnnyy-lab/APCS1to3/main/graders/" + filename,
            "https://johnnyy-lab.github.io/APCS1to3/graders/" + filename,
            "https://cdn.jsdelivr.net/gh/johnnyy-lab/APCS1to3@main/graders/" + filename,
        ],
    )
    os.environ["APCS_MOCK_CHECK_ONLY"] = "1"
    namespace = {"__builtins__": __builtins__, "__name__": "apcs_grader"}
    try:
        exec(text, namespace)
    finally:
        os.environ.pop("APCS_MOCK_CHECK_ONLY", None)
    return namespace


def exam_verdict(total):
    """150 分以上才寫出通過線。"""
    if total >= 150:
        return "達到 150 分通過線。"
    return "這份成績低於通過線。"


def exam_score_one(problem, student_env):
    """單題對拍。通過 100 分，否則 0 分。"""
    ans_name = problem.get("answer_name") or problem["function"]
    check_key = problem["function"]
    fn = student_env.get(ans_name)
    if not callable(fn):
        return 0, "找不到函式 " + ans_name + "。請依題目簽名定義它，並執行該作答格。"
    namespace = exam_load_grader(problem["grader"])
    checker = namespace.get(problem["check"])
    if not callable(checker):
        return 0, problem["id"] + " 的評分程式缺少 " + problem["check"] + "。"
    try:
        result = checker({check_key: fn})
    except Exception as exc:
        return 0, "評分時發生錯誤：" + type(exc).__name__ + ": " + str(exc)
    if not (isinstance(result, tuple) and len(result) == 2):
        return 0, "評分程式回傳格式不正確。"
    ok, message = result
    return (100 if ok else 0), str(message)


def exam_grade_report(paper, student_env, started_at, redrawn, now=None):
    """組成三題成績單。"""
    if not isinstance(paper, list) or len(paper) != 3 or not started_at:
        return "請先執行「開始」。"
    clock = time.time() if now is None else now
    elapsed = max(0, int(clock - started_at))
    minutes, seconds = divmod(elapsed, 60)
    lines = []
    if redrawn:
        lines.append("這是重新抽出的題組。作答格裡可能還留著上一份的函式，本次只評這三題。")
        lines.append("")
    total = 0
    for index, problem in enumerate(paper, 1):
        score, message = exam_score_one(problem, student_env)
        total += score
        lines.append(
            "第 " + str(index) + " 題 "
            + problem["id"] + " " + problem["code"] + " " + problem["title"]
            + "：" + str(score) + " 分"
        )
        lines.append(message)
        lines.append("")
    lines.append("總分 " + str(total) + " / 300")
    lines.append(exam_verdict(total))
    lines.append("用時 " + str(minutes) + " 分 " + f"{seconds:02d}" + " 秒。")
    if elapsed > 120 * 60:
        lines.append("已超過 120 分鐘，這次成績仍照算，並註記逾時。")
    return "\n".join(lines)


if not isinstance(globals().get("exam_paper"), list) or not globals().get("exam_started_at"):
    print("請先執行「開始」。")
else:
    print(exam_grade_report(
        exam_paper,
        globals(),
        exam_started_at,
        bool(globals().get("exam_was_redrawn")),
    ))
````
