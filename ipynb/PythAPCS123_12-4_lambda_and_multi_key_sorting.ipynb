### 12.4.1 為什麼需要 `lambda`？免命名的臨時輕量函數（一分鐘掌握 `lambda x: ...`）

#### 1. 生活故事比喻：便利貼紙條 vs 蓋章簽約的正式聘書
在上一節中，我們用 `def` 定義函數，就像是人事部門起草了一份正式的員工合約：給他一個正式的職稱名字（如 `def get_second`）、註冊在公司員工名冊中、佔用一個固定位置。
但如果你今天只是在辦公桌上要計算一個簡單的數字，隨手在身旁抽出一張「黃色便利貼（Post-it Note）」，上面寫著「把輸入的值乘以 2」，算完這一步之後隨手揉掉扔進回收桶，你根本不需要大費周章幫這張便利貼舉行隆重的命名儀式！
在 Python 語言中，**`lambda` 就是這張功能強大、免命名的隨身便利貼**！

#### 2. 底層運作機制：`def` 與 `lambda` 的精準對照
`lambda` 的語法結構極度精簡：
```python
lambda 參數名稱: 運算表達式
```
讓我們將它與一般的 `def` 進行一對一對照：
- 傳統寫法：
  ```python
  def square(x):
      return x ** 2
  ```
- `lambda` 一行化寫法：
  ```python
  lambda x: x ** 2
  ```
看到差別了嗎？
1. 關鍵字由 `def` 換成 `lambda`。
2. 省略了函數名稱，直接寫參數清單（例如 `x`）。
3. 用冒號 `:` 分隔參數與主體。
4. **自動回傳！** 冒號右邊的表達式算出來的結果，會自動被 `return` 出去，因此**絕對不能顯式寫出 `return` 關鍵字**！

#### 3. 初學者常見陷阱：在 lambda 冒號右側寫 `return` 或複雜陳述句
初學同學最容易犯的報錯：
```python
bad_lambda = lambda x: return x + 1  # 語法錯誤！SyntaxError: invalid syntax
```
請牢記鐵律：**`lambda` 的冒號右側只能放單一「表達式（Expression）」，不能放 `return`、不能放賦值等號 `=`、也不能放多行迴圈！** 它的天職就是提供最純粹、最輕量的一行運算。

#### 4. APCS 實戰視野
`lambda` 是 Python 撰寫競賽程式碼時最受喜愛的工具。它讓你的程式碼不用在前面塞滿 5、6 個只用一次的微型 `def`，讓所有排序邏輯緊湊地收納在呼叫現場，極大地提升了答題與除錯的流暢度。

````python
# 範例 12.4.1：一行搞定！lambda 匿名函式語法
# 1. 將傳統具名函式與 lambda 進行對照
def demo_add_five(x):
    return x + 5

demo_lafv = lambda x: x + 5

print("具名函式計算 10 + 5 =", demo_add_five(10))
print("lambda 運算 10 + 5 =", demo_lafv(10))

# 2. 在 sorted() 的 key 參數直接內聯（Inline）定義 lambda
demo_ns124 = [4, 15, 2, 28, 9, 31]
print("\n原始數列：", demo_ns124)

# 依照「與 10 的距離（絕對值）」排序
demo_dst10 = sorted(demo_ns124, key=lambda x: abs(x - 10))
print("依與 10 的距離由近到遠排序：", demo_dst10)
````

### 12.4.2 二維串列與元組客製排序：`key=lambda x: x[1]`（依特定欄位排序）

#### 1. 生活故事比喻：試算表軟體（Excel）的「排序列」選擇
在日常使用 Google 試算表或 Excel 時，如果你的表格有兩欄：A 欄是「產品品名」，B 欄是「售價」。
當你選取整個表格按下「排序」按鈕時，軟體一定會跳出一個視窗問你：「請問您希望依據哪一欄來排序？」
你可以自由勾選「依據 B 欄（售價）由小到大排序」。
在 Python 程式碼中，`key=lambda x: x[1]` 就是你用程式碼勾選「依據第二欄（索引 1）進行排序」的指令！

#### 2. 底層運作機制：結構解構與索引投影
當我們有一組二維資料，例如元組清單 `items = [("筆電", 32000), ("滑鼠", 600), ("螢幕", 8500)]`：
如果呼叫 `items.sort(key=lambda x: x[1])`：
1. 傳入 `x = ("筆電", 32000)`，`x[1]` 抽取數值 `32000`。
2. 傳入 `x = ("滑鼠", 600)`，`x[1]` 抽取數值 `600`。
3. 傳入 `x = ("螢幕", 8500)`，`x[1]` 抽取數值 `8500`。
排序演算法比較這三個數值：$600 < 8500 < 32000$。
原二維元素被重新擺放到位，輸出：`[("滑鼠", 600), ("螢幕", 8500), ("筆電", 32000)]`！

#### 3. 初學者常見陷阱：欄位索引越界（IndexError）
如果你的二維清單中，有些子元素包含 2 個項目，但有些子元素只有 1 個項目（甚至是空串列）：
呼叫 `lambda x: x[1]` 時，遇到只有 1 個項目的子串列就會當場拋出 `IndexError: list index out of range`！因此在多維排序前，請務必確保資料格式具備一致的欄位長度。

#### 4. APCS 實戰視野
在 APCS 歷屆實作題目中，輸入的資料幾乎 80% 以上都是結構化資料（如座標 `(x, y)`、線段 `(l, r)`、成績 `(name, score)`）。熟練運用 `lambda x: x[1]` 或任意欄位索引，是處理所有二維陣列題目的必備核心基本功。

````python
# 範例 12.4.2：多元組結構依特定欄位排序
# 學生名冊：(姓名, 國文成績, 數學成績)
demo_s124 = [
    ("小明", 82, 95),
    ("小華", 90, 88),
    ("小美", 75, 98),
    ("小英", 95, 76)
]
print("原始名冊：")
for demo_s in demo_s124:
    print(f"  {demo_s[0]:<4} 國文: {demo_s[1]:2d}, 數學: {demo_s[2]:2d}")

# 1. 依數學成績（索引 2）降序排列
demo_mdsc = sorted(demo_s124, key=lambda record: record[2], reverse=True)
print("\n依數學成績高低排名：", demo_mdsc)

# 2. 依國文成績（索引 1）升序排列
demo_chns = sorted(demo_s124, key=lambda record: record[1])
print("依國文成績低到高排序：", demo_chns)
````

### 12.4.3 多準則雙鍵值排序（元組打包技巧）：首要條件升序、次要條件升序 `key=lambda x: (x[0], x[1])`

#### 1. 生活故事比喻：大考分發的同分比序規則
在國中教育會考或大學學測的分發比序中，常常遇到這樣的情境：
「先比總積分，積分高的優先錄取；如果兩位同學總積分一模一樣，再啟動第二輪比序，比國文單科成績；若國文再同分，最後比數學成績。」
這種「一關一關過、先比第一條件，若平手才啟用第二條件」的決策流程，在電腦科學中稱為**多準則複合排序（Multi-criteria Sorting）**。
在 Python 中，我們該如何教電腦進行這種「平手時啟動次要條件」的比序呢？答案超乎想像地優雅——**直接把多個特徵打包成一個元組（Tuple）回傳！**

#### 2. 底層運作機制：元組的天生字典序比大小哲學
在第十章我們學過：**Python 元組（Tuple）比大小天生遵循字典序！**
當電腦比較兩個元組 `(A1, B1) < (A2, B2)` 時：
1. 它會先比對第 0 個位置：如果 `A1 != A2`，直接由 `A1` 和 `A2` 的勝負決定大小，後面的 `B` 完全不看！
2. 只有當 `A1 == A2`（第一條件平手打成平局）時，電腦才會自動移向第 1 個位置，比對 `B1` 與 `B2`！
因此，當我們寫出：
```python
key=lambda x: (x[0], x[1])
```
排序引擎拿到的特徵鑰匙就是一個包含兩項的元組。它會自動在第 0 欄相同時，平滑無縫地啟動第 1 欄的比對！

#### 3. 初學者常見陷阱：在 lambda 冒號後面忘記加元組小括號
初學同學常犯的語法筆誤：
```python
key=lambda x: x[0], x[1]  # 致命錯誤！
```
逗號在 Python 語法解析中的優先級很低。如果不加外層小括號，直譯器會把 `key=lambda x: x[0]` 當作第一個引數，而把後面的 `x[1]` 當作傳給 `sort()` 的第二個未知引數，當場報錯 `NameError` 或 `TypeError`！請務必將多個欄位用小括號牢牢包裝成元組：`(x[0], x[1])`。

#### 4. APCS 實戰視野
APCS 題目中最經典的二維座標排序規格：「將所有座標點由左至右（X 座標升序）排列；若 X 座標相同，則由下至上（Y 座標升序）排列」。只要寫 `key=lambda pt: (pt[0], pt[1])`，瞬間完全滿足題目所有繁瑣的同分比序要求！

````python
# 範例 12.4.3：多準則鍵值排序（打包元組技巧）
# 平面座標點清單：(X 座標, Y 座標)
demo_pt124 = [(3, 5), (1, 8), (3, 2), (2, 4), (1, 3), (3, 7)]
print("原始座標點清單：", demo_pt124)

# 需求：先按 X 升序；若 X 相同，再按 Y 升序
demo_srtxy = sorted(demo_pt124, key=lambda pt: (pt[0], pt[1]))
print("\n先看 X、再看 Y 雙重升序排序結果：", demo_srtxy)
````

### 12.4.4 多準則混合升降序排序：首要條件依成績降序、次要條件依學號升序 `key=lambda x: (-x[1], x[0])`

#### 1. 生活故事比喻：奧運獎牌榜與學業榮譽榜的「一降一升」
在全校學期成績總結算時，教務處訂下了鐵律：
- 第一準則：成績越高越好（由大到小降序排列）。
- 第二準則：如果兩人分數完全相同，依學號由小到大（由小到大升序排列，學號小的排在前面）。
初學者一看到這個需求立刻傻眼：「如果我加上 `reverse=True`，成績是降序了沒錯，但學號也會跟著被反轉成降序！如果我不加 `reverse=True`，學號是升序了，但成績卻變成由低到高！天下哪有一隻眼睛看升序、另一隻眼睛看降序的魔法？」
有的！這正是結合我們在 12.2.5 節學到的**負號取反技巧（`-x`）**與元組打包的終極融合技！

#### 2. 底層運作機制：負號倒影帶來的維度獨立控制
回憶一下：在升序排序中，數值越小排在越前面。
如果我們希望「成績越大越排在前面」，我們只要在成績前面硬生生加上一個負號：`-成績`！
- 成績 100 分的人，特徵值為 `-100`。
- 成績 90 分的人，特徵值為 `-90`。
因為 $-100 < -90$，在預設的升序引擎眼中，`-100` 小於 `-90`，因此 100 分的人順理成章搶先排在最前面！
而學號部分，我們維持正數 `學號` 不動。
於是，特徵元組設計為：
```python
key=lambda x: (-x[1], x[0])
```
- 第一項：`-成績`，達成**成績降序**！
- 第二項：`學號`，若成績平手，達成**學號升序**！
完全不需要動用全域的 `reverse=True`，輕輕鬆鬆實現各維度隨心所欲的獨立升降序控制！

#### 3. 初學者常見陷阱：對字串欄位加負號引發報錯
再次強烈提醒：**負號運算子 `-` 只能施加在數字（整數 `int` 或浮點數 `float`）身上！**
如果學號是字串 `"S001"`，千萬不要寫 `-x[0]`，否則直譯器會立刻拋出 `TypeError: bad operand type for unary -: 'str'`。

#### 4. APCS 實戰視野
「高分者在前，同分者依輸入序或編號小者在前」是 APCS 實作考題出現頻率最高、殺傷力最大的題型模板。只要掌握 `(-score, id)` 這一招神仙寫法，考場上任何混合升降序難題都能在 3 秒內迎刃而解！

````python
# 範例 12.4.4：混合升降序技巧（成績降序、學號升序）
# 學生名冊格式：(學號, 成績, 姓名)
demo_smix = [
    (101, 90, "小明"),
    (102, 85, "小華"),
    (103, 72, "小美"),
    (104, 90, "小英"),
    (105, 85, "小王")
]
print("原始名冊：", demo_smix)

# 成績降序（加負號 -record[1]）、學號升序（正號 record[0]）
demo_rmix = sorted(demo_smix, key=lambda record: (-record[1], record[0]))
print("\n成績降序、學號升序排序結果：", demo_rmix)
````

### 12.4.6 APCS 考場經典：平面座標點依原點距離平方 `x**2 + y**2` 排序

#### 1. 生活故事比喻：雷達掃描與颱風暴風圈中心
想像你是一位中央氣象局的雷達監控員，雷達螢幕正中心是台北市（原點 $(0, 0)$）。
現在海面上出現了許多個熱帶氣旋或雨胞觀測站，每個觀測站有其相對座標 $(x, y)$。
局長下達緊急指令：「請立刻將所有觀測站，依照『離台北市中心直線距離由近到遠』排定監控優先順序！」
直覺上，我們要計算歐幾里得直線距離 $\sqrt{x^2 + y^2}$。
但身為頂尖程式設計師，我們需要大費周章去呼叫開根號運算嗎？
完全不需要！因為如果 $A^2 < B^2$，那麼在正數世界裡 $A < B$ 保證絕對成立！
直接拿「距離的平方 $x^2 + y^2$」作為排序的 key，不僅速度極快，更能避開浮點數開根號的精度誤差！

#### 2. 底層運作機制：代數轉換與無損精度計算
在 APCS 考場中，只要題目要求「依距離大小排序」，**一律比對距離平方**：
對於平面座標點 `pt = (x, y)`：
- 原點距離平方公式：$x^2 + y^2$，在 Python 中寫為 `pt[0]**2 + pt[1]**2`。
如果座標包含正負號（例如第四象限 $(3, -4)$ 與第二象限 $(-4, 3)$）：
- 平方運算會自動消除負號：$3^2 + (-4)^2 = 9 + 16 = 25$。
這完美符合了幾何上對稱距離的定義！

#### 3. 初學者常見陷阱：負數次方的括號陷阱
在第二章我們學過運算優先級：`-4**2` 會被解讀為 `-(4**2) == -16`！
但在座標元組中取值 `pt[0]**2`，因為 `pt[0]` 已經是單一負整數，例如 `(-4)**2 == 16`，這是正確的。
但請注意：千萬不要寫成 `pt[0] * pt[0] + pt[1] * pt[1]` 以外的多餘括號打架，直接寫 `pt[0]**2 + pt[1]**2` 或乘法連乘是最穩健的寫法。

#### 4. APCS 實戰視野
此題型是 APCS 實作第二級至第三級的超高頻幾何考點。題目通常還會附加次要條件：「若離原點距離相同，則 X 座標較小者在前；若 X 再相同，則 Y 座標較小者在前」。
結合我們剛剛學會的元組打包，一行代碼就能擊穿考題：
`key=lambda pt: (pt[0]**2 + pt[1]**2, pt[0], pt[1])`！

````python
# 範例 12.4.6：座標幾何距離平方投影排序
demo_pt_gm = [(3, 4), (1, 1), (-2, -2), (0, 5), (-3, 4), (2, 1)]
print("原始座標清單：", demo_pt_gm)

# 依距離原點 (0, 0) 之平方距離由近到遠排序
demo_sdst = sorted(demo_pt_gm, key=lambda pt: pt[0]**2 + pt[1]**2)
print("\n依離原點距離由近到遠排序：", demo_sdst)
````

````python
# ==========================================
# [4] Code 練習題 12.4.6
# 任務：
# 給定一組二維座標點 pt124。
# 請使用 sorted 配合 lambda 將所有座標點依「距離原點 (0,0) 的距離平方（x^2 + y^2）」從小到大排序，
# 並輸出排序後的串列，以及距離原點最近的點與其距離平方。
#
# 【公開測資 1】
# pt124 = [(4, 3), (1, 2), (0, 3), (-1, -1)]
# 預期輸出：
# 距離排序結果： [(-1, -1), (1, 2), (0, 3), (4, 3)]
# 最近點： (-1, -1) (距離平方: 2)
#
# 【公開測資 2】
# pt124 = [(10, 0), (0, 1)]
# 預期輸出：
# 距離排序結果： [(0, 1), (10, 0)]
# 最近點： (0, 1) (距離平方: 1)
# ==========================================
# 請在下方撰寫程式碼並測試：
pt124 = [(4, 3), (1, 2), (0, 3), (-1, -1)]
srt_pt = sorted(pt124, key=lambda p: p[0]**2 + p[1]**2)
closest = srt_pt[0]
dist_sq = closest[0]**2 + closest[1]**2
print("距離排序結果：", srt_pt)
print(f"最近點： {closest} (距離平方: {dist_sq})")
````

````python
# ==========================================
# [5] Code 挑戰題 12.4.6
# 任務說明：
# 某尋寶遊戲在地圖上放置了 5 個寶箱，座標分別為：
# chests = [(5, 12), (0, -10), (-6, -8), (9, 0), (7, 7)]
# 玩家目前的位置在起點 (1, 2)。
# 請計算每個寶箱離玩家目前位置的「相對距離平方（(x - 1)**2 + (y - 2)**2）」，
# 並將寶箱清單依離玩家目前位置由近到遠排序後印出。
# （本題為自由挑戰題，無公開測資，請依題目情境自行思考並撰寫完整程式碼）
# ==========================================
# 請在下方撰寫你的程式碼：
chests = [(5, 12), (0, -10), (-6, -8), (9, 0), (7, 7)]
px, py = 1, 2
srt_chst = sorted(chests, key=lambda c: (c[0] - px)**2 + (c[1] - py)**2)
print(f"從玩家位置 ({px}, {py}) 出發的尋寶最佳順序：")
for c in srt_chst:
    d2 = (c[0] - px)**2 + (c[1] - py)**2
    print(f"  寶箱座標: {c:<9} -> 距離平方: {d2}")
````
