### 11.6.1 遞迴第一步：認識「函數呼叫自己」與鏡中鏡意象

#### 1. 生活故事比喻：理髮廳的兩面大鏡子與俄羅斯套娃
當你走進理髮廳坐在椅子上，正前方與正後方各有一面大鏡子。當你望向前方的鏡子時，你會看到鏡子裡面映出了後面的鏡子，而後面的鏡子裡面又映出了前方的鏡子……鏡中鏡、身外身，影像無窮無盡地往深處延伸！
另一個經典比喻是「俄羅斯套娃（Matryoshka）」。當你拿起一個精美的大木偶，打開它的肚子，裡面裝著一個一模一樣但縮小了一號的中木偶；你再打開中木偶，裡面又有一個更小的小木偶。
在程式設計的世界中，這種**「為了解決一個大問題，在函數執行過程中，呼叫一個結構一模一樣、但規模縮小了一號的自己」**的技巧，就叫做**遞迴（Recursion）**！

#### 2. 底層運作機制：遞（Forward）與迴（Return）的雙向旅程
「遞迴」這兩個字，精準道出了它包含的兩個完全相反方向的生命旅程：
- **「遞（Dive / Push）」**：去程。大問題無法直接算出答案，於是大問題把任務拆成縮小版的小問題，向下委託給下一個自己。此時上一層處於等待狀態，一層層往下探尋。
- **「迴（Climb Back / Pop）」**：回程。當下潛到最微小的基底狀態有了確定答案後，答案開始一層層往上交接、層層計算組裝，最終回到最初發起呼叫的最外層函數！
- **關鍵本質**：當函數呼叫自己時，電腦直譯器並不是「跳回原本那行代碼重跑」，而是在記憶體中**建立了一個全新的獨立函數分身**！每個分身都有自己獨立的參數與區域變數，互不相干。

#### 3. 初學者常見陷阱：把遞迴當成迴圈（Loop）
很多初學同學會困惑：「這跟 while 迴圈有什麼兩樣？」
兩者的思維模型完全不同：
- **迴圈（Iteration）**：是一個人在同一個房間裡，看著同一張計數紙，不斷用橡皮擦擦掉舊數字、寫上新數字（原地狀態更新）。
- **遞迴（Recursion）**：是一個人走進新房間，留下一封信給外面的自己；裡面的人又走進更新的房間……最後房間裡的人把答案寫在信封上，一層層往外傳出來！

#### 4. APCS 實戰視野
遞迴是 APCS 實作題第二級邁向第三級、第四級不可逾越的龍門。樹狀結構走訪、深度優先搜尋（DFS）、二分搜尋、分治演算法（Divide and Conquer），全數以遞迴為根本基石。

````python
# 範例 11.6.1：最直觀的遞迴倒數計時器

def countdown(n):
    # 印出當前所在的層級
    print(f"  [遞推中] 當前數字: {n}")
    
    # 煞車條件：數到 0 就停下來
    if n == 0:
        print("  🚀 倒數結束，火箭發射！")
        return
    
    # 核心：函數呼叫了縮小規模的自己 (n - 1)
    countdown(n - 1)
    
    # 注意：這行是在「回程」時才會被執行的！
    print(f"  [折返中] 回到了層級: {n}")

# 主程式啟動遞迴
print("[主程式] 準備開始倒數...")
countdown(3)
print("[主程式] 遞迴旅程圓滿結束！")
````

### 11.6.2 呼叫堆疊（Call Stack）物理模型：棧針推入（Push）與彈出（Pop）過程

#### 1. 生活故事比喻：自助餐餐廳的彈簧托盤架
走進大學的自助餐廳，在取餐處有一座金屬的彈簧托盤架。
廚房洗碗阿姨洗好第一個紅色托盤，由上方用力向下壓進架子裡（Push，壓入）；接著洗好第二個藍色托盤，壓在紅色托盤上方；再洗好第三個黃色托盤，壓在最上面。
此時學生走進來取餐，學生伸手只能拿到哪一個托盤？**只能拿最上面的黃色托盤（Pop，彈出）**！
最底下那個最早被放進去的紅色托盤，必須耐心等待上面所有後來放進去的托盤全被客人拿走之後，才終於能夠重見天日！這就是電腦資料結構中最著名的鐵律——**「後進先出（Last In, First Out，簡稱 LIFO）」**！

#### 2. 底層運作機制：堆疊框架（Stack Frame）的微觀世界
在電腦記憶體中，Python 直譯器為每一次函數呼叫都維護著一個專屬的**「呼叫堆疊（Call Stack）」**：
1. **推入（Push）**：當程式呼叫一個函數（例如 `f(3)`）時，系統在 Stack 最頂端壓入一個「堆疊框架（Stack Frame）」，記錄傳入的引數 `n=3`、區域變數與目前執行行數。
2. **暫停凍結**：當 `f(3)` 內部呼叫了 `f(2)`，`f(3)` 的執行被**瞬間凍結**，留在棧底等待；新的 `f(2)` Frame 壓在它上面。
3. **深入基底**：接著壓入 `f(1)`、`f(0)`。此時 Stack 累積了 4 個框架。
4. **彈出甦醒（Pop）**：當最頂端的 `f(0)` 執行完畢 `return` 時，`f(0)` 的 Frame 被立刻彈出銷毀！
   此時，暫停在下方的 `f(1)` 重新甦醒，接續執行呼叫語句後方的程式碼，執行完畢後也彈出；接著輪到 `f(2)`、最後輪到 `f(3)` 彈出！

#### 3. 初學者常見迷思：忽略了「呼叫後方的回程代碼」
初學同學最常犯的直覺錯誤是：以為函數呼叫了自己之後，該函數就此結束。
請務必記住：**呼叫自己只是一次「中途暫停」！**
當深層的分身完成任務彈出後，上一層函數會從暫停點醒來，繼續執行呼叫語句**後面**剩下的所有代碼！這就是為什麼遞迴具有天然的「逆序對稱」特徵：
- 寫在遞迴呼叫**前面**的指令：按照 $1 \rightarrow 2 \rightarrow 3$ 的順序執行（去程）。
- 寫在遞迴呼叫**後面**的指令：按照 $3 \rightarrow 2 \rightarrow 1$ 的倒序執行（回程）！

#### 4. APCS 實戰視野
在 APCS 觀念題中，遞迴代碼追蹤題是每屆必考題型。解題時千萬不要憑空在大腦裡空想，只要在草稿紙邊緣畫出一個垂直的「長條方木盒（Call Stack）」，一層層把框架寫上去、再一層層由上往下擦掉，答案立刻水落石出！

````python
# 範例 11.6.2：透視 Call Stack 的進出對稱印記

def explore_cave(depth):
    # --- 【去程：Push 階段】---
    print(f"👉 [入洞 Push] 正在下潛至第 {depth} 層洞穴...")
    
    if depth == 3:
        print(f"  💎 抵達第 {depth} 層洞底，發現神祕水晶，準備折返！")
    else:
        # 暫停當前層，呼叫下一層
        explore_cave(depth + 1)
        
    # --- 【回程：Pop 階段】---
    # 注意：這行是在下一層彈出後，甦醒接續執行的代碼！
    print(f"👈 [出洞 Pop]  離開第 {depth} 層洞穴，返回上一層！")

# 觀察輸出順序的對稱美感
print("=== 冒險開始 ===")
explore_cave(1)
print("=== 安全重返地面 ===")
````

### 11.6.3 終止條件（Base Case）：遞迴能停下來的唯一煞車皮

#### 1. 生活故事比喻：下坡賽車的煞車皮與安全氣囊
想像一輛賽車在下坡山路上風馳電掣。每跑完一公里，速度就加快一倍（問題規模遞減，進入更深層的遞迴）。
如果這輛賽車上完全沒有安裝煞車皮（沒有終止條件 Base Case），那麼當賽車抵達終點線時，它絕不可能自己停下來，而是會以時速三百公里狂暴衝出護欄、墜入無底深淵！
在遞迴的世界中，**終止條件（Base Case，或稱基底條件）**就是這輛賽車的唯一煞車皮。它明確規定：「當車輪碰到終點線時，立刻踩死煞車，絕對不准再踩油門呼叫下一層！」

#### 2. 底層運作機制：遞迴函數的黃金雙重結構
任何一個能正常終止的遞迴函數，其代碼內部必須嚴格分為互斥的兩大區塊：
1. **Base Case（終止條件 / 基底條件）**：
   - 通常置於函數開頭的第一行。
   - 檢查當前參數是否已經到達「最小的、已知的、無需再拆解的平凡狀態」（例如 $n == 0$、$n == 1$ 或字串長度為 0）。
   - **一旦命中，立即執行 `return` 回傳具體答案，絕不再呼叫自己！**
2. **Recursive Case（遞迴推進步驟）**：
   - 當未命中 Base Case 時執行。
   - 將當前問題規模縮小（例如傳入 $n - 1$ 或切片字串），並呼叫自己。

#### 3. 初學者常見陷阱與致命 Bug
- **陷阱一：忘記在 Base Case 寫 `return`**  
  初學同學常寫出：
  ```python
  def bad_func(n):
      if n == 0:
          print("到底了")  # 致命漏寫 return！
      bad_func(n - 1)      # 依然會被執行，煞車徹底失靈！
  ```
  這會讓程式印出「到底了」之後，繼續往下傳入 -1、-2……一路暴衝直到崩潰！
- **陷阱二：終止條件過於脆弱**  
  如果條件寫 `if n == 0:`，而呼叫者不小心傳入了負數（如 -1），或者遞迴步進是 `n - 2` 恰好從 3 跳過 0 變成 -1，那麼 `n == 0` 永遠不會成立！
  防禦性設計必須使用不等式：**`if n <= 0:`**，為程式築起堅不可摧的安全防波堤。

#### 4. APCS 實戰視野
撰寫任何遞迴程式碼時，請在腦海中默念黃金第一鐵律：**「動手先寫 Base Case，再寫 Recursive Case」**！先確保程式能安全著陸，再去構思如何化繁為簡。

### 11.6.4 經典範例：階乘累乘 $N!$ 與數列累加的呼叫與歸還路徑追蹤

#### 1. 生活故事比喻：大將軍的階梯公文批示
古代戰場上，大將軍想要計算五軍聯合作戰的編隊可能數 $5!$（5的階乘）。
大將軍對副將說：「你去把 $4!$ 算出來回報給我，我收到後乘上 5，就是全軍答案！」
副將轉頭對參謀長說：「你去把 $3!$ 算出來回報給我，我乘上 4 回報給將軍！」
公文一層層往下簽發，最後到了前線最基層的列兵手中。列兵打開最高作戰守則（Base Case），手冊上白紙黑字寫著：**「規定 $1! = 1$」**。
列兵大喊：「報告長官，$1!$ 的答案是 1！」
公文開始反向層層回傳批示：
參謀拿到 1，計算 $2 \times 1 = 2$ 回報副將；副將計算 $3 \times 2 = 6$ 回報中將；中將計算 $4 \times 6 = 24$ 回報大將軍；大將軍最後計算 $5 \times 24 = 120$！

#### 2. 底層運作機制：數學歸納與回傳值組裝
階乘函數的嚴格數學定義：
$$N! = \begin{cases} 1 & \text{if } N \le 1 \quad (\text{Base Case}) \\ N \times (N-1)! & \text{if } N > 1 \quad (\text{Recursive Case}) \end{cases}$$

觀察它的執行路徑，每個 Stack Frame 的回傳與計算如下：
1. `fact(4)` 等待 `4 * fact(3)`
2. `fact(3)` 等待 `3 * fact(2)`
3. `fact(2)` 等待 `2 * fact(1)`
4. `fact(1)` 命中 Base Case，直接回傳 `1`！
5. **回程組裝階段**：
   - `fact(2)` 收到 1，計算 $2 \times 1 = 2$ 並回傳。
   - `fact(3)` 收到 2，計算 $3 \times 2 = 6$ 並回傳。
   - `fact(4)` 收到 6，計算 $4 \times 6 = 24$ 並回傳給主程式！

#### 3. 初學者常見陷阱：累乘與累加的 Base Case 混淆
請特別注意運算的**單位元素（Identity Element）**：
- **累加和（Sum of $1 \dots N$）**：Base Case 是 `if n == 0: return 0`（0 加任何數不變）。
- **累乘積（Factorial $N!$）**：Base Case 絕對**不能 return 0**！若誤寫 `return 0`，因為 $X \times 0 = 0$，整串乘積在回程時會全盤被歸零，最終結果變成 0！累乘基底必須是 **1**！

#### 4. APCS 實戰視野
APCS 觀念題中，追蹤數列遞迴公式（如 $f(n) = n + f(n-1)$ 或帶係數的遞推）是每年必考的核心考點。掌握階梯推導與倒帶回溯，能讓你在考場上迅速秒殺這類題目。

````python
# 範例 11.6.4：階乘累乘與數列累加的完整遞迴實作

def factorial(n):
    # Base Case: 1! 或 0! 為 1
    if n <= 1:
        print(f"  [觸發基底] fact({n}) = 1 (準備開始回溯計算)")
        return 1
        
    print(f"  [深入委託] fact({n}) 等待 {n} * fact({n-1})...")
    sub_result = factorial(n - 1)
    
    current_result = n * sub_result
    print(f"  [回溯結算] fact({n}) = {n} * {sub_result} = {current_result}")
    return current_result

def sum_1_to_n(n):
    # Base Case: 加到 0 時總和為 0
    if n <= 0:
        return 0
    # Recursive Case
    return n + sum_1_to_n(n - 1)

print("=== 階乘計算 fact(4) 追蹤 ===")
ans_fact = factorial(4)
print(f"最終計算結果 4! = {ans_fact}\n")

print("=== 數列累加計算 sum(1..5) ===")
ans_sum = sum_1_to_n(5)
print(f"最終計算結果 1+2+3+4+5 = {ans_sum}")
````

### 11.6.5 字串與串列的線性遞迴：反轉字串與倒序輸出

#### 1. 生活故事比喻：拆解一串字母珍珠項鍊
想像你手上有一串串著英文字母的珍珠項鍊，上面刻著 `"cat"`。你想要用遞迴的方式把它反轉成 `"tac"`。
最直覺的手法是：
你把項鍊上的第一顆珍珠 `'c'` 輕輕解下來放在桌子最右邊；
接著把剩下的項鍊 `"at"` 交給身邊的小夥伴，對他說：「幫我把這條剩下的短項鍊反轉過來！」
小夥伴把短項鍊反轉成了 `"ta"` 還給你；
最後，你把剛才拔下來的第一顆珍珠 `'c'` 接在 `"ta"` 的最後面，拼在一起就成了 `"tac"`！
大問題「反轉長度為 3 的字串」，成功被化解為「反轉長度為 2 的子字串，再加上一個字元」！

#### 2. 底層運作機制：序列切片的遞迴結構
在 Python 中，字串與串列都具備強大的切片功能（Slicing），天生非常適合用遞迴處理：
- **字串反轉函數 `rev_str(s)`**：
  - **Base Case**：當字串長度小於等於 1 時（空字串 `""` 或單字元 `"a"`），反轉後的結果就是它自己，直接 `return s`！
  - **Recursive Case**：`return rev_str(s[1:]) + s[0]`  
    （剩下的部分反轉後，把第一個字元黏在尾端！）
- **串列加總函數 `rec_list_sum(arr)`**：
  - **Base Case**：當串列為空 `not arr` 時，總和為 0，`return 0`。
  - **Recursive Case**：`return arr[0] + rec_list_sum(arr[1:])`  
    （取出第一個元素，加上剩下子串列的總和！）

#### 3. 初學者常見陷阱：切片索引越界防護
字串切片 `s[1:]` 在 Python 中非常安全，就算長度只有 1，`s[1:]` 也會安全產生空字串 `""`；
但是，索引存取 `s[0]` 若用在空字串上，會立刻引發致命的 `IndexError: string index out of range`！
因此，在寫序列遞迴時，**Base Case 必須先檢查長度 `len(s) <= 1`**，確保進入 Recursive Case 時 `s` 絕對至少有 1 個字元，才能安全取用 `s[0]`。

#### 4. APCS 實戰視野
將遞迴從「純數字計算」延伸到「字串與串列結構」，是跨向高階資料結構（如鏈結串列 Linked List、二元樹遍歷）的關鍵一步。理解這種由前向後剝皮、再由後向前組裝的模式，能極大拓寬你的演算法思維。

````python
# 範例 11.6.5：字串反轉與迴文檢測的遞迴實現

def reverse_string(s):
    # Base Case: 長度為 0 或 1 時，反轉即為自身
    if len(s) <= 1:
        return s
    
    # Recursive Case: 剩下子字串反轉 + 第一個字元
    return reverse_string(s[1:]) + s[0]

def is_palindrome_rec(s):
    # 遞迴判斷迴文（對稱字串）
    # Base Case: 空字串或單字元必定對稱
    if len(s) <= 1:
        return True
    # 頭尾字元若不同，立刻判定非迴文
    if s[0] != s[-1]:
        return False
    # 頭尾相同，繼續檢查去除頭尾後的子字串
    return is_palindrome_rec(s[1:-1])

test_word = "python"
rev_word = reverse_string(test_word)
print(f"原字串: {test_word} -> 遞迴反轉: {rev_word}")

palindrome_test = "racecar"
print(f"'{palindrome_test}' 是否為迴文? {is_palindrome_rec(palindrome_test)}")
````

### 11.6.6 遞迴深度限制與防爆：`RecursionError: maximum recursion depth exceeded`

#### 1. 生活故事比喻：深海潛水艇的深度防護閥
一艘研究用深海潛水艇下潛探索馬里亞納海溝。每下潛 10 公尺，艇身外殼承受的水壓就劇增一分（每深入一層遞迴，Call Stack 就多消耗一塊記憶體）。
如果這艘潛水艇的鋼板極限是 1000 公尺，當儀表板顯示下潛深度達到 1000 公尺時，警報器會瘋狂發出尖銳鳴響，自動防爆閥門會強行切斷下潛動力，強制啟動上浮程序！
在 Python 直譯器中，同樣安裝了這樣一個不可侵犯的**防爆安全閥**——**遞迴深度上限（Recursion Limit）**！
如果你的程式碼不小心寫成了無窮遞迴，或者下潛深度過深，Python 不會任由它耗盡記憶體導致電腦作業系統崩潰，而會主動拔除插頭，擲出響亮的警報：`RecursionError: maximum recursion depth exceeded`！

#### 2. 底層運作機制：遞迴深度限制與 `sys` 模組
- **查詢預設上限**：
  透過標準函式庫 `sys.getrecursionlimit()`，可以查詢到 Python 預設的遞迴深度上限（通常為 **1000** 層左右）。
- **為什麼會有 1000 層的限制？**  
  因為每一次函數呼叫，直譯器都必須在 C 語言層級的堆疊（C Call Stack）上分配空間。如果沒有這個保險絲，過深的遞迴會引發 C 語言堆疊溢位（Stack Overflow），直接導致整個 Python 直譯器程序瞬間閃退當機（Segmentation Fault），甚至連拋出錯誤訊息的機會都沒有！

#### 3. 初學者常見陷阱與 APCS 考場的 Runtime Error (RE)
在 APCS 競賽中，這是初學選手最常踩中的慘烈陷阱：
- 題目測資規模給定 $N = 10^5$（10 萬）。
- 同學寫了一個線性遞迴去走訪這 10 萬筆資料。在自己電腦上測小範例 $N = 10$ 完全正確；
- 一送上 APCS 評判系統，遇到大測資時，程式在呼叫到第 1000 層時當場引爆 `RecursionError`，系統判讀為 **Runtime Error (RE)**，整組子任務 0 分！

#### 4. APCS 實戰視野：遞迴與迴圈的時空權衡（Trade-off）
面對線性推進的問題，請銘記競賽設計原則：
- **線性問題（如單純計數、陣列走訪、等差求和）**：一律優先使用 **`for` 迴圈或 `while` 迴圈**！迴圈的時間複雜度是 $O(N)$，空間複雜度是極致省電的 $O(1)$，即使 $N = 10^7$ 也絕不可能爆棧！
- **樹狀結構、分治演算法（如二分搜尋、歸併排序、樹遍歷）**：因為每一步問題規模折半，樹的深度只有 $\log_2 N$（例如 $N = 10^6$ 時深度僅約 20 層），遠低於 1000，這才是遞迴大顯身手的最佳舞台！

````python
# 範例 11.6.6：查詢遞迴上限與捕捉 RecursionError 防爆示範

import sys

# 1. 查詢 Python 預設遞迴上限
current_limit = sys.getrecursionlimit()
print(f"[系統資訊] 目前 Python 預設最大遞迴深度: {current_limit} 層\n")

# 2. 設計一個故意過度下潛的函數
def dive_too_deep(n):
    return dive_too_deep(n + 1)

print("=== 模擬超深遞迴觸發防爆安全閥 ===")
try:
    dive_too_deep(1)
except RecursionError as e:
    print(f"🛑 成功攔截異常: {e}")
    print("說明：Python 自動切斷了無窮遞迴，保護電腦記憶體未被耗盡！")
````

## 11.6 單元重點回顧與自我檢驗

恭喜你跨過了電腦科學中最具代表性的壯麗門檻——**遞迴（Recursion）**！你已經從單純的循序迴圈思維，躍升為能夠在腦海中模擬「函數呼叫分身、呼叫堆疊壓入與彈出」的抽象演算法思維。

### 核心觀念複習清單
1. **遞迴的核心意象**：
   - 「函數呼叫自己」如同鏡中鏡與俄羅斯套娃，大問題肚子裡裝著規模縮小的同結構小問題。
   - 每次呼叫都是直譯器在記憶體中建立一個全新的獨立函數副本，絕非原代碼重跑。
2. **呼叫堆疊（Call Stack）物理模型**：
   - 後進先出（LIFO）的托盤架機制。每層呼叫將 Stack Frame 壓入（Push）；遇到 return 彈出（Pop）。
   - 呼叫點前方的代碼在「去程」順序執行；呼叫點後方的代碼在「回程」倒序執行。
3. **終止條件（Base Case）煞車皮**：
   - 遞迴能安全停下來的唯一基石。嚴格遵守「先寫 Base Case，再寫 Recursive Case」黃金法則。
   - Base Case 務必落實 `return`，並使用 `<=` 不等式構建防禦性邊界。
4. **數學歸納與經典階乘**：
   - 累乘 Base Case 必須回傳乘法單位元素 1；累加 Base Case 回傳 0。
   - 掌握公文層層下發與答案層層倒帶回報的計算路徑。
5. **序列資料的線性遞迴**：
   - 字串切片 `s[1:]` 與串列 `arr[1:]` 每次削去一個元素，搭配 Base Case 形成精煉的序列逆序與處理模型。
6. **`RecursionError` 與時空權衡**：
   - Python 預設約 1000 層的遞迴深度防爆閥，防止記憶體溢位崩潰。
   - APCS 競賽鐵律：線性單純問題優先用迴圈（空間 $O(1)$，絕不爆棧）；樹狀與分治問題才用遞迴！
