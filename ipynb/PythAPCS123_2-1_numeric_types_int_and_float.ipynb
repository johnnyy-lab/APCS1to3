### 🧱 2-1-1 數字世界的兩大宗族——整數積木（int）與浮點果汁（float）

親愛的初學同學們，歡迎來到第二章的數字世界！在第一章裡，我們學會了如何拿貼有名牌的神奇置物箱來裝資料。但是，你有沒有想過：箱子裡裝進去的「數字」，其實長相與性格有著天壤之別呢？

在 Python 的奇幻王國裡，所有資料都有它的「**資料型態（Data Type）**」。今天我們要認識數字界勢力最龐大的兩大家族：**整數（Integer，簡稱 int）** 與 **浮點數（Floating Point，簡稱 float）**！

你可以把這兩大家族想像成兩種截然不同的物品：
1. **整數（int）就像「硬邦邦的樂高積木」**：例如 `5`, `0`, `-12`, `100`。一顆就是一顆，結結實實、完完整整，絕對沒有切半或碎片！
2. **浮點數（float）就像「量杯裡的果汁」**：例如 `3.14`, `0.5`, `-2.8`。它身上永遠帶著一顆圓圓的小數點 `.`，可以精細地量出小數點後面的刻度！

這裡隱藏著一個所有初學者必須注意的關鍵小秘密：**只要看見小數點 `.`，就算小數點後面只有一個 0，它也百分之百是浮點數！**
例如數字 `4` 是正統的整數積木，但寫成 `4.0` 就瞬間變成了浮點水果汁！分清楚這兩種家族的長相，是我們寫出正確算式的第一步喔！

```python
# [2] Code 範例區：展示整數與浮點數的外觀差異
a = 4      # 這是乾淨俐落的整數 int
b = 4.0    # 只要帶有小數點，就是浮點數 float

print(a)   # 印出純整數 4
print(b)   # 印出帶有小數點的浮點數 4.0
```

### 🔬 2-1-2 查驗身分的照妖鏡——認識內建型態檢查工具 type()

既然整數和浮點數長得有點像，那如果我們面對一個神秘的變數箱子，不知道裡面裝的到底是積木還是果汁，該怎麼辦呢？

這時候，Python 給了我們一面無比靈驗的神器——「**照妖鏡函數 `type()`**」！
你可以把 `type()` 想像成機場海關的「**超強 X 光掃描儀**」！
只要你把任何變數箱子送進 `type(變數)` 的傳送帶裡，它就會立刻透視箱子底部的資料結構，並把它的真實身份標籤印給你看：
* 如果裡面裝的是整數，照妖鏡會顯示 `<class 'int'>`（代表它是 Integer 整數家族的成員）。
* 如果裡面裝的是浮點數，照妖鏡會顯示 `<class 'float'>`（代表它是 Floating-point 浮點數家族的成員）。

例如執行 `print(type(5))`，螢幕就會吐出 `<class 'int'>`；執行 `print(type(5.0))`，螢幕就會吐出 `<class 'float'>`！
在日後寫大型程式或 APCS 除錯時，每當懷疑變數型態有鬼，隨時拿出 `type()` 照一照，真相立刻大白！

```python
# [2] Code 範例區：使用 type() 照妖鏡檢驗變數的身分型態
x = 100
y = 3.14

print(type(x))  # 印出 <class 'int'>，證實 x 是整數
print(type(y))  # 印出 <class 'float'>，證實 y 是浮點數
```

### 🎨 2-1-3 染紅效應——整數與浮點數混合運算自動晉升為浮點數

現在我們要來揭曉一個神奇的「化學反應」：如果把硬邦邦的整數積木，跟帶有小數點的浮點水果汁放在同一個算式裡加減乘除，會發生什麼事？

吳邦一老師在講義第 8 頁特別指出：
「**浮點數與整數運算後會變成浮點數！**」
例如：
* `3.2 + 4` 會得到 `7.2`
* `4 * 1.0` 會得到 `4.0`

你可以把這個現象想像成「**一滴紅墨水染紅整盆清水**」！
在 Python 運算規則中，只要算式中有哪怕「任何一個」微小的浮點數存在，電腦為了保留小數點後面珍貴的精確刻度，就會啟動自動升級機制，把整個算式的所有整數全部融化改造成浮點數！

所以：
* `5 + 2` 是兩個整數相加，結果是整數 `7`。
* 但 `5 + 2.0`，因為出現了 `2.0` 這滴小數點紅墨水，結果就會被整盆染紅，變成浮點數 `7.0`！
* 同樣地，即使是相乘：`4 * 1.0`，雖然數值大小還是 4，但因為乘上了 `1.0`，結果就永遠變成了浮點數 `4.0`。

記住這個「一滴墨水染紅整盆」的特性，才不會在不知不覺中把整數變成了浮點數喔！

```python
# [2] Code 範例區：展示整數與浮點數混合運算自動晉升為 float
result1 = 3.2 + 4    # 浮點數 + 整數
print(result1)       # 印出 7.2

result2 = 4 * 1.0    # 整數 * 浮點數
print(result2)       # 印出 4.0（已晉升為浮點數）
print(type(result2)) # 印出 <class 'float'>
```

### ⚠️ 2-1-4 APCS 考場的致命失分陷阱——4.0 與 4 大不相同（Wrong Answer 的眼淚）

在學完數值型態後，我們要來面對全台灣國高中生在 APCS 考場上最痛心疾首的「**血淚教訓**」！

吳邦一老師在講義第 8 頁用極其嚴肅的語氣警告大家：
「**在自動裁判的考試與比賽中，答案是 4.0 與 4 會是不一樣的！如果答案的格式要求整數，那麼輸出 4.0 很可能就是錯的，所以必須分清楚整數或是浮點數！**」

為什麼會這樣呢？
在 APCS 或 ZeroJudge 的線上自動評判系統（Online Judge，簡稱 OJ）中，坐在電腦前批改考卷的不是溫柔的人類老師，而是一台**冷酷無情、逐字逐句精確比對字元的評判機器人**！

想像一下評判機器人的眼光：
* 當題目要求你輸出整數答案 `4` 時，機器人期待看到的文字字元只有一個：`'4'`（長度為 1）。
* 如果你的程式因為運算不小心產生了浮點數，印出了 `4.0`，機器人看到的是三個字元：`'4'`, `'.'`, `'0'`（長度為 3）！

雖然在人類數學課上看來 $4 = 4.0$ 毫無疑問，但裁判機器人只看字元格式！多了一個小數點與零，機器人就會立刻亮出紅牌，毫不留情地宣判：**WA（Wrong Answer，答案錯誤，0分！）**

這是在 APCS 考場上，無數考生明明算法正確、邏輯通順，卻莫名其妙痛失分數的最大隱形殺手！
因此，解題時一定要瞪大雙眼看清楚題目要求：**題目要整數，就絕對不能輸出帶有小數點的浮點數！**

```python
# [2] Code 範例區：展示文字比對下 4 與 4.0 的殘酷差異
ans_int = 4
ans_float = 4.0

# 在人類看來數值相等，但在裁判系統印出字串時完全不同：
print(ans_int)    # 輸出：4（題目要求的整數格式，AC 通過！）
print(ans_float)  # 輸出：4.0（若題目要求整數，這會被判定為 WA 零分！）
```
