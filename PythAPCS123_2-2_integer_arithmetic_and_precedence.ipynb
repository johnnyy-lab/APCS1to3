### ➕ 2-2-1 整數加法與減法（+、-）——數值增減與負數運算

國一的同學們，大家在數學課都學過正負數的加減法對吧？在 Python 的魔法世界裡，加法符號就是我們最熟悉的加號 `+`，減法符號就是減號 `-`！

你可以把加減法想像成「**電玩冒險遊戲中的血量增減**」：
1. **加法（`+`）就像喝下補血藥水**：例如玩家原本有 100 點生命值，喝了紅藥水增加 30 點，寫成 `hp = 100 + 30`，生命值就變成了 130 點！
2. **減法（`-`）就像受到怪物攻擊扣血**：如果被史萊姆撞了一下扣 25 點，寫成 `hp = 130 - 25`，生命值就剩下 105 點。

除了拿來做兩數之間的加減運算之外，減號 `-` 還可以直接放在數字前面代表「**負數（Negative Numbers）**」，例如高山上的氣溫是 `-5` 度。而且 Python 完全懂得數學上的「**負負得正**」法則喔！
如果你寫 `15 - (-5)`，意思是從 15 度減去負 5 度，Python 會精準算出溫差為 `20`！

**⚠️ 初學者常見錯誤陷阱**：
在數學課本上，我們有時會看到 `--5` 這樣的簡寫；但在 Python 中，千萬不要把兩個減號黏在一起寫得模稜兩可。為了讓程式碼清楚好讀，當你要減去一個負數時，**強烈建議在負數外面加上小括號**（例如寫成 `a - (-b)`），這樣不僅直譯器看得懂，你自己回頭檢查時也不會眼花看錯！

**🎯 APCS 考場實務提示**：
在 APCS 題目中，整數的加法與減法運算後的結果「**保證仍然是純整數（int）**」，永遠不會平白無故冒出小數點！請大膽安心使用。

```python
# [2] Code 範例區：展示整數加法、減法與負負得正運算
hp = 100
potion = 30
damage = 25

# 加法：喝藥水補血
hp = hp + potion
print(hp)        # 印出 130

# 減法：受到攻擊扣血
hp = hp - damage
print(hp)        # 印出 105

# 負數運算：計算平地 15 度與玉山山頂 -5 度的溫差（負負得正）
temp_diff = 15 - (-5)
print(temp_diff) # 印出 20
```

### ✖️ 2-2-2 整數乘法（*）與代數陷阱——星號連乘與嚴禁省略乘號

接下來我們要來認識能讓數字迅速爆發翻倍的利器——「**乘法（Multiplication）**」！

在國小數學課本中，我們習慣寫乘號「$\times$」；但是在標準電腦鍵盤上，根本沒有這個符號！
因此在 Python（以及全世界所有的程式語言）中，我們一律使用「**米字星號 `*`**」（鍵盤上的 Shift + 數字鍵 8）來代表乘法運算子！例如買 5 本每本 40 元的筆記本，就要寫成 `40 * 5`。

**💥 國中生初學 APCS 最容易踩到的「超級代數陷阱」**：
國一的大家在數學課剛學會代數符號，常常會習慣性地把乘號「**偷偷省略掉**」，例如：
* 數學課寫 $2a$ 代表 $2 \times a$
* 數學課寫 $3(x + y)$ 代表 $3 \times (x + y)$
* 數學課寫 $ab$ 代表 $a \times b$

**請注意！這在 Python 中是 100% 絕對禁止的違法行為！**
1. 如果你在 Python 寫 `2a`：直譯器會以為你想命名一個以數字開頭的非法變數，立刻無情噴出 `SyntaxError: invalid decimal literal`（無效的十進位數字語法錯誤）！
2. 如果你在 Python 寫 `3(x + y)`：因為小括號接在數字後面在 Python 代表「呼叫函數」，直譯器會誤以為你有個叫做 `3` 的函數，立刻當場大崩潰噴出 `TypeError: 'int' object is not callable`！
3. 如果你在 Python 寫 `ab`：直譯器會以為這是一個叫做 `ab` 的全新變數，而不是 `a` 乘以 `b`！

**🎯 鐵則口訣：星號一個都不能少！**
在寫 Python 程式時，所有的相乘關係，必須乖乖、老老實實打出星號 `*`：
`2 * a`、`3 * (x + y)`、`a * b`。牢記這個好習慣，保證你在 APCS 考場上少碰壁十次！

```python
# [2] Code 範例區：展示乘法運算子 * 的正確用法與嚴禁省略乘號
unit_price = 45
count = 6

# 正確使用星號 * 進行乘法計算
total = unit_price * count
print(total)  # 印出 270

# 代數運算示範：計算 2*a + 3*b（絕不能寫 2a + 3b）
a = 5
b = 8
ans = 2 * a + 3 * b
print(ans)    # 2*5 + 3*8 = 10 + 24 = 34
```

### 🚦 2-2-3 運算子優先級鐵則——先乘後加減與由左至右結合性

當一個算式中同時出現了加法、減法與乘法時，電腦會先算哪一個呢？
這時候就必須請出電腦王國的交通號誌法規——「**運算子優先級（Operator Precedence）**」！

Python 的運算規則和國中數學課完全一模一樣，遵守著最核心的鐵律：
👉 **「先乘除，後加減！」**

你可以把這想像成鐵路軌道上的路權等級：
* **乘法（`*`）是最高等級的「特快高鐵列車」**，擁有優先通行的特權！
* **加減法（`+`、`-`）是速度較慢的「普通平快車」**，必須禮讓高鐵先行。

例如在算式 `2 + 3 * 4` 中：
特快車 `3 * 4` 會先呼嘯而過算出 `12`，接著平快車才進站執行 `2 + 12`，最終答案是 `14`！
如果誤以為電腦是從左到右呆呆地算，算成 `(2 + 3) * 4 = 20`，那可就差了十萬八千里囉！

**那麼，如果算式裡遇到「同等級」的運算子，誰先算？**
當加法與減法排在一起（例如 `10 - 4 + 2`）時，它們的路權完全相同。此時 Python 會遵守「**由左至右（Left-to-Right Associativity，左結合性）**」的排隊規則：
1. 先算最左邊的 `10 - 4 = 6`
2. 再算後面的 `6 + 2 = 8`
千萬不要自己腦補把加法先算成 `10 - (4 + 2) = 4`，這樣會直接掉入邏輯運算的致命大坑！

**🎯 APCS 解題叮嚀**：
在 APCS 第一大題的程式閱讀觀念題中，教授非常喜歡出長算式來考驗考生的耐心。只要冷靜拿起筆，先把所有的 `*` 乘法圈起來先算，剩下的同級加減由左往右推進，就能穩穩拿到分數！

```python
# [2] Code 範例區：展示先乘後加減與由左至右的結合順序
# 範例 1：先乘後加
ans1 = 2 + 3 * 4
print(ans1)  # 2 + 12 = 14（絕對不是 20）

# 範例 2：兩組乘法先行，再進行中間加法
ans2 = 3 * 4 + 5 * 2
print(ans2)  # 12 + 10 = 22

# 範例 3：同級加減由左向右依序計算
ans3 = 10 - 4 + 2
print(ans3)  # 先算 10-4=6，再算 6+2=8（絕對不是 4）
```

### 👑 2-2-4 運算順序的王牌指揮官——小括號 (...) 搶先執行權

學完了「先乘後加減」，你心裡可能會有個大疑問：
「老師！如果我的算式真的想要『先做加法，再做乘法』，難道電腦就不准嗎？」

當然准！這時候我們就要隆重介紹掌控運算順序的最高統治者——「**小括號 `(...)`**」！

你可以把小括號想像成主題樂園裡的「**VIP 快速通關手環**」！
不論算式裡是地位多麼卑微的加法還是減法，**只要被戴上了小括號手環，它的優先級瞬間飆升至全宇宙最高等級**！
就連橫行霸道的乘法特快車，看到戴著小括號手環的算式，也必須恭恭敬敬地在旁邊立正站好，等括號裡面的算式全部算完之後，才能輪到乘法！

讓我們來看看加了小括號後的驚人威力對比：
* 沒有小括號：`2 + 3 * 4` $\rightarrow$ 乘法先算 $\rightarrow$ `2 + 12 = 14`
* 戴上小括號：`(2 + 3) * 4` $\rightarrow$ 括號加法搶先 $\rightarrow$ `5 * 4 = 20`

**🚨 APCS 經典失分現場：梯形面積公式大災難**
大家還記得國小學過的梯形面積公式嗎？「**（上底 ＋ 下底）$\times$ 高 $\div$ 2**」。
在歷年 APCS 實作測驗中，無數初學者往往一緊張就寫成：
`area = top + bottom * height // 2`
結果因為乘除法先算，電腦只把下底拿去乘以高，再加上上底，答案徹底面目全非，立刻收到慘烈的 WA（Wrong Answer，答案錯誤）！
正確寫法一定要用小括號保護上底加下底：
`area = (top + bottom) * height // 2`

**⚠️ 語法報錯警訊**：
小括號一定要「**成對出現（有開必有關）**」！如果你只寫了左括號卻忘了補右括號（例如 `x = (5 + 3 * 2`），直譯器會在行尾當場拋出 `SyntaxError: '(' was never closed`，提醒你括號沒有關閉喔！

```python
# [2] Code 範例區：展示小括號 () 改變運算順序的強大威力
a = 2
b = 3
c = 4

# 沒有小括號：先乘後加
ans_no_paren = a + b * c
print(ans_no_paren)  # 2 + 12 = 14

# 戴上小括號手環：括號內優先搶先計算
ans_with_paren = (a + b) * c
print(ans_with_paren) # 5 * 4 = 20

# 梯形面積加倍值範例：(上底 + 下底) * 高
top = 4
bottom = 6
height = 5
double_area = (top + bottom) * height
print(double_area)    # (4 + 6) * 5 = 10 * 5 = 50
```

### 🪆 2-2-5 多重巢狀小括號與括號大忌——由內而外層層拆解

在國一數學課遇到很長、很複雜的四則運算式時，課本上常常會看到三種不同形狀的括號依序包在一起：
「大括號 $\{$ $\}$ 包著 中括號 $[$ $]$，中括號再包著 小括號 $($ $)$」。

這時候，所有剛剛轉到 Python 寫程式的同學，最常犯的重大錯誤就是把數學課本的算式「照單全收」！

**🚫 程式設計第一大忌：Python 算式中嚴禁使用中括號 `[]` 或大括號 `{}`！**
請大家在心中默念三次這個超級禁忌！在 Python 語言裡：
1. **中括號 `[...]` 在 Python 中不是算術括號**：如果你在算式裡寫 `[2 + 3] * 4`，Python 不會算出 20，而是會產生非純數值的特殊資料結構！型態直接跑掉！
2. **大括號 `{...}` 也是專屬保留物**：它是第十章要學的「**集合（Set）**」或「**字典（Dict）**」專用符號！如果你寫 `{2 + 3} * 4`，Python 會直接氣到吐血報錯 `TypeError: unsupported operand type(s) for *: 'set' and 'int'`！

**💡 唯一解法：在 Python 算式中，無論包了多少層，通通只能用小括號 `()`！**
例如數學上的算式：
$2 \times [ 3 + (4 - 1) \times 5 ]$
在 Python 裡必須全面改寫為純小括號：
`2 * (3 + (4 - 1) * 5)`

**🪆 運作心法：像剝開俄羅斯套娃一樣，由內而外層層求值！**
當算式出現多重小括號時，電腦的解題流程就像拆解俄羅斯套娃：
1. 找到**最深處、最內層**的小套娃 `(4 - 1)` 先算，得到 `3`。
2. 算式簡化為 `2 * (3 + 3 * 5)`，接著處理外層括號內的乘法 `3 * 5 = 15` 與加法 `3 + 15 = 18`。
3. 最後拆掉所有括號，執行最外層的 `2 * 18 = 36`！

只要謹記「**只用小括號**」與「**由內而外**」兩大心法，再複雜的競賽算式也難不倒你！

```python
# [2] Code 範例區：展示多重小括號 ((...)) 由內而外的運算拆解
# 數學式：2 * [ 3 + (4 - 1) * 5 ]
# Python 規範：嚴禁中括號，一律使用小括號！
result = 2 * (3 + (4 - 1) * 5)
print(result)  # 2 * (3 + 3*5) = 2 * (3 + 15) = 2 * 18 = 36

# 雙層巢狀括號實例：先加總兩數、再乘上倍率、最後減掉扣打後再加倍
a, b, c, d, e = 2, 3, 4, 8, 3
complex_ans = ((a + b) * c - d) * e
print(complex_ans)  # ((2+3)*4 - 8)*3 = (5*4 - 8)*3 = (20 - 8)*3 = 12*3 = 36
```
