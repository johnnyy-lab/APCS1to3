### 🕰️ 6-5-1 雙重迴圈心智模型：時鐘的分針與時針——外層帶動內層的維度展開

**生活故事：機械鐘錶的齒輪與運動會班級進場**  
請初學同學抬頭看一看教室或客廳牆上的機械時鐘。  
當時針指著 1 點時，時針會靜靜停在那裡；與此同時，修長的分針必須在鐘面上「滴答滴答」辛勤地走滿整整 60 個刻度（從 0 分一路走到 59 分）。  
只有當分針完整跑滿一整圈，齒輪咬合轉動，時針才會「喀噠」一聲前進一步，指向 2 點！  
接著，時針又停在 2 點不動，分針再度從第 0 分鐘出發，老老實實地重新跑滿 60 個刻度！  
又例如學校運動會的開幕典禮：大會司令台廣播「第 1 班進場」（外層時針），第 1 班的 30 位同學（內層分針）依序跑步通過司令台向全場揮手；等這 30 位同學全員走過之後，司令台才廣播「第 2 班進場」，第 2 班的 30 位同學再從第 1 位到第 30 位完整跑過一次！

**運作機制——外層動一步，內層跑全套：**  
這就是「雙重迴圈（Nested Loops，又稱巢狀迴圈）」最核心的心智模型！  
所謂「巢狀」，在程式世界中就是「大盒子裡面套小盒子」，也就是將一個完整的 `for`（或 `while`）迴圈，直接塞進另一個迴圈的縮排區塊中。  
- **外層迴圈（Outer Loop）**：扮演**「時針」**的角色，主導宏觀的大進程（例如第幾點鐘、第幾樓層、第幾個班級）。
- **內層迴圈（Inner Loop）**：扮演**「分針」**的角色，主導微觀的細節（例如第幾分鐘、第幾間房、第幾號同學）。  
- **核心執行次數鐵則**：外層迴圈每推進 1 回合，內層迴圈就必須**「從頭到尾完整執行一整輪」**！如果外層要跑 $M$ 輪，內層每輪跑 $N$ 步，那麼縮排在最內層的程式碼總共會被扎扎實實執行 $M 	imes N$ 次！  
這就是將原本單純的一維數線，向外延伸出第二個維度，在二維平面上展開了無限可能！

**常見錯誤陷阱：**  
- **步伐速度認知混淆**：初學者最常犯的觀念誤區，是誤以為「外層走一步，內層也跟著走一步」。請牢牢記住：**外層走一步，內層跑全套！** 內層沒跑完，外層絕對不准動！
- **縮排層次混亂**：內層迴圈的開頭必須相對於外層向右縮排 4 個空白；而內層迴圈內部要執行的指令，則必須縮排 8 個空白（雙重縮排）。層級一旦錯位，程式邏輯就會差之千里。

````python
# [2] Code 範例區：時鐘心智模型——外層時針推進，內層分針跑滿一整圈
print("=== 迷你時鐘運轉模擬（1 點到 2 點，每 15 分鐘報時一次）===")

for hour in range(1, 3):  # 外層時針：走訪 1 點與 2 點（共 2 輪）
    print(f"--> 【時針推進】現在是大進程：{hour} 點整！")
    
    for minute in range(0, 60, 15):  # 內層分針：每 15 分鐘走一刻度（每輪跑 4 步）
        print(f"    分針走動中：{hour} 點 {minute:02d} 分")

print("=== 模擬結束！外層走 2 輪，內層每輪走 4 步，總共執行 2 * 4 = 8 次報時！ ===")
````

````python
# ==========================================
# [4] Code 練習題
# 任務說明：
# 保全人員巡邏大樓房間：
# 大樓共有 total_floors 層（樓層編號 1 到 total_floors）。
# 每層樓共有 rooms_per_floor 間房間（房間編號 1 到 rooms_per_floor）。
# 請使用雙重 for 迴圈走訪所有房間：
# 外層走訪樓層 floor，內層走訪房間 room，
# 每巡查一間房間，印出 "巡查房間：F{floor}-R{room}"，
# 並將巡查過的房間總數累加至 inspected_count。
#
# 【公開測試資料 1】
# 設定：total_floors = 2, rooms_per_floor = 3
# 預期：總巡查房間數 inspected_count = 6
#
# 【公開測試資料 2】
# 設定：total_floors = 3, rooms_per_floor = 4
# 預期：總巡查房間數 inspected_count = 12
# ==========================================
# 請在下方撰寫你的程式碼並執行測試：

total_floors = 2
rooms_per_floor = 3
inspected_count = 0

for floor in range(1, total_floors + 1):
    for room in range(1, rooms_per_floor + 1):
        print(f"巡查房間：F{floor}-R{room}")
        inspected_count += 1

print(f"巡查完畢，總房間數：{inspected_count}")

# 公開測資驗證：
assert inspected_count == 6, "公開測資 1 驗證失敗"
print("公開測資 1 通過！")
````

````python
# ==========================================
# [5] Code 挑戰題
# 任務說明：
# 籃球投籃自主訓練：
# 選手進行 num_sets = 3 組投籃訓練。
# 每一組固定投 shots_per_set = 5 球。
# 訓練規則：
# 外層走訪組數 set_idx（1 到 num_sets），
# 內層走訪球數 shot_idx（1 到 shots_per_set）。
# 若（組數 + 球數）為偶數（即 (set_idx + shot_idx) % 2 == 0），
# 代表該球順利投進，得分 score 累加 2 分，進球數 hits 累加 1；
# 否則代表未投進，不得分。
# 請計算 3 組結束後的總投球數 total_shots、進球數 hits 與總得分 score。
# （本題為自由挑戰題，無公開測資，請依題目情境自行思考並撰寫完整程式碼）
# ==========================================
# 請在下方撰寫你的程式碼：

num_sets = 3
shots_per_set = 5
total_shots = 0
hits = 0
score = 0

for set_idx in range(1, num_sets + 1):
    for shot_idx in range(1, shots_per_set + 1):
        total_shots += 1
        if (set_idx + shot_idx) % 2 == 0:
            hits += 1
            score += 2

print(f"訓練結算：總投球 {total_shots} 球，進球 {hits} 球，總得分 {score} 分")
assert total_shots == 15 and hits == 8 and score == 16, "挑戰題計算錯誤！"
print("挑戰題 6.5.1 驗證通過！")
````

### 🏷️ 6-5-2 迴圈計數器命名鐵律：外層 r 內層 c（或 i, j），嚴禁同名覆蓋（Shadowing）

**生活故事：同名同姓的班長與混亂的點名冊**  
如果一個班級裡，班長叫「陳小明」，而排在隊伍裡的另一位同學也叫「陳小明」。  
班導師站在台上說：「陳小明，你去把陳小明叫過來！」這時全班必定面面相覷、陷入大混亂。  
更糟糕的是，如果外層指派大明去拿點名簿，大明走到一半，突然旁邊殺出另一個小明硬是搶走了「小明」這個名牌，等搶完之後，原本大明到底點到第幾個人了？記憶完全被抹除摧毀！  
在程式中，這就是極其危險且令人崩潰的「同名遮蔽（Variable Shadowing）」！

**運作機制——嚴禁同名覆蓋的鐵律：**  
在 Python 的雙重迴圈中，**外層迴圈與內層迴圈「絕對、絕對不能使用同一個變數名稱」**！  
看看這個反面教材慘劇：
```python
# 致命錯誤示範：外層與內層都搶著用 i！
for i in range(3):       # 外層想跑 0, 1, 2
    for i in range(3):   # 內層竟然也叫 i！
        print(i)
```
這段代碼執行時會發生什麼事？  
1. 當外層啟動第 0 輪，`i = 0`；
2. 程式進入內層，內層又宣告了 `for i in range(3)`，把變數箱子 `i` 強行搶走，並依序改成了 0, 1, 2；
3. 當內層跑完時，變數 `i` 裡面的值停留在 2；
4. 此時回到外層迴圈，外層直譯器以為原本的 `i` 已經跑完了，整個迴圈提早夭折猝死！原本預期執行 $3 	imes 3 = 9$ 次，最後卻只跑了 3 次！

**命名黃金準則：**  
為了讓程式碼清晰好讀且絕不踩腳，請各位學習者銘記兩套業界通用的黃金命名規範：
1. **網格 / 矩陣 / 棋盤情境（強烈推薦）**：
   - 外層變數命名為 **`r`**（代表 **Row，橫列 / 行**）；
   - 內層變數命名為 **`c`**（代表 **Column，直欄 / 列**）；
   - 任何位置的座標一律表示為 `(r, c)`，橫縱維度一眼看穿，永遠不會混淆！
2. **純數學 / 抽象枚舉情境**：
   - 外層使用 **`i`**，內層使用 **`j`**；如果有第三層，則使用 **`k`**。

**APCS 考試實務建議：**  
APCS 歷屆實作題中，有大量題目圍繞在二維地圖與矩陣走訪。考場緊張時，手滑把內層變數誤打成外層名稱是常見的失分陷阱。一律採用 `r` 與 `c` 命名，能讓你從大腦底層直接杜絕同名遮蔽，寫出穩如泰山的程式碼！

````python
# [2] Code 範例區：正確命名 r 與 c，清楚走訪二維座標網格
print("=== 走訪 2 列 3 欄的二維座標網格 (r, c) ===")

total_points = 0

# 外層 r 代表橫列（Row 0 到 Row 1）
for r in range(2):
    # 內層 c 代表直欄（Column 0 到 Column 2）
    for c in range(3):
        print(f"當前座標點：(r={r}, c={c})")
        total_points += 1

print(f"=== 網格走訪完畢，總共拜訪了 {total_points} 個獨立座標點！ ===")
````

````python
# ==========================================
# [3] Code 填空題
# 任務說明：
# 某位同學原本寫了一段程式想要枚舉 1 到 3 的雙變數配對 (i, j)，
# 但他不小心把內層迴圈的變數名稱也寫成了 i！
# 請將內層迴圈修正為使用獨立變數 j，
# 並補齊配對印出指令，確保印出 3 * 3 = 9 組獨立配對！
# ==========================================

pair_count = 0

for i in range(1, 4):
    # 請將內層計數器正確命名為 j（避免與外層 i 衝突）
    for ___ in range(1, 4):
        print(f"配對組合：({i}, {___})")
        pair_count += 1

print(f"總配對組數：{pair_count}（必須剛好為 9 組）")
````

````python
# ==========================================
# [4] Code 練習題
# 任務說明：
# 二維座標乘積篩選：
# 給定網格列數 max_r 與欄數 max_c。
# 請使用雙重迴圈走訪所有座標點 (r, c)，其中：
# r 的範圍為 1 到 max_r，c 的範圍為 1 到 max_c。
# 檢查座標乘積 product = r * c：
# 若 product >= threshold，則將合格格子計數器 qualified_cells 累加 1。
#
# 【公開測試資料 1】
# 設定：max_r = 3, max_c = 4, threshold = 6
# 合格乘積點：(2,3)=6, (2,4)=8, (3,2)=6, (3,3)=9, (3,4)=12
# 預期：qualified_cells = 5
#
# 【公開測試資料 2】
# 設定：max_r = 2, max_c = 3, threshold = 4
# 合格乘積點：(2,2)=4, (2,3)=6
# 預期：qualified_cells = 2
# ==========================================
# 請在下方撰寫你的程式碼並執行測試：

max_r = 3
max_c = 4
threshold = 6
qualified_cells = 0

for r in range(1, max_r + 1):
    for c in range(1, max_c + 1):
        product = r * c
        if product >= threshold:
            qualified_cells += 1

print("合格格子總數：", qualified_cells)

# 公開測資驗證：
assert qualified_cells == 5, "公開測資 1 驗證失敗"
print("公開測資 1 通過！")
````

````python
# ==========================================
# [5] Code 挑戰題
# 任務說明：
# 棋盤黑白染色統計：
# 一個 R 列 C 欄的西洋棋盤，座標以 (r, c) 表示（r 從 0 到 R-1，c 從 0 到 C-1）。
# 國際象棋的著色規則為：
# 若 (r + c) % 2 == 0，該格為「白格（White）」；
# 否則為「黑格（Black）」。
# 給定 R = 4, C = 5，請撰寫雙重迴圈（外層 r，內層 c），
# 分別統計整個棋盤中「白格總數 white_count」與「黑格總數 black_count」，
# 並驗證 white_count + black_count 是否等於總格數 R * C。
# （本題為自由挑戰題，無公開測資，請依題目情境自行思考並撰寫完整程式碼）
# ==========================================
# 請在下方撰寫你的程式碼：

R = 4
C = 5
white_count = 0
black_count = 0

for r in range(R):
    for c in range(C):
        if (r + c) % 2 == 0:
            white_count += 1
        else:
            black_count += 1

print(f"棋盤統計：白格 {white_count} 格，黑格 {black_count} 格，總計 {white_count + black_count} 格")
assert white_count == 10 and black_count == 10 and (white_count + black_count == 20), "挑戰題計算錯誤！"
print("挑戰題 6.5.2 驗證通過！")
````

### 🖨️ 6-5-3 精密排版輸出基本功：end=' ' 水平間隔與 print() 換行節奏掌控

**生活故事：老式機械打字機的擊字與「叮～換行拉桿」**  
在電腦螢幕還沒發明的年代，人們使用機械打字機寫信。  
每當打字員按下一顆字母鍵，打字機的金屬字槌就會重重敲在紙上，接著字車自動「向右滑動一格」，讓下一個字母緊鄰在右側水平排開；  
當打字員打完一整行文字、來到紙張最右邊時，打字機會發出清脆的「叮～」提示聲。這時，打字員必須親手推動「換行拉桿（Carriage Return）」，字車才會「喀噠」一聲滑回最左側，同時滾筒將白紙向上捲動一行，準備開始打下一行！

**運作機制——水平鋪開與垂直換行的黃金三步拍檔：**  
在 Python 中，`print()` 函數預設在印完資料後會自動附帶換行符號 `
`。  
如果我們想在螢幕上印出一個像方陣一樣漂亮的二維表格，就必須完美重現打字機的節奏：
1. **內層水平鋪開（字車右移）**：  
   在內層迴圈中，我們使用 `print(數值, end=' ')`。透過將結尾參數指定為單一空格 `end=' '`，我們強制取消了自動換行，讓同一列的所有數字水平向右緊密排開！
2. **內層結束後精準換行（推動換行拉桿）**：  
   當內層迴圈的所有欄位全部印完時，此時代表這一列的工作已經大功告成！我們必須退回到**外層迴圈的縮排層級**，呼叫一次不帶任何參數的 `print()`！這個空呼叫的唯一使命，就是送出一個換行符號，讓光標回到最左側並跳到下一行！
3. **外層推進下一列**：  
   外層迴圈前進到下一輪，再度啟動內層迴圈，開始印製下一列的內容。

**常見錯誤陷阱：**  
- **陷阱一：換行位置放錯層級（嚴重變形）**：初學者最常把負責換行的 `print()` 誤放進了內層迴圈裡面！結果每印一個數字就換行一次，原本寬廣的矩陣直接被撕裂成一條筆直細長的一維垂直長條。
- **陷阱二：忘記呼叫換行（萬里長城）**：完全忘記在外層寫 `print()`，導致整個矩陣的所有數字全部連綿不斷地印在同一個單行上。

**APCS 考試實務建議：**  
APCS 與線上評判系統（如 ZeroJudge）對輸出格式要求達到像素級的精確度！多一個換行、少一個空格都會被判定為 Presentation Error（格式錯誤）甚至 Wrong Answer。牢記「內層水平加 end，外層結束補換行」，是二維排版通關的不二法門！

````python
# [2] Code 範例區：掌握 end=' ' 與 print() 換行節奏，打印 3 列 4 欄星號矩陣
print("=== 打印 3 列 4 欄的星號矩陣 ===")

H = 3  # 高度（列數）
W = 4  # 寬度（欄數）

for r in range(H):
    # 內層迴圈：負責印出同一列的 4 顆星號，水平鋪開
    for c in range(W):
        print("*", end=" ")
    
    # 關鍵時刻：內層結束後，在外層縮排處執行換行！
    print()

print("=== 矩陣排版完成！ ===")
````

### ✖️ 6-5-4 九九乘法表排版大作戰：格式化對齊工整表格（二維網格雛形）

**生活故事：九九乘法表的「成年禮」與歪七扭八的排版災難**  
九九乘法表是全世界每一位程式冒險者在學習雙重迴圈時，必定親手攻克的「經典里程碑」！  
我們在國小課本或墊板上看到的九九乘法表，排列得整齊方正、嚴絲合縫。  
然而，當初學者第一次嘗試用程式印出九九乘法表時，常常會驚呼：「天啊，為什麼我的表格像骨牌倒塌一樣東倒西歪？」  
原因非常簡單：因為乘積的長度不一致！像 $1 	imes 1 = 1$ 乘積只有 1 位數，而 $2 	imes 5 = 10$ 乘積卻有 2 位數。多了一個字元的寬度，後面的所有算式就被硬生生往右推擠了一格，整齊的表格瞬間變成了參差不齊的鋸齒狀！

**運作機制——f-string 寬度格式化對齊神技：**  
要讓乘法表像軍隊踢正步一樣整齊對齊，我們需要給乘積設定**「固定佔位寬度」**！  
在第四章我們學過 f-string，這裡我們將解鎖它的排版超能力：
- 語法：`f"{i}*{j}={i*j:2d}"`
- **`:2d` 的奧秘**：意思是將整數（decimal）格式化為**固定佔用 2 個字元的寬度**！
  - 如果算出來是個位數（例如 6），Python 會自動在左側補上一個空格，變成 `' 6'`（共佔 2 格）；
  - 如果算出來是兩位數（例如 12），它本來就佔 2 格，維持 `'12'`。
這樣一來，無論乘積是 1 位數還是 2 位數，每一個算式印出來的長度完全一樣長！再搭配欄位之間的間距（例如以跳格符號 `	` 或固定兩個空格隔開），就能輕鬆打造出完美對齊的二維矩陣！

**乘法表雙重迴圈架構拆解：**
```python
for i in range(1, 10):       # 外層 i：被乘數（代表第幾列）
    for j in range(1, 10):   # 內層 j：乘數（代表同一列水平展開）
        print(f"{i}*{j}={i*j:2d}", end="  ")
    print()                  # 一整列乘完，換行！
```

**常見錯誤陷阱：**  
- **乘數與被乘數內外層混淆**：如果是直式乘法表（例如第 1 欄全是 1 的乘法，第 2 欄全是 2 的乘法），外層應該是 $j$ 還是 $i$？看清楚題目要求的是「逐列印出」還是「逐欄印出」，切換外層與內層的變數即可改變表格方向。

**APCS 考試實務建議：**  
格式化佔位符 `:2d`, `:3d`, `:4d` 是處理 APCS 報表輸出、日曆排版、二維地圖視覺化最可靠的神器，初學者務必熟練運用！

````python
# [2] Code 範例區：迷你 3x3 九九乘法表——展示 :2d 格式化對齊的威力
print("=== 迷你 3x3 乘法表（格式化完美對齊）===")

for i in range(1, 4):  # 被乘數 1 到 3
    for j in range(1, 4):  # 乘數 1 到 3
        # 使用 :2d 讓乘積固定佔 2 格，結尾留兩個空格間距
        print(f"{i}*{j}={i*j:2d}", end="  ")
    print()  # 該列印製完成，換行

print("=== 排版結束，每一直欄垂直切齊！ ===")
````

### 📐 6-5-5 內層邊界與外層連動：隨外層行數動態改變內層跑動次數

**生活故事：體操隊伍的三角形階梯隊形**  
運動會表演時，大會排出了氣勢磅礴的階梯隊形：  
隊長一個人英姿颯爽站在第 1 排（1 個人）；  
兩位副隊長站在第 2 排（2 個人）；  
第三排站了 3 個人……到了第 5 排，整整齊齊站了 5 個人！  
請初學同學仔細思考：每一排要站幾個人，是一個固定不變的常數嗎？  
不是！每一排要站的人數，**「完全取決於當前是第幾排」**！第幾排，裡面就站幾個人！

**運作機制——動態內層邊界展開幾何世界：**  
在先前的章節中，內層迴圈的次數都是一個死板板的固定常數（例如 `range(5)`），因此每一列印出來的長度都一模一樣，組成了方方正正的矩形。  
如果我們想打破矩形的束縛，印出「階梯形」或「三角形」，關鍵就在於：**將內層迴圈的範圍上限，直接綁定外層迴圈的計數變數！**  
請看這段核心結構：
```python
for i in range(1, N + 1):      # 外層 i 從 1 到 N
    for j in range(1, i + 1):  # 內層上限直接由 i 決定！
        print(j, end=" ")
    print()
```
讓我們以放慢鏡頭逐步追蹤直譯器的執行歷程：
- 當外層 $i = 1$ 時：內層跑 `range(1, 2)`，執行 **1 次**，印出 `1`；
- 當外層 $i = 2$ 時：內層跑 `range(1, 3)`，執行 **2 次**，印出 `1 2`；
- 當外層 $i = 3$ 時：內層跑 `range(1, 4)`，執行 **3 次**，印出 `1 2 3`；
- 當外層 $i = N$ 時：內層跑 `range(1, N + 1)`，執行 **$N$ 次**！
內層的跑動步數隨著外層的節奏「動態伸縮」，這就是二維圖形與矩陣三角走訪的靈魂所在！

**常見錯誤陷阱——差一錯誤（Off-by-one Error）：**  
- **忘記加 1 導致少跑一格**：寫成 `range(1, i)`。因為 Python 的 `range(start, stop)` 是「左閉右開（含頭不含尾）」，如果寫到 `i`，實際上只會跑到 `i - 1`！當 $i = 1$ 時 `range(1, 1)` 直接是空的，第一列就變成了空行！
- **從 0 開始沒調整好**：如果外層寫 `for i in range(N):`（$i$ 從 0 開始），內層寫 `range(i)`，在 $i = 0$ 的那一輪內層完全不會跑！

**APCS 考試實務建議：**  
這種「內層上限受外層控制」的動態架構，在後續氣泡排序法（每一輪比對次數遞減）、二維陣列的上三角與下三角走訪、以及求字串的所有連續子字串中無處不在。掌握外層變數傳遞給內層邊界的代數技巧，是突破演算法初階門檻的關鍵！

````python
# [2] Code 範例區：內層邊界隨外層動態改變——階梯數字三角展開
print("=== 階梯數字三角展開（N = 4）===")

N = 4
for i in range(1, N + 1):  # 外層 i 代表當前列數（1 到 4）
    print(f"第 {i} 列（跑 {i} 次）：", end="")
    for j in range(1, i + 1):  # 內層 j 受外層 i 控制，走訪 1 到 i
        print(j, end=" ")
    print()  # 每列結束換行

print("=== 階梯展開完成！ ===")
````

### 🤝 6-5-6 雙重枚舉（Brute Force）：數對 (i, j) 與所有不重複組合 (i < j) 生成

**生活故事：班級同學兩兩握手與羽球雙打配對**  
想像班上有 4 位同學（編號 1, 2, 3, 4）參加迎新活動，每個人都要跟其他同學互相握手一次。  
小明（1號）跟小華（2號）握手，和小華（2號）跟小明（1號）握手，是不是同一件事？當然是同一件事！我們絕不需要在登記簿上重複記錄兩次！  
而且，任何人都不可能「自己伸出雙手跟自己握手」（1號跟1號握手沒有意義）。  
那麼，我們該如何寫出一個程式，既能把全班所有兩兩互相握手的可能全部找齊，又保證「絕對不漏掉、也絕對不重複握兩次」呢？

**運作機制——鎖定 i < j 的不重複組合枚舉：**  
如果我們寫出最天真的雙重迴圈：
```python
# 天真寫法：所有可能無腦枚舉（排列）
for i in range(1, N + 1):
    for j in range(1, N + 1):
        print(f"({i}, {j})")
```
這段程式會產生 $N^2$ 種結果。其中包含了：
1. 自己配自己：例如 $(1, 1), (2, 2)$；
2. 反向重複配對：例如出現了 $(1, 2)$，後面又會出現 $(2, 1)$！

**黃金解法：強制約束 $j$ 從 $i + 1$ 開始出發！**  
請仔細觀察這段精妙的神仙寫法：
```python
for i in range(1, N + 1):
    for j in range(i + 1, N + 1):  # 內層起點直接是 i + 1！
        print(f"({i}, {j})")
```
因為內層的 $j$ 永遠從 $i + 1$ 起跑，這項設計帶來了三大無懈可擊的優勢：
- **保證 $i < j$**：$j$ 必定嚴格大於 $i$，因此永遠不可能發生 $i == j$（自動消除自己配自己）；
- **保證每組組合只出現一次**：較小的號碼永遠在左邊，較大的號碼永遠在右邊（有了 $(1, 2)$ 就絕不可能出現 $(2, 1)$）；
- **運算量當場減半**：配對總數精確等於高中數學的組合公式 $C(N, 2) = rac{N(N-1)}{2}$！

**常見錯誤陷阱：**  
初學者常在內層仍然寫 `for j in range(1, N + 1):`，然後在裡面用 `if i != j:` 去過濾。這樣做雖然濾掉了自身，但仍然會做兩倍次數的運算，而且會印出重複的反向配對！

**APCS 考試實務建議：**  
在 APCS 考場上，這項技巧被稱為**「暴力枚舉法（Brute Force）」**的最基礎原型！當題目要求「在 $N$ 個物品中找出兩個物品滿足特定條件（如兩數和為目標值、兩點曼哈頓距離最近）」且 $N$ 不大時，直接枚舉所有 $i < j$ 的數對就是拿滿基本分的王道解法！

````python
# [2] Code 範例區：枚舉 1 到 4 的所有不重複兩兩配對 (i < j)
print("=== 枚舉 4 位同學的所有不重複握手配對 ===")

N = 4
pair_count = 0

for i in range(1, N + 1):
    # 核心精髓：j 從 i + 1 開始，保證 i < j 且不重複！
    for j in range(i + 1, N + 1):
        print(f"握手配對：({i}, {j})")
        pair_count += 1

print(f"=== 枚舉完畢！總握手次數為：{pair_count} 次（符合 4*3/2 = 6） ===")
````

### 🚪 6-5-7 巢狀迴圈中的 break 範圍陷阱：只跳一層的特性與跳出多層迴圈之旗標技巧

**生活故事：教室的門 vs 學校的大門**  
想像你正在教學大樓 3 樓的某間教室裡參加考試。  
突然間，你肚子非常痛，不得不舉手向老師報告，並立刻「奪門而出（衝出教室門）」！  
這時請大家停下來想一想：當你衝出教室門的那一瞬間，你人現在在哪裡？  
你人依然站在 3 樓的走廊上！你並沒有因為衝出教室門，就瞬間直接掉到學校校門口外！  
如果你要完全離開整座校園回家休息，你不僅要走出教室門（內層），還必須沿著走廊走到樓梯口，再一步步衝出一樓的校園大門（外層）才行！

**運作機制——break 只能終止直接包住它的那一層：**  
在 Python 中，**`break` 的爆炸威力是嚴格受限的：它「只會跳出它身處的那一個迴圈」！**  
如果在雙重迴圈的內層寫了一個 `break`：
```python
for r in range(3):       # 外層迴圈
    for c in range(3):   # 內層迴圈
        if c == 1:
            break        # 只會跳出內層 c 迴圈！
    # 程式跳到這裡，外層 r 迴圈依然會繼續執行下一輪！
```
這導致初學者常常陷入崩潰：「我明明在裡面寫了 `break`，為什麼整個程式還在一直跑？」因為外層迴圈毫髮無傷，它依然精神抖擻地拉著內層啟動下一輪！

**如何實現「連鎖煞車，雙層徹底跳出」？——布林旗標法（Flag Cascade）：**  
如果我們的任務是「在二維表格中尋找某個寶藏，一旦找到就立刻結束全部搜尋」，正確的標準解法是運用**布林旗標（Flag）進行內外連鎖反應**：
1. **宣告旗標**：在雙重迴圈外層準備一個 `found = False` 警報器；
2. **內層觸發與第一道煞車**：在內層一旦找到目標，立刻將旗標拉起 `found = True`，並執行內層的 `break`（衝出教室門）；
3. **外層感應與第二道煞車**：在內層迴圈剛結束、外層縮排的位置，立刻檢查旗標：`if found: break`（看見警報拉起，立刻衝出學校大門）！
透過這兩道環環相扣的煞車機制，就能優雅、安全地瞬間終止多層巢狀迴圈！

**常見錯誤陷阱：**  
- **單層 break 導致找到答案後被洗掉**：例如搜尋第一個滿足條件的數對，結果因為外層沒跳出，變數在後面的回合被更新成別的數值，導致答案錯誤。
- **忘記在外層放 `if found: break`**：旗標設成 True 了，但外層沒有檢查，導致外層依然跑完全程，浪費了寶貴的執行時間。

**APCS 考試實務建議：**  
在 APCS 歷屆實作題中，例如在 $N 	imes M$ 的二維字元地圖中搜尋主角的起始座標、判斷迷宮是否已找到終點，布林旗標連鎖雙層跳出是保障答案正確且杜絕超時 TLE 的必備經典套路！

````python
# [2] Code 範例區：對比只跳內層 vs 布林旗標連鎖跳出雙層
print("=== 情境一：只寫內層 break（外層依然繼續跑）===")
for r in range(1, 4):
    for c in range(1, 4):
        if r == 2 and c == 2:
            print(f"  [觸發內層煞車] 找到目標 (r={r}, c={c})，執行內層 break！")
            break
    print(f"--> 外層第 {r} 輪結束（外層並未終止）")

print()
print("=== 情境二：運用布林旗標 found 實現雙層連鎖徹底跳出 ===")
found = False
target_r, target_c = 0, 0

for r in range(1, 4):
    for c in range(1, 4):
        if r == 2 and c == 2:
            print(f"  [鎖定目標] 找到寶藏 (r={r}, c={c})！拉起警報並煞車！")
            found = True
            target_r, target_c = r, c
            break  # 第一道煞車：跳出內層
    
    # 關鍵連鎖：外層檢查旗標
    if found:
        print(f"--> 外層感應到警報已觸發，執行第二道煞車，終止外層！")
        break

print(f"最終鎖定寶藏位置：({target_r}, {target_c})")
````

**生活故事：魔術方塊的體積爆炸與巨型沙灘**  
想像一個三階魔術方塊，每條邊長有 3 個小方塊，整顆魔術方塊總共由 $3 	imes 3 	imes 3 = 27$ 個小方塊組成，看起來十分小巧；  
如果今天有一顆十階魔術方塊，每條邊長有 10 個小方塊，總方塊數就會暴增到 $10 	imes 10 	imes 10 = 1000$ 個！  
那如果每條邊長有 1000 個小方塊呢？整顆立方體將包含 $1000 	imes 1000 	imes 1000 = 10$ 億個小方塊！  
三次方增長（Cubic Growth）的威力就像滾雪球一樣，一旦維度增加，運算量會以驚人的幾何級數瞬間失控！

**運作機制——時間複雜度量級與考場運算上限：**  
在撰寫迴圈時，我們必須具備估算電腦執行次數的「時間複雜度（Time Complexity）」概念：
- **一重迴圈（線性，Linear）**：跑 $N$ 次，記作 $O(N)$。若 $N = 10^6$（一百萬），電腦約 0.05 秒即可跑完。
- **雙重迴圈（平方，Quadratic）**：跑 $N 	imes N = N^2$ 次，記作 $O(N^2)$。若 $N = 1000$，總步數為 $1000^2 = 10^6$（一百萬次），電腦依然能在 0.1 秒內輕鬆搞定。
- **三重迴圈（立方，Cubic）**：外層、中層、內層三重嵌套，總步數高達 $N 	imes N 	imes N = N^3$ 次，記作 $O(N^3)$！  
  - 當 $N = 100$ 時：$100^3 = 10^6$（一百萬次），可以安全過關；
  - 但當 $N = 1000$ 時：$1000^3 = 10^9$（十億次）！

**APCS 考場的 1 秒鐘生死線：**  
在 APCS 評測系統中，Python 每 1 秒鐘大約只能執行 $10^7$（一千萬）到 $3 	imes 10^7$ 次基本運算指令。  
如果你面對一道測資規模 $N = 1000$ 的題目，卻寫出了一個三重迴圈，你的程式需要整整跑 30 秒以上！評測系統會在第 1.0 秒時無情地將你的程式強制中斷，並噴出巨大的紅字：**TLE（Time Limit Exceeded，程式執行逾時）**！

**常見錯誤陷阱：**  
初學者看到題目給了三個變數（例如三個數相加等於某值），就理所當然地寫出三層 `for` 迴圈。在解題前，**請務必先看一眼題目末尾的「測資範圍（Constraints）」**！

**APCS 考試實務建議：**  
審題先看 $N$！  
- 若 $N \le 50 \sim 100$，三重迴圈 $O(N^3)$ 是完全合法的暴力首選；  
- 若 $N \ge 1000$，請立刻打消三重迴圈的念頭！通常第三層變數可以透過代數關係直接算出（例如 $k = target - i - j$），當場把複雜度降階到 $O(N^2)$，這就是演算法優化的神奇魅力！

````python
# [2] Code 範例區：體驗三重迴圈運算量的大爆炸（計數器量測）
print("=== 比較 N = 5 時，一重、雙重、三重迴圈的執行次數 ===")

N = 5

# 1. 一重迴圈
count_1 = 0
for i in range(N):
    count_1 += 1

# 2. 雙重迴圈
count_2 = 0
for i in range(N):
    for j in range(N):
        count_2 += 1

# 3. 三重迴圈
count_3 = 0
for i in range(N):
    for j in range(N):
        for k in range(N):
            count_3 += 1

print(f"N = {N} 時：")
print(f"一重迴圈執行次數：{count_1:4d} 次 (N)")
print(f"雙重迴圈執行次數：{count_2:4d} 次 (N^2)")
print(f"三重迴圈執行次數：{count_3:4d} 次 (N^3)")
print("=== 想像若 N = 1000，三重迴圈將會達到 1,000,000,000 次！ ===")
````

````python
# ==========================================
# [4] Code 練習題
# 任務說明：
# 三角形三邊長整數解枚舉：
# 給定三角形周長 P = perimeter。
# 我們要在滿足 1 <= a <= b <= c < P 的範圍內，
# 尋找所有滿足「三邊和恰為 P（a + b + c == P）」且「滿足三角形不等式（兩短邊之和大於最長邊：a + b > c）」的整數三角形邊長組合 (a, b, c)。
# 請使用三重迴圈進行枚舉，並統計合法三角形組合總數 valid_triangles。
#
# 【公開測試資料 1】
# 設定：P = 12
# 合法三角形組合：(2, 5, 5), (3, 4, 5), (4, 4, 4)
# 預期：valid_triangles = 3
#
# 【公開測試資料 2】
# 設定：P = 15
# 合法三角形組合：(1, 7, 7), (2, 6, 7), (3, 5, 7), (3, 6, 6), (4, 4, 7), (4, 5, 6), (5, 5, 5)
# 預期：valid_triangles = 7
# ==========================================
# 請在下方撰寫你的程式碼並執行測試：

P = 12
valid_triangles = 0

for a in range(1, P):
    for b in range(a, P):
        for c in range(b, P):
            if a + b + c == P and a + b > c:
                print(f"找到合法三角形：a={a}, b={b}, c={c}")
                valid_triangles += 1

print("合法三角形總數：", valid_triangles)

# 公開測資驗證：
assert valid_triangles == 3, "公開測資 1 驗證失敗"
print("公開測資 1 通過！")
````

````python
# ==========================================
# [5] Code 挑戰題
# 任務說明：
# 湊零錢硬幣枚舉（三重迴圈與二重優化思維）：
# 身上有面額 1 元、5 元、10 元的硬幣各若干枚。
# 現在要精準湊出總金額 total_amount = 25 元。
# 請使用三重迴圈枚舉：
# num_10（10元硬幣個數，範圍 0 到 2）
# num_5（5元硬幣個數，範圍 0 到 5）
# num_1（1元硬幣個數，範圍 0 到 25）
# 條件：num_10 * 10 + num_5 * 5 + num_1 * 1 == total_amount。
# 統計共有多少種不同的湊錢組合方案 coin_ways。
# （本題為自由挑戰題，無公開測資，請依題目情境自行思考並撰寫完整程式碼）
# ==========================================
# 請在下方撰寫你的程式碼：

total_amount = 25
coin_ways = 0

for num_10 in range(0, 3):
    for num_5 in range(0, 6):
        for num_1 in range(0, 26):
            if num_10 * 10 + num_5 * 5 + num_1 == total_amount:
                coin_ways += 1

print(f"湊齊 {total_amount} 元的硬幣組合總數共有：{coin_ways} 種")
# 驗證方案數：
# 10元=0: 5元可選 0~5 (6種)
# 10元=1: 剩餘15元，5元可選 0~3 (4種)
# 10元=2: 剩餘5元，5元可選 0~1 (2種)
# 總計 = 6 + 4 + 2 = 12 種
assert coin_ways == 12, "挑戰題計算錯誤！"
print("挑戰題 6.5.8 驗證通過！")
````

1. **時鐘心智模型**：
   - 外層時針推進一小步，內層分針跑滿一整圈；執行總次數為外層次數乘內層次數（$M 	imes N$）。
2. **命名鐵律防同名遮蔽**：
   - 橫列寫 `r`，直欄寫 `c`（或數學 `i, j`），嚴禁內外層共用同一個變數名稱，守護座標純淨度。
3. **精密二維排版節奏**：
   - 內層水平輸出指定 `end=' '`，外層每列結束呼叫空 `print()` 執行換行，告別變形與擠線。
4. **格式化寬度對齊**：
   - 活用 f-string 佔位符 `f"{val:2d}"`，消除位數長短落差，打造方正工整的二維數據矩陣。
5. **動態內層邊界**：
   - 將內層範圍終點綁定外層變數（如 `range(1, i + 1)`），解鎖階梯形與三角形的幾何密碼。
6. **不重複組合枚舉（$i < j$）**：
   - 內層起點強制設為 $i + 1$，自動消除自我配對與相反重複，計算量精確減半（$O(N^2 / 2)$）。
7. **布林旗標連鎖雙層煞車**：
   - `break` 只跳直接身處的一層；透過布林旗標 `found` 內外呼應，實現找到目標瞬間徹底停機。
8. **$O(N^3)$ 三重迴圈量級警戒**：
   - 牢記 APCS 考場 1 秒約 $10^7$ 次上限。當 $N \ge 1000$ 時三重迴圈必爆 TLE，必須善用代數降階！
