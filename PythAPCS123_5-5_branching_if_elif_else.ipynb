### 🚩 5-5-1 單向選擇 `if` 指令——冒號與縮排程式區塊的誕生

**生活比喻：火車站的補票檢查哨**  
想像你搭火車抵達終點站出站。對於一般持有合法車票的人，閘門直接放行；但站務員眼睛很尖，「**如果（if）**你持有的車票區間不足（條件為真）」，他就會把你引導到一旁的補票處補繳差額（執行特定指令區塊）！如果車票完全沒問題（條件為假），補票這件事就根本不會發生，直接順利出站！

這就是**單向選擇（Single Selection）**——只有在條件成立時，才額外執行特定的加碼動作。

**語法結構與核心機制：**
```python
if 條件運算式:
    條件成立時要執行的程式碼區塊
```
1. **冒號 `:` 的宣告性**：
   在 `if 條件` 的最後面，**必須加上一個英數半形冒號 `:`**！這個冒號在 Python 中代表：「*注意！接下來要展開一個附屬於此條件的程式區塊！*」
2. **縮排（Indentation）界定結構**：
   在第一章 1-6 節中我們學過縮排。被 `if` 管轄的指令行，前面必須統一空 4 個半形空格（或一個 Tab 鍵）。只要縮排縮在裡面，就代表「受該 if 保護」；當縮排退回到最左側時，代表分支結束，回歸主幹道！

**教材範例（吳邦一老師講義第 19 頁）：**
```python
if 2 < 3:
    print(5)
```
因為 `2 < 3` 為 `True`，所以電腦會執行縮排內的 `print(5)`，在螢幕上印出 5。

**常見錯誤陷阱：**
1. **遺漏冒號 `:`**：這是初學者第一大失誤，會引發 `SyntaxError: expected ':'`。
2. **縮排空格不一致**：有時空 2 格、有時空 4 格，會引發 `IndentationError`。請一律維持 4 個半形空白！

**APCS 考試實務建議：**
單向 `if` 常用於「更新極值（例如 `if x > max_val: max_val = x`）」或「特例加扣分處理」。


````python
# [2] Code 範例區：展示單向選擇 if 指令與縮排結構

# 1. 吳邦一老師講義第 19 頁經典範例
if 2 < 3:
    print("講義範例：條件 2 < 3 成立，印出 5")

# 2. 實務情境：超商滿千折百單向回饋
total_bill = 1250
print("--- 購物結帳前 ---")
print("原始消費金額：", total_bill)

# 如果滿 1000 元，觸發折價 100 元優惠
if total_bill >= 1000:
    print("恭喜達到滿千門檻！享有折價 100 元回饋！")
    total_bill = total_bill - 100  # 縮排內部指令

print("--- 最終實付金額 ---")
print("應付總額：", total_bill)
````

### ⚖️ 5-5-2 雙向互斥選擇 `if-else`——非此即彼的命運分水嶺

**生活比喻：道路叉路口的左轉與右轉**  
想像汽車開到一個丁字路口，正前方是一面石壁，你只能選擇向左轉或者向右轉，絕不可能同時走兩條路，也不可能停在半空中。  
- 如果往左（`if` 條件成立），進入蔚藍海岸公路。
- **否則（`else`，條件不成立）**，必然進入翠綠森林大道。

這就是**雙向互斥選擇（Two-Way Selection，`if-else`）**！

**語法架構與核心規則：**
```python
if 條件運算式:
    條件成立時執行的區塊 A
else:
    條件不成立時執行的區塊 B
```
1. **完全互斥（Mutually Exclusive）**：
   區塊 A 與區塊 B 必定**恰好只會執行其中一個**！電腦永遠不會同時執行 A 與 B，也絕對不會兩個都不執行。
2. **`else` 後面絕對不能加條件式**：
   初學者常犯的致命毛病是在 `else` 後面又偷加條件，例如寫成 `else score < 60:`。請記住：**`else` 的意思就是「其餘所有上述沒提到的狀況」**，它本身就涵蓋了條件不成立的所有剩下情況，因此 `else` 後方直接接冒號 `:`，千萬不能加任何條件！
3. **對齊與縮排層級**：
   `if` 與對應的 `else` 必須位於**相同的水平對齊縮排線**上！

**教材範例（吳邦一老師講義第 20 頁）：**
```python
if a < b:
    c = b
    d = b - a
else:
    c = a
    d = a - b
```
這段代碼讓 `c` 永遠存放較大的數值，而 `d` 永遠存放較大減較小的正向差值（大減小），展現了雙向互斥賦值的經典魅力。

**APCS 考試實務建議：**
及格/不及格、奇數/偶數、獲勝/落敗等二分法問題，均為 `if-else` 的標準舞台。


````python
# [2] Code 範例區：展示雙向互斥選擇 if-else 架構

# 1. 吳邦一老師講義第 20 頁範例：求兩數之極大值與大減小的正差值
a = 7
b = 15

if a < b:
    c = b        # 較大值存入 c
    d = b - a    # 正差值存入 d
else:
    c = a
    d = a - b

print("a =", a, ", b =", b)
print("兩數中較大者 c =", c)
print("兩數相差正差值 d =", d)

# 2. 實務情境：及格與不及格判定
score = 58
print("--- 考試成績審查 ---")
if score >= 60:
    print("審查結果：恭喜及格 PASS！")
else:
    print("審查結果：成績不及格 FAIL，需參加補考！")
````

### 🪜 5-5-3 多向互斥選擇 `if-elif-else`——階梯式分級審查

**生活比喻：自強號、莒光號與區間車的分道月台**  
當乘客來到鐵路樞紐，廣播依據車票等級引導月台：
- 第一優先：「持自強號車票者，請往第一月台。」
- 否則第二優先：「**如果持有（elif）**莒光號車票者，請往第二月台。」
- 否則第三優先：「**如果持有（elif）**區間車票者，請往第三月台。」
- 「**其餘所有乘客（else）**，請至服務台洽詢。」

在 Python 中，`elif` 是 **`else if`（否則如果）** 的極速合體縮寫！它建構起一套**多向互斥的階梯式決策系統**。

**語法結構與運作順序：**
```python
if 條件 1:
    區塊 1
elif 條件 2:
    區塊 2
elif 條件 3:
    區塊 3
else:
    前面全都不符合時執行的備用區塊
```
**階梯運作三部曲（至關重要！）：**
1. **由上而下循序檢驗**：電腦從最上方的 `if` 開始比對。
2. **命中即跳離（First Match Wins）**：只要上方**有任何一個條件被滿足並執行了該區塊，整個 `if-elif-else` 階梯立刻結束！** 後面所有剩下的 `elif` 與 `else` 會被電腦直接略過，看都不看！
3. **安全備用 `else`**：若上方所有 `if` 與 `elif` 全數落空，才會由最後的 `else` 兜底執行。

**教材經典核心範例（吳邦一老師講義第 21 頁）：**
```python
if score > 90:
    grade = 'A'
elif score > 80:
    grade = 'B'
elif score > 70:
    grade = 'C'
elif score > 60:
    grade = 'D'
else:
    grade = 'F'
```
請大家特別注意第二個條件 `elif score > 80:`：這裡為什麼不需要寫 `80 < score <= 90`？  
**因為如果 score 真的大於 90，它早在第一回合的 `if score > 90:` 就被攔截帶走了！** 能活著抵達第二階梯的人，分數必然小於等於 90！這正是 `if-elif` 階梯架構最迷人、最精簡的特性。

**APCS 考試實務建議：**
在設計階梯時，**條件順序必須具備嚴謹的單調性（由嚴格到寬鬆，或由大到小）**！若順序排錯（例如把 `score > 60` 放在最頂層），95 分的同學就會過早被誤判為 D 等第！


````python
# [2] Code 範例區：展示多向互斥選擇 if-elif-else 階梯架構

# 1. 吳邦一老師講義第 21 頁成績等第劃分範例
score = 85

if score > 90:
    grade = 'A'
elif score > 80:
    grade = 'B'
elif score > 70:
    grade = 'C'
elif score > 60:
    grade = 'D'
else:
    grade = 'F'

print("--- 成績等第測試 ---")
print("學生實測得分：", score)
print("階梯評定等第：", grade)  # B

# 2. 數值正負與零三相判定
x = 0
if x > 0:
    print("x 是正數")
elif x < 0:
    print("x 是負數")
else:
    print("x 恰好等於零 ZERO")
````

### ⚠️ 5-5-4 獨立多重 `if` vs 互斥 `if-elif`——破解初學者最常犯的覆蓋天坑

**生活比喻：多道獨立安檢 vs 單一專屬通道**  
- **獨立多重 `if`（走過每一扇門）**：就像機場出境，你必須先通過「身分證檢查（if 1）」，接著通過「隨身行李 X 光機（if 2）」，最後通過「金屬探測門（if 3）」。每一道檢驗都是獨立的，每一扇門你都必須走一遍！
- **互斥 `if-elif`（多扇門中只進一扇）**：就像醫院掛號，你依據病情選擇內科、外科或眼科。一旦進入了內科診間，其他科別的門你就不會再去敲！

**初學者最容易踩中的連鎖覆蓋重大 Bug：**
看下面這段錯誤程式碼：
```python
# 🚨 嚴重錯誤示範：使用多個獨立 if
score = 95
if score >= 60:
    grade = 'Pass'
if score >= 90:
    grade = 'A'
```
這段代碼執行完後，`grade` 確實是 `'A'`，看起來好像沒問題對吧？  
但如果順序稍微倒過來：
```python
# 🚨 驚悚悲劇發生！
score = 95
if score >= 90:
    grade = 'A'
if score >= 60:   # 95 分也滿足 >= 60！
    grade = 'Pass'
```
結果：95 分的同學最終拿到的不是 `'A'`，而是被第二個獨立 `if` 粗暴地**覆蓋**成了 `'Pass'`！  
因為兩個 `if` 各自獨立，電腦執行完第一個 `if` 之後，依然會強制去跑第二個 `if`！

**核心口訣與使用時機：**
1. **只要各選項之間「互斥」（只能選一個結果）**：請一律使用 **`if-elif-else`**！
2. **多個條件「可能同時成立且需要同時執行」**（例如檢查是否需要加購配件、檢查各項防護是否啟用）：才使用**多個獨立的 `if`**！

**APCS 考試實務建議：**
90% 的初學者邏輯錯誤（WA），都是因為誤用獨立 `if` 代替 `elif`，導致變數值被後面的條件非預期覆蓋。請在動筆前明確思考：這些條件是「二選一」還是「並行不悖」！


````python
# [2] Code 範例區：展示獨立 if 與互斥 elif 的巨大差異

score = 95

# 1. 🚨 錯誤示範：使用獨立 if 引發的後續覆蓋
grade_wrong = "None"
if score >= 90:
    grade_wrong = "A"
if score >= 60:
    grade_wrong = "Pass"  # 95 分也大於等於 60，慘遭覆蓋！

print("獨立 if 的覆蓋悲劇：score = 95，最終等第竟然變成：", grade_wrong) # Pass (錯誤！)

# 2. ✅ 正確示範：使用互斥 if-elif，命中即跳離
grade_correct = "None"
if score >= 90:
    grade_correct = "A"    # 命中此處，後續 elif 全數略過！
elif score >= 60:
    grade_correct = "Pass"

print("互斥 if-elif 正確防禦：score = 95，最終等第正確維持：", grade_correct) # A (正確！)
````

### ⚡ 5-5-5 單行 if 簡寫語法——俐落優雅的競技程式快速寫法

**生活比喻：便條紙快速備忘 vs 正式長篇公文**  
當你想提醒自己出門帶雨傘時，你不會在便條紙上規規矩矩分四行寫抬頭和公文，而是直接一行字：「*若下雨則帶傘*」。  
在 Python 程式語言中，如果條件成立時**只有單獨一行指令要執行**，Python 允許我們直接把該指令寫在冒號 `:` 的同一行後方！

**語法結構與教材範例（吳邦一老師講義第 19 頁）：**
```python
# 講義原著範例：單行 if
if 2 < 3: print(5)

# 講義原著範例：單行極值判定（第 20 頁）
if a < b: c = a
else: c = b
```

**單行簡寫的優缺點分析：**
1. **優點（簡短俐落）**：
   在 APCS 程式競賽中，爭分奪秒是常態。單行簡寫能省去按下 Enter 與縮排的繁瑣，讓簡單的賦值或旗標開關一氣呵成。
2. **缺點與注意事項**：
   - 冒號後方**只能放簡單的單一語句**。如果有多條指令，強烈建議回歸標準的換行縮排結構。
   - 請勿在單行中硬塞過於複雜的巢狀邏輯，過度追求短代碼反而會毀滅程式的可讀性。

**APCS 考試實務建議：**
在實作題中，例如「*發現負數時轉正：`if x < 0: x = -x`*」或「*發現更大值時更新：`if val > max_v: max_v = val`*」，單行簡寫是頂尖競賽選手極為鍾愛的洗鍊風格。


````python
# [2] Code 範例區：展示單行 if 與單行 if-else 簡寫

# 1. 吳邦一老師講義第 19 頁原著範例
if 2 < 3: print("講義單行 if 示範：5")

# 2. 吳邦一老師講義第 20 頁單行極值更新
a = 15
b = 8
if a < b: c = a
else: c = b
print("較小值 c =", c)

# 3. 實務情境：負數自動轉正（絕對值手刻單行版）
x = -42
if x < 0: x = -x
print("負數轉正結果 x =", x)
````
