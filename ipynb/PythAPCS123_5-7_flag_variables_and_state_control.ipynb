### 🚩 5-7-1 什麼是旗標變數——程式記憶狀態的儀表板指示燈

**生活比喻：汽車儀表板上的「引擎故障燈」**  
當你發動汽車時，儀表板上的引擎故障燈是熄滅的（狀態為 `False`，代表一切安好）。在行駛過程中，車載電腦會持續感測各個部件：
- 檢查水溫是否過高？
- 檢查機油壓力是否不足？
- 檢查含氧感知器是否異常？
只要其中任何一個部件亮起紅燈，故障燈就會被點亮並保持常亮（狀態被更新為 `True`）！當駕駛抵達修車廠時，技師不需要拆開所有螺絲，只要看一眼儀表板上的故障燈有沒有亮，就能立刻知道「這趟旅程中是否有異常發生」！

在 Python 程式語言中，這種**「用來記錄某件事情是否曾經發生、或者某個狀態是否成立」的布林變數，就叫做「旗標變數（Flag Variable）」**！

**旗標變數的命名藝術：**
好的旗標變數名稱，本身就能說明它的職責。在業界與競賽中，最常見的優秀旗標命名有：
- `found = False`（是否找到了目標？）
- `has_error = False`（是否有發生錯誤？）
- `is_valid = True`（目前資料是否有效？）
- `matched = False`（是否有成功匹配？）

**常見錯誤陷阱：**
- **隨意使用無意義的單字母命名**：初學者喜歡把旗標取名為 `f` 或 `t`，寫到後面自己都忘記 `f=True` 到底是代表「成功」還是「失敗」。請務必使用具備明確語意的英文名稱！

**APCS 考試實務建議：**
旗標變數是演算法狀態機（State Machine）的基礎核心。養成建立旗標的習慣，可以讓你的程式架構從混亂的多頭馬車，昇華為層次分明的專業代碼。


````python
# [2] Code 範例區：展示旗標變數的定義與狀態記錄

# 1. 建立旗標變數：預設為 False（尚未發現目標）
target_found = False

item1 = 15
item2 = 88
item3 = 42
target = 88

# 逐一檢查各項目，若等於 target 則將旗標立起（設為 True）
if item1 == target:
    target_found = True
if item2 == target:
    target_found = True
if item3 == target:
    target_found = True

# 最終依據旗標狀態進行單一回報
print("目標數值：", target)
print("三個項目數值：", item1, item2, item3)
print("是否在項目中成功找到目標：", target_found)  # True
````

### 🔄 5-7-2 旗標的生命週期——歸零初始化、條件觸發與狀態鎖定

**生活比喻：火警警報器的自鎖機制**  
在大樓火警系統中：
1. **第一階段（初始化）**：系統開機時，警報開關被重置為安全狀態（`alarm = False`）。
2. **第二階段（條件觸發）**：當三樓的煙霧感測器探測到濃煙時，開關被推上（`alarm = True`）。
3. **第三階段（狀態鎖定）**：此時即使四樓和五樓沒有煙霧，警報開關**絕不會因為其他樓層沒事而被降回去！** 它會保持鎖定在 `True`，直到整棟大樓所有人安全疏散！

**旗標的單向鎖定（One-way Latch）：**
在許多演算法情境中，旗標具備「只升不降」的特徵。
```python
# 1. 歸零初始化
found = False

# 2. 條件觸發更新（只在命中時設為 True）
if condition_1:
    found = True
if condition_2:
    found = True
# 🚨 絕對不要在 else 裡面把 found 設回 False！
```

**初學者最常犯的重大邏輯漏洞：**
```python
# 🚨 驚悚錯誤示範！
found = False
if item1 == target:
    found = True
else:
    found = False  # 沒問題

if item2 == target:
    found = True
else:
    found = False  # 💥 毀滅性災難！
```
如果 `item1` 是目標（設為 True），但 `item2` 不是目標，第二個 `else` 就會把 `found` 硬生生**重置回 False**！前面找到的成果瞬間化為烏有！  
請牢記鐵律：**除非題目有特殊重置需求，否則「偵測到即成立」的旗標，絕對不要在 else 裡將它改回 False！**

**APCS 考試實務建議：**
牢牢記住旗標的標準三步流程：**「事前歸零 ➔ 命中立旗 ➔ 事後統籌」**，這是競賽得分的護身符。


````python
# [2] Code 範例區：展示旗標單向鎖定與避免誤重置

target = 100
a = 100
b = 50
c = 30

# 正確寫法：只在命中時立旗，不可在未命中時清空旗標！
found = False

if a == target:
    found = True

if b == target:
    found = True

if c == target:
    found = True

print("a =", a, ", b =", b, ", c =", c)
print("是否在序列中偵測到 100（正確鎖定）：", found)  # True
````

### 🚪 5-7-3 單一出口原則——以旗標統籌多分支最終輸出

**生活比喻：結帳櫃台的單一收據印表機**  
在大型超市購物時，如果蔬果區印一張發票、肉品區印一張發票、日用品區又印一張發票，顧客出門時手上會拿著十幾張零碎的紙條，混亂不堪。  
現代化的收銀系統是在各個貨架只負責「掃描紀錄購物車清單（變更旗標或資料狀態）」，最後當顧客推車走到大門口的總收銀台時，由總收銀機**統籌印出一張完整明瞭的總收據（單一出口）**！

**軟體工程金律：單一出口原則（Single Exit Principle）**  
初學者最常犯的寫法是在程式的各個角落到處塞 `print()`：
```python
# 🚨 容易混亂的散落式輸出
if score >= 90:
    print("PASS WITH A")
elif score >= 60:
    print("PASS")
else:
    print("FAIL")
```
如果明天老師突然說：「*所有輸出文字前面都要加上 [學生姓名]*」，你就必須跑到每一個分支裡面逐一修改 `print()`，一旦漏改一個就會引發 Bug！

**以旗標/結果變數統籌的優雅架構：**
```python
# ✅ 專業級單一出口架構
result = "FAIL"  # 預設狀態

if score >= 90:
    result = "PASS WITH A"
elif score >= 60:
    result = "PASS"

# 全程式唯一的出口！
print(f"[{name}] {result}")
```
無論前面有五個分支還是十個分支，**資料歸資料處理，最後交給單一出口統籌印出**！這能讓程式邏輯與輸出展示完全解耦。

**APCS 考試實務建議：**
在 APCS 評測中，輸出格式的要求極端嚴苛（多一個空白或少一個換行都會被判定 WA）。採用單一出口架構，能讓你輕鬆掌控最外層的輸出格式。


````python
# [2] Code 範例區：展示單一出口原則架構

score = 75

# 1. 以單一狀態變數統籌
final_status = "UNKNOWN"

if score >= 90:
    final_status = "EXCELLENT"
elif score >= 60:
    final_status = "PASSED"
else:
    final_status = "FAILED"

# 全程式唯一的輸出點（單一出口）
print("--- 單一出口規範印出 ---")
print("學生考核評定結果：", final_status)
````

### 🚥 5-7-4 多旗標狀態組合——多項指標聯合會審

**生活比喻：太空梭發射前的三道綠燈會審**  
在航太發射中心，指揮官倒數計時前，螢幕上有三盞大燈：
1. 第一盞燈：推進器燃料壓力是否正常（`fuel_ok = True`）
2. 第二盞燈：發射場高空氣候是否合規（`weather_ok = True`）
3. 第三盞燈：地面遙測通訊是否暢通（`telecom_ok = True`）
指揮官只有在看到**三盞綠燈全部亮起**時，才會下達「發射（`LAUNCH`）」的終極指令！若有任何一盞燈未亮，立刻中止發射（`ABORT`）！

**多旗標協同運作機制：**
當一個系統有多個不同維度的健康狀態時，我們可以分別宣告專屬的布林旗標：
```python
flag_a = False
flag_b = False

# 各自獨立檢驗維度 A
if 條件 A:
    flag_a = True

# 各自獨立檢驗維度 B
if 條件 B:
    flag_b = True

# 聯合會審
if flag_a and flag_b:
    print("ALL PASS")
else:
    print("CHECK FAILED")
```

**優點解析：**
1. **解耦清晰**：每個旗標專注於自己的檢查邏輯，互不干擾。
2. **易於擴充**：如果要追加第四個檢查項目，只需多開一盞燈 `flag_d`，整體架構完全不需大改。

**APCS 考試實務建議：**
在許多字串或陣列分析題目中，常要求同時滿足「長度足夠」、「包含特定字母」與「不含違規符號」多項條件，多旗標組合是最佳解題模板。

### 🏆 5-7-5 APCS 實作經典模式——c461 前哨戰：IMPOSSIBLE 無解防禦輸出

**生活比喻：懸賞佈告欄的「查無此人」**  
城門口的捕快拿著通緝畫像比對進城的百姓：
- 他依序比對：這個人像不像嫌疑犯 A？像就通報「MATCH A」；像不像嫌疑犯 B？像就通報「MATCH B」。
- 一整天下來，如果捕快「**至少抓到了一個人**」，他就收工回府。
- 但如果經過整整一天的嚴格排查，**連一個符合的人都沒看見（一個解都沒有）**，捕快就必須在報告書上鄭重蓋下大印：「**IMPOSSIBLE（查無可疑人員）**」！

**APCS 經典題型震撼登場：c461. 邏輯運算子**  
在真實的 APCS 檢定真題中（例如 c461 題），題目給定輸入與輸出，要求我們依序檢驗可能是 `AND`、`OR` 或 `XOR` 中的哪幾種運算。  
可能同時有多個運算符合（例如符合 AND 又符合 OR，兩者都要印出），**但若三者全數不符時，必須印出 "IMPOSSIBLE"！**

**這正是旗標變數最榮耀的實戰舞台！**
```python
# 旗標變數：記錄是否有任何一種情況成功命中
has_any_match = False

# 獨立檢驗情況 1
if 條件 1:
    print("情況 1 命中")
    has_any_match = True

# 獨立檢驗情況 2
if 條件 2:
    print("情況 2 命中")
    has_any_match = True

# 🚨 終極無解防禦！
if not has_any_match:
    print("IMPOSSIBLE")
```
看！這段代碼何其優雅！  
前面每一個 `if` 只要有命中，就會各自印出，並順手將旗標立起（`has_any_match = True`）；  
如果前面全軍覆沒，旗標就會一直保持寂靜的 `False`；最後一行 `if not has_any_match` 就會帥氣地兜底印出 `IMPOSSIBLE`！

**APCS 考試實務建議：**
「有解逐一印出，無解印出 IMPOSSIBLE」是 APCS 實作題出題頻率極高的標準範式。掌握此模板，等於直接拿下真題的核心解題骨架！


````python
# [2] Code 範例區：展示 APCS c461 經典 IMPOSSIBLE 無解防禦輸出架構

# 模擬尋找符合條件的倍數
target = 15

has_solution = False

# 檢驗是否能被 2 整除
if target % 2 == 0:
    print("DIVISIBLE BY 2")
    has_solution = True

# 檢驗是否能被 3 整除
if target % 3 == 0:
    print("DIVISIBLE BY 3")
    has_solution = True

# 檢驗是否能被 5 整除
if target % 5 == 0:
    print("DIVISIBLE BY 5")
    has_solution = True

# 終極兜底防禦：若前面一個因數都沒找到，印出 IMPOSSIBLE
if not has_solution:
    print("IMPOSSIBLE")
````
