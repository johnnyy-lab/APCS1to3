### 📐 6-6-1 幾何排版核心心法：列數 r 與行數 c 的代數關係式推導

**生活故事：泥水匠砌磚與方格紙上的像素畫**  
請初學同學想像一位專業的泥水匠正在砌一面造型紅磚牆。  
他絕不會拿起磚頭隨意亂堆，而是先拿出一張藍圖，上面清楚標註著：「第 1 層砌 1 塊磚，第 2 層砌 3 塊磚，第 3 層砌 5 塊磚……」  
又或者我們在方格紙上畫像素風的馬賽克圖案，每一個小方格要不要塗黑，都取決於它位在「第幾列（垂直高度 $r$）」與「第幾欄（水平位置 $c$）」！  
在終端機黑底白字的螢幕上用星號 `*` 排版圖案，本質上就是在方格紙上作畫！

**運作機制——幾何排版的三步推導心法：**  
許多初學者在寫星號排版時，常常「憑感覺亂加減」：代碼跑出來少一顆星，就加個 1；多換了一行，就刪個 1，結果越改越亂。  
真正頂尖的程式設計師，靠的是嚴謹的**「代數關係推導三步法」**：
1. **第一步：建立座標草稿表**：  
   在紙上畫出目標圖形，將垂直方向的列號標記為 $r$（通常習慣由上而下標為 $1, 2, 3, \dots, H$）。
2. **第二步：找出「星號數量」與「列號 $r$」的數學關係式**：  
   觀察每一列星號的數量規律：
   - 如果第 1 列有 1 顆，第 2 列有 2 顆，第 3 列有 3 顆 ➔ 星號數 $= r$；
   - 如果第 1 列有 2 顆，第 2 列有 4 顆，第 3 列有 6 顆 ➔ 星號數 $= 2r$；
   - 如果第 1 列有 1 顆，第 2 列有 3 顆，第 3 列有 5 顆 ➔ 星號數 $= 2r - 1$（奇數數列！）。
3. **第三步：將代數式化為內層迴圈的 `range()` 邊界**：  
   若每一列要印 $K$ 顆星號，內層迴圈直接寫 `for c in range(K): print('*', end='')`！

**常見錯誤陷阱：**  
- **急於敲鍵盤而不動筆推導**：沒有在草稿紙上列出 $r$ 與數量的對照表，直接在代碼裡微調，導致陷入死胡同。
- **忘記歸納為代數通式**：把每一列寫成獨立的 `if r == 1: ... elif r == 2: ...`，完全失去了迴圈自動化的意義。

**APCS 考試實務建議：**  
在 APCS 觀念題中，經常考查「給定一段巢狀迴圈代碼，問螢幕上會印出幾顆星號」或「下列哪一個選項能印出特定幾何圖案」。只要熟練代數關係式推導，就能在 10 秒內秒殺題目！

````python
# [2] Code 範例區：代數關係式推導展示——每列印出 2 * r 顆星號
print("=== 代數關係推導展示：每列印出 2 * r 顆星號（H = 4）===")

H = 4
for r in range(1, H + 1):
    stars_count = 2 * r  # 代數關係式：星號數量為列數的 2 倍
    print(f"列號 r={r}（預期印 {stars_count} 顆星）：", end="")
    
    # 內層迴圈依據代數式跑動 stars_count 次
    for c in range(stars_count):
        print("*", end="")
    print()  # 該列結束換行

print("=== 規律圖案列印完成！ ===")
````

### 🖼️ 6-6-2 等長矩形與空心矩形：邊框條件（首列、末列、首欄、末欄）判定實戰

**生活故事：實心相框 vs 裝著玻璃的空心相框**  
想像我們木工課做相框。  
如果我們拿一整塊紮實的實木板，把整片塗成黑色，那就是一個「實心矩形」；  
但如果我們要放照片，我們只會把最外圍的「上邊框、下邊框、左邊框、右邊框」用木條框起來，中間挖空裝上透明玻璃！  
站在方格陣列的視角來看，什麼樣的格子屬於「外框」？什麼樣的格子屬於「中間的透明玻璃」呢？

**運作機制——邊框四界判定的邏輯：**  
給定一個高度為 $H$、寬度為 $W$ 的網格，座標為 $(r, c)$（其中列號 $r$ 從 0 到 $H-1$，欄號 $c$ 從 0 到 $W-1$）：
- **實心矩形**：非常單純，外層走訪列 $r$，內層走訪欄 $c$，每一個格子無條件全部印出星號 `print('*', end='')`！
- **空心矩形**：不是每個格子都印星號！我們必須加入一道條件分支門檻：
  - **什麼時候該印星號 `*`？**  
    只要當前格子滿足以下四種「邊界條件」中的**任一種**即可（以邏輯 `or` 連接）：
    1. **最頂層（首列）**：`r == 0`
    2. **最底層（末列）**：`r == H - 1`
    3. **最左側（首欄）**：`c == 0`
    4. **最右側（末欄）**：`c == W - 1`
    綜合邊框條件判定式：`if r == 0 or r == H - 1 or c == 0 or c == W - 1:`
  - **其他所有內部格子**：印出一個空格 `' '`（透明玻璃）！
  - 每一列印完後，在外層執行換行 `print()`。

**常見錯誤陷阱：**  
- **誤把 `or` 寫成 `and`**：如果寫成 `and`，代表必須同時是首列又是末列（只有高度為 1 時才會成立），導致圖形中間完全碎裂。
- **空白與星號寬度不對齊**：印星號時如果加了空格（如 `* ` 佔 2 格），那麼內部的空格也必須印 2 格（`'  '`），否則右邊框會嚴重歪斜錯位！

**APCS 考試實務建議：**  
在 APCS 矩陣與地圖模擬題中，「邊界加框法（Padding）」與「合法走訪邊界防禦（$0 \le r < H$ and $0 \le c < W$）」是兵家必爭之地。空心矩形的邊框判定，正是邊界條件思維最經典的啟蒙實戰！

````python
# [2] Code 範例區：空心矩形排版——邊框四界（首末列、首末欄）判定
print("=== 空心矩形排版（高度 H = 4, 寬度 W = 6）===")

H = 4
W = 6

for r in range(H):
    for c in range(W):
        # 邊界判定：首列、末列、首欄、末欄印星號，其餘內部印空格
        if r == 0 or r == H - 1 or c == 0 or c == W - 1:
            print("*", end="")
        else:
            print(" ", end="")
    print()  # 列尾換行

print("=== 空心矩形排版完成！ ===")
````

### 📐 6-6-3 靠左直角三角形：1 ~ N 顆星號的規律展開

**生活故事：貼牆堆疊的樂高階梯**  
請初學同學想像你在房間的牆角玩樂高積木。  
你把積木緊緊貼著左邊的牆面往上堆疊：  
最上面第 1 層只放 1 塊積木；  
第 2 層緊貼著左牆放 2 塊積木；  
第 3 層放 3 塊積木……到了最底層第 $N$ 層，整整齊齊放了 $N$ 塊積木！  
從側面看過去，左邊是一條垂直筆直的直線，斜邊則一路向右下方平緩延伸，這就是最經典的「靠左直角三角形」！

**運作機制——列號與星號數量的一對一對應：**  
靠左直角三角形是所有幾何排版中最直觀、最純粹的原型圖案。  
- **外層迴圈**：控制「垂直層數」，列號 $r$ 從 1 走到 $N$（代表第 1 列到第 $N$ 列）；
- **內層迴圈**：控制「水平星號」，觀察每一列的星號數量：
  - 第 1 列有 1 顆星；
  - 第 2 列有 2 顆星；
  - 第 3 列有 3 顆星；
  - 第 $r$ 列正好有 **$r$ 顆星號**！
- 程式核心結構：
  ```python
  for r in range(1, N + 1):      # 外層 r 代表第幾列
      for c in range(r):         # 內層跑 r 次
          print('*', end='')
      print()                    # 列尾換行
  ```
- **總星號數量公式**：整座三角形的星號總數，正好就是著名的高斯等差級數求和：$1 + 2 + 3 + \dots + N = rac{N(N+1)}{2}$！

**常見錯誤陷阱：**  
- **從 0 開始造成的幽靈空白行**：如果外層寫 `for r in range(N):`（$r$ 走訪 0 到 $N-1$），內層寫 `for c in range(r):`，那麼當 $r = 0$ 時，內層迴圈一次都不會跑，直接換行！這會導致三角形頂端平白多出一個空行，讓整個圖形的高度變成 $N+1$ 行！
- **忘記加 `end=''`**：預設的換行沒有取消，導致星號垂直掉落變成直線。

**APCS 考試實務建議：**

### 📉 6-6-4 倒直角三角形：N ~ 1 顆星號的遞減走訪

**生活故事：遊樂園的重力溜滑梯與倒數沙漏**  
想像我們在水上樂園玩高空滑水道。  
最頂端的平台最寬廣，可以同時容納許多人；隨著滑水道一層層往下俯衝，水道越來越窄，直到最後一個出口只剩下一道窄窄的水流！  
又例如倒數計時的沙漏，上層一開始裝滿了 $N$ 顆星塵，每一秒過去，頂端就少掉一顆，星塵數量呈現 $N, N-1, N-2, \dots, 1$ 規律遞減，這就是「倒直角三角形」！

**運作機制——兩種經典的遞減代數思維：**  
要印出星號數量逐列減少的倒三角形，在程式設計中有兩種完全等價但視角不同的經典寫法：
1. **第一種：外層倒數法（負步進迴圈）**  
   直接讓外層迴圈從 $N$ 倒數回 1：
   ```python
   for r in range(N, 0, -1):   # r 依序為 N, N-1, ..., 1
       for c in range(r):      # 內層直接印 r 顆星！
           print('*', end='')
       print()
   ```
   這種寫法極度直觀：外層變數 $r$ 就直接代表「這一列要印幾顆星」！
2. **第二種：代數轉換法（正向外層 + 差值轉換）**  
   外層仍然維持傳統的順序走訪 $r = 0, 1, \dots, N-1$；  
   但透過代數關係式將內層個數轉換為 **$N - r$**！
   - 當 $r = 0$ 時，印 $N - 0 = N$ 顆星；
   - 當 $r = 1$ 時，印 $N - 1$ 顆星；
   - 當 $r = N - 1$ 時，印 $N - (N - 1) = 1$ 顆星！

這兩種思維在各類演算法中各有千秋。熟練在「負步進」與「代數差值 $N - r$」之間自如切換，能大幅拓寬你的解題視野！

**常見錯誤陷阱：**  
- **負步進忘了第三個參數 `-1`**：寫成 `range(N, 0)`。因為預設步進是 $+1$，從 $N$ 到 0 用正步進永遠走不到，迴圈直接形同虛設、一次都不跑！
- **結束條件差一錯誤**：寫成 `range(N, 1, -1)`，結果最後只跑到 2，漏掉了最後一顆星號的那一列！

**APCS 考試實務建議：**  
在動態規劃（DP）由後往前更新狀態、或是倒序走訪二維網格時，倒序遞減走訪是高頻出現的操作，務必將 `range(N, 0, -1)` 的邊界刻進骨子裡！

````python
# [2] Code 範例區：倒直角三角形的兩種寫法對比展示
N = 4

print("=== 寫法一：負步進外層迴圈（直觀倒數）===")
for r in range(N, 0, -1):  # r 依序為 4, 3, 2, 1
    for c in range(r):
        print("*", end="")
    print()

print()
print("=== 寫法二：代數差值轉換法（N - r）===")
for r in range(N):  # r 依序為 0, 1, 2, 3
    stars = N - r   # 代數計算：4 - r
    for c in range(stars):
        print("*", end="")
    print()
````

**運作機制——雙重平行內迴圈的完美配合：**

讓我們在草稿紙上分析高度 $N = 4$ 時，每一列的空間分配：
- 列號 $r = 1$：需要 **3 個空格**，**1 顆星號**（$3 + 1 = 4$）
- 列號 $r = 2$：需要 **2 個空格**，**2 顆星號**（$2 + 2 = 4$）
- 列號 $r = 3$：需要 **1 個空格**，**3 顆星號**（$1 + 3 = 4$）
- 列號 $r = 4$：需要 **0 個空格**，**4 顆星號**（$0 + 4 = 4$）
看到了嗎？這項代數規律無比優雅：
- **前置空格數**：$spaces = N - r$！
- **星號數量**：$stars = r$！

**程式架構——平行雙內迴圈（Brothers Loop）：**  
外層迴圈每走一列，裡面依序由兩個**平級的兄弟內迴圈**接力完成：
1. **第一個內層迴圈**：跑 $N - r$ 次，專門印空格 `print(' ', end='')`！
2. **第二個內層迴圈**：跑 $r$ 次，專門印星號 `print('*', end='')`！
3. **外層負責換行**：兩個內層都跑完後，執行一次 `print()`！

**常見錯誤陷阱——誤寫成三重巢狀迴圈（致命慘劇）：**  
初學者常手滑把「星號迴圈」縮排到了「空格迴圈」的肚子裡！這導致每印一個空格就跑整批星號，整個畫面瞬間崩壞成外星密碼。記住：空格迴圈與星號迴圈是**平起平坐的兄弟**，縮排層級必須完全一致！

**APCS 考試實務建議：**  
「先印前置空白、再印實體內容」的排版模型，是對稱金字塔、菱形、以及各式縮排對齊的根本核心，務必親手將空格與星號的代數推導化為肌肉記憶！

````python
# [2] Code 範例區：靠右直角三角形——前置空格與星號雙內迴圈示範
print("=== 靠右直角三角形（高度 N = 4）===")

N = 4
for r in range(1, N + 1):
    spaces = N - r  # 代數推導：空格數量
    stars = r       # 代數推導：星號數量
    
    # 兄弟迴圈 1：專職印出前置空格
    for s in range(spaces):
        print(" ", end="")
        
    # 兄弟迴圈 2：專職印出星號（與兄弟迴圈 1 同一縮排層次）
    for c in range(stars):
        print("*", end="")
        
    # 該列任務全數達成，換行！
    print()

print("=== 靠右對齊排版完成！ ===")
````

### 🔺 6-6-6 正三角形與金字塔：奇數星號（2i - 1）與左右對稱排版

**生活故事：古埃及的雄偉金字塔與聖誕樹尖端**  
請初學同學閉上眼睛，想像古埃及吉薩高地上的大金字塔，或者客廳裡擺放的聖誕樹。  
從正前方看過去，它的左右兩側完全對稱：  
塔頂尖端只有一個中心點，往下每一層均勻向左右兩邊同步擴展！  
為什麼它能保持完美對稱？因為每一層的星號數量，都必須是**「奇數（1, 3, 5, 7, ...）」**！唯有奇數，才能擁有一個正中央的「軸心」，讓左右兩側掛上一樣多的星號！

**運作機制——金字塔的兩大代數核心公式：**  
設金字塔的總高度為 $N$，列號記為 $i$（從 1 到 $N$）：
1. **前置空格數（置中關鍵）**：  
   為了讓塔尖對齊在正中央，頂層（第 1 列）需要最多的空格，隨著層數往下，空格逐列遞減！  
   公式：**$spaces = N - i$**！
2. **星號數量（奇數數列）**：  
   第 1 列有 1 顆星（$2 	imes 1 - 1 = 1$）；  
   第 2 列有 3 顆星（$2 	imes 2 - 1 = 3$）；  
   第 3 列有 5 顆星（$2 	imes 3 - 1 = 5$）；  
   第 $i$ 列正好有 **$2i - 1$ 顆星號**！

**排版小秘密（右側空格完全不需要印！）：**  
很多初學者會困惑：「老師，星號左邊印了空格，那星號右邊要不要印空格對稱回來？」  
答案是：**完全不需要！**  
因為螢幕在每一列印完星號後，只要呼叫 `print()` 換行，游標就會直接跳到下一行的最左端！右側本來就是一片虛無，不需要白白浪費電腦時間去印看不見的空格！

**常見錯誤陷阱：**  
- **誤把奇數公式寫成 $2i + 1$**：如果寫成 $2i + 1$，當 $i = 1$ 時第 1 列就印了 3 顆星，塔尖直接被削平變成平頭梯形！
- **計算空格時誤用除法產生小數**：不要寫 `(width - stars) / 2`，單斜線除法會產生 float 導致 range 報出 TypeError！請直接使用整數公式 $N - i$。

**APCS 考試實務建議：**

### 💎 6-6-7 菱形與沙漏形：上下半部對稱拆解與組合拼裝心法

**生活故事：璀璨的南非鑽石與神奇的時間沙漏**

鑽石的形狀是一個標準的「菱形（Diamond）」；  
而沙漏則是上下兩頭寬廣、中間細腰緊縮的「沙漏形（Hourglass）」。  
初學者初次看到菱形時，往往會倒吸一口氣：「天啊，這又要遞增、又要遞減、空格還要忽多忽少，一個迴圈怎麼可能搞得定？」  
別慌！在演算法大師的眼中，菱形根本不需要用複雜的單一迴圈硬寫，我們只需要施展**「分而治之（Divide and Conquer）」**的心法：在最寬的腰線處「一刀切開」！

**運作機制——分段拼裝心法（上半部 + 下半部）：**  
把複雜大問題切分成兩個已知的小問題，這就是程式設計最極致的優雅！  
設菱形的半徑大小為 $N$（例如 $N = 3$）：
1. **上半部（包含中間最長腰線）**：  
   這不就是我們剛剛在 6-6-6 剛學會的「正三角形金字塔」嗎？  
   高度為 $N$，列號 $i$ 從 1 到 $N$：
   - 空格數 $= N - i$
   - 星號數 $= 2i - 1$
2. **下半部（收尾尖端）**：  
   這就是一個「倒金字塔」！  
   請特別注意：**下半部的高度只有 $N - 1$ 行！**（因為中間最寬的那一列已經在上半部的最後一行印過了，下半部絕不能再印一次！）  
   讓倒數列號 $i$ 從 $N - 1$ 一路倒數回 1：
   - 空格數 $= N - i$
   - 星號數 $= 2i - 1$
透過「兩個順序執行的獨立頂層迴圈」，分別負責上半部與下半部，璀璨的菱形立刻就被我們無痛拼裝出來了！  
同理，如果我們先印倒金字塔（上半部），接著印正金字塔（下半部），就是經典的沙漏形！

**常見錯誤陷阱——雙下巴慘劇（腰線重複）：**  
最常見的失分點，就是下半部也跑了 $N$ 行，導致中間最寬的那一列被硬生生重複印了兩次！菱形中間多了一圈肥肥的贅肉，破壞了完美的幾何對稱。

**APCS 考試實務建議：**  
分段處理（Piecewise Processing）是面對 APCS 各類複雜題型（如分段計費、分段函數、分多階段模擬）不可或缺的解題哲學。學會將視覺圖形拆為獨立階段，你的程式碼結構將永遠保持清晰！

````python
# [2] Code 範例區：菱形分段拼裝示範（N = 3，共 2*N - 1 = 5 列）
print("=== 璀璨鑽石菱形排版（N = 3）===")

N = 3

# === 第一階段：上半部（包含中間最寬列，高度 N 行）===
for i in range(1, N + 1):
    # 前置空格
    for s in range(N - i):
        print(" ", end="")
    # 奇數星號
    for c in range(2 * i - 1):
        print("*", end="")
    print()

# === 第二階段：下半部（高度 N - 1 行，由 N-1 倒數至 1）===
for i in range(N - 1, 0, -1):
    # 前置空格
    for s in range(N - i):
        print(" ", end="")
    # 奇數星號
    for c in range(2 * i - 1):
        print("*", end="")
    print()

print("=== 菱形完美合體！ ===")
````

````python
# ==========================================
# [5] Code 挑戰題
# 任務說明：
# 中心曼哈頓距離漸變方陣：
# 給定奇數邊長 N = 5。中心點座標為 center = N // 2 = 2。
# 每個格子 (r, c)（r, c 皆從 0 到 N-1）的數值為其到中心點的曼哈頓距離：
# dist = abs(r - center) + abs(c - center)
# 請排版印出該 5x5 數值方陣（以 f"{dist:2d} " 格式化對齊），
# 並計算整張方陣所有曼哈頓距離的總和 total_dist_sum。
# （本題為自由挑戰題，無公開測資，請依題目情境自行思考並撰寫完整程式碼）
# ==========================================
# 請在下方撰寫你的程式碼：

N = 5
center = N // 2
total_dist_sum = 0

for r in range(N):
    for c in range(N):
        dist = abs(r - center) + abs(c - center)
        print(f"{dist:2d}", end=" ")
        total_dist_sum += dist
    print()

print(f"全方陣曼哈頓距離總和：{total_dist_sum}")
# 中心點(0距離): 1格
# 距離1: 4格 (4*1=4)
# 距離2: 8格 (8*2=16)
# 距離3: 8格 (8*3=24)
# 距離4: 4格 (4*4=16)
# 總和 = 0 + 4 + 16 + 24 + 16 = 60
assert total_dist_sum == 60, "挑戰題計算錯誤！"
print("挑戰題 6.6.8 驗證通過！")
````
