### 11.8.1 自頂向下拆題思維（Top-Down Decomposition）：先寫主流程骨架，再填細節函數

#### 1. 生活故事比喻：摩天大樓的鋼骨骨架 vs 砌磚填土
在建造一座現代摩天大樓時，專業建築工程總監絕不會一來到工地就蹲在地上砌第一塊紅磚。
如果一開始就埋頭砌磚，砌到第十層才發現地基歪了、電梯井沒留空間，整棟樓就只能推倒重來！
建築總監的第一步，是先在藍圖上規劃大樓骨架，立起巍峨的**「鋼骨大樑（Main Skeleton）」**：哪裡是主入口、哪裡是承重牆、哪裡是電梯井。
大骨架穩固確立之後，水電工、泥水工、木工再分別進入各個專屬樓層，填補各房間的磚瓦與管線。
在程式設計中，面對 APCS 繁雜的綜合大題時，這種**「先寫主流程 `solve()`，把複雜子任務先假想成一個個現成的魔法工具函數，最後再去逐一實現子函數」**的頂尖心法，就叫做**自頂向下拆題法（Top-Down Decomposition）**！

#### 2. 底層運作機制：由抽象到具體的模組化三步曲
- **第一步：宏觀抽象（寫出主幹）**  
  在主程式或 `solve()` 函數中，用最口語、最清晰的函數名稱把整個解題故事串聯起來：
  ```python
  def solve():
      data = read_input()                # 讀取並格式化輸入
      valid_data = filter_valid(data)     # 過濾合法項目
      ans = find_best_candidate(valid_data) # 計算最佳解答
      print_solution(ans)                # 輸出格式化結果
  ```
- **第二步：約定介面（輸入與回傳規格）**  
  確定每個子函數需要什麼參數（如：座標、數列），以及需要回傳什麼結果（如：布林值、極值數值）。
- **第三步：微觀落實（填補細節）**  
  將每個子函數當成一個個獨立的單一職責小黑盒子，逐一撰寫內部代碼。

#### 3. 初學者常見陷阱：「義大利麵條代碼（Spaghetti Code）」
初學同學最容易犯的毛病是：把輸入讀取、字串切割、多重判斷、邊界檢查、運算與輸出，全部攪成一團塞在一個長達 80 行的巨大迴圈裡！
這種代碼就像一盤糾纏不清的義大利麵，一旦答案算錯，你根本不知道是第幾行的哪一個變數出問題，除錯時只能茫然盯著螢幕嘆氣。

#### 4. APCS 實戰視野
自頂向下拆題法是征服 APCS 實作題第二級到第四級的必修武功。當你把一道大題目拆成 3 到 4 個各只有 5 行的小函數時，你的大腦一次只需要專注於一個微小的邏輯點，不僅編碼速度翻倍，更幾乎不可能寫出 Bug！

````python
# 範例 11.8.1：自頂向下拆題法範例——學期成績結算系統

# --- 第一層：主流程骨架（一目了然的業務邏輯）---
def process_semester_grades(raw_scores):
    # 1. 清洗數據
    clean_list = sanitize_scores(raw_scores)
    # 2. 篩選及格
    passed_list = extract_passing_scores(clean_list)
    # 3. 計算平均
    avg_score = compute_average(passed_list)
    return passed_list, avg_score

# --- 第二層：逐一落實各子函數 ---
def sanitize_scores(scores):
    """過濾負數與缺考異常值"""
    return [s for s in scores if 0 <= s <= 100]

def extract_passing_scores(scores):
    """篩選大於等於 60 分的及格名單"""
    return [s for s in scores if s >= 60]

def compute_average(scores):
    """計算平均分，防禦空串列除以零"""
    if not scores:
        return 0.0
    return sum(scores) / len(scores)

# 主程式呼叫
raw_data = [85, 40, -5, 95, 60, 105, 75]
passed, avg = process_semester_grades(raw_data)
print(f"有效及格成績清單: {passed}")
print(f"及格學生平均成績: {avg:.1f} 分")
````

### 11.8.2 必備輔助函式庫一：二維網格安全邊界檢查 `in_bound(r, c, H, W)`

#### 1. 生活故事比喻：懸崖邊緣的雷達電子防護圍籬
在所有二維網格遊戲中（例如迷宮尋寶、地圖探路、棋盤行走），探險機器人最常發生的慘劇就是「一腳踩空，摔出地圖邊界粉身碎骨」（引發 `IndexError: list index out of range`）！
想像你在探測機器人前進時，每次在上下左右四個方向移動前，都必須在程式碼裡重複寫一遍：
`0 <= nr < H and 0 <= nc < W`。
如果整道題目有 5 個地方需要移動探測，你就必須把這串又臭又長的不等式敲 5 次！只要其中有一次不小心漏打了 `<` 或把長寬 $H, W$ 打反，整個程式就會當場崩潰！
最優雅的做法，就是在工具箱裡放一個隨身專用的「雷達電子防護圍籬」——**`in_bound(r, c, H, W)`**！

#### 2. 底層運作機制：判定函數（Predicate Function）的黃金封裝
在 Python 中，將邊界檢查封裝為純函數（Pure Function）：
```python
def in_bound(r, c, H, W):
    """檢查座標 (r, c) 是否合法落在 H x W 網格內部"""
    return 0 <= r < H and 0 <= c < W
```
- **極簡高效**：只有一行，利用 Python 獨家的「連續比較式（Chained Comparisons）」，直接回傳布林值 `True` 或 `False`。
- **搭配方向向量使用**：
  ```python
  directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]  # 上下左右
  for dr, dc in directions:
      nr, nc = r + dr, c + dc
      if in_bound(nr, nc, H, W):
          # 安全讀取 grid[nr][nc]，絕無越界風險！
          ...
  ```

#### 3. 初學者常見陷阱：$H$（高度/列數）與 $W$（寬度/行數）對應混淆
初學者最常犯的粗心失誤：
- 把 $r$（Row 列數）跟 $W$（Width 寬度）比對；
- 把 $c$（Column 行數）跟 $H$（Height 高度）比對。
當網格是長方形（例如 $3 \times 5$）而非正方形時，這種混淆會引發隱蔽而致命的 WA 或 RE！
將其封裝進 `in_bound(r, c, H, W)` 之後，在函數內部寫好一次，整個程式其他地方只要無腦呼叫它，徹底杜絕粗心！

#### 4. APCS 實戰視野
APCS 歷屆實作題中，例如 e287（機器人的路徑）、g596（骨牌連線）、f313（人口遷移），幾乎所有涉及二維矩陣走訪的題目，都必須以 `in_bound` 作為第一道守護神。考場一開局先敲下這行，穩健拿分！

````python
# 範例 11.8.2：in_bound 安全網格邊界檢查與相鄰格子走訪

def in_bound(r, c, H, W):
    """二維網格安全邊界檢查判定函數"""
    return 0 <= r < H and 0 <= c < W

# 模擬 3x4 的地圖
grid = [
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 25, 35]
]
H = len(grid)
W = len(grid[0])

def get_neighbors(r, c):
    """安全獲取 (r, c) 周圍上下左右合法相鄰格子的數值"""
    neighbors = []
    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)] # 上下左右
    for dr, dc in directions:
        nr, nc = r + dr, c + dc
        if in_bound(nr, nc, H, W):
            neighbors.append(grid[nr][nc])
    return neighbors

# 測試邊角座標 (0, 0)
print(f"邊角 (0, 0) 的安全相鄰數值: {get_neighbors(0, 0)}") # 只有右與下
# 測試中心座標 (1, 1)
print(f"中心 (1, 1) 的安全相鄰數值: {get_neighbors(1, 1)}") # 上下左右皆有
````

### 11.8.3 必備輔助函式庫二：幾何與網格曼哈頓距離 / 歐式距離計算函數

#### 1. 生活故事比喻：曼哈頓棋盤街道的計程車計價器
在紐約曼哈頓城區，所有街道都是筆直工整的九宮格街區（Grid）。
計程車司機想要載客人從座標 $(x_1, y_1)$ 到達 $(x_2, y_2)$，計程車絕不可能像超人一樣斜向穿牆飛過去！司機必須先老老實實水平開 $|x_1 - x_2|$ 條街，再垂直開 $|y_1 - y_2|$ 條街。
這種沿著網格垂直橫向行進的總距離，就叫做**曼哈頓距離（Manhattan Distance）**！
而如果是一隻鳥在天空中直線飛過兩點，量出的直尺直線距離，就叫做**歐式距離（Euclidean Distance）**。

#### 2. 底層運作機制：幾何公式的極簡模組化封裝
在 APCS 競賽中，座標距離計算出現頻率極高，封裝成標準工具庫能隨時調用：
- **曼哈頓距離函數**：
  ```python
  def manhattan_dist(p1, p2):
      """計算 p1(x1, y1) 與 p2(x2, y2) 之間的曼哈頓距離"""
      return abs(p1[0] - p2[0]) + abs(p1[1] - p2[1])
  ```
- **歐式距離平方函數（防精度誤差神器）**：
  ```python
  def euclidean_dist_sq(p1, p2):
      """計算 p1 與 p2 之間直線距離的平方"""
      return (p1[0] - p2[0])**2 + (p1[1] - p2[1])**2
  ```

#### 3. 初學者常見陷阱：比大小時不必要地開根號
初學同學最常犯的重大失誤：在尋找「誰離原點最近」時，興沖沖地寫了 `dist = ((x1-x2)**2 + (y1-y2)**2) ** 0.5`。
**這在競賽中是大忌！**
- 開根號運算會將乾淨的整數轉化為**浮點數（float）**，隨之帶來二進位小數點的**精度誤差（Precision Error）**！在比對距離是否相等或比大小時，極微小的誤差就會導致 Wrong Answer！
- 競賽真理：若 $A^2 < B^2$，則必有 $A < B$！**在比較兩點距離遠近時，永遠只比較「距離的平方（dist_sq）」**，百分之百維持純整數運算，速度更快且精度絕對無懈可擊！

#### 4. APCS 實戰視野
外送員最短路徑配送、基地台最近涵蓋範圍、平面雷達目標鎖定等題型中，曼哈頓距離與歐式距離平方是不可或缺的解題基石。

````python
# 範例 11.8.3：曼哈頓距離與歐式距離平方的封裝與應用

def manhattan_dist(p1, p2):
    """計算兩點的曼哈頓距離 |x1-x2| + |y1-y2|"""
    return abs(p1[0] - p2[0]) + abs(p1[1] - p2[1])

def euclidean_dist_sq(p1, p2):
    """計算兩點歐式距離平方 (x1-x2)^2 + (y1-y2)^2 (純整數防誤差)"""
    return (p1[0] - p2[0])**2 + (p1[1] - p2[1])**2

# 實戰情境：尋找距離主角 (0, 0) 最近的補給站
player = (0, 0)
stations = [("補給站A", (3, 4)), ("補給站B", (1, 6)), ("補給站C", (2, 2))]

# 1. 曼哈頓距離評比
closest_manhattan = min(stations, key=lambda s: manhattan_dist(player, s[1]))
print(f"曼哈頓最近補給站: {closest_manhattan[0]} (距離: {manhattan_dist(player, closest_manhattan[1])})")

# 2. 歐式直線距離評比（比平方）
closest_euclidean = min(stations, key=lambda s: euclidean_dist_sq(player, s[1]))
print(f"直線最近補給站: {closest_euclidean[0]} (距離平方: {euclidean_dist_sq(player, closest_euclidean[1])})")
````

#### 3. 初學者常見陷阱：邊界漏防與開根號漏 +1
- **陷阱一：0 與 1 的判斷**  
  0 和 1 都**不是質數**！許多初學者忘記在第一行加上 `if n < 2: return False`，導致 1 被錯誤判為質數！
- **陷阱二：`range` 邊界漏寫 `+1`**  
  若寫 `range(2, int(n**0.5))`，當 $n=4$ 時，`int(4**0.5) = 2`，`range(2, 2)` 為空迴圈，4 就會被誤判成質數！務必寫為 `int(n**0.5) + 1`！

````python
# 範例 11.8.4：數論三大工具函數實作與綜合篩選

def is_prime(n):
    if n < 2: return False
    for d in range(2, int(n**0.5) + 1):
        if n % d == 0: return False
    return True

def is_palindrome_num(n):
    s = str(n)
    return s == s[::-1]

def count_divisors(n):
    if n <= 0: return 0
    cnt = 0
    for d in range(1, int(n**0.5) + 1):
        if n % d == 0:
            cnt += 1 if d * d == n else 2
    return cnt

# 實戰應用：找出 10 到 200 之間所有的「迴文質數」
pal_primes = [x for x in range(10, 200) if is_prime(x) and is_palindrome_num(x)]
print(f"10~200 之間的迴文質數: {pal_primes}")

# 統計 36 的因數個數
print(f"36 的因數總個數: {count_divisors(36)} (1,2,3,4,6,9,12,18,36 共 9 個)")
````

在組裝多個輔助函數時，初學者常犯的錯誤是：

#### 4. APCS 實戰視野
這就是 APCS 高分滿分選手的真實解題風貌：
先用 2 分鐘在最上方貼好預先鍛造好的黃金工具函數，再用 3 分鐘寫出骨架清晰的主程式，5 分鐘直接奪下 AC！

````python
# 範例 11.8.5：APCS 綜合模擬題實戰——挑選最強能量基地台

# === 武器庫：黃金輔助函式庫 ===
def in_bound(r, c, H, W):
    return 0 <= r < H and 0 <= c < W

def manhattan_dist(p1, p2):
    return abs(p1[0] - p2[0]) + abs(p1[1] - p2[1])

def is_prime(n):
    if n < 2: return False
    for d in range(2, int(n**0.5) + 1):
        if n % d == 0: return False
    return True

# === 主流程：乾淨俐落的篩選與極值維護 ===
def find_best_power_station(grid, max_dist):
    H = len(grid)
    W = len(grid[0])
    hq_loc = (0, 0)
    best_loc = None
    best_power = -1
    
    for r in range(H):
        for c in range(W):
            power = grid[r][c]
            # 一行流整合三大條件審查！
            if is_prime(power) and manhattan_dist(hq_loc, (r, c)) <= max_dist:
                if power > best_power:
                    best_power = power
                    best_loc = (r, c)
                    
    return best_loc, best_power

# 主程式測試
grid_map = [
    [ 4, 11,  8],
    [13, 20, 29],
    [ 6, 17, 31]
]
loc, pwr = find_best_power_station(grid_map, max_dist=2)
print(f"總部距離 2 以內之最強質數基地台座標: {loc}，能量值: {pwr}")
# (0, 1) 的 11 (距離 1)、(1, 0) 的 13 (距離 1)、(1, 1) 的 20 非質數、(2, 0) 的 6 非質數
# 最強符合條件者為 (1, 0)，能量 13！
````

### 11.8.6 模組化除錯策略：單獨測試子函數（Unit Test 雛形）快速定位 Bug

#### 1. 生活故事比喻：火箭發射前的零件獨立通電測試
工程師在組裝一枚太空火箭時，絕不可能把幾萬個零件全部焊死在一起之後，才第一次按下點火發射鈕！
萬一火箭在半空中爆炸，在濃煙火海中，工程師根本無法知道究竟是燃料閥門卡死、陀螺儀失靈、還是點火器故障。
航太工業的唯一標準是：**「零件獨立通電測試」**。
在裝上火箭之前，先在工作台上單獨給燃料閥門通電 100 次確認開關正常；單獨測試陀螺儀感應器；單獨點火測試噴嘴。所有模組拿到合格驗證標籤後，才正式組裝。
在程式設計中，這套救命心法就叫做**單元測試（Unit Testing）**！

#### 2. 底層運作機制：考場 3 秒鐘除錯定位 SOP
在 APCS 考場上，最慘烈的畫面就是：送出程式碼後，系統無情顯示 Wrong Answer (WA)，選手慌了手腳，開始在 100 行的主程式中到處盲目亂改，結果越改越錯，最後整組崩潰。
學會模組化之後，你的除錯 SOP 應該是降維打擊式的**「隔離偵錯」**：
1. **先測底層工具**：不要去跑完整測資，直接在代碼最下方，對每個子函數餵入最極端的測試邊界：
   ```python
   # 快速單元測試
   print("測試 is_prime(1):", is_prime(1))       # 正確應為 False
   print("測試 is_prime(2):", is_prime(2))       # 正確應為 True
   print("測試 in_bound(-1, 0):", in_bound(-1, 0, 3, 3)) # 正確應為 False
   print("測試 in_bound(3, 3):", in_bound(3, 3, 3, 3))   # 正確應為 False
   ```
2. **只要 3 秒鐘**：按下執行，若發現 `is_prime(1)` 竟然印出了 `True`，嫌疑犯立刻被當場逮捕！
3. **精準修復**：你只要花 5 秒鐘修復 `is_prime` 中的邊界判斷，整道大題目立刻滿分通過！

#### 3. 初學者常見陷阱：滿螢幕 print 的「資訊淹沒」
初學者在除錯時，常在主迴圈裡加滿 `print(r, c, x, y, temp, ans)`。
一執行下去，螢幕瞬間刷出好幾千行雜亂數字，眼睛看花了也找不到哪裡算錯。
記住：**「除錯要從最小的單元測起，由底向上隔離排查！」**

#### 4. APCS 實戰視野
養成「寫好一個子函數，順手在下方 print 一行驗證」的肌肉記憶。這不僅能大幅提升代碼的一次通過率（AC Rate），更能讓你在考場高壓緊張的倒數時刻，擁有強大而穩定的心理素質！

````python
# 範例 11.8.6：極速單元測試與自我診斷腳本演示

def is_prime(n):
    # 假設某同學寫錯了，漏掉了小於 2 的檢查
    # if n < 2: return False  <-- 故意註解掉這行重現 bug
    for d in range(2, int(n**0.5) + 1):
        if n % d == 0: return False
    return True

# === 考場極速單元診斷 SOP ===
def run_unit_tests():
    print("=== 正在啟動子模組極速單元檢測 ===")
    test_cases = [
        (1, False, "邊界值 1 絕不是質數"),
        (2, True,  "最小偶數質數 2"),
        (4, False, "合數 4"),
        (17, True, "經典質數 17")
    ]
    for val, expected, desc in test_cases:
        actual = is_prime(val)
        status = "✅ 通過" if actual == expected else "❌ 警報抓出 BUG！"
        print(f"  檢驗 is_prime({val:2d}) -> 結果: {str(actual):5s} | 預期: {str(expected):5s} [{status}] ({desc})")

run_unit_tests()
print("\n💡 看到紅燈警報後，立刻知道是 is_prime(1) 出了問題，精準修復！")
````
