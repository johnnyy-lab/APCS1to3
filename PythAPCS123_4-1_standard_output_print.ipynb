### 📢 4-1-1 print() 的初體驗——電腦的發聲大喇叭與字串輸出

想像電腦是一個超級聰明的機器人，它肚子裡有無數個記憶體置物櫃，運算速度比光速還快。但是，機器人如果沒有嘴巴，就算他在肚子裡算出了宇宙大秘密，外面的我們也什麼都看不到！而 **`print()` 函數** 就是接在機器人嘴巴上的「高分貝大喇叭」。只要你把想說的話交給 `print()`，喇叭就會把這些話原封不動地傳送到電腦螢幕（標準輸出 stdout）上讓我們讀取。這也是全球每一位程式設計師在學習新語言時，寫出的第一個經典程式——向全世界打招呼！

**運作機制與底層原理：**
當我們在小括號裡放進用引號包起來的文字時，例如 `print("Hello, World!")`，引號內部的文字在 Python 中被稱為「字串（String）」。引號的作用就像是「文字專用包裹箱」，用來告訴 Python 直譯器：「請把箱子裡面的這段話當成純文字訊息，直接送給大喇叭播報，不要試圖把它當成指令去執行！」
在 Python 中，你可以使用成對的雙引號 `"..."`，也可以使用成對的單引號 `'...'`，兩者的效果完全一模一樣，只要前後一致即可。

**常見錯誤陷阱（初學者必讀三大地雷）：**
1. **大小寫敏銳陷阱（`Print` vs `print`）**：在 Python 魔法王國中，大小寫是住在不同套房的人！內建的輸出函數一律都是全小寫的 `print()`。如果你不小心把首字寫成大寫 `Print()`，電腦會完全認不出它是誰，立刻報出 `NameError: name 'Print' is not defined`（找不到名為 Print 的東西）！
2. **引號遺漏陷阱**：如果你寫了 `print(Hello)` 卻忘了加引號，電腦不會知道這是文字，而是會以為你要去置物櫃尋找一個名叫 `Hello` 的變數名牌貼紙，如果找不到就會引發 `NameError`！
3. **括號失蹤或未成對**：括號必須成雙成對 `()`，漏掉右括號會導致 `SyntaxError: '(' was never closed`。

**APCS 考試實務建議：**
在 APCS 與各類線上題庫（Online Judge, OJ）中，自動評判系統完全是透過「標準輸出」來檢查你的答案。在 Python 中，你所有辛苦計算出來的成果，最終都必須依靠 `print()` 送出。若多了一個錯字、大小寫不符或少了一個標點符號，OJ 都會毫不留情地給出 **Wrong Answer (WA)**！因此，精準掌控輸出的每一個字元，是程式競技最基礎的第一步。


````python
# [2] Code 範例區：展示 print() 基本文字輸出功能

# 範例 1：使用雙引號輸出經典英文招呼語
print("Hello, World!")

# 範例 2：使用單引號輸出親切的中文問候
print('歡迎來到 Python 與 APCS 程式世界！')

# 範例 3：將文字先存入變數名牌中，再交給大喇叭播報
greeting_message = "你好，親愛的程式小勇士！"
print(greeting_message)
````

### 🔢 4-1-2 print() 輸出數值與即時算術——讓喇叭播報運算結果

大喇叭不僅能把現成的話講出來，還自帶一台「即時計算器」！想像大喇叭的麥克風孔前裝了一台微型收銀機，當你把「10 塊錢的鉛筆加 15 塊錢的橡皮擦」的算式丟進麥克風時，大喇叭不會呆呆地唸出「10 加 15」，而是在肚子裡「叮！」一聲算出 25，然後直接向全場大喊：「25」！

**運作機制與底層原理：**
`print()` 的小括號在 Python 中具有「求值優先權」。當我們在括號裡放入數值或算術運算式時（例如 `print(15 + 25)`），Python 的直譯器會遵守先前的運算順序規則：
1. **優先計算小括號內部的算術表示式（Expression Evaluation）**：將 `15 + 25` 求值得到單一數值 `40`。
2. **傳遞求值結果**：接著將最終的求值結果 `40` 傳遞給 `print()` 函數。
3. **轉換輸出**：`print()` 將整數 `40` 轉換為螢幕文字輸出。
無論是整數（`int`）、浮點數（`float`），或是複雜的四則混合運算（例如 `100 - 4 * 5`），甚至先前學過的次方 `2 ** 10` 與整除 `17 // 3`，都可以直接放進 `print()` 的括號中即時播報！

**常見錯誤陷阱（文字 vs 算式的致命混淆）：**
這是許多國一初學者最容易掉入的陷阱：
- `print("10 + 20")`：因為有引號，Python 把它當成純文字字串，螢幕上會一字不差地印出 `10 + 20`。
- `print(10 + 20)`：沒有引號，Python 把它當成數學運算式，螢幕上會印出計算結果 `30`！
請務必分清楚：你想印的是「算式題目本身」，還是「算式的答案」！

**APCS 考試實務建議：**
在 APCS 第一級與第二級的基礎題目中，許多題目都是單純的公式代入與數值計算（例如計算購買總額、找零計算、求平均或全距等）。掌握在 `print()` 內部直接進行運算，或者將算好的變數直接交給 `print()`，能夠大幅精簡程式碼行數，讓邏輯一目了然！


````python
# [2] Code 範例區：展示 print() 輸出數值與即時算術運算

# 範例 1：直接輸出整數與浮點數
print(100)
print(3.14159)

# 範例 2：在 print() 內部即時進行四則混合運算
print(25 + 75)          # 先算 25+75，印出 100
print(100 - 8 * 5)      # 先乘除後加減，印出 60

# 範例 3：印出整除商數與取餘數
print(17 // 5)          # 印出整除商數 3
print(17 % 5)           # 印出餘數 2

# 範例 4：比較「加了引號」與「沒加引號」的差別
print("5 + 3")          # 印出純文字：5 + 3
print(5 + 3)            # 印出計算結果：8
````

### 🪄 4-1-3 逗號 `,` 串聯多個項目——預設單一空格隔開的神奇魔術

想像大喇叭不只是一次只能吃一口話的單音喇叭，它還有一條「多節車廂托盤」！你可以一口氣把好幾個不同的物品（提示文字、數字、計算結果）放在托盤的各個小隔間裡，每個隔間之間用「半形逗號 `,`」隔開。當大喇叭把托盤吸進去準備廣播時，它會非常貼心地在每個隔間物品之間自動放上一塊「軟綿綿的半形空白海綿」，整整齊齊、清清楚楚地依序播報出來！

**運作機制與底層原理：**
在 Python 中，傳遞給函數的資料稱為「引數（Arguments）」。`print()` 允許我們傳入任意多個引數，各項目之間必須以「逗號 `,`」分隔，例如：
`print("國文:", 95, "數學:", 100)`
它的底層運作機制包含兩個極其強大的特性：
1. **預設單一空白間隔**：Python 在印完第一個項目後，會**自動輸出一個半形空白字元 `' '`**，然後再印下一個項目。因此你不需要在文字字串裡面自己手動狂敲空格！
2. **跨型態自動相容**：在第 3-1 節我們學過，如果用字串加法 `+` 串接文字與數字（例如 `"得分:" + 95`），直譯器會立刻報出 `TypeError: can only concatenate str (not "int") to str`；但在 `print()` 中使用逗號 `,` 分隔，Python 會自動將整數、浮點數等不同型態安全轉換為文字顯示，完全不需手動 `str()` 轉型！

**常見錯誤陷阱（標點符號的致命細節）：**
1. **中英文輸入法全形逗號陷阱**：在寫程式碼時，分隔項目的逗號必須是英文半形逗號 `,`。如果輸入法忘記切換，打成中文全形逗號 `，`，Python 會直接崩潰並跳出 `SyntaxError: invalid character '，'`！
2. **誤以為項目會黏在一起**：有些同學以為 `print("A", "B")` 會印出 `AB`，但實際上它會印出 `A B`（中間有一個空白）！如果想要黏在一起或改成別的符號，必須等到下一節（4-2）學到 `sep` 參數才能調整。

**APCS 考試實務建議：**
在 APCS 與各類程式競賽中，超過 80% 的輸出要求都是：「輸出兩數，中間以一個空白隔開」（例如 APCS c294 題要求由小到大輸出三邊長以空白隔開）。此時，只要寫 `print(a, b, c)`，就天然完全符合題目的格式要求，是最常用、最穩健的競賽基本功！


````python
# [2] Code 範例區：展示 print() 逗號串聯多個項目與預設空白間隔

# 範例 1：串聯提示文字與數值變數
player_name = "Alex"
player_score = 98
print("玩家姓名:", player_name, "最終得分:", player_score)
# 輸出：玩家姓名: Alex 最終得分: 98

# 範例 2：串聯多個數字（各數字間天然以單一空白隔開）
side_a = 3
side_b = 4
side_c = 5
print(side_a, side_b, side_c)
# 輸出：3 4 5

# 範例 3：文字、變數與即時算術混合輸出
item = "蘋果"
price = 25
count = 6
print(item, "共", count, "顆，總計", price * count, "元")
# 輸出：蘋果 共 6 顆，總計 150 元
````

### ↵ 4-1-4 print() 的自動換行特質——每喊一次話就移至新的一行

想像你手裡拿著一台傳統的打字機，每當你敲完一行字、按下右邊的「換行板手」，打字機的滾筒就會「喀噠！」一聲，自動把紙張往上捲一行，並且把打字頭推回最左邊的開頭位置。在 Python 中，每一次執行 `print()`，大喇叭講完話之後，都會自動幫你按下這根「換行板手」！

**運作機制與底層原理：**
`print()` 在預設情況下，每當把小括號內的所有內容播報完畢後，都會在最後面默默追加一個**不可見的「換行符號（Newline Character，在電腦內部表示為 `\n`）」**。
這意味著：
- 當前 `print()` 輸出的內容會在當前行印出。
- 印完後游標立即跳到「下一行的第 1 個字元位置」。
- 因此，如果你寫了三個連續的 `print()` 指令：
  ```python
  print("第一行")
  print("第二行")
  print("第三行")
  ```
  它們絕對不會黏在同一行，而是會整齊劃一地排成上下三行！

**常見錯誤陷阱（同行的幻想迷思）：**
許多初學程式的同學常誤以為：「我寫在不同行的指令，會不會印在同一行？」或是「我該怎麼讓兩次印出的東西接在同一行？」
請記住：**一行 `print()` 就是獨佔一整行**！在預設情況下，Python 不會讓兩次呼叫的內容接在同一行。如果想取消自動換行，必須仰賴 4.3 節將正式登場的 `end` 參數。

**APCS 考試實務建議：**
線上評判系統（OJ）對「行數（Line Count）」的檢查是極其嚴格的。若題目規定：「輸出第一行包含兩個整數，第二行包含一個答案」，這就明確提示你必須依序呼叫兩次 `print()`！如果少呼叫一次導致全部印在同一行，OJ 會當場判定為 WA；如果多呼叫了多餘的換行，也可能會判定格式錯誤。因此，每一次 `print()` 伴隨的「換行特質」，是每位選手必須烙印在腦海中的直覺本能！

### 📄 4-1-5 不帶引數的 print()——空白行與段落視覺排版術

想像你在讀一本厚厚的故事書，如果整本書幾萬個字全密密麻麻擠成一團，連一個段落空行都沒有，眼睛一定會酸得受不了！這時候，作家都會在章節或段落之間留出一個「空白行」，讓讀者深吸一口氣、眼睛休息一下。在 Python 的程式世界裡，大喇叭也很懂這種排版美學，當你呼叫它卻什麼話都不交代時，它就會優雅地在畫面上空出整整一行！

**運作機制與底層原理：**
當我們寫下 `print()` 並且**小括號裡面空空如也、完全不傳入任何引數**時，Python 會執行什麼動作？
正如前一小節所學：`print()` 每次執行的標準動作就是「印出內容 ＋ 追加換行」。
現在既然內容是空的（沒有字串也沒有數值），那麼大喇叭做的事情就只剩下最後一步：**印出一個換行符號（`\n`）**！
因此，螢幕游標會直接往下跳一行，在兩段文字之間留下一個完完整整、乾乾淨淨的「空白行（Blank Line）」。

**常見錯誤陷阱與 APCS 考場生死門：**
1. **誤用引號包空白**：`print()` 與 `print(" ")` 在畫面上看起來都是空行，但後者其實多輸出了一個不可見的半形空白字元！在一般的視覺閱讀中沒有影響，但在要求絕對純淨空行的自動評判情境中，可能造成字元比對差異。
2. **APCS 嚴格警訊（切勿畫蛇添足）**：在自己電腦上寫程式練習時，多印空行能讓版面美觀；**但在 APCS 實作題或 LeetCode、ZeroJudge 等競技題庫中，除非題目明確要求「每筆測資輸出後請輸出一行空白行」，否則絕對禁止擅自加上 `print()` 空行**！多印了沒要求的空白行，OJ 會當場宣判格式錯誤（Presentation Error）或直接判定 WA！

**實務排版總結：**
善用 `print()` 進行段落區隔，是設計主控台（Console）應用程式、文字遊戲介面與日常資料報表時不可或缺的排版技巧！


````python
# [2] Code 範例區：展示不帶引數的 print() 產生空行與段落區隔

# 範例 1：段落之間插入空行
print("【第一章：初入魔法森林】")
print("小勇士走進了幽暗的森林深處...")
print()  # 不放任何引數，輸出一個空白行
print("【第二章：遇見智慧精靈】")
print("精靈問道：『小勇士，你準備好學習 Python 了嗎？』")

# 範例 2：收據明細與結算區塊分割
print("--- 購物明細 ---")
print("商品 A: 100 元")
print("商品 B: 250 元")
print()  # 插入視覺空行
print("--- 總計結算 ---")
print("應付總額: 350 元")
````
