### 🐷 6-2-1 累加器模式（Accumulator）基礎——小豬撲滿的「歸零」與「逐回合投入」

**生活比喻：小豬撲滿存錢日記**  
想像你桌上有一個透明的小豬撲滿。月初第一天，撲滿裡面空空如也，餘額是 0 元（這叫做「歸零初始化」）。  
接下來的每一天，你依序往撲滿裡投錢：第一天投 10 元、第二天投 20 元、第三天投 30 元……  
每投一次，撲滿裡的最新金額就是「原本的金額 加上 今天投入的金額」！  
直到月底，你才會把撲滿拿起來看一眼總共存了多少錢（這叫做「迴圈結束後的最終結算」）！

在 Python 程式語言中，這種**「在迴圈前設為 0，每回合累加一個數值，迴圈結束後得到總和」的變數，就稱為「累加器（Accumulator）」**！
其核心標準寫法為：
```python
total = 0  # 步驟一：出發前必須先清空歸零
for i in range(1, n + 1):
    total += i  # 步驟二：每回合使用複合賦值運算子更新累加（total = total + i）
print("總和：", total)  # 步驟三：縮排退回最外層，只在最後印出最終總額
```

**初學者最容易掉入的兩大致命陷阱：**
1. **陷阱一：把歸零寫在迴圈肚子裡（內部歸零災難）**  
   若把 `total = 0` 寫在 `for` 縮排內部，每跑一輪撲滿就被倒空一次！跑完迴圈後，`total` 永遠只保留最後一輪加進去的數值，先前的努力全數蒸發！
2. **陷阱二：把結算輸出寫在迴圈肚子裡（過度縮排噪音）**  
   若把 `print(total)` 寫在縮排內，程式每加一次就會印一次過程，如果走訪 1000 回合就會洗版 1000 行，在 APCS 自動評判中會立刻被判定答案格式錯誤（WA）！

**APCS 考試實務建議：**
累加器變數名稱通常習慣取名為 `total`、`sum_val` 或 `ans`，請務必避免使用 Python 內建關鍵字 `sum` 當作變數名稱，否則會覆蓋掉內建的加總函數！

````python
# [2] Code 範例區：展示累加器求 1 到 5 連續整數總和

# 步驟一：迴圈前歸零初始化
total = 0
n = 5

# 步驟二：迴圈中逐回合更新累加（1 到 n，stop 設定為 n + 1）
for i in range(1, n + 1):
    total += i  # 等同於 total = total + i

# 步驟三：迴圈結束後，縮排退回最外層輸出最終總和
print("1 到", n, "的整數總和為：", total)  # 輸出 15
````

### 🎯 6-2-2 計數器模式（Counter）基礎——特定事件發生的「打勾計次機」

**生活比喻：球賽裁判的按鈕計分器**  
在排球或羽球比賽中，裁判手中握著一個金屬計次器。比賽開始前，計次器歸零顯示為 0。  
球員在場上激烈來回，裁判「並不是每一次球過網就按一下」，而只有在滿足特定條件時（例如：A 隊球落地得分），裁判才會按下按鈕「咔嗒」一聲，讓計數器上的數字加 1（`count += 1`）！  
整場比賽結束後，計數器上的數字就是「得分事件總共發生的次數」！

在程式設計中，這種**「在迴圈前初始化為 0，每當符合特定條件時就加 1（`count += 1`）」的變數，就稱為「計數器（Counter）」**！

**累加器 vs 計數器的本質差異（新手必備對照鏡）：**
- **累加器（Accumulator）**：`total += x` ➔ 關心的是**「數量有多大、數值總和是多少」**，每次加的是變數 `x` 本身的值！
- **計數器（Counter）**：`count += 1` ➔ 關心的是**「發生了幾次、有幾個符合資格」**，每次固定只加常數 `1`！

**語法架構標準骨架：**
```python
count = 0  # 步驟一：計數器事前歸零
for x in range(1, n + 1):
    if 滿足特定條件:  # 步驟二：檢驗條件
        count += 1  # 步驟三：條件成立，次數加 1
print("符合條件總次數：", count)  # 步驟四：迴圈結束後單一輸出
```

**常見錯誤陷阱：**
- **把 `count += 1` 誤寫成 `count += x`**：初學者常常頭腦打結，明明題目要求「統計有幾個奇數」，結果程式碼卻把奇數的值給加了進去，造成答案爆增！請牢記：算個數永遠只加 1！

**APCS 考試實務建議：**
計數器常命名為 `count`、`cnt` 或 `match_count`。在 APCS 各種過濾題（例如統計符合預算的商品件數、統計及格人數）中，計數器是最不可或缺的解題骨架。

````python
# [2] Code 範例區：展示計數器模式統計 1 到 20 之間能被 3 整除的數字個數

count_div_by_3 = 0

for num in range(1, 21):
    # 檢驗條件：能否被 3 整除（餘數為 0）
    if num % 3 == 0:
        count_div_by_3 += 1  # 滿足條件，次數加 1

print("1 到 20 之間能被 3 整除的數字共有：", count_div_by_3, "個")  # 6 個（3,6,9,12,15,18）
````

### 📥 6-2-3 結合使用者輸入的動態累加——連續讀入 n 筆測資的流式計算

**生活比喻：超商收銀員刷條碼**  
當你去便利商店買了 5 件商品，收銀員不會預先知道每件商品的價格。  
收銀員的操作流程非常標準：  
1. 收銀機總金額先歸零（`total = 0`）。
2. 拿起第 1 件商品，刷條碼（讀入價格），立刻加進總金額。
3. 拿起第 2 件商品，刷條碼，立刻加進總金額……
4. 5 件商品全部刷完後，收銀機螢幕亮出應付總金額！

在前面的章節中，我們走訪的數值都是由 `range()` 固定生成的數列。  
但在真實 APCS 考試中，絕大多數題目是：**「第一行給定資料筆數 $n$，接下來的 $n$ 行每行各有一個數值」**！  
面對這種未知資料，我們不需要使用串列（`list`）把全部數字存下來，而是可以運用**「流式處理（Streaming）」**的絕技：**邊讀入、邊累加、邊丟棄**！

**語法架構標準骨架：**
```python
n = int(input())  # 先讀入總筆數
total = 0  # 累加器歸零

for _ in range(n):  # 重複執行 n 次
    val = int(input())  # 每一輪讀入一個全新的數值
    total += val  # 即時投入累加器

print(total)  # 最終印出總和
```

**底層細節剖析：底線變數 `_` 的專業用法**  
請注意上面迴圈中的 `for _ in range(n):`。  
在 Python 業界慣例中，如果我們**只需要這個迴圈重複執行 $n$ 次，但根本不在乎當前的迴圈索引值**時，我們習慣使用單一底線 `_` 作為迴圈變數名稱！這清楚告訴讀代碼的人：「這裡只是在做次數控制，不需要用到變數值」。

**常見錯誤陷阱：**
- **把 `n = int(input())` 寫進迴圈內部**：初學者若不小心把讀取 $n$ 也放進迴圈，程式每回合都會停下來等待輸入 $n$，導致輸入順序徹底錯亂崩潰！

**APCS 考試實務建議：**
「讀一筆、算一筆」的流式處理只佔用極微量的記憶體（空間複雜度為常數 $O(1)$），即使測資有 100 萬行，程式也不會因為記憶體爆掉而當機！

````python
# [2] Code 範例區：展示連續讀入 n 個數字並即時計算總和
# 模擬輸入：假設先輸入筆數 3，接著依序輸入 10, 20, 30

total_score = 0
n = int(input())  # 讀入筆數

for _ in range(n):
    score = int(input())  # 逐行讀入整數
    total_score += score  # 即時累加

print("總得分為：", total_score)
````

### ⚖️ 6-2-4 累加與計數的協同作戰——條件加總、求平均值與防除零防禦

**生活比喻：班級及格學生統計與平均分計算**  
期中考結束後，老師請小幫手統計班上「及格同學的表現」：  
- 小幫手手中拿著兩個計分本：一個是「及格總分小豬撲滿（`pass_total = 0`）」，另一個是「及格人數計數器（`pass_count = 0`）」。  
- 念到一位同學的分數時，如果大於等於 60 分：撲滿裡加上這位同學的分數（`pass_total += score`），計數器也按一下加 1（`pass_count += 1`）！  
- 若未滿 60 分，兩邊都不動作。  
- 當全班成績登記完畢，及格平均分數就是 $\frac{\text{及格總分}}{\text{及格人數}}$！

將「累加器」與「計數器」同時放入同一個迴圈中進行**雙變數協同維護**，是統計學運算的核心基本功。

**最關鍵的系統防禦：防止除以零（ZeroDivisionError）！**  
如果這場考試超級艱難，全班「沒有任何一位同學及格」（即 `pass_count == 0`）呢？  
如果此時你直接執行：
```python
avg = pass_total / pass_count  # 致命崩潰！
```
Python 會立刻當場崩潰，擲出 `ZeroDivisionError: division by zero`，在線上評判系統中拿到刺眼的 `RE`（Runtime Error）！  
因此，**任何計算平均值的程式碼前方，都必須加上防禦性判斷**：
```python
if pass_count > 0:
    avg = pass_total / pass_count
    print("平均分：", avg)
else:
    print("無合格資料")
```

**常見錯誤陷阱：**
- **平均值整數除法 vs 浮點數除法**：注意題目要求是「精確小數平均」還是「無條件捨去整數平均」。精確小數請使用單斜線 `/`；若題目要求整數平均請使用雙斜線 `//`。

**APCS 考試實務建議：**
在 APCS 實作題中（如 f605 購買力），題目常要求篩選出特定價差條件的商品並計算「件數」與「總花費」。這正是累加器與計數器協同作戰的經典原型！

````python
# [2] Code 範例區：展示累加器與計數器協同統計及格同學表現

n = int(input())  # 學生人數
pass_total = 0    # 及格總分累加器
pass_count = 0    # 及格人數計數器

for _ in range(n):
    score = int(input())
    if score >= 60:
        pass_total += score  # 累加數值
        pass_count += 1      # 累計次數

print("及格人數：", pass_count)
print("及格總分：", pass_total)

# 防除零防禦檢查
if pass_count > 0:
    avg = pass_total / pass_count
    print("及格平均分：", avg)
else:
    print("無人及格")
````

### 🏆 6-2-5 迴圈動態極值維護——擂台盟主挑戰模式與常數空間串流求極值

**生活比喻：武林比武大會的「衛冕擂台主」**  
在比武招親或擂台爭霸賽中，舞台中央站著一位目前的「擂台盟主」。  
每當下一位挑戰者跳上台：  
- 裁判比對兩人的武功：如果挑戰者武功更高強（`challenger > current_champion`），衛冕者立刻退場，挑戰者成為「新任擂台盟主」（`current_champion = challenger`）！  
- 如果挑戰者武功較弱，盟主維持原位，挑戰者黯然退場。  
當所有參賽者全部挑戰完畢，最後站在台上的那個人，毫無疑問就是「全場武功最高強的天下第一（最大值）」！

在資料量龐大時，很多初學者以為一定要把幾萬個數字放進清單裡才能比大小。  
事實上，**只要利用「擂台主變數」，在迴圈中逐一比對更新，就能在不用串列的情況下，以超光速求得最大值與最小值**！

**擂台主的初始值設定智慧：**
如果我們要找最大值 `max_val`，在比賽開始前，擂台主應該先設多大？  
- **方法一（最安全通用的專業手法）：直接拿「第一筆輸入的資料」當作初始擂台主！**  
  先讀入第 1 個數字給 `max_val`，接下來迴圈只需要跑剩下的 $n-1$ 次，完全不用擔心預設值設錯。
- **方法二：設定為理論上的「極小下限值」**（例如題目保證皆為正整數時，可設為 `max_val = -1` 或 `0`）。

**語法核心標準寫法：**
```python
n = int(input())
# 先讀第一筆當作初始擂台盟主
max_val = int(input())

# 剩下的 n - 1 筆逐一上台挑戰
for _ in range(n - 1):
    val = int(input())
    if val > max_val:      # 挑戰成功！
        max_val = val      # 換新盟主！

print("最大值為：", max_val)
```

**常見錯誤陷阱：**
- **找最大值卻把初始值設得太大**：如果題目有負數（如 -50, -20, -10），若將 `max_val` 初始設為 0，結果比完之後答案竟然是 0（原本資料裡根本沒有 0）！這就是為什麼「以第一筆資料當作初始值」是最穩健安全的黃金法則！

**APCS 考試實務建議：**
此模式也是第三章 `max()` 函數在迴圈中的極致體現：`max_val = max(max_val, val)`。在 APCS c294、f312 等大題中，求全域極值皆依賴此核心邏輯！

````python
# [2] Code 範例區：展示在迴圈中動態維護最大值（擂台盟主模式）

# 模擬情境：讀入 4 位參賽者的投籃得分，找出最高分
n = 4
print("請輸入", n, "位選手的分數（第一筆為初始盟主）：")

first_score = int(input())
best_score = first_score  # 初始盟主

# 走訪剩下的 n - 1 位挑戰者
for _ in range(n - 1):
    challenger = int(input())
    # 若挑戰者分數更高，更新盟主
    if challenger > best_score:
        best_score = challenger

print("全場最高分為：", best_score)
````
