### 📦 8-1-1 串列概念、中括號建立與異質容器特性

**生活比喻：多寶格藥櫃與火車貨物列車**  
走進傳統中藥行，你會看到整面牆壁都是整齊的小抽屜，每個小抽屜拉開都能放特定的藥材；又或者像一列貨物列車，車頭後面拉著第一節、第二節、第三節車廂，每節車廂都可以裝載不同的貨物。  
在 Python 中，這種「用一條繩子串起多個獨立物品」的容器，就叫做 **串列（List）**！  
宣告串列的標誌性符號是**一對方括號 `[ ]`**，裡面的各個元素之間用**逗號 `,`** 隔開。

**底層運作機制與異質容器（Heterogeneous）特性**  
在 C/C++ 或 Java 等傳統語言中，陣列往往被嚴格限制「全體元素必須是同一種型態」（例如整數陣列只能放整數）；  
但 Python 的串列具備極高的包容性，被稱為**「異質容器（Heterogeneous Container）」**！  
你可以建立純整數串列：`scores = [90, 85, 100]`；  
也可以建立純文字串列：`names = ["Alice", "Bob", "Charlie"]`；  
甚至可以在同一個串列中混合放置整數、浮點數與字串：`mixed = [1, "Hero", 99.5]`！  
若裡面什麼都不放，就稱為**空串列（Empty List）**：`empty_list = []`，長度為 0。

**常見錯誤陷阱：括號打錯與少打逗號**  
初學者宣告串列時常誤用小括號 `(1, 2, 3)`（這在 Python 中是元組 Tuple）或花括號 `{1, 2, 3}`（集合 Set）。請務必牢記：**串列專屬的標誌是方括號 `[ ]`**！另外，元素之間若漏打逗號（如 `[1 2 3]`），會直接引發 `SyntaxError: invalid syntax`！

### 🎯 8-1-2 串列正負向索引：精準讀取各隔間物品

**生活比喻：火車各節車廂編號與倒數檢票**  
在單元 7-3 中，我們學到了字串可以用號碼牌存取單一字元。令人興奮的是，串列的索引規則與字串**完全一模一樣、無縫銜接**！  
想像串列是一列掛著 4 節車廂的列車 `a = [100, 200, 300, 400]`：
- **正向索引（從 0 開始）**：
  - `a[0]` 是第 1 節車廂（`100`）
  - `a[1]` 是第 2 節車廂（`200`）
  - `a[2]` 是第 3 節車廂（`300`）
  - `a[3]` 是第 4 節車廂（`400`，即 `len(a)-1`）
- **負向索引（從 -1 倒數）**：
  - `a[-1]` 永遠精準抓取最後一節車廂（`400`）！
  - `a[-2]` 是倒數第二節車廂（`300`）！

**底層運作機制與語意提升**  
雖然索引語法相同，但兩者取出的東西層次不同：  
- 字串取索引 `s[i]` 取出來的永遠是**單一字元**（長度為 1 的字串）。  
- 串列取索引 `a[i]` 取出來的是**存放在該位置的完整獨立物件**！如果放的是整數，取出來就是貨真價實的 `int`，可以直接拿去加減乘除，完全不需要轉型！

**常見錯誤陷阱：依然不能超界（IndexError）**  
長度為 4 的串列，若存取 `a[4]` 或 `a[-5]`，Python 直譯器同樣會立刻暴斃：  
**`IndexError: list index out of range`（串列索引超出範圍！）**  
永遠牢記合法索引上界是 `len(a) - 1`。

**APCS 考試實務建議：**  
在競技題中，當資料被讀入串列後，讀取首筆資料 `a[0]`、末筆資料 `a[-1]` 是最日常的操作。善用負向索引可以讓代碼免於寫出落落長的 `a[len(a)-1]`，顯著降低打字錯誤。

````python
# [2] Code 範例區：展示串列的正負向索引存取與型態保留
# 說明：觀察取出元素後的運算能力與末位負向索引。

prices = [50, 120, 80, 200, 30]

# 正向索引存取
first_price = prices[0]
second_price = prices[1]
print("第一項商品價格:", first_price)
print("第二項商品價格:", second_price)

# 取出元素可直接進行數學運算 (因為它就是 int)
total = first_price + second_price
print("前兩項合計:", total)

# 負向索引存取最後一項
last_price = prices[-1]
print("最後一項商品價格:", last_price)
````

### 🔄 8-1-3 串列可變性（Mutable）：就地替換元素（a[i] = val）

**生活比喻：可擦拭白板 vs 刻在石碑上的文字**  
還記得我們在 7-3 單元學過字串是「不可變的（Immutable）」嗎？字串一旦寫好就像刻在花崗岩石碑上的墓誌銘，如果寫錯了，你不能拿立可白塗掉單一個字母，試圖執行 `s[0] = 'B'` 會立刻被直譯器狠狠打臉！  
但是，串列完全不一樣！串列是一塊**「隨時可用板擦擦掉重寫的磁性白板」**！  
串列擁有極為尊貴的特性——**可變性（Mutable）**！你可以隨時指定任意一個抽屜編號，把裡面的舊東西丟掉，換上一個全新的寶物！

**底層運作機制：就地賦值修改（Item Assignment）**  
語法：`a[index] = new_value`
- 電腦會直接找到該索引對應的記憶體抽屜，將原本存放的物件參照切換為 `new_value`。
- **串列本身沒有被重建，記憶體身分證（`id`）完全不變**，這叫做「就地修改（In-place Modification）」！
- 範例：  
  `colors = ["Red", "Green", "Blue"]`  
  `colors[1] = "Yellow"`  # 把原本的 Green 就地換成 Yellow！  
  此時 `colors` 的內容變成 `["Red", "Yellow", "Blue"]`！

**兩數交換（Swap）的極致應用：**  
如果我們想交換串列中第 0 個元素與第 2 個元素的位置：  
在第一章我們學過多變數交換，這招在串列中依然神效：  
`a[0], a[2] = a[2], a[0]`  
一行代碼直接完成就地乾坤大挪移！

**常見錯誤陷阱：不能用賦值來「憑空新增」元素**  
初學者最容易誤解的一點：  
`a = [10, 20]`（長度為 2）  
`a[2] = 30` ❌  
這會瞬間引發 `IndexError: list assignment index out of range`！  
`a[i] = val` 只能用來**替換已經存在的抽屜**，不能用來把櫃子憑空變長！若要加長，必須使用後續學到的 `.append()` 方法。

### 🏗️ 8-1-4 固定長度預設值初始化：[0] * n 快速造表

**生活比喻：飯店客房預先鋪設與一鍵建造置物櫃**  
想像一家新建成的飯店有 100 間客房，在客人入住前，經理需要把所有房間的燈全部關閉、鋪上乾淨的白床單。如果經理必須一間一間手動填寫「未入住」，那得耗費大量時間。  
在寫程式時，我們經常需要一個「固定大小、全部填滿預設值（例如 0）」的串列，來當作計數器或紀錄表（例如記錄 0~9 號數字出現的次數）。  
在第二章我們學過字串乘法 `"*" * 10`；太棒了！**串列同樣支援乘法 `*` 運算子**！

**底層運作機制：一維串列的快速乘法初始化**  
語法：`a = [預設值] * 長度`
- `[0] * 5`：會瞬間打造出一個包含 5 個 0 的全新串列：`[0, 0, 0, 0, 0]`！
- `[False] * 4`：打造出一個布林旗標串列：`[False, False, False, False]`！
- 乘法建立之後，因為串列具有可變性，你就可以合法且安心地對每個既有抽屜執行賦值更新：  
  ```python
  counts = [0] * 10  # 準備 0 到 9 的計數櫃子
  counts[3] += 1     # 3 號出現了一次，合法更新！
  ```

**常見錯誤陷阱：二維陣列的乘法地雷（第九章特訓預告）**  
在**一維串列**中，`[0] * n` 是完全安全且最高效的標準寫法；但在第九章學習「二維陣列（List of Lists）」時，絕對不能寫 `[[0]*n] * m`，那會引發致命的共用參照陷阱！在此處處理一維純數值時，大家可以百分之百安心使用 `[0] * n`。

**APCS 考試實務建議：**  
在統計頻率（Bucket Sort 桶子計數法）或標記特定號碼是否出現過（如標記 1~N 號選手是否簽到）時，使用 `table = [0] * (n + 1)` 預先分配空間，是 APCS 解題中最不可或缺的高頻標準起手式。

### 🌟 8-1-5 競技輸出必殺技：星號解包輸出 print(*a)

**生活比喻：拆開快遞包裹與展示櫃陳列**  
想像你從網路上訂購了一箱模型公仔。如果快遞送來時，你直接把整整齊齊連同紙箱、膠帶封條與保麗龍泡棉直接擺在客廳桌上，客人會覺得這是一個「未拆封的包裹」；  
只有拿美工刀把箱子劃開，把裡面的公仔一個一個拿出來擺在展示架上，它們才是獨立完美的展品！  
在 Python 中：
- 如果你直接執行 `print(a)`：螢幕上會印出帶有方括號與逗號的原始容器外觀，例如 **`[10, 20, 30]`**！  
在絕大多數的線上裁判系統（Online Judge / APCS）中，題目都嚴格要求：**「請輸出各數值，彼此以一個空白隔開，行末不帶方括號與逗號！」**  
很多新手為了解開這個外殼，辛苦地寫 `for` 迴圈搭配 `end=" "`，一不小心行末還多了一個空格導致格式錯誤；  
而在 Python 魔法王國中，有一顆全宇宙最強大的解包炸彈——**星號解包 `print(*a)`**！

**底層運作機制：星號 `*` 的拆箱超能力**  
語法：`print(*a)`
- 星號 `*` 放在串列前面時，代表**「解包（Unpacking）」**！
- 它會把串列的外層方括號徹底剝掉，把裡面的所有元素「一個一個單獨拿出來」，依序傳給 `print()`！
- `print(*[10, 20, 30])` 底層等價於：`print(10, 20, 30)`！
- 由於 `print()` 多參數輸出預設就是以單一空白間隔，所以印出來就是完美無暇的：  
  **`10 20 30`**！
- 還能搭配 `sep` 參數自訂分隔符號：  
  `print(*a, sep=",")` ➔ 輸出 `10,20,30`！

**APCS 考試實務建議：**  
吳邦一老師講義第 12 頁特別強調 `print(*a)` 的高效解題價值！這是一道分水嶺語法——熟練運用 `print(*a)` 的同學，能在 1 秒之內優雅解決 95% 以上的串列格式化輸出需求，代碼既短又絕不踩雷！

````python
# [2] Code 範例區：展示 print(a) 與 print(*a) 星號解包輸出的巨大差別
# 說明：比較原始包裹輸出與剝除外殼後的乾淨競技格式輸出。

data = [10, 20, 30, 40, 50]

print("=== 一般 print(data) 輸出 (帶中括號與逗號) ===")
print(data)

print("\n=== 星號解包 print(*data) 輸出 (OJ 標準格式) ===")
print(*data)

print("\n=== 搭配 sep=' - ' 自訂連接符號 ===")
print(*data, sep=" - ")
````
