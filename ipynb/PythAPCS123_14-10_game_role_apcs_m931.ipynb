---

### 🪜 階梯 1：14.10.1 題意解析與「第二高」幾何戰力模型

#### 💡 核心概念導引
在現代多數遊戲數值設計中，單一特化數值往往會引發失衡（例如只有極端高攻擊卻極低防禦的「玻璃大砲」，或只有極端高防禦卻毫無傷害的「移動沙包」）。
本題所採用的「平方和指標」：
$$\text{Power} = \text{Attack}^2 + \text{Defense}^2$$
在二維直角坐標系中，恰好對應於將角色視為點 $(\text{Attack}, \text{Defense})$ 時，該點到原點 $(0, 0)$ 的**歐幾里得距離平方（Euclidean Distance Squared）**！

在幾何直觀上，離原點越遠的角色，其戰力圓半徑越大，代表綜合強度越高。
而本題最核心的思維轉折在於：題目不要距離原點最遠的第一名（最大值），而是要精準找出**距離原點第二遠的「榜眼角色（第二大值）」**！

```text
【角色幾何戰力示意圖】
  Defense ^
          |          ★ 角色 3 (4, 9) -> Power = 97 [第一高/金牌]
        8 |     ● 角色 1 (5, 8) -> Power = 89 [第二高/銀牌 ★ 目標]
          |          
          |     ▲ 角色 4 (7, 1) -> Power = 50 [第三高/銅牌]
        3 |   ■ 角色 2 (2, 3) -> Power = 13 [第四高]
        0 +-----------------------------------> Attack
          0     2    4  5    7
```

#### 🎯 本階梯學習目標
1. 掌握平方和運算在 Python 中的標準語法（`atk**2 + defn**2` 或 `atk*atk + defn*defn`）。
2. 理解「第二高」的相對大小關係：在 $n$ 個相異元素中，它小於且僅小於 1 個元素，但大於其餘 $n-2$ 個元素。

````python
# 範例 1：單一角色的平方和戰力計算函數
def calculate_power(atk: int, defn: int) -> int:
    """計算角色的綜合能力指標：Attack^2 + Defense^2"""
    return atk ** 2 + defn ** 2

# 測試官方範例一中的 4 位角色
test_roles = [
    (5, 8),
    (2, 3),
    (4, 9),
    (7, 1)
]

print("=== 4 位角色戰力計算展示 ===")
for idx, (a, d) in enumerate(test_roles, start=1):
    p = calculate_power(a, d)
    print(f"角色 {idx}：攻擊力 {a:2d}, 防禦力 {d:2d} -> 綜合能力 Power = {p:2d}")
````

````python
# 填空題 1：完成平方和計算函數並比對兩角色誰的戰力更高
def get_power(attack: int, defense: int) -> int:
    # 請填入平方和計算公式：attack 的平方加上 defense 的平方
    return attack ___ 2 + defense ___ 2

# 測試資料：勇者 A (3, 4) 與 勇者 B (1, 6)
power_a = get_power(3, 4)  # 3^2 + 4^2 = 9 + 16 = 25
power_b = get_power(1, 6)  # 1^2 + 6^2 = 1 + 36 = 37

print(f"勇者 A 戰力: {power_a}, 勇者 B 戰力: {power_b}")
if power_b > power_a:
    print("勇者 B 的綜合能力更高！")
````

````python
# 練習題 1：給定 3 位角色的 (atk, def)，找出戰力最高的數值
def find_highest_power(candidates):
    max_p = -1
    for a, d in candidates:
        p = a * a + d * d
        if p > max_p:
            max_p = p
    return max_p

# 公開測試組 1
group1 = [(1, 2), (3, 1), (2, 2)]
print("測試組 1 最高戰力:", find_highest_power(group1))  # (3,1)->10

# 公開測試組 2
group2 = [(6, 8), (7, 1), (5, 5)]
print("測試組 2 最高戰力:", find_highest_power(group2))  # (6,8)->100
````

````python
# 挑戰題 1：驗證「平方和相等」的幾何圓形概念
# 判斷兩組 (atk, def) 是否擁有完全相同的戰力半徑
def have_same_power(r1, r2):
    p1 = r1[0]**2 + r1[1]**2
    p2 = r2[0]**2 + r2[1]**2
    return p1 == p2

# 測試 (3, 4) 與 (4, 3) -> 戰力皆為 25
print("(3,4) 與 (4,3) 戰力是否相同:", have_same_power((3, 4), (4, 3)))
# 測試 (0, 5) 與 (3, 4) -> 戰力皆為 25
print("(0,5) 與 (3,4) 戰力是否相同:", have_same_power((0, 5), (3, 4)))
````

### 🪜 階梯 2：14.10.2 多行雙整數輸入讀取與資料關聯綁定

#### 💡 核心概念導引
在程式設計初學者最常犯的經典失誤中，**「資料脫鉤（Data Decoupling）」**名列前茅！
許多同學在解這題時，寫出了如下的直覺邏輯：
```python
# ❌ 初學者常見致命失誤：資料脫鉤
powers = []
for _ in range(n):
    a, d = map(int, input().split())
    powers.append(a**2 + d**2)

powers.sort()
second_power = powers[-2]
# 😱 此時驚覺：我知道第二高戰力是 89，但我完全不知道 89 是誰的攻擊力與防禦力算出來的！
```

要徹底根絕這種「顧此失彼」的窘境，最優雅且關鍵的程式心法就是**「資料三位一體綁定（Data Binding via Tuple）」**！
我們不只儲存戰力數值，而是將每一個角色的完整資訊打包成一個不可分割的**元組（Tuple）**：
$$\text{Role} = (\text{Power}, \text{Attack}, \text{Defense})$$

```text
【資料三位一體綁定結構圖】
  roles = [
     ( 89,  5,  8 ),   <-- 角色 1：(戰力, 攻擊, 防禦)
     ( 13,  2,  3 ),   <-- 角色 2
     ( 97,  4,  9 ),   <-- 角色 3
     ( 50,  7,  1 )    <-- 角色 4
  ]
```

為什麼要把 `Power` 放在元組的最前面（第 0 個位置）呢？這正是我們在下一個微階梯要揭曉的 Python 魔法！

#### 🎯 本階梯學習目標
1. 熟練多行雙整數的讀取與映射：`map(int, line.split())`。
2. 建立資料打包思維：將關聯數值封裝為元組 `(power, atk, defn)` 並加入串列管理。

````python
# 範例 2：將多行輸入讀取並封裝為三元組串列
raw_input_lines = [
    "4",
    "5 8",
    "2 3",
    "4 9",
    "7 1"
]

n = int(raw_input_lines[0])
roles = []

for i in range(1, n + 1):
    atk, defn = map(int, raw_input_lines[i].split())
    power = atk ** 2 + defn ** 2
    # 將 (power, atk, defn) 打包為元組加入清單
    roles.append((power, atk, defn))

print("封裝完成的角色資料清單：")
for r in roles:
    print(f"  戰力: {r[0]:2d} | 原始攻擊: {r[1]}, 原始防禦: {r[2]}")
````

````python
# 填空題 2：讀取行字串並封裝為 (power, atk, defn) 元組
sample_line = "6 8"

# 請填入拆解字串並轉為兩整數之語法
atk, defn = map(int, sample_line.___())
p = atk ** 2 + defn ** 2

# 請填入將 p, atk, defn 封裝為元組加入 roles 串列之語法
roles_list = []
roles_list.___( (p, atk, defn) )

print("封裝結果:", roles_list)
````

````python
# 練習題 2：模擬從字串列表中讀入 3 名角色並印出所有角色的原始攻防
def parse_roles_from_lines(lines):
    result = []
    n = int(lines[0])
    for idx in range(1, n + 1):
        a, d = map(int, lines[idx].split())
        result.append((a*a + d*d, a, d))
    return result

test_lines = ["3", "10 0", "0 10", "6 8"]
parsed = parse_roles_from_lines(test_lines)
print("解析角色總數:", len(parsed))
for p, a, d in parsed:
    print(f"  戰力: {p} -> (Atk: {a}, Def: {d})")
````

````python
# 挑戰題 2：列表生成式（List Comprehension）極簡單行封裝
# 在競賽中，熟練的選手能用一行將資料流直接轉化為綁定好的元組串列
raw_pairs = [(5, 8), (2, 3), (4, 9), (7, 1)]
roles_one_liner = [(a*a + d*d, a, d) for a, d in raw_pairs]
print("單行封裝結果:", roles_one_liner)
````

### 🪜 階梯 3：14.10.3 策略一基礎：Python 多元組（Tuple）大小比對與排序機制

#### 💡 核心概念導引
在前面我們特意將元組設計為 `(power, atk, defn)`，把 `power` 放第一位。這背後的根本原理是什麼？
這源自於 Python 語言中極度強大且直觀的**「字典序比較規則（Lexicographical Comparison）」**！

當 Python 比較兩個元組時，其比對順序為：
1. 先比對索引 `[0]` 的元素：如果數值不同，就由該元素的大小直接決定整個元組的大小！
2. 只有當索引 `[0]` 的元素完全相同時，才會接續比較索引 `[1]` 的元素，依此類推。

例如：
- `(97, 4, 9) > (89, 5, 8)`：因為第一項 `97 > 89`，Python **立刻判定前者較大**，根本不需要看後面的攻擊與防禦力！
- `(89, 5, 8) > (50, 7, 1)`：同樣因為 `89 > 50`，立刻判定前者較大。

```text
【元組字典序比對示意】
   ( 97,  4,  9 )
     VS
   ( 89,  5,  8 )
     ^
     | 第一項 97 > 89 立即決勝負！後續元素不影響大小關係。
```

由於本題官方明確保證：**「所有角色的綜合能力指標（平方和）皆相異」**，
這意味著所有元組的第一項 `power` 絕對不會重複！
因此，當我們直接對包含 `(power, atk, defn)` 的串列執行排序時，Python 會**百分之百純粹依據綜合戰力 `power` 進行排序**，完全不必自訂複雜的 `key=lambda` 排序函數！

#### 🎯 本階梯學習目標
1. 徹底理解 Python 元組的字典序比較原則。

### 🪜 階梯 4：14.10.4 策略一實現：全清單收集、`sort()` 排序與倒數第二項索引 `[-2]`

#### 💡 核心概念導引
掌握了元組排序機制後，我們就能立刻實現考場上**最直覺、程式碼最短、最不容易出錯的秒殺解法——全清單排序法**！

檢視題目規格：候選角色數量 $n$ 滿足 $3 \le n \le 20$。
這個資料規模非常小！即便到了子題組二，也僅僅只有 20 名角色。
對現代電腦而言，排序 20 筆資料耗時不到 $0.00001$ 秒。因此，在 APCS 考場上，只要資料規模允許，選擇**最不易寫出臭蟲的代碼**就是最高原則！

#### 🔍 負數索引的神奇魔力：`roles[-2]`
當我們將包含 $n$ 個元組的串列使用 `roles.sort()` 進行由小到大排序後：
- `roles[0]`：全場實力最弱的角色（最後一名）。
- `roles[-1]`：全場實力最強的角色（第一高 / 金牌冠軍）。
- **`roles[-2]`：全場實力第二強的角色（第二高 / 銀牌亞軍）**！

```text
【排序後的串列索引結構】
  roles = [
     (13, 2, 3),   <-- roles[0]  (第 4 名)
     (50, 7, 1),   <-- roles[1]  (第 3 名 / 銅牌)
     (89, 5, 8),   <-- roles[-2] (第 2 名 / 銀牌 ★ 目標答案！)
     (97, 4, 9)    <-- roles[-1] (第 1 名 / 金牌)
  ]
```

取得 `target = roles[-2]` 後，目標角色的原始攻擊力就是 `target[1]`，防禦力就是 `target[2]`，完全不用費心反查！

#### 🎯 本階梯學習目標
1. 完整實現策略一：讀取 ➔ 封裝 ➔ `roles.sort()` ➔ 提取 `roles[-2]`。
2. 熟練使用 Python 負數索引 `[-2]` 直取倒數第二個元素。

````python
# 範例 4：完整策略一（全清單排序法）解題流程
def solve_by_sorting(role_pairs):
    # 1. 建立角色三元組串列
    roles = []
    for a, d in role_pairs:
        power = a * a + d * d
        roles.append((power, a, d))
    
    # 2. 由小到大排序（預設依 power 比對）
    roles.sort()
    
    # 3. 取得倒數第二個（第二高戰力）角色
    second_best = roles[-2]
    
    # second_best 結構為 (power, atk, defn)
    return second_best[1], second_best[2]

# 測試官方範例一
sample_1 = [(5, 8), (2, 3), (4, 9), (7, 1)]
ans_atk, ans_def = solve_by_sorting(sample_1)
print("官方範例一輸出結果:", ans_atk, ans_def)  # 預期輸出: 5 8

# 測試官方範例二
sample_2 = [(1, 1), (2, 2), (3, 3)]
ans_atk, ans_def = solve_by_sorting(sample_2)
print("官方範例二輸出結果:", ans_atk, ans_def)  # 預期輸出: 2 2
````

### 🪜 階梯 5：14.10.5 策略二核心：雙變數極值動態交替推擠模型

#### 💡 核心概念導引
雖然排序法非常方便，但身為立志衝刺 APCS 實作 3 級以上的高手，我們必須掌握所有演算法中最經典的模式——**「雙變數極值動態維護（Two-Variable Running Maximums）」**！

想像一個武術錦標賽的擂台，擂台上隨時維持兩把榮譽交椅：
- **`max1`（金牌寶座）**：記錄全場目前為止綜合戰力**最高（第 1 高）**的角色。
- **`max2`（銀牌寶座）**：記錄全場目前為止綜合戰力**第二高（第 2 高）**的角色。

每當一位新角色登場時，只可能發生以下三種情境之一：

```text
【推擠模型三部曲】

情境 A：新挑戰者超強，擊敗了目前的金牌！ (power > max1[0])
  步驟 1：原金牌被推擠退位，降級為新的銀牌！ -> max2 = max1  (⚠️ 這步絕對要先做！)
  步驟 2：新挑戰者榮登金牌寶座！             -> max1 = new_role

情境 B：新挑戰者沒贏過金牌，但擊敗了銀牌！ (power > max2[0])
  步驟 1：金牌不受影響，依然穩坐第一。
  步驟 2：新挑戰者奪取銀牌寶座！             -> max2 = new_role

情境 C：新挑戰者連銀牌都打不過 (power <= max2[0])
  步驟 1：直接淘汰，擂台交椅維持不變。
```

#### 💣 考場致命大失誤：推擠順序顛倒！
許多同學在寫情境 A 時，一時心急寫成：
```python
# ❌ 致命慘劇：原金牌直接被覆蓋蒸發！
if power > max1[0]:
    max1 = new_role  # 完蛋！原來的金牌被洗掉了！
    max2 = max1      # 此時 max2 複製到的是新角色，而不是舊金牌！
```
請牢記口訣：**「新人登頂，舊王退居二線；先讓位（`max2 = max1`），再就職（`max1 = new`）！」**

#### 🎯 本階梯學習目標
1. 徹底理解雙極值動態交替推擠的狀態轉移條件。
2. 建立正確的賦值順序防禦意識，嚴防舊極值被提早覆蓋蒸發。

````python
# 範例 5：手把手追蹤雙變數極值推擠動態過程
def trace_two_max(roles_stream):
    # 初始化擂台：戰力設為 -1 表示交椅初始為空
    max1 = (-1, 0, 0)  # (power, atk, defn)
    max2 = (-1, 0, 0)
    
    print("=== 雙變數動態推擠擂台追蹤 ===")
    for step, (a, d) in enumerate(roles_stream, start=1):
        p = a * a + d * d
        cur = (p, a, d)
        
        if p > max1[0]:
            # 情境 A：超越金牌
            max2 = max1        # 舊金牌退居銀牌
            max1 = cur         # 新人登頂金牌
            action = f"超越金牌！舊金牌退居銀牌"
        elif p > max2[0]:
            # 情境 B：超越銀牌
            max2 = cur         # 新人奪取銀牌
            action = f"挑戰銀牌成功！更新銀牌"
        else:
            # 情境 C：被淘汰
            action = f"實力不足，直接淘汰"
            
        print(f"新角色 {step}: (攻{a},防{d},戰力{p:2d}) ➔ {action}")
        print(f"   目前金牌: 戰力{max1[0]:2d}(攻{max1[1]},防{max1[2]}) | 銀牌: 戰力{max2[0]:2d}(攻{max2[1]},防{max2[2]})")
    
    return max2

# 追蹤官方範例一
trace_two_max([(5, 8), (2, 3), (4, 9), (7, 1)])
````

### 🪜 階梯 6：14.10.6 策略二實現：$O(n)$ 一次走訪與空間複雜度 $O(1)$ 串流解法

#### 💡 核心概念導引
現在，我們要將前面的推擠模型昇華為競技程式設計中的神技——**「線上串流解法（Online Streaming Algorithm）」**！

什麼叫串流解法？
在策略一中，我們把所有 $n$ 筆資料全部存進串列 `roles = []` 中，這需要消耗 $O(n)$ 的記憶體空間。
然而，在策略二中，我們**完全不需要開串列儲存所有角色**！
我們可以在讀入每一行資料的同時，當場計算戰力、當場推進擂台比較；比對完畢後，這筆資料的使命就結束了，記憶體中始終**只保留 `max1` 與 `max2` 兩個變數**！

```text
【串流處理流程圖】
  輸入資料流 ──> 讀取一行 (atk, def)
                    │
                    ▼
             計算 power = a*a + d*d
                    │
                    ▼
          與 max1, max2 動態推擠比對 ──> 更新交椅
                    │
                    ▼ （不儲存！直接丟棄，迎向下一筆）
  全數讀完後 ──> max2 即為最終答案！
```

#### 🌟 串流解法的終極威力
- **時間複雜度**：$O(n)$，只巡迴資料一次，完全免去排序的時間消耗！
- **空間複雜度**：$O(1)$，不論 $n$ 是一百、一萬還是一億，所佔用的記憶體大小完全恆定！

#### 🎯 本階梯學習目標
1. 實作免串列、即時讀入即時比較的 $O(1)$ 空間串流程式碼架構。
2. 體會線性掃描在極大規模資料流中的極速與省記憶體優勢。

````python
# 範例 6：空間複雜度 O(1) 串流解法完整實作
def solve_streaming(lines):
    """
    以串流方式走訪資料，不建立全域角色清單
    時間複雜度：O(n)
    額外空間複雜度：O(1)
    """
    n = int(lines[0])
    
    # 維護前兩大極值 (power, atk, defn)
    max1 = (-1, 0, 0)
    max2 = (-1, 0, 0)
    
    for i in range(1, n + 1):
        atk, defn = map(int, lines[i].split())
        power = atk * atk + defn * defn
        
        if power > max1[0]:
            max2 = max1
            max1 = (power, atk, defn)
        elif power > max2[0]:
            max2 = (power, atk, defn)
            
    # 返回第二大角色的攻擊與防禦力
    return max2[1], max2[2]

# 驗證官方測資一
data1 = ["4", "5 8", "2 3", "4 9", "7 1"]
print("串流解法測試一:", solve_streaming(data1))  # (5, 8)

# 驗證官方測資二
data2 = ["3", "1 1", "2 2", "3 3"]
print("串流解法測試二:", solve_streaming(data2))  # (2, 2)
````

````python
# 範例 8：雙演算法自動化對拍測試（Stress Testing）
# 生成 100 組隨機測試案例，比對「排序法」與「串流走訪法」是否答案永遠完全一致
import random

def test_cross_validation(num_cases=100):
    random.seed(2024)
    all_pass = True
    
    for case_id in range(1, num_cases + 1):
        n = random.randint(3, 20)
        # 隨機生成 n 個平方和互不相同的角色
        used_powers = set()
        roles = []
        while len(roles) < n:
            a = random.randint(0, 100)
            d = random.randint(0, 100)
            p = a * a + d * d
            if p not in used_powers:
                used_powers.add(p)
                roles.append((a, d))
                
        # 策略一解法
        sorted_roles = sorted([(a*a + d*d, a, d) for a, d in roles])
        ans1 = (sorted_roles[-2][1], sorted_roles[-2][2])
        
        # 策略二解法
        m1, m2 = (-1, 0, 0), (-1, 0, 0)
        for a, d in roles:
            p = a*a + d*d
            if p > m1[0]:
                m2 = m1
                m1 = (p, a, d)
            elif p > m2[0]:
                m2 = (p, a, d)
        ans2 = (m2[1], m2[2])
        
        if ans1 != ans2:
            print(f"❌ 發現不一致！Case {case_id}: 排序法={ans1}, 走訪法={ans2}")
            all_pass = False
            break
            
    if all_pass:
        print(f"🎉 完美對拍通過！全部 {num_cases} 組隨機測資，兩種演算法輸出 100% 吻合！")

test_cross_validation(100)
````

### 📝 版本一：APCS 官方實作考場專用版 —— 淺顯易懂一般版（新手友善推薦）

* **適用情境**：APCS 正式考試現場、初學者課堂練習。
* **教學與設計理念**：
  1. **步驟平鋪直敘、依序讀取**：先讀入角色總人數 `n`，再以 `for` 迴圈依序讀取每位角色的攻擊力與防禦力。
  2. **計算戰力並綁定資料**：計算 `power = atk * atk + df * df`，將 `(power, atk, df)` 作為多元組存入 `roles` 清單。
  3. **利用 sort() 排序取倒數第二位**：Python 預設以多元組第一項（戰力）從小到大排序，直接取倒數第二位 `roles[-2]` 即可精準取得第二高戰力角色。
  4. **穩健 100% 滿分 AC**：無需維護複雜的雙變數交替推擠指標，最直觀、最不易緊張出錯，考場穩拿滿分！


````python
# ==============================================================================
# 📝 版本一：APCS 官方實作考場專用版 —— 淺顯易懂一般版（新手友善推薦）
# 適用情境：APCS 正式考場單筆測試案例，全清單收集排序、步驟明確、穩健滿分
# ==============================================================================

# 步驟 1：讀取角色總人數 n
n = int(input())

# 步驟 2：建立儲存所有角色資料的清單
roles = []

# 步驟 3：逐一讀取每位角色的攻擊力與防禦力，並計算戰力
for i in range(n):
    atk, df = map(int, input().split())
    power = atk * atk + df * df
    # 將 (戰力, 攻擊力, 防禦力) 作為元組加入清單
    roles.append((power, atk, df))

# 步驟 4：由小到大依戰力進行排序
roles.sort()

# 步驟 5：取得戰力第二高的角色（排序後倒數第二個元素 index -2）
second_role = roles[-2]
second_atk = second_role[1]
second_df = second_role[2]

# 步驟 6：依題目要求印出該角色的攻擊力與防禦力（空格隔開）
print(second_atk, second_df)
````
