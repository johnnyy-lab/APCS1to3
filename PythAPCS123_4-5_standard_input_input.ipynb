### 👂 4-5-1 input() 的本質——電腦的靈敏耳朵與「輸入一律為字串」的鐵律

想像一下學校教務處的公文收發窗口。當你在窗口遞進去一張紙條，不管上面寫的是文字「請假申請」、英文字母「ABC」，還是數字「100」，收發老師都是把它當成「一封公文信件（文字資料）」收下來，而絕不可能因為紙條寫了 100，就把紙條當成 100 塊真正的硬幣拿去買便當！
在 Python 魔法王國裡，標準輸入函數 `input()` 扮演的正是這個公文收發窗口，也就是**電腦的「靈敏耳朵」**！

**運作機制與底層原理：**
當 Python 直譯器執行到 `input()` 這行指令時，固定發生以下三個關鍵步驟：
1. **停頓等待（阻塞機制 Blocking）**：電腦會立刻停下所有後續指令的執行，進入耐心等待模式，畫面會出現一個閃爍的游標或輸入文字框，等待使用者在鍵盤上敲打字元。
2. **收納整行資料**：直到使用者敲下鍵盤上的 **Enter 鍵**（換行鍵），電腦便認定這句話已經說完，瞬間將剛才輸入的所有按鍵內容打包成一個包裹。
3. **鐵律：一律回傳字串（str）**：這是所有 Python 初學者最必須牢記的黃金準則——**不論使用者在鍵盤上按下了什麼，`input()` 拿到的包裹資料型態「一律是文字字串（str）」**！即使你輸入的是 `2026`、`99` 甚至是負數 `-5`，在電腦眼中它們全都被引號牢牢鎖住，成為了純文字 `"2026"` 與 `"99"`！
4. **賦值貼上名牌**：我們必須使用變數與等號（如 `user_data = input()`），將這包文字包裹好好存放到記憶體置物櫃中，否則輸入的資料就會像風吹過一樣瞬間消失。

**常見錯誤陷阱（初學者必讀三大地雷）：**
1. **以為輸入數字就能直接做算術**：如果直接拿 `input()` 的結果去相加，例如 `ans = a + b`，電腦會執行字串串接（文字接龍），`"5" + "3"` 會變成 `"53"` 而不是 `8`！
2. **單獨呼叫 input() 忘記用變數接住**：有些同學只寫了一行 `input()`，使用者輸入了資料，但後面程式卻找不到名牌可以拿出來用，導致白白讀取。
3. **重複呼叫兩次 input() 導致需要按兩次 Enter**：如果寫 `print(input())` 接著又寫 `x = input()`，這代表要使用者在鍵盤上連續輸入兩次，造成互動邏輯混亂！

### 🔢 4-5-2 字串轉整數 int(input())——破解 "10" + "20" = "1020" 陷阱，解鎖真正整數計算

想像你去夜市買了兩袋炸雞，第一袋紙袋上面用奇異筆寫著「10」顆，第二袋紙袋上寫著「20」顆。如果老闆算總顆數時，不是把兩袋紙袋裡面的炸雞倒在盤子裡數出「30 顆」，而是直接把兩個紙袋的字樣拼在一起說「先生你總共買了 1020 顆！」，你一定會覺得老闆在開玩笑！
可是在 Python 的世界裡，如果你寫 `a = input()` 和 `b = input()`，然後直接算 `a + b`，電腦就會嚴肅地做出這件荒謬的事！因為前面說過，`input()` 拿到的是裝在引號紙袋裡的「文字」，而文字碰上加號 `+`，執行的就是文字接龍！想要進行真正的數學計算，我們就必須把紙袋拆開，把文字轉型為真正的純整數——這就是經典的 **`int(input())` 組合技**！

**運作機制與底層原理：**
當我們撰寫 `num = int(input())` 時，這是一個標準的**巢狀函數呼叫（Nested Function Call）**，Python 永遠嚴格遵守「**由內而外求值**」法則：
1. **第一步（內層聽取）**：內層的 `input()` 先執行，停下來等待使用者敲鍵盤。假設使用者輸入了 `25` 並按 Enter，內層函數會產出一個字串物件 `"25"`。
2. **第二步（外層拆裝）**：外層的 `int()` 函數立刻接過這個產物，執行型態轉換。它檢查字串內容是否全為合法數字，確定無誤後，撕開引號外皮，將文字 `"25"` 轉換為二進位存放的純整數 `25`！
3. **第三步（賦值貼牌）**：賦值等號 `=` 將這個真正的整數數值放進左邊的變數置物櫃中。
經過這三道工序，變數 `num` 就能自由自在地進行加、減、乘、整除 `//`、取餘數 `%` 以及次方 `**` 等所有算術運算了！

**常見錯誤陷阱（初學者必讀三大地雷）：**
1. **忘記包外層 int() 導致加法變接龍**：初學者最常漏寫 `int()`，以為打的是數字電腦就知道是數字，導致 `3 + 5` 算出來變成 `35`。
2. **字串含有小數點卻丟給 int() 導致報錯**：如果使用者輸入了 `3.14`，直接丟給 `int()` 會瞬間引發 `ValueError: invalid literal for int() with base 10: '3.14'`！請記住：`int()` 只能解析純整數的文字，小數點必須留給後面要教的 `float()`。
3. **輸入含有中文、空格或英文**：若輸入 `"100元"` 或 `"abc"`，同樣會觸發 `ValueError`，因為這些文字無法轉換為整數。

在 APCS 第一題實作真題中（例如三角形辨別、秘密差、購買力等），讀取整數是出現頻率最高的起手式。熟練撰寫 `int(input())` 並清楚其求值順序，是所有程式競賽選手必備的肌肉記憶！

### 📐 4-5-3 小數輸入 float(input())——精準量測身高體重與實數幾何計算

在科學實驗室裡，物理學家在量測單擺週期、化學家在滴定溶液容量、或是體育老師在測量選手的百米賽跑秒數時，世界上的真實數據幾乎都不是完美的整數。如果某位同學的身高是 168.5 公分，如果你用上一小節學到的 `int(input())` 去讀取，直譯器會立刻報警當機！
面對真實世界中無處不在的小數點（實數），我們必須派出另一位轉型大將軍——**`float(input())`**！

**運作機制與底層原理：**
`float()` 函數具備強大的小數點字串辨識能力：
1. **精確解析小數點**：當使用者鍵入 `"168.5"` 並按下 Enter 時，內層的 `input()` 先拿到字串 `"168.5"`。外層的 `float()` 能夠精準識別中間的數值小數點 `.`，將其轉換為遵循 IEEE 754 雙精度浮點數格式的數值 `168.5`。
2. **整數向上相容**：有趣的是，如果使用者輸入的是整數 `"50"`，交給 `float()` 處理也不會出錯，它會自動將其升級為帶小數點的浮點數 `50.0`。
3. **配合 round() 或第四章 sep='' 精巧輸出**：浮點數運算往往會產生很長的小數點位數（例如除法運算），在輸出時常搭配第三章學過的 `round(數值, 位數)`，或者利用 `sep=''` 與文字單位無縫黏合，產出兼具精確度與專業美感的報表！

**常見錯誤陷阱（初學者必讀三大地雷）：**
1. **使用者手滑輸入單位字元**：例如鍵入 `"168.5cm"` 或 `"52.3kg"`，因為包含了非數字的英文字母，`float()` 會直接拋出 `ValueError`。在電腦讀取純數值時，絕對不可以輸入單位！
2. **輸入多個小數點**：例如誤打成 `"3.1.4"`，這不是合法的實數，同樣會造成 `ValueError`。
3. **浮點數精準度錯覺**：在第二章我們學過電腦以二進位表示十進位小數時的限制（例如 `0.1 + 0.2` 不完全等於 `0.3`），因此小數點運算結果若要作為最後結論輸出，記得適度使用 `round()` 修剪末端雜訊！

### 📑 4-5-4 循序多次調用 input() 讀入多行單一資料——分行讀入矩形長寬與商品收據

想像你去手搖飲料店點餐，櫃台店員絕不可能只聽你說一個字就跑去搖飲料。店員一定會先問第一句：「甜度要多少？」（你回答：半糖，按一次 Enter）；店員接著又問第二句：「冰塊要多少？」（你回答：去冰，又按一次 Enter）。如果店員不等你講第二句就急著結帳，那點單絕對會大亂！
在 Python 程式中也是完全一樣的道理：**每一次呼叫 `input()`，就代表電腦洗耳恭聽「一整行」！**
當使用者在鍵盤上敲下 Enter 鍵，這一行輸入就正式結束；如果你的題目需要讀取兩行、三行甚至更多行資料，你就必須在程式碼中**依序呼叫多次 `input()`**，將每一行內容分別接引至各自專屬的變數名牌置物櫃中！

**運作機制與底層原理：**
Python 的輸入串流（Input Stream）運作機制宛如一條輸送帶：
1. **行尾截斷與指標推移**：使用者按下 Enter 鍵時，輸入串流中會產生一個不可見的換行符號 `\n`。第一次調用 `input()` 時，直譯器會把輸送帶上第一行（直到 `\n` 前）的文字打包回傳，並自動把輸送帶指標移動到第二行的最開頭。
2. **多行循序排隊**：當程式執行到下一行程式碼的第二個 `input()` 時，直譯器再次進入等待，讀取輸送帶上的第二行文字。
3. **型態依序轉換**：每一行可以依照題意套用不同的轉型包裝。例如第一行讀取學生姓名（保持 `str`），第二行讀取段考分數（套用 `int(input())`），井然有序、互不干擾！

**常見錯誤陷阱（初學者必讀三大地雷）：**
1. **變數讀入順序與測資顛倒**：這是最嚴重的致命失誤！如果題目規定第一行給的是商品名稱（文字）、第二行給的是價格（整數），結果你的程式第一行寫 `price = int(input())`，第二行寫 `name = input()`，第一行文字一被餵給 `int()` 就會立刻炸出 `ValueError` 導致程式當場終止崩潰！務必嚴格對照題目格式，一對一順序排列！
2. **以為一次 input() 可以直接讀入兩行**：有些同學以為寫一行 `a, b = input()` 就能同時抓取兩行，這在語法上是完全錯誤的，必須分成兩行各自呼叫 `input()`。
3. **少寫一次 input() 導致測資吃不完**：題目給了三行，程式只寫了兩次 `input()`，剩下的資料被孤立在串流中，導致計算短少。

**APCS 考試實務建議：**
在 APCS 許多實作真題中（例如 e286 籃球比賽連續給出四行每節比分、f579 購物車多行操作），輸入資料都是分行送達的。養成「對照題目輸入範例，一行對應一個 input()」的良好習慣，是建立多筆資料輸入架構的基本功！

### 🚫 4-5-5 競技程式 APCS / OJ 生死線——嚴禁在 input() 括號內放入提示文字的「純淨讀入」哲學

很多剛在學校電腦課接觸 Python 的同學，在練習寫個人小工具時，很喜歡寫成這樣：
```python

```
在終端機上互動時，螢幕印出親切的引導提示，看起來非常溫馨貼心。但是，**如果你帶著這個習慣去參加 APCS 檢定，或是把這段程式碼提交到 ZeroJudge 等線上裁判系統（Online Judge），系統會毫不留情地直接判你滿江紅的 `Wrong Answer (WA 答案錯誤)` 或是 `Presentation Error (格式錯誤)`！**
為什麼會這樣？這就是所有程式初學者最常踩中、也最感到委屈困惑的「隱形大雷區」！

**運作機制與底層原理：**
要明白為什麼不能加提示字串，我們必須透視線上評判系統（Online Judge）的工作原理：
1. **裁判是冷酷無情的自動化機器人**：OJ 評判你的代碼時，背後是一套全自動化比對程序。它將標準測試資料灌入你的程式標準輸入（stdin），然後捕捉你的程式在標準輸出（stdout）印出的每一個字元，並將其與官方的「標準答案」進行**逐字、逐行、連空白與換行都一模一樣的嚴格比對**。
2. **prompt 參數的底層真面目**：當你寫下 `input("請輸入數值：")` 時，Python 直譯器在等待輸入前，會**強制把引號內的文字直接發送到標準輸出（stdout）**！這意味著，原本題目只要求印出一個數字 `42`，你的程式卻印出了 `請輸入數值：42`！
3. **評判機器人的判定**：機器人看到多出了「`請輸入數值：`」這串字，它並不知道這是給人類看的友善提示，只會認定「輸出內容與答案不符」，毫不猶豫判定零分！

2. **擅自印出多餘的裝飾線或說明文字**：例如題目只要求輸出兩數之和，同學卻自己加戲寫 `print("兩數相加的答案是：", ans)`，同樣會全盤被判錯！題目要什麼，就精準印出什麼！
3. **誤以為 Colab 輸入需要提示文字才看得懂**：在 Colab 互動練習時，上方跳出的輸入框即使沒有提示文字，游標依然會正常等待你輸入。

面對這種情況，單純呼叫 `input()` 只能抓下一整串文字，如果直接丟給 `int()` 更會立刻噴出錯誤！該如何一次把同一行的多個數字拆分開來？
在下一個單元「**單元 4-6：單行多數值切割與映射（split, map）**」中，我們將解鎖 Python 競賽最威猛的一行流讀入大絕招！我們下一課見！
