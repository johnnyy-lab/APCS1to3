### 📏 7-3-1 字串長度函數 len() 與空字串邊界

**生活比喻：火車車廂清點員與珍珠計數器**  
當一列長長的火車開進月台時，站務員如果想知道這列火車總共掛了幾節車廂，他不需要把每節車廂拆開，只需要拿計數器從頭到尾點一遍。  
在 Python 中，專門負責清點字串「裡面總共有多少個字元」的官方計數器，就是內建函數 **`len()`**（取自英文 Length 長度的縮寫）。無論字串裡面裝的是英文字母、中文漢字、標點符號、甚至是看不見的空白字元，`len()` 都會公正無私地將其計為「1 個字元長度」！

**底層運作機制與 O(1) 極速查詢**  
很多初學者以為，當我們呼叫 `len(s)` 時，電腦是從第一個字元開始一個個慢慢數到最後一個字元。  
其實不然！在 Python 直譯器內部，每個字串物件在誕生的瞬間，記憶體頭部就已經悄悄記錄了該字串的精確長度。因此呼叫 `len()` 時，電腦只是瞬間讀取現成的長度數值，速度是 $O(1)$ 常數時間，即便字串長達數十萬字，也是一眨眼就得到結果！

**空字串的邊界特性：**  
- 普通字串：`len("Python")` 結果為 `6`；`len("中文字")` 結果為 `3`。
- 包含空格：`len("A B")` 結果為 `3`（中間的空格也算一個字元！）。
- **空字串（Empty String）**：`len("")` 結果為 **`0`**！這是所有字串長度的下限，也是條件檢查中判斷「使用者是否什麼都沒輸入」的常用判斷式（`if len(s) == 0:`）。

**常見錯誤陷阱：對整數或浮點數使用 len()**  
數字型態（`int`, `float`）並不是由字元組成的容器，如果寫出 `len(12345)` ❌，直譯器會立刻報錯：  
**`TypeError: object of type 'int' has no len()`**  
若想計算一個數字有幾位數，必須先用 `str()` 轉為字串：`len(str(12345))` 得到 `5` ✅！

````python
# [2] Code 範例區：展示 len() 函數測量各種字串長度的效果
# 說明：觀察英文、中文、空格與空字串的長度計算。

s1 = "APCS"
s2 = "程式設計"
s3 = "Hello World"
s4 = ""

print("s1 的長度:", len(s1))
print("s2 的長度:", len(s2))
print("s3 的長度 (包含空格):", len(s3))
print("s4 (空字串) 的長度:", len(s4))

# 測量大數的十進位位數：先轉成 str 再取 len
big_number = 9876543210123456789
digits_count = len(str(big_number))
print("大數的位數:", digits_count)
````

### 🎯 7-3-2 正向索引（Positive Indexing）與 IndexError 超界防禦

**生活比喻：置物櫃號碼牌與從 0 開始的電腦傳統**  
走進圖書館或游泳池的置物區，你會看到一整排相連的置物櫃。  
在人類的習慣中，第一個置物櫃通常被貼上「1 號」；但在電腦科學的魔法世界中，工程師們習慣**「從 0 號開始編號」**！  
想像字串 `s = "MAGIC"` 是一排置物櫃：
- `s[0]` 放著 `'M'`（第 1 個字元）
- `s[1]` 放著 `'A'`（第 2 個字元）
- `s[2]` 放著 `'G'`（第 3 個字元）
- `s[3]` 放著 `'I'`（第 4 個字元）
- `s[4]` 放著 `'C'`（第 5 個字元，也是最後一個字元）

這種透過中括號加號碼 `s[index]` 取用特定字元的語法，就叫做 **索引（Indexing）**。

**底層運作機制與 0-based 編號核心公式**  
若一個字串的長度為 $N = len(s)$：  
- 最開頭的第一個字元索引一定是 **`0`**。
- **最末尾的最後一個字元索引一定是 `N - 1`**（即 `len(s) - 1`）！

**常見錯誤陷阱：IndexError 超界大崩潰（Off-by-one 經典錯誤）**  
初學者最常犯的致命直覺錯誤是：  
`s = "CAT"`（長度為 3）  
`last_char = s[3]` ❌  
程式會瞬間暴斃並丟出紅色警報：  
**`IndexError: string index out of range`（字串索引超出範圍！）**  
因為 `"CAT"` 的合法索引只有 `0, 1, 2`，根本沒有 3 號置物櫃！試圖伸手去開不存在的櫃子，作業系統就會立刻重罰。

**APCS 考試實務建議：**  
在設計迴圈走訪字串或讀取首尾字元時，永遠牢記末位字元是 `s[len(s)-1]` 而非 `s[len(s)]`。在競賽線上裁判系統中，`IndexError` 是導致程式拿到 RE（Runtime Error）評定的最主要原因之一，必須保持高度警戒。

````python
# [2] Code 範例區：展示正向索引存取字元與邊界計算
# 說明：觀察 0-based 索引如何定位第一個與最後一個字元。

word = "PYTHON"
n = len(word)

print("完整單字:", word)
print("字串長度:", n)

# 存取第一個字元 (索引 0)
print("第一個字元 word[0]:", word[0])

# 存取最後一個字元 (索引 n - 1)
print("最後一個字元 word[n-1]:", word[n - 1])

# 存取中間字元 (例如索引 2)
print("第三個字元 word[2]:", word[2])
````

### 🔙 7-3-3 負向索引（Negative Indexing）：倒數定位技巧

**生活比喻：排隊隊伍從最後一名往回倒數**  
想像遊樂園排隊搭雲霄飛車，如果隊伍排了 100 個人，你想跟「最後一名」說話，難道你得先跑到隊伍最前端，從 1 慢慢數到 100 才知道誰是最後一個嗎？  
當然不用！你只需要直接拍拍排在最後面那位的肩膀說：「倒數第一名，換你了！」。  
在許多程式語言中，要拿最後一個字元，必須繁瑣地寫 `s[len(s) - 1]`；但 Python 貼心地為我們設計了全宇宙最優雅的語法——**負向索引（Negative Indexing）**！

**底層運作機制與負向倒數標號法則**  
Python 允許在索引中使用負整數：
- **`s[-1]`**：永遠代表**最後一個字元**（倒數第 1 個）！
- **`s[-2]`**：代表**倒數第二個字元**！
- **`s[-3]`**：代表**倒數第三個字元**！
- 直到 **`s[-len(s)]`**：剛好倒數到最開頭的第一個字元！

以 `s = "HERO"` 為例：
```text
正向索引：   0    1    2    3
字元內容：   H    E    R    O
負向索引：  -4   -3   -2   -1
```
`s[0]` 與 `s[-4]` 指向同一個 `'H'`，而 `s[3]` 與 `s[-1]` 指向同一個 `'O'`！

**常見錯誤陷阱：負向超界（Negative Out of Range）**  
負向索引同樣受到邊界保護！若長度為 4 的字串，你存取了 `s[-5]`，電腦會因為倒數過頭而同樣觸發：  
**`IndexError: string index out of range`**！

**APCS 考試實務建議：**  
在題意要求檢查副檔名（例如最後一個字元是不是小寫字母）、判斷句尾標點符號（如最後是不是問號 `?`）、或者數字奇偶性判定時，直接使用 `s[-1]` 不需要計算字串長度，代碼既精準又具備最高防禦力。

````python
# [2] Code 範例區：展示負向倒數索引的便攜性
# 說明：使用 s[-1] 與 s[-2] 免算長度秒殺末位定位。

sentence = "Where is the treasure?"

# 取出最後一個字元 (檢查標點符號)
last_char = sentence[-1]
print("最後一個字元 sentence[-1]:", last_char)

# 取出倒數第二個字元
second_last = sentence[-2]
print("倒數第二個字元 sentence[-2]:", second_last)

# 驗證 s[-len(sentence)] 是否等於 s[0]
print("倒數最前端 s[-len]:", sentence[-len(sentence)])
print("正向第一個 s[0]  :", sentence[0])
````

### 🔒 7-3-4 字串不可變性（Immutable）：就地竄改的致命陷阱

**生活比喻：凝固的琥珀與鑄鐵硬幣**  
想像一隻被封存在松樹脂裡的遠古小昆蟲，經過數百萬年凝固成了琥珀。你可以把它拿在手上觀賞、用尺量它的長度、用放大鏡仔細看它觸角的紋路；但是，你**絕對不可能**在不打碎琥珀的情況下，伸手把裡面昆蟲的翅膀單獨拔下來換掉！  
在 Python 的物件家族中，字串（`str`）就像這塊堅硬無比的琥珀。字串擁有非常重要的天賦屬性——**不可變性（Immutable）**！

**底層運作機制：為何字串無法就地修改？**  
很多從其他語言（如 C 語言 char 陣列）轉過來，或是剛學完索引的同學，會直覺地想要這樣修改字串：  
`s = "Cat"`  
`s[0] = "B"`  # 想要把 'C' 換成 'B'，變成 "Bat" ❌  
當直譯器執行這行賦值時，會立刻嚴厲制止並拋出型態異常：  
**`TypeError: 'str' object does not support item assignment`（字串物件不支援單點賦值修改！）**  
在 Python 的記憶體設計中，字串一旦被創造出來，其內部的每個字元就永遠被焊死了，沒有任何人能就地更改它的一分一毫！

### 🚶 7-3-5 兩大字串走訪模式：字元走訪 vs 索引走訪

**生活比喻：挨家挨戶問候 vs 拿著號碼牌名冊逐戶點名**  
當郵差來到一條巷子送信時，有兩種工作模式：
1. **逐戶走訪（只看人）**：郵差走過每一棟房子，把信件直接親手交給門口站著的人（「你好，信給你！」）。他不需要管對方的門牌號碼到底幾號。
2. **索引走訪（看門牌號碼）**：郵差手裡拿著一張「0 號到 9 號」的清冊，他站在 3 號門口，核對門牌，甚至還能轉頭看看 2 號鄰居或 4 號鄰居的狀況！

在 Python 迴圈中走訪字串，正好完全對應這兩種經典模式：

**模式一：直接走訪字元（`for ch in s:`）——清爽直觀**  
- 語法：`for ch in s:`
- 運作：迴圈會自動從第 0 個字元依序把字元裝進變數 `ch`，直到最後一個字元結束。
- 適用時機：**當你只需要檢查每個字元本身**（例如統計字母 'A' 出現幾次、檢查是否有數字字元），不需要知道它排在第幾個位置時。

**模式二：利用索引範圍走訪（`for i in range(len(s)):`）——掌控全場**  
- 語法：`for i in range(len(s)):` 搭配 `s[i]` 取值。
- 運作：迴圈變數 `i` 會依序產生 `0, 1, 2, ..., len(s)-1`。
- 適用時機：**當你的邏輯需要知道「位置」**（例如只檢查偶數位置的字元、或是需要比較相鄰字元 `s[i]` 與 `s[i+1]` 是否相同）時。

**常見錯誤陷阱：在索引走訪中與相鄰元素比對時超界**  
在進行相鄰比對（如 `if s[i] == s[i+1]:`）時，迴圈終點若寫成 `range(len(s))`，在最後一輪時 `i+1` 就會直接撞牆引發 `IndexError`！相鄰比對時，迴圈上限務必設定為 `range(len(s) - 1)`。

**APCS 考試實務建議：**  
APCS 實作真題中，超過 70% 的文字題目都需要用到這兩種走訪模式！學會根據題目需求靈活切換「純字元走訪」與「索引走訪」，是通往高分級分的絕對關鍵。

````python
# [2] Code 範例區：展示兩種走訪字串的標準模式
# 說明：模式一直接取字元，模式二透過索引控制精確位置。

target = "CODE"

print("=== 模式一：直接走訪字元 (for ch in s) ===")
for ch in target:
    print("讀取到字元:", ch)

print("\n=== 模式二：索引走訪 (for i in range(len(s))) ===")
for i in range(len(target)):
    print("門牌號碼 i =", i, "-> 字元為:", target[i])
````

````python
# ==========================================
# [3] Code 填空題
# 任務說明：
# 請觀察底下的字母計數器，使用「模式一：直接字元走訪」，
# 走訪字串中的每個字元，統計單字中 'a'（小寫 a）出現的總次數！
# ==========================================

banana = "banana"
count_a = 0

# 請填入字元走訪的關鍵字與變數名稱：
for ch ___ banana:
    if ch == 'a':
        count_a += 1

# 驗證區（請勿更動下方代碼）
if count_a == 3:
    print("恭喜答對！成功使用 for ch in s 走訪字元！")
else:
    print("請檢查 for 迴圈的語法是否為 for ch in banana 喔！")
````
