### 14.7.1 題意解析：商品價格波動全距與購買決策過濾

#### 🛒 生活化比喻：特價獵手的抄底哲學
想像你是一位平時喜歡在各大電商平台比價的購物達人。在觀察心儀的商品時，如果某件衣服每天的售價都是平穩的 500 元，這說明它目前處於常態定價，沒有特別的優惠折扣。
但是，如果某款球鞋在三天之內，價格曾飆升到 1200 元，但也曾在限時快閃活動中暴跌至 400 元，兩者之間的極端價差高達 800 元！這種劇烈的價格波動，往往意味著市場出現了大幅折讓的「特賣跳樓出清」或是搶購熱潮。身為精明的買家，只要價差達到心目中的期待門檻 $d$，就是進場購買的最佳時機！

#### 📊 統計學概念：全距（Range）的意義
在統計學中，一組數據中「最大值減去最小值」的差距被稱為**全距（Range）**：
$$\text{Range} = \text{Max} - \text{Min}$$
全距是衡量一組數值分散程度或波動幅度最直觀、計算最迅速的指標。
在本題中，每一項商品都只有 3 天的價格記錄 $(p_1, p_2, p_3)$：
- 我們首先找出這 3 個數字中的最大值 $\text{Max}$ 與最小值 $\text{Min}$；
- 兩者相減即可得出該商品的價格震盪全距 $\text{diff} = \text{Max} - \text{Min}$；
- 接著與門檻值 $d$ 進行比較：**只有當 $\text{diff} \ge d$ 時，才觸發購買行動！**

#### ⚠️ 關鍵細節：大於等於（`>=`）的邊界防線
初學同學在審閱題意時，必須特別留意中文敘述中的字眼：
題目要求「最高價與最低價之差**大於或等於**門檻值 $d$」。
這意味著如果某商品的差值剛好等於門檻（例如門檻 $d = 6$，差值剛好也是 6），**這件商品仍然必須被購買**！
如果程式碼誤寫成了嚴格大於 `diff > d`，在剛好相等的邊界測資上就會遺漏購買，進而導致數量與總金額雙雙出錯而拿到 WA（Wrong Answer）。

````python
# ==============================================================================
# 14.7.1 [2] Code 範例區：計算三天價格全距並進行購買決策判定
# ==============================================================================

# 設定波動門檻值 d
d = 10

# 測試兩項不同商品的 3 天價格
item_A_prices = [10, 20, 30]
item_B_prices = [10, 12, 10]

print(f"【購買決策演示】波動門檻 d = {d}")
print("-" * 55)

for name, p in [("商品 A", item_A_prices), ("商品 B", item_B_prices)]:
    mx = max(p)  # 找出最高價
    mn = min(p)  # 找出最低價
    diff = mx - mn
    is_buy = (diff >= d)  # 判定是否大於或等於門檻
    print(f"{name} 價格：{p} | 最高價：{mx}，最低價：{mn} | 價差：{diff}")
    print(f"  --> 是否符合購買條件 (diff >= {d})？ {is_buy}")

print("-" * 55)
# 驗證結果：商品 A 價差 20 >= 10 符合購買；商品 B 價差 2 < 10 放棄購買！
````

### 14.7.2 單行三數讀取與極值計算（`max - min >= d`）

#### 📥 單行多數值的 Python 讀取慣用法
在 APCS 考場上，輸入資料中每一行包含「以空格隔開的 3 個整數價格」，例如：
`24 27 21`
初學同學在處理這種輸入時，有兩種最常見的寫法：

1. **寫法一：三個變數直接解包（Unpacking）**：
   ```python
   p1, p2, p3 = map(int, input().split())
   ```
   這種寫法非常直觀，如果後續需要特別針對第一天、第二天做單獨運算時很方便。

2. **寫法二：轉換為串列儲存（List Wrapping，本單元強烈推薦 ⭐）**：
   ```python
   prices = list(map(int, input().split()))
   ```
   為什麼寫法二更具優勢？因為把數值裝入串列後，我們可以直接將整個容器傳給 Python 原生內建函數：
   - 最高價：`max(prices)`
   - 最低價：`min(prices)`
   - 總價格：`sum(prices)`
   不需要手寫複雜的三數比大小巢狀 `if` 分支，程式碼只有短短的一行，兼具高可讀性與極速執行效能！

#### ⚡ 內建函數的執行效率
Python 的 `max()` 與 `min()` 是以底層 C 語言高度最佳化實現的內建函式，對於只有 3 個元素的迷你串列，呼叫耗時僅需數十奈秒（Nanoseconds），效能完全不用擔心。
因此，一項商品的價格波動差值可以直接精簡寫為：
```python
if max(prices) - min(prices) >= d:
    # 觸發購買決策
```
這樣的語法表達力極強，任何閱讀程式碼的人都能一秒理解你的演算法邏輯！

````python
# ==============================================================================
# 14.7.2 [2] Code 範例區：將字串轉為整數串列並計算 max 與 min
# ==============================================================================

# 模擬 input().split() 讀入的單行價格字串
raw_input_line = "24 27 21"

# 1. 字串分割並轉為整數串列
prices = list(map(int, raw_input_line.split()))
print(f"轉換後的整數串列：{prices}")

# 2. 調用內建函數快速提取極值
highest = max(prices)
lowest = min(prices)
diff = highest - lowest

print(f"最高價格：{highest}")
print(f"最低價格：{lowest}")
print(f"價格全距 (最高 - 最低)：{diff}")

# 3. 驗證比對：27 - 21 = 6，完全吻合！
````

### 14.7.3 整數除法平均值計算：`sum(prices) // 3`

#### 💰 購入成本的定義與計算公式
當某項商品通過價格波動審核（$\text{max} - \text{min} \ge d$）之後，題目規定：
> 購買該商品的費用為**該商品近 3 天價格的平均值**。

計算平均值在數學上是三者相加除以 3：
$$\text{Average} = \frac{p_1 + p_2 + p_3}{3}$$
在 Python 語言中，我們可以利用 `sum(prices)` 瞬間將串列中的所有價格加總。

#### 🚨 世紀陷阱：浮點數除法 `/` vs 整數除法 `//`
很多初學同學在算平均時，習慣順手寫下：
```python
avg = sum(prices) / 3  # ❌ 危險的浮點數除法！
```
在 Python 3 中，單斜線 `/` 是**浮點數除法（True Division）**，即使總和可以被 3 整除，其運算結果也必定會強制轉換為浮點數（`float`）！
例如：
`(24 + 27 + 21) / 3 = 72 / 3 = 24.0`
如果你的程式輸出 `24.0`，在線上評判系統（Online Judge）嚴格的字元比對裁判眼裡：
- 官方標準答案要求輸出整數：`1 24`
- 你的程式輸出包含小數點：`1 24.0`
系統會毫不留情地判定為 **Presentation Error (PE)** 或 **Wrong Answer (WA)**！

#### 🛡️ 安全解法：雙斜線整數除法 `//`
題目在規格中已經非常貼心地做出保證：**所有符合購買條件的商品，其三天價格平均值保證必為整數**。
因此，我們應該堅決使用雙斜線 `//`（地板除法，Floor Division）：
```python
avg = sum(prices) // 3  # ✅ 產生乾淨俐落的純整數 int，例如 24
```
使用 `//` 不僅能夠保證輸出型態完全符合整數要求，還能避免任何浮點數二進位精度誤差，是競技程式設計中的標準防衛姿勢！

````python
# ==============================================================================
# 14.7.3 [2] Code 範例區：對比浮點數除法 / 與整數除法 // 的輸出差異
# ==============================================================================

prices = [24, 27, 21]
total_sum = sum(prices)

print(f"三天價格串列：{prices}，總和 sum = {total_sum}")
print("-" * 55)

# ❌ 使用浮點數除法 /
float_avg = total_sum / 3
print(f"❌ 浮點數除法 (/) 結果：{float_avg} | 型態：{type(float_avg)}")
print("  --> 警告：若印出 24.0 會被評測系統視為格式錯誤 WA！")

# ✅ 使用整數除法 //
int_avg = total_sum // 3
print(f"✅ 整數除法 (//) 結果：{int_avg} | 型態：{type(int_avg)}")
print("  --> 完美符合官方格式要求，安全滿分！")
print("-" * 55)
````

### 14.7.4 雙計數器模式：購買總件數與花費總金額

#### 📊 競技程式經典架構：雙指標統計模式
在 APCS 許多統計題目中，我們常常需要在走訪過程中同時維護兩種不同的累計量：
1. **頻率計數器（Counter）**：統計「符合條件的事件發生了幾次」（例如：購買了幾件商品）；
2. **數值累加器（Accumulator）**：統計「符合條件事件所產生的數值總和」（例如：購買這些商品的總花費）。

這種「計數器 + 累加器」的雙搭檔，我們統稱為**雙計數器模式（Dual Counter Pattern）**。

#### ⚙️ 雙計數器的初始化與狀態轉移
在處理任何資料之前，必須先在迴圈外部設定好乾淨的初始狀態：
```python
total_items = 0  # 記錄購買總件數，初始為 0
total_cost = 0   # 記錄購買總花費，初始為 0
```
當我們審核某一項商品時：
- 如果全距 $\text{max} - \text{min} < d$：條件不符，什麼都不做（略過）；
- 如果全距 $\text{max} - \text{min} \ge d$：符合購買資格！
  - 件數計數器加一：`total_items += 1`
  - 金額累加器加上該項商品的平均價：`total_cost += sum(prices) // 3`

#### 📤 輸出的字元級對齊規範
題目要求：「輸出僅有一行，包含兩個整數，兩數之間以一個半形空格隔開」。
在 Python 中，呼叫 `print(a, b)` 時，預設的參數分隔符號 `sep` 就是單一半形空格 `' '`，結尾 `end` 預設就是換行字元 `'\n'`。
因此，我們只需一行最樸素的語法：
```python
print(total_items, total_cost)
```
就能自動輸出完全符合官方評測格式的答案，既安全又不需要手動拼接字串！

````python
# ==============================================================================
# 14.7.4 [2] Code 範例區：雙計數器模式（件數累加與花費累計）
# ==============================================================================

# 模擬官方範例 2 的情境：門檻 d = 10，共有兩件商品
d = 10
all_items_prices = [
    [10, 20, 30],  # 商品 1：價差 20 >= 10（買！平均價 20）
    [10, 12, 10]   # 商品 2：價差 2 < 10（不買）
]

# 1. 初始化雙計數器
total_items = 0
total_cost = 0

# 2. 逐一審核每項商品
for idx, prices in enumerate(all_items_prices, 1):
    diff = max(prices) - min(prices)
    if diff >= d:
        avg_price = sum(prices) // 3
        total_items += 1         # 件數 + 1
        total_cost += avg_price  # 花費累加
        print(f"商品 {idx}：符合條件！購入價格 = {avg_price}")
    else:
        print(f"商品 {idx}：波動不足，放棄採購。")

# 3. 輸出最終統計結果
print("-" * 50)
print(f"APCS 標準格式輸出：{total_items} {total_cost}")
````

### 14.7.5 逐筆處理與多行輸入迴圈完整 AC 實作

#### 🌊 在線處理演算法（Online Algorithm）的威力
在處理多行輸入的競賽題目時，初學者常會產生一種直覺：
「我是不是應該先開一個超大的二維串列，把所有 $n$ 行的價格全部 `.append()` 存起來，之後再寫第二個迴圈慢慢算？」
答案是：**完全不需要！**

這種「讀入一筆資料、立刻進行計算、更新統計量後就拋棄該筆原始記憶體」的技術，在計算機科學中稱為**在線演算法（Online Algorithm）**或**串流處理（Stream Processing）**：
1. **記憶體極致節省**：我們每次只需要維持「當前這一行」的 3 個整數變數，空間複雜度永遠是輕量的 $O(1)$；
2. **程式結構極簡**：只需一個單層的 `for _ in range(n):` 迴圈，邊讀邊算，直覺順暢；
3. **考場打字負擔最小**：整套程式連同輸入與輸出，僅僅 10～12 行代碼即可全數搞定！

#### 🧩 手把手拼裝滿分流程
讓我們回顧組裝步驟：
1. 讀入第 1 行的總數量與門檻：`n, d = map(int, input().split())`
2. 初始化雙計數器：`buy_count = 0`, `total_cost = 0`
3. 展開 $n$ 次迴圈：`for _ in range(n):`
   - 讀入當行 3 天價格：`prices = list(map(int, input().split()))`
   - 判斷波動門檻：`if max(prices) - min(prices) >= d:`
   - 同步更新雙計數器：`buy_count += 1`、`total_cost += sum(prices) // 3`
4. 迴圈結算後一行印出：`print(buy_count, total_cost)`

沒有任何多餘的包裝，這就是 APCS 考場取得滿分最精準、最快速的標準模板！

````python
# ==============================================================================
# 14.7.5 [2] Code 範例區：完整在線串流處理演算法（以範例 2 模擬輸入）
# ==============================================================================

# 模擬第一行：n = 2, d = 10
n, d = 2, 10

# 模擬接下來的 n 行輸入串流
input_stream = [
    "10 20 30",
    "10 12 10"
]

# 1. 初始化統計變數
buy_count = 0
total_cost = 0

# 2. 逐筆在線處理每一行商品
for line in input_stream:
    prices = list(map(int, line.split()))
    # 判定波動門檻
    if max(prices) - min(prices) >= d:
        buy_count += 1
        total_cost += sum(prices) // 3

# 3. 輸出滿分結果
print(f"APCS 評判系統預期輸出：{buy_count} {total_cost}")
````

### 14.7.6 一件都沒買（件數為 0、花費為 0）之邊界檢驗與 WA 排查

在 APCS 評測系統中，主辦單位最喜歡設計「冷門極限案例（Corner Cases）」來考驗學生的程式健壯度。
對於本題而言，最經典的極端案例就是：**市場上沒有任何一件商品達到購買門檻！**

- 門檻 $d = 100$ 設定得極高，但所有商品的價差頂多只有 10 或 20；

### 📝 版本一：APCS 官方實作考場專用版 —— 淺顯易懂一般版（新手友善推薦）

* **適用情境**：APCS 正式考試現場、初學者課堂練習。
* **教學與設計理念**：
  1. **步驟平鋪直敘、拆解詳細**：先讀取商品總數量 `n` 與價格波動門檻 `d`，接著初始化購買總件數 `buy_count` 與花費金額 `total_cost`。
  2. **標準迴圈逐筆處理**：使用 `for` 迴圈搭配直觀變數讀取每項商品的三天價格，計算最高價 `max(prices)` 與最低價 `min(prices)` 的差額。
  3. **穩健 100% 滿分 AC**：嚴格檢查價差 `diff >= d`，平均花費使用整數除法 `// 3` 累加，變數語意一清二楚，考場零壓力輕鬆拿滿分！


````python
# ==============================================================================
# 📝 版本一：APCS 官方實作考場專用版 —— 淺顯易懂一般版（新手友善推薦）
# 適用情境：APCS 正式考場單筆測試案例，平鋪直敘、步驟明確、穩健滿分
# ==============================================================================

# 步驟 1：讀取第一行商品總數量 n 與價格波動門檻 d
first_line = input().split()
n = int(first_line[0])
d = int(first_line[1])

# 步驟 2：初始化雙計數器（購買件數與總花費金額）
buy_count = 0
total_cost = 0

# 步驟 3：逐一巡視 n 項商品近 3 天的價格
for i in range(n):
    # 讀取當前商品的三天價格
    price_strings = input().split()
    p1 = int(price_strings[0])
    p2 = int(price_strings[1])
    p3 = int(price_strings[2])
    prices = [p1, p2, p3]
    
    # 計算三天最高價與最低價的差額
    highest = max(prices)
    lowest = min(prices)
    diff = highest - lowest
    
    # 判斷波動差額是否達到購買門檻 d
    if diff >= d:
        buy_count += 1
        # 題目保證總和可被 3 整除，使用整數除法 // 保持純整數
        avg_price = (p1 + p2 + p3) // 3
        total_cost += avg_price

# 步驟 4：依題目要求印出「購買件數」與「總花費」（以空格隔開）
print(buy_count, total_cost)
````

### ⚡ 版本二：APCS 官方實作考場專用版 —— 極簡高效精煉版（進階高手推薦）

* **適用情境**：APCS 現場正式檢定考試（追求極速敲碼、極致優雅的高階模式）。
* **教學與設計理念**：
  1. **列表推導式一次性讀取篩選**：利用 `[p for p in items if max(p) - min(p) >= d]` 一行萃取出所有符合購買門檻的商品價格！
  2. **高階聚合運算直接輸出**：購買件數即為 `len(bought)`，總花費以生成器運算式 `sum(sum(p) // 3 for p in bought)` 直接求和。
  3. **僅僅 5 行極速奪分**：全篇僅 5 行核心代碼，結構短小精悍，大幅節省考場作答與除錯時間！


````python
# ==============================================================================
# ⚡ 版本二：APCS 官方實作考場專用版 —— 極簡高效精煉版（進階高手推薦）
# 適用情境：APCS 正式考試現場，極簡 5 行高效通關
# ==============================================================================

# 步驟 1：讀取 n 與 d
n, d = map(int, input().split())

# 步驟 2：推導式讀入所有商品並篩選出符合門檻者
items = [list(map(int, input().split())) for _ in range(n)]
bought = [p for p in items if max(p) - min(p) >= d]

# 步驟 3：輸出購買件數與總平均花費
print(len(bought), sum(sum(p) // 3 for p in bought))
````
