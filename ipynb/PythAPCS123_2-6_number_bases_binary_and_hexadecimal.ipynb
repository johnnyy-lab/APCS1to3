### 💡 2-6-1 電腦的原生母語——二進位常數表示法（0b）

各位學習者在日常生活中買文具、算分數，用的都是「**十進位（Decimal）**」，也就是每逢 10 就進一位（數字只有 0 到 9）。
但是，電腦大腦內部的晶片是由無數個微小的電晶體構成的，電晶體只有兩種狀態：**通電（開 / 1）** 與 **斷電（關 / 0）**！這就是為什麼電腦的母語是「**二進位（Binary）**」！

**二進位是怎麼計數的呢？**
二進位是「**逢 2 就進一位**」！每一位的權重都是 2 的次方：
* 第 1 位（最右邊）：權重為 $2^0 = 1$
* 第 2 位：權重為 $2^1 = 2$
* 第 3 位：權重為 $2^2 = 4$
* 第 4 位：權重為 $2^3 = 8$
例如二進位的 `101`，代表的就是：$1 \times 4 + 0 \times 2 + 1 \times 1 = 5$！

**在 Python 程式碼中，該如何直接宣告一個二進位常數？**
吳邦一老師在講義第 10 頁特別指引：只要在數字開頭加上前綴「**`0b`**」或「**`0B`**」（數字 0 接英文字母 b，b 代表 Binary 二進位）！
例如宣告：
```python
x = 0b10101
```
這段二進位常數存進電腦記憶體後，**它本質上就是一個標準純正的整數（int）**！當你用 `print(x)` 印出時，Python 會自動將它翻譯成人類看得懂的十進位整數 `21`！

**⚠️ 語法報錯禁忌**：
二進位的前綴是數字 0 接字母 b，後面**絕對只能出現數字 0 與 1**！
如果你手滑寫了 `0b102`，直譯器會立刻報錯 `SyntaxError: invalid digit '2' in binary literal`（二進位常數中出現無效的數字 2）！

````python
# [2] Code 範例區：展示二進位常數 0b 的宣告與十進位輸出
# 範例 1：0b101 代表 1*4 + 0*2 + 1*1 = 5
num_a = 0b101
print(num_a)        # 印出 10 進位的 5
print(type(num_a))  # <class 'int'>，本質依然是整數

# 範例 2：0b1111 代表 8 + 4 + 2 + 1 = 15
num_b = 0b1111
print(num_b)        # 印出 15

# 範例 3：講義範例 0b10101 代表 16 + 4 + 1 = 21
num_c = 0b10101
print(num_c)        # 印出 21
````

### 💎 2-6-2 資訊科學的高速密碼——十六進位常數表示法（0x）

二進位雖然完美契合電腦硬體，但它有一個超級大缺點：**字串實在太長了！**
例如一個簡單的數字 255，二進位要寫成 `11111111`（整整 8 個數字），如果是一億，那長度簡直會讓人眼花撩亂！
為了解決這個問題，工程師們發明了資訊科學的超強壓縮法寶——「**十六進位（Hexadecimal）**」！

**為什麼選十六進位？**
因為 $2^4 = 16$！這代表**每 4 個二進位位元（Bit），剛好可以完美濃縮成「1 個」十六進位字元**！
原本長長的一串二進位，瞬間縮短為四分之一的長度，閱讀起來無比優雅簡潔！

**十六進位的 16 個數字符號**：
十六進位是「**逢 16 進一位**」。但我們平時只有 0 到 9 十個阿拉伯數字，超過 9 的數字怎麼辦？
十六進位請來了 6 位英文字母大將支援：
* `0 ~ 9`：代表數值 0 到 9
* `a` 或 `A`：代表 10
* `b` 或 `B`：代表 11
* `c` 或 `C`：代表 12
* `d` 或 `D`：代表 13
* `e` 或 `E`：代表 14
* `f` 或 `F`：代表 15

**在 Python 程式碼中如何宣告十六進位常數？**
吳邦一老師在講義第 10 頁介紹：使用前綴「**`0x`**」或「**`0X`**」（數字 0 接英文字母 x，x 代表 Hexadecimal 十六進位）！
例如：
1. `0x41`：代表 $4 \times 16 + 1 = 65$（在 ASCII 編碼中，65 剛好就是大寫英文字母 `'A'`！）
2. `0xff`：代表 $15 \times 16 + 15 = 255$（網頁色彩 RGB 代碼中全亮飽和的紅色 `#FF0000`，最大亮度就是 255！）
3. `0x10`：代表 $1 \times 16 + 0 = 16$（逢 16 就進位變成 10）！

十六進位字母大小寫完全通用（`0xff` 與 `0xFF` 完全一樣），是日後探索網路協定、記憶體位址與圖形色彩不可或缺的核心工具！

````python
# [2] Code 範例區：展示十六進位常數 0x 的宣告與十進位輸出
# 範例 1：ASCII 字母 'A' 的十六進位編碼 0x41
char_a = 0x41
print(char_a)        # 印出 65（4 * 16 + 1 = 65）

# 範例 2：RGB 滿分色值 0xff
max_color = 0xff
print(max_color)     # 印出 255（15 * 16 + 15 = 255）

# 範例 3：字母大小寫皆合法通用
val_lower = 0x2f
val_upper = 0x2F
print(val_lower)     # 2*16 + 15 = 47
print(val_upper)     # 47
print(val_lower == val_upper)  # True
````

### 🔄 2-6-3 十進位與進位制的雙向翻譯機——bin() 與 hex() 內建轉換函數

在前面兩節中，我們學會了「由我們寫下 `0b` 或 `0x`，電腦自動幫我們翻譯成十進位整數」。
但如果反過來呢？如果我們手上有一個普通的十進位整數（例如 `42`），我們想要親眼看看它的二進位長相，或是十六進位長相，該怎麼做？

Python 貼心地為我們準備了兩位超級「**進位制即時口譯官**」：

**1. 二進位口譯官：`bin(整數)`**
* 傳入任何整數，它會立刻將該數字翻譯成二進位表示法的文字！
* 例如執行 `bin(13)`，會回傳字串 `'0b1101'`！

**2. 十六進位口譯官：`hex(整數)`**
* 傳入任何整數，它會立刻將該數字翻譯成十六進位表示法的文字！
* 例如執行 `hex(255)`，會回傳字串 `'0xff'`！

**🔍 關鍵型態細節：回傳型態是「字串（str）」！**
請大家特別留意：`bin()` 與 `hex()` 轉換出來的結果型態是「**字串（str，帶有引號）**」！
而且字串的開頭會非常貼心地自帶標籤：
* 二進位前面必定帶有 `'0b'`
* 十六進位前面必定帶有 `'0x'`

在 APCS 競賽或日常除錯時，如果你算完某個複雜數字，想快速檢驗某個位元到底有沒有亮起（是 0 還是 1），只要把數字丟進 `bin()` 裡印出來，真相瞬間一目了然！

````python
# [2] Code 範例區：展示 bin() 與 hex() 的雙向轉換功能
number = 42

# 使用 bin() 轉換為二進位字串
bin_str = bin(number)
print(bin_str)        # 印出 '0b101010'（32 + 8 + 2 = 42）
print(type(bin_str))  # <class 'str'>

# 使用 hex() 轉換為十六進位字串
hex_str = hex(number)
print(hex_str)        # 印出 '0x2a'（2 * 16 + 10 = 42）
print(type(hex_str))  # <class 'str'>

# 驗證：轉換後的字串可以直接放回 Python 當成數值
print(0b101010 == 42) # True
print(0x2a == 42)     # True
````

### 🎛️ 2-6-4 APCS 實戰視野：進位制在狀態開關與位元遮罩（Bitmask）的解題先導

學到這裡，你可能會好奇地問：
「老師！平常寫程式用十進位好好的，為什麼 APCS 考試和競技演算法要特別介紹 `0b` 與 `0x` 呢？」

吳邦一老師在講義第 17.1 節（位元運算）為我們揭示了高階演算法的精髓：
👉 **「在競技程式中，整數常常被拿來代表一組物品的選取狀態（狀態壓縮）！」**

**🎒 經典情境：背包寶物選取狀態**
假設勇者眼前有 4 樣神奇寶物：【寶劍、盾牌、藥水、地圖】。
我們可以用一個 4 位元的二進位數字來精準記錄勇者拿了什麼：
* `1` 代表「有拿」
* `0` 代表「沒拿」

如果勇者拿了【寶劍(1)、沒拿盾牌(0)、拿了藥水(1)、沒拿地圖(0)】：
* 如果你把這個狀態存成十進位數字 `10`，別人看程式碼根本不知道 10 是什麼意思！
* 但如果你在程式中直接寫：
  ```python
  bag_status = 0b1010
  ```
  任何讀程式的人一眼看過去，第 4 樣有拿、第 3 樣沒拿、第 2 樣有拿、第 1 樣沒拿，**狀態開關清清楚楚、一目了然**！

**位元遮罩（Bitmask）的威力**：
用一個簡單的整數就能一口氣儲存 30 個開關或選項的狀態，不僅記憶體占用極小，執行速度更達到毫秒級！
學會以 `0b` 與 `0x` 觀察狀態，就是通往單元 2-7「位元運算子（AND、OR、XOR、位移）」的黃金入場券！

````python
# [2] Code 範例區：展示使用 0b 直觀表達物品選取狀態與開關集合
# 4 件物品：[寶劍, 盾牌, 藥水, 地圖]
# 選取狀態：選了寶劍(1)與藥水(1)，未選盾牌(0)與地圖(0)
bag_state = 0b1010
print(bag_state)  # 儲存於電腦中，數值為十進位 10

# 檢驗：4 件全選的滿格狀態（全部為 1）
full_state = 0b1111
print(full_state) # 印出 15

# 十六進位狀態遮罩：在進階系統中常用 0xFF 表示 8 個開關全開
mask_8bits = 0xff
print(mask_8bits) # 印出 255
````

2. **十六進位常數（0x）**：二進位的 4 倍壓縮包，前綴 `0x`（如 `0x41 = 65`、`0xff = 255`），在 ASCII 編碼與色彩工程扮演不可取代的角色。
3. **雙向翻譯神器（bin / hex）**：`bin()` 與 `hex()` 輕鬆將整數翻譯為自帶標籤的字串，直觀透視數值位元長相。
4. **APCS 狀態壓縮先導（Bitmask）**：用二進位代表物品挑選與開關集合，以最少記憶體達成極致可讀性！
