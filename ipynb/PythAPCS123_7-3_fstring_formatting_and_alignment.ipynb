### 🎫 7-3-1 填空車票——`f"..."` 把變數嵌進字串

想像你在寫一張成績通知單。紙上印好了固定的句子：「同學 ____ 的分數是 ____ 分。」兩個空格要填人名和分數。單元 7-2 的做法，是把句子剪成好幾段，再用加號黏回去：

```python
notice = "同學 " + name + " 的分數是 " + str(score) + " 分。"
```

人名是字串，可以直接黏；分數是整數，不先變成字串就會被電腦拒絕，丟出 `TypeError`。空格、括號、引號一多，眼睛很容易漏掉其中一個。

格式化字串（f-string）把這張通知單變成一張完整的車票。做法只有兩步：

1. 在開頭的引號前面加一個小寫字母 `f`。沒有這個 `f`，大括號就只是普通文字。
2. 空格的位置改寫成 `{變數名}`。電腦印出這行時，會把大括號連同裡面的名字換成變數現在裝的值。

```python
notice = f"同學 {name} 的分數是 {score} 分。"
```

分數不用再呼叫 `str()`。f-string 在印出之前，會自己把數字轉成文字。引號仍然要成對，單引號或雙引號都可以，規則和單元 7-1 一樣。

**它沒有改掉原來的變數。** `name` 還是那個人名，`score` 還是那個數字。f-string 只是做出一個新的字串交給你。你可以先存進 `notice` 再 `print`，也可以直接 `print(f"...")`。

**考場用法：** 輸出若要夾著文字說明，例如「答案是 42」，用 f-string 比一段一段加號更不容易漏空格。這一節先只放變數，下一小節才把算式放進大括號。



````python
# [2] Code 範例區：用 f-string 把變數嵌進句子
name = "小明"
score = 90

# 引號前的 f，以及大括號裡的變數名
notice = f"同學 {name} 的分數是 {score} 分。"
print(notice)

# 也可以直接印，不必先存起來
print(f"你好，{name}")
````

### 🧮 7-3-2 大括號裡的小算盤——先算完再印出

上一小節的大括號裡只放變數名，像在車票空格貼上名牌。其實大括號裡面可以放一條小型算式。電腦會先把算式算完，再把結果填進句子。

```python
print(f"{a} 加 {b} 等於 {a + b}")
```

閱讀順序是：先看 `{a}` 和 `{b}` 各自的值，再看 `{a + b}`。加號是在大括號裡面做的，不是在引號外面把字串黏起來。所以這裡不需要 `str()`，也不需要單元 7-2 那種一長串加號。

可以放進大括號的，是你已經學過的運算：`+`、`-`、`*`、`/`、`//`、`%`，以及單元 2.5 的 `round(數值, 位數)`。例如 `{round(total / 3, 2)}` 會先做除法、再四捨五入，最後才變成文字。

**括號要數清楚。** f-string 自己的大括號包住整段算式。如果算式裡還有函式，函式的小括號必須在大括號結束之前就閉合：`{round(x, 2)}` 是對的，`{round(x, 2)` 少了一個 `}` 就會在執行前直接語法錯誤。

**不要在這一節把整段程式塞進大括號。** 大括號適合「一個結果」。若你要先判斷、再迴圈，那些步驟仍寫在 f-string 外面，最後只把答案變數放進去。這樣除錯時才知道錯在算式，還是錯在句子。



````python
# [2] Code 範例區：大括號裡先算完再填進句子
a = 8
b = 5
print(f"{a} 加 {b} 等於 {a + b}")

price = 19.9
count = 3
print(f"三件一共 {round(price * count, 2)} 元")
````

### ✂️ 7-3-3 小數固定位數——`{:.2f}` 四捨五入並自動補零

單元 2.5 的 `round(7.5, 2)` 得到的是數字 `7.5`。`print` 印出來也是 `7.5`，不會畫出尾端那個沒有意義的 0。考卷若規定「輸出到小數點後兩位」，裁判比的是畫面上的字，`7.5` 和 `7.50` 會被看成不同的答案。

f-string 的冒號就是在交代「這個空格要怎麼畫」。寫法是：

```python
print(f"{number:.2f}")
```

拆開來記：

* `{` 與 `}` 仍然是車票空格。
* `number` 是要印的值。
* `:` 後面開始是格式說明，不是數學的除法。
* `.2` 表示小數點後留下 2 位。
* `f` 表示這是浮點數的畫法（float），不夠的位數用 0 補上。

所以 `7.5` 會被畫成 `7.50`，`10 / 3` 會先變成很長的小數，再四捨五入畫成 `3.33`。要留三位就寫 `{:.3f}`，要留一位就寫 `{:.1f}`。

**它和 `round` 的差別：** `round` 交還的是數字，方便接著運算；`{:.2f}` 交還的是已經排好的文字，專門給 `print` 或存成字串。若你還要把結果拿去加減，先用 `round` 留下數字；若你只是要交卷，用 `{:.2f}`。

**常見抄錯：** 把冒號寫到外面，或寫成 `{.2f}` 卻忘了變數名。空格裡必須先有值，才有冒號。正確順序永遠是 `{值:格式}`。



````python
# [2] Code 範例區：小數固定印出指定位數
number = 7.5
print(f"{number:.2f}")   # 7.50，尾端補零

val = 3.1415926
print(f"{val:.2f}")      # 3.14
print(f"{val:.4f}")      # 3.1416

print(f"{10 / 3:.2f}")   # 3.33
````

### 📐 7-3-4 整數對齊寬度——`{:4d}` 預留整齊欄位

成績單或矩陣若每個數字長短不一，直接換行印，個位數和十位數會擠成一團。f-string 可以先向右預留固定的字元寬度，短的數字左邊補空白，像把數字放進同樣寬的格子。

```python
print(f"{n:4d}")
```

* `4` 是這一格至少要佔 4 個字元。
* `d` 表示這是整數（decimal integer）的畫法。
* 數字若不到 4 位，左邊補空白，不是補 0。`7` 會變成三個空白再加 `7`。
* 數字若已經比 4 位還長，格子會撐開，不會把數字砍掉。`10000` 仍然完整印出。

直著看三行 `{:4d}`，個位數會落在同一條直線上。這就是以後單元 9.2 印矩陣時，`f"{val:4d}"` 的來源。現在先在一維的數字上練會，不必碰串列。

**小數不要用 `d`。** `d` 只接受整數。把 `3.5` 放進 `{:4d}` 會產生 `ValueError`。小數對齊用上一小節的 `f`，例如 `{:.2f}`。整數對齊用 `d`。

**寬度算的是字元，不是「幾位數的感覺」。** 負號也佔一格。`-7` 用寬度 4，會是兩個空白、一個負號、再一個 7。考卷若要對齊含負數的欄位，記得把負號算進去。



````python
# [2] Code 範例區：整數預留固定寬度
print(f"{7:4d}")
print(f"{42:4d}")
print(f"{100:4d}")

# 負號也算一格
print(f"{-7:4d}")
````

### 🚧 7-3-5 三種常見翻車——漏寫 f、括號不成對、引號撞車

f-string 很好寫，也很容易在三個小地方翻車。這一節把這三種錯誤分開看，每一種都只改一個地方就能修好。

**第一種：忘了引號前面的 `f`。**  
寫成 `"分數是 {score}"` 時，電腦不知道大括號是空格，就會把 `{score}` 這六個字元原樣印出來。畫面上看起來像印了程式碼，而不是分數。修好的方法只有一個：讓引號前面有 `f`。

**第二種：大括號沒有成對。**  
`f"合計 {a + b"` 少了右邊的 `}`。這不是執行到一半才錯，而是語法錯誤 `SyntaxError`，整格程式都不會開始跑。數括號的方法和數引號一樣：每一個 `{` 都要有一個 `}` 在它後面，而且要包住完整的算式。

**第三種：引號撞車。**  
f-string 若用雙引號包住整句，句子裡面又想放雙引號，電腦會以為字串提前結束。和單元 7-1 相同，可以改成外層單引號、內層雙引號，或是外層雙引號、內層單引號。例如：`f"他說：'你好'"`。

**考場自檢三問：** 引號前有沒有 `f`？大括號有沒有成對？句子裡的引號有沒有和外層用同一種而沒有跳脫？三問都過了，再看格式是 `f` 還是 `d`。13.2 會把這些錯誤放進語法除錯清單，到時候請回到這一節對照。



````python
# [2] Code 範例區：對照漏寫 f 的原樣輸出，以及修好的寫法
score = 80

# 沒有 f 時，大括號會被當成普通文字
print("分數是 {score}")

# 加上 f 之後，才會填入變數的值
print(f"分數是 {score}")

# 外層雙引號、內層單引號，避免引號撞車
print(f"他說：'分數是 {score}'")
````
