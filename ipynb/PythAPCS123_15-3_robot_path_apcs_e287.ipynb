### 📝 題目描述（Problem Description）
有一個探勘機器人被放置在一個 $R \times C$ 的矩形網格地圖上。地圖中的每一個格子都有一個整數數值，且**整張地圖中所有格子的數值皆不相同**。

機器人的移動規則如下：
1. **起點選擇**：機器人會從**全地圖中數值最小**的格子出發。
2. **移動決策**：每一步，機器人會觀察目前所在位置的**上下左右四個相鄰格子**。在所有「在邊界內」且「尚未走過」的相鄰格子中，挑選**數值最小**的那一格移動過去。
3. **停止條件**：若當前位置周圍四個方向的相鄰格子都「已經走過」或「超出地圖邊界」，機器人便無法繼續移動，任務即告終止。

請撰寫一個程式，計算機器人從出發到停止移動所經過的**所有格子的數值總和**（包含起點與終點格）。

---

### 📥 輸入格式（Input Format）
第一行包含兩個以空白隔開的正整數 $R$ 與 $C$，分別代表地圖的列數（Row）與行數（Column）。  
接下來的 $R$ 行，每行包含 $C$ 個非負整數，依序代表地圖中每一格的地形數值。

---

### 📤 輸出格式（Output Format）
請輸出一個整數，代表機器人所走過所有格子的數值總和。

---

### ⚠️ 測資範圍與評分規範（Constraints）
* $1 \le R, C \le 100$
* 格子數值均為介於 $0$ 到 $10^6$ 之間的整數
* 題目保證全地圖中**所有格子的數值皆不相同**，因此最小起點與每次移動方向皆唯一確定。

#### 🏞️ 生活直觀比喻：低谷探險家

### 15.3.2 全網格雙重走訪：精確鎖定全局最小值座標 `(min_r, min_c)` 作為起點

#### 🔍 二維極值搜尋演算法
在二維網格中尋找起點，必須進行雙層迴圈走訪：
1. 初始化 `min_val = float('inf')`、`start_r = 0`、`start_c = 0`。
2. 外層迴圈枚舉列號 $i$（$0 \le i < R$），內層迴圈枚舉行號 $j$（$0 \le j < C$）。
3. 當 `grid[i][j] < min_val` 時，同步更新最小值與座標。


````python
# =====================================================================
# [2] Code 範例：全網格搜尋最小值座標
# =====================================================================

grid = [
    [9, 1, 8, 14],
    [7, 2, 5, 13],
    [6, 3, 4, 12]
]
r = len(grid)
c = len(grid[0])

min_val = float('inf')
start_r, start_c = -1, -1

for i in range(r):
    for j in range(c):
        if grid[i][j] < min_val:
            min_val = grid[i][j]
            start_r, start_c = i, j

print(f"找到全域起點：({start_r}, {start_c})，數值為 {min_val}")
````

### 15.3.4 邊界防護與已拜訪標記：`visited` 布林矩陣防止迴圈打轉

#### 🚫 兩大安全防護門檻：
1. **邊界防護（Boundary Guard）**：  
   相鄰座標 $(nr, nc)$ 必須滿足 $0 \le nr < R$ 且 $0 \le nc < C$，嚴防 `IndexError` 與負數索引倒取。
2. **拜訪標記（Visited Flag）**：  
   使用 `visited = [[False] * C for _ in range(R)]`。每走一步將該格設為 `True`，避免機器人在兩格之間無限橫跳打轉！


````python
# =====================================================================
# [2] Code 範例：結合邊界與布林標記之有效鄰居過濾
# =====================================================================

r, c = 3, 3
visited = [[False] * c for _ in range(r)]
visited[0][0] = True  # 起點已走過

cur_r, cur_c = 0, 1
dr = [-1, 1, 0, 0]
dc = [0, 0, -1, 1]

for k in range(4):
    nr = cur_r + dr[k]
    nc = cur_c + dc[k]
    if 0 <= nr < r and 0 <= nc < c:
        status = "已造訪過" if visited[nr][nc] else "可前進 (合法未造訪)"
    else:
        status = "越界無效"
    print(f"方向 ({nr:2d}, {nc:2d}) ➔ {status}")
````

````python
# =====================================================================
# [4] Code 練習題：合法未造訪鄰居計數
# 任務說明：實作 count_valid_neighbors(cr, cc, r, c, visited)
# =====================================================================

def count_valid_neighbors(cr, cc, r, c, visited):
    cnt = 0
    for dr, dc in ((-1,0), (1,0), (0,-1), (0,1)):
        nr, nc = cr + dr, cc + dc
        if 0 <= nr < r and 0 <= nc < c and not visited[nr][nc]:
            cnt += 1
    return cnt

vis = [[False]*2 for _ in range(2)]
vis[0][0] = True
assert count_valid_neighbors(0, 0, 2, 2, vis) == 2
print("✅ 練習題所有 assert 測試通過！")
````

### 15.3.5 四鄰居極小值決策引擎：有效鄰居數值比較與最佳下一步判定

#### 🎯 單步最佳化挑選機制
在當前位置 $(cur\_r, cur\_c)$：
1. 宣告 `smallest_neighbor = float('inf')`、`next_r = -1`、`next_c = -1`。
2. 走訪四個方向，凡滿足「未越界」且「尚未走過」者，比對其數值：
   若 `grid[nr][nc] < smallest_neighbor`，則更新候選目的地。
3. 四方向皆檢查完後，若 `next_r == -1`，代表**四面受阻無路可走**！


````python
# =====================================================================
# [2] Code 範例：單步挑選最小未走鄰居展示
# =====================================================================

grid = [
    [9, 1, 8, 14],
    [7, 2, 5, 13],
    [6, 3, 4, 12]
]
r, c = len(grid), len(grid[0])
visited = [[False] * c for _ in range(r)]
visited[0][1] = True  # 當前在 (0, 1) 數值 1

cur_r, cur_c = 0, 1
smallest_val = float('inf')
next_r, next_c = -1, -1

for dr, dc in ((-1,0), (1,0), (0,-1), (0,1)):
    nr, nc = cur_r + dr, cur_c + dc
    if 0 <= nr < r and 0 <= nc < c and not visited[nr][nc]:
        if grid[nr][nc] < smallest_val:
            smallest_val = grid[nr][nc]
            next_r, next_c = nr, nc

print(f"在 (0, 1) 數值 1 處，選出最小鄰居為 ({next_r}, {next_c}) 數值 {smallest_val}")
````

### 15.3.7 解法升級：哨兵外圍加框法（Sentinel Padding）免邊界檢查極速實作

#### 🏰 什麼是哨兵加框法？
在二維網格尋路中，每次都要檢查 `0 <= nr < r and 0 <= nc < c` 顯得十分繁瑣。
如果我們在整張地圖的最外圍，包上一圈**極大值哨兵（例如 $\infty$）**：
* 擴展後的網格大小為 $(R+2) \times (C+2)$。
* 原始地圖座標 $(r, c)$ 對應到加框地圖 $(r+1, c+1)$。
* 外圍一圈哨兵因為數值為 $\infty$，**永遠不可能成為最小值候選人**，因此完全不需要寫邊界檢查，程式碼極致精煉！


````python
# =====================================================================
# [2] Code 範例：哨兵加框法建立與走訪示範
# =====================================================================

raw_grid = [
    [9, 1, 8, 14],
    [7, 2, 5, 13],
    [6, 3, 4, 12]
]
r, c = len(raw_grid), len(raw_grid[0])
INF = float('inf')

# 上下左右包一圈 INF
padded = [[INF] * (c + 2)]
for row in raw_grid:
    padded.append([INF] + row + [INF])
padded.append([INF] * (c + 2))

print(f"加框後網格尺寸：{len(padded)} x {len(padded[0])}")
print("第一列哨兵：", padded[0])
````

````python
# =====================================================================
# [3] Code 填空題：補全加框列之長度
# 任務：請將下方的 ___ 替換為 c + 2
# =====================================================================

c_val = 4
top_sentinel_len = len([float('inf')] * (___))
print("頂部哨兵列長度：", top_sentinel_len)  # 應為 6
````

````python
# =====================================================================
# [4] Code 練習題：以哨兵加框法實作 e287
# 任務說明：實作 solve_e287_sentinel(r, c, grid)
# =====================================================================

def solve_e287_sentinel(r, c, grid):
    INF = float('inf')
    pad = [[INF] * (c + 2)]
    for row in grid:
        pad.append([INF] + row + [INF])
    pad.append([INF] * (c + 2))
    
    # 找起點 (在 1..r 與 1..c 中尋找)
    min_v = INF
    cr, cc = 1, 1
    for i in range(1, r + 1):
        for j in range(1, c + 1):
            if pad[i][j] < min_v:
                min_v = pad[i][j]
                cr, cc = i, j
                
    vis = [[False] * (c + 2) for _ in range(r + 2)]
    vis[cr][cc] = True
    ans = pad[cr][cc]
    
    while True:
        best_r, best_c = -1, -1
        lowest = INF
        for dr, dc in ((-1,0), (1,0), (0,-1), (0,1)):
            nr, nc = cr + dr, cc + dc
            # 無需檢查邊界！因為哨兵值為 INF 且未拜訪，若皆大於等於 lowest 則不選
            if not vis[nr][nc] and pad[nr][nc] < lowest:
                lowest = pad[nr][nc]
                best_r, best_c = nr, nc
        if best_r == -1 or lowest == INF:
            break
        cr, cc = best_r, best_c
        vis[cr][cc] = True
        ans += pad[cr][cc]
    return ans

assert solve_e287_sentinel(3, 4, [[9, 1, 8, 14], [7, 2, 5, 13], [6, 3, 4, 12]]) == 62
print("✅ 練習題哨兵法驗證通過！")
````

### 📝 版本一：APCS 官方實作考場專用版 —— 淺顯易懂一般版（新手友善推薦）

* **適用情境**：APCS 正式考試現場、初學者課堂練習。
* **教學與設計理念**：
  1. **步驟平鋪直敘、拆解詳細**：讀入 $R, C$ 與網格後，以最直觀的雙層 `for` 迴圈找出全域最小值座標作為起點。
  2. **標準位移向量與安全檢查**：定義 `dr = [-1, 1, 0, 0]` 與 `dc = [0, 0, -1, 1]`，逐一檢查四方向鄰居，選出未拜訪之最小數值。
  3. **穩健 100% 滿分 AC**：使用 `while True` 驅動，無路可走時立即 `break`，累加路徑數值並輸出，考場高壓下零失誤奪得滿分！


````python
# ==============================================================================
# 📝 版本一：APCS 官方實作考場專用版 —— 淺顯易懂一般版（新手友善推薦）
# 適用情境：APCS 正式考場單筆測試案例，平鋪直敘、步驟明確、穩健滿分
# ==============================================================================

# 步驟 1：讀取矩陣列數 r 與行數 c
r, c = map(int, input().split())

# 步驟 2：讀取完整的二維網格地圖
grid = []
for _ in range(r):
    grid.append(list(map(int, input().split())))

# 步驟 3：全域搜尋起點（數值最小的格子）
min_val = float('inf')
start_r = 0
start_c = 0
for i in range(r):
    for j in range(c):
        if grid[i][j] < min_val:
            min_val = grid[i][j]
            start_r = i
            start_c = j

# 步驟 4：初始化走訪狀態與拜訪標記陣列
visited = [[False] * c for _ in range(r)]
visited[start_r][start_c] = True

cur_r = start_r
cur_c = start_c
total_sum = grid[start_r][start_c]

# 步驟 5：定義四個方向向量（上、下、左、右）
dr = [-1, 1, 0, 0]
dc = [0, 0, -1, 1]

# 步驟 6：模擬機器人貪婪移動
while True:
    next_r = -1
    next_c = -1
    smallest_neighbor = float('inf')
    
    # 檢查上下左右四個相鄰格子
    for k in range(4):
        nr = cur_r + dr[k]
        nc = cur_c + dc[k]
        
        # 邊界防禦：確保不越界且尚未走過
        if 0 <= nr < r and 0 <= nc < c and not visited[nr][nc]:
            if grid[nr][nc] < smallest_neighbor:
                smallest_neighbor = grid[nr][nc]
                next_r = nr
                next_c = nc
                
    # 若四個方向皆無路可走（無未走過的合法鄰居），停止移動
    if next_r == -1:
        break
        
    # 走到下一個最小鄰居
    cur_r = next_r
    cur_c = next_c
    visited[cur_r][cur_c] = True
    total_sum += grid[cur_r][cur_c]

# 步驟 7：輸出路徑上的數值總和
print(total_sum)
````

````python
# ==============================================================================
# ⚡ 版本二：APCS 官方實作考場專用版 —— 極簡高效精煉版（進階高手推薦）
# 適用情境：APCS 正式考試現場，極簡 16 行高效貪婪模擬
# ==============================================================================

r, c = map(int, input().split())
g = [list(map(int, input().split())) for _ in range(r)]
vis = [[False] * c for _ in range(r)]

# 一行找出全地圖最小值的起點座標 (cr, cc)
cr, cc = min(((i, j) for i in range(r) for j in range(c)), key=lambda p: g[p[0]][p[1]])
vis[cr][cc] = True
ans = g[cr][cc]

while True:
    # 找出四方向合法且未走過之鄰居清單
    nbrs = [(cr + dr, cc + dc) for dr, dc in ((-1, 0), (1, 0), (0, -1), (0, 1))
            if 0 <= cr + dr < r and 0 <= cc + dc < c and not vis[cr + dr][cc + dc]]
    if not nbrs:
        break
    cr, cc = min(nbrs, key=lambda p: g[p[0]][p[1]])
    vis[cr][cc] = True
    ans += g[cr][cc]

print(ans)
````

````python
# ==============================================================================
# 🌐 版本三：ZeroJudge 線上評判萬用 AC 版（支援多筆測資 EOF 循環）
# ==============================================================================
import sys

def solve_robot_path(r, c, grid):
    """
    機器人的路徑核心解題演算法
    :param r: 列數
    :param c: 行數
    :param grid: 二維整數網格
    :return: 走訪路徑數值總和
    """
    # 1. 尋找全圖最小值起點
    min_val = float('inf')
    start_r, start_c = 0, 0
    for i in range(r):
        for j in range(c):
            if grid[i][j] < min_val:
                min_val = grid[i][j]
                start_r, start_c = i, j

    # 2. 初始化走訪標記與狀態
    visited = [[False] * c for _ in range(r)]
    visited[start_r][start_c] = True
    cur_r, cur_c = start_r, start_c
    total_sum = grid[start_r][start_c]

    dr = [-1, 1, 0, 0]
    dc = [0, 0, -1, 1]

    # 3. 模擬貪婪移動
    while True:
        next_r, next_c = -1, -1
        smallest_neighbor = float('inf')
        for k in range(4):
            nr = cur_r + dr[k]
            nc = cur_c + dc[k]
            if 0 <= nr < r and 0 <= nc < c and not visited[nr][nc]:
                if grid[nr][nc] < smallest_neighbor:
                    smallest_neighbor = grid[nr][nc]
                    next_r, next_c = nr, nc
        if next_r == -1:
            break
        cur_r, cur_c = next_r, next_c
        visited[cur_r][cur_c] = True
        total_sum += grid[cur_r][cur_c]

    return total_sum

# ---------------------------------------------------------------------
# 【ZeroJudge 官方提交程式碼範本】
# 若要在 ZeroJudge 提交，請複製下方函式內容至解題系統：
# def main():
#     tokens = sys.stdin.read().split()
#     if not tokens: return
#     idx = 0
#     while idx < len(tokens):
#         r = int(tokens[idx])
#         c = int(tokens[idx+1])
#         idx += 2
#         grid = []
#         for _ in range(r):
#             grid.append([int(x) for x in tokens[idx:idx+c]])
#             idx += c
#         print(solve_robot_path(r, c, grid))
# ---------------------------------------------------------------------

# =====================================================================
# 🧪 Colab 本地自動化測試檢驗套件（點擊播放鍵自動執行）
# =====================================================================
test_cases = [
    {
        "name": "官方範例一 (1x7 單行地圖)",
        "r": 1, "c": 7,
        "grid": [[1, 2, 3, 4, 5, 6, 7]],
        "expected": 28
    },
    {
        "name": "官方範例二 (3x4 經典貪婪走訪)",
        "r": 3, "c": 4,
        "grid": [
            [9, 1, 8, 14],
            [7, 2, 5, 13],
            [6, 3, 4, 12]
        ],
        "expected": 62
    },
    {
        "name": "極端邊界測資 (1x1 單格起點即終點)",
        "r": 1, "c": 1,
        "grid": [[99]],
        "expected": 99
    },
    {
        "name": "2x2 螺旋路徑測資",
        "r": 2, "c": 2,
        "grid": [
            [1, 2],
            [4, 3]
        ],
        "expected": 10  # 1 -> 2 -> 3 -> 4 = 10
    }
]

print("=== e287. 機器人的路徑 全自動化測試報告 ===")
all_passed = True
for tc in test_cases:
    actual = solve_robot_path(tc["r"], tc["c"], tc["grid"])
    passed = (actual == tc["expected"])
    status = "✅ PASS" if passed else "❌ FAIL"
    if not passed:
        all_passed = False
    print(f"{status} | {tc['name']} ➔ 預期: {tc['expected']}, 實際: {actual}")

if all_passed:
    print("\n🎉 恭喜！所有官方與極限範例測試全數通過！可安心提交至 APCS 考場與 ZeroJudge！");
else:
    print("\n⚠️ 有測試資料未通過，請檢查四方向判斷或起點極值搜尋邏輯！");
````
