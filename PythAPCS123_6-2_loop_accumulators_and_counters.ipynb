### 🐷 6-2-1 累加器（Accumulator）模式：存錢筒模型與初始值歸零（total = 0）

**生活故事：陶瓷小豬撲滿存錢筒**  
你買了一隻全新可愛的陶瓷小豬撲滿。在開始存錢的第一天，你一定會先搖一搖撲滿，確認裡面是空空如也的（餘額為 0 元）。如果原本裡面就殘留了別人的 50 元假鈔，你存了一整年最後敲開撲滿算出的總金額就徹底不準了！在接下來的每一天，你回家後就把口袋裡的零錢投進撲滿（`total += coin`）。到了月底，你只要看一眼撲滿，裡面裝的就是這一個月所有硬幣的累計總額！

**運作機制與累加器三部曲：**  
在 Python 中，計算數值總和的標準架構稱為**「累加器模式（Accumulator Pattern）」**：  
1. **事前歸零**：在進入迴圈前，必須宣告一個累加變數並設初始值為 0（`total = 0`）。
2. **逐輪累投**：在迴圈內部，每一輪將當前數值加到累加器中（`total += val`，即 `total = total + val`）。
3. **事後結算**：迴圈結束後，在**沒有縮排**的最外層輸出或使用最終的 `total`。

**常見錯誤陷阱：**  
- **世紀大災難一：把 `total = 0` 寫在迴圈「內部」**！如果寫在迴圈內，每一輪剛加完數值，下一輪開始又被強制歸零，最後印出來的 `total` 永遠只會是最後一筆的數字！
- **世紀大災難二：忘記初始化**！若沒有寫 `total = 0` 就直接在迴圈中執行 `total += val`，直譯器會立刻崩潰並噴出 `NameError: name 'total' is not defined`。

**APCS 考試實務建議：**  
計算 1 到 N 的連續整數總和（高斯求和公式 $\frac{N(N+1)}{2}$ 的程式驗證）、數列總得分等，累加器是出現頻率高達 90% 的絕對基礎功。


````python
# [2] Code 範例區：累加器計算 1 加到 5 的總和

# 第一步：在迴圈外宣告累加器並歸零
total = 0

# 第二步：迴圈逐輪累加
print("開始存錢累加：")
for coin in range(1, 6):
    total += coin
    print("投入", coin, "元，當前撲滿總額：", total)

# 第三步：迴圈外輸出最終總結算
print("-" * 30)
print("1 到 5 的累積總和為：", total)  # 15
````

### 🔘 6-2-2 計數器（Counter）模式：特定事件發生的「打勾計次機（count += 1）」

**生活故事：展覽館門口的金屬計次器**  
保全大叔站在展覽館入口，手裡拿著一顆圓形金屬計次器（Clicker）。每當有一位參觀者通過安檢門走進展覽館，大叔的大拇指就用力按一下按鈕，「喀噠！」一聲，計數器上的數字就增加 1。大叔完全不在乎走進來的人體重是 40 公斤還是 80 公斤，他唯一關心的只有一件事——「進來的人數到底增加了幾位」！

**運作機制與累加器的本質差異：**  
初學者最容易混淆「累加器」與「計數器」：  
- **累加器（Accumulator）**：每次加上具體的數值 `total += val`（數值存錢筒）。
- **計數器（Counter）**：每次只增加固定次數 `count += 1`（打勾計次機）。

計數器標準流程：  
1. 在迴圈前宣告 `count = 0`。
2. 在迴圈中走訪每一項，當滿足特定事件或條件時，觸發 `count += 1`。
3. 迴圈結束後，`count` 的值即為事件發生的總次數！

**常見錯誤陷阱：**  
- **手滑寫成覆蓋賦值 `count = 1`**：少寫了一個加號 `+`，導致不管條件觸發了幾百次，`count` 每次都被重置為 1，最後印出來永遠是 1！請務必寫 `count += 1`。
- **將計數與累加搞混**：例如題目要統計「有幾個及格成績」，卻寫成了 `count += score`（這算成了及格總分，而非及格人數！）。

**APCS 考試實務建議：**  
統計因數個數（質數判定的重要基礎）、統計滿分人數、統計特定字元出現次數等，計數器模式是不可或缺的解題武器。


````python
# [2] Code 範例區：計數器統計 1 到 20 之中 3 的倍數個數

count = 0  # 初始化計數器為 0

print("走訪 1 到 20 並統計 3 的倍數：")
for num in range(1, 21):
    if num % 3 == 0:
        count += 1  # 每次發現 3 的倍數，計數器就按一下 (+1)
        print("發現 3 的倍數：", num, "(目前累計", count, "個)")

print("-" * 30)
print("1 到 20 之中，3 的倍數總共有：", count, "個")  # 6 個 (3, 6, 9, 12, 15, 18)
````

````python
# ==========================================
# [4] Code 練習題
# 任務說明：
# 因數計數器（Factor Counter）：
# 給定一個正整數 target。
# 請使用 for 迴圈檢查 1 到 target 之間的所有整數 i，
# 若 target 能被 i 整除（target % i == 0），計數器 count 加 1。
# 迴圈結束後輸出："Number of factors of target: count"
#
# 【公開測試資料 1】
# 設定：target = 6  (因數有 1, 2, 3, 6 共 4 個)
# 預期輸出：
# Number of factors of 6: 4
#
# 【公開測試資料 2】
# 設定：target = 7  (質數，因數只有 1, 7 共 2 個)
# 預期輸出：
# Number of factors of 7: 2
# ==========================================
# 請在下方撰寫你的程式碼並執行測試：

target = 6
count = 0
for i in range(1, target + 1):
    if target % i == 0:
        count += 1
print("Number of factors of ", target, ": ", count, sep="")
````

### ❄️ 6-2-3 累乘器（Product）模式：階乘 N! 與初始值設 1 的關鍵守則

**生活故事：滾雪球與細菌指數分裂**  
有一種傳奇的魔法細菌，每過一個小時就會自我分裂成原來的 2 倍。如果你在培養皿裡一開始放入「0 隻細菌」，無論分裂多少個小時，0 乘以 2 永遠還是 0！只有當你一開始放入「1 隻細菌」作為初始種子，經過 3 個小時後才會變成 $1 \times 2 \times 2 \times 2 = 8$ 隻！乘法的世界與加法完全不同，加法的起點是 0，但**乘法的起點絕對是 1**！

**運作機制與階乘連乘：**  
數學上的**階乘（Factorial）**定義為從 1 連乘到 $N$：  
$$N! = 1 \times 2 \times 3 \times \dots \times N$$
在 Python 中實作累乘器模式時，必須牢記：  
1. **初始值設為 1**：`product = 1`（因為 1 是乘法的單位元素，$x \times 1 = x$）。
2. **複合乘法賦值**：每一輪使用 `product *= i`（相當於 `product = product * i`）。
3. **走訪起點必須從 1 開始**：`range(1, N + 1)`。

**常見錯誤陷阱：**  
- **習慣性手滑宣告 `product = 0`**：這是一場悲劇！因為任何數乘以 0 都會變成 0，只要初始值設了 0，整個迴圈不管跑幾千次，最後算出的乘積永遠都是 0！
- **range 忘記設起點寫成 `range(N + 1)`**：預設起點是 0，第一輪執行 `product *= 0`，原本辛苦維持的乘積一瞬間化為烏有！

**APCS 考試實務建議：**  
排列組合（$P_m^n$、$C_m^n$）、二進位權重推導、連乘積運算時，累乘器都是必備基本功。在 Python 中整數具有任意精度（不會像 C/C++ 一樣超過 20! 就整數溢位），這讓 Python 在數論連乘上具有得天獨厚的優勢！


````python
# [2] Code 範例區：累乘器計算 5! (5 的階乘 = 1 * 2 * 3 * 4 * 5)

n = 5
factorial = 1  # 乘法初始種子必須為 1！

print("計算", n, "階乘的逐步過程：")
for i in range(1, n + 1):
    factorial *= i
    print("乘上", i, "，當前乘積為：", factorial)

print("-" * 30)
print(n, "! 的最終結果為：", factorial)  # 120
````

### 🍎 6-2-4 條件複合累加：奇數和、偶數和與 3/5 倍數篩選加總

**生活故事：智慧果園的水果分揀輸送帶**  
果農伯伯採收了一大批蘋果，透過自動化輸送帶送入包裝廠。輸送帶上有一顆紅外線感應器，當一顆蘋果滾過去時，感應器會量測重量：如果這顆蘋果重量超過 250 公克（符合「特級果」條件），機械手臂就會把它推進特級箱，並且將重量累計到「特級箱總重」中；如果是一般蘋果，就直接放行，不計入特級統計！這就是程式中的「條件複合累加」。

**運作機制與結構嵌套：**  
條件複合累加是將第五章學習的 `if` 條件分支，嵌套在第六章的 `for` 迴圈之中：  
```python
total = 0
for x in range(1, 101):
    if x % 2 == 1:       # 守門員：只有奇數才能通過
        total += x       # 只有通過審查的數值才會被加總
```  
迴圈每一輪負責提供一個候選數值，`if` 擔任嚴格的守門員，只有當條件為真時，累加器或計數器才會被觸發！

**常見錯誤陷阱：**  
- **縮排層次混淆（最常見的格式 WA）**：  
  - `for` 是最外層（縮排 0 空格）。
  - `if` 是迴圈內部（縮排 4 空格）。
  - 累加動作 `total += x` 必須縮在 `if` 裡面（縮排 8 空格）。
  - **最後的 `print(total)` 必須退回最外層（縮排 0 空格）**！如果把 `print` 誤放在迴圈裡面，每一輪都會印出一次中途結果，在線上評判（OJ）中會直接吃到 Wrong Answer！

**APCS 考試實務建議：**  
在 APCS 歷屆題目中，「計算區間內符合某些複合條件（如既是 3 的倍數但不是 5 的倍數）的數值總和」是極為高頻出現的典型題型。


````python
# [2] Code 範例區：計算 1 到 10 之間的奇數和與偶數和

odd_sum = 0   # 奇數存錢筒
even_sum = 0  # 偶數存錢筒

for num in range(1, 11):
    if num % 2 == 1:
        odd_sum += num   # 奇數加總
    else:
        even_sum += num  # 偶數加總

# 迴圈結束後，統一在外層輸出
print("1 到 10 奇數和 (1+3+5+7+9) =", odd_sum)    # 25
print("1 到 10 偶數和 (2+4+6+8+10) =", even_sum)  # 30
````

### 📊 6-2-5 迴圈流式平均值計算：累加與計數協同作戰及防除零保護

**生活故事：段考班級及格學生平均分計算**  
學校段考結束後，數學老師想要計算「及格同學的平均分數」。老師需要兩名小幫手：第一位小幫手拿著計算機，每看到一張及格考卷就把分數加進去（負責維護**總分累加器**）；第二位小幫手拿著點名本，每看到一張及格考卷就畫上一筆「正」字（負責維護**及格人數計數器**）。所有考卷批改完成後，老師用總分除以總人數，算出漂亮又客觀的平均分！

**運作機制與協同作戰：**  
平均值的數學公式是：  
$$\text{平均值（Average）} = \frac{\text{總和（Sum）}}{\text{項數（Count）}}$$
因此在程式中，我們在迴圈外部同時建立兩個夥伴：  
1. `total = 0`（總和累加器）
2. `count = 0`（項數計數器）
每當符合條件的數據進來時，兩者同步更新：`total += score` 且 `count += 1`。  
迴圈結束後，計算 `average = total / count`。

**常見錯誤陷阱（致命的除以零 RE）：**  
- **`ZeroDivisionError: division by zero`**！想像一下，如果這場考試題目超難，全班竟然「沒有任何人及格」（`count == 0`）！這時如果你的程式不管三七二十一直接執行 `total / count`，電腦會瞬間崩潰暴斃，線上評判立刻判定為 **Runtime Error (RE)**！  
- **必備防禦 SOP**：在相除前，必須先以 `if count > 0:` 進行保護檢查；若 `count == 0`，則輸出特定提示（例如 `0.0` 或 `None`）。

**APCS 考試實務建議：**  
在競賽評測中，評判系統非常喜歡出「完全沒有符合條件之極端邊界測資（Corner Cases）」。只要養成計算平均值前先防除零的反射神經，就能保證程式穩若磐石。


````python
# [2] Code 範例區：計算 1 到 10 中偶數的平均值（附防除零保護）

total = 0
count = 0

for num in range(1, 11):
    if num % 2 == 0:
        total += num
        count += 1

# 迴圈結束後，嚴格執行防除零衛語保護
if count > 0:
    average = total / count
    print("偶數總和：", total)
    print("偶數個數：", count)
    print("偶數平均值：", average)  # 30 / 5 = 6.0
else:
    print("沒有找到任何符合條件的偶數！")
````

### 🥊 6-2-6 擂台盟主法（極值維護）：負無窮大初始化與單趟走訪求最大/最小值

**生活故事：天下第一武道大會的衛冕盟主**  
天下第一武道大會正式開打。比賽剛開始時，第一位登場的選手直接走上擂台，戴上象徵榮譽的「盟主王冠（Max）」。接下來，第二位挑戰者上台挑戰：如果挑戰者的戰鬥力比擂台上的現任盟主更高，現任盟主就得乖乖退位，由挑戰者搶下王冠成為新任衛冕盟主；如果挑戰者打輸了，現任盟主繼續霸氣衛冕！等到所有參賽選手輪番挑戰完畢後，留在擂台上的那個人，必定就是全場戰鬥力最高的大冠軍！

**運作機制與極值維護：**  
想要在一串連續輸入或數列中找出「最大值（Maximum）」或「最小值（Minimum）」，最經典的高效技巧就是**「擂台盟主法（Max / Min Tracking）」**：  
1. **建立擂台盟主變數**：`max_val` 或 `min_val`。
2. **單趟走訪比武**：每一輪來了一個新數值 `x`：  
   - 若想求最大值：若 `x > max_val`，則盟主換人做 `max_val = x`（或寫 `max_val = max(max_val, x)`）。
   - 若想求最小值：若 `x < min_val`，則更換 `min_val = x`（或寫 `min_val = min(min_val, x)`）。

**常見錯誤陷阱（世紀大盲點）：**  
- **求最大值時隨手把 `max_val` 初始化設為 0！**  
  初學者最容易寫出 `max_val = 0`。想像一下，如果今天輸入的數值「全都是負數」（例如北極的氣溫：`-10, -5, -20`），程式執行完畢後，印出的最大值竟然是 `0`！可是這串數據裡根本連一個 0 都沒有，直接造成全盤皆輸的 WA！  
- **專業正解**：  
  - 做法 A：將「第一筆實際輸入的資料」直接設為初始盟主！
  - 做法 B：將最大值初始化為負無窮大 `max_val = float('-inf')`，最小值初始化為正無窮大 `min_val = float('inf')`。

**APCS 考試實務建議：**  
擂台盟主法只需要 $O(N)$ 時間與 $O(1)$ 空間，不需要開任何串列陣列暫存，是流式計算中最高效的經典演算法骨架。


````python
# [2] Code 範例區：擂台盟主法尋找最大值與最小值

# 使用負無窮大與正無窮大作為安全起點
max_score = float('-inf')  # 初始設為負無窮大，任何數都能贏它
min_score = float('inf')   # 初始設為正無窮大，任何數都能小於它

# 模擬走訪一組數值（含負數）
print("擂台比武開始：")
for val in (-15, -3, -42, -7, -2):
    if val > max_score:
        max_score = val  # 挑戰成功，換新盟主
    if val < min_score:
        min_score = val  # 發現更小值
    print("走訪數值:", val, "-> 目前最高:", max_score, "目前最低:", min_score)

print("-" * 30)
print("全場最高盟主：", max_score)  # -2
print("全場最低守門：", min_score)  # -42
````

### 📥 6-2-7 連續 N 筆資料動態輸入累加：結合 input() 的流式處理模式

**生活故事：超商收銀員刷條碼結帳**  
當你抱著 3 瓶飲料到便利商店櫃檯結帳時，店員首先在收銀螢幕上確認了商品數量，接著拿起手上的條碼掃描槍，「嗶！35元」、「嗶！20元」、「嗶！25元」，每刷過一瓶飲料，收銀螢幕上的應付總金額就即時跳動累加！店員完全不需要拿一張紙把這 3 瓶飲料的條碼編號全部抄下來才開始算，而是「來一件、刷一件、加一件」！這就是電腦科學中極其強悍的**「流式資料處理（Streaming Processing）」**！

**運作機制與標準考場骨架：**  
在競技程式 APCS 中，極高比例的測資格式為：  
```
第一行：給定一個整數 N（代表接下來有 N 筆資料）
接下來的 N 行：每行包含一個整數
```  
對應的標準 Python 處理骨架為：  
```python
n = int(input())     # 1. 先讀取總筆數
total = 0            # 2. 宣告累加器歸零
for _ in range(n):   # 3. 精確重複執行 n 次
    val = int(input())  # 4. 每一輪讀取一筆
    total += val        # 5. 立刻即時累加
print(total)         # 6. 迴圈結束後輸出結果
```  
這種寫法的記憶體空間複雜度只有驚人的 $O(1)$！即使 $N = 1000000$（一百萬筆資料），記憶體中也始終只保存當前那一個數字，永遠不會引發記憶體超限（MLE）！

**常見錯誤陷阱：**  
- **漏讀第一行的 N**：直接把第一行的 N 當成第一筆資料加進去了，導致答案全錯！
- **在迴圈內寫了兩次 `input()`**：一次拿去判斷、一次拿去累加，導致資料每隔一行被跳過一次，最後提早遇到檔案結尾崩潰（EOFError）！

**APCS 考試實務建議：**  
這是 APCS 實作題最經典的起手式之一。請反覆練習這套「先讀 N、for 跑 N 次、每次讀一筆即時處理」的肌肉記憶！

- 全天的平均氣溫是多少？
- 氣溫超過 30 度的酷熱時數有幾個小時？
這份報表不是分開跑五次程式算出來的，而是在**「同一趟迴圈走訪中，同時並行維護多個指標變數」**！這就是頂級演算法工程師的多指標協同匯總能力。

**運作機制與全距（Range）計算：**  
統計學中的**全距（Range）**定義為：  
$$\text{全距} = \text{最大值（Max）} - \text{最小值（Min）}$$
全距代表一組數據的分散程度（波動幅度）。  
在單一迴圈中，我們可以同時維護：  
1. `total`（加總器）
2. `count`（計數器）
3. `max_val`（最大盟主）
4. `min_val`（最小守門員）
5. `special_count`（條件計數器）  
一趟走訪全部搞定，時空效率達到了最極致的境地！

**常見錯誤陷阱：**  
- **變數過多導致手滑抄錯**：例如在更新 `min_val` 時，手滑寫成了 `max_val = min(max_val, x)`，導致最大值被徹底污染破壞。
- **忘記全距是大減小**：誤寫成 `min_val - max_val` 導致全距出現負數。

**APCS 考試實務建議：**  
在 APCS 實作題第二題至第三題中，題目常要求選手同時回報多個統計特徵（如最高值、總和與最大差值）。掌握單趟多指標並行維護，能讓你的程式碼結構清晰俐落，毫無贅字！


````python
# [2] Code 範例區：單趟走訪同時維護多項統計指標

# 模擬 5 筆氣溫數據：(28, 33, 26, 35, 28)
total_temp = 0
count = 0
max_temp = float('-inf')
min_temp = float('inf')
hot_hours = 0  # 超過 30 度的酷熱時段計數器

for temp in (28, 33, 26, 35, 28):
    total_temp += temp
    count += 1
    if temp > max_temp:
        max_temp = temp
    if temp < min_temp:
        min_temp = temp
    if temp >= 30:
        hot_hours += 1

# 結算全距與平均
temp_range = max_temp - min_temp
avg_temp = total_temp / count

print("=== 一日氣候綜合觀測簡報 ===")
print("最高溫度：", max_temp, "°C")
print("最低溫度：", min_temp, "°C")
print("溫差全距：", temp_range, "°C")
print("平均溫度：", avg_temp, "°C")
print("酷熱時數 (>=30°C)：", hot_hours, "小時")
````

````python
# ==========================================
# [3] Code 填空題
# 任務說明：
# 請走訪數列 (12, 45, 19, 88, 30)，
# 同步找出最大值 max_v 與最小值 min_v，並計算全距 diff = max_v - min_v。
#
# 請在空格處填入最小值更新判斷 if x < min_v，以及全距計算公式 max_v - min_v！
# ==========================================

max_v = float('-inf')
min_v = float('inf')

for x in (12, 45, 19, 88, 30):
    if x > max_v:
        max_v = x
    # 第一空：最小值檢查
    if x ___ min_v:
        min_v = x

# 第二空：全距為最大值減去最小值
diff = max_v ___ min_v
print("最大值:", max_v, ", 最小值:", min_v, ", 全距:", diff)
````

````python
# ==========================================
# [4] Code 練習題
# 任務說明：
# 數列全距與極值報表：
# 輸入第一行整數 n 代表資料個數（n >= 2）。
# 接下來 n 行每行輸入一個整數。
# 請輸出兩行：
# 第一行輸出："Max: max_val, Min: min_val"
# 第二行輸出："Range: range_val"
#
# 【公開測試資料 1】
# 輸入：
# 3
# 10
# 25
# 15
# 預期輸出：
# Max: 25, Min: 10
# Range: 15
#
# 【公開測試資料 2】
# 輸入：
# 2
# -5
# -12
# 預期輸出：
# Max: -5, Min: -12
# Range: 7
# ==========================================
# 請在下方撰寫你的程式碼並執行測試：

n = int(input())
max_val = float('-inf')
min_val = float('inf')
for _ in range(n):
    num = int(input())
    if num > max_val:
        max_val = num
    if num < min_val:
        min_val = num

print("Max: ", max_val, ", Min: ", min_val, sep="")
print("Range: ", max_val - min_val, sep="")
````
