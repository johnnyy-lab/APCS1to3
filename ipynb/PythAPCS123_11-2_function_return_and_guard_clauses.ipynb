### 11.2.1 `return` 關鍵字：將函數計算結果產出並交接給外部變數

如果說參數是函數接收原料的「輸入口」，那麼 `return` 關鍵字就是函數將加工完成的產品送出的**「唯一合法出口」**！在 Python 中，`return` 承擔了兩大不可撼動的職責：
1. **立即終止函數的執行**：一旦直譯器執行到 `return` 這一行，函數內部後續的所有程式碼將會被**立刻中斷且完全忽略**，控制權會瞬間脫離函數本體，跳回到呼叫處。
2. **交接成果給運算式**：`return` 後方緊跟的運算式或數值，會成為整個函數呼叫運算式的「最終求值結果」。

外部主程式可以直接使用單向賦值等號，將函數回傳的值接收並儲存到變數中，例如 `total = calc_sum(10, 20)`。更強大的是，由於「函數呼叫本身就是一個運算式」，你可以把帶有回傳值的函數呼叫直接嵌入到更大的算式中，例如 `avg = (calc_sum(10, 20) + calc_sum(30, 40)) / 4`，甚至作為另一個函數的引數！這種像樂高積木一樣自由組合的特性，正是現代軟體架構與演算法設計中「模組可組合性（Composability）」的靈魂所在。


````python
# 範例 11.2.1：利用 return 產出計算結果並交接給外部變數
def calc_rect_area(width, height):
    # 計算面積並透過 return 將數值送出
    area = width * height
    return area
    # 注意：return 之後的程式碼永遠不會被執行（稱為 Dead Code）
    print("這行永遠不會被印出！")

# 1. 呼叫函數並用變數接收回傳結果
room_a = calc_rect_area(5, 8)
room_b = calc_rect_area(4, 6)

print(f"房間 A 面積：{room_a} 平方公尺")
print(f"房間 B 面積：{room_b} 平方公尺")

# 2. 將函數呼叫直接代入數學運算式中（可組合性）
total_area = room_a + room_b
print(f"兩房間總面積：{total_area} 平方公尺")

# 3. 鏈式嵌入運算：直接將呼叫當成數字參與計算
diff = abs(calc_rect_area(5, 8) - calc_rect_area(4, 6))
print(f"兩房間面積差距：{diff} 平方公尺")
````

### 11.2.2 世紀迷思剖析：`return`（資料輸出）vs `print()`（終端列印）

在程式教學現場中，初學者最常犯、且後果最慘烈的概念混淆，就是分不清 **`return`** 與 **`print()`** 的本質區別！在許多剛接觸函數的學生眼中，兩者似乎都是「把結果拿出來」，但它們在電腦底層的運作目標截然不同：

1. **`print()` 是給「人類眼睛看」的終端顯示器**：
   - 它將資料轉成文字字符，輸出到作業系統的標準輸出（stdout，即終端黑色螢幕）。
   - **程式碼本身「拿不到」`print()` 印出來的東西！** `print()` 執行完畢後的自身回傳值永遠是 `None`。
2. **`return` 是給「程式碼邏輯用」的數據傳遞管道**：
   - 它將變數記憶體內的真實資料物件（如整數、浮點數、串列等）交給外部的主程式。
   - **`return`「不會在螢幕上印出任何東西」！** 人類肉眼完全看不見回傳過程，但後續的程式碼可以立即對該數值進行加法、乘法或條件比對。

如果在需要回傳資料的函數中誤寫成了 `print()`，當你在外部嘗試以 `res = my_func()` 接收時，`res` 只會得到無辜的 `None`。若你接著寫 `res + 10`，直譯器就會無情地炸出 `TypeError: unsupported operand type(s) for +: 'NoneType' and 'int'`！因此請務必牢記鐵則：**「要算數交棒用 return，要螢幕顯示才用 print」**。


````python
# 範例 11.2.2：殘酷對決——誤用 print 導致的 TypeError 大災難
# 錯誤示範：以為 print 就是輸出結果
def bad_add(a, b):
    print(a + b) # 只是印在螢幕上，但沒有交棒！

# 正確示範：使用 return 交棒數據
def good_add(a, b):
    return a + b # 真正將數值交接回主程式

print("=== 1. 測試錯誤寫法 bad_add ===")
ans_bad = bad_add(10, 20)
print("螢幕上確實印出了 30，但是 ans_bad 變數的實際內容卻是：", ans_bad)
print("ans_bad 的型態是：", type(ans_bad))

# 嘗試繼續拿 ans_bad 乘上 2，會發生什麼事？
try:
    impossible_calc = ans_bad * 2
except TypeError as err:
    print("【報錯攔截】致命錯誤！", err)

print("\n=== 2. 測試正確寫法 good_add ===")
ans_good = good_add(10, 20)
print("螢幕上不會自動印出任何東西，但 ans_good 真正拿到了數字：", ans_good)
# 可以自由參與後續所有運算
double_result = ans_good * 2
print("順利進行後續倍增計算：", double_result)
````

### 11.2.3 衛語句（Guard Clauses）：利用提前 `return` 進行非法條件防禦攔截

在編寫複雜的業務邏輯或演算法時，我們常常需要先檢查輸入的合法性。初學者往往習慣將合法邏輯包裹在一層又一層的 `if ... else` 結構中：
```python
# 巢狀向右傾斜的寫法（反模式）
def process(x):
    if x > 0:
        if x < 100:
            if x % 2 == 0:
                # 真正的核心邏輯被深深縮排在第四層！
                return x * 2
    return -1
```
這種代碼在軟體工程中被稱為**「箭頭型反模式（Arrow Anti-pattern）」**：隨著條件增多，縮排不斷向右擠壓，使得主要邏輯深埋在層層括號或縮排底層，極難閱讀與維護。

專業競賽選手與資深工程師極力推薦的解法稱為**「衛語句（Guard Clauses，守衛語句）」**。其核心心法是**「逆向防守，儘早退出（Fail Fast）」**：
在函數的最開頭幾行，優先檢查所有「非法、異常或邊界情況」，一旦發現條件不符，立刻使用 `return` 提早終止並退場！
```python
# 優雅清爽的衛語句寫法
def process(x):
    if x <= 0 or x >= 100 or x % 2 != 0:
        return -1  # 守衛把關：不合格直接踢出
    
    # 通過守衛後，核心邏輯保持在最乾淨的外層！
    return x * 2
```
善用衛語句能徹底消除多餘的 `else` 與層層縮排，使程式碼扁平直觀，不僅除錯更容易，更是 APCS 考場上防止 `ZeroDivisionError` 或邊界崩潰的絕佳武器。


````python
# 範例 11.2.3：使用衛語句實現安全除法器與門檻檢驗
def safe_divide(numerator, denominator):
    # 衛語句 1：除數為 0 是數學禁忌，立刻攔截退出
    if denominator == 0:
        print("[GUARD] 警告：除數為 0，終止計算！")
        return None
    
    # 衛語句 2：分子若為負數，視為非法測資提早退出
    if numerator < 0:
        print("[GUARD] 警告：分子不可為負數！")
        return None
    
    # 順利通過上述所有警衛把關，核心除法邏輯在最外層優雅執行
    return numerator / denominator

print("合法計算：100 / 4 =", safe_divide(100, 4))
print("違規計算 1：100 / 0 =", safe_divide(100, 0))
print("違規計算 2：-50 / 5 =", safe_divide(-50, 5))
````

````python
# 練習題 11.2.3：網格座標合法性防守檢查
# 題目說明：
# 在高度為 H、寬度為 W 的網格地圖中，合法列座標範圍為 0 <= r < H，行座標為 0 <= c < W。
# 請定義函數 get_grid_value_safe(grid, r, c)：
# 1. 取得地圖高度 H = len(grid) 與寬度 W = len(grid[0])。
# 2. 使用衛語句：若 r 或 c 越界（r < 0 或 r >= H 或 c < 0 或 c >= W），立刻提早回傳 -1。
# 3. 若未越界，回傳該格數值 grid[r][c]。
# 完成後請通過公開測資驗證。

# 請在下方撰寫你的程式碼：
def get_grid_value_safe(grid, r, c):
    H = len(grid)
    W = len(grid[0])
    # 衛語句提早阻斷越界
    if r < 0 or r >= H or c < 0 or c >= W:
        return -1
    return grid[r][c]

# 公開測資驗證：
test_map = [
    [10, 20, 30],
    [40, 50, 60]
]
assert get_grid_value_safe(test_map, 1, 2) == 60, "測資 1 錯誤"
assert get_grid_value_safe(test_map, 2, 0) == -1, "測資 2 越界檢查錯誤"
assert get_grid_value_safe(test_map, -1, 1) == -1, "測資 3 負數越界檢查錯誤"

print("練習題 11.2.3 驗證通過！")
````

````python
# 挑戰題 11.2.3：會員折價券多重守衛檢驗
# 題目說明：
# 定義 calc_coupon_discount(total_amount, coupon_code):
# 衛語句 1：若 total_amount <= 0，回傳 0。
# 衛語句 2：若 coupon_code 不是 "SAVE50" 且不是 "SAVE100"，回傳 0。
# 衛語句 3：若 coupon_code 是 "SAVE100" 但消費總額未達 1000 元，回傳 0（未達門檻）。
# 若通過所有檢驗：
#   coupon_code 是 "SAVE50" 回傳 50
#   coupon_code 是 "SAVE100" 回傳 100

def calc_coupon_discount(total_amount, coupon_code):
    if total_amount <= 0:
        return 0
    if coupon_code != "SAVE50" and coupon_code != "SAVE100":
        return 0
    if coupon_code == "SAVE100" and total_amount < 1000:
        return 0
    
    if coupon_code == "SAVE50":
        return 50
    return 100

assert calc_coupon_discount(500, "SAVE50") == 50
assert calc_coupon_discount(800, "SAVE100") == 0
assert calc_coupon_discount(1200, "SAVE100") == 100
assert calc_coupon_discount(-10, "SAVE50") == 0
print("挑戰題 11.2.3 挑戰成功！")
````

### 11.2.4 多條件分支回傳：`if-elif-else` 中的 return 覆蓋完整性

在處理分級、分類或多種情境的決策問題時，我們常會在函數內部搭配 `if-elif-else` 分支結構，並在各個分支中分別執行 `return`。

這時必須特別注意一個致命的陷阱——**「分支回傳遺漏（Missing Return）」**！
請思考以下情況：如果你寫了：
```python
def classify(x):
    if x > 0:
        return "Positive"
    elif x < 0:
        return "Negative"
    # 如果 x 剛好等於 0 呢？
```
當傳入 `x = 0` 時，程式既不滿足 `if` 也不滿足 `elif`，直譯器會直接滑過整個分支結構到達函數尾端。如同 11.1 所述，當函數執行到底部卻沒有碰上任何 `return` 時，它會**靜悄悄地回傳 `None`！**
呼叫端原本期待收到代表狀態的字串，結果卻意外接到了 `None`，若後續拿它進行字串拼接（例如 `classify(x) + " Number"`），整個系統就會立刻崩潰。

因此，在設計多條件回傳時，務必遵守以下兩大設計準則之一：
1. **備齊 `else` 收底**：在所有 `elif` 的最後，務必加上一個負責收拾所有未盡可能性的 `else: return 預設值`。
2. **底層兜底回傳（Default Fallback）**：將最普遍、最常規或無條件的預設結果，寫在整個分支結構下方的最外層直接 `return 預設值`。這能確保無論資料多麼極端，函數的出口都永遠 100% 被縝密覆蓋！


````python
# 範例 11.2.4：多分支回傳完整性與 None 漏接防禦
# 瑕疵寫法：遺漏了 60 分以下的情況，會意外回傳 None
def flawed_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 60:
        return "C"

print("85 分評等：", flawed_grade(85))
print("50 分評等：", flawed_grade(50)) # 印出 None！

# 完美防禦寫法：使用 else 保證 100% 覆蓋
def secure_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 60:
        return "C"
    else:
        return "F" # 兜底保證，絕不漏接！

print("\n修正後 50 分評等：", secure_grade(50)) # 成功回傳 'F'
````

### 11.2.5 函數多值回傳技巧：逗號自動打包為 Tuple 與呼叫端多變數解包

在許多傳統程式語言（如 C/C++ 或 Java）中，一個函數在語法上被嚴格限制只能回傳「單一一個數值」。如果想同時拿到最小值、最大值與平均值，往往必須大費周章地建立專門的結構體或物件。

然而，Python 擁有全世界公認最優雅的黑科技——**「多值回傳」**！
在 Python 中，只要在 `return` 後方使用**逗號 `,`** 串接多個數值，直譯器就會自動幫我們完成以下兩大魔法：
1. **自動打包（Automatic Packing）**：當你寫下 `return a, b, c` 時，Python 會在背後自動將這三個數值封裝為一個**不可變元組（Tuple）**，即 `(a, b, c)`。
2. **呼叫端多元解包（Multiple Unpacking）**：在呼叫端，我們可以直接用相同數量的變數並排在等號左側，一氣呵成地同時接收：
   ```python
   low, high, avg = get_statistics([10, 20, 30])
   ```
   等號右側回傳的元組，會依序將各個隔間的值精準解開並賦予左側的對應變數！

這項特性在演算法解題中極為常見。例如二維平面中的座標轉換、數論運算中同時求商數與餘數（如內建的 `divmod` 原型）、或是模擬賽事中同時產出主隊得分與客隊得分。只要記得：**等號左側接收的變數個數，必須與函數回傳的數值個數完全吻合**，否則會觸發 `ValueError: too many/not enough values to unpack` 的天平不對稱報錯！


````python
# 範例 11.2.5：多值回傳的元組打包與解包接收
def get_array_stats(numbers):
    # 同時計算最小值、最大值與總和
    minimum = min(numbers)
    maximum = max(numbers)
    total_sum = sum(numbers)
    # 用逗號分隔，自動打包為 Tuple 回傳！
    return minimum, maximum, total_sum

scores = [85, 92, 78, 96, 88]

# 1. 呼叫端解包：用三個獨立變數同時接收
low, high, total = get_array_stats(scores)
print(f"最低分：{low}，最高分：{high}，總得分：{total}")

# 2. 探查底層：如果只用一個變數接收，它究竟是什麼？
single_holder = get_array_stats(scores)
print("\n只用一個變數接收時的內容：", single_holder)
print("型態證實：", type(single_holder)) # 證明本質確實是 Tuple！
````

### 11.2.6 函數回傳布林值：構建高可讀性的判定函數（Predicate Functions）

在演算法競賽與實務開發中，有一類極為高頻出現的特殊函數：它們的任務是「檢驗某個條件是否成立」，並且永遠只回傳**布林值（`True` 或 `False`）**。在資訊科學中，這類函數被正式稱為**「謂詞函數 / 判定函數（Predicate Functions）」**。

在 Python 命名慣例中，判定函數的名稱通常以 `is_`、`has_` 或 `can_` 作為前綴，例如：
- `is_prime(n)`：判定數字 $n$ 是否為質數。
- `is_leap_year(year)`：判定年份是否為閏年。
- `is_valid_triangle(a, b, c)`：判定三邊長是否能構成合法三角形。

撰寫判定函數時，初學者常常會寫出冗長繁複的結構：
```python
# 初學者囉嗦寫法
def is_even(n):
    if n % 2 == 0:
        return True
    else:
        return False
```
請敏銳地觀察：比較式 `n % 2 == 0` 本身運算出來的結果，**就已經是一個道道地地的布林值（True 或 False）了**！根本不需要再外包一層 `if-else`。
因此，高手的精簡優雅寫法只需要一行：
```python
# 高手俐落一行流
def is_even(n):
    return n % 2 == 0
```
當我們在主程式中使用判定函數時，條件句會變成 `if is_even(x):`，程式碼讀起來就如同流暢的英文散文一般，可讀性提升數倍，更是將複雜題意拆解為模組化積木的關鍵起手式！


````python
# 範例 11.2.6：閏年判定函數的一行流美學與主流程應用
# 西元年份閏年規則：
# 1. 能被 400 整除者為閏年；
# 2. 或者能被 4 整除但「不能」被 100 整除者亦為閏年。
def is_leap_year(year):
    # 直接回傳複合布林運算式的真假值！
    return (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0)

# 在主流程中使用：讀起來宛如自然語言
test_years = [2000, 1900, 2024, 2023]
for y in test_years:
    if is_leap_year(y):
        print(f"西元 {y} 年 是閏年（366 天）")
    else:
        print(f"西元 {y} 年 是平年（365 天）")
````
