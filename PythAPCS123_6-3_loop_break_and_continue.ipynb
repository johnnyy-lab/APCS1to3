### 🛑 6-3-1 `break` 基礎：緊急煞車踏板——條件滿足立即跳離整個迴圈

**生活比喻：工廠流水線的緊急停機鈕**  
想像你是一家餅乾工廠的品檢員。傳送帶上排滿了 100 箱餅乾，機器正一箱一箱往前送（就像 `for i in range(100)`）。你的任務是檢查品質，但如果中途發現了任何一箱含有致命毒素或嚴重瑕疵，身為專業的品檢員，你不可能讓機器繼續運轉！你會立刻毫不猶豫地一巴掌拍在牆上的「紅色緊急煞車鈕」（這就是 `break`）！整條傳送帶瞬間徹底停擺，後續所有還沒送過來的餅乾全部不再檢查，直接下班！

**底層運作機制與語法解析**  
在 Python 程式語言中，`break` 是一個專門用來**「提前打破並終止整個迴圈」**的關鍵字。  
當電腦在迴圈內部執行到 `break` 時：
1. 它會**立刻停止當前回合的所有動作**。
2. 它會**永久跳出目前所在的這一層 `for` 迴圈**，絕對不會再去讀取 `range()` 接下來產生的任何數值！
3. 程式流程會直接跳到與 `for` 語句對齊的同一層外部程式碼繼續往下執行。

其核心標準骨架如下：
```python
for i in range(1, 101):
    if i == 5:
        break  # 當 i 數到 5 時，立刻拉下手煞車！
    print(i)
print("迴圈結束，繼續前進！")
```
在上述程式中，畫面上只會印出 1, 2, 3, 4。當 `i` 變成 5 時，條件成立觸發 `break`，連 5 都還來不及被 `print(i)` 印出，迴圈就已經徹底煙消雲散了！

**初學者最容易掉入的兩大致命陷阱：**
1. **陷阱一：在迴圈外孤零零地寫 `break`（語法報錯災難）**  
   `break` 的唯一宿命就是打破迴圈！如果你在沒有任何 `for`（或未來的 `while`）包覆的平坦程式碼或純 `if` 區塊中寫下 `break`，Python 會直接砸出嚴重語法錯誤：`SyntaxError: 'break' outside loop`！
2. **陷阱二：無條件 `break`（迴圈夭折悲劇）**  
   初學同學常忘記把 `break` 放在 `if` 條件判斷的肚子裡，直接把 `break` 寫在 `for` 的第一行。結果迴圈才剛邁出第 1 步，就立刻踩下煞車慘死在起跑點，完全失去迴圈反覆執行的價值！

**APCS 考試實務建議：**
在 APCS 與各類程式競賽中，題目通常會有嚴格的執行時間限制（通常是 1 秒）。如果某個目標一旦找到就不需要再搜尋（例如找到第一個合法答案或因數），及時使用 `break` 跳出迴圈能節省數萬至數百萬次的無謂計算，這是避免拿到「超時錯誤（TLE, Time Limit Exceeded）」的必備關鍵技術！

````python
# [2] Code 範例區：展示 break 提前中斷迴圈
# 目標：在 1 到 10 的數字中，尋找第一個平方大於 20 的整數

print("=== 尋找第一個平方大於 20 的數字 ===")

for i in range(1, 11):
    square = i * i
    print("目前檢查數字：", i, "，平方值為：", square)
    if square > 20:
        print("💡 找到目標了！", i, "的平方是", square, "> 20，立刻中斷迴圈！")
        break  # 煞車踏板：不再檢查後續的 6, 7, 8, 9, 10！

print("=== 迴圈已完全結束 ===")
````

### 🦘 6-3-2 `continue` 基礎：跨欄跳步踏板——跳過當前回合剩餘指令，直奔下一輪

**生活比喻：運動會的跨欄跑道與水果挑揀**  
想像你正在操場上參加 400 公尺跨欄障礙賽。跑道上每隔一段距離就立著一個欄架。當你跑到欄架前時，你不會像大卡車撞牆一樣直接停下來放棄比賽（那叫 `break`）！你會輕盈地凌空一躍，優雅跨過障礙，繼續朝著下一個目標奮力衝刺！  
再想像你在水果行幫柳丁貼認證標籤，檯面上有一整排柳丁等待處理。如果拿起來看到一顆長斑點的爛柳丁，你只要把它丟進垃圾桶，「略過貼標籤與裝箱步驟」，立刻伸手去拿下一顆柳丁！  
這就是 `continue` 的精神：**「跳過當前回合剩下的瑣事，直接進入下一回合！」**

**底層運作機制與語法解析**  
`continue` 關鍵字的作用是**「提早結束當前回合（Iteration），直接跳到下一輪」**。  
當 Python 在迴圈內部碰上 `continue` 時：
1. 它**絕不會摧毀或跳出整個迴圈**。
2. 它會**立即放棄當前回合中，排在 `continue` 後方的所有指令**。
3. 它會讓迴圈控制變數自動抓取 `range()` 的「下一個數值」，並且從迴圈的第一行指令重新開始下一輪循環！

其核心標準骨架如下：
```python
for i in range(1, 6):
    if i == 3:
        continue  # 遇到 3 就跨欄跳過！
    print(i)
```
執行這段程式，輸出結果會是：
```text
1
2
4
5
```
請大家睜大眼睛觀察：只有數字 3 憑空消失了！因為當 `i == 3` 時，`continue` 發動咒語，直接把後面的 `print(i)` 略過，瞬移到下一輪 `i = 4` 繼續執行！

**初學者最容易掉入的兩大致命陷阱：**
1. **陷阱一：把狀態更新放在 `continue` 後方（資料漏算災難）**  
   如果迴圈中有些動作是「不論如何每一輪都必須做」的（例如回合計數器 `round_count += 1`），千萬不能隨便擺在 `continue` 後面！否則一旦被跳過，該回合的更新就會完全被略過，造成難以察覺的邏輯臭蟲（Bug）。
2. **陷阱二：混淆 `break` 與 `continue` 的影響範圍**  
   初學同學常分不清這兩者：記住口訣——「**`break` 是把迴圈連根拔起直接拆台；`continue` 是該回合轉身就走、下回合準時報到！**」

**APCS 考試實務建議：**
在競賽與實務中，`continue` 最常用於「資料預先過濾（Filtering）」。當讀取到不符合資格的資料（例如偶數、負數、無效值）時，立刻一行 `continue` 略過，能讓後續的主要運算邏輯不用被包裹在深深的 `if-else` 巢狀結構中，代碼更加乾淨俐落！

````python
# [2] Code 範例區：展示 continue 跳過特定回合
# 目標：走訪 1 到 10，跳過所有 3 的倍數，只印出不是 3 的倍數的數字

print("=== 走訪 1 到 10（排除 3 的倍數）===")

for i in range(1, 11):
    # 檢查是否為 3 的倍數
    if i % 3 == 0:
        print(">> 偵測到", i, "是 3 的倍數，跳過本回合剩餘動作！")
        continue  # 跨欄跳步：直接進入下一輪，下方指令不會被執行
    
    print("合格數字：", i)

print("=== 走訪結束 ===")
````

### ⚔️ 6-3-3 `break` vs `continue` 深度對決——「提早收工」與「跳過換下一輪」的本質差異

**生活比喻：電影院驗票閘門的兩種突發狀況**  
想像你站在一家熱門電影院的入口驗票閘門前，門外大排長龍站著 50 位滿心期待的觀眾（`range(1, 51)`），由你負責逐一檢查票根：
- **突發狀況 A（`break` 模式：警報響起，立刻封館）**：  
  當驗到第 3 位觀眾時，他的背包裡居然掉出了一枚未爆彈！你大驚失色，立刻按下緊急廣播：「有危險！關閉大門！（`break`）」這時，不只是第 3 位觀眾不能看，**後方排隊的第 4 位到第 50 位觀眾全數被驅離，整個驗票作業原地終止**！
- **突發狀況 B（`continue` 模式：單人沒帶票，下一位請進）**：  
  當驗到第 3 位觀眾時，他摸了摸口袋發現忘記帶票。你客氣地說：「抱歉，請你先到旁邊補票，下一位！（`continue`）」這時，**只有第 3 位觀眾被略過沒能進場，排在後面的第 4 位到第 50 位觀眾依然井然有序地繼續上前驗票**！

**底層原理與流程圖對比**  
- **`break` 的流程圖路徑**：  
  執行 `break` ➔ 破壞迴圈控制結構 ➔ 直接將程式執行指針（Program Counter）丟到整個迴圈的最下方 ➔ 迴圈生命週期徹底畫下句點。
- **`continue` 的流程圖路徑**：  
  執行 `continue` ➔ 不破壞迴圈控制結構 ➔ 將執行指針瞬間拉回到迴圈的「頂端條件判斷處」➔ 計數變數前進一步，開始嶄新的下一回合。

讓我們用一段精確的代碼來觀察兩者對同一組數字 `1 到 5` 產生的天壤之別：
```python

    if i == 3:

    print(i, end=" ")
# 印出結果：1 2 （遇到 3 整條斷掉，後面的 4, 5 永無天日）

# 測試 continue

        continue

# 印出結果：1 2 4 5 （唯獨跳過 3，後續的 4, 5 正常輸出）
```

**常見錯誤陷阱：**
在題目敘述中，常常會出現「忽略」、「跳過」、「終止」、「停止」等關鍵字。  
很多同學常常一看到「如果 ... 就不要加」就隨手寫了 `break`，結果把後面幾千筆合法測資一口氣全部腰斬，慘遭 WA（Wrong Answer）；或者看到「一旦發現 ... 就不再處理」卻誤寫成 `continue`，導致後面成千上萬筆無效資料繼續被讀入與重複計算，慘遭 TLE（Time Limit Exceeded）！

**APCS 實務建議：解題審題心法**
- 看到題意寫著：**「遇到某條件即結束輸入 / 不再處理後續資料 / 提早找出答案」** ➔ 毫不猶豫使用 `break`！
- 看到題意寫著：**「若某數為負數則不予計算 / 排除特定例外 / 略過無效項目」** ➔ 毫不猶豫使用 `continue`！

````python
# [2] Code 範例區：同一組數據下 break 與 continue 的動態對比展示

print("--- 實驗 1：使用 break 提前中止 ---")
for i in range(1, 6):
    if i == 3:
        print("[遇見 3，觸發 break！]")
        break
    print("執行回合：", i)
print("實驗 1 結束，此時迴圈外繼續執行。\n")

print("--- 實驗 2：使用 continue 跳過本輪 ---")
for i in range(1, 6):
    if i == 3:
        print("[遇見 3，觸發 continue！]")
        continue
    print("執行回合：", i)
print("實驗 2 結束，全部輪次檢驗完畢。")
````

### 🚩 6-3-4 `break` 搭配布林旗標（Flag）的高效搜尋模式——「找到就停」的線性搜尋與質數檢驗

**生活比喻：圖書館尋找特定藏書與立起旗幟**  
想像你走進一間典藏百萬冊圖書的國家圖書館，要找一本編號為 `777` 的珍貴古籍。你在長長的書架前一本一本拿下來核對（這叫「線性搜尋」）。  
在開始找之前，你在手心寫上「尚未找到」（這叫做布林旗標初始化：`found = False`）。  
你從第 1 本開始看，看到第 38 本時，你驚喜地發現：「天啊！這本就是 777 號古籍！」  
這時候，你會做兩件事：
1. **立起成功旗幟**：把手心翻開改成「已經找到！」（`found = True`）。
2. **抱書離館，立刻下班**：毫不猶豫地抱著書走向櫃台借閱，踩下煞車（`break`）！  
請問：你還會繼續站在書架前，把剩下 999,962 本書一本一本拿下來檢查完嗎？當然不可能！

**底層運作機制：旗標變數（Flag）與提早中斷的黃金拍檔**  
在第五章 5-7 單元中，我們學習了「布林旗標變數（Flag）」。當旗標遇上 `for` 迴圈與 `break`，便誕生了演算法世界中最經典、最無懈可擊的**「提早結束搜尋模式（Early-Exit Search Pattern）」**：
```python
found = False  # 步驟一：出發前旗標設為 False
target_val = 0

for i in range(1, n + 1):
    if 條件滿足:
        found = True       # 步驟二：找到了，立起旗幟！
        target_val = i    # 記錄目標物
        break             # 步驟三：功德圓滿，立即中斷！不再浪費 CPU 效能

# 步驟四：退回最外層，檢查旗標狀態
if found:
    print("順利尋獲目標：", target_val)
else:
    print("在搜尋範圍內未找到目標！")
```

**質數判定（Prime Number Test）的極致提早加速：**  
什麼是質數？大於 1 的正整數中，除了 1 和它本身之外「沒有任何其他因數」的數就是質數。  
如果要檢驗數字 $n$（例如 35）是否為質數：
- 我們只要讓迴圈從 2 跑到 $n - 1$。
- 只要在中途抓到「任何一個數字可以整除 $n$（`n % i == 0`）」，就鐵證如山證明它絕對不是質數！
- 此時立刻將旗標 `is_prime = False` 並毫不客氣地 `break`！
- 不需要把所有因數找齊，找到第一個反例就足以給它判死刑！

**初學者最容易掉入的兩大致命陷阱：**
1. **陷阱一：在迴圈肚子裡亂寫 `else: print("找不到")`（過早判定慘劇）**  
   初學者最常見的錯誤，就是把 `else` 接在迴圈內部的 `if` 後面：
   ```python
   # 錯誤示範！千萬不要這樣寫！
   for i in range(1, 10):
       if i == target:
           print("找到了")
           break
       else:
           print("找不到")  # 慘劇：只要第 1 個數字不是 target，就直接大喊找不到！
   ```
   請牢牢記住：**「只要還有數字沒檢查完，你就沒有資格宣布找不到！」** 只有跑完所有機會都沒觸發旗標，才能在迴圈外宣告找不到！
2. **陷阱二：忘了立旗標就 `break`**  
   若只寫了 `break` 卻忘了把 `found` 設為 `True`，迴圈結束後程式根本無法辨別是「中途找到煞車跳出」還是「跑完全部數字失望結束」。

**APCS 考試實務建議：**
在 APCS 實作題中，多數搜尋與檢驗題型（如檢查是否存在指定數字、檢查兩數是否有公因數等）都必須使用「旗標 + break」架構，代碼結構清晰且執行速度最快。

````python
# [2] Code 範例區：展示 break 搭配旗標進行質數檢驗
# 檢驗數字 35 是否為質數

n = 35
is_prime = True  # 步驟一：出發前預設它是質數

print("開始檢驗整數", n, "是否為質數...")

for i in range(2, n):
    if n % i == 0:
        print("💡 抓到了！", n, "可以被", i, "整除（", n, "=", i, "*", n // i, "）！")
        is_prime = False  # 步驟二：找到因數，推翻質數假設！
        break             # 步驟三：立即中斷，不需要再檢查後續因數！

# 步驟四：迴圈外依據旗標總結判定
if is_prime and n > 1:
    print(">> 結論：", n, "是質數（Prime）！")
else:
    print(">> 結論：", n, "不是質數（Composite）！")
````

### 🛡️ 6-3-5 `continue` 在串流數據清洗與安全過濾中的實戰應用——挑選符合條件的純淨數據

**生活比喻：頂級淨水廠的多層防護濾網**  
想像你是一座現代化淨水廠的總工程師。從河川引入的原水水流中，夾雜著各種奇形怪狀的雜質：有些含有大石頭（負數異常值）、有些含有懸浮泥沙（非指定範圍）、有些含有化學漂白劑（極端異常值）。  
在進入最後的核心「殺菌加總水庫」之前，管線上安裝了三道全自動攔截濾網：
- 第一道濾網檢測：「偵測到大石頭！直接排放衝走，不進主水庫！（`continue`）」
- 第二道濾網檢測：「偵測到超標懸浮泥沙！直接排放衝走，不進主水庫！（`continue`）」  
只有通過重重檢驗、純淨無瑕的合格水源，才會順利流到最後的總水庫中進行累加與供水！

**底層原理：守衛條件（Guard Clauses）模式讓程式碼扁平清爽**  
在實際撰寫程式或解 APCS 題目時，我們常常需要對輸入的每一筆資料進行嚴格的合法性檢查。  
如果不使用 `continue`，我們的程式碼往往會長成這樣：
```python
# 傳統多層巢狀寫法（縮排深、像洋蔥或金字塔，不易閱讀）
for _ in range(n):
    score = int(input())
    if score >= 0:
        if score <= 100:
            if score >= 60:
                valid_sum += score
                pass_count += 1
```
這種層層包裹的縮排被軟體工程界稱為「箭頭型代碼（Arrow Anti-Pattern）」。  
而如果改用 **`continue` 作為守衛條件（Guard Clause）**，程式碼瞬間變得一目了然：
```python
# 守衛條件優雅寫法（及早過濾、縮排扁平、乾淨俐落）
for _ in range(n):
    score = int(input())
    # 守衛 1：排除超出合理範圍的異常成績
    if score < 0 or score > 100:
        continue
    # 守衛 2：排除不及格成績
    if score < 60:
        continue
    
    # 真正的主核心邏輯：能夠活著走到這裡的，絕對都是 60~100 的及格好成績！
    valid_sum += score
    pass_count += 1
```

**常見錯誤陷阱：**
在設計守衛過濾條件時，初學者常犯的邏輯錯誤是「**條件邏輯寫反**」。  
例如：我們明明想要「保留大於 0 的數字」，但在寫守衛時，要過濾掉的是它的反面條件——「`if x <= 0: continue`」！若不小心寫成 `if x > 0: continue`，就會把所有合格的好數據全部丟棄，反而只剩下壞數據被累加！

**APCS 考試實務建議：**
在 APCS 題目中，資料清洗（Data Sanitization）非常普遍。使用 `continue` 能讓你的解題思維專注在「挑出不合格項目先踢掉」，剩下的區塊就能安心處理題目的核心演算法，大幅減少邏輯筆誤與除錯時間！

````python
# [2] Code 範例區：展示 continue 在數據清洗中的守衛條件應用
# 目標：走訪 1 到 20，計算所有「不是 2 的倍數」且「不是 5 的倍數」的純淨數字總和

print("=== 數據清洗過濾展示（走訪 1 到 20）===")

pure_sum = 0
pure_count = 0

for num in range(1, 21):
    # 守衛條件 1：過濾 2 的倍數
    if num % 2 == 0:
        continue
    
    # 守衛條件 2：過濾 5 的倍數
    if num % 5 == 0:
        continue
    
    # 通過雙重守衛的純淨數據
    print("通過檢驗的純淨數字：", num)
    pure_sum += num
    pure_count += 1

print("--- 統計結果 ---")
print("合格純淨數字個數：", pure_count)
print("合格純淨數字總和：", pure_sum)
````
