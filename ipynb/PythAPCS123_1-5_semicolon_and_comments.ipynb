### 🚆 1-5-2 單行多指令（;）——省空間的車廂連接器

在 Python 中，通常一條指令就單獨佔據一行，每按一次 Enter 鍵，就是一道新的命令。
但是，有時候我們只是想做一些非常簡短的動作（例如設定兩個變數的起點），如果每一個動作都要換一行，整個螢幕滑鼠要滾動很久才能看完。

這時候，Python 提供了一個小工具——「**分號（`;`）**」！
吳邦一老師在講義第 5 頁指出：
「**多行指令可以分號隔開放在同一行，執行順序是由左到右一一執行。**」

例如：
```python
a = 5; b = 3
```

你可以把分號想像成「**火車車廂的連接器掛鉤**」！
`a = 5` 是一節載著 5 的車廂，`b = 3` 是一節載著 3 的車廂。分號 `;` 就像掛鉤把兩節車廂串在同一軌道上。直譯器火車頭開過去時，完全不會混淆，它會嚴格遵守「**由左至右、一節一節依序執行**」的規則，先執行左邊的 `a = 5`，執行完畢後再執行右邊的 `b = 3`！

````python
# [2] Code 範例區：利用分號 ; 將兩道賦值指令寫在同一行
a = 5; b = 3
print(a)
print(b)
````

### 📏 1-5-3 分號的優雅紀律——過猶不及與一組關聯原則

既然分號 `;` 可以把好幾行指令擠成一行，那我們是不是可以把整支程式 100 行指令，全部用分號串成一條又臭又長的貪食蛇呢？

答案是「**絕對不行！千萬別這麼做！**」
吳邦一老師在講義第 9 頁給了大家一段語重心長的高手箴言：
「**不要把指令寫得太長，萬一有錯搞不清楚錯在哪裡，而且可讀性差，通常我們只把『一組』指令放在同一行！**」

想像一下你出門穿衣服的整理原則：
你可以把一雙襪子捲在一起收進同一個抽屜格（這叫『一組關聯』，非常合理）；但是如果你硬要把帽子、上衣、外套、褲子、皮帶、鞋子全部硬塞進同一隻襪子裡，那整團衣服只會變成解不開的死結！

在程式設計裡，如果一行代碼塞了太多指令，當直譯器報錯時，它只會告訴你「第 5 行出錯」，但第 5 行如果有 8 個指令，你根本分不清楚到底是哪一個指令在搞鬼！

**最佳實務原則**：
只有像 `i = 1; ans = 0` 這種「同時初始化設定一對變數」的情境，才適合放在同一行；其他複雜的運算或控制結構，請老老實實分行撰寫，保持清晰易讀！

````python
# [2] Code 範例區：優雅示範！只把緊密關聯的一組變數初始化放在同一行
i = 1; ans = 0  # 良好示範：迴圈或演算法啟動前的乾淨初始化
ans += i
print(ans)
````

### 🛑 1-5-4 註解的除錯魔法——暫停嫌疑犯的「程式碼封印術」

註解除了用來寫備忘錄之外，還有一個全天下所有工程師天天都在使用的神奇超能力——「**程式碼封印術（Comment Out / 註解除錯法）**」！

想像一下：你寫了一支包含 10 行代碼的程式，按下去執行後，畫面突然噴出一長串紅色錯誤警告。這時候你心中懷疑：「一定是第 4 行那行奇怪的算式在搗蛋！」
但是，你又不敢隨便把第 4 行直接按 Delete 鍵刪掉，因為萬一它是無辜的，你待會還要花時間重新打字背出來，那多冤枉啊！

這時候，只要在第 4 行的最前方輕輕加上一個「`#`」符號：
```python
# bad_line = 10 / 0  (這行被封印了！)
```

唰！一瞬間，就像警察在犯罪現場拉起了「黃色封鎖線」一樣，直譯器瞬間把這行嫌疑犯當成透明空氣，繞過它繼續執行後面的代碼！
如果程式順利跑通了，就代表你的直覺是正確的；如果程式還是報錯，你只要把 `#` 刪掉，就能立刻解除封印、毫髮無損地復原！
這種用 `#` 暫時切換程式碼開關的技巧，是 APCS 考場上最快速、最安心的除錯神技喔！

````python
# [2] Code 範例區：展示使用 # 封印有錯誤的程式碼行
val = 100

# 封印有問題的算式（若拿掉 # 執行 10/0 會引發 ZeroDivisionError 崩潰！）
# error_calc = val / 0

print(val)  # 程式順利運行通過！
````

### 📐 1-5-5 長指令優雅折行——小括號 `(...)` 隱式換行與反斜線（`\`）防禦

**生活比喻：超長捲軸與折頁書本的閱讀藝術**  
在前面 1-5-3 中，我們學到了吳邦一老師的叮嚀：「不要把指令寫得太長，可讀性差」。
但是，想像一下這個日常情境：你在替學校園遊會算帳，或者在遊戲中計算角色身上十幾樣裝備的數值總和。就算我們每個變數都取了清晰的好名字，光是把這些變數全部用加號 `+` 連在一起，整道算式就會瞬間飆破 100 個字元，遠遠超出螢幕的右側邊界！這時你必須用滑鼠水平捲動好幾公尺才能看完全貌，就像一張攤開三公尺長的手工捲軸一樣，閱讀起來非常痛苦。
那麼，當一道指令「本身就是很長」的時候，我們能不能把它像折疊書本一樣，優雅地拆成多行呢？答案是「**當然可以！這就是 Python 的折行術（Line Continuation）！**」

**底層運作機制：兩大換行秘技**  
在 Python 中，一般情況下直接按下 Enter 換行代表一道指令結束；但如果我們想告訴電腦「這道算式還沒算完，請看下一行」，有兩種核心方式：
1. **小括號隱式換行（Implicit Line Continuation，PEP 8 官方最強烈推薦！）**：
   - Python 語法有一條非常溫柔且聰明的規則：**凡是在圓括號 `(...)`、方括號 `[...]` 或花括號 `{...}` 內部，換行符號會被直譯器自動忽略！**
   - 直譯器會靜靜等待你打出右括號 `)` 閉合為止。因此，只要用一對小括號 `(...)` 把長算式包起來，你就可以在加號 `+` 後面放心地按下 Enter 換行，每行還能自由縮排對齊，代碼瞬間變得賞心悅目！
2. **反斜線顯式換行（Explicit Line Continuation，`\`）**：
   - 在行尾打上一個反斜線 `\`，代表向直譯器發出信號：「這行指令還沒講完，請接續下一行」。

**常見錯誤陷阱：反斜線後方的「致命隱形空格」**  
⚠️ **為什麼全世界的 Python 高手都一致推薦小括號，而盡量避開反斜線呢？**
因為反斜線 `\` 隱藏著一個極為致命的考場地雷：**在反斜線的後方，絕對不能有任何字元，連一個看不見的空白鍵（Space）都不行！**
如果你手滑在 `\` 後面敲了一下空白鍵才按 Enter，Python 就會立刻噴出令人一頭霧水的崩潰警告：`SyntaxError: unexpected character after line continuation character`！初學者往往盯著螢幕除錯半小時，完全看不出肉眼可見的代碼到底錯在哪裡。
此外，如果使用小括號時**忘了打結尾的右括號 `)`**，直譯器會一路往後吞噬代碼，最終拋出 `SyntaxError: unexpected EOF while parsing`。因此，**養成成對輸入括號 `()` 的習慣，並優先使用小括號折行**，是最穩健的專業防禦手法！

**APCS 考試實務建議：**  
在競技程式設計與 APCS 考場上，長公式（如多項加權總分、曼哈頓座標加總）十分常見。透過小括號折行讓每一行長度保持在 80 個字元內，不僅除錯時能逐行核對變數名稱與正負號，也能徹底避免因螢幕自動換行導致的眼花看錯！
