### 🔍 3-2-1 ord() 字元身分證——探訪 ASCII 編碼與連續性特徵

想像你來到了一個國際機場的海關櫃台。每位旅客（英文字母、數字符號）走上海關前，海關人員都會拿出一台「身分證掃描器」，對著旅客「逼」一聲，螢幕上就會跳出這位旅客的專屬身分證號碼。在 Python 中，這台神奇的掃描器就是內建函數 **`ord()`**（Ordinal 的縮寫，代表序數、編號）！

**運作機制與三大連續性特徵：**
吳邦一老師在講義第 46 頁與第 50 頁特別強調：初學者**完全不需要死記整張 ASCII 表**！你只需要記住三個最核心的「連續性」黃金基準點：
1. **大寫英文字母 `'A'` ～ `'Z'` 是連續的**：`ord('A')` 是 **`65`**，`'B'` 就是 `66`，一直連續排到 `'Z'`（`90`）。
2. **小寫英文字母 `'a'` ～ `'z'` 是連續的**：`ord('a')` 是 **`97`**，`'b'` 就是 `98`，一直連續排到 `'z'`（`122`）。
3. **數字字元 `'0'` ～ `'9'` 也是連續的**：`ord('0')` 是 **`48`**，一直排到 `'9'`（`57`）。注意：字元 `'0'` 的編碼可不是 `0`，而是 `48` 喔！

**常見錯誤陷阱（重要！報錯解析）：**
`ord()` 的肚子非常小，**一次只能吃「長度剛好為 1 的單一字元」**！
如果你不小心傳入兩個以上的字母，例如 `ord('AB')` 或是傳入空字串 `ord('')`，Python 就會立刻噴出：
`TypeError: ord() expected a character, but string of length 2 found`！
請牢記：它專屬為單一字元服務，不能對整個單字或句子直接使用喔！


```python
# [2] Code 範例區：使用 ord() 查驗字元的 ASCII 身分證字號

# 範例 1：大寫字母與小寫字母基準點
print(ord('A'))  # 印出 65（大寫基準）
print(ord('B'))  # 印出 66（連續緊接）
print(ord('a'))  # 印出 97（小寫基準）
print(ord('b'))  # 印出 98

# 範例 2：數字字元的 ASCII 編碼
print(ord('0'))  # 印出 48（注意：不是 0！）
print(ord('1'))  # 印出 49

# 範例 3：查驗特殊標點符號
print(ord('!'))  # 印出 33
print(ord(' '))  # 印出 32（空白鍵也是一個合法字元）
```

### 🧙‍♂️ 3-2-2 chr() 編碼變身術——數字召喚字元

如果說 `ord()` 是把字元還原成數字的身分證掃描器，那麼它的孿生兄弟 **`chr()`**（Character 的縮寫）就是一座神奇的「字元 3D 列印召喚機」！
只要你把一個整數編號放進 `chr()` 肚子裡，它就能像變魔術一樣，精準將對應的字元召喚到螢幕上！

**運作機制與底層原理：**
- 輸入整數 `65` 呼叫 `chr(65)`，召喚出大寫英文字母 **`'A'`**。
- 輸入整數 `97` 呼叫 `chr(97)`，召喚出小寫英文字母 **`'a'`**。
- 輸入整數 `48` 呼叫 `chr(48)`，召喚出數字字元 **`'0'`**。
- `ord()` 與 `chr()` 是一對完美的互逆操作（Inverse Operations）：
  `chr(ord('K'))` 會回到 `'K'`；而 `ord(chr(88))` 算出來剛好就是 `88`！

**常見錯誤陷阱（ValueError 邊界守則）：**
`chr()` 只接受代表 Unicode / ASCII 的合法非負整數（ASCII 範圍通常在 0 到 127 之間）。
如果你傳入負數，例如 `chr(-1)`，或是傳入超過範圍的荒謬數字，Python 會立刻發出警告：
`ValueError: chr() arg not in range(0x110000)`！
另外，傳給 `chr()` 的必須是**整數**，傳入小數（如 `chr(65.0)`）或字串（如 `chr('65')`）會引發 `TypeError`，務必確認傳入的是純整數喔！


```python
# [2] Code 範例區：展示 chr() 召喚字元與互逆操作

# 範例 1：使用數字召喚英文字母
print(chr(65))   # 印出 'A'
print(chr(66))   # 印出 'B'
print(chr(97))   # 印出 'a'

# 範例 2：連續數字召喚字母接龍
start_code = 70  # 70 是 'F'
print(chr(start_code), chr(start_code + 1), chr(start_code + 2))  # 印出 F G H

# 範例 3：ord 與 chr 互逆驗證
my_char = 'X'
reconstructed = chr(ord(my_char))
print(reconstructed == my_char)  # 印出 True！完全復原！
```

### ⚖️ 3-2-3 經典差值 32（0x20）——純手工大小寫無痛轉換

在英文字母的世界裡，大寫字母與小寫字母就像是住在不同樓層的孿生兄弟。你有沒有算過，他們樓層之間相差幾公尺呢？
讓我們來做個簡單的減法：
小寫 `'a'` 的編碼是 `97`，大寫 `'A'` 的編碼是 `65`：
$$97 - 65 = 32$$
再檢查一組：小寫 `'b'`（`98`）減去大寫 `'B'`（`66`），依然是 **`32`**！
在十六進位制中，`32` 剛好就是 **`0x20`**（$2 \times 16 = 32$）！

**吳邦一老師講義第 10 頁與第 46 頁的經典必殺技：**
所有的英文字母，大寫與小寫之間**永遠嚴格相差 32**！這項特性是計算機歷史上最偉大的精巧設計之一：
- **大寫轉小寫**：只要把大寫編碼「加上 32」，再召喚成字元：
  `chr(ord('C') + 32)` ➔ 就會無痛變成 `'c'`！
- **小寫轉大寫**：只要把小寫編碼「減去 32」，再召喚成字元：
  `chr(ord('g') - 32)` ➔ 就會俐落變身為 `'G'`！

**APCS 考場實務價值：**
很多初學者在寫競賽題目時，若忘記了字串內建的 `.upper()` 或 `.lower()`，只要心中默念「差值 32」，單靠 `ord()` 與 `chr()` 就能純手工打造出 100% 正確的大小寫轉換器，在考場上無往不利！


```python
# [2] Code 範例區：展示大小寫字母差值 32 的手工轉換魔法

# 範例 1：驗證大小寫差值永遠為 32（十六進位 0x20）
print(ord('a') - ord('A'))  # 印出 32
print(ord('z') - ord('Z'))  # 印出 32

# 範例 2：大寫轉小寫（+32）
upper_char = 'E'
lower_char = chr(ord(upper_char) + 32)
print(lower_char)           # 印出 'e'

# 範例 3：小寫轉大寫（-32）
lower_letter = 'q'
upper_letter = chr(ord(lower_letter) - 32)
print(upper_letter)         # 印出 'Q'
```

### 📊 3-2-4 字母相對索引 ord(c) - ord('A')——26 個字母計數器起手式

在後續章節與 APCS 考題中，有一種無比高頻的題目：
「輸入一篇文章，統計 26 個英文字母每個字母出現了幾次？」
面對 26 個英文字母，難道我們要宣告 26 個變數 `count_a, count_b, ... count_z` 嗎？
當然不行！那樣寫會讓程式碼冗長到無法收拾！

**吳邦一老師講義第 26 頁與第 46 頁的提示秘笈：**
在後續進階章節中，我們可以用 26 個置物櫃格子，讓編號 `0` 代表 `'A'`，索引 `1` 代表 `'B'`，...，索引 `25` 代表 `'Z'`！
但問題來了：如何把英文字母快速變成 `0` 到 `25` 的整數索引呢？
核心公式正是：**「字母相對索引（Relative Indexing）」**！
$$\text{index} = \text{ord}(c) - \text{ord}('A')$$
- 當 $c = 'A'$ 時：$65 - 65 = \mathbf{0}$（正好對應第一個格子！）
- 當 $c = 'B'$ 時：$66 - 65 = \mathbf{1}$（正好對應第二個格子！）
- 當 $c = 'Z'$ 時：$90 - 65 = \mathbf{25}$（正好對應最後一個格子！）

如果是小寫字母，公式只要改成減去小寫基準 `'a'`：
$$\text{index} = \text{ord}(c) - \text{ord}('a')$$
掌握了這道公式，你就擁有了打通字元統計與陣列計數器之間的終極橋樑！


```python
# [2] Code 範例區：展示字母相對索引計算與 26 字母計數器原理

# 範例 1：大寫字母轉 0~25 相對序號
print(ord('A') - ord('A'))  # 印出 0（A 是第 0 號）
print(ord('B') - ord('A'))  # 印出 1（B 是第 1 號）
print(ord('Z') - ord('A'))  # 印出 25（Z 是第 25 號）

# 範例 2：小寫字母轉 0~25 相對序號
print(ord('a') - ord('a'))  # 印出 0
print(ord('c') - ord('a'))  # 印出 2

# 範例 3：查驗任意英文字母在字母表中的位置
char_sample = 'D'
idx = ord(char_sample) - ord('A')
print("D 的字母編號（0 起算）:", idx)        # 印出 3
print("若從第 1 個起算（自然序號）:", idx + 1)  # 印出 4
```

### 🛡️ 3-2-5 凱撒密碼位移——字元循環輪替加密小魔法

在西元前古羅馬時代，尤利烏斯·凱撒（Julius Caesar）將軍為了傳送軍事機密，發明了一種簡單卻強大的加密方法：把信中的每一個英文字母，在字母表上「往後位移固定步數」！
例如，如果約定位移量為 `3`，那麼字母 `'A'` 就會變成 `'D'`，字母 `'B'` 就會變成 `'E'`！這就是資訊科學中最著名的「**凱撒密碼（Caesar Cipher）**」！

**運作機制與底層算式：**
透過今天學到的 `ord()` 與 `chr()`，我們可以在 Python 中優雅實作這項傳奇密碼：
1. 先用 `ord(c)` 取得字母的 ASCII 身分證。
2. 加上位移步數 `shift`，得到新的編號：`new_code = ord(c) + shift`。
3. 用 `chr(new_code)` 召喚出加密後的密文字元！

**如果位移超過 'Z' 該怎麼辦？（循環求模進階思維）**
如果字母是 `'Y'`，往後移 3 步就會衝出大寫字母範圍。在進階演算法中，我們會結合前一小節的相對索引與模除 `% 26` 達成無縫圓環輪替：
$$\text{cipher} = \text{chr}((\text{ord}(c) - \text{ord}('A') + \text{shift}) \% 26 + \text{ord}('A'))$$
在基礎階段，只要理解「**字元 ➔ 編號 ➔ 數學運算 ➔ 字元**」這套標準處理流，你就已經掌握了高階文字加密與字串處理的核心本質！


```python
# [2] Code 範例區：展示凱撒密碼字元位移加密

# 範例 1：基本字母位移加密
origin_char = 'A'
shift_step = 3
cipher_ascii = ord(origin_char) + shift_step
cipher_char = chr(cipher_ascii)
print(cipher_char)  # 印出 'D'（A 往後移 3 步）

# 範例 2：解密操作（反向倒退 3 步）
decrypted_char = chr(ord(cipher_char) - shift_step)
print(decrypted_char)  # 成功復原為 'A'！
```
