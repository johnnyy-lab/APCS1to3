### 🔄 6-3-1 什麼是 while 迴圈？「會重複執行的 if」心智模型

**生活故事：水餃大胃王與手機低電量警示**  
想像你在吃水餃，你的大腦控制邏輯通常不是「我今天要精確吃 23 顆水餃」（這是 `for` 迴圈的已知次數思維），而是「只要我肚子還覺得餓（條件為真），我就繼續吃下一顆水餃」！直到肚子飽了（條件變成假），你才會放下筷子停止進食。  
又例如手機的充電警示器：「只要電量低於 20%（條件為真），就每隔一分鐘閃爍紅燈警告」；直到你插上充電線讓電量回升到 20% 以上，警示才會解除。

**運作機制與心智模型：**  
在 Python 中，`while` 英文原意是「當……的時候」。它的語法結構與單向分支 `if` 極為相似：
```python
while 條件判斷式:
    縮排執行的迴圈主體
```
兩者的唯一差別在於：
- `if` 是**「一次性檢查」**：條件為 True 就執行裡面的區塊一次，執行完就離開。
- `while` 則是**「重複性檢查」**：條件為 True 就執行裡面的區塊；**執行完畢後，控制權會立刻跳回開頭，再次檢查條件！**只要條件依然為 True，就會周而復始不斷循環；直到某一次檢查時條件變成了 False，才會跳出迴圈，繼續往下執行主程式。

**常見錯誤陷阱：**  
- **把 while 誤當成 if**：初學者有時會誤以為 while 執行完一次就結束了，忘記它會自動迴轉再次檢查條件。
- **縮排錯誤**：while 下方的指令必須統一縮排 4 個空格，縮排結束即代表跳出迴圈後的代碼。

**APCS 考試實務建議：**  
當題目的重複次數「無法在開始前直接算出」，而是「依據某個數值狀態（如數字還沒除盡、餘數不為 0、指標還沒相遇）」來決定何時停機時，`while` 迴圈就是首選武器！


````python
# [2] Code 範例區：體驗「會重複執行的 if」——吃水餃模型
hungry_level = 3  # 飢餓指數初始為 3

print("=== 準備開始用餐 ===")
# 只要飢餓指數大於 0，就重複執行
while hungry_level > 0:
    print(f"吃了一顆美味水餃！目前飢餓指數剩餘：{hungry_level}")
    hungry_level -= 1  # 每吃一顆，飢餓指數減少 1

print("=== 肚子吃飽了，停止進食！ ===")
print(f"最終飢餓指數：{hungry_level}")
````

````python
# ==========================================
# [4] Code 練習題
# 任務說明：
# 某款益智遊戲的魔王體力值為 boss_hp。
# 勇者每發動一次攻擊，魔王體力就會減少 15 點。
# 請撰寫一個 while 迴圈，只要魔王體力 boss_hp 大於 0，
# 就持續扣除 15 點體力，並將攻擊次數 attack_count 累加 1。
# 迴圈結束後，請印出總共攻擊了幾次。
#
# 【公開測試資料 1】
# 設定：boss_hp = 40, attack_count = 0
# 運算過程：40 -> 25 -> 10 -> -5 (停止)
# 預期 attack_count = 3
#
# 【公開測試資料 2】
# 設定：boss_hp = 60, attack_count = 0
# 運算過程：60 -> 45 -> 30 -> 15 -> 0 (停止)
# 預期 attack_count = 4
# ==========================================
# 請在下方撰寫你的程式碼並執行測試：

boss_hp = 40
attack_count = 0

while boss_hp > 0:
    boss_hp -= 15
    attack_count += 1

print("總攻擊次數：", attack_count)

# 公開測資驗證：
assert attack_count == 3, "公開測資 1 驗證失敗"
print("公開測資 1 通過！")
````

### ⚙️ 6-3-2 while 迴圈的三大黃金要件：初始值、終止條件、更新步進（缺一不可！）

**生活故事：操場長跑的三大要素**  
學校舉辦 1000 公尺耐力長跑測驗，任何一個能順利完成測驗的跑者都必須具備三件事：
1. **站上起跑線（初始值）**：知道自己從 0 公尺出發；
2. **盯緊終點線（終止條件）**：知道只要「跑動距離 < 1000」就得繼續邁開雙腿；
3. **一步一步往前跑（更新步進）**：雙腳必須持續邁進（每次前進 1 公尺），如果跑者站在原地不動，他就永遠抵達不了終點！

**運作機制與三大黃金支柱：**  
每一個正確且能安全停機的 `while` 迴圈，背後都必須緊密依賴以下三大要件：
1. **要件一：迴圈控制變數的初始值（Initialization）**  
   在進入 `while` 迴圈之前的那一行，必須先為控制變數設定初始狀態（例如 `n = 1` 或 `num = 12345`）。若未初始化就直接在條件式使用，會當場引發 `NameError`。
2. **要件二：迴圈繼續執行的條件式（Condition）**  
   位於 `while` 關鍵字後方的布林判斷式（例如 `n <= 5`）。只要條件為 True 迴圈就運轉，條件一旦為 False 迴圈立刻煞車。
3. **要件三：迴圈內部的推進步進（Update / Step）**  
   在迴圈內部的縮排程式碼中，**必須至少有一行程式碼負責改變控制變數的值**（例如 `n += 1` 或 `num //= 10`），讓變數的狀態逐步朝著「終止條件」的方向逼近！

**常見錯誤陷阱：**  
- **頭號世紀大災難——漏寫步進（Update）！**  
  初學者最常犯的致命錯誤就是寫了 `while i <= 10:`，但在迴圈主體內忘記寫 `i += 1`。由於 `i` 的值永遠固定在 1，`1 <= 10` 永遠為 True，直譯器就會陷入永無止境的「無窮迴圈（Infinite Loop）」，CPU 使用率飆高且程式陷入假死狀態！

**APCS 考試實務建議：**  
在考場撰寫任何 `while` 迴圈時，請在心中默念口訣：**「先設起點、盯緊條件、別忘邁步」**。寫完冒號的第一件事，就先確認迴圈最底部有沒有寫推進指令！


````python
# [2] Code 範例區：檢視三大黃金要件的協同運作
# 要件一：起跑線（初始值宣告）
count = 1

print("=== 倒數計時即將開始 ===")

# 要件二：終點線（條件檢查）
while count <= 5:
    print(f"正在執行第 {count} 次巡檢作業...")
    
    # 要件三：邁開步伐（更新步進！如果少了這行，程式會永遠停在 count = 1）
    count += 1

print("=== 巡檢作業全部圓滿完成！ ===")
print(f"迴圈結束後 count 的殘留值為：{count}")
````

````python
# ==========================================
# [4] Code 練習題
# 任務說明：
# 請使用 while 迴圈計算從 1 加到 n 的總和（sum_val）。
# 宣告變數 cur = 1，當 cur <= n 時，將 cur 累加進 sum_val，
# 並且每輪將 cur 增加 1。
#
# 【公開測試資料 1】
# 設定：n = 5
# 預期輸出：sum_val = 15 (1 + 2 + 3 + 4 + 5)
#
# 【公開測試資料 2】
# 設定：n = 10
# 預期輸出：sum_val = 55 (1 + 2 + ... + 10)
# ==========================================
# 請在下方撰寫你的程式碼並執行測試：

n = 5
cur = 1
sum_val = 0

while cur <= n:
    sum_val += cur
    cur += 1

print(f"1 到 {n} 的總和為：{sum_val}")

# 測資驗證：
assert sum_val == 15, "公開測資 1 驗證失敗"
print("公開測資 1 通過！")
````

### ⚖️ 6-3-3 for vs while 深度對決：已知次數 vs 未知次數的題目選型指南

**生活故事：搭公車與投販賣機**  
想像兩種不同的生活情境：
- **情境 A（搭公車）**：你看著站牌路線圖，明確知道從「火車站」坐到「學校」剛好經過 5 站，這時你的心理模型是數 1, 2, 3, 4, 5 站下車（這是典型的 `for` 迴圈）。
- **情境 B（自動販賣機投零錢）**：一罐可樂 35 元，你從口袋掏出一把零錢，你並不知道要掏幾次，你只知道「只要累積投進去的金額還不到 35 元，我就得繼續翻口袋投零錢」（這是典型的 `while` 迴圈）。

**核心對決與選型決策樹：**  
在 Python 中，所有 `for` 迴圈理論上都能改寫成 `while` 迴圈，反之亦然。但為什麼我們需要同時擁有這兩種武器？關鍵就在於**「代碼語意清晰度與可維護性」**！
- **什麼時候選 `for` 迴圈？（已知次數 / 固定範圍 / 序列走訪）**
  - 題目開頭明講：「給定正整數 $N$，接下來有 $N$ 筆資料」；
  - 走訪明確的數學區間（例如走訪 1 到 100 中的所有奇數）；
  - 走訪固定字串中的每一個字元。  
  *優勢*：`for` 搭配 `range()` 會自動初始化與自動步進，絕不會發生忘記寫 `i += 1` 的無窮迴圈慘劇！
- **什麼時候選 `while` 迴圈？（未知次數 / 狀態驅動 / 逼近目標）**
  - 重複次數無法預測，完全取決於運算過程中的數據變化（如整數除法直到商為 0）；
  - 連續逼近某個數學邊界；
  - 事件驅動的動態模擬遊戲。

**常見錯誤陷阱：**  
- **殺雞用牛刀**：明知道要跑 10 次，卻寫了 5 行帶有 `i = 0`, `while i < 10:`, `i += 1` 的繁瑣代碼。這不僅代碼冗長，更增加了打錯字引發死結的風險。

**APCS 考試實務建議：**  
審題第一步：先問自己「這個迴圈要跑幾圈是題目一開始就能算出來的嗎？」能算出來就毫不猶豫用 `for`；算不出來、由數值狀態決定何時收工，就果斷拔出 `while`！


````python
# [2] Code 範例區：同一任務的 for 與 while 對照
print("=== 任務：印出 1 到 4 的平方 ===")

# 方式 1：for 迴圈（已知次數，簡潔安全）
print("【for 迴圈寫法】")
for i in range(1, 5):
    print(f"{i} 的平方是 {i**2}")

# 方式 2：while 迴圈（手動控制起點、條件、步進）
print("\n【while 迴圈寫法】")
i = 1
while i <= 4:
    print(f"{i} 的平方是 {i**2}")
    i += 1
````

````python
# ==========================================
# [4] Code 練習題
# 任務說明：
# 【選型判斷實戰】
# 給定起始資金 money = 1000 元。
# 每天生活費開銷不固定：第 1 天花 80 元，第 2 天花 100 元，第 3 天花 120 元...
# 即第 day 天的生活費為 80 + (day - 1) * 20 元。
# 請使用 while 迴圈計算：這筆資金最長可以支撐幾天（max_days）？
# （當資金不足以支付當天生活費時即停止）
#
# 【公開測試資料 1】
# 設定：money = 1000, day = 1
# 第 1 天花 80 (剩 920)
# 第 2 天花 100 (剩 820)
# 第 3 天花 120 (剩 700)
# 第 4 天花 140 (剩 560)
# 第 5 天花 160 (剩 400)
# 第 6 天花 180 (剩 220)
# 第 7 天花 200 (剩 20)
# 第 8 天需 220 (不足，停止)
# 預期：可支撐 7 天
# ==========================================
# 請在下方撰寫你的程式碼並執行測試：

money = 1000
day = 1

while True:
    cost = 80 + (day - 1) * 20
    if money < cost:
        break
    money -= cost
    day += 1

max_days = day - 1
print(f"最多可支撐 {max_days} 天，最後剩下 {money} 元")

assert max_days == 7, "公開測資 1 驗證失敗"
print("公開測資 1 通過！")
````

### 🧅 6-3-4 APCS 核心基本功一：整數數位剝皮術（// 10 與 % 10 逐位拆解與加總）

**生活故事：剝洋蔥與拆解存錢筒**  
想像你抱著一顆洋蔥，你想看看洋蔥最裡面的核心，唯一的辦法就是一層一層由外向內把洋蔥皮剝掉；又或者你打破了存錢筒，想要把裡面的硬幣算清楚，你也是一顆一顆拿出來統計。  
當 APCS 題目給定一個很大的整數（例如 `38495`），需要你求出「所有位數的總和」（$3+8+4+9+5 = 29$）時，如果我們不用字串轉換，要如何純靠數學運算達成呢？

**運作機制——整除與取餘數的黃金組合：**  
在十進位制中，任何整數都可以被兩把神奇的手術刀精確拆解：
1. **手術刀一：`n % 10`（取出最右側的個位數）**  
   任何數除以 10 的餘數，剛好就是該數字當前最末尾的「個位數」！  
   例如：`38495 % 10` 的結果就是 `5`。
2. **手術刀二：`n = n // 10`（將整數向右截斷一位，剝掉個位數）**  
   整除 10 的商數，會直接無條件丟棄個位數，把剩下的高位數往右平移一位！  
   例如：`38495 // 10` 的結果就是 `3849`。

將這兩把刀放入 `while n > 0:` 的迴圈中，每一次迴圈就取下一個個位數並剝掉一層，直到整顆數字被削成 0 為止！這套俐落乾脆的技巧在演算法世界被尊稱為**「數位剝皮術（Digit Extraction）」**。

**常見錯誤陷阱：**  
- **邊界案例 $n = 0$**：若輸入數字恰好是 0，`while n > 0:` 會一次都不執行！若題目要求統計總位數，必須先特判 0 的情況。
- **誤用小數除法 `/`**：切記一定要使用雙斜線整除 `//`，若誤用單斜線 `/` 會產生浮點數精度災難。

**APCS 考試實務建議：**  
數位剝皮術是 APCS 實作題最經典的起手式之一（例如歷屆真題 c290「秘密差」、位數重組題）。熟練此架構，能在不耗費記憶體轉型字串的情況下，以純常數空間極速破解！


````python
# [2] Code 範例區：整數數位剝皮術全流程拆解
num = 38495
temp = num
digit_sum = 0
step = 1

print(f"=== 開始對整數 {num} 進行數位剝皮 ===")

while temp > 0:
    digit = temp % 10          # 刀法 1：取下當前最末位個位數
    digit_sum += digit         # 將個位數累加
    print(f"第 {step} 輪：取出個位數 [{digit}]，累加和變為 {digit_sum}")
    temp = temp // 10          # 刀法 2：將數字右移一位（剝掉末尾）
    print(f"         剩餘高位部分：{temp}")
    step += 1

print("=== 剝皮完成！ ===")
print(f"整數 {num} 的各位數字總和為：{digit_sum}")
````

### 🪞 6-3-5 APCS 核心基本功二：反轉整數與回文數檢驗（純數值無字串解法）

**生活故事：魔術倒車與迴文對稱**  
如果一個數字正著讀和倒著讀完全一樣，例如 `12321` 或 `1331`，我們就稱它為**「回文數（Palindrome Number）」**。  
很多初學者在解回文題時，第一反應是將數字轉成字串，然後用字串切片倒著排。然而在資訊科學高階考核或受限環境中，面試官與競賽評判往往會加上一句致命限制：**「請勿將數字轉為字串，請以純數值演算法在常數空間內完成！」**

**運作機制——左移十倍加上新個位數：**  
如何把一個整數完全倒過來？  
我們延續上一關的數位剝皮術，但這次多準備一個容器變數 `reversed_num = 0`：
1. 每當我們用 `n % 10` 拔下 `n` 的個位數 `digit` 時；
2. 我們先將原本的 `reversed_num` 乘以 10（等於將原本累積的數字整體「往左挪動一個十進位」）；
3. 然後把剛剛拔下來的 `digit` 填入那個空出來的個位數：  
   `reversed_num = reversed_num * 10 + digit`！
4. 最後同樣執行 `n //= 10`。

舉例來說，將 `123` 反轉：
- 第一輪：拔下 3，`rev = 0 * 10 + 3 = 3`
- 第二輪：拔下 2，`rev = 3 * 10 + 2 = 32`
- 第三輪：拔下 1，`rev = 32 * 10 + 1 = 321`！  
最後，只要檢查 `reversed_num == 原始數值`，就能立刻判斷出是否為回文數！

**常見錯誤陷阱：**  
- **破壞了原始數值**：在執行 `while n > 0` 的過程中，`n` 最後會變成 0。如果你沒有在一開始先用一個備份變數（如 `original_n = n`）存下來，最後你就無法拿反轉後的數字跟原始數字進行比對了！

**APCS 考試實務建議：**  
純數值反轉演算法的時間複雜度是 $O(\log_{10} n)$，空間複雜度是極致的 $O(1)$。這項演算法完全展現了你對進位制底層數學的透徹理解，是 APCS 解題必備的滿分技能！


````python
# [2] Code 範例區：純數值反轉整數演算法
num = 12345
original_num = num   # 關鍵防護：先備份原始數值
reversed_num = 0

print(f"=== 原始整數：{num} ===")

while num > 0:
    digit = num % 10
    reversed_num = reversed_num * 10 + digit
    num //= 10
    print(f"抽出 [{digit}] -> 當前反轉累積值：{reversed_num}")

print(f"\n反轉結果：{reversed_num}")
print(f"是否為對稱回文數？ {reversed_num == original_num}")
````

### 📉 6-3-6 對數級折半縮小演算法：二進位轉換與考拉茲猜想（3n + 1）

**生活故事：神秘的冰雹猜想（Collatz Conjecture）**  
在數學界有一個著名且迷人的未解之謎，稱為「考拉茲猜想」（又稱 3n + 1 猜想或冰雹猜想）：  
隨便挑選任何一個正整數 $n$：
- 如果 $n$ 是偶數，就把牠除以 2；
- 如果 $n$ 是奇數，就把牠乘上 3 再加 1。  
數學家猜測：無論一開始選的數字有多巨大，數字如同冰雹般在雲層中忽上忽下碰撞，**最終都必定會墜落收斂到 1**！但究竟需要跳躍幾次？事先完全無人知曉！

**運作機制——對數級折半的威力：**  
當一個演算法每經過一輪運算，就把問題的規模「折半（除以 2）」時，其數字縮小的速度會是指數級的反向——也就是**對數級（Logarithmic Scale，$\log_2 N$）**！  
例如，數字 100 萬（$10^6$）每次折半除以 2，只需要大約 20 次就會歸零！這正是未來我們在第十二章要學習的「二分搜尋法（Binary Search）」之所以神速的底層原理。

除了考拉茲猜想外，**「十進位轉二進位」**也是完全相同的折半架構：  
每次用 `n % 2` 得到當前的二進位位元（0 或 1），然後用 `n //= 2` 將數字折半，直到 `n == 0` 為止。

**常見錯誤陷阱：**  
- **整除 vs 小數除法**：在偶數除以 2 時，務必寫 `n = n // 2` 而非 `n = n / 2`！若寫成單斜線除法，數字會自動晉升為浮點數，當數字很大時會產生精度遺失，且無法再正確判定 `% 2 == 0`。

**APCS 考試實務建議：**  
考拉茲猜想與二進位拆解是 APCS 最愛考的經典條件迴圈題型。熟練 `while n > 1:` 與偶奇分支，能讓你秒殺此類動態收斂題！


````python
# [2] Code 範例區：考拉茲猜想（3n + 1）的冰雹軌跡
n = 26
step_count = 0

print(f"=== 探索考拉茲軌跡：起始數字 {n} ===")
print(f"起點：{n}", end="")

while n != 1:
    if n % 2 == 0:
        n = n // 2       # 偶數折半
    else:
        n = 3 * n + 1    # 奇數三倍加一
    print(f" -> {n}", end="")
    step_count += 1

print(f"\n=== 成功收斂到 1！總共歷經 {step_count} 步 ===")
````

````python
# ==========================================
# [4] Code 練習題
# 任務說明：
# 請計算輸入數字 n 經過考拉茲猜想變換過程中出現的「最高峰值（max_value）」。
#
# 【公開測試資料 1】
# 設定：n = 7
# 軌跡：7 -> 22 -> 11 -> 34 -> 17 -> 52 -> 26 -> 13 -> 40 -> 20 -> 10 -> 5 -> 16 -> 8 -> 4 -> 2 -> 1
# 預期峰值：max_val = 52
#
# 【公開測試資料 2】
# 設定：n = 10
# 軌跡：10 -> 5 -> 16 -> 8 -> 4 -> 2 -> 1
# 預期峰值：max_val = 16
# ==========================================
# 請在下方撰寫你的程式碼並執行測試：

n = 7
max_val = n

while n != 1:
    if n % 2 == 0:
        n //= 2
    else:
        n = 3 * n + 1
    if n > max_val:
        max_val = n

print(f"最高峰值為：{max_val}")

assert max_val == 52, "公開測資 1 驗證失敗"
print("公開測資 1 通過！")
````

### 🏛️ 6-3-7 古希臘智慧：歐幾里得輾轉相除法（求兩數最大公因數 GCD）

**生活故事：正方形地磚鋪設問題**  
想像你有一間長 48 公尺、寬 18 公尺的長方形大客廳，你想要買一批尺寸完全相同的正方形大理石地磚，把地板鋪得整整齊齊，而且「每一塊地磚都不能被切割裁減」。  
請問這塊正方形地磚的最大邊長是多少？這在數學上，就是要求 48 與 18 的**「最大公因數（Greatest Common Divisor，簡稱 GCD）」**！

**運作機制——千古神算法「輾轉相除法」：**  
西元前 300 年，古希臘數學家歐幾里得發現了一個天才般的幾何定理：  
**「兩數 $a$ 和 $b$ 的最大公因數，剛好等於 $b$ 和『$a$ 除以 $b$ 的餘數』的最大公因數！」**  
也就是說：$\gcd(a, b) = \gcd(b, a \% b)$！

這代表我們可以不斷用較小的數去整除較大的數求餘數，並將問題替換成更小的一組數，直到餘數變成 0 為止！此時除數就是我們要找的 GCD！  
在 Python 中，結合第一章學過的神奇語法「多變數同時賦值交換」，輾轉相除法只需要短短 2 行代碼：
```python
while b != 0:
    a, b = b, a % b
```
當迴圈結束跳出時，最後留下來的 `a` 就是最大公因數！

**常見錯誤陷阱：**  
- **擔心 $a$ 比 $b$ 小怎麼辦？**  
  初學者常常花大篇幅先寫 `if a < b:` 去交換兩者。但請看數學的奇妙之處：如果一開始 $a = 18, b = 48$，第一輪執行 `a % b` 得到 $18 \% 48 = 18$；賦值後變成 $a = 48, b = 18$！**直譯器在第一輪就自動幫你完成大小校正了**，完全不需要額外的 if 判斷！

**APCS 考試實務建議：**  
APCS 題庫中所有涉及分數約分、齒輪轉動齒數比、以及求最小公倍數（$LCM = a \times b // GCD$）的題目，全都可以靠這 2 行輾轉相除法秒殺！它是競賽選手必背的黃金公式。


````python
# [2] Code 範例區：歐幾里得輾轉相除法極速演繹
a = 48
b = 18
orig_a, orig_b = a, b

print(f"=== 求 GCD({a}, {b}) 的輾轉相除過程 ===")

round_idx = 1
while b != 0:
    rem = a % b
    print(f"第 {round_idx} 輪：{a} 除以 {b} 的餘數為 {rem} -> 更新為 a={b}, b={rem}")
    a, b = b, rem
    round_idx += 1

gcd_val = a
print(f"\n=== 計算完畢！ ===")
print(f"{orig_a} 與 {orig_b} 的最大公因數（GCD）為：{gcd_val}")

# 延伸應用：求最小公倍數（LCM）
lcm_val = (orig_a * orig_b) // gcd_val
print(f"兩數的最小公倍數（LCM）為：{lcm_val}")
````

````python
# ==========================================
# [4] Code 練習題
# 任務說明：
# 在 APCS 模擬題中，要求將分數 numerator / denominator 進行最簡約分。
# 請先使用輾轉相除法求出兩者的最大公因數 g，
# 接著將分子與分母同時整除 g，求出最簡分子 simple_num 與最簡分母 simple_den。
#
# 【公開測試資料 1】
# 設定：num = 36, den = 24
# GCD = 12，最簡分數為 3 / 2
#
# 【公開測試資料 2】
# 設定：num = 15, den = 28
# GCD = 1，最簡分數為 15 / 28
# ==========================================
# 請在下方撰寫你的程式碼並執行測試：

num = 36
den = 24

a, b = num, den
while b != 0:
    a, b = b, a % b

g = a
simple_num = num // g
simple_den = den // g

print(f"約分結果：{num}/{den} = {simple_num}/{simple_den}")

assert simple_num == 3 and simple_den == 2, "公開測資 1 驗證失敗"
print("公開測資 1 通過！")
````

### 🛑 6-3-8 無窮迴圈的觸發成因、終端機強制中斷（Ctrl+C）與防暴走自救指南

**生活故事：遊樂園的摩天輪煞車失靈**  
想像你坐上了遊樂園的刺激大轉輪，原本規定轉 3 圈就要停下來讓遊客下車。結果控制室的操作員不小心把煞車電掣拆掉了，摩天輪就這樣以超高速永無止境地旋轉下去！這時唯一能解救大家的，就是立刻按下控制台牆上的「緊急斷電大紅鈕」！

**運作機制——無窮迴圈的成因與緊急自救：**  
在撰寫 `while` 迴圈時，直譯器只會忠實地執行指令。如果判斷條件永遠無法變成 False，程式就會陷入**「無窮迴圈（Infinite Loop / Dead Loop）」**！  
常見的四大成因：
1. **忘記寫步進更新**（如漏寫 `i += 1`）；
2. **步進方向完全相反**（例如原本應該 `i += 1` 逼近 10，卻誤寫成 `i -= 1`，數值離 10 越來越遠）；
3. **終止條件設計自相矛盾**（例如 `while x > 0 and x < 0:`）；
4. **浮點數精度比對陷阱**（例如 `while x != 1.0:`，因為二進位小數精度誤差，`x` 可能永遠只會是 `0.9999999999` 而剛好跳過 `1.0`）。

**如何緊急自救？**
- 在終端機命令列（Terminal / CMD）中執行卡死時：立刻按下鍵盤組合鍵 **`Ctrl + C`**，強制拋出 `KeyboardInterrupt` 終斷程式！
- 在 Jupyter Notebook 或 Google Colab 中卡死時：點擊上方工具列的 **「中斷核心（Interrupt Kernel / 黑色正方形暫停鍵）」**！

**APCS 考試實務建議（防暴走安全閥門）：**  
在競賽中，若題目邏輯複雜，選手們常用**「計數安全閥（Safety Counter）」**：  
宣告一個安全計數器 `guard = 0`，每輪 `guard += 1`。若 `guard > 1000000`（超過一百萬次），就強制印出警示並跳出，避免在線上裁判系統中白白浪費時間慘遭「Time Limit Exceeded（TLE）」！


````python
# [2] Code 範例區：演示安全閥門（Safety Guard）如何阻斷潛在死結
# 模擬一個可能因為邏輯瑕疵而永無止境的迴圈
val = 1
safety_guard = 0
MAX_SAFE_STEPS = 10  # 設定最多只允許跑 10 步

print("=== 啟動帶有安全閥門的監控迴圈 ===")

while val > 0:
    # 安全閥門檢查：一旦超過安全上限立刻強制跳出
    safety_guard += 1
    if safety_guard > MAX_SAFE_STEPS:
        print(f"[SECURITY ALERT] 偵測到疑似無窮迴圈！在第 {safety_guard} 步強制煞車！")
        break
    
    print(f"第 {safety_guard} 輪安全執行中，val = {val}")
    # 假設工程師手滑，val 始終沒有被減到 0 以下

print(f"=== 程式安全脫身，避免系統當機！ ===")
````

````python
# ==========================================
# [4] Code 練習題
# 任務說明：
# 請為一個數值跳躍模擬加上「安全計數器」。
# 數值從 n = 1 出發，每輪進行 n = (n * 3) % 11 的變換。
# 若 n 等於 0 則正常停止；
# 同時設定 steps = 0，每輪 steps += 1，若 steps 達到 20 步仍未歸零，
# 則主動中斷跳出並將 terminated_by_guard 設為 True。
#
# 【公開測試資料 1】
# 設定初始：n = 1, steps = 0, terminated_by_guard = False
# 預期輸出：因在 11 步內循環，最後會被安全計數器攔截 (terminated_by_guard = True)
# ==========================================
# 請在下方撰寫你的程式碼並執行測試：

n = 1
steps = 0
terminated_by_guard = False

while n != 0:
    steps += 1
    if steps >= 20:
        terminated_by_guard = True
        break
    n = (n * 3) % 11

print(f"總共跑了 {steps} 步，是否被安全閥門攔截：{terminated_by_guard}")

assert terminated_by_guard is True and steps == 20, "公開測資 1 驗證失敗"
print("公開測資 1 通過！")
````

````python
# ==========================================
# [5] Code 挑戰題
# 任務說明：
# 防無窮迴圈的雙重條件審查：
# 請撰寫一個 while 迴圈，同時具備「實質業務條件」與「防暴走上限」：
# 初始 energy = 5，每輪 energy 增加 7。
# 迴圈繼續條件：(energy % 13 != 0) 且 (loop_count < 100)。
# 驗證此迴圈是否能在 100 輪之內找到滿足能被 13 整除的 energy 數值？
# ==========================================
# 請在下方撰寫你的程式碼：

energy = 5
loop_count = 0

while (energy % 13 != 0) and (loop_count < 100):
    energy += 7
    loop_count += 1

is_divisible = (energy % 13 == 0)
print(f"在第 {loop_count} 輪找到能量值 {energy}，是否能被 13 整除：{is_divisible}")
assert is_divisible is True and energy % 13 == 0, "挑戰題計算錯誤！"
print("挑戰題 6.3.8 驗證通過！")
````
