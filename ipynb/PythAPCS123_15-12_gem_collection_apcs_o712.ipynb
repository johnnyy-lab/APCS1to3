### 2. 完整題目描述（Problem Description）
小田設計了一個機器人來玩蒐集寶石的遊戲，遊戲的場地是一個 $M \times N$ 的二維陣列，每個元素 $(i, j)$ 對應在第 $i$ 橫列第 $j$ 直行的格子，其中橫列由上而下，直行由左往右，皆由 0 開始編號。某些格子是牆壁（值為 -1），機器人無法進入，其他的格子機器人可以進入，這些格子內有若干寶石，但某些格子的寶石數量可能是 0，場地四周視為被牆壁包圍，也就是機器人不能走出界外。

遊戲開始時，機器人從編號 $(r, c)$ 的格子開始，初始的方向為面向東方（右方），初始得分與擁有的寶石數量皆為 0。機器人會記錄目前的得分以及行進的方向。

機器人的行動規則如下，請計算出機器人在停止時所獲得的寶石數量，其中 $K$ 是一個輸入的正整數：
1. **如果走到的格子內無寶石（寶石數量等於 0），則停止遊戲**；否則
2. **將得分加上該格子目前的寶石數量，並取得一顆寶石。該格子的剩餘寶石數量就會減少一顆**；
3. **如果總得分是 $K$ 的倍數，則向右轉（順時針轉 90 度）**；
4. **若前方是牆壁或者前方是界外，則向右轉；持續本步驟直到前方的格子在場地範圍內且非牆壁**；
5. **進入前方的格子，並回到步驟 1 繼續**。

下圖為題目說明範例（場地為 $4 \times 5$ 二維陣列，起點在 $(2, 1)$，起始朝東，$K=4$）：

```text
           直行 (col)
        0     1     2     3     4
      ┌─────┬─────┬─────┬─────┬─────┐
   0  │  2  │  0  │  1  │ 1 ↱ │ 1 ↰ │
      ├─────┼─────┼─────┼───┼─┼─────┤
横 1  │  2  │ █-1 │  0  │ 2 ↑ │ █-1 │
列    ├─────┼─────┼─────┼───┼─┼─────┤
(row) │  0  │ 3 → │ 2 → │ 3 ↵ │  0  │  (起點 (2,1)，初始面向東)
   2  ├─────┼─────┼─────┼───┼─┼─────┤
   3  │ █-1 │  1  │ █-1 │ 3 ↰ │  0  │
      └─────┴─────┴─────┴─────┴─────┘
      註：█-1 代表障礙物（牆壁），箭頭表示機器人行進軌跡
```

機器人一開始會行經 $(2, 1) \to (2, 2) \to (2, 3)$，此時得分為 $3 + 2 + 3 = 8$，為 $K=4$ 的倍數，因此會右轉，方向改為朝南。接下來往南一步走到 $(3, 3)$，此時得分為 $8 + 3 = 11$，但往前是界外，因此會右轉朝西，此時前方 $(3, 2)$ 的位置為牆壁，所以繼續右轉朝北。

再前進時到達 $(2, 3)$，這個格子一開始有 3 顆寶石，但之前經過一次拿走一顆，所以只剩下 2 顆，故總得分為 $11 + 2 = 13$。接著會往北走兩次，總得分分別為 15 與 16，在 $(0, 3)$ 時因為總得分 16 為 4 的倍數，故右轉到 $(0, 4)$，然後會右轉兩次朝西，最後走到 $(0, 3)$ 時，該格子的寶石已被取走剩下 0 顆寶石，遊戲停止。總共獲得的寶石數量即是停止前所走的總步數加上起始位置的 1 顆，一共是 8 顆。

⚠️ **官方重要提示**：在原地右轉時並不算再次走到該格子。

## 15.12.1 題意解析與網格機器人動態系統心智模型（座標、朝向、累計得分與寶石庫存）

### 💡 觀念說明：動態遊戲模擬（Game Simulation）的心智模型
在 APCS 實作試題中，「模擬題」考驗的是學習者能否將題目規定的一連串遊戲規則，精確無誤地轉化為程式碼中的**狀態變數（State Variables）**與**狀態轉移（State Transitions）**。

我們可以將本題想像成「一台裝備了寶石探測器與方向舵的採礦車」，在一個 $M \times N$ 的矩陣迷宮中巡邏：
1. **空間位置座標 $(r, c)$**：目前採礦車正停在第幾列（$0 \le r < M$）與第幾行（$0 \le c < N$）。
2. **目前行進朝向 $d$**：車頭正面向東、南、西還是北。
3. **累計總得分 `total_score`**：從起點出發以來，所有採集到格子內的「初始數值」加總。這個分數是判斷是否觸發「順時針主動右轉」的關鍵。
4. **累計寶石總數 `gems`**：採礦車車廂內裝了多少顆寶石。每成功採集一次就加 1，這也是最後題目要求輸出的答案。
5. **動態地圖庫存 `grid[r][c]`**：迷宮不是靜態不變的！採礦車每開進去採集一次，該格子的寶石數量就會被扣減 1（即 `grid[r][c] -= 1`）。這意味著同一格如果未來被二次、三次經過，格子內的寶石會持續變少，直到變成 0。

```text
 ┌──────────────────────────────────────────────────────────────┐
 │                  機器人動態模擬系統心智模型                    │
 ├──────────────────────────────────────────────────────────────┤
 │  [空間狀態]  當前座標 (r, c)  +  當前朝向 d (東南西北)       │
 │  [得分狀態]  累計得分 total_score (用以觸發倍數右轉)          │
 │  [戰利品]    採集寶石總數 gems (本題唯一輸出)                │
 │  [場地狀態]  網格矩陣 grid[r][c] (具有原地扣減 1 的副作用)     │
 └──────────────────────────────────────────────────────────────┘
```

⚠️ **考場致命先決條件：停機條件檢驗時機**
遊戲的終止條件是：**「如果走到的格子內無寶石（即寶石數量等於 0），則停止遊戲」**。
這告訴我們：採礦車每踏入一個新格子（包括一開始的起點 $(r, c)$），**第一件要做的事情就是檢查該格寶石數是否為 0**！如果是 0，遊戲立即鳴笛收工，不採集、不加分、不轉向、不前進！

````python
# 15.12.1 範例展示：單一格子的狀態更新與副作用模擬
# 展示機器人踏入格子時，得分累加、寶石獲取與地圖原地扣減的運作

# 模擬一個 2x2 的迷宮片段
grid = [
    [3, 1],
    [2, 0]
]

r, c = 0, 0          # 目前機器人在 (0, 0)
total_score = 0      # 初始總得分為 0
gems = 0             # 初始獲得寶石數為 0

print(f"踏入前：座標=({r}, {c}), 格內寶石={grid[r][c]}, 得分={total_score}, 寶石總數={gems}")

# 執行規則 1 與規則 2：
if grid[r][c] == 0:
    print("格子內無寶石，停止遊戲！")
else:
    # 步驟 A: 將得分加上該格子目前的寶石數量
    total_score += grid[r][c]
    # 步驟 B: 取得一顆寶石
    gems += 1
    # 步驟 C: 該格子的剩餘寶石數量減少一顆
    grid[r][c] -= 1

print(f"踏入後：座標=({r}, {c}), 格內剩餘={grid[r][c]}, 得分={total_score}, 寶石總數={gems}")
````

#### 1. 順時針轉向順序與座標位移向量
本題規定機器人**「向右轉」意即順時針旋轉 90 度**，且初始方向面向**東方**。
我們將順時針的四個方向依序編號為 $0, 1, 2, 3$：

| 方向編號 $d$ | 中文方向 | 列位移 $dr$ | 行位移 $dc$ | 座標向量 $(dr, dc)$ |
| :---: | :---: | :---: | :---: | :---: |

## 15.12.3 核心行動規則一：停機檢驗（目前格子寶石數 $= 0$ 即停止）與得分加總/寶石扣減

### 💡 觀念說明：規則 1 與規則 2 的原子級操作順序
當機器人踏入一個格子 $(r, c)$ 時，題目給予的順序極為清晰，我們必須在程式碼中嚴格遵循此時序，不可顛倒：

#### 步驟 1：停機檢驗（Termination Check）
```python
if grid[r][c] == 0:
    break  # 停止遊戲！不再進行後續任何操作
```
* **為什麼要先檢查？**  
  題目明載：「如果走到的格子內無寶石，則停止遊戲；否則...」。如果當前格子原本就是 0 顆寶石，或者之前被採集過導致現在剩下 0 顆，機器人一踩上去就會立刻停機，**不能再獲得任何寶石與得分**！

#### 步驟 2：採集三部曲（Acquisition Triad）
若 `grid[r][c] > 0`，則依序執行：
1. **得分累加**：`total_score += grid[r][c]`  
   ⚠️ **注意**：加的是**採集前**該格子目前的寶石數量！例如格內有 3 顆寶石，則得分增加 3 分。
2. **寶石入庫**：`gems += 1`  
   ⚠️ **注意**：無論該格子原本有幾顆寶石，**每次踏入只能取走 1 顆寶石**！
3. **場地扣減**：`grid[r][c] -= 1`  
   該格子的剩餘寶石數量減 1。

#### ⚠️ 考場常見致命失分地雷
* **地雷 A（加分加錯）**：誤以為得分也是只加 1 分（寫成 `total_score += 1`）。這會徹底破壞後續規則 3 的倍數判定！
* **地雷 B（時序倒置）**：先執行了 `grid[r][c] -= 1`，才去累加得分 `total_score += grid[r][c]`。這樣每次加的分數都會憑空少算 1 分！

````python
# 15.12.3 範例展示：採集函式的標準實作與狀態輸出
def collect_gem(r, c, grid, total_score, gems):
    # 規則 1: 停機檢驗
    if grid[r][c] == 0:
        return False, total_score, gems  # False 代表應停止遊戲
    
    # 規則 2: 採集三部曲
    # 1. 將得分加上該格子目前的寶石數量
    total_score += grid[r][c]
    # 2. 取得一顆寶石
    gems += 1
    # 3. 該格剩餘寶石數量減少一顆
    grid[r][c] -= 1
    
    return True, total_score, gems

# 模擬一個格子 (1, 1) 原有 2 顆寶石
grid_demo = [[0, 0], [0, 2]]
score = 0
my_gems = 0

print(f"初始格內寶石: {grid_demo[1][1]}")

# 第一次經過 (1, 1)
ok, score, my_gems = collect_gem(1, 1, grid_demo, score, my_gems)
print(f"第 1 次採集 -> 成功: {ok}, 累計得分: {score}, 寶石總數: {my_gems}, 格內剩餘: {grid_demo[1][1]}")

# 第二次經過 (1, 1)
ok, score, my_gems = collect_gem(1, 1, grid_demo, score, my_gems)
print(f"第 2 次採集 -> 成功: {ok}, 累計得分: {score}, 寶石總數: {my_gems}, 格內剩餘: {grid_demo[1][1]}")

# 第三次經過 (1, 1)
ok, score, my_gems = collect_gem(1, 1, grid_demo, score, my_gems)
print(f"第 3 次採集 -> 成功: {ok}, 累計得分: {score}, 寶石總數: {my_gems}, 格內剩餘: {grid_demo[1][1]}")
````

## 15.12.5 核心行動規則三：前進探測與被動避障右轉（撞牆 -1 或界外持續轉向直到前方暢通）

### 💡 觀念說明：射線探測（Raycasting）與被動避障機制
在完成步驟 3 的「主動倍數轉向」後，機器人準備向前邁進一步。
但此時前方可能是**場地邊界之外（界外）**，或者是一堵**堅硬的石牆（值為 -1）**！

題目規則 4 明確規定：
> **「若前方是牆壁或者前方是界外，則向右轉；持續本步驟直到前方的格子在場地範圍內且非牆壁。」**

這是一個標準的**被動避障探測迴圈**。

#### 1. 前方格子安全檢驗條件
給定當前座標 $(r, c)$ 與朝向 $d$，預計前進的目標格子為：
$$nr = r + dirs[d][0], \quad nc = c + dirs[d][1]$$

一個格子被視為**安全可行走**，必須同時滿足三大條件：
1. 列座標未出界：$0 \le nr < M$
2. 行座標未出界：$0 \le nc < N$
3. 非障礙物牆壁：$grid[nr][nc] \neq -1$

用 Python 寫成條件式就是：
```python
is_safe = (0 <= nr < M and 0 <= nc < N and grid[nr][nc] != -1)
```

#### 2. 被動避障迴圈架構
只要前方不安全，就必須順時針向右轉 90 度，重新探測新的前方，**直到前方安全為止**：
```python
while True:
    nr = r + dirs[d][0]
    nc = c + dirs[d][1]
    # 檢查前方是否在界內且非牆壁
    if 0 <= nr < M and 0 <= nc < N and grid[nr][nc] != -1:
        break  # 前方暢通！結束避障迴圈
    # 否則持續右轉 90 度
    d = (d + 1) % 4
```

⚠️ **考場安全防呆原則**：
雖然題目保證「起始點不會四周均無路可走，機器人最終一定會停下來」，但若在迷宮中四周全被牆壁包圍（死胡同），上述迴圈會陷入無窮迴圈（Infinite Loop）。高分選手通常會加上 `turns < 4` 的安全計數器，一旦 4 個方向皆受阻就安全退出。

````python
# 15.12.5 範例展示：避障探測迴圈實作
M, N = 3, 3
grid = [
    [0,  1,  0],
    [2,  3, -1],  # (1, 2) 是牆壁
    [0, -1,  0]   # (2, 1) 是牆壁
]

dirs = [(0, 1), (1, 0), (0, -1), (-1, 0)]
dir_names = ["東", "南", "西", "北"]

r, c = 1, 1  # 機器人在中心 (1, 1)
d = 0        # 初始朝東

print(f"機器人位於 ({r}, {c})，初始朝向: {dir_names[d]}")

# 執行避障探測迴圈：
turns = 0
while turns < 4:
    nr = r + dirs[d][0]
    nc = c + dirs[d][1]
    
    # 檢查是否在界內且不是牆
    if 0 <= nr < M and 0 <= nc < N and grid[nr][nc] != -1:
        print(f"-> 探測成功！朝向 {dir_names[d]} 前方為 ({nr}, {nc})，值為 {grid[nr][nc]}，可通行！")
        break
    else:
        obstacle_type = "出界" if not (0 <= nr < M and 0 <= nc < N) else "撞牆(-1)"
        print(f"-> 朝向 {dir_names[d]} 前方 ({nr}, {nc}) {obstacle_type}，觸發避障右轉！")
        d = (d + 1) % 4
        turns += 1
````

## 15.12.6 考場致命細節辨析：原地轉向不重複計算步數與寶石、避免死迴圈防護

### 💡 觀念說明：考場兩大致命失分盲區辨析
在 APCS 考場上，許多同學明明實作出了狀態機，但卻在特定測資拿到 WA 或 TLE，主要原因就在於以下兩大細節：

#### 1. 致命盲區一：「原地轉向」絕對不等於「再次踏入」！
官方題本中以特大字體特別標註：
> **⚠️「請注意，在原地右轉時並不算再次走到該格子。」**

* **錯誤思維**：有些同學將「每次迴圈」與「一次採集」綁定在一起。當機器人發現前方有牆而原地轉向時，迴圈重新跑了一輪，程式碼竟然又把當前格子的分數加了一次、寶石多拿了一顆、格子寶石又扣了一顆！
* **正確認知**：
  * **原地轉向只是改變朝向變數 $d$ 的數值**，機器人的雙腳依然站在原本的座標 $(r, c)$，地圖數值與累計寶石完全不動！
  * 只有在步驟 5**「真正跨出步伐進入前方格子」**時，才會在下一個回合執行步驟 1 與步驟 2 的採集運算！

#### 2. 致命盲區二：兩大轉向機制的連續疊加
題目中存在兩種完全獨立的轉向動機：
1. **主動轉向（規則 3）**：因為累計總得分是 $K$ 的倍數而觸發。
2. **被動避障轉向（規則 4）**：因為前方是牆壁或出界而觸發。

這兩種轉向是**可能在同一步驟中連續觸發的**！
* 例如範例一：在 $M=1$ 的通道中，原本朝東前進，得分剛好成為 $K$ 的倍數，因此觸發「主動轉向」變成朝**南**。
* 但轉成朝南後，前方是下邊界（界外），因此立刻觸發「被動避障轉向」變成朝**西**！
* 這等於在同一個格子上連續轉了兩次 90 度（直接調頭 180 度）！
如果同學把這兩種轉向混在同一條 `if-elif` 裡面，就會漏轉一次 90 度，導致大批測資全數潰敗！

````python
# 15.12.6 範例展示：主動轉向與被動避障連續疊加的標準流程
dirs = [(0, 1), (1, 0), (0, -1), (-1, 0)]
dir_names = ["東", "南", "西", "北"]

M, N = 1, 5
grid = [[1, 2, 3, 2, 0]]  # 單列狹長地圖
K = 3

r, c = 0, 1  # 機器人在 (0, 1)
d = 0        # 朝東
total_score = 1  # 原有得分 1
gems = 1

print(f"--- 踏入 (0, 1) 前：朝向={dir_names[d]}, 得分={total_score}, 寶石={gems} ---")

# 規則 2: 採集
total_score += grid[r][c]  # 加 2 分 -> 變成 3 分
gems += 1
grid[r][c] -= 1
print(f"採集後：得分={total_score}, 寶石={gems}, 格內剩餘={grid[r][c]}")

# 規則 3: 主動倍數轉向
if total_score % K == 0:
    d = (d + 1) % 4
    print(f"-> 觸發規則 3 (主動倍數轉向)：轉向 90 度為 {dir_names[d]}")

# 規則 4: 被動避障轉向
while True:
    nr = r + dirs[d][0]
    nc = c + dirs[d][1]
    if 0 <= nr < M and 0 <= nc < N and grid[nr][nc] != -1:
        print(f"-> 前方暢通！確認前進方向為 {dir_names[d]}，目標格 ({nr}, {nc})")
        break
    print(f"-> 前方 ({nr}, {nc}) 不通，觸發規則 4 (避障轉向)：右轉 90 度")
    d = (d + 1) % 4
````

## 15.12.7 完整狀態機主迴圈 AC 實作：分步模擬、總步數計算與最終寶石總數統計

#### 滿分主迴圈執行藍圖
```text
                     開始遊戲 (起點 r, c, 朝向 d=0 東)
                                  │
                   ┌──────────────┴──────────────┐
                   ▼                             │
    [規則 1] grid[r][c] == 0 ? ───(是)───► 停止遊戲，輸出 gems
                   │ (否)                        ▲
                   ▼                             │
    [規則 2] 得分加總: total_score += grid[r][c]  │
             寶石入手: gems += 1                  │

    [規則 3] total_score % K == 0 ?              │
```

#### 官方範例二完整手動追蹤驗證
* 起點 $(2, 1)$，面向東，$K=4$。
  1. 踏入 $(2, 1)$：格值 3，得分 3，寶石 1，格值變 2。朝向東，前方 $(2, 2)$ 安全。
  2. 踏入 $(2, 2)$：格值 2，得分 5，寶石 2，格值變 1。朝向東，前方 $(2, 3)$ 安全。
  3. 踏入 $(2, 3)$：格值 3，得分 8（為 4 倍數！轉南），寶石 3，格值變 2。朝向南，前方 $(3, 3)$ 安全。
  4. 踏入 $(3, 3)$：格值 3，得分 11，寶石 4，格值變 2。朝南出界轉西、$(3, 2)$ 牆壁轉北。朝向北，前方 $(2, 3)$ 安全。
  5. 踏入 $(2, 3)$：格值剩 2，得分 13，寶石 5，格值變 1。朝向北，前方 $(1, 3)$ 安全。
  6. 踏入 $(1, 3)$：格值 2，得分 15，寶石 6，格值變 1。朝向北，前方 $(0, 3)$ 安全。
  7. 踏入 $(0, 3)$：格值 1，得分 16（為 4 倍數！轉東），寶石 7，格值變 0。朝向東，前方 $(0, 4)$ 安全。
  8. 踏入 $(0, 4)$：格值 1，得分 17，寶石 8，格值變 0。朝東出界轉南出界轉西。朝向西，前方 $(0, 3)$ 安全。
  9. 踏入 $(0, 3)$：格值為 0！觸發規則 1 停機！
* 最終累計獲得寶石數：**8 顆**！與官方標準答案 100% 吻合！

````python
# 15.12.7 範例展示：完整解題函式與官方範例二動態軌跡輸出
def solve_gem_collector(M, N, K, r, c, grid, verbose=False):
    dirs = [(0, 1), (1, 0), (0, -1), (-1, 0)]
    dir_names = ["東", "南", "西", "北"]
    d = 0  # 初始朝東
    total_score = 0
    gems = 0
    step = 0
    
    while True:
        step += 1
        # 規則 1: 停機檢驗
        if grid[r][c] == 0:
            if verbose:
                print(f"步驟 {step}: 踏入 ({r}, {c})，格內寶石為 0 -> 停止遊戲！")
            break
            
        # 規則 2: 採集三部曲
        val = grid[r][c]
        total_score += val
        gems += 1
        grid[r][c] -= 1
        
        # 規則 3: 主動倍數右轉
        r3_turned = False
        if total_score % K == 0:
            d = (d + 1) % 4
            r3_turned = True
            
        # 規則 4: 被動避障連續右轉
        r4_turns = 0
        while r4_turns < 4:
            nr = r + dirs[d][0]
            nc = c + dirs[d][1]
            if 0 <= nr < M and 0 <= nc < N and grid[nr][nc] != -1:
                break
            d = (d + 1) % 4
            r4_turns += 1
            
        if verbose:
            print(f"步驟 {step}: 在 ({r}, {c}) 採集 (格值 {val}->{grid[r][c]}), 得分={total_score}, 寶石={gems}, 倍數右轉={r3_turned}, 避障右轉={r4_turns}次, 下一步朝 {dir_names[d]} 往 ({nr}, {nc})")
            
        # 規則 5: 進入前方格子
        r, c = nr, nc
        
    return gems

# 測試官方範例二：
demo_grid = [
    [2, 0, 1, 1, 1],
    [2, -1, 0, 2, -1],
    [0, 3, 2, 3, 0],
    [-1, 1, -1, 3, 0]
]
ans = solve_gem_collector(4, 5, 4, 2, 1, demo_grid, verbose=True)
print(f"\n官方範例二最終結果: 蒐集到 {ans} 顆寶石 (預期: 8)")
````

## 15.12.8 子題組特判分析（$M=1$ 狹長網格連續右轉兩次即迴轉）、極端邊界測試與常見 WA 排查

### 💡 觀念說明：子題組設計心機與演算法複雜度證明

#### 1. 深入剖析子題組一（60 分）：$M = 1$ 的幾何特性
在官方題本的評分說明中，明確提示：
> **「第 1 子題組 60 分：$M = 1$。請留意 $M = 1$ 時，如遇到右轉必然會連續轉兩次，也就是說右轉即是迴轉。」**

* **為什麼右轉等於迴轉？**
  * 當場地只有橫列第 0 列時（$M=1$），上下方完全沒有空間，任何向南（$dr=+1$）或向北（$dr=-1$）的移動都會**立刻出界**！
  * 當原本朝東前進時，若觸發右轉變朝南，南邊出界立即再右轉變朝西！
  * 同理，原本朝西前進時，若觸發右轉變朝北，北邊出界立即再右轉變朝東！
  * **結論**：只要轉向，就相當於 180 度掉頭（迴轉）！
* **重要啟發**：我們在 15.12.5 所設計的**「通用被動避障探測迴圈」**，完全不需要針對 $M=1$ 寫任何特殊的 `if M == 1:` 特判分支！因為 `while` 迴圈在探測到南向出界時，自動會再 `d = (d + 1) % 4` 轉向西，**全自動且天然地完成 180 度迴轉**！

#### 2. 時間與空間複雜度嚴格證明
* **空間複雜度**：$O(M \times N)$，最多 $100 \times 100 = 10,000$ 個整數，佔用記憶體小於 1 MB，遠低於 256 MB 限制。
* **時間複雜度**：
  * 格子內寶石數量介於 $0$ 到 $K-1$ 之間，而 $K \le 20$。
  * 整個地圖上的寶石總數至多為 $M \times N \times (K-1) \le 100 \times 100 \times 19 = 1.9 \times 10^5$ 顆。
  * 機器人每跨出一步踏入有效格子，**必定會消耗 1 顆寶石**！
  * 由於寶石數量只減不增，且總量有限，因此整個模擬的主迴圈最多只會執行 $1.9 \times 10^5$ 次！
  * 在 Python 中執行 $1.9 \times 10^5$ 次簡單加減運算僅需約 **0.03 秒**，保證穩穩通過 1.0 秒的時間限制，絕不可能發生 TLE！

#### 3. 考場常見 WA 地雷總清單
1. **起點寶石為 0**：如果起點格子寶石就是 0，答案應直接輸出 `0`。若沒在進入時先檢查就直接加分，會算出錯誤的 1。
2. **原地轉向重複採集**：每次避障右轉時誤跑採集邏輯，導致格子寶石被重複扣減。
3. **輸入讀取未處理空白換行**：二維陣列讀取時若遇多餘空行未安全過濾，會拋出 `IndexError`。

### 📝 版本一：APCS 官方實作考場專用版 —— 淺顯易懂一般版（新手友善推薦）

* **適用情境**：APCS 正式考試現場（題目保證單筆測資輸入）。
* **教學與設計理念**：
  1. **平鋪直敘、步驟展開**：使用標準的 `input().split()` 分別讀取各行參數與網格矩陣，變數命名如白話文（`dirs`, `total_score`, `gems`）。
  2. **完全對齊題目 1~5 步驟**：主迴圈中的 `if` 與 `while` 嚴格按照題目敘述的五大規則循序編寫，即使在考場緊張氛圍下，也能隨時隨地與題本逐條核對，除錯零障礙！
  3. **穩健安全 100% 滿分**：無任何冷門黑魔法，保證 100% AC 通過考場所有測資。

````python
# ==============================================================================
# 📝 版本一：APCS 官方實作考場專用版 —— 淺顯易懂一般版（新手友善推薦）
# 適用情境：APCS 正式考場單筆輸入保證，平鋪直敘、拆解詳細、穩健滿分
# ==============================================================================

def solve_apcs_version1():
    # 步驟 1：讀取首行參數 M, N, K, r, c
    line1 = input().split()
    if not line1:
        return
    M = int(line1[0])
    N = int(line1[1])
    K = int(line1[2])
    r = int(line1[3])
    c = int(line1[4])
    
    # 步驟 2：逐行讀取 M 列網格內容
    grid = []
    for _ in range(M):
        row = [int(x) for x in input().split()]
        grid.append(row)
        
    # 步驟 3：定義順時針方向向量：東(0), 南(1), 西(2), 北(3)
    dirs = [(0, 1), (1, 0), (0, -1), (-1, 0)]
    d = 0  # 初始朝向面向東方
    
    total_score = 0  # 累計總得分
    gems = 0         # 蒐集到的寶石總數
    
    # 步驟 4：啟動狀態機主迴圈
    while True:
        # 規則 1：如果走到的格子內無寶石，則停止遊戲
        if grid[r][c] == 0:
            break
            
        # 規則 2：得分加上目前格內寶石數，取得一顆寶石，格內寶石減一
        total_score += grid[r][c]
        gems += 1
        grid[r][c] -= 1
        
        # 規則 3：如果總得分是 K 的倍數，向右轉 90 度
        if total_score % K == 0:
            d = (d + 1) % 4
            
        # 規則 4：若前方是牆壁(-1)或界外，持續向右轉直到前方暢通
        while True:
            nr = r + dirs[d][0]
            nc = c + dirs[d][1]
            # 檢查目標格是否在界內且非牆壁
            if 0 <= nr < M and 0 <= nc < N and grid[nr][nc] != -1:
                break
            d = (d + 1) % 4
            
        # 規則 5：進入前方的格子，繼續下一輪
        r = nr
        c = nc
        
    # 步驟 5：輸出最後蒐集到的寶石總數
    print(gems)

# 在本機或考場執行時取消下行註解即可：
# solve_apcs_version1()
````

### ⚡ 版本二：APCS 官方實作考場專用版 —— 極簡高效精煉版（進階高手推薦）

* **適用情境**：APCS 正式考試現場（追求極速敲碼、代碼洗鍊的競賽高手）。
* **教學與設計理念**：
  1. **串流解析器（Stream Tokenizer）**：使用 `iter(sys.stdin.read().split())` 一次性將所有輸入轉為迭代器，自動免疫多餘空白、換行與格式雜訊，讀取僅需 3 行。
  2. **高階時鐘向量與簡潔避障**：利用 Python 簡潔的元組解包與短路邏輯，以不到 20 行的洗鍊架構完成全套狀態機模擬，展現高階極致效能！

````python
# ==============================================================================
# ⚡ 版本二：APCS 官方實作考場專用版 —— 極簡高效精煉版（進階高手推薦）
# 適用情境：APCS 正式考場，展現 Python 極短程式碼之高階表現力與極致效能
# ==============================================================================

import sys

def solve_apcs_version2():
    it = iter(sys.stdin.read().split())
    try:
        M, N, K, r, c = [int(next(it)) for _ in range(5)]
    except StopIteration:
        return
    grid = [[int(next(it)) for _ in range(N)] for _ in range(M)]
    
    dirs = [(0, 1), (1, 0), (0, -1), (-1, 0)]
    d = score = gems = 0
    
    while grid[r][c] > 0:
        score += grid[r][c]
        gems += 1
        grid[r][c] -= 1
        if score % K == 0:
            d = (d + 1) % 4
        while not (0 <= r + dirs[d][0] < M and 0 <= c + dirs[d][1] < N and grid[r + dirs[d][0]][c + dirs[d][1]] != -1):
            d = (d + 1) % 4
        r += dirs[d][0]
        c += dirs[d][1]
        
    print(gems)

# 在本機或考場執行時取消下行註解即可：
# solve_apcs_version2()
````
