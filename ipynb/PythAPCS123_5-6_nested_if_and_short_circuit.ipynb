### 🪆 5-6-1 巢狀分支結構——多層次縮排與條件俄羅斯套娃

**生活比喻：機場貴賓室的雙重驗證閘門**  
當旅客抵達機場：
1. **外層大門**：第一關先檢驗是否有國際線機票。若無機票，連機場管制區都進不去（外層 `else` 直接離開）。
2. **內層貴賓室小門**：通過第一關進入候機大廳後，旅客走到貴賓室門口，服務人員再次檢驗是否持有「黑卡貴賓證（內層 `if`）」。如果是黑卡會員，引導進入享受香檳點心；否則（內層 `else`），請在一般候機座椅候機。

這種「**在分支裡面，又包了一層完整分支**」的架構，在程式設計中稱為**巢狀結構（Nested Structure）**！

**語法縮排與層次幾何學：**
```python
if 外層條件:
    外層區塊指令
    if 內層條件:
        內層區塊指令（縮排 8 個空格！）
    else:
        內層 else 指令（縮排 8 個空格！）
else:
    外層 else 指令
```
1. **縮排的層次累加**：
   外層 `if` 縮排 4 個空格；進入內層 `if` 後，裡面的指令必須**再縮排 4 個空格（共 8 個空格）**！
2. **else 的就近對齊原則**：
   `else` 永遠與它**同一縮排層級**且最近的 `if` 結為一對！縮排對齊決定了誰跟誰是一組，差一個空格就會引發嚴重的邏輯偏離。

**教材經典範例（吳邦一老師講義第 21 頁）：**
```python
if x > 5:
    print(x)
    if y > 3:
        print(y)
    else:
        print(y + 8)
else:
    print(x + 3)
```
- 若 $x = 7, y = 5$：$x > 5$ 成立，印出 7；接著 $y > 3$ 成立，印出 5。
- 若 $x = 7, y = 2$：$x > 5$ 成立，印出 7；接著 $y > 3$ 不成立，印出 $2 + 8 = 10$。
- 若 $x = 4$：$x > 5$ 不成立，直接跳到最外層 else，印出 $4 + 3 = 7$！

**APCS 考試實務建議：**
在題目有「大類分類，細類再分級」時，巢狀結構能清晰地分流資料。但請注意巢狀層級盡量不要超過 3 層，否則過深的縮排會讓代碼難以維護。


````python
# [2] Code 範例區：展示吳邦一老師講義第 21 頁巢狀 if 範例

# 測試組 1：x = 7, y = 2
x = 7
y = 2
print("--- 測試情境 1：x = 7, y = 2 ---")
if x > 5:
    print("外層命中，x =", x)
    if y > 3:
        print("內層命中，y =", y)
    else:
        print("內層 else 命中，y + 8 =", y + 8)  # 2 + 8 = 10
else:
    print("外層 else 命中，x + 3 =", x + 3)

# 測試組 2：x = 4, y = 9
x = 4
y = 9
print("--- 測試情境 2：x = 4, y = 9 ---")
if x > 5:
    print("外層命中，x =", x)
    if y > 3:
        print("內層命中，y =", y)
    else:
        print("內層 else 命中，y + 8 =", y + 8)
else:
    print("外層 else 命中，x + 3 =", x + 3)  # 4 + 3 = 7 (內層直接跳過！)
````

### 🔄 5-6-2 巢狀 if 與 and 複合條件——雙向轉換與可讀性權衡

**生活比喻：雙道旋轉門 vs 同時刷卡閘門**  
- **巢狀結構**：你走過第一道旋轉門進入小走廊，接著再刷卡推開第二道旋轉門。
- **單一 `and` 複合條件**：在一道大門前，要求你左手按指紋、右手刷門禁卡，兩者同時滿足大門才打開。

在邏輯本質上，**「兩層嵌套的單向 `if`」與「單一 `if 條件A and 條件B`」是百分之百等價的**！

**等價代碼對照：**
```python
# 形式一：巢狀 if
if A:
    if B:
        做事情()

# 形式二：扁平化 and
if A and B:
    做事情()
```

**如何選擇？工程實務的權衡天平：**
1. **何時該扁平化為 `A and B`（推薦）**：
   若兩個條件都沒有各自獨立的 `else` 備用處置，只是單純要「兩者同時滿足才動作」，**強烈建議使用 `A and B`**！這樣可以減少不必要的縮排，讓代碼更扁平、乾淨、優雅（遵守 Python 禪理：*Flat is better than nested*）。
2. **何時必須保留巢狀 `if`**：
   當 $A$ 失敗與 $B$ 失敗需要給予**完全不同的錯誤提示**（例如提示「帳號不存在」vs「密碼錯誤」）時，就必須依靠巢狀 `if-else` 的各自分支！

**APCS 考試實務建議：**
在解演算法題時，若只要在兩條件均滿足時執行更新，扁平的 `and` 可以大幅降低縮排錯誤（IndentationError）的機率。


````python
# [2] Code 範例區：展示巢狀 if 與 and 複合條件的雙向等價轉換

score = 85
attendance = 95

# 寫法一：巢狀 if 結構
print("--- 寫法一：巢狀 if ---")
if score >= 80:
    if attendance >= 90:
        print("巢狀判定：成績達標且出席優良，頒發榮譽證書！")

# 寫法二：扁平化 and 結構（更精簡現代！）
print("--- 寫法二：扁平化 and ---")
if (score >= 80) and (attendance >= 90):
    print("扁平判定：成績達標且出席優良，頒發榮譽證書！")
````

### 🛡️ 5-6-3 and 短路求值機制——除數防零檢查（b != 0 and a / b > 1）

**生活比喻：保全巡邏的雷區警戒哨**  
想像探險隊前方有一片雷區，隊伍第一守則是：「*先確認前方有沒有地雷（條件一），如果沒有地雷，才踩上去測量深度（條件二）！*」  
如果第一眼就看到地雷赫然在目（條件一為 `False`），隊長會立刻吹哨：「*危險！立刻停止前進！*」後面的所有人絕不會去踩地雷，隊伍全體安全！

這就是電腦科學中極其重要的**短路求值（Short-circuit Evaluation）**機制！

**運作機制與底層原理：**
在執行 `A and B` 時，電腦是**由左至右**逐一評估：
1. 先計算左側條件 $A$。
2. **關鍵短路點**：如果 $A$ 的結果是 **`False`**，依據 `and` 的真值表，無論後面 $B$ 是真是假，整個式子注定已經是 `False` 了！
3. 因此，**Python 直譯器會立刻停止運算，直接回傳 `False`，完全略過右側條件 $B$ 的執行！**

**教材原著範例（吳邦一老師講義第 21 頁）：**
在數學運算中，若分母為零（例如 `a / 0`），電腦會立即崩潰並引發致命錯誤：`ZeroDivisionError: division by zero`！  
但如果我們這樣寫：
```python
b != 0 and a / b > 1
```
讓我們來看當 $b = 0$ 時會發生什麼奇蹟：
1. 電腦先看左邊：`b != 0` ➔ `0 != 0` 為 **`False`**！
2. 電腦觸發 **`and` 短路機制**：既然左邊是 `False`，整體必為 `False`，後面的 `a / b > 1` **直接被跳過，根本不會被執行！**
3. 結果：程式**完全沒有報錯，平平安安地回傳了 `False`**！

**APCS 考試實務建議：**
「除數防零檢查」與後續章節的「索引防出界檢查（`i < len(a) and a[i] == target`）」是 APCS 中最核心的防呆防爆防線！務必將安全守門條件放在 `and` 的**最左側**！


````python
# [2] Code 範例區：展示 and 短路求值與除數防零保護（吳邦一老師講義第 21 頁）

a = 10
b = 0

# 1. 🚨 如果單獨執行 a / b 會怎樣？
# print(a / b > 1)  # 這行若解除註解會立刻拋出 ZeroDivisionError 導致程式崩潰！

# 2. ✅ 利用 and 短路求值進行安全防禦
# 當 b == 0 時，左側 (b != 0) 為 False，右側 (a / b > 1) 直接被短路跳過，絕不執行！
is_valid_ratio = (b != 0) and (a / b > 1)
print("a =", a, ", b =", b)
print("除數防零安全保護：(b != 0 and a / b > 1) 結果：", is_valid_ratio)  # False (安全無崩潰！)

# 3. 當 b != 0 時正常求值
b_valid = 2
is_valid_ratio2 = (b_valid != 0) and (a / b_valid > 1)  # 10 / 2 = 5 > 1 -> True
print("當 b = 2 時：", is_valid_ratio2)  # True
````

**生活比喻：保全快速通道與備用鑰匙**  
想像一棟大樓有兩道身份驗證：
- 「持有一級VIP黑卡（條件一）」或者「去櫃台輸入訪客登記碼（條件二）」。
- 當一位貴賓大步走來，門禁感應器瞬間感應到他胸前的黑卡（條件一為 `True`），閘門立刻在 0.01 秒內打開！保全人員絕不會再叫他停下來去填寫訪客登記碼（條件二被跳過！）。

這就是 **`or` 運算子的短路求值（Or Short-circuit）**！

**運作機制與底層原理：**
在執行 `A or B` 時：
1. 電腦先計算左側條件 $A$。
2. **關鍵短路點**：如果 $A$ 的結果是 **`True`**，依據 `or` 的真值表，只要有一邊為真，整體就注定是 `True` 了！
3. 因此，**Python 直譯器會立刻停止運算，直接回傳 `True`，完全不再理會右側條件 $B$！**

**危險防禦示範：**
```python
# 假設 x 等於 0
result = (x == 0) or (100 / x > 2)
```
1. 電腦先看左側：`x == 0` 為 **`True`**！
2. 觸發 `or` 短路求值：既然左側已成立，整體必為 `True`，右側可能引發除以零崩潰的 `100 / x > 2` **直接被略過！**
3. 程式安全運行，得到 `True`！

**常見錯誤陷阱：**
順序千萬不可放反！如果寫成 `(100 / x > 2) or (x == 0)`，電腦會先去執行左邊的除以零，在短路機制生效之前就直接引發崩潰報錯！

**APCS 考試實務建議：**
將「最可能成立的條件」或「最容易計算且能兜底的特例」放在 `or` 的左邊，不僅能避免危險運算，還能加速程式執行時間！


````python
# [2] Code 範例區：展示 or 短路求值防禦

x = 0

# 1. 利用 or 短路保護可能發生除以零的危險運算
# 左側 (x == 0) 為 True，右側 (10 / x > 1) 瞬間被短路略過，平安無事！
safe_eval = (x == 0) or (10 / x > 1)
print("x =", x)
print("or 短路防禦：(x == 0 or 10 / x > 1) 結果：", safe_eval)  # True

# 2. 實務情境：管理員免檢通行
is_admin = True
# 假設驗證普通員工資格需要複雜或危險的檢查，管理員可直接短路過關
can_enter = is_admin or (x != 0 and 100 // x > 5)
print("管理員身分直接短路過關：", can_enter)  # True
````

### 🛡️ 5-6-5 APCS 邊界條件防崩潰雙重防線——先合法驗證，再執行運算

**生活比喻：出海航行前的氣象與油量安檢**  
一艘遠洋漁船準備出航前，船長一定會先看兩張檢查表：
1. 第一防線（合法性邊界）：今天外海有沒有海嘯警報？油箱是否有加滿？
2. 第二防線（實質作業）：若第一防線安全無虞，才啟動引擎、出海撒網作業！
如果海嘯警報大響，漁船連港口都不會出，根本不需要去承受風浪翻船的危險！

**APCS 競賽的生命線：保護程式絕不 RE（Runtime Error）**  
在線上自動裁判系統中，選手最害怕的評判結果之一就是 **`RE`（執行時期錯誤）**。  
通常造成 RE 的頭號元兇就是：
- 拿了無效的數字去當除數或模除（ZeroDivisionError）
- 在非法邊界內嘗試計算根號負數（ValueError）

**雙重防線標準架構：**
```python
if 合法性檢驗門檻:
    執行高風險實質運算
else:
    輸出錯誤代碼或備援安全數值
```
或者一行化短路防線：
```python
合法邊界 and 高風險條件式
```

**APCS 實戰防禦心法：**
永遠在腦海中問自己：「*這行算式有沒有可能除以零？有沒有可能超出題目的有效範圍？*」  
只要將邊界條件作為第一道防線，你的程式碼就能在面對黑心測資與邊界特例時屹立不搖！


````python
# [2] Code 範例區：展示 APCS 邊界條件雙重防線設計

# 模擬折扣除法計算：避免總件數為 0 引發平均計算崩潰
total_cost = 500
item_count = 0

print("--- 雙重防線防禦展示 ---")
# 第一防線：先檢查 item_count > 0
if item_count > 0:
    unit_cost = total_cost // item_count
    print("每件平均成本：", unit_cost)
else:
    print("警報：總件數為零，無法計算平均，安全避開除以零崩潰！")

# 短路雙重防線一行化展示
is_cheap = (item_count > 0) and (total_cost // item_count < 100)
print("是否件數合法且平均單價低於 100：", is_cheap)  # False
````
