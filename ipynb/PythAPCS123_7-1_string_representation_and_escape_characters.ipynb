### 🔤 7-1-1 單引號 ' ' 與雙引號 " " 的表示法與成對配對原則

**生活比喻：字串項鍊與頭尾相扣的珠寶盒**  
想像你手中有許多美麗的珍珠，每顆珍珠上都刻著一個字母或符號。如果把珍珠散落在桌上，風一吹就會滾落得無影無蹤；只有拿一條細線將它們串在一起，並在項鍊的最左端與最右端扣上相匹配的珠寶扣環，它們才會成為一件完整的首飾。  
在 Python 中，字串就像這條項鍊，而扣環就是「引號」！只要被一對引號包裹起來的任何內容（不管是英文字母、中文漢字、數字還是標點符號），在 Python 眼中都是純粹的文字。

**底層運作機制與成對配對規則**  
Python 提供了兩種最常用的引號標記：
1. **單引號**：`'Hello, APCS!'`
2. **雙引號**：`"Hello, APCS!"`

在 Python 直譯器底層，使用單引號和使用雙引號所建立出來的字串物件是**完全一模一樣、沒有任何高低之分**的（使用 `type()` 檢查兩者，型態皆為 `<class 'str'>`）。  
但直譯器有一條不可違背的鐵律——**「引號必須完全同型且成對配對」**！
- 如果你以單引號開頭，就必須以單引號結尾（如 `'Python'`）。
- 如果你以雙引號開頭，就必須以雙引號結尾（如 `"Python"`）。

**常見錯誤陷阱：混用引號導致的未閉合字串錯誤**  
初學者最常犯的粗心失誤之一，就是開頭打了單引號，結尾卻順手打了雙引號，例如：  
`greeting = 'Hello"`  
當 Python 直譯器讀到單引號時，會一路往右尋找下一個單引號來閉合字串。當它看到雙引號時，會把它當成一般文字繼續往下找，直到這行程式碼結束都找不到單引號，就會立刻拋出嚴厲的語法錯誤：  
**`SyntaxError: unterminated string literal`（未閉合的字串常數！）**  
看到這個報錯時，第一時間檢查字串的左右兩端引號是否型態一致即可解決。

### 📦 7-1-2 引號巢狀包夾技巧（外單內雙、外雙內單）

**生活比喻：夾心餅乾與俄羅斯套娃**  
在日常書寫與英文對話中，引號是極為常見的符號。例如英文的縮寫及所有格常帶有撇號（單引號）：`It's a great book`、`Teacher's apple`；而在中文故事或小說對話中，人物說出的台詞通常會被雙引號包覆：「小明說："大家加油！"」。  
如果我們想在 Python 程式中完整記錄這句話，該怎麼辦？如果直接寫 `'It's sunny'`，就像是用同樣形狀的鑰匙插進門孔，Python 會以為第一個撇號就是收尾扣環，導致文字被腰斬！

**底層運作機制與八字真言**  
Python 直譯器在掃描程式碼時，它的邏輯非常純粹：**「它只在尋找與最外層開頭完全相符的那一個引號」**！  
因此，如果我們在最外層使用了**雙引號**，那麼在遇到下一個雙引號之前，文字中間出現的任何**單引號**，直譯器都會把它們當作普普通通的文字珍珠，完全不會把它們誤判為字串結束！  
同理，若最外層使用的是**單引號**，內部就可以隨心所欲地容納**雙引號**！  

請務必牢記這八字真言：  
👉 **「外單內雙，外雙內單」**！
- **外雙內單**：`message = "It's raining outside."`（外層雙引號，內部可直接放單引號 `'`）
- **外單內雙**：`quote = 'The master said: "Never give up!"'`（外層單引號，內部可直接放雙引號 `"`）

**常見錯誤陷阱：引號型態打架導致的 SyntaxError**  
看看這行經典錯誤：  
`print('I'm fine.')`  
在 Python 眼中，這行被切成了三段：
1. `'I'`：合法的字串。
2. `m fine.'`：既不是合法的變數名，也不是運算子，最後還有一個孤苦伶仃的結尾引號！  
這會引發 `SyntaxError: invalid syntax`。只要把外層改為雙引號 `print("I'm fine.")`，問題瞬間迎刃而解！

APCS 檢定中有時會要求印出包含特定對話格式的字樣（如輸出帶有引號的字串標籤）。善用「外單內雙」或「外雙內單」的技巧，比起使用跳脫符號更不易眼花出錯，能使程式碼清爽又好讀。

### ⚡ 7-1-3 常用跳脫字元：\n（換行）與 \t（水平跳格 Tab）

**生活比喻：鍵盤上的隱形幽靈與精靈排版師**  
當我們在鍵盤上打字寫報告時，按下 `Enter` 鍵會跳到下一行，按下 `Tab` 鍵會讓游標跳到下一個對齊格。這些按鍵雖然會產生實質的排版動作，但螢幕上並不會印出一個具體的「符號形狀」，它們就像是一群看不見的隱形幽靈！  
那麼，在寫 Python 單行程式碼時，我們該如何把「跳到下一行」或「空格對齊」打包進字串裡面呢？  
這時就需要呼叫 **「跳脫字元（Escape Character）」**！反斜線 `\` 就是召喚魔法的魔杖，當它出現在特定字母前面時，就會賦予該字母不可思議的排版魔力！

**底層運作機制：`\n` 與 `\t` 的威力**  
在 Python 字串中，反斜線 `\` 會讓緊隨其後的字元脫離原有的純文字字面含義：
1. **`\n`（Newline，換行字元）**：
   - 代表「換行」。當 `print()` 讀到 `\n` 時，游標會立刻移動到下一行的最左端開始輸出。
   - 範例：`"Line 1\nLine 2"` 會被輸出為兩行！
2. **`\t`（Tab，水平跳格字元）**：
   - 代表「水平跳格（Tab）」。它會讓文字游標自動跳到下一個制表位（通常為 4 或 8 個字元的整數倍位置），非常適合用來製作簡潔工整的成績單或數據表格。
   - 範例：`"Name\tScore\nBob\t100"` 能夠讓欄位整齊垂直對齊！

**常見錯誤陷阱：斜線方向搞錯與多餘空行**  
初學者最容易犯的錯誤就是「斜線方向顛倒」：  
- 正斜線 `/`（除法運算子，右上到左下）：寫成 `/n` 或 `/t` 只會原封不動印出字母 `/n`，完全無法換行！  
- 反斜線 `\`（跳脫字元，左上到右下，通常在鍵盤 Backspace 鍵下方）：只有 `\n` 與 `\t` 才是合法的跳脫指令！  
另外，`print()` 函數本身預設在最後就會自動換行一次；如果你的字串最後一個字元已經放了 `\n`，印出來時就會產生額外的一整行空白行，在嚴格比對輸出的評判系統中可能導致判定失敗。

**APCS 考試實務建議：**  
在解題時，若題目要求輸出多行結果，傳統寫法需要寫很多個 `print()`；而善用 `\n`，我們可以把多行文字組裝在同一個字串中，只呼叫一次 `print()` 就能整批輸出，這在測資筆數非常龐大時能有效減少 I/O 操作時間。

```python
# [2] Code 範例區：展示 \n（換行）與 \t（水平跳格）的排版效果
# 說明：觀察字串中的 \n 如何切換新行，以及 \t 如何整齊對齊欄位。

# 範例 1：使用 \n 達成單一字串輸出多行
poem = "春眠不覺曉，\n處處聞啼鳥。\n夜來風雨聲，\n花落知多少。"
print("=== \n 換行效果 ===")
print(poem)

# 範例 2：使用 \t 製作整齊的戰力成績單
table = "Role\tHP\tMP\nKnight\t120\t30\nMage\t60\t150"
print("\n=== \t 水平對齊效果 ===")
print(table)
```

### 🛡️ 7-1-4 跳脫自身與引號：\\、\' 與 \" 的特殊防護

**生活比喻：穿上防護衣的反斜線與雙重保險**  
在前面的學習中，我們見識到了反斜線 `\` 具有神奇的魔法力量，它能夠把平凡的字母 `n` 變成換行、把字母 `t` 變成跳格。  
但是，這也引來了一個大麻煩：**「如果我只是想印出一個普普通通的反斜線符號自身呢？」**  
舉個最著名的例子：在 Windows 電腦中，檔案資料夾的路徑通常以反斜線分隔，例如 `C:\new_folder\test.txt`。如果直接把這個路徑寫進字串裡，Python 一看到 `\n` 就會自作聰明地把它變成換行，一看到 `\t` 就會把它變成跳格，結果整個路徑被炸得面目全非！

**底層運作機制：用兩個反斜線代表自身**  
為了解決這個符號衝突，Python 規定了一套自我防護機制：
1. **`\\`（反斜線自身）**：
   - 前面那根反斜線是魔杖，後面那根反斜線是目標。意思是：「魔杖請解除下一個字元的魔法，回歸平凡的反斜線文字！」。
   - 螢幕上只會顯示出**單一根反斜線** `\`。
   - 範例：`"C:\\new_folder\\test.txt"` 才能正確印出 `C:\new_folder\test.txt`。
2. **`\'` 與 `\"`（跳脫引號）**：
   - 如果你在單引號字串中，非得使用單引號不可；或者在雙引號字串中非得使用雙引號不可，只要在引號前加上反斜線（如 `\'` 或 `\"`），直譯器就會明白「這只是普通引號文字，不是字串的結尾扣環」！
   - 範例：`'It\'s amazing!'`（完全合法，不會報錯）。

**常見錯誤陷阱：字串末端單獨反斜線的致命崩潰**  
初學者極度容易踩中一個隱形地雷：**字串的最末端不能是單獨一個反斜線！**  
例如：`path = "C:\test\"`  
大家以為結尾的雙引號是字串結尾，但因為它前方緊挨著一個反斜線 `\"`，Python 會認定你是在「跳脫雙引號」！結果字串一直沒有閉合，最後再次觸發：  
**`SyntaxError: unterminated string literal`**  
若結尾需要印出反斜線，必須打兩根：`"C:\\test\\"`！

**APCS 考試實務建議：**  
在輸出 ASCII 藝術圖形（如幾何斜線、三角形、顏文字或迷宮地圖）時，反斜線是常見的組成部件。若題目給定的範例輸出含有 `\`，在 Python 程式碼中必須以 `\\` 雙重書寫，否則不僅會少印字元，還可能引發意外的轉義錯誤。

```python
# [2] Code 範例區：展示 \\、\' 與 \" 的特殊防護技巧
# 說明：正確使用 \\ 印出路徑，以及用 \' 和 \" 實現在同型引號中印出引號。

# 範例 1：使用 \\ 輸出真實的檔案路徑
file_path = "C:\\Program Files\\Python313\\python.exe"
print("檔案路徑範例:", file_path)

# 範例 2：在單引號中使用 \' 保持單引號合法性
message = 'It\'s a secret that shouldn\'t be told.'
print("跳脫單引號範例:", message)

# 範例 3：印出經典的趣味顏文字（包含反斜線）
shrug = "¯\\_(ツ)_/¯"
print("顏文字範例:", shrug)
```

### 📜 7-1-5 三引號 ''' 與 """ 多行文字與原始排版

**生活比喻：全景巨幅畫布與整頁羊皮紙手稿**  
如果我們要記錄一整首古典詩詞、一封長長的信件、或者一個 RPG 遊戲開頭冗長的世界觀介紹，如果每寫一行都要手動敲一個 `\n`，或者寫幾十個 `print()`，那程式碼看起來會支離破碎、非常雜亂。  
這時，Python 為我們準備了一面超級畫布——**「三引號（Triple Quotes）」**！  
只要在開頭與結尾各連續放上三個引號（`'''` 或 `"""`），這個區塊就像一張展開的羊皮紙，你可以直接在裡面按 Enter 換行、打空格縮排，你在螢幕上看到的樣子，印出來就是完全一模一樣的樣子！

**底層運作機制：所見即所得的多行字串**  
1. **語法規格**：
   - 使用三個單引號：`'''第一行
第二行'''`
   - 使用三個雙引號：`"""第一行
第二行"""`
2. **自動保留換行與空白**：
   - 在三引號內部，所有鍵盤敲下的換行符號都會被 Python 直譯器忠實地收納為 `\n`，所有空格縮排也都會原封不動保留。
3. **多行註解（Docstring）用途**：
   - 如果三引號字串單獨擺在程式碼中，沒有賦值給任何變數，Python 直譯器執行時會完全忽略它，因此它也常被當作長篇說明的「多行註解」使用。

**常見錯誤陷阱：三引號計數失誤與頭尾空行**  
- **引號數量必須剛好是 3**：開頭 3 個，結尾就必須剛好 3 個。少一個變成雙引號會引發語法錯誤，多打一個也會使配對錯亂。
- **首行的隱形換行**：如果在開頭三引號之後立刻按 Enter 換行，字串開頭就會自帶一個看不見的 `\n` 空行。如果你不希望第一行是空的，文字要緊接著三引號後面開始寫，或者在三引號後加上反斜線 `\` 取消首行換行。

在競技程式設計或 APCS 練習時，當我們需要測試多行文字或繪製大型選單時，三引號能讓我們在最少的程式碼行數內完成複雜排版，是非常實用的代碼整潔利器。
