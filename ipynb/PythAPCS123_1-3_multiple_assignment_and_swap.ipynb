### 👐 1-3-1 多變數同時賦值——雙手齊發的打包與解包

在前面的章節裡，如果我們想給兩個變數箱子裝入寶物，我們必須寫兩行指令：
```python
a = 5
b = 3
```
這就像是你一次只能用一隻手搬一個寶物，先搬好 5 送進 a 箱，再轉身搬 3 送進 b 箱。

但是在 Python 的神奇魔法世界中，小助手有一雙靈巧的雙手！Python 允許我們在等號兩邊用「**逗號（,）**」把好幾個變數與數值串起來，一次完成賦值：
```python
a, b = 5, 3
```

這行神奇指令是怎麼運作的呢？
1. **右側先打包**：直譯器小助手看到等號右邊有 `5, 3`，會先把它們像一顆雙黃蛋一樣打包放在一個隨身托盤上。
2. **左側對應拆開**：接著看到等號左側有 `a, b` 兩個箱子，小助手就按照從左到右的順序，將托盤上的第一個寶物（5）精準放進第一個箱子 `a`，將第二個寶物（3）放進第二個箱子 `b`！

這在程式專業術語中稱為「**多重賦值（Multiple Assignment）**」或「**元組解包（Tuple Unpacking）**」。透過逗號，我們只要一行簡潔俐落的代碼，就能一口氣設定好幾組變數，既好看又省力喔！

````python
# [2] Code 範例區：使用逗號同時給 demo_num1 與 demo_num2 賦值，並印出內容
demo_num1, demo_num2 = 5, 3
print(demo_num1)
print(demo_num2)
````

### ⚖️ 1-3-2 等號兩側天平必須對稱——個數不符的 ValueError 警報

在享受多重賦值的便利時，吳邦一老師在講義第 5 頁特別嚴格叮嚀了一句話：
「**等號左右可以有逗號分開的若干個，但個數必須一樣！**」

讓我們來看看如果兩邊個數不一樣，會發生什麼慘劇：

**情況一：右邊寶物太少，左邊箱子分不夠**
```python

```
此時直譯器小助手看到左邊準備了 3 個箱子，托盤上卻只端來 2 個寶物，第 3 個箱子空等不到東西，電腦就會大發脾氣報警：

（白話文：寶物不夠分啦！我要 3 個，你只給我 2 個！）

**情況二：右邊寶物太多，左邊箱子裝不下**

### 🥛 1-3-3 傳統變數交換的慘痛悲劇——為什麼需要第三個空杯子 temp？

現在我們來看一個經典的程式謎題：
假設你有兩個變數：`a` 裝著 5，`b` 裝著 3。如果我們希望「**把 a 和 b 的內容對調**」，讓 a 變成 3、b 變成 5，初學者通常會怎麼寫？

很多人會天真地寫出以下兩行代碼：
```python
a = b
b = a
```
吳邦一老師在講義第 5 頁犀利地問道：「**這會將 a 與 b 的內容交換嗎？為何？**」

讓我們用「兩杯飲料」來還原這個車禍現場：
想像你的左手拿著一杯鮮奶（`a`），右手拿著一杯紅茶（`b`）。
1. 當你執行第一行 `a = b` 時，相當於你直接把右手那杯紅茶倒進左手的杯子裡！這時候，左手原本的鮮奶被活生生沖洗稀釋、徹底覆蓋掉了！現在兩隻手拿的「全部都是紅茶」！
2. 接著你執行第二行 `b = a`，把左手的飲料倒回右手，結果倒來倒去還是紅茶！原本珍貴的鮮奶永遠消失了，根本沒有完成交換！

在 C 語言、Java 或傳統程式語言中，要解決這個悲劇，我們必須向桌子借「**第三個空杯子**」，在程式中通常叫做 `t` 或 `temp`（暫存變數 Temporary）：
```python
t = a  # 步驟 1：先將鮮奶暫存在空杯子 t
a = b  # 步驟 2：放心地把紅茶倒進 a 杯
b = t  # 步驟 3：把暫存杯裡的鮮奶倒回 b 杯！大功告成！
```
請大家親自執行看看，深刻體會為什麼覆蓋特性會需要第三個變數小幫手！

````python
# [2] Code 範例區：傳統交換三步驟（利用暫存變數 demo_tmp）
demo_va = 5
demo_vb = 3

demo_tmp = demo_va  # 第一步：把 demo_va 的舊值 5 保存在 demo_tmp 箱子
demo_va = demo_vb  # 第二步：把 demo_vb 的值 3 覆蓋進 demo_va（此時 demo_va 變為 3）
demo_vb = demo_tmp  # 第三步：把 demo_tmp 裡的 5 覆蓋進 demo_vb（此時 demo_vb 變為 5）

print(demo_va)  # 印出 3
print(demo_vb)  # 印出 5
````

### 🎩 1-3-4 Python 專屬的瞬間乾坤大挪移——`a, b = b, a` 變數交換

傳統的暫存變數 `t = a; a = b; b = t` 雖然能解決問題，但要寫三行代碼，還要額外多開一個 `t` 箱子，是不是覺得有點麻煩又累贅呢？

身為最優雅現代的 Python，給了全世界程式設計師一個超級大驚喜！在 Python 裡，要將兩數交換，**只需要一行指令**：
```python
a, b = b, a
```

哇！這到底是什麼神奇的魔法？為什麼這樣寫不會像剛才倒牛奶一樣被覆蓋掉呢？
吳邦一老師在講義第 5 頁為大家揭曉了秘密：
「**記得等號的運作順序：先計算右邊的值，再放入左邊的記憶體！**」

讓我們用慢動作再次分解這奇蹟的一瞬間：
假設一開始 `a = 5, b = 3`：
1. **第一步（先看右邊托盤）**：電腦小助手先看等號右邊的 `b, a`。這時候它迅速去偷看現在箱子裡的值——`b` 目前是 3，`a` 目前是 5。小助手把 `(3, 5)` 這兩個珍貴的舊數值穩穩當當地放在隱形托盤上！
2. **第二步（再存左邊箱子）**：托盤端穩之後，小助手才啟動左邊的等號傳送帶！托盤上的第一個數字 3 倒進左邊的第一個箱子 `a`；托盤上的第二個數字 5 倒進左邊的第二個箱子 `b`！

看懂了嗎？因為右邊的舊值早就在記憶體後台被完整打包保存了，所以在賦值給左邊時，根本不會互相踩踏覆蓋！
這行語法是 Python 最優雅的招牌魔法，不論是兩人交換座位還是數值互換，一行指令就能乾淨俐落完成！

````python
# [2] Code 範例區：Python 一行神技 demo_swap1, demo_swap2 = demo_swap2, demo_swap1 瞬間交換
demo_swap1 = 5
demo_swap2 = 3

# 一行乾坤大挪移！
demo_swap1, demo_swap2 = demo_swap2, demo_swap1

print(demo_swap1)  # demo_swap1 成功變成 3
print(demo_swap2)  # demo_swap2 成功變成 5
````

### 🤹 1-3-5 三人傳球大輪替——多變數循環交換（Cyclic Swap）

學會了兩個人交換，那如果是三個人、甚至四個人要進行「**大風吹輪流換位置**」呢？

吳邦一老師在講義第 5 頁出了一道非常精彩的思考題：
「**如果要將 a, b, c 三個變數循環交換（a 變成 b, b 變成 c, c 變成 a），應該如何下指令？**」

如果用剛剛學到的「先看右邊打包托盤，再存左邊箱子」的原理，這題對現在的你來說簡直易如反掌！
讓我們來對齊目標：
* 左邊的第一個箱子是 `a` ➔ 題目要求它變成 `b` 的值。
* 左邊的第二個箱子是 `b` ➔ 題目要求它變成 `c` 的值。
* 左邊的第三個箱子是 `c` ➔ 題目要求它變成 `a` 的值。

所以只要把左邊與右邊像鏡子一樣排好：
```python
a, b, c = b, c, a
```

運作過程是：右邊先把 `(b原本值, c原本值, a原本值)` 放在托盤上，然後同一時間倒進左邊的 `a, b, c` 箱子！
完全不需要宣告任何複雜的暫存變數，一個指令就能完成三人的大風吹輪替！這就是 Python 語言令無數程式設計師驚豔讚嘆的魔力所在！

````python
# [2] Code 範例區：三個變數循環交換 demo_r1 變 demo_r2, demo_r2 變 demo_r3, demo_r3 變 demo_r1
demo_r1, demo_r2, demo_r3 = 1, 2, 3

# 大風吹循環輪替！
demo_r1, demo_r2, demo_r3 = demo_r2, demo_r3, demo_r1

print(demo_r1)  # demo_r1 得到原本 demo_r2 的值：2
print(demo_r2)  # demo_r2 得到原本 demo_r3 的值：3
print(demo_r3)  # demo_r3 得到原本 demo_r1 的值：1
````

````python
#@title 🧪 點擊此處取得單元 1-3 學習成效自動評分報告 { display-mode: "form" }
# ------------------------------------------------------------------------------
# 官方智慧全環境評分載入器（支援：本地開發庫優先 ➔ 公開正式庫雙重備援）
# ------------------------------------------------------------------------------
import os
import sys
import json
import logging
import urllib.request

# 靜音內部 httplib2 警告
logging.getLogger("google_auth_httplib2").setLevel(logging.ERROR)

# 1. 互動式輸入學員姓名與學號（分開獨立欄位，按 Enter 採用預設值）
print("👤【學員身分登記】（自學測試可直接按 Enter 鍵使用預設值）：")
student_name = input("請輸入您的姓名 [預設: 自學冒險者]：").strip() or "自學冒險者"
student_id = input("請輸入學號或座號 [預設: 無座號]：").strip() or "無座號"

# 2. 自動嘗試獲取 Google 帳號官方身分認證（真實 Email 與 Google 暱稱）
google_email = None
google_name = None
try:
    from google.colab import auth
    print("🔐 正在連結 Google 帳號進行身分認證（若跳出授權視窗，請點選允許）...")
    auth.authenticate_user()

    # 方法 1: 使用 Google Drive API v3 (about:get) 抓取真實 displayName 與 emailAddress
    # Colab ADC 預設包含 drive 權限，能 100% 精準獲取 Google 帳號中文顯示名稱（如：葉明昌）
    try:
        import google.auth
        from googleapiclient.discovery import build
        creds, _ = google.auth.default()
        drive_service = build('drive', 'v3', credentials=creds, cache_discovery=False)
        about = drive_service.about().get(fields="user(displayName,emailAddress)").execute()
        u_obj = about.get("user", {})
        google_name = u_obj.get("displayName")
        google_email = u_obj.get("emailAddress")
    except Exception:
        pass

    # 方法 2: 若 Drive API 失敗，嘗試透過 OAuth2 userinfo / tokeninfo 備援
    if not google_email or not google_name:
        try:
            import google.auth
            from google.auth.transport.requests import Request
            credentials, _ = google.auth.default()
            credentials.refresh(Request())
            token = credentials.token

            for ep in [
                f"https://www.googleapis.com/oauth2/v3/tokeninfo?access_token={token}",
                "https://www.googleapis.com/oauth2/v3/userinfo",
                "https://www.googleapis.com/oauth2/v2/userinfo"
            ]:
                try:
                    req = urllib.request.Request(
                        ep,
                        headers={"Authorization": f"Bearer {token}"}
                    )
                    with urllib.request.urlopen(req, timeout=5) as resp:
                        u_data = json.loads(resp.read().decode('utf-8'))
                        if not google_email:
                            google_email = u_data.get("email") or u_data.get("emailAddress")
                        if not google_name:
                            google_name = (
                                u_data.get("name")
                                or u_data.get("given_name")
                                or u_data.get("displayName")
                            )
                        if google_email and google_name:
                            break
                except Exception:
                    continue
        except Exception:
            pass

    # 方法 3: 若只有 email 但 Google API 沒給 name，以 email 前綴作為暱稱回退
    if google_email and not google_name:
        google_name = google_email.split("@")[0]

    if google_email:
        print(f"✅ 身分認證成功！已綁定 Google 帳號：{google_name} ({google_email})")
except Exception:
    # 非 Colab 環境、本機離線測試或略過授權
    pass

# 3. 身分資料獨立注入環境變數（學號與姓名完全分開）
os.environ["STUDENT_NAME"] = student_name
os.environ["STUDENT_ID"] = student_id
if google_email:
    os.environ["GOOGLE_EMAIL"] = google_email
if google_name:
    os.environ["GOOGLE_NAME"] = google_name

# 4. 評分核心腳本載入（本機優先 ➔ 遠端記憶體載入，絕不產生實體暫存檔）
# 找不到來源就換下一條。內容已拿到但執行失敗時，停止改試其他來源。
# 評分腳本放在獨立命名空間執行，成功後也不把名稱寫回筆記本。答案讀執行前的學生存放格快照。
_loaded = False
_load_error = None

# 4.1 優先嘗試讀取開發庫/本機檔案（本機離線開發測試零延遲）
_local_paths = [
    "web/graders/grade_1_3.py",
    "../graders/grade_1_3.py",
    "../../graders/grade_1_3.py",
    "../web/graders/grade_1_3.py",
    "../../web/graders/grade_1_3.py"
]
for _lp in _local_paths:
    if not os.path.exists(_lp):
        continue
    try:
        with open(_lp, "r", encoding="utf-8") as _f:
            _code = _f.read()
        _grader_ns = {"__builtins__": __builtins__, "__name__": "apcs_grader", "__STUDENT_ENV__": dict(globals())}
        try:
            exec(_code, _grader_ns)
        finally:
            del _grader_ns
        os.environ["GRADER_SOURCE"] = _lp
        _loaded = True
        break
    except Exception as _exc:
        _load_error = _exc
        break

# 4.2 若在雲端環境（Colab），自動透過公開正式庫雙重備援以純記憶體載入
if not _loaded and _load_error is None:
    _grader_urls = [
        "https://raw.githubusercontent.com/johnnyy-lab/APCS1to3/main/graders/grade_1_3.py",
        "https://johnnyy-lab.github.io/APCS1to3/graders/grade_1_3.py",
        "https://raw.githubusercontent.com/johnnyy-lab/APCS1to3-dev/main/web/graders/grade_1_3.py"
    ]
    for _url in _grader_urls:
        try:
            _req = urllib.request.Request(_url, headers={"User-Agent": "Mozilla/5.0"})
            with urllib.request.urlopen(_req, timeout=10) as _resp:
                _code = _resp.read().decode("utf-8")
        except Exception:
            continue
        try:
            _grader_ns = {"__builtins__": __builtins__, "__name__": "apcs_grader", "__STUDENT_ENV__": dict(globals())}
            try:
                exec(_code, _grader_ns)
            finally:
                del _grader_ns
            os.environ["GRADER_SOURCE"] = _url
            _loaded = True
            break
        except Exception as _exc:
            _load_error = _exc
            break

if not _loaded:
    if _load_error is not None:
        print("❌ 評分程式載入失敗，請稍後重新執行。若仍失敗，請把這個畫面交給老師。")
        if os.environ.get("GRADER_DEBUG", "").strip().lower() in ("1", "true", "yes", "on"):
            _reason = str(_load_error).strip().splitlines()
            _line = _reason[0][:180] if _reason else ""
            if _line:
                print("（開發診斷：" + type(_load_error).__name__ + "：" + _line + "）")
            else:
                print("（開發診斷：" + type(_load_error).__name__ + "）")
    else:
        print("❌ 評分系統連線異常，請確認網路連線狀態（或稍後重新執行）")
````
