### 🏷️ 1-1-3 變數命名規則一——名牌只能使用英文字母、數字與底線（_）

既然要給箱子貼名牌，那名牌上的字可以隨心所欲地亂塗亂畫嗎？答案當然是「不行」！Python 的管理員機器人是個很有規矩、做事一絲不苟的小幫手，它的光學掃描儀只看得懂三種筆劃字元：
1. **英文字母**：大寫 `A` 到 `Z`、小寫 `a` 到 `z` 都可以。
2. **阿拉伯數字**：`0` 到 `9` 的數字都可以使用。
3. **底線符號**：鍵盤上的貼地底線 `_`（按住 Shift 加上減號鍵）。

很多剛接觸程式的初學同學最容易犯的初學者錯誤，就是在變數名牌裡面隨意加了「空格」，例如寫成：
`my score = 90`（這是嚴重大忌！）
你想想看，如果名牌中間破了一個大空格，管理員機器人在閱讀程式碼時，就會以為 `my` 是一道命令，而 `score` 又是另一道命令，兩邊對不起來，腦袋瞬間打結而噴出 `SyntaxError`（語法錯誤）！

同樣地，如果把名牌寫成 `my-score`（使用了減號），電腦會誤以為你想拿變數 `my` 去「減掉」變數 `score` 呢！
那麼，當我們想用好幾個英文單字拼成一個語意清清楚楚的名牌時，該怎麼辦呢？這時候，貼地的「**底線 `_`**」就是全宇宙最強的黏著劑！例如寫成 `my_score`、`player_1_level` 或 `test_number`。這種用底線把單字串連起來的漂亮格式，在程式界被形象地稱為「**蛇形命名法（snake_case）**」，也是 APCS 考試與專業工程師最推薦的寫法喔！

````python
# [2] Code 範例區：合法使用英文字母、數字與底線組合（蛇形命名法）
team_member_1 = 25
print(team_member_1)
````

### 🚫 1-1-4 變數命名規則二——頭號禁忌：開頭絕對不能是數字！

現在請所有同學拿出螢光筆，在你的筆記本上用力畫上三顆大星星！因為接下來這條規則，是每年全台灣程式初學者、甚至 APCS 考生最容易不小心踩到的經典超級大地雷——「**變數名牌的第一個字元，絕對不能是阿拉伯數字！**」

也就是說，你可以把變數命名為 `box1`、`player2`、`stage_3` 或是 `test99`（數字放在英文字母後面，完全合法且很常見）；但是你**千千萬萬不能**寫成 `1box`、`2player`、`3_stage` 或 `99test`！

為什麼 Python 會有這麼固執的怪脾氣呢？這其實是為了保護直譯器小助手的運算秩序。電腦在閱讀任何一行程式碼時，習慣由左至右一個字一個字掃描。當直譯器眼睛看到第一個字元是 `1`、`2` 或 `9` 這樣的阿拉伯數字時，它的大腦神經會瞬間切換成「數學模式」，心裡開心地預期：「太好了！主人接下來要輸入一個數學常數，準備讓我做加減乘除囉！」
結果讀到第二個字元時，眼前突然冷不防冒出英文字母 `b` 或 `p`，直譯器當場嚇傻：「天啊！這世界上哪有長了英文字母翅膀的怪胎數字啊？」
於是直譯器瞬間當機，憤怒地丟出 `SyntaxError: invalid decimal literal`（無效的十進位數值）錯誤警告。

因此，請大家深深印在腦海中：**數字永遠只能乖乖當跟班，放在英文字母或底線的後面，絕對不能搶當開頭的第一個字！**

````python
# [2] Code 範例區：數字必須擺在英文字母後面（如 stage1）
stage1 = 100
print(stage1)

# 警告小叮嚀：若嘗試寫成 1stage = 100，
# Python 直譯器會直接丟出 SyntaxError: invalid decimal literal 喔！
````

### 👑 1-1-5 變數命名規則三——大小寫大不同（超敏銳的辨識眼）

在學校的英語課上，英文老師一定告訴過大家：不管是首字大寫的 `Apple` 還是全小寫的 `apple`，在中文裡都代表紅通通的水果「蘋果」。但是在 Python 的神奇世界裡，事情可完全不是這樣！

Python 的置物櫃管理員配備了一副「**極度敏銳、絕不妥協的大小寫辨識眼**」（在專業術語中稱為 **Case-sensitive，大小寫敏感**）。在大腦嚴謹的 Python 眼裡，大寫英文字母跟小寫英文字母，是長相完全不同、血統完全獨立的兩個個體！

讓我們用一個生動的角色扮演來理解：
想像一下，穿著隨興休閒便服的小寫 `score`，是一位平民小英雄；而頭戴閃亮皇冠、身穿華麗披風的大寫 `Score`，則是尊貴的帝國國王；至於全身都金光閃閃的全大寫 `SCORE`，更是神聖的創世神！這三者在記憶體大樓裡面住的是三間完全不同的獨立套房，門牌貼紙長得不一樣，裡面的寶物也完全各自獨立、互不干涉。

如果你寫了：
```python
score = 80
```
把 80 分存進了平民小英雄的房間，轉頭卻在程式中寫 `print(Score)`（首字不小心大寫），電腦只會一臉無辜地對你搖搖頭，拋出 `NameError` 大喊找不到 `Score` 的房間！

這條規則告訴我們：
1. **輸入變數名時，手眼要高度專注**：特別留意鍵盤上的 Caps Lock（大小寫鎖定鍵）以及 Shift 鍵，絕不能手滑混用。
2. **通篇保持一致**：如果你一開始決定變數叫小寫 `ans`，後面呼叫時就必須百分之百精確地使用小寫 `ans`！

````python
# [2] Code 範例區：示範大小寫是完全獨立、互不影響的兩個置物箱
score = 80
Score = 95

print(score)  # 印出小寫 score 的內容：80
print(Score)  # 印出大寫 Score 的內容：95
````

### 🪄 1-1-6 變數命名規則四——避開魔法咒語（保留關鍵字與內建名稱）

來到 1-1 單元的最後一個小絕招！我們要學習如何避開 Python 魔法王國裡的「**神聖禁忌咒語**」。

在 Python 語言誕生之初，創作者就先挑選了三十幾個具有強大魔法超能力的單字，保留給系統專屬使用，這些詞彙在電腦科學裡稱為「**保留字 / 關鍵字（Keywords）**」。
例如：
* `for` 與 `while`：是用來發動「旋轉分身、重複迴圈」的專屬咒語。
* `if` 與 `else`：是用來進行「命運十字路口、條件分支抉擇」的專屬咒語。
* `def`：是用來「創造新技能、打造新函數」的專利指令。
* `True` 與 `False`：是代表「真理與謬誤」的專屬布林值名牌。

如果你異想天開，寫出 `for = 10` 或是 `if = 5` 這樣的程式碼，Python 系統會立刻陷入混亂大當機！因為它分不清楚你到底是要發動迴圈咒語，還是要給箱子貼名牌，因此會直接亮出紅牌 `SyntaxError: invalid syntax` 拒絕執行！

此外，吳邦一老師在《PythAPCS123》講義中給了大家一個價值連城的經驗談：「**最好不要取完整的英文字，可以用簡稱或加上編號，例如 test1, arr2, foo, bar。**」
這是為什麼呢？因為像 `print`（電腦說話的嘴巴）、`sum`（總和算盤）、`max`（挑最大值雷達）這些單字雖然在語法上允許被賦值，但它們是 Python 內建的珍貴公用工具！如果你寫了 `print = 10`，你就像親手把膠帶黏在電腦的嘴巴上，讓電腦從此變成啞巴，後面所有的 `print()` 指令都會全部壞掉！

因此，為了保證程式在 APCS 考場上萬無一失，請牢記老師的忠告：利用底線修飾、縮寫或加上數字編號（如 `my_sum`、`max_val`、`box_for`），就能保證你的程式永遠不會撞到魔法咒語囉！

````python
# [2] Code 範例區：避免撞到關鍵字 if 或 for，加上修飾字或編號
my_choice_if = 1
for_count = 5

print(my_choice_if)
print(for_count)

# 警告：若直接寫 if = 1 或 for = 5，會引發 SyntaxError: invalid syntax
````
