### 12.2.1 參數 `reverse=True`：從大到小反向降序排列

#### 1. 生活故事比喻：頒獎典禮的金銀銅牌排行榜
在奧運百米賽跑決賽結束後，大會公布成績時，第一名金牌得主永遠印在最上方，接著是第二名銀牌、第三名銅牌，依此類推。但在射擊或打電動比賽中，分數最高的玩家才是冠軍！
如果我們想將全班考試成績由高到低（降序）排出榜單，我們不需要先由小到大排完再手動倒著抄一遍。
Python 的排序工具貼心地內建了一個「倒車檔按鈕」——**`reverse=True`**！只要在呼叫排序時按下這個開關，原本由小到大的升序傳送帶就會瞬間轉向，直接由大到小輸出！

#### 2. 底層運作機制：`reverse` 參數的通用性
不論是串列的原地排序方法 `list.sort()`，還是內建的獨立函數 `sorted()`，都一致支援具名參數 `reverse=True`：
- `numbers.sort(reverse=True)`：在原串列記憶體空間中，直接將元素由大到小原地重新排列。
- `ranked = sorted(numbers, reverse=True)`：不影響原串列，回傳一個全新由大到小排列的新串列。
預設情況下，`reverse` 參數的值為 `False`（代表升序）。當顯式指定 `reverse=True` 時，底層比較邏輯在每一次決策時反轉大於小於的判定，從而達成嚴格的降序排列（Descending Order）。

#### 3. 初學者常見陷阱：單詞拼寫錯誤與大小寫混淆
初學者最常發生的語法錯誤是拼錯字：
```python
scores = [60, 80, 70]
scores.sort(reverse=true)   # 報錯 NameError: name 'true' is not defined (首字未大寫)
scores.sort(reversed=True)  # 報錯 TypeError: 'reversed' is an invalid keyword argument (多打了 d)
```
請特別留意：參數名稱是純名詞 **`reverse`**（無 `d`），而其賦予的布林值首字母必須大寫為 **`True`**！

#### 4. APCS 實戰視野
在 APCS 貪婪演算法（Greedy Algorithm）題型中，如背包問題挑選價值最高物品、面額找零優先拿最大硬幣，前置第一步永遠是將待選清單「由大到小降序排序」。熟練 `reverse=True` 是迅速切入貪婪策略的基本功。

````python
# 範例 12.2.1：使用 reverse=True 進行由大到小降序排序
# 1. 使用 sorted() 產生降序新串列（電玩積分榜）
points = [1200, 3500, 800, 4200, 2100]
leaderboard = sorted(points, reverse=True)

print("原始分數清單：", points)
print("降序排行榜（最高分在前）：", leaderboard)
print("冠軍得分：", leaderboard[0])
print("最後一名得分：", leaderboard[-1])

# 2. 使用 .sort(reverse=True) 進行原地降序修改
inventory_weights = [15.5, 42.0, 8.3, 27.1]
print("\n原地排序前負重：", inventory_weights)
inventory_weights.sort(reverse=True)
print("原地降序排序後負重：", inventory_weights)
````

### 12.2.2 數值排序 vs 字串字典序（Lexicographical Order）：ASCII 編碼順序

#### 1. 生活故事比喻：實體字典的翻閱順序
當你在翻閱一本厚厚的英文字典查詢單字時，你找單字的規則是什麼？
你會先看第一個字母：以 `a` 開頭的字（如 `apple`）絕對排在以 `b` 開頭的字（如 `banana`）前面。
如果兩個單字的第一個字母一模一樣呢？例如 `cat` 和 `car`，你就會把目光移向第二個字母，第二個字母都是 `a`，於是你繼續看第三個字母：因為 `r` 排在 `t` 前面，所以 `car` 一定排在 `cat` 前面。
這種**「由左至右、逐字比對」**的排列規則，在電腦科學中就叫做**字典序（Lexicographical Order）**！

#### 2. 底層運作機制：字串比大小的背後是 ASCII 數值比對
在 Python 內部，當我們比較兩個字串 `"cat" < "dog"` 時，電腦並不懂人類語言的含義。它做的事情是：
1. 拿出兩字串索引為 0 的字元：`'c'` vs `'d'`。
2. 查詢它們的底層編碼：`ord('c')` 是 99，`ord('d')` 是 100。
3. 因為 $99 < 100$，比對當場宣告結束！`"cat"` 小於 `"dog"`。
只要在前方的某個位置分出了勝負，後面的所有字元完全不具任何影響力！如果一個字串是另一個字串的前綴（例如 `"app"` 與 `"apple"`），則較短的字串會被判定為較小。

#### 3. 初學者常見陷阱：以為字串會看長度來比大小
許多初學同學會直覺以為「字串越長，排序時就排在越後面」，例如誤以為 `"zoo"`（長度 3）比 `"apple"`（長度 5）還要小。
但事實上，字典序完全不看字串的總長度！因為 `'z'` 的編碼（122）大於 `'a'` 的編碼（97），所以單單一個字元 `'z'` 就足以打敗整串 `"apple"`：`"zoo" > "apple"` 是百分之百的真理。

#### 4. APCS 實戰視野
在 APCS 與各類資訊競賽中，字串排序題目極為常見（如學生姓名清單字典序排序、字串前綴查詢）。深刻理解「逐字元 ASCII 比較、前字元決定勝負」的字典序特性，是寫出正確字串邏輯的根本關鍵。

````python
# 範例 12.2.2：字串字典序排序與逐字元比對體驗
fruits = ["banana", "apple", "cherry", "avocado", "blueberry"]

# 觀察字典序升序排序
sorted_fruits = sorted(fruits)
print("水果單字字典序排序結果：")
for f in sorted_fruits:
    print(f"  {f} (首字母編碼: {ord(f[0])})")

# 觀察前綴字串比對：短字串排在前方
words = ["app", "apple", "application"]
print("\n前綴關係單字排序：", sorted(words))

# 破除長度迷思："zoo" 雖然只有三個字，但比 "watermelon" 還大
print("\n'zoo' > 'watermelon' 的比對結果：", "zoo" > "watermelon")
````

### 12.2.3 考場世紀大陷阱：字串 `'100'` 小於 `'2'` 的成因與數值轉型 `int` 必要性

#### 1. 生活故事比喻：門牌號碼貼紙與銀行存款
想像你手上有兩張寫著數字的名牌貼紙：一張寫著 `"100"`，另一張寫著 `"2"`。
如果你把這兩張貼紙拿去問銀行行員：「哪一個數字代表的錢比較多？」行員一定毫不猶豫回答：「當然是一百元比二元多！」
但如果你把這兩張貼紙拿去給圖書館的英文字母檔案管理員，管理員戴著眼鏡逐字檢驗：
- 看到第一張的第一個字元是 `'1'`（字元編碼 49）。
- 看到第二張的第一個字元是 `'2'`（字元編碼 50）。
管理員立刻蓋章宣判：「因為 `'1'` 排在 `'2'` 前面，所以 `"100"` 比 `"2"` 還要小！」
這就是無數程式學習者在 APCS 考場上留下悔恨眼淚的**「文字數字字典序大陷阱」**！

#### 2. 底層運作機制：`"100" < "2"` 的真實真相
當串列儲存的是文字型態（`str`）時，直譯器絕對不會把它當成數值計算！
在進行比較時：
`"100"[0]` 是 `'1'`，其 `ord('1') == 49`。  
`"2"[0]` 是 `'2'`，其 `ord('2') == 50`。  
因為 $49 < 50$，電腦比對完第一個字元就立刻結束判定，得出 `"100" < "2"` 為 `True`！
如果你有一組文字串列 `["10", "2", "1", "20", "100"]`，直接對它排序會得到荒謬的結果：
`['1', '10', '100', '2', '20']`！所有 1 開頭的字串通通擠在前面，2 開頭的全部落在後面。

#### 3. 初學者常見陷阱：從 `input().split()` 讀入後忘記轉型 `int`
在競賽中，當題目輸入一行整數時，許多同學常這樣寫：
```python
# 致命錯誤！
raw_nums = input().split()  # 得到的是字串串列，例如 ['15', '3', '100', '9']
raw_nums.sort()            # 結果是 ['100', '15', '3', '9']，徹底錯亂！
```
因為 `input().split()` 切割出來的產物**百分之百是字串**！如果沒有透過 `int()` 進行數值轉型，後續的任何比大小、求最大值或排序，全部都會按照字串字典序運作，導致全盤皆輸。

#### 4. APCS 實戰視野
防禦此陷阱的唯一黃金鐵律：**讀入數值資料時，永遠使用列表生成式或 map 轉為整數！**
即：`nums = [int(x) for x in input().split()]`。只有讓資料回歸真正的 `int` 整數型態，排序時電腦才會依據數值量值（Magnitude）比較，一百才會真正大於二！

````python
# 範例 12.2.3：字串數字字典序陷阱與整數轉型防禦
# 1. 驚人的字串字典序比較
str_list = ["100", "2", "15", "30", "5"]
print("未轉型的字串串列：", str_list)

str_sorted = sorted(str_list)
print("直接對字串排序（大災難！）：", str_sorted)
print("驗證 '100' < '2' 是否成立：", "100" < "2")  # 驚人地輸出 True！

# 2. 正確的解法：數值轉型防禦
# 利用列表生成式將每一個元素轉為 int
int_list = [int(x) for x in str_list]
print("\n轉型為 int 後的數值串列：", int_list)

int_sorted = sorted(int_list)
print("數值排序後的正確結果：", int_sorted)
print("正確的最小值：", int_sorted[0], "，最大值：", int_sorted[-1])
````

### 12.2.4 大小寫混合字串排序規則：大寫字母（65~90）永遠排在小寫字母（97~122）前面

#### 1. 生活故事比喻：戴皇冠的國王與平民百姓
想像在古代王國的盛大遊行中，有一條鐵律：只要是頭戴皇冠的貴族國王（大寫字母），不管他的名字多簡短，遊行時永遠走在最前方的貴賓特區；而平民百姓（小寫字母），不管他的功勳再顯赫，隊列都必須排在貴族特區的後面。
在 ASCII 編碼世界中，也是完全一樣的階級制度！
大寫英文字母 `'A' ~ 'Z'` 的編碼範圍是 **65 到 90**。  
小寫英文字母 `'a' ~ 'z'` 的編碼範圍是 **97 到 122**。  
因為 65 到 90 永遠小於 97 到 122，所以**「任何大寫字母，在排序時永遠排在任何小寫字母前面」**！

#### 2. 底層運作機制：`"Zebra"` 竟然排在 `"apple"` 前面！
這又是一個反直覺的認知衝擊：
在一般的英文字典中，`apple` 一定排在 `zebra` 前面。但在電腦世界中，如果你比對 `"Zebra"` 與 `"apple"`：
- `'Z'` 的 ASCII 編碼是 90。
- `'a'` 的 ASCII 編碼是 97。
因為 $90 < 97$，直譯器判定 `"Zebra" < "apple"`！
因此，若對 `["apple", "Banana", "cat", "Apple"]` 進行排序，結果會是大寫字詞先就位，接著才是小寫字詞：
`['Apple', 'Banana', 'apple', 'cat']`！

#### 3. 初學者常見陷阱：預期「不分大小寫」卻得到「大寫集中在最前」
初學同學在處理英文人名、國名或單字表時，常常被這種現象搞得一頭霧水：「為什麼明明名單裡有小寫的 a 開頭單字，卻排在大寫 Z 開頭單字的後面？」
這就是因為沒有考慮到 ASCII 編碼的絕對大小。純粹呼叫 `sorted(words)` 執行的就是最標準、冷酷無情的 ASCII 碼比對（也稱為 Case-sensitive 區分大小寫排序）。

APCS 觀念題常以大小寫混合的字串清單為考題，測試學生是否真正理解 ASCII 編碼順序。而在字串比對的實作題中，若題目明確要求「字母順序（不分大小寫）」，我們就必須引入大小寫轉換技術，這也是我們稍後 12.2.6 即將初探的關鍵武器！

### 12.2.5 負號取反技巧：利用 `-x` 在純數值排序中進行升降序快速切換

#### 1. 生活故事比喻：數線上的照後鏡與倒影
在國中數學課本中，我們學過「數線」：向右為正、向左為負。
如果原點右邊有三個點：$2, 5, 9$，顯然 $2 < 5 < 9$。
現在想像原點立了一面魔鏡，把這些數字全部「乘上負號（加個負號取反）」映照到原點左側，它們變成了：$-2, -5, -9$。
這時候大小關係發生了奇妙的反轉！
在負數世界裡：$-9 < -5 < -2$！原本最大的正數 $9$，加上負號後變成了最深、最小的負數 $-9$！
利用這個純數學的**「負號反轉特性」**，我們可以在完全不更動排序演算法升序核心的情況下，輕巧實現降序的效果！

#### 2. 底層運作機制：特徵值反向映射
當我們想把一組數值由大到小排序時，除了使用 `reverse=True`，在許多高階場合（特別是後續第 12.4 節即將學到的多準則排序：第一條件大到小、第二條件小到大），`reverse=True` 會把所有條件一起反向，無法單獨針對某一個欄位降序。
這時候，**負號取反 `-x`** 就是最犀利的武器！
- 如果你讓電腦比對 `-x`，原本數值越大者，`-x` 的值就越小，於是在升序排序時就會被搶先排在最前面！
例如，比較兩分數 $100$ 與 $60$：
因為 $-100 < -60$，若以負號值作為評判標準，電腦就會把 $100$ 排在前面。

#### 3. 初學者常見陷阱：試圖對字串加上負號引發 TypeError
負號取反技巧雖然極度優雅，但請注意它的適用範圍：**僅限純數值型態（int, float）！**
如果你對字串寫 `-s`：
```python
name = "Alice"
print(-name)  # 致命報錯！TypeError: bad operand type for unary -: 'str'
```
字串是沒有負號運算子的！初學者在多欄位排序時若不小心對字串欄位加負號，程式就會立即崩潰。

#### 4. APCS 實戰視野
在 APCS 的多欄位結構題中（例如：輸入學生成績與學號，要求「成績高者在前；若成績相同，學號小者在前」），這種「一降一升」的混合需求，幾乎 100% 的競賽選手都是利用負號技巧（將成績加上負號 `-score`）一瞬間輕鬆解決！

````python
# 範例 12.2.5：負號取反技巧原理演示
numbers = [15, 3, 99, 42, 8]
print("原始數值串列：", numbers)

# 觀察取負號後的數值分布
neg_numbers = [-x for x in numbers]
print("各元素取負號後的數列：", neg_numbers)

# 觀察取負號數列升序排序後的結果
neg_sorted = sorted(neg_numbers)
print("取負號數列由小到大排序：", neg_sorted)

# 再次取負號還原回原數值，奇蹟發生：原數值已經是由大到小降序了！
restored_descending = [-x for x in neg_sorted]
print("還原後的降序結果：", restored_descending)

# 撰寫一個簡單的取負號特徵函數
def make_negative(val):
    return -val

# 將 make_negative 作為排序標準傳入 key
# （提早感受：利用特徵值反轉達成降序）
custom_descending = sorted(numbers, key=make_negative)
print("透過 key=make_negative 達成的降序：", custom_descending)
````

### 12.2.6 字母大小寫不敏感排序初探：初探 `key=str.lower`

#### 1. 生活故事比喻：同字視同仁的英語詞彙表
想像你是一位英文老師，要編印一本給全校同學背誦的單字手冊。名單中有首字母大寫的專有名詞（如 `"Apple"`, `"Bob"`），也有一般小寫單字（如 `"ant"`, `"banana"`）。
如果你按照上一節學到的預設 ASCII 規則印出來，學生翻閱時會發瘋：大寫專有名詞通通堆在前半本，小寫單字全部擠在後半本！
身為老師，你希望的是：**不管是大寫的 A 還是小寫的 a，一律當成同一個字母看待！** `ant` 和 `Apple` 應該緊緊排在一起，接著才是 `banana` 和 `Bob`。
這種「無視大小寫差異，一律以小寫形態進行順序評判」的模式，在程式設計中稱為**大小寫不敏感排序（Case-insensitive Sort）**。

#### 2. 底層運作機制：`key=` 參數的透視鏡概念
在 Python 中，`sort()` 和 `sorted()` 提供了一個神級參數：**`key=`**。
當我們傳入 `key=str.lower` 時，Python 在進行元素大小比較時，會默默把每個單字先放進 `str.lower()` 機器中「透視」出它全小寫時的長相。
例如：
- 看到 `"Apple"`，透視出 `"apple"`。
- 看到 `"ant"`，透視出 `"ant"`。
- 看到 `"Bob"`，透視出 `"bob"`。
排序演算法在幕後比對的是這些**透視出來的特徵值**！但最棒的是：**排序完成後，清單中保留的依然是單字原本漂亮的原始大小寫長相！**

#### 3. 初學者常見陷阱：呼叫 `str.lower()` 時多打了括號 `()`
初學同學最常犯的語法錯誤：
```python
words = ["Banana", "apple"]
# 致命錯誤！
sorted(words, key=str.lower())  # 報錯 TypeError: descriptor 'lower' of 'str' object needs an argument
```
請千萬記住：傳給 `key=` 的必須是**函數本身的名牌（Function Object）**，而不是立刻執行呼叫它！所以後面**絕對不能加小括號 `()`**！你只是把工具（`str.lower`）交給排序工人，讓工人去對每個元素逐一加工，而不是由你現在立刻執行。

#### 4. APCS 實戰視野
初探 `key=str.lower` 是我們跨入進階客製化排序的「第一座橋樑」！這為下一小節（12.3 節）我們正式全面展開「自訂特徵函數與客製化 Key」立下了最完美的緩坡鋪墊。

````python
# 範例 12.2.6：key=str.lower 實現大小寫不敏感排序
words = ["banana", "Apple", "cat", "Avocado", "Blueberry", "ant"]

print("1. 原始清單：", words)

# 比較 A：標準 ASCII 排序（大寫全部衝到最前面）
ascii_sorted = sorted(words)
print("2. 預設 ASCII 排序（區分大小寫）：", ascii_sorted)

# 比較 B：使用 key=str.lower 進行大小寫不敏感排序
case_insensitive_sorted = sorted(words, key=str.lower)
print("3. key=str.lower 排序（字典序真實順序）：", case_insensitive_sorted)

# 驗證：原始字串的大小寫完整被保留下來！
````

在本單元中，你征服了程式競賽中最容易讓人陰溝裡翻船的諸多順序陷阱：
- **降序排列工具**：
  - 透過 `reverse=True` 參數，`sort()` 與 `sorted()` 皆可瞬間倒轉為由大到小降序排列。
- **字串字典序（Lexicographical Order）核心**：
  - 逐字元比對 ASCII 編碼，一旦在前方位數分出勝負，後方字元完全不影響結果。
  - 大寫字母（65~90）永遠排在小寫字母（97~122）之前。
- **考場世紀大黑洞防禦**：
  - 字串 `'100'` 比 `'2'` 小！讀入數字後務必使用 `int()` 轉型，切勿直接對文字數字排序。
- **負號取反技巧 `-x`**：
  - 利用純數學正負倒影，在不更動升序引擎的情況下實現數值降序，為後續多準則排序立下奇兵。
- **大小寫不敏感排序初探**：
  - 透過 `key=str.lower`（不加小括號），以透視特徵值比大小，同時保留原始字串大小寫。
