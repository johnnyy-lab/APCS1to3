### 10.3.1 為什麼需要字典？從「數值索引」邁向「任意名牌索引」的認知躍升

在過去學習一維串列（List）時，我們存取任何資料，都必須依賴「整數編號」：`a[0]`, `a[1]`, `a[2]`。
這就像學校的置物櫃，每間隔間外面都只標著 `0 號`, `1 號`, `2 號`……

#### 😫 串列在真實查詢中的痛苦：
想像全班有 50 位同學，你想查詢「Alice」的數學成績：
- 如果用串列存放：你必須先把所有人名存一個串列 `names`、成績存另一個串列 `scores`。
- 然後你得用一個 `for` 迴圈從第 0 號一個個看過去，直到看到 `"Alice"` 停下來（$O(N)$ 線性搜尋）。
- 如果班上有 100 萬人，查詢一次就要找 100 萬步！這在 APCS 考場上會直接超時（TLE）！

#### 📖 救星登場：真實世界的查字典（Dictionary）
現實生活中，當你想查英文單字 `apple` 的中文時，你會從字典第 1 頁一頁頁翻嗎？當然不會！
你會直接翻到開頭為 `a` 的詞條，**用單字本身作為索引**，瞬間查到意思！

在 Python 中，這種以「**文字或任意資料本身作為專屬名牌**」的容器，就叫做 **字典（Dictionary）**！
- 字典是由一個個 **鍵值對（Key-Value Pair）** 組成的。
- **鍵（Key，名牌 / 鑰匙）**：例如 `"Alice"`、`"apple"`、`"0912-345678"`。
- **值（Value，內容物 / 寶物）**：名牌所對應的資料，例如 `95`（成績）、`"蘋果"`（解釋）。

有了字典，我們不再需要記住「Alice 是幾號櫃子」，只要寫 `scores["Alice"]`，電腦就會以極速 $O(1)$ 瞬間取出 `95`！


````python
# 範例 10.3.1：串列查詢 vs 字典極速直達對比

# 1. 傳統串列做法：需要維護雙串列並以迴圈比對
names = ["Alice", "Bob", "Charlie"]
scores = [95, 88, 92]

target = "Charlie"
charlie_score = None
for i in range(len(names)):
    if names[i] == target:
        charlie_score = scores[i]
        break

print("串列迴圈搜尋結果：", target, "的成績是", charlie_score)

# 2. 現代字典做法：名牌直接對應內容物！
score_dict = {
    "Alice": 95,
    "Bob": 88,
    "Charlie": 92
}

print()
print("字典直接存取結果：", target, "的成績是", score_dict[target])
print("字典完整結構：", score_dict)
````

### 10.3.2 大括號鍵值對語法 `{key: value}` 與空字典建立 `{}`

在 Python 中，字典的標誌性外觀是**大括號 `{...}`**。
其基本定義語法為：
```python
my_dict = {
    key1: value1,
    key2: value2,
    key3: value3
}
```
- **冒號 `:`**：連接在「名牌（Key）」與「內容物（Value）」之間。
- **逗號 `,`**：分隔不同的鍵值對組合。
- **空字典**：最直接且最常用的方式就是一對空大括號 `d = {}`，也可以使用型態建構子 `d = dict()`。

#### ⚠️ 關鍵觀念：鍵的唯一性（Uniqueness）與覆蓋特性
在一個字典中，**鍵（Key）是絕對唯一的！** 不可能同時存在兩個完全相同的名牌！
如果你在宣告或賦值時，不小心寫入了兩個相同的 Key：
```python
grades = {"Alice": 80, "Alice": 95}
```
Python 不會報錯，但**後面的鍵值對會毫不留情地將前面的值「覆蓋抹除」**！最終字典裡只有 `{"Alice": 95}`，長度 `len(grades)` 為 1！
這一點與真實生活完全一致：同一個抽屜只能貼一張專屬名牌，貼上新標籤，舊紀錄就被取代。


````python
# 範例 10.3.2：字典建立語法、空字典與重複鍵覆蓋現象

# 1. 建立空字典
d_empty = {}
print("空字典：", d_empty, "，型態：", type(d_empty), "，長度：", len(d_empty))

# 2. 標準多鍵值對字典
fruit_prices = {
    "apple": 30,
    "banana": 15,
    "cherry": 50
}
print("水果價目表：", fruit_prices, "，共有幾種水果：", len(fruit_prices))

# 3. 重複鍵的覆蓋實驗
duplicate_demo = {
    "A": 10,
    "B": 20,
    "A": 99  # "A" 重複出現！99 會覆蓋 10
}
print()
print("重複鍵覆蓋後的字典：", duplicate_demo)
print("長度只有：", len(duplicate_demo), "（'A' 只被計算一次！）")
````

### 10.3.3 鍵的合法型態限制（Key 必須是 Hashable：整數、字串、Tuple，嚴禁 List）

現在我們來到字典最關鍵的底層規則：**到底什麼東西可以當作字典的「名牌（Key）」？**

在上一節 10-2 中，我們花了整整一節學習「**可雜湊性（Hashability）**」。現在這個伏筆正式派上用場了！
因為字典底層仰賴雜湊表（Hash Table）來實現超光速查詢，所以：
**只有可雜湊（Hashable）的不可變物件，才有資格成為字典的 Key！**

#### 🔑 合法的 Key 種類（通行無阻）：
1. **整數與浮點數**：`{1: "One", 2: "Two"}`
2. **文字字串**：`{"Alice": 95, "Bob": 88}`
3. **不可變元組（Tuple）**：`{(0, 1): "地雷", (2, 3): "金幣"}`（超級強大！）

#### ❌ 違法的 Key 種類（絕對死刑）：
**串列（List）絕對不能當 Key！**
如果你嘗試寫出：
```python
bad_dict = {[1, 2]: "座標點"}
```
Python 直譯器會立刻拋出極其嚴厲的報錯：
`TypeError: unhashable type: 'list'`！
因為串列隨時會被修改，一旦內容變動，雜湊值就徹底迷失，字典就找不到東西了！

至於**值（Value）**呢？Value 完全沒有任何型態限制！你要在 Value 裡面放整數、字串、串列、甚至嵌套另一個字典，全部 100% 合法自由！


````python
# 範例 10.3.3：各類合法 Key 示範 vs 串列作 Key 的報錯攔截

# 1. 字串與整數作為 Key
user_ages = {"David": 18, 101: 20}
print("字串與整數鍵：", user_ages)

# 2. 元組作為 Key（APCS 網格座標標記必殺技！）
grid_map = {
    (0, 0): "起點",
    (1, 2): "寶箱",
    (3, 4): "終點"
}
target_coord = (1, 2)
print()
print("座標元組鍵：", grid_map)
print(f"座標 {target_coord} 上的物體是：{grid_map[target_coord]}")

# 3. 捕捉串列作為 Key 的 TypeError 崩潰
try:
    illegal_dict = {}
    illegal_key = [1, 2]
    illegal_dict[illegal_key] = "Crash!"
except TypeError as e:
    print()
    print("❌ 攔截到錯誤：", e)
    print("說明：串列是可變物件，絕對不可作為字典的 Key！")
````

````python
# 練習 10.3.3：網格寶物定位查詢機
# 題目說明：輸入三行。
# 前兩行每行包含三個數值：列座標 r（整數）、行座標 c（整數）、寶物代號 item（字串）。
# 請將兩筆資料以元組 (r, c) 為 Key、寶物代號為 Value 存入字典 treasure_map。
# 第三行輸入一個查詢座標 qr, qc（兩個整數）。
# 輸出該座標上的寶物代號（題目保證查詢座標一定存在）。

# 【公開測試資料 1】
# 0 1 GOLD
# 2 3 GEM
# 0 1
# 輸出：
# GOLD

# 【公開測試資料 2】
# 1 1 SWORD
# 3 4 SHIELD
# 3 4
# 輸出：
# SHIELD

# 請在此處撰寫你的程式碼：
p1 = input().split()
p2 = input().split()
qr, qc = map(int, input().split())

treasure_map = {
    (int(p1[0]), int(p1[1])): p1[2],
    (int(p2[0]), int(p2[1])): p2[2]
}

query_pt = (qr, qc)
print(treasure_map[query_pt])
````

````python
# 挑戰 10.3.3：座標點重疊價值累加器
# 題目說明：輸入一個整數 N。
# 接著有 N 行，每行輸入兩個整數 r, c 代表點位座標，以及一個整數 val 代表該點數值。
# 請以 (r, c) 為鍵建立字典。若座標點重複出現，請將數值「累加」上去！
# 最後輸入查詢座標 qr, qc，輸出該點的總數值；若無該點輸出 0。
# 本題無公開測試資料，請自行測試。

# 請在此處撰寫你的程式碼：
n = int(input())
m = {}
for _ in range(n):
    r, c, v = map(int, input().split())
    pt = (r, c)
    if pt in m:
        m[pt] += v
    else:
        m[pt] = v

qr, qc = map(int, input().split())
q_pt = (qr, qc)
if q_pt in m:
    print(m[q_pt])
else:
    print(0)
````

### 10.3.4 鍵值存取 `d[key]` 與未定義鍵引發 `KeyError` 的致命陷阱

使用中括號 `d[key]` 存取資料非常直觀，就像串列的索引一樣。但請特別當心：**這是初學者在 APCS 考場最常陣亡的地方！**

#### 💥 致命陷阱：KeyError
在串列中，如果索引超出範圍，會引發 `IndexError`；
而在字典中，**如果中括號裡的 Key 根本不存在於字典中**，Python 直譯器會立刻拋出中斷錯誤：
```python
menu = {"tea": 30}
print(menu["coffee"])  # 💥 崩潰！KeyError: 'coffee'
```
一旦引發 `KeyError`，你的程式就會直接死機，在線上評判系統（OJ）上被判定為 **`RE`（Runtime Error，執行時期錯誤）**，整題直接抱鴨蛋！

#### 🛡️ 防禦心法：先用 `in` 照一下！
在不知道該 Key 是否存在的情況下，絕對不要冒然使用中括號硬取！
最經典的防呆方式，就是利用 **`in` 運算子** 進行存在性檢查：
```python
if "coffee" in menu:
    print(menu["coffee"])
else:
    print("今日無販售咖啡！")
```
養成「**先查驗、再存取**」的防禦性編程習慣，是寫出零崩潰競賽程式的基礎修養！


````python
# 範例 10.3.4：KeyError 崩潰重現與 if in 安全防禦

inventory = {
    "sword": 1,
    "shield": 2,
    "potion": 5
}

print("當前背包內容：", inventory)

# 1. 存取存在的鍵：順暢成功
print("水水數量：", inventory["potion"])

# 2. 存取不存在的鍵（若解除註解會直接引發 KeyError 崩潰）：
# print(inventory["bow"])  # KeyError: 'bow'

# 3. 安全防禦標準寫法：使用 in 運算子預先判斷
query_item = "bow"

print()
if query_item in inventory:
    print(f"找到裝備 {query_item}，數量為：{inventory[query_item]}")
else:
    print(f"背包中找不到裝備：{query_item}（安全避開 KeyError！）")
````

````python
# 練習 10.3.4：安全圖書館借閱查詢機
# 題目說明：第一行輸入一個整數 N 代表館藏書籍數量。
# 接著有 N 行，每行兩個字串：書名 與 借閱者姓名。
# 請將資料存入字典 books（書名為 Key，借閱者為 Value）。
# 最後一行輸入欲查詢的書名 query_book。
# 若該書在館藏中，輸出借閱者姓名；若不在館藏中，輸出 "NOT FOUND"。

# 【公開測試資料 1】
# 2
# Python Alice
# Algorithm Bob
# Python
# 輸出：
# Alice

# 【公開測試資料 2】
# 2
# Math Charlie
# Science David
# English
# 輸出：
# NOT FOUND

# 請在此處撰寫你的程式碼：
n = int(input())
books = {}
for _ in range(n):
    title, borrower = input().split()
    books[title] = borrower

query_book = input().strip()

if query_book in books:
    print(books[query_book])
else:
    print("NOT FOUND")
````

### 10.3.5 新增與修改鍵值對：`d[key] = new_val` 的就地更新機制

在串列中，要新增元素必須用 `.append()`，要修改元素必須用 `a[i] = val`，增與改的語法完全不同。
但在字典中，**新增與修改的語法是完全合而為一的！**
```python
d[key] = new_val
```

#### 🔄 Python 的自動雙向判斷邏輯：
當你執行這行賦值指令時，Python 會在內部進行自動分流：
1. **若 `key` 原本「不存在」** ➔ 自動在字典中**建立全新鍵值對**！
2. **若 `key` 原本「已經存在」** ➔ **就地覆蓋修改**該鍵原本所對應的值！

#### ⚠️ 常見避坑：`d[key] += 1` 的初值天坑！
在做頻率統計（例如算投票數）時，大家最喜歡寫 `counts[name] += 1`。
但是請注意！`counts[name] += 1` 的等價展開是：
`counts[name] = counts[name] + 1`！
如果 `name` 是第一次出現，等號右邊在嘗試讀取 `counts[name]` 的那一瞬間，就會因為找不到而**立即引發 `KeyError` 崩潰**！
因此，標準的就地累加寫法必須搭配條件判斷：
```python
if name not in counts:
    counts[name] = 1  # 第一次出現，初值設為 1
else:
    counts[name] += 1  # 已經出現過，安心累加 1
```
掌握這個邏輯，你就掌握了 APCS 中最核心的頻率計數器原型！


````python
# 範例 10.3.5：動態新增、就地修改與安全頻率累加實戰

stock = {"apple": 10, "banana": 20}
print("初始庫存：", stock)

# 1. 修改已存在的鍵
stock["apple"] = 15
print("修改 apple 庫存後：", stock)

# 2. 新增原本不存在的鍵
stock["orange"] = 30
print("新增 orange 庫存後：", stock)

# 3. 實戰：投票計票器（初值安全防禦模式）
votes = ["Alice", "Bob", "Alice", "Charlie", "Alice", "Bob"]
vote_counts = {}

print()
print("--- 逐票開票歷程 ---")
for person in votes:
    if person not in vote_counts:
        vote_counts[person] = 1  # 第一次得票，初值為 1
    else:
        vote_counts[person] += 1  # 再次得票，累加 1
    print(f"投給 {person:<7} ➔ 當前計票狀態：{vote_counts}")
````

````python
# 練習 10.3.5：動態商品庫存管理員
# 題目說明：輸入一個整數 N 代表操作次數。
# 接著有 N 行，每行包含：商品名稱（字串）與進出貨數量（整數，正數代表進貨，負數代表出貨）。
# 請使用字典維護各商品的最終庫存量（初始庫存皆為 0）。
# 題目保證所有操作結束後，最後一行會輸入一個欲查詢的商品名稱。
# 請輸出該商品的最終庫存量（若該商品從未出現過，輸出 0）。

# 【公開測試資料 1】
# 3
# Pen 10
# Pen -3
# Book 5
# Pen
# 輸出：
# 7

# 【公開測試資料 2】
# 2
# Apple 50
# Banana 20
# Orange
# 輸出：
# 0

# 請在此處撰寫你的程式碼：
n = int(input())
inv = {}

for _ in range(n):
    name, qty = input().split()
    qty = int(qty)
    if name not in inv:
        inv[name] = qty
    else:
        inv[name] += qty

q = input().strip()
if q in inv:
    print(inv[q])
else:
    print(0)
````

### 10.3.6 刪除鍵值對：`del d[key]` 與 `.pop(key, default)` 安全彈出

當某筆資料過期、或在模擬遊戲中某個角色死亡時，我們需要將對應的鍵值對從字典中移除。在 Python 中，主要有兩大刪除利器：

#### 1. 暴力刪除指令：`del d[key]`
- 語法：`del my_dict["Alice"]`
- 特性：乾脆俐落直接刪除，不回傳任何值。
- 致命風險：如果欲刪除的 `key` 根本不存在，會**立即觸發 `KeyError` 崩潰報錯**！

#### 2. 安全彈出方法：`d.pop(key, default)`（強烈推薦！）
- 語法：`val = my_dict.pop("Alice")`
- 特性：不僅把資料從字典中移除，還能**順便將被刪除的值回傳給變數**！
- 🛡️ **帶預設值的終極防禦**：
  若寫 `val = my_dict.pop(key, None)` 或 `val = my_dict.pop(key, -1)`：
  - 如果 key 存在：成功移除並回傳原本的值。
  - **如果 key 不存在：完全不會報錯，而是默默回傳你設定的預設值！**

#### 3. 一鍵全部清空：`d.clear()`
- 將字典裡的所有內容一次清空為 `{}`，但保留字典物件本體。


````python
# 範例 10.3.6：del 指令 vs pop() 安全彈出實務對比

scores = {"Alice": 95, "Bob": 88, "Charlie": 92}
print("原始成績單：", scores)

# 1. 使用 del 刪除鍵值對
del scores["Bob"]
print("del 刪除 Bob 後：", scores)

# 2. 使用 pop() 提取並刪除
charlie_val = scores.pop("Charlie")
print(f"pop() 提取出 Charlie 的分數 {charlie_val}，剩餘字典：", scores)

# 3. 測試安全防禦 pop(key, default)
# 嘗試刪除不存在的 "David"
david_val = scores.pop("David", "不存在無須刪除")
print()
print("安全彈出結果：", david_val)
print("字典依然完好如初：", scores)

# 4. 一鍵清空
scores.clear()
print("clear() 清空後：", scores)
````

````python
# 練習 10.3.6：會員登出與註銷系統
# 題目說明：第一行輸入一個整數 N 代表現有會員數量。
# 接著有 N 行，每行包含：會員帳號（字串）與點數（整數）。
# 請建立字典 members。
# 隨後輸入一個欲註銷的帳號 remove_id。
# 請使用 pop(remove_id, -1) 嘗試註銷該會員：
# 若該會員存在，輸出其結算點數；
# 若該會員不存在，輸出 -1。
# 最後一行輸出註銷後剩餘的會員總人數 len(members)。

# 【公開測試資料 1】
# 3
# userA 100
# userB 250
# userC 50
# userB
# 輸出：
# 250
# 2

# 【公開測試資料 2】
# 2
# vip1 999
# vip2 888
# guest
# 輸出：
# -1
# 2

# 請在此處撰寫你的程式碼：
n = int(input())
members = {}
for _ in range(n):
    acc, pts = input().split()
    members[acc] = int(pts)

remove_id = input().strip()
val = members.pop(remove_id, -1)

print(val)
print(len(members))
````

## 10.3 學習總結與核心技能檢核

在單元 10-3 中，我們正式進入了 Python 雜湊容器的核心殿堂——**字典（Dictionary）**：

1. **從數值索引到任意名牌索引**：
   - 字典以鍵值對（Key-Value）組織資料，實現了跳脫整數限制的直觀直達查詢。
2. **大括號語法與鍵唯一性**：
   - 使用 `{k: v}` 定義，空字典為 `{}`；字典中同一個 Key 絕對唯一，重複給值會被就地覆蓋。
3. **Key 的可雜湊性（Hashable）限制**：
   - 只有不可變物件（int, str, tuple）能當 Key，串列（List）絕對嚴禁作為 Key！
4. **防禦 KeyError 致命崩潰**：
   - 存取未定義的 Key 會直接噴出 `KeyError`，務必先用 `if key in d:` 進行安全驗證。
5. **增改合一機制**：
   - `d[key] = val` 同時具備「不存在則新增、已存在則覆蓋」的雙重能力；做頻率計數時務必注意第一次出現的初值處理。
6. **優雅安全刪除**：
   - 優先使用帶預設值的 `d.pop(key, default)`，兼具「提取舊值」與「防報錯避坑」雙重優勢！
