### 🛑 6-4-1 break：緊急煞車踏板——滿足特定條件立刻終止並跳出迴圈

**生活故事：流水線瑕疵檢驗與找鑰匙**  
想像工廠的高速輸送帶正在源源不絕地運送罐頭，品質檢驗員只要一發現某個罐頭「嚴重凹陷發霉」，就會立刻拍下紅色的「緊急停車鈕（E-STOP）」，整條產線立刻瞬間停擺，絕對不會繼續讓後面 1000 個罐頭繼續白白通過；  
又例如你在口袋裡摸鑰匙，你的目標是「找到鑰匙開門」，只要指尖摸到了鑰匙，你就會立刻把手抽出來，絕不會傻傻地繼續把口袋裡剩下的發票、零錢、衛生紙全部摸過一遍！

**運作機制——瞬間切斷迴圈：**  
在 Python 迴圈中，**`break`（中斷 / 煞車）** 關鍵字扮演的就是這顆緊急煞車鈕！  
它的執行規則極端霸道且不可阻擋：
- 當直譯器在迴圈內部執行到 `break` 這一行時，當前迴圈會**「立刻被強制殺死」**！
- 本輪在 `break` 之後的所有指令被全部捨棄；
- 後續所有原本還沒跑的剩餘回合也**全部被立刻取消**；
- 程式的控制權會瞬間**「跳出迴圈」**，直接降落到迴圈縮排外面的下一行繼續執行！

通常 `break` 都會緊密搭配 `if` 條件式使用：`if 某個極端條件成立: break`。如果沒有 `if` 守護，只要程式一碰到無條件的 `break`，迴圈就會在跑完第一輪後當場斃命！

**常見錯誤陷阱：**  
- **單純寫 break 忘記放進 if**：直接寫在迴圈體內，導致迴圈永遠只跑 1 次就退場。
- **以為 break 能跳出多層迴圈**：請注意，在雙重或多重巢狀迴圈中，一個 `break` **「只會跳出它身處的那一層迴圈」**，外層的迴圈依然會繼續運轉！

**APCS 考試實務建議：**  
在 APCS 評測系統中，時間限制通常是極其嚴格的 1 秒鐘（TLE 警戒線）。只要找到答案或目標，果斷使用 `break` 提前停機，能為你的程式省下高達數百萬次的無謂計算，是化解超時危機的救命神器！


````python
# [2] Code 範例區：體驗 break 的緊急煞車機制
print("=== 開始尋找第一個能被 7 整除的雙位數 ===")

for num in range(10, 100):
    print(f"檢查中：{num}")
    if num % 7 == 0:
        print(f"--> 大功告成！找到目標數字：{num}，立刻緊急煞車！")
        break  # 強制跳出迴圈，15 以後的數字完全不看！

print("=== 迴圈已終止，主程式繼續前進 ===")
print(f"最後鎖定的 num 值為：{num}")
````

````python
# ==========================================
# [4] Code 練習題
# 任務說明：
# 給定目標幸運數字 target = 37。
# 請使用 for 迴圈走訪 range(1, 101)：
# 每一輪檢查當前數字 i 是否等於 target：
# 若等於，將布林變數 found 設為 True，記錄尋獲時的步數 steps = i，並立刻 break！
#
# 【公開測試資料 1】
# 設定：target = 37
# 預期：found = True, steps = 37
#
# 【公開測試資料 2】
# 設定：target = 150 (不在 1~100 範圍內)
# 預期：found = False, steps = 0
# ==========================================
# 請在下方撰寫你的程式碼並執行測試：

target = 37
found = False
steps = 0

for i in range(1, 101):
    if i == target:
        found = True
        steps = i
        break

print(f"搜尋結果：found={found}, steps={steps}")

assert found is True and steps == 37, "公開測資 1 驗證失敗"
print("公開測資 1 通過！")
````

### ⏭️ 6-4-2 continue：換下一個——跳過當前回合剩餘指令，直奔下一輪

**生活故事：跨欄運動與機場綠色通道免檢**  
想像你在操場進行 110 公尺跨欄跑，當你跑到柵欄前，你的動作是「奮力一跳，跨過障礙」，然後腳步不停歇，**繼續朝下一個欄架奔馳**！你絕對不會因為眼前有個障礙，就直接脫下釘鞋棄賽回家（那是 `break`）；  
又例如國際海關安檢通道，海關人員看到沒有申報物品的旅客，會示意「免檢放行，請走綠色通道」，立刻轉身呼喚下一位旅客上前檢查。

**運作機制——跳過本輪，直奔下輪：**  
如果說 `break` 是「整部機器直接斷電」，那麼 **`continue`（繼續 / 跳步）** 就是「跳過這首歌，直接播下一首」！  
當直譯器在迴圈中碰上 `continue` 時：
- 它會**立刻中斷「當前這一個回合」**；
- 本回合在 `continue` 後方剩下的程式碼，**完全略過不執行**；
- 但是——**整個迴圈「絕對不會結束」！**
- 直譯器會立刻回到迴圈最上方，自動將迴圈變數推向下一個元素，神采奕奕地開始**「下一回合」**！

在數據清洗與過濾篩選的場景中，`continue` 是最優雅的神兵利器。例如「只要遇到偶數就跳過，只印奇數」、「只要遇到負數就略過，只累加正數」。

**常見錯誤陷阱——在 while 迴圈中使用 continue 的世紀大坑！**  
在 `for` 迴圈中使用 `continue` 非常安全，因為 `for` 會自動切換下一個值。  
但是在 `while` 迴圈中，**如果你把 `i += 1`（步進更新）寫在 `continue` 的正後方**，那麼當 `continue` 發動時，`i += 1` 會被殘忍地跳過！這導致 `i` 的值永遠不會改變，程式瞬間失控陷入無窮迴圈！在 `while` 內使用 `continue` 時，務必在執行 `continue` 前先手動推進變數！

**APCS 考試實務建議：**  
善用 `continue` 可以有效減少深層的 `if-else` 巢狀結構。把不符合資格的特例在最上方用 `if ...: continue` 剔除，能讓後續的主要處理邏輯保持平整乾淨！


````python
# [2] Code 範例區：continue 排除特定特例走訪
print("=== 走訪 1 到 10，排除所有 3 的倍數 ===")

for n in range(1, 11):
    if n % 3 == 0:
        print(f"[跳過] 數字 {n} 是 3 的倍數，跳過不處理！")
        continue  # 立刻略過下方印出核心指令，直接進入下一輪！
    
    # 只有通過上方檢查的數字才會執行到這裡
    print(f"--> [合格] 處理優質數字：{n}，其十倍值為 {n * 10}")

print("=== 走訪完畢！ ===")
````

````python
# ==========================================
# [4] Code 練習題
# 任務說明：
# 計算非負數與非五倍數之和：
# 請走訪整數 i 從 1 到 20：
# 1. 若 i 是 5 的倍數，使用 continue 跳過不計；
# 2. 其餘數字累加至 sum_filtered。
# 迴圈結束後，請印出 sum_filtered。
#
# 【公開測試資料 1】
# 範圍 1 到 20，跳過 5, 10, 15, 20
# 1~20 總和為 210，扣除 (5+10+15+20 = 50)
# 預期：sum_filtered = 160
# ==========================================
# 請在下方撰寫你的程式碼並執行測試：

sum_filtered = 0

for i in range(1, 21):
    if i % 5 == 0:
        continue
    sum_filtered += i

print("篩選後總和為：", sum_filtered)

assert sum_filtered == 160, "公開測資 1 驗證失敗"
print("公開測資 1 通過！")
````

### 🔀 6-4-3 break vs continue 流程跳轉路徑視覺化圖解對比

**生活故事：商店提早打烊 vs 客人沒戴口罩**  
讓我們把兩者放在同一個情境下對比，你將會一輩子記住它們的差異：
- **`break` 的場景（商店失火或提早打烊）**：廣播響起「本店發生火災，請所有人立刻撤離！」所有人立刻放下商品衝出大門，整間店直接拉下鐵捲門停止營業，**後面的所有客人完全不准再進來**！
- **`continue` 的場景（防疫管制）**：店員站在門口檢查體溫，排到第三位客人時發現他沒戴口罩，店員說：「先生不好意思，請您離開！」這位客人被拒絕入場（跳過），**但排在第四位、第五位的客人繼續正常上前檢查入場**！

**底層跳轉路徑本質對照表：**  
| 比較維度 | `break` | `continue` |
| :--- | :--- | :--- |
| **白話意義** | 緊急煞車，提早收工不玩了 | 換下一個，跨步跳過這回合 |
| **當前這輪剩下的指令** | 全部放棄，不執行 | 全部放棄，不執行 |
| **後續未跑的所有回合** | **全部取消，完全不跑！** | **繼續正常執行！** |
| **控制權跳轉目標** | **跳出整個迴圈**，去迴圈外面的下一行 | **跳回迴圈頂端**，推進變數開始下一輪 |
| **最常見搭配** | 線性搜尋「找到目標就停」、偵測嚴重異常 | 髒數據清洗、排除無效條件、過濾特例 |

**常見錯誤陷阱：**  
初學者在手寫考卷或快速敲代碼時，常常「腦子想著換下一個，手指卻敲出了 break」！  
這會造成原本只是想略過一筆無效資料，結果整個資料處理流直接被腰斬斷裂，導致線上評判系統輸出答案殘缺不全，痛失大分。

**APCS 考試實務建議：**  
只要你寫下 `break`，請立刻在腦中確認：「我是真的要結束這整個任務了嗎？」若只是「這個特例我不想要」，那百分之百要使用 `continue`！


````python
# [2] Code 範例區：同一數列下 break 與 continue 的行為對決
print("=== 對照實驗：走訪 1 到 6，在 i == 3 時分別觸發 ===")

# 實驗組 A：使用 break
print("\n【實驗 A：觸發 break】")
for i in range(1, 7):
    if i == 3:
        print(f"i = {i}，觸發 break！")
        break
    print(f"  執行中：i = {i}")

# 實驗組 B：使用 continue
print("\n【實驗 B：觸發 continue】")
for i in range(1, 7):
    if i == 3:
        print(f"i = {i}，觸發 continue！")
        continue
    print(f"  執行中：i = {i}")
````

### 🎯 6-4-4 線性搜尋與布林旗標（Flag）：「找到就停」的質數檢驗演算法

**生活故事：驗貨員查驗與質數的孤獨**  
什麼是「質數（Prime Number）」？除了 1 和自己以外，無法被任何其他正整數整除的數字（如 2, 3, 5, 7, 11, 13...）。  
如果有人拿著一個超大數字 $n = 99999999999$ 問你：「請問它是質數嗎？」  
你根本不需要花幾天幾夜把所有因數都算出來！你只要心裡先預設「它是質數」，然後開始從 2, 3, 4... 逐一檢查。只要在第 2 步發現它能被 3 整除，你就拿到了它「不是質數」的鐵證！這時立刻宣告它不是質數，並立刻停機（break）！這就是**「找到就停」**的極致效率。

**運作機制——旗標變數 + 開根號加速 + break：**  
在第五章中我們學過「布林旗標（Flag）」，它與 `break` 是天造地設的黃金搭檔：
1. **起手式：假設為真**  
   先立起一個旗標：`is_prime = True`（若 $n < 2$ 則直接設為 False）。
2. **搜尋因數：地毯式走訪**  
   走訪可能整除 $n$ 的潛在因數 $d$。數學上，因數都是成雙成對出現的（例如 $36 = 4 \times 9$，一個小於等於 6，另一個大於等於 6）。因此我們根本**不需要檢查到 $n-1$，只要檢查到 $\sqrt{n}$（開根號）即可！**
3. **一擊必殺：找到因數立即煞車**  
   一旦 `n % d == 0`，代表抓到了除盡的證據：
   立刻 `is_prime = False`，緊接著觸發 **`break`**！
   剩餘成千上萬個數字完全不用再跑，直接收工！

**常見錯誤陷阱：**  
- **漏判 0 和 1**：0 和 1 既不是質數也不是合數，必須在迴圈前先防禦；
- **忘記開根號加 1**：`range(2, int(n**0.5) + 1)`，因為 range 是「左閉右開」，如果少了 `+ 1`，當檢查 4 或 9 時會剛好漏掉根號本身！

**APCS 考試實務建議：**  
若不用開根號加速與 `break`，檢驗一個 $10^9$ 的數字需要花 10 秒（TLE 爆掉）；加上開根號與 `break`，只需檢查到 31622 次（約 0.001 秒 AC 通過）！這是 APCS 最經典的效能飛躍實例。


````python
# [2] Code 範例區：質數檢驗演算法（開根號加速 + break）
n = 97  # 待檢驗數字
is_prime = True

print(f"=== 啟動質數檢驗：目標數字 {n} ===")

if n < 2:
    is_prime = False
else:
    limit = int(n ** 0.5)  # 數學定理：只需檢查到根號 n
    print(f"根號上限為：{limit}，只需檢查 2 到 {limit}")
    
    for d in range(2, limit + 1):
        if n % d == 0:
            print(f"--> 在 d = {d} 處發現能整除！不是質數！")
            is_prime = False
            break  # 關鍵：找到鐵證，立刻煞車！

if is_prime:
    print(f"恭喜！{n} 確實是一個孤獨而美麗的【質數】！")
else:
    print(f"遺憾！{n} 是【合數】，非質數！")
````

### 🛡️ 6-4-5 continue 在髒資料過濾與跳過無效測資之防守應用

**生活故事：農夫採收草莓與過濾瑕疵品**  
草莓園採收季到了，輸送帶上一顆一顆草莓快速滑過。優良的包裝工人看到表面爛掉、發霉或太小的青色草莓，會一把抓起扔進廚餘桶（這顆跳過不包裝），然後眼睛立刻盯緊下一顆送過來的草莓！  
如果工人因為看到一顆爛草莓就把整座工廠停工關門（那是 `break`），老闆一定會氣得跳腳！面對源源不絕的輸入串流，「壞的踢掉、好的處理」正是 `continue` 最光芒四射的舞台。

**運作機制——衛語句（Guard Clauses）防守哲學：**  
在真實世界的數據處理與 APCS 考題中，輸入資料往往夾雜著雜訊（稱為髒資料，Dirty Data），例如：
- 學生成績輸入為負數（如 `-10`）或超過 100 分（如 `150`）；
- 體重或年齡輸入為 0；
- 異常的無效測資。

如果在處理核心邏輯前，把所有條件用巢狀 `if ... else` 一層包一層，代碼會向右傾斜極難閱讀。優雅高手的寫法是將 `continue` 當成**「安全警衛」**：  
在迴圈體的最上方寫上**「防守衛語句」**：
```python
for data in stream:
    # 警衛把關：非法或無效資料立刻踢出！
    if data < 0 or data > 100:
        continue
    
    # 通過警衛把關的純淨合法資料，在最外層優雅計算！
    valid_total += data
```
這樣能徹底保護後續的運算邏輯，遠離 `ZeroDivisionError` 或邏輯失真！

**常見錯誤陷阱：**  
- **計數器放錯位置**：如果你想統計「總共有幾筆合法資料」，記得把 `valid_count += 1` 放在 `continue` 之後！若放錯在前面，髒資料也會被誤算進去。

**APCS 考試實務建議：**  
當題目敘述寫道：「若輸入為負值則略過不計，只統計正數平均值」時，這就是最標準的 `continue` 召喚訊號！


````python
# [2] Code 範例區：衛語句過濾髒資料與防除零保護
# 模擬逐筆讀入的成績數據（包含負數與超過 100 的異常值）
# 為了零提前依賴串列，我們以多變數模擬數據流
print("=== 啟動成績資料清洗過濾管線 ===")

valid_sum = 0
valid_count = 0

# 模擬 5 筆輸入成績：85, -5, 92, 120, 78
for round_i in range(1, 6):
    # 產生測試成績：依序為 85, -5, 92, 120, 78
    if round_i == 1: score = 85
    elif round_i == 2: score = -5
    elif round_i == 3: score = 92
    elif round_i == 4: score = 120
    else: score = 78
    
    # 衛語句防守：跳過非法成績
    if score < 0 or score > 100:
        print(f"[警報] 第 {round_i} 筆成績 {score} 為非法髒資料，予以剔除！")
        continue
    
    # 合法資料才進入核心處理
    valid_sum += score
    valid_count += 1
    print(f"--> 第 {round_i} 筆成績 {score} 合法，納入統計。")

print("=== 資料清洗完畢 ===")
if valid_count > 0:
    avg = valid_sum / valid_count
    print(f"合法總分：{valid_sum}，有效人數：{valid_count}，平均成績：{avg:.1f}")
````

### 🌟 6-4-6 Python 特色語法：for ... else 與 while ... else（正常走完才執行的語意）

**生活故事：地毯式搜尋失物與「都沒找到」的公告**  
想像你在整間教室尋找遺失的手錶，你翻遍了第一排、第二排……一直翻到最後一排：
- 如果中間任何一個抽屜「找到了手錶」，你會高興地大喊一聲「找到了！」並立刻停下翻找（觸發 `break` 離開）；
- 但如果你把全班所有的座位**「完完整整、一處不漏地全部翻遍了，卻完全沒看見手錶」**，這時你才會走出教室貼出一張公告說：「全班皆已翻遍，確認【都沒找到】！」

這句「確認都沒找到」的話，有一個嚴格的前提——**你必須是在沒有中途被 break 打斷、完整搜遍所有角落後才能說出口！**

**運作機制——Python 獨步天下的迴圈 else 語法：**  
在 C/C++ 或 Java 中沒有這個語法，但在 Python 中，`for` 迴圈與 `while` 迴圈居然可以搭配一個從屬的 `else:` 區塊！  
```python
for 變數 in 序列:
    if 找到目標:
        print("找到了！")
        break
else:
    # 只有當上面的迴圈「正常走完最後一輪」（完全沒碰到 break）時才會執行！
    print("全部走完，都沒找到！")
```
它的觸發鐵律如下：
1. **如果迴圈中途被 `break` 終止** ➔ `else:` 區塊會**直接被略過不執行**！
2. **如果迴圈順利走完全程**（從來沒有被任何 break 煞車過） ➔ 直譯器就會**自動進入並執行 `else:` 區塊**！

這項語法徹底消滅了過去為了記錄「到底有沒有找到」而額外宣告的布林旗標變數，讓「搜尋無果（Not Found）」的邏輯表達得如同神來之筆！

**常見錯誤陷阱：**  
- **縮排對齊錯誤**：初學者最常把這個 `else:` 縮排對齊到了內部的 `if` 上！請務必張大眼睛確認：這個 `else:` 的縮排是與最外層的 `for` 或 `while` **垂直一線齊平的**！

**APCS 考試實務建議：**  
在質數檢驗或線性搜尋題中，熟練使用 `for...else` 可以少寫 2 行旗標宣告，代碼緊湊優雅，充分展現 Pythonic 的編程境界！


````python
# [2] Code 範例區：體驗 for...else 的奇妙魔法
print("=== 實驗 1：搜尋目標存在的情境 ===")
for x in range(1, 6):
    print(f"  檢查數字：{x}")
    if x == 3:
        print("--> 成功尋獲目標 3！觸發 break！")
        break
else:
    print("【else 區塊】全部檢查完畢，沒有找到目標！")
# 注意：因為碰到了 break，上方 else 區塊被完全略過！

print("\n=== 實驗 2：搜尋目標不存在的情境 ===")
for x in range(1, 6):
    print(f"  檢查數字：{x}")
    if x == 99:
        print("--> 找到目標 99！觸發 break！")
        break
else:
    print("【else 區塊】太棒了！1 到 5 全數正常走完，完全沒碰到 break！")
# 注意：因為從未碰到 break，迴圈正常結束後自動執行 else！
````

### 🔄 6-4-7 while True 搭配 break：動態事件驅動與自由出口架構

**生活故事：自動提款機（ATM）與販賣機操作界面**  
當你走到路邊的 ATM 提款機前，螢幕上顯示著歡迎畫面；當你插卡、輸入密碼、查詢餘額、提領現金後，螢幕不會直接當機關閉，而是問你：「您還需要其他服務嗎？」  
如果你按「繼續」，它就跳回選單；直到你按下「退卡離開」，ATM 才會退出主循環並吐出卡片。  
這種「先運作、隨時依據用戶動態輸入或內部狀態，自由決定何時離開」的架構，在程式設計中被稱為**「事件驅動循環（Event Loop）」**。

**運作機制——自由出口架構（Free Exit Pattern）：**  
傳統的 `while 條件:` 嚴格規定「出口必須在迴圈的最開頭」。但很多時候，我們必須「先做完一部分運算，拿到結果之後，才能判斷要不要停機」！  
這時，專業工程師與競賽選手最常採用的架構就是：
```python
while True:
    # 步驟 1：執行運算或讀取輸入
    ...
    
    # 步驟 2：在迴圈的中間或任何位置，依據條件決定是否退場
    if 退出條件成立:
        break
    
    # 步驟 3：後續繼續處理的運算
    ...
```
`while True` 建立了一個永不熄火的引擎，而內部的 `if ...: break` 則成為了可以隨意擺放在任何位置的**「任意門 / 自由出口」**！

**常見錯誤陷阱：**  
- **忘記提供 break 出口**：寫了 `while True:` 卻完全沒寫任何 `break`，或者 `break` 身處在一個永遠不可能達成的條件式裡，這會引發真正的無窮迴圈死結！

**APCS 考試實務建議：**  
在 APCS 考試中，有大量題目要求「持續讀入整數，直到輸入為 0 時停止計算」（哨兵值題型）。使用 `while True` 搭配 `if n == 0: break` 是最直觀、代碼行數最少、且最不容易出錯的考場速解模板！


````python
# [2] Code 範例區：while True 搭配 break 模擬哨兵值讀取
# 模擬輸入串流：50, 30, 20, 0（0 代表結束標記 Sentinel）
print("=== 啟動購物車計費收銀機（輸入 0 結算） ===")

total_bill = 0
item_idx = 1

while True:
    # 模擬逐筆輸入價格（依序為 50, 30, 20, 0）
    if item_idx == 1: price = 50
    elif item_idx == 2: price = 30
    elif item_idx == 3: price = 20
    else: price = 0
    
    print(f"輸入第 {item_idx} 件商品價格：{price}")
    
    # 核心自由出口：若價格為 0，代表輸入結束，立刻 break！
    if price == 0:
        print("[哨兵訊號觸發] 偵測到 0，停止計費，結算帳單！")
        break
    
    total_bill += price
    item_idx += 1

print(f"=== 最終結帳總金額：{total_bill} 元 ===")
````

````python
# ==========================================
# [4] Code 練習題
# 任務說明：
# 模擬未知步數的倍數逼近：
# 從 val = 1 出發，在 while True 迴圈中，
# 每輪將 val 乘以 2（val *= 2），並將步數 steps 增加 1。
# 當 val 大於等於 1000 時，立刻 break 退出。
# 請計算出達到 1000 以上時的最小 2 的次方數 val 與所需步數 steps。
#
# 【公開測試資料 1】
# 起點 val = 1
# 2^10 = 1024 >= 1000
# 預期：val = 1024, steps = 10
# ==========================================
# 請在下方撰寫你的程式碼並執行測試：

val = 1
steps = 0

while True:
    val *= 2
    steps += 1
    if val >= 1000:
        break

print(f"經過 {steps} 步，數值達到 {val}（首次大於等於 1000）")

assert val == 1024 and steps == 10, "公開測資 1 驗證失敗"
print("公開測資 1 通過！")
````

### ⚠️ 6-4-8 考場手滑陷阱：誤在 if 內放錯縮排導致只跑第 1 輪就退出的常見 bug

**生活故事：只查了一張證件就關門的警衛**  
想像大考考場門口排著長長的人龍，監考老師站在門口檢查准考證。  
正確的做法是：「如果第一個學生准考證合格，放行；如果第一個學生准考證有問題，要求補辦；**無論如何，接著檢查第二個、第三個學生！**」  
但是如果監考老師手滑犯糊塗，檢查完第一個合格學生後，順手就「把考場大門反鎖並宣布檢查完畢」，後面上百名排隊的考生全都被擋在門外！

**運作機制——慘絕人寰的「縮排手滑」大 Bug：**  
在 Python 中，縮排不僅僅是排版，它**實質決定了程式的從屬邏輯結構**！  
初學者在寫搜尋迴圈時，最常犯的崩潰錯誤如下：
```python
# 致命錯誤示範：縮排手滑！
for i in range(1, 100):
    if i == target:
        print("找到了！")
        break
    else:
        # 手滑大災難：誤把 break 寫在 else 裡面！
        break  # 第一輪如果不是 target，程式就直接在此殉職！
```
請仔細追蹤直譯器的執行路徑：
- 當 $i = 1$ 時，`1 == target` 為 False；
- 程式進入 `else:` 分支，碰到了裡面的 `break`！
- 迴圈當場停止，直接跳出！
結果：這個原本要檢查 100 次的迴圈，**居然「只執行了第 1 輪」就草草結束**！初學者還百思不得其解：「我明明寫了 range(1, 100)，為什麼它只跑一次就不跑了？」

**APCS 考試實務防禦 SOP（考場除錯三問）：**  
1. **輸出跑了嗎？** 如果發現迴圈明明有資料，卻好像只印出第 1 筆，請立刻檢查是不是在 `else` 或無條件處放了 `break`！
2. **縮排對齊了嗎？** `break` 與 `continue` 永遠必須嚴格受控於明確的 `if` 條件式內部！
3. **單步手動追蹤（Trace）**：拿 $i = 1, i = 2$ 在腦海或草稿紙上走一次，確保第 1 輪不匹配時能順利推進到第 2 輪！


````python
# [2] Code 範例區：抓出「只跑一輪就猝死」的縮排 Bug
target = 5
print(f"=== 尋找目標數字：{target} ===")

# 錯誤示範（註解展示邏輯）：
# for x in range(1, 10):
#     if x == target:
#         print("找到了")
#         break
#     else:
#         break  # 致命手滑！x=1 不是 5 就立刻退場了！

# 正確示範：else 絕不能無腦放 break！
found = False
for x in range(1, 10):
    print(f"正常檢查第 {x} 號元素...")
    if x == target:
        print(f"--> 成功鎖定目標 {x}！")
        found = True
        break
    # 注意：如果不等於 target，什麼都不要做，讓迴圈自然推進！

print(f"檢驗結果：是否成功找到？{found}")
````

## 🏁 單元 6-4 總結與 APCS 解題心法回顧

太精彩了！恭喜你完成了迴圈流程控制的最強進階修練！讓我們回顧這八大核心心法：

1. **`break` 緊急煞車**：
   - 條件滿足立刻終止並跳出當前迴圈，大幅縮短運算時間，避免 TLE。
2. **`continue` 跨步跳輪**：
   - 跳過當前回合剩餘指令，直奔下一輪，是數據清洗與衛語句過濾的最佳利器。
3. **`break` vs `continue` 底層差異**：
   - `break` 射向迴圈外部，後續所有回合全部取消；`continue` 射向迴圈頂端，後續回合照常運行。
4. **質數檢驗標準模板**：
   - 布林旗標 + 開根號加速（$\sqrt{N}$）+ 找到因數立即 `break`，達到 $O(\sqrt{N})$ 高效解。
5. **衛語句防守**：
   - 在迴圈頂端優先排除髒資料，維持核心運算結構扁平優雅。
6. **`for...else` / `while...else` 正常結束語法**：
   - 只有當迴圈「完全沒碰到 break 正常走完全程」時才執行 else，優雅表達「搜尋無果」。
7. **`while True` 自由出口架構**：
   - 打造事件驅動引擎，將出口設置在迴圈中段或末尾，是處理哨兵值輸入的最強模板。
8. **防範縮排手滑**：
   - 警惕誤把 break 放在 else 內導致只跑一輪就猝死的隱形 Bug。
