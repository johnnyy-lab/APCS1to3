### ➕ 1-4-1 複合加法賦值（+=）——神奇的快速加料按鈕

在之前的單元中，如果我們想把箱子裡的數字增加 10，我們必須寫：
```python
x = x + 10
```
吳邦一老師在講義第 6 頁特別解釋：類似 `x = x + 10` 這樣的指令，可以把它直接解釋成「**將 10 加進去 x 中**」。

因為在寫程式與 APCS 解題時，「把某個數值加進現有變數」這件事實在太常發生了（例如遊戲得分、跑步圈數累加、商品總價加總）！為了讓程式碼更簡潔俐落，Python 提供了一個超棒的縮寫法：
```python
x += 10
```

你可以把 `+=` 想像成手搖飲店的「**快速加珍珠按鈕**」！
你不需要把原本那杯奶茶倒出來、重新加料、再倒回杯子；只要對著杯子按下 `+= 10` 按鈕，10 顆黑糖珍珠就會自動撲通撲通掉進原本的杯子裡，跟原本的茶融為一體！

請注意：`+=` 這兩個符號必須緊緊黏在一起，中間絕對不能加空格喔！

````python
# [2] Code 範例區：使用 += 進行自我累加
score = 50
print(score)  # 原本分數：50

score += 10   # 相當於 score = score + 10
print(score)  # 加料後分數：60
````

### ➖ 1-4-2 複合減法賦值（-=）——受到攻擊扣血與右側先算鐵則

學會了快速加珍珠的 `+=`，那反過來「扣除、扣血、花費金幣」該怎麼辦呢？沒錯，就是 `x -= 5`！這完全等同於 `x = x - 5`。

但是，這裡隱藏著一個國中初學者最常跌倒的超級大陷阱！請看吳邦一老師在講義第 6 頁舉出的經典範例：
```python
x = 4
y = 5
x -= 3 + y
```
請問執行完畢後，`x` 的值會變成多少呢？

很多同學會直覺地在心裡心算：「$x$ 先減去 3，再加上 $y$ 嗎？」
**大錯特錯！** 請千萬不要忘記我們在 1-2 學過的黃金鐵律：**等號（包括所有複合等號）永遠先算右邊！**

直譯器小助手的真正解讀方式是：
1. **第一步（右側先算乾淨）**：先看右側算式 `3 + y`，因為 $y = 5$，所以右側是 $3 + 5 = 8$！
2. **第二步（整體一次扣除）**：整行代碼其實等價於：
   `x = x - (3 + y)` 也就是 `x = 4 - 8 = -4`！

你可以把這想像成 RPG 冒險遊戲中的傷害結算：
怪物對你同時發動了「物理攻擊 3 點」與「魔法攻擊 y 點」。你的英雄護盾會先在怪物手上把兩種傷害加總為 8 點，然後一口氣從你的生命值 $x$ 裡面扣除 8 點！
牢記右側相當於自帶一個「隱形大括號 `( )`」，就能保證你在 APCS 考試中永遠不會算錯！

````python
# [2] Code 範例區：展示 -= 以及右側算式先完整加總再扣除的過程
x = 4
y = 5

# 相當於 x = x - (3 + y) = 4 - 8 = -4
x -= 3 + y
print(x)
````

### ✖️ 1-4-3 複合乘法賦值（*=）——攻擊力爆發與倍數升級

第三個登場的超人招式是「**複合乘法賦值（`*=`）**」！
`x *= 2` 的意思就是 `x = x * 2`。

這就像是超級瑪利歐吃到了超級蘑菇，或者是七龍珠的角色發動界王拳，戰鬥力在瞬間「**翻倍狂飆**」！

讓我們來看看吳邦一老師在講義第 6 頁的精彩進階例子：
```python
x = 4
x *= 2 * (3 + 10)
```
這裡同樣徹底貫徹了「**右邊先算**」的法則！
1. 電腦先算右邊的乘法算式：括號裡的 $3 + 10 = 13$，接著 $2 \times 13 = 26$。
2. 右邊完全算好得出 26 後，才啟動 `*=`，將原本的 $x$（也就是 4）乘以 26：$4 \times 26 = 104$！
3. 最後把 104 存回 $x$ 箱子中。

複合乘法在計算複利利息、縮放遊戲畫面大小、或是計算幾何級數時，都是超級實用且快速的利器！

````python
# [2] Code 範例區：展示 *= 乘法賦值與右側括號優先計算
x = 4

# 相當於 x = x * (2 * (3 + 10)) = 4 * 26 = 104
x *= 2 * (3 + 10)
print(x)
````

### ➗ 1-4-4 複合整除與取餘數（//=, %=）——位數縮小與循環分堆

接下來這兩個複合運算子，是 APCS 考試中最核心、考過最多次的「**拆解整數神技**」！

1. **整數除法賦值（`//=`）**：
   `x //= 4` 相當於 `x = x // 4`（整除取商數，無條件捨去小數點）。
   在 APCS 處理大整數時，最經典的招式就是：
   `n //= 10`
   這行指令就像是一把神奇剪刀，會把整數 $n$ 的個位數直接「剪掉」，讓整個數字往右縮小一位（例如 1234 變成 123）！

2. **取餘數賦值（`%=`）**：
   `x %= 3` 相當於 `x = x % 3`（求除以 3 的餘數）。
   你可以把 `%=` 想像成時鐘轉盤或排隊分組！不管數字原本有多大，只要一經過 `%= 3`，它就會立刻被乖乖鎖定在 0、1、2 這三種可能之中！

讓我們在 IDLE 或 Colab 裡面親自動手試試看，這兩道除法縮寫會如何迅速改變變數的值！

````python
# [2] Code 範例區：示範 //= 整除取商 與 %= 模除取餘數
num = 1234
num //= 10   # 相當於 num = num // 10，去除最後一位數
print(num)   # 印出 123

days = 17
days %= 7    # 相當於 days = days % 7，計算 17 天除以 7 天（一週）後的餘數
print(days)  # 印出 3
````

### 🚀 1-4-5 複合次方賦值（**=）——威力驚人的幾何連乘

最後一個複合賦值大絕招是「**次方賦值（`**=`）**」！
`x **= 2` 相當於 `x = x ** 2`（將 $x$ 變成自己的平方）。

吳邦一老師在講義第 7 頁給出了一行非常高深的指令：
```python
x = 2
y = 3
x **= y + 1
```
看到這裡，相信經過前面訓練的國一同學們，已經能一眼看穿它的秘密了！
1. **右邊先算**：先算右邊的式子 `y + 1`，因為 $y = 3$，所以 $3 + 1 = 4$！
2. **次方運算**：這行指令等同於 `x = x ** 4`，也就是 $2$ 的 $4$ 次方：$2 \times 2 \times 2 \times 2 = 16$！
3. **裝箱保存**：最後將 16 裝入 $x$ 箱子中保存。

次方賦值的增長速度非常驚人，在數學次方題、二進位次方計算、或幾何級數運算中，`**=` 是最帥氣的一行寫法！

````python
# [2] Code 範例區：展示 **= 次方賦值運算
x = 2
y = 3

# 相當於 x = x ** (y + 1) = 2 ** 4 = 16
x **= y + 1
print(x)
````

````python
# ==========================================
# [3] Code 填空題
# 任務說明：
# 請使用 **= 次方賦值運算子，將 base 變數直接計算為自己的 3 次方（立方）：
# ==========================================

base = 4
base ___ 3  # 相當於 base = base ** 3 = 4 * 4 * 4
print(base) # 預期印出 64
````
