### 📦 2-3-1 整數除法雙斜線（//）——只留整數商數的無情去尾刀

國一的大家在國小數學都學過除法直式運算，例如：$17 \div 5 = 3 \dots 2$。
在算式中，`3` 叫做「商數」，`2` 叫做「餘數」。
在 Python 中，如果我們只想拿到除完後的「**整數商數**」，該用什麼運算子呢？
答案就是打兩次斜線的——「**整數除法運算子 `//`（Floor Division / 整除取商）**」！

你可以把 `//` 想像成水果包裝工廠裡的「**自動裝箱打包機**」：
假設果園今天採收了 17 顆香甜的紅富士蘋果，每 5 顆裝成一個完整高級禮盒。
當打包機啟動執行 `17 // 5` 時，它會喀擦一聲迅速包裝出 `3` 個完整禮盒！
至於那剩下的 2 顆因為裝不滿一整盒，打包機就像一把冷酷無情的去尾刀，**直接自動丟棄、完全不理會**，只回傳打包好的箱數 `3`！

**⭐ 整數除法的重要特質：純正的整數血統（int）**：
當除數與被除數都是整數時，`//` 算出來的結果「**保證 100% 是純整數（int）**」！
例如 `17 // 5` 的答案就是純整數 `3`，絕對不會出現任何小數點（不是 `3.4`，也不是 `3.0`）！
這在程式設計中非常實用，能確保運算出來的數值永遠是乾淨俐落的純整數。

**⚠️ 區別預告：雙斜線 `//` vs 單斜線 `/`**：
在 Python 裡，單斜線 `/` 是浮點數除法，`17 / 5` 會算出帶有小數點的 `3.4`（我們會在 2-5 單元深入探討）；
但只要你在題目中看到「算有幾打、有幾箱、能坐滿幾輛車」，請務必直接聯想到雙斜線 `//`！

```python
# [2] Code 範例區：展示整數除法 // 只取商數的特性
apples = 17
box_size = 5

# 使用 // 計算能裝滿幾個完整箱子
boxes = apples // box_size
print(boxes)  # 印出 3（零頭 2 顆被自動去尾）

# 驗證型態：結果為純正的整數 <class 'int'>
print(type(boxes))
```

### 🔍 2-3-2 模除取餘數百分比（%）——撿回落單零頭的餘數探測器

在上一節中，打包機無情地把裝不滿一盒的零散蘋果丟在一旁。但是，身為精打細算的老闆，那些剩下的蘋果到底有幾顆呢？
這時候，我們就要隆重介紹雙斜線的好搭檔——「**模除取餘數運算子 `%`（Modulus）**」！

**🚨 初學者最大符號迷思破解**：
請大家特別注意！鍵盤上的 Shift + 數字鍵 5 雖然是百分比符號 `%`，但**在 Python 程式語言中，它完全不是數學打折（Percentage）的意思**！
如果你寫 `100 % 10`，它絕對不是在算 100 的 10%，而是在算「**100 除以 10 之後剩下的餘數**」！

你可以把 `%` 想像成裝箱輸送帶底部的「**落單零頭探測器**」：
當 17 顆蘋果以每盒 5 顆裝箱完後，探測器啟動 `17 % 5`，就會立刻嗶嗶嗶偵測出剩下落單的 `2` 顆蘋果！

**📐 除法黃金還原公式**：
在數學上，被除數 $A$、除數 $B$、商數 $Q$ 與餘數 $R$ 永遠滿足一個完美的守恆公式：
$$A = (A // B) \times B + (A \% B)$$
例如：$17 = (17 // 5) \times 5 + (17 \% 5) = 3 \times 5 + 2 = 15 + 2 = 17$！
這代表只要有商數 `//` 與餘數 `%`，我們就能毫無遺漏地完整還原所有的數字！

**餘數的數學邊界特質**：
除以任何正整數 $B$，算出來的餘數必定大於等於 0，而且「**嚴格小於 $B$**」！
例如除以 5 的餘數只可能是 `0, 1, 2, 3, 4`，絕對不可能出現 5 或更大。

**💥 致命大陷阱：除以 0 的死穴（ZeroDivisionError）**：
在所有數學與程式中，除數絕對不能為 0！如果你寫 `17 // 0` 或 `17 % 0`，Python 直譯器會當場引發大爆炸，噴出 `ZeroDivisionError: integer division or modulo by zero`！在 APCS 上傳系統中這會直接判定為 RE（Runtime Error，執行時期錯誤），不可不慎！

```python
# [2] Code 範例區：展示 % 模除取餘數與除法還原公式
candies = 23
kids = 4

# 每位小朋友分到的糖果數（商數）
each_kid = candies // kids
# 分完後剩下的糖果數（餘數）
leftover = candies % kids

print(each_kid)  # 印出 5（每人 5 顆）
print(leftover)  # 印出 3（剩下 3 顆落單）

# 驗證除法還原公式：商數 * 除數 + 餘數 == 原總數
rebuilt = each_kid * kids + leftover
print(rebuilt)   # 5 * 4 + 3 = 23（完美還原！）
```

### 🎡 2-3-3 奇偶數判定與分組循環——餘數的週期魔法

學完了基本的 `%` 運算，大家可別以為餘數只能拿來算分剩下的零食！
在演算法的世界裡，餘數 `%` 擁有兩大無人不知、無人不曉的超能力：

**⚡ 超能力一：一秒照出奇數與偶數（Even or Odd）**
國一數學告訴我們：只要能被 2 整除的數就是偶數，不能被 2 整除的就是奇數。
在 Python 中，只要把任何整數拿來 `% 2`：
* 如果 `n % 2 == 0`：餘數是 0，代表它百分之百是「**偶數（Even）**」！例如 `8 % 2` 得到 `0`，`0 % 2` 也是 `0`。
* 如果 `n % 2 == 1`：餘數是 1，代表除以 2 剩 1，它百分之百是「**奇數（Odd）**」！例如 `9 % 2` 得到 `1`。
這是未來學習第 5 章 `if` 判斷式時，最常用、速度最快的神級判別公式！

**🎡 超能力二：週期轉盤與輪流排班（Cycle）**
你可以把餘數想像成「**旋轉木馬或是時鐘指針**」！當指針轉了一圈之後，又會回到起點重新開始。
例如：
1. **星期幾推算**：一週固定有 7 天。如果今天是星期三（設定為 3），那麼經過 10 天後是星期幾呢？
   只要算 `(3 + 10) % 7`，答案是 `6`，代表那天就是星期六！
2. **4 位玩家輪流出牌**：
   第 0 回合是 0 號、第 1 回合是 1 號、第 2 回合是 2 號、第 3 回合是 3 號、第 4 回合又輪回 0 號……
   只要寫 `round_num % 4`，數字就會像摩天輪一樣永遠在 `0, 1, 2, 3` 之間規律跳動，永遠不會出界！

在 APCS 考題中，舉凡循環紅綠燈、轉盤遊戲、分組隊伍，核心全是利用這個週期餘數魔法！

```python
# [2] Code 範例區：展示奇偶數判定與週期循環運算
# 示範 1：使用 % 2 檢驗奇偶數
num1 = 18
num2 = 25
print(num1 % 2)  # 印出 0（偶數）
print(num2 % 2)  # 印出 1（奇數）

# 示範 2：星期幾推進循環（以 7 為週期）
today = 3        # 星期三
pass_days = 10   # 經過 10 天
future_day = (today + pass_days) % 7
print(future_day) # 印出 6（星期六）

# 示範 3：4 人輪流遊戲出牌者編號
turn = 9
player = turn % 4
print(player)    # 印出 1（第 9 回合輪到 1 號玩家）
```

### 🧅 2-3-4 神奇位數拆解術——取出個位數（% 10）與右移剝皮（// 10）

現在，我們要來解鎖全套教材中最經典、在 APCS 考場上出現頻率最高的核心絕技——「**整數各個位數拆解術**」！
在程式設計與演算法中，題目常常會給我們一個很大、很長的整數，例如 `358`，要求我們分別取出它的「個位數 8」、「十位數 5」以及「百位數 3」來做數學運算。

很多剛學程式的同學會想把數字轉成文字來切，但在競技程式中，最正統、執行速度最快的解法，就是出動我們的「**除法雙神刀**」：

**🔪 第一把神刀：夾娃娃機爪子——`% 10` 取出個位數**
任何十進位的整數除以 10，它的餘數永遠剛好就是「**最右邊的那顆個位數**」！
* `358 % 10` $\rightarrow$ 得到 `8`
* `47 % 10` $\rightarrow$ 得到 `7`
* `9 % 10` $\rightarrow$ 得到 `9`
你看！就像夾娃娃機的機械爪一樣，一爪伸下去，精準抓出最後一個數字！

**🔪 第二把神刀：洋蔥剝皮刀——`// 10` 右移砍掉個位數**
任何十進位的整數整除 10，最右邊的個位數就會被瞬間砍掉、數字自動向右縮排一位！
* `358 // 10` $\rightarrow$ 得到 `35`（個位數 8 被砍掉了！）
* `35 // 10` $\rightarrow$ 得到 `3`（十位數 5 被砍掉了！）
* `3 // 10` $\rightarrow$ 得到 `0`（最後縮減為 0！）

**🪓 兩把神刀的完美連環合體技**：
想要拆解一個三位數 `num = 358`：
1. 第一步（抓個位）：`digit1 = num % 10` $\rightarrow$ 抓出 `8`！
2. 第二步（砍個位）：`num = num // 10` $\rightarrow$ 數字縮小為 `35`！
3. 第三步（抓十位）：`digit2 = num % 10` $\rightarrow$ 抓出 `5`！
4. 第四步（砍十位）：`num = num // 10` $\rightarrow$ 數字縮小為 `3`！
5. 第五步（抓百位）：`digit3 = num % 10` $\rightarrow$ 抓出 `3`！

透過這套「抓一位、砍一位」的優雅規律，再多位數的數字在我們眼前都像剝洋蔥一樣，被拆解得一清二楚！

```python
# [2] Code 範例區：展示 % 10 與 // 10 逐步拆解三位數
num = 358
print(num)  # 原數字：358

# 第 1 步：取出個位數，並右移剝除
d_units = num % 10
num = num // 10

# 第 2 步：取出十位數，並右移剝除
d_tens = num % 10
num = num // 10

# 第 3 步：取出百位數
d_hundreds = num % 10

print(d_units)     # 印出 8
print(d_tens)      # 印出 5
print(d_hundreds)  # 印出 3
```

### ⏰ 2-3-5 時間與單位的複合進位換算——商與餘數的黃金雙重奏

經過前面的修煉，我們已經分別熟練了 `//`（算完整單位）與 `%`（算零頭殘額）。
現在，我們要來看這兩位搭檔合體時，如何展現出如同天籟般的「**黃金雙重奏**」！

在生活應用與 APCS 考題中，最常遇見的需求就是「**單位進位換算**」！

**場景一：影片總秒數換算為「幾分幾秒」**
假設有一段精彩的籃球比賽短片，總播放時間是 `195` 秒。
如果我們想在畫面上顯示「幾分幾秒」，該怎麼算？
因為 1 分鐘固定有 60 秒：
* **完整的分鐘數**：用整數除法 `195 // 60`，算出 `3` 分鐘！
* **剩下的秒數**：用模除取餘數 `195 % 60`，算出剩下的 `15` 秒！
合起來就是漂亮的 `3 分 15 秒`！完全不需要繁瑣的多次手動扣除！

**場景二：自動販賣機零錢找零（百元鈔與十元硬幣）**
你投幣買了一罐飲料，機器需要找零 `270` 元：
* **先換成最大面額百元鈔**：`270 // 100` $\rightarrow$ 吐出 `2` 張百元鈔！
* **剩下百元找不開的零頭**：`remain = 270 % 100` $\rightarrow$ 剩下 `70` 元。
* **再把零頭換成十元硬幣**：`remain // 10` $\rightarrow$ 吐出 `7` 枚十元硬幣！

**🎯 APCS 考場提速秘訣**：
遇到公分轉公尺、秒轉時分秒、天數轉週天數時，永遠記得：**「大單位用 `//`，零散尾數用 `%`」**！
一行算式瞬間擊破，時間複雜度是極速的 $O(1)$，又快又準確！

```python
# [2] Code 範例區：展示商數與餘數雙重奏進行時間與零錢換算
# 範例 1：秒數換算為幾分幾秒
total_seconds = 195
minutes = total_seconds // 60  # 算完整分鐘
seconds = total_seconds % 60   # 算剩餘秒數
print(minutes)  # 印出 3
print(seconds)  # 印出 15

# 範例 2：270 元換算百元鈔與十元硬幣
change = 270
bills_100 = change // 100      # 2 張百元
coins_10 = (change % 100) // 10 # 7 枚十元
print(bills_100)
print(coins_10)
```
