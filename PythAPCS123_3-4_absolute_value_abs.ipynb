### 🛡️ 3-4-1 認識 abs() 絕對值函數——數線上的無向距離守護者

想像你站在操場的起跑點（座標 0）。向右跑 50 公尺，你的位移是 $+50$；向左跑 50 公尺，你的位移是 $-50$。
但如果教練問你：「你今天跑步的『步數距離』是多少？」
不管是向左還是向右，你實實在在消耗熱量跑出來的距離**永遠都是正的 50 公尺**！
在 Python 中，負責拔除方向與負號、只保留純粹距離大小的內建函數就是 **`abs()`**（Absolute Value 的縮寫）！

**運作機制與底層特性：**
- **正數與零保持不變**：`abs(10)` 是 `10`，`abs(0)` 是 `0`。
- **負數消除負號轉正**：`abs(-42)` 會把前面的減號一秒拔除，回傳正整數 `42`！
- **同時支援整數與浮點數**：`abs(-3.14)` 會精準回傳浮點數 `3.14`。

**常見錯誤陷阱（非數值報錯）：**
`abs()` 專門為數值（整數、小數）服務。如果你把包含引號的字串傳給它（例如 `abs('-5')`），Python 會毫不客氣地拋出：
`TypeError: bad operand type for abs(): 'str'`！
請記得：如果資料來自使用者輸入 `input()`，必須先透過 `int()` 或 `float()` 轉型為數值，才能交給 `abs()` 處理喔！


````python
# [2] Code 範例區：展示 abs() 基本數值絕對值運算

# 範例 1：整數與負數絕對值
print(abs(15))     # 印出 15（正數不變）
print(abs(-42))    # 印出 42（負數拔除負號變正數）
print(abs(0))      # 印出 0

# 範例 2：浮點數絕對值
temp_drift = -3.75
print(abs(temp_drift))  # 印出 3.75

# 範例 3：搭配算術運算
val = -8 * 5
print(abs(val))    # -40 的絕對值為 40
````

### ✂️ 3-4-2 兩數無向差值計算——一行求得差距，取代正負號手動調整

在生活與科學計算中，我們無時無刻不在計算「相差多少」：
* 哥哥身高 175 公分，弟弟身高 162 公分，兩人相差幾公分？
* 今天最高溫 28 度，清晨最低溫 19 度，溫差是多少？

**初學者的計算煩惱：**
在數學算式中，如果我們用大數減小數（$175 - 162 = 13$），結果是正數。
但如果不知道誰大誰小，直接相減（$162 - 175 = -13$），就會跑出令人頭痛的負號！
如果每次都要手動檢查負號、再乘以 $-1$ 轉正，算式會顯得相當繁瑣。

**abs() 的一行解法：**
有了絕對值函數，我們完全不用在乎減數與被減數的順序！
不論是 `abs(175 - 162)` 還是 `abs(162 - 175)`，通通都會得到純正的距離正值 **`13`**！
一行算式直接搞定差值，乾淨俐落！

````python
# [2] Code 範例區：利用 abs() 計算兩數無向差值

# 給定兩位同學的體重
weight_a = 52
weight_b = 65

# 傳統困境：直接相減為負數
raw_diff = weight_a - weight_b
print(raw_diff)  # 印出 -13

# 一行精簡解法：使用 abs() 求純距離差值
diff = abs(weight_a - weight_b)
print(diff)      # 印出 13

# 即使顛倒相減順序，結果也完全相同
print(abs(weight_b - weight_a))  # 同樣印出 13
````

### 🏙️ 3-4-3 曼哈頓距離（兩點網格距離）計算——abs(x1 - x2) + abs(y1 - y2)

在美國紐約的曼哈頓島上，所有的街道都像棋盤方格一樣整整齊齊。計程車司機想要從一個十字路口開到另一個十字路口，不能像鳥兒一樣穿牆斜切飛行，只能乖乖沿著街道「水平開幾條街、再垂直開幾條街」。
這種在網格方格上只能沿上下左右走的距離計算方式，在資訊科學中被尊稱為「**曼哈頓距離（Manhattan Distance / 市區街區距離）**」！

**核心公式拆解：**
給定二維平面上的起點 $(x_1, y_1)$ 與終點 $(x_2, y_2)$：
- 水平需要走的橫向街區數：$|x_1 - x_2|$
- 垂直需要走的縱向街區數：$|y_1 - y_2|$
- 總曼哈頓距離公式：
$$D = |x_1 - x_2| + |y_1 - y_2|$$
在 Python 中，只要寫成極度直觀的一行：
`distance = abs(x1 - x2) + abs(y1 - y2)`！

**APCS 競技舞台的關鍵地位：**
在 APCS 的第 2 題與第 3 題網格模擬題中（如機器人移動、棋盤追蹤、人口遷移），「計算兩個座標點的最短步數」幾乎每次必考！
只要記住這道雙重 `abs()` 的黃金拍檔公式，你在網格世界中就能精準算出行走步數，再也不怕被橫縱座標搞得眼花撩亂！


````python
# [2] Code 範例區：展示曼哈頓距離計算

# 範例 1：計算起點與終點的曼哈頓距離
x1, y1 = 1, 2  # 起點座標
x2, y2 = 4, 6  # 終點座標

# 水平差值 + 垂直差值
dist = abs(x1 - x2) + abs(y1 - y2)
print("曼哈頓距離:", dist)  # |1-4| + |2-6| = 3 + 4 = 7

# 範例 2：對稱性檢驗（反過來算也一樣）
reverse_dist = abs(x2 - x1) + abs(y2 - y1)
print(dist == reverse_dist)  # 印出 True
````

### 🎯 3-4-4 兩組總和差距計算實務——abs(odd_sum - even_sum) 的純數學化身

在許多數據分析、競賽統計與演算法中，我們常常會分別計算出兩組數據的總和：
例如「奇數項數據總和 $A$」與「偶數項數據總和 $B$」，或是「主隊上半場得分 $A$」與「客隊上半場得分 $B$」。
題目最後的核心要求常常是：**「請輸出兩組總和的相差絕對值 $|A - B|$」！**

**演算法高手的純數學解法：**
如果不知道哪一邊總和較大，最乾淨、最不拖泥帶水的寫法，正是將兩組總和的差距直接交給 `abs()` 處理：
```python
print(abs(odd_sum - even_sum))
```
這短短一行，融合了我們在前面單元學到的整數運算、變數賦值，以及今天單元 3-4 的絕對值防護！
這正是資訊科學中「利用內建工具大幅精簡邏輯」的最佳典範！

````python
# [2] Code 範例區：兩組總和差值絕對值計算示範

# 模擬計算後的兩組數值總和
odd_sum = 15   # 奇數項加總
even_sum = 22  # 偶數項加總

# 步驟 1：兩數相減（此時 diff 為 -7）
diff = odd_sum - even_sum

# 步驟 2：利用 abs() 直接輸出純差值
print(abs(diff))  # 成功印出正整數 7！
````
