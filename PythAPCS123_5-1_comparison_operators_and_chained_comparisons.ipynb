### 🎢 5-1-1 數值大小比較——大於（`>`）、小於（`<`）、大於等於（`>=`）、小於等於（`<=`）

**生活比喻：遊樂園的身高安全檢測桿**  
當你來到主題樂園準備搭乘雲霄飛車時，入口處總會立著一根醒目的身高檢測桿。檢測員的手勢只有兩種可能：
- 若身高達到標線，點頭放行（代表條件成立，在程式世界稱為 **`True`**）。
- 若身高不足標線，搖頭攔阻（代表條件不成立，在程式世界稱為 **`False`**）。

這根身高檢測桿，在 Python 中就是**比較運算子（Comparison Operators）**。當我們寫下 `height >= 120` 時，我們並不是在計算某個數值，而是在向電腦提出一個「是非題」！電腦在審視左右兩側數值後，只會回答我們：到底這件事是「真的（`True`）」還是「假的（`False`）」。

**運作機制與符號清單：**
Python 提供四個最常用的數值順序比較運算子：
1. `>` （嚴格大於）：只有左邊數值「嚴格超過」右邊時才成立。例如 `7 > 3` 結果為 `True`；`5 > 5` 結果為 `False`。
2. `<` （嚴格小於）：只有左邊數值「嚴格低於」右邊時才成立。例如 `2 < 8` 結果為 `True`；`6 < 6` 結果為 `False`。
3. `>=` （大於等於）：左邊數值大於「或者等於」右邊時皆成立。例如 `5 >= 5` 結果為 `True`；`6 >= 5` 結果為 `True`；`4 >= 5` 結果為 `False`。
4. `<=` （小於等於）：左邊數值小於「或者等於」右邊時皆成立。例如 `7 <= 7` 結果為 `True`；`3 <= 7` 結果為 `True`；`8 <= 7` 結果為 `False`。

**常見錯誤陷阱（初學者常見語法地雷）：**
1. **符號順序顛倒**：
   在鍵盤輸入時，必須將大於號或小於號寫在前面，等號永遠放在後面（`>=`、`<=`）。初學者常因口語「等於大於」或鍵盤手誤寫成 `=>` 或 `=<`，這在 Python 中屬於非法語法，會立刻拋出 `SyntaxError: invalid syntax` 致命錯誤！
2. **符號中間留有空格**：
   `>=` 與 `<=` 是一個不可分割的雙字元運算子，千萬不能寫成 `> =` 或 `< =`，否則同樣會導致語法解析失敗。

**APCS 考試實務建議：**
在 APCS 與各類演算法競賽中，題目對於邊界值（Boundary Conditions）的定義極為嚴格，例如「*分數達 60 分（含）以上*」必須使用 `>= 60`；若寫成 `> 60`，在臨界測資（剛好 60 分）就會產生邏輯錯誤（Wrong Answer）。審題時務必看清楚「是否包含端點」！


````python
# [2] Code 範例區：展示數值大小比較運算子（>, <, >=, <=）

# 1. 宣告檢驗基準與測試數值
pass_threshold = 60
score_alpha = 75
score_beta = 60
score_gamma = 59

# 2. 進行嚴格大小比較（> 與 <）
print("--- 嚴格大小比較 ---")
print("75 分嚴格大於 60 分：", score_alpha > pass_threshold)  # True
print("60 分嚴格大於 60 分：", score_beta > pass_threshold)   # False (邊界未達嚴格大於)
print("59 分嚴格小於 60 分：", score_gamma < pass_threshold)  # True

# 3. 進行包含邊界的比較（>= 與 <=）
print("--- 包含邊界的大於等於與小於等於 ---")
print("75 分大於等於 60 分：", score_alpha >= pass_threshold) # True
print("60 分大於等於 60 分：", score_beta >= pass_threshold)  # True (剛好達標成立！)
print("59 分大於等於 60 分：", score_gamma >= pass_threshold) # False

# 4. 比較運算結果可以直接存入變數（布林變數）
is_gamma_failing = score_gamma < pass_threshold
print("Gamma 是否未達及格門檻：", is_gamma_failing)
````

### ⚖️ 5-1-2 嚴格相等與不相等——雙等號（`==`）與驚嘆號等號（`!=`）

**生活比喻：身分證防偽掃描器 vs 置物櫃貼名牌**  
在第一章中，我們學過單等號 `=` 是「賦值（Assignment）」——把右邊的物品放進左邊貼有名牌的箱子裡。這是一種**動作命令**（例如 `box = 100`，箱子請裝進 100）。  
而現在我們要學習的雙等號 `==`，則是一台「防偽驗證掃描器」！掃描器只會拿著兩張卡片左右比對，向你回報：「*這兩張卡片的號碼有一樣嗎？*」這是一個**真偽提問**。

在 Python 程式語言中，這兩者肩負著完全截然不同的神聖使命：
1. **雙等號（`==`，等於判斷）**：
   用來提問「左邊的值是否與右邊的值相等？」。若相等則回傳 `True`，不相等則回傳 `False`。
   例如：`5 == 5` 回傳 `True`；`5 == 8` 回傳 `False`。
2. **驚嘆號等號（`!=`，不等於判斷）**：
   在電腦符號學中，驚嘆號 `!` 代表「否定（Not）」。因此 `!=` 就是用來提問「左邊的值是否『不等於』右邊的值？」。若兩者不同則回傳 `True`，相同則回傳 `False`。
   例如：`5 != 8` 回傳 `True`；`5 != 5` 回傳 `False`。

**常見錯誤陷阱（初學者第一大天坑）：**
1. **把雙等號 `==` 誤寫為單等號 `=`**：
   這是所有程式初學者最容易踩中的經典天坑！  
   如果你在想比對兩數時寫成了 `x = y`，電腦不會進行比對，而是直接把 `y` 的值粗暴地覆寫進 `x` 變數中，同時在許多運算情境下引發 `SyntaxError: invalid syntax` 語法報錯！  
   請務必牢記心法：**一個等號是塞值（命令），兩個等號是比值（提問）**！
2. **浮點數精確度比對陷阱**：
   在第二章 2-5 節中我們學過電腦二進位浮點數存在微小精度誤差，例如 `0.1 + 0.2 == 0.3` 在 Python 中會回傳 `False`（因為 `0.1 + 0.2` 實際為 `0.30000000000000004`）！因此在 APCS 中，除非題目明確要求整數比較，否則請盡量避免對浮點數直接使用 `==` 比對。

**APCS 考試實務建議：**
`==` 與 `!=` 常用於檢驗「狀態碼」、「旗標比對」以及「餘數是否為零（例如判斷偶數 `n % 2 == 0`）」。善用 `==` 可以迅速取得精準的真假狀態。


````python
# [2] Code 範例區：展示雙等號（==）與不等於（!=）運算子

target_pin = 1234
input_pin_correct = 1234
input_pin_wrong = 9999

# 1. 雙等號 == 檢驗是否完全相同
print("--- 雙等號 == 等於驗證 ---")
is_match_1 = (input_pin_correct == target_pin)
is_match_2 = (input_pin_wrong == target_pin)
print("正確密碼比對結果：", is_match_1)  # True
print("錯誤密碼比對結果：", is_match_2)  # False

# 2. 驚嘆號等號 != 檢驗是否不同
print("--- 不等於 != 差異驗證 ---")
has_error_1 = (input_pin_correct != target_pin)
has_error_2 = (input_pin_wrong != target_pin)
print("正確密碼是否有誤：", has_error_1)  # False (無誤)
print("錯誤密碼是否有誤：", has_error_2)  # True (確實有誤)

# 3. 結合第二章取餘數運算 % 檢驗奇偶數
test_num = 18
is_even = (test_num % 2 == 0)
is_odd = (test_num % 2 != 0)
print("--- 數值整除與奇偶特性 ---")
print("18 除以 2 的餘數是否為 0（是否為偶數）：", is_even)  # True
print("18 除以 2 的餘數是否不為 0（是否為奇數）：", is_odd)   # False
````

### 🍳 5-1-3 運算式與運算優先級——先算術運算，再進行大小比較

**生活比喻：料理大賽的評審評分**  
在一場廚藝競賽中，兩位廚師正在準備端出各自的招牌菜。裁判並不會在廚師切菜時就急著判定勝負，而是必須等到：
1. 一號廚師完成所有的烹煮、調味並結算出總分（完成算術運算式）。
2. 二號廚師也完成所有的烹煮、調味並結算出總分（完成算術運算式）。
3. 裁判才吹哨站出來，把兩碗菜餚擺在天平上比大小（進行比較運算）。

這正是電腦處理運算時的**優先級法則（Operator Precedence）**！在 Python 中，**所有四則算術運算子（`+`, `-`, `*`, `/`, `//`, `%`, `**`）的優先等級，都嚴格高於比較運算子（`>`, `<`, `==`, `!=`, `>=`, `<=`）**！

**運作機制深入解析：**
看這個經典範例（吳邦一老師講義第 17 頁）：
```python
x = 7
y = 3
result = (x + y) > (x * y)
```
電腦在執行時的步驟如下：
1. 步驟一（算術運算優先）：先計算左側的加法 `x + y`，得到 `7 + 3 = 10`。
2. 步驟二（算術運算優先）：再計算右側的乘法 `x * y`，得到 `7 * 3 = 21`。
3. 步驟三（最後比較求值）：把兩邊算出來的純量進行比較，即 `10 > 21`。
4. 最終答案為 `False`，並將此布林值存入 `result`！

即使我們不寫小括號 `x + y > x * y`，Python 的語法規則依然會先算加法與乘法，最後才做 `>` 比較。  
但是，**「加上小括號」是頂尖程式設計師與 APCS 考生的優良習慣**！小括號不僅能清晰劃分邏輯區塊，還能徹底杜絕因記憶模糊而產生的優先級混淆。

### 📖 5-1-4 文字字串的比較——字典序（Lexicographical Order）與大小寫 ASCII

**生活比喻：圖書館英文字典排架順序**  
當你在牛津英文字典中查單字時，為什麼 `apple` 一定排在 `banana` 的前面？因為開頭字母 `a` 比 `b` 更早出現在字母表中！那如果兩個單字前幾個字母都一樣呢？例如 `cat` 和 `car`，字典會往後比對第三個字母，因為 `r` 排在 `t` 前面，所以 `car` 比 `cat` 更小。

在 Python 裡，**文字字串（String）也可以使用所有比較運算子（`==`, `!=`, `<`, `>`, `<=`, `>=`）！**  
其比對規則正是模擬字典的**「字典序（Lexicographical Order）」**，其背後的根本依據就是我們在第三章 3-2 節所學過的 **ASCII 編碼值（`ord()` 函數）**！

**教材範例與核心運作原理（吳邦一老師講義第 17 頁）：**
讓我們來拆解講義中的經典範例：
```python
s1 = 'Python'
s2 = 'python'
s3 = 'algorithm'
```
1. **`s1 == s2` 為 `False`（大小寫極為敏感）**：
   因為第一個字元 `s1[0]` 是大寫 `'P'`（ASCII 為 80），而 `s2[0]` 是小寫 `'p'`（ASCII 為 112）。兩者編碼不同，因此不相等！
2. **`s1 <= s2` 為 `True`（大寫字母永遠小於小寫字母）**：
   在 ASCII 碼表中，所有英文大寫字母（A~Z 為 65~90）都排在小寫字母（a~z 為 97~122）的前面！因此大寫 `'P'`（80）小於小寫 `'p'`（112），所以 `'Python' < 'python'` 成立！
3. **`s1 <= s3` 為 `True`（大寫 P 與小寫 a 的比對）**：
   比對 `'Python'` 與 `'algorithm'` 時，電腦首先比對各自的第一個字元：大寫 `'P'`（80）與小寫 `'a'`（97）。因為 80 < 97，所以 `'Python' < 'algorithm'` 同樣成立！

**常見錯誤陷阱（字串比較的超級核彈級地雷）：**
- **數字以字串形式比對的巨大災難**：
  若你忘了將 `input()` 轉型為 `int`，拿兩個「數字字串」去比大小：
  ```python
  '100' < '20'   # 結果竟然是 True！
  ```
  初學者看到這裡常大驚失色：「*100 怎麼可能比 20 小？！*」  
  這是因為電腦是比對**第一個字元**：字串 `'100'` 的開頭字元是 `'1'`（ASCII 49），而字串 `'20'` 的開頭字元是 `'2'`（ASCII 50）。因為 49 < 50，電腦就直接判定 `'100' < '20'` 為 `True`！  
  **鐵律：只要是想比數值大小，絕對必須先用 `int()` 或 `float()` 轉型，千萬不可拿字串直接比大小！**

**APCS 考試實務建議：**
字串的字典序比對是後續字串搜尋、排序（第 12 章）以及各類文字處理題型的基礎核心。掌握 ASCII 編碼大小關係，就能一眼看穿字串比對的真實底層行為。


````python
# [2] Code 範例區：展示文字字串的字典序比較與常見陷阱

# 1. 吳邦一老師講義第 17 頁經典範例
s1 = 'Python'
s2 = 'python'
s3 = 'algorithm'

print("s1 =", s1, ", s2 =", s2, ", s3 =", s3)
print("s1 == s2（大小寫比對）：", s1 == s2)   # False
print("s1 <= s2（大寫 P vs 小寫 p）：", s1 <= s2) # True (80 <= 112)
print("s1 <= s3（大寫 P vs 小寫 a）：", s1 <= s3) # True (80 <= 97)

# 2. 驗證字元 ASCII 碼（第三章 ord 函數複習）
print("--- 查驗字元 ASCII 編碼 ---")
print("'P' 的 ASCII 碼：", ord('P'))  # 80
print("'p' 的 ASCII 碼：", ord('p'))  # 112
print("'a' 的 ASCII 碼：", ord('a'))  # 97

# 3. 🚨 極重要警告：數字字串的比對地雷示範
str_num1 = '100'
str_num2 = '20'
int_num1 = int(str_num1)
int_num2 = int(str_num2)

print("--- 數字字串 vs 真實整數比對 ---")
print("字串比對 '100' < '20'：", str_num1 < str_num2)  # True！(陷阱！)
print("整數比對  100  <  20 ：", int_num1 < int_num2)  # False (符合常理)
````

### 🔗 5-1-5 Python 獨家黑科技——連續比較式（Chained Comparisons，如 `1 < x < 9`）

**生活比喻：數學課本的數線區間夾擠**  
在數學課上，當老師在黑板畫出一條數線，標示一個數字 $x$ 必須大於 1 同時小於 9 時，我們會很自然地寫下：
$$1 < x < 9$$
然而，如果你去學習 C++、Java 或 JavaScript 等語言，老師會嚴肅地警告你：「*千萬不能這樣寫！這樣寫會引發嚴重的運算錯誤！*」因為在那些語言中，電腦會先算 `1 < x` 得到真假值 `1`，再拿 `1 < 9` 繼續算，造成荒謬的邏輯錯誤。

但是——**在 Python 魔法王國中，你可以百分之百依照數學直覺書寫！**  
Python 原生支援極其優雅強大的**「連續比較式（Chained Comparisons）」**（吳邦一老師講義第 17 頁重點語法）。

**運作機制深入剖析：**
當你在 Python 中寫下：
```python
1 < x < 9
```
Python 直譯器在底層會自動將其拆解為兩個獨立的比較：
1. 先檢驗 `1 < x` 是否成立。
2. 同時檢驗 `x < 9` 是否成立。
3. 只有當前後兩個比較**同時成立（兩者皆為 `True`）**時，整體結果才為 `True`；只要其中任何一個不成立，結果就是 `False`！

**連續比較式的多元應用情境：**
1. **數值區間範圍檢查（極高頻應用）**：
   - 檢驗分數是否在 0 到 100 分的合法區間：`0 <= score <= 100`。
   - 檢驗水溫是否在適宜飲用的 40 到 60 度之間：`40 <= temp <= 60`。
2. **三數完全相等檢驗**：
   - 在傳統語言中要寫兩次比對，但在 Python 中可以直接寫：`a == b == c`！只有當 `a == b` 且 `b == c` 時才會回傳 `True`。
3. **等差或由小到大遞增數列檢驗**：
   - 檢驗三個輸入數字是否嚴格遞增：`num1 < num2 < num3`。

**常見錯誤陷阱（避免寫出混亂無章的折線比較）：**
雖然 Python 語法允許混搭比較符號，例如 `a < b > c`（代表 `a < b` 且 `b > c`），但這種「忽大忽小」的寫法極度容易讓閱讀程式碼的人產生認知混淆。  
**程式風格最佳實踐**：請一律維持**方向一致性**（全部朝向同一個方向，例如統一使用小於 `<` 與 `<=`），保持與數線一致的直覺性！

**APCS 考試實務建議：**
連續比較式是 Python 在 APCS 實作題中的殺手級武器！它能以極度簡潔的一行代碼完成邊界檢驗，大幅降低打字錯誤與括號漏寫的機率。


````python
# [2] Code 範例區：展示 Python 連續比較式（Chained Comparisons）

# 1. 吳邦一老師講義第 17 頁範例：1 < x < 9
x = 7
print("測試數值 x =", x)
print("1 < x < 9 是否成立：", 1 < x < 9)  # 1 < 7 成立且 7 < 9 成立 -> True

x_out = 15
print("測試數值 x_out =", x_out)
print("1 < x_out < 9 是否成立：", 1 < x_out < 9)  # 15 < 9 不成立 -> False

# 2. 應用情境一：學生成績閉區間檢驗（0 <= score <= 100）
score = 85
is_valid_score = (0 <= score <= 100)
print("--- 成績有效性檢驗 ---")
print("85 分是否在 0 ~ 100 分合法區間：", is_valid_score)  # True

invalid_score = 105
print("105 分是否在 0 ~ 100 分合法區間：", 0 <= invalid_score <= 100)  # False

# 3. 應用情境二：三數全等檢驗（a == b == c）
a, b, c = 5, 5, 5
print("--- 三數全等檢驗 ---")
print("三數 5, 5, 5 是否完全相同：", a == b == c)  # True

d, e, f = 5, 5, 8
print("三數 5, 5, 8 是否完全相同：", d == e == f)  # False (e == f 不成立)
````
