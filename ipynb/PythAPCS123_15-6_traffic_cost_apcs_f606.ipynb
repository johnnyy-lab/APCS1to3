### 📖 題目敘述（Problem Description）
一家跨國雲端服務公司擁有 $n$ 個伺服器與 $m$ 個基地台（城市）。  
每個伺服器運作時，會產生欲發送至各基地台的流量。我們用一個 $n \times m$ 的二維陣列 $Q$ 來表示：  
$Q[i][j]$ 代表**伺服器 $i$** 預計要傳送到**基地台 $j$** 的流量大小（$0 \le i < n, 0 \le j < m$）。

為了營運管理，這 $n$ 個伺服器需要被指派放置在特定的基地台。  
題目給定 $k$ 種伺服器放置方案。在每種方案中，伺服器 $i$ 會被放置在基地台 $p_i$（$0 \le p_i < m$）。

#### 💰 費用計算規則：
當多個伺服器被放置在同一個基地台時，**必須先將所有從基地台 $u$ 傳送到基地台 $v$ 的流量加總合併**，再計算傳輸費用：
1. **同基地台內部傳輸（$u = v$）**：  
   每單位流量費用為 **1 元**。
2. **跨基地台傳輸（$u \ne v$）**：  
   * 前 **1000** 單位流量（含 1000），每單位流量費用為 **3 元**。
   * 超過 **1000** 單位的部分，每單位流量費用享有優惠，為 **2 元**。
   * 舉例：若從基地台 $u$ 到基地台 $v$ 的總流量為 1100，則費用為 $1000 \times 3 + (1100 - 1000) \times 2 = 3000 + 200 = 3200$ 元。

請在給定的 $k$ 種放置方案中，計算每種方案的總傳輸花費，並輸出**花費最少的方案金額**。

* 接下來 $k$ 行，每行包含 $n$ 個整數，代表一種方案中各伺服器被分配到的基地台編號 $p_0, p_1, \dots, p_{n-1}$。

### 15.6.2 二維流量矩陣 Q[i][j] 讀取與伺服器指派陣列對照

#### 📐 矩陣座標對照模型：

### 15.6.3 矩陣聚合演算法：從「伺服器對伺服器」轉換為「基地台對基地台」流量矩陣

#### 🌟 解題大關鍵：聚合矩陣（Traffic Aggregation Matrix）
因為費用是依據**實體基地台之間的傳輸總流量**計算，所以計算費用的第一步：  
**必須先建立一個 $m \times m$ 的實體基地台流量矩陣 `traffic`**！

演算法流程：
1. 初始化 `traffic = [[0] * m for _ in range(m)]`。
2. 走訪每台伺服器 $i$（$0 \le i < n$）：
   * 獲取其起點基地台 $u = p[i]$。
   * 走訪每個目標基地台 $v$（$0 \le v < m$）：
     * `traffic[u][v] += Q[i][v]`
3. 經過此聚合，原本 $n \times m$ 的伺服器矩陣就轉換成了 $m \times m$ 的實體路徑流量矩陣！


````python
# =====================================================================
# [2] Code 範例：方案 1 的矩陣聚合實作
# =====================================================================

scheme_1 = [0, 0] # 兩台伺服器都在站點 0
traffic_1 = [[0] * m for _ in range(m)]

for i in range(n):
    u = scheme_1[i]
    for v in range(m):
        traffic_1[u][v] += Q[i][v]

print("=== 方案 1 實體基地台流量矩陣 (3x3) ===")
for row in traffic_1:
    print(row)
````

### 15.6.5 階梯式計費規則二：站點跨機房非線性超額分段費用計算（門檻 1000 轉折點）

#### 🌐 跨城市骨幹傳輸費率（Inter-station Traffic）：
當 $u \ne v$ 時，資料必須跨越實體長途幹線網路，採**階梯式累進優惠**：
* **第一段（$\le 1000$ 單位）**：每單位收取 **3 元**。
* **第二段（$> 1000$ 單位）**：前 1000 單位收 3 元，**超出的部分每單位降價為 2 元**。

#### 🧮 數學分段函數式：
$$\text{Cost}(F) = \begin{cases} 
F \times 3, & \text{if } F \le 1000 \\[6pt]
1000 \times 3 + (F - 1000) \times 2 = 3000 + (F - 1000) \times 2, & \text{if } F > 1000 
\end{cases}$$

```text
費用
 ▲                      / (斜率 = 2，超出部分每單位 2 元)
 │                     /
 │                    /
 │  3000 ───────────●
 │                 / (斜率 = 3，每單位 3 元)
 │                /
 └───────────────┴────────▶ 流量
 0              1000
```


````python
# =====================================================================
# [2] Code 範例：跨站傳輸分段計費函式
# =====================================================================

def calculate_inter_cost(flow):
    if flow <= 1000:
        return flow * 3
    else:
        return 1000 * 3 + (flow - 1000) * 2

print("流量 500 單位費用：", calculate_inter_cost(500))   # 1500
print("流量 1000 單位費用：", calculate_inter_cost(1000)) # 3000
print("流量 1100 單位費用：", calculate_inter_cost(1100)) # 3200
````

### 15.6.6 單一方案成本封裝：模組化評估函數與 k 種分流方案動態枚舉

#### 🧩 模組化整合架構：
給定一個方案 $p$，計算該方案總花費的完整流程如下：
1. 宣告 $m \times m$ 矩陣 `traffic`，初值全為 0。
2. 雙重迴圈聚合每台伺服器的流量至 `traffic[p[i]][v]`。
3. 遍歷整個 $m \times m$ 的 `traffic` 矩陣：
   * 若流量為 0，略過。
   * 若 $u == v$：累加 $\text{flow} \times 1$。
   * 若 $u \ne v$：累加分段跨站費用。
4. 回傳該方案的總花費。


````python
# =====================================================================
# [2] Code 範例：單一方案成本評估函式封裝
# =====================================================================

def evaluate_scheme_cost(scheme, Q_mat, n_val, m_val):
    # 步驟 1: 聚合流量
    traffic = [[0] * m_val for _ in range(m_val)]
    for i in range(n_val):
        u = scheme[i]
        for v in range(m_val):
            traffic[u][v] += Q_mat[i][v]
            
    # 步驟 2: 計算總傳輸費用
    total_cost = 0
    for u in range(m_val):
        for v in range(m_val):
            f = traffic[u][v]
            if f == 0: continue
            if u == v:
                total_cost += f * 1
            else:
                if f <= 1000:
                    total_cost += f * 3
                else:
                    total_cost += 3000 + (f - 1000) * 2
    return total_cost

cost_1 = evaluate_scheme_cost([0, 0], Q, n, m)
cost_2 = evaluate_scheme_cost([0, 1], Q, n, m)
cost_3 = evaluate_scheme_cost([0, 2], Q, n, m)
print("方案 1 花費：", cost_1)  # 257
print("方案 2 花費：", cost_2)  # 217
print("方案 3 花費：", cost_3)  # 261
````

### 15.6.7 全局最佳化維護：擂台法維護最低營運花費與完整 AC 程式碼實作

#### 🏆 擂台法極小值維護：
我們不需要將所有方案的花費都存入清單，而是可以在枚舉 $k$ 種方案的過程中：
1. 設定 `min_cost = float('inf')`。
2. 每讀取或計算完一個方案 `cur_cost`，立即執行 `if cur_cost < min_cost: min_cost = cur_cost`。
3. 遍歷完畢後，輸出 `min_cost`。

這樣做不僅節省記憶體，也能確保在串流輸入時即時完成運算！


````python
# =====================================================================
# [2] Code 範例：完整擂台法最佳化實作
# =====================================================================

def solve_f606_stream(n_val, m_val, k_val, Q_mat, schemes_list):
    min_cost = float('inf')
    for s in schemes_list:
        c = evaluate_scheme_cost(s, Q_mat, n_val, m_val)
        if c < min_cost:
            min_cost = c
    return min_cost

best_ans = solve_f606_stream(n, m, k, Q, schemes)
print("最低營運花費：", best_ans)  # 217
````

### 15.6.8 維度混淆防禦（n 與 m 邊界不對稱）、整數運算效能分析與常見 WA 陷阱排查

#### ⏱️ 複雜度精準估算：

  * 每個方案需要：

### 📝 版本一：APCS 官方實作考場專用版 —— 淺顯易懂一般版（新手友善推薦）

* **適用情境**：APCS 正式考試現場、初學者課堂練習。
* **教學與設計理念**：
  1. **步驟平鋪直敘、拆解詳細**：依序完成「讀取 $n, m, k$」、「讀取 $Q$ 矩陣」、「逐方案聚合流量」、「累加計算花費」四大步驟。
  2. **標準邊界防禦與分段計費**：清晰判斷 `u == v` 與 `u != v`，階梯式超額公式步驟明朗，絕不混淆。
  3. **穩健 100% 滿分 AC**：考場高壓下最容易理解與除錯，零失誤奪取滿分！


````python
# ==============================================================================
# 📝 版本一：APCS 官方實作考場專用版 —— 淺顯易懂一般版（新手友善推薦）
# 適用情境：APCS 正式考場單筆測試案例，平鋪直敘、步驟明確、穩健滿分
# ==============================================================================

# 步驟 1：讀取伺服器數 n、基地台數 m 與方案數 k
n, m, k = map(int, input().split())

# 步驟 2：讀取 n 個伺服器傳送到各基地台的流量矩陣 Q (n x m)
Q = []
for _ in range(n):
    Q.append(list(map(int, input().split())))

min_total_cost = float('inf')

# 步驟 3：逐一讀取並評估 k 種伺服器放置方案
for _ in range(k):
    scheme = list(map(int, input().split()))
    
    # 建立目前方案下，基地台對基地台的流量聚合矩陣 traffic (m x m)
    traffic = [[0] * m for _ in range(m)]
    for i in range(n):
        source_station = scheme[i]
        for dest_station in range(m):
            traffic[source_station][dest_station] += Q[i][dest_station]
            
    # 計算該方案的營運總傳輸花費
    current_cost = 0
    for u in range(m):
        for v in range(m):
            f = traffic[u][v]
            if f == 0:
                continue
            
            # 規則 1：同基地台內部傳輸（每單位 1 元）
            if u == v:
                current_cost += f * 1
            # 規則 2：跨基地台長途傳輸（前 1000 單位每單位 3 元，超出每單位 2 元）
            else:
                if f <= 1000:
                    current_cost += f * 3
                else:
                    current_cost += 3000 + (f - 1000) * 2
                    
    # 擂台法維護最低花費
    if current_cost < min_total_cost:
        min_total_cost = current_cost

# 步驟 4：輸出最低營運花費
print(min_total_cost)
````

````python
# ==============================================================================
# ⚡ 版本二：APCS 官方實作考場專用版 —— 極簡高效精煉版（進階高手推薦）
# 適用情境：APCS 正式考試現場，極簡 18 行高效最佳化
# ==============================================================================

n, m, k = map(int, input().split())
Q = [list(map(int, input().split())) for _ in range(n)]
ans = float('inf')

for _ in range(k):
    p = list(map(int, input().split()))
    t = [[0] * m for _ in range(m)]
    for i, u in enumerate(p):
        for v, flow in enumerate(Q[i]):
            t[u][v] += flow
            
    cost = sum(f if u == v else (f * 3 if f <= 1000 else 3000 + (f - 1000) * 2)
               for u in range(m) for v, f in enumerate(t[u]) if f)
    if cost < ans:
        ans = cost

print(ans)
````

````python
# ==============================================================================
# 🌐 版本三：ZeroJudge 線上評判萬用 AC 版（支援多筆測資 EOF 循環）
# ==============================================================================
import sys

def solve_traffic_cost(n, m, k, Q, schemes):
    """
    流量成本最佳化核心演算法函式
    :param n: 伺服器數量
    :param m: 基地台數量
    :param k: 方案數量
    :param Q: 伺服器流量矩陣 (n x m)
    :param schemes: 放置方案二維清單 (k x n)
    :return: 方案最低傳輸費用
    """
    min_cost = float('inf')
    for p in schemes:
        t = [[0] * m for _ in range(m)]
        for i, u in enumerate(p):
            for v, flow in enumerate(Q[i]):
                t[u][v] += flow
        
        cost = 0
        for u in range(m):
            for v in range(m):
                f = t[u][v]
                if not f: continue
                if u == v:
                    cost += f
                else:
                    if f <= 1000:
                        cost += f * 3
                    else:
                        cost += 3000 + (f - 1000) * 2
        if cost < min_cost:
            min_cost = cost
            
    return min_cost

# ---------------------------------------------------------------------
# 【ZeroJudge 官方提交程式碼範本】
# 若要在 ZeroJudge 提交，請複製下方函式內容至解題系統：
# def main():
#     tokens = sys.stdin.read().split()
#     if not tokens: return
#     idx = 0
#     while idx < len(tokens):
#         n = int(tokens[idx])
#         m = int(tokens[idx+1])
#         k = int(tokens[idx+2])
#         idx += 3
#         Q = []
#         for _ in range(n):
#             Q.append([int(x) for x in tokens[idx:idx+m]])
#             idx += m
#         schemes = []
#         for _ in range(k):
#             schemes.append([int(x) for x in tokens[idx:idx+n]])
#             idx += n
#         print(solve_traffic_cost(n, m, k, Q, schemes))
# ---------------------------------------------------------------------

# =====================================================================
# 🧪 Colab 本地自動化測試檢驗套件（點擊播放鍵自動執行）
# =====================================================================
test_cases = [
    {
        "name": "官方範例一 (2 伺服器, 3 基地台, 3 方案)",
        "n": 2, "m": 3, "k": 3,
        "Q": [
            [30, 23, 23],
            [5, 25, 3]
        ],
        "schemes": [
            [0, 0],
            [0, 1],
            [0, 2]
        ],
        "expected": 217
    },
    {
        "name": "官方範例二 (3 伺服器, 4 基地台, 5 方案, 超千優惠)",
        "n": 3, "m": 4, "k": 5,
        "Q": [
            [500, 400, 800, 200],
            [500, 400, 100, 600],
            [450, 420, 800, 790]
        ],
        "schemes": [
            [0, 0, 0],
            [0, 1, 2],
            [0, 2, 2],
            [2, 1, 2],
            [1, 1, 1]
        ],
        "expected": 13470
    },
    {
        "name": "邊界極限測資 (1 伺服器, 1 基地台, 1 方案)",
        "n": 1, "m": 1, "k": 1,
        "Q": [[999]],
        "schemes": [[0]],
        "expected": 999
    }
]

print("=== f606. 流量 全自動化測試報告 ===")
all_passed = True
for tc in test_cases:
    actual = solve_traffic_cost(tc["n"], tc["m"], tc["k"], tc["Q"], tc["schemes"])
    passed = (actual == tc["expected"])
    status = "✅ PASS" if passed else "❌ FAIL"
    if not passed:
        all_passed = False
    print(f"{status} | {tc['name']} ➔ 預期: {tc['expected']}, 實際: {actual}")

if all_passed:
    print("\n🎉 恭喜！所有官方與邊界壓力測試案例 100% 全數通過！可安心提交至 APCS 考場與 ZeroJudge！")
else:
    print("\n⚠️ 有測試資料未通過，請檢查矩陣聚合或階梯計費公式！")
````
