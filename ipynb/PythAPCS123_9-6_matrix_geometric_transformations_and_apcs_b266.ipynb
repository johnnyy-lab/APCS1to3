### 9.6.1 矩陣水平翻轉（Horizontal Flip / 左右對調）

在影像處理、相機濾鏡或電玩遊戲中，最常見的視覺特效之一就是「**水平翻轉（左右鏡射）**」。當一張圖片左右顛倒時，原本在畫面左邊的物體會跑到右邊，右邊的物體則跑到左邊。

將水平翻轉放到二維矩陣（$R \times C$）的世界中，讓我們觀察每一格座標的代數變化：
- 原本在第 $r$ 列、第 $c$ 行的元素，翻轉之後：
  - **列索引（Row）保持不變**：依然位於第 $r$ 列！
  - **行索引（Column）左右對調**：原本在第 $0$ 行跑到第 $C - 1$ 行，第 $1$ 行跑到第 $C - 2$ 行……
  - 一般化座標公式：舊位置 $(r, c)$ 會移動到新位置 **$(r, C - 1 - c)$**。

在 Python 中，實現水平翻轉有兩種極具啟發性的思維：
1. **切片語法（Slicing）**：因為每一橫列本身就是一個一維串列，我們可以直接對每一列進行 `row[::-1]` 步進倒序取出，產生全新的翻轉橫列！
2. **雙指標對調（In-place Swap）**：走訪每一列的前半段（$c < C // 2$），將第 $c$ 格與對稱位置第 $C - 1 - c$ 格進行數值交換：
   `row[c], row[C - 1 - c] = row[C - 1 - c], row[c]`。

水平翻轉具有一個極重要的性質：**對同一個矩陣連續做兩次水平翻轉，矩陣會完全復原成原狀**！這在後續逆推演算法中扮演了關鍵角色。


````python
# 範例 9.6.1：使用串列切片語法實現 2x3 矩陣的水平翻轉

demo_m096 = [
    [1, 2, 3],
    [4, 5, 6]
]

demo_R096 = len(demo_m096)
demo_C096 = len(demo_m096[0])

print("原始矩陣：")
for demo_row in demo_m096:
    print(*demo_row)

# 建立翻轉後的新矩陣：每一列都透過 [::-1] 倒序產生
demo_flpph = []
for demo_row in demo_m096:
    demo_flpph.append(demo_row[::-1])

print()
print("水平翻轉後的矩陣：")
for demo_row in demo_flpph:
    print(*demo_row)
````

### 9.6.2 矩陣垂直翻轉（Vertical Flip / 上下顛倒）

學會了左右對調，接下來我們要看另一種常見變換——「**垂直翻轉（上下顛倒）**」，如同湖面上的倒影或鏡子中的倒立影像。

讓我們觀察一個 $R \times C$ 矩陣進行垂直翻轉時的座標特徵：
- 原本在第 $r$ 列、第 $c$ 行的元素，翻轉之後：
  - **行索引（Column）保持不變**：元素依然停留在第 $c$ 直行！
  - **列索引（Row）上下顛倒**：最頂端的第 $0$ 列跑到最底端的第 $R - 1$ 列，第 $1$ 列跑到第 $R - 2$ 列……
  - 一般化座標公式：舊位置 $(r, c)$ 會移動到新位置 **$(R - 1 - r, c)$**。

在 Python 中，垂直翻轉的實作比水平翻轉更加簡潔！因為二維陣列的外層就是「橫列的串列」，所以整張地圖上下倒過來，等價於**把外層串列中的所有橫列倒序排列**：
```python
flipped_v = grid[::-1]  # 只要一行切片！
```
或者利用雙指標原地對調最外層的列物件：
```python
for r in range(R // 2):
    grid[r], grid[R - 1 - r] = grid[R - 1 - r], grid[r]
```
請特別注意：垂直翻轉同樣具有**自逆性（Self-Inverse）**——連續做兩次垂直翻轉，矩陣必定復原！
在 APCS b266 考題中，「操作 1」定義的就是這種翻轉變換。


````python
# 範例 9.6.2：使用切片語法實現 3x3 矩陣的垂直翻轉

demo_mat8 = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print("原始矩陣：")
for demo_row in demo_mat8:
    print(*demo_row)

# 垂直翻轉：直接對外層串列使用 [::-1] 倒序
demo_flppv = demo_mat8[::-1]

print()
print("垂直翻轉（上下顛倒）後的矩陣：")
for demo_row in demo_flppv:
    print(*demo_row)
````

### 9.6.3 矩陣轉置操作（Matrix Transpose / 行列互換）

在線性代數與資料表處理中，有一種極為根本的幾何變換叫做「**轉置（Transpose，記為 $A^T$）**」。

轉置的核心操作是：**將矩陣的橫列（Row）與直行（Column）全面對調**！
- 原本的第一橫列，轉置後變成第一直行；
- 原本的第二橫列，轉置後變成第二直行；
- 依此類推。

這會帶來矩陣結構的兩個重大質變：
1. **尺寸翻轉（Dimension Swap）**：
   如果原本矩陣的大小是 **$R$ 列 $C$ 行**，轉置後的新矩陣大小必定變成 **$C$ 列 $R$ 行**！
   例如：$2 \times 3$ 的矩陣轉置後會變成 $3 \times 2$ 的矩陣。
2. **座標映射（Coordinate Swap）**：
   原本在第 $r$ 列第 $c$ 行的數值，轉置後會跑到新矩陣的第 $c$ 列第 $r$ 行：
   $$new\_grid[c][r] = old\_grid[r][c]$$

實作轉置時，最安全穩健的步驟是：
- 第一步：先量測原始矩陣的尺寸 $R$ 與 $C$。
- 第二步：宣告一個尺寸為 **$C$ 列 $R$ 行** 的全新零矩陣：
  `transposed = [[0] * R for _ in range(C)]`（務必注意列與行的順序互換！）。
- 第三步：用雙重迴圈走訪原始矩陣，將 `grid[r][c]` 填入 `transposed[c][r]`。


````python
# 範例 9.6.3：將 2x3 矩陣轉置為 3x2 矩陣

demo_mat9 = [
    [1, 2, 3],
    [4, 5, 6]
]

demo_R8 = len(demo_mat9)       # 原矩陣列數: 2
demo_C4 = len(demo_mat9[0])    # 原矩陣行數: 3

print(f"原始矩陣（尺寸 {demo_R8} x {demo_C4}）：")
for demo_row in demo_mat9:
    print(*demo_row)

# 建立轉置矩陣：新列數為 C，新行數為 R
demo_trns = [[0] * demo_R8 for _ in range(demo_C4)]

# 雙重走訪填入對應數值
for demo_r in range(demo_R8):
    for demo_c in range(demo_C4):
        demo_trns[demo_c][demo_r] = demo_mat9[demo_r][demo_c]

print()
print(f"轉置後矩陣（尺寸 {demo_C4} x {demo_R8}）：")
for demo_row in demo_trns:
    print(*demo_row)
````

### 9.6.4 矩陣旋轉 90 度（順時針與逆時針）

在手機旋轉螢幕、俄羅斯方塊或地圖旋轉時，最核心的幾何動作就是「**旋轉 90 度**」。

旋轉看似複雜，但只要掌握前兩節學過的「轉置」與「翻轉」，旋轉就變得非常直觀！數學上有個非常優雅的合成定理：
1. **順時針旋轉 90 度（Clockwise 90°）**：
   - 方式 A（複合變換）：**先將矩陣「上下垂直翻轉」，接著再做「轉置」**！
     或者：**先做「轉置」，接著再將每列「左右水平翻轉」**！
   - 方式 B（座標直達）：
     原始 $R \times C$ 矩陣旋轉後變成 $C \times R$ 矩陣。
     原座標 $(r, c)$ 會被旋轉到新座標：
     $$new\_grid[c][R - 1 - r] = old\_grid[r][c]$$
2. **逆時針旋轉 90 度（Counter-Clockwise 90°）**：
   - 方式 A（複合變換）：**先「轉置」，接著再做「上下垂直翻轉」**！
   - 方式 B（座標直達）：
     原座標 $(r, c)$ 旋轉到新座標：
     $$new\_grid[C - 1 - c][r] = old\_grid[r][c]$$

在 **APCS b266 矩陣轉換** 中，「操作 0」定義的是「旋轉」操作。理解了這兩套變換公式，面對考題的幾何旋轉便能胸有成竹。


````python
# 範例 9.6.4：將 2x3 矩陣順時針旋轉 90 度

demo_mat10 = [
    [1, 2, 3],
    [4, 5, 6]
]

demo_R9 = len(demo_mat10)       # 2
demo_C5 = len(demo_mat10[0])    # 3

print("原始矩陣 (2x3)：")
for demo_row in demo_mat10:
    print(*demo_row)

# 旋轉後的新矩陣尺寸必定為 C 列 R 行 (3x2)
demo_rcw = [[0] * demo_R9 for _ in range(demo_C5)]

for demo_r in range(demo_R9):
    for demo_c in range(demo_C5):
        # 順時針旋轉座標公式：新列為 c，新行為 R - 1 - r
        demo_rcw[demo_c][demo_R9 - 1 - demo_r] = demo_mat10[demo_r][demo_c]

print()
print("順時針旋轉 90 度後 (3x2)：")
for demo_row in demo_rcw:
    print(*demo_row)
````

### 9.6.5 連續操作與變數尺寸動態追蹤

在真實的幾何操作競賽題目中，題目很少只要求做「一次」旋轉或翻轉，而是會給定一個操作指令序列（例如先旋轉、再翻轉、又旋轉、再旋轉……）。

這時，許多初學者會踩進一個非常隱蔽的「**維度僵化陷阱**」：
在程式最開始，我們用 `R, C = map(int, input().split())` 讀取了初始矩陣的長寬。
但只要程式執行了一次「旋轉」或「轉置」，矩陣的列數與行數就**互相交換了**！
如果下一個操作依然沿用舊的 `R` 和 `C` 變數去跑迴圈，就會立刻觸發 `IndexError` 索引越界或產生扭曲的形變！

**動態追蹤法則（Dynamic Dimension Tracking）**：
每次完成任何幾何變換並更新 `grid` 之後，必須立刻重新量測當前的真實尺寸：
```python
# 每次操作完後，立即更新當前長寬：
R = len(grid)
C = len(grid[0])
```
只要保持這個好習慣，無論矩陣經過多少次不同方向的變換、長方形如何翻滾旋轉，程式的迴圈邊界永遠都能百分之百精準契合當下的陣列尺寸。


````python
# 範例 9.6.5：連續執行多個幾何操作並動態維護矩陣尺寸

demo_gd096 = [
    [1, 2, 3],
    [4, 5, 6]
]

# 操作指令：0 代表順時針旋轉 90 度，1 代表上下垂直翻轉
demo_cmmn = [0, 1, 0]

print("初始尺寸：", len(demo_gd096), "x", len(demo_gd096[0]))

demo_s096 = 1
for demo_cmd in demo_cmmn:
    demo_R = len(demo_gd096)
    demo_C = len(demo_gd096[0])
    
    if demo_cmd == 0:
        # 順時針旋轉 90 度 (尺寸變為 C x R)
        demo_new_grid = [[0] * demo_R for _ in range(demo_C)]
        for demo_r in range(demo_R):
            for demo_c in range(demo_C):
                demo_new_grid[demo_c][demo_R - 1 - demo_r] = demo_gd096[demo_r][demo_c]
        demo_gd096 = demo_new_grid
        demo_op_name = "順時針旋轉 90 度"
    elif demo_cmd == 1:
        # 垂直翻轉 (尺寸保持 R x C)
        demo_gd096 = demo_gd096[::-1]
        demo_op_name = "垂直翻轉"
        
    print(f"步驟 {demo_s096}（{demo_op_name}）後尺寸變為：{len(demo_gd096)} x {len(demo_gd096[0])}")
    demo_s096 += 1

print()
print("最終矩陣：")
for demo_row in demo_gd096:
    print(*demo_row)
````

### 9.6.6 APCS b266 矩陣轉換：逆操作還原心法（APCS 實戰原型）

現在，我們來到本單元的高峰——**APCS 實作真題 b266 矩陣轉換**！

#### 📜 題目情境與核心難題：
題目告訴我們：原本有一個原始矩陣，經過了一連串的「操作指令」（0 代表旋轉，1 代表翻轉）之後，變成了目前的「最終矩陣」。
題目的目標是：**給定最終矩陣與操作紀錄，求回最初的「原始矩陣」**！

這是一個典型的「**逆推問題（Reverse Engineering）**」！
要逆推回原始狀態，我們必須掌握兩大逆推心法：
1. **指令順序倒著做（Reverse Command Order）**：
   想像你出門的動作是：先穿襪子，再穿鞋子。
   當你回家要復原時，順序必然相反：**必須先脫鞋子，才能脫襪子**！
   因此，原本的操作序列若是 `[op1, op2, op3]`，我們要逆推，就必須由後往前依序執行：
   `for op in reversed(ops):` 或 `for op in ops[::-1]:`。
2. **單一步驟做反向操作（Inverse Operation）**：
   - **操作 1（翻轉）的反向操作**：
     翻轉具有自逆性（翻兩次就回到原狀），所以翻轉的反向操作就是「**翻轉自己**」！
   - **操作 0（旋轉）的反向操作**：
     APCS b266 題目的正向操作 0 定義為「順時針旋轉 90 度並翻轉」或「旋轉」。在官方測資規範中：
     正向若是「逆時針旋轉 90 度」，則逆向就是「順時針旋轉 90 度」；
     若正向為「順時針旋轉 90 度」，則逆推時就必須執行「**逆時針旋轉 90 度**」！
     （或者利用特性：逆時針旋轉 90 度 等價於 連續順時針旋轉 3 次！）。

掌握了「倒序讀取指令」與「執行反向幾何操作」，整道 APCS 中級真題便迎刃而解！


````python
# 範例 9.6.6：APCS b266 逆操作還原演示

# 假設最後的矩陣為：
demo_fnlgd = [
    [6, 3],
    [5, 2],
    [4, 1]
]

# 原本正向執行的操作是：先操作 0 (順時針 90 度)，再操作 1 (垂直翻轉)
demo_oops = [0, 1]

print("最終矩陣：")
for demo_row in demo_fnlgd:
    print(*demo_row)

# 逆推核心第一步：指令序列倒序！
demo_rops = demo_oops[::-1]  # 變成 [1, 0]
demo_cg096 = demo_fnlgd

for demo_op in demo_rops:
    demo_curr_R = len(demo_cg096)
    demo_curr_C = len(demo_cg096[0])
    
    if demo_op == 1:
        # 操作 1 的逆操作是：垂直翻轉本身！
        demo_cg096 = demo_cg096[::-1]
    elif demo_op == 0:
        # 正向若為順時針旋轉 90 度，逆操作就是「逆時針旋轉 90 度」
        # 尺寸變為 curr_C x curr_R，新座標為 (curr_C - 1 - c, r)
        demo_inv_grid = [[0] * demo_curr_R for _ in range(demo_curr_C)]
        for demo_r in range(demo_curr_R):
            for demo_c in range(demo_curr_C):
                demo_inv_grid[demo_curr_C - 1 - demo_c][demo_r] = demo_cg096[demo_r][demo_c]
        demo_cg096 = demo_inv_grid

print()
print("還原後的原始矩陣：")
for demo_row in demo_cg096:
    print(*demo_row)
````

## 本單元重點回顧與核心心法

在單元 9-6 中，我們完整掌握了二維矩陣的幾何變換技術，並成功征服了 APCS b266 實作真題的核心逆推思維：

1. **水平翻轉（左右對調）**：
   - 座標對應：$(r, c) \rightarrow (r, C - 1 - c)$。
   - 語法心法：`[row[::-1] for row in grid]`。
2. **垂直翻轉（上下顛倒）**：
   - 座標對應：$(r, c) \rightarrow (R - 1 - r, c)$。
   - 語法心法：`grid[::-1]`。
3. **矩陣轉置（行列互換）**：
   - 維度變化：$R \times C \rightarrow C \times R$。
   - 座標對應：$new\_grid[c][r] = old\_grid[r][c]$。
4. **旋轉 90 度**：
   - 順時針：先上下翻轉再轉置，座標為 $new\_grid[c][R - 1 - r] = old\_grid[r][c]$。
   - 逆時針：先轉置再上下翻轉，座標為 $new\_grid[C - 1 - c][r] = old\_grid[r][c]$。
5. **動態尺寸追蹤法則**：
   - 任何涉及旋轉或轉置的操作後，長寬會對調，必須隨時以 `R, C = len(grid), len(grid[0])` 更新維度變數。
6. **APCS b266 逆推雙大心法**：
   - **指令順序倒序**：`for op in ops[::-1]:`。
   - **單步動作反向**：翻轉的逆操作為翻轉；順時針旋轉的逆操作為逆時針旋轉。
