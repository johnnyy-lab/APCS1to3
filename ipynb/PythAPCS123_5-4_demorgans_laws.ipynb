### 🧩 5-4-1 口語否定與數理邏輯陷阱——「非（A 且 B）」的思維迷宮

**生活比喻：健康檢查的「非雙重異常」判定**  
某體檢標準規定：「*若受檢者『並非（血壓超標 且 血糖超標）』，即屬於非高危險族群。*」  
許多初學同學常會誤以為這句話的意思是：「血壓正常，同時血糖也必須正常」。  
但請仔細推敲：
- 如果一個人「血壓超標，但血糖正常」，他算不算「血壓與血糖雙雙超標」？不算！所以他符合「並非雙雙超標」！
- 只有當一個人**兩項指標同時爆表**時，「血壓超標 且 血糖超標」才成立；只要其中一項指標正常，他就脫離了雙重高危險名單！

**數理邏輯與口語的鴻溝：**
在邏輯表示式中：
```python
not (A and B)
```
代表對「$A$ 與 $B$ 同時成立」這整件事進行否定。  
依據真值表，只要 $A$ 與 $B$ 之中**有任何一個不成立**（或者兩個都不成立），整體就是 `True`！

**真值表驗證：**
| $A$ | $B$ | $A 	ext{ and } B$ | $	ext{not } (A 	ext{ and } B)$ | 狀態解析 |
| :---: | :---: | :---: | :---: | :--- |
| `True` | `True` | `True` | **`False`** | 雙雙成立，否定後為 False |
| `True` | `False` | `False` | **`True`** | 右側打破，否定後為 True |
| `False` | `True` | `False` | **`True`** | 左側打破，否定後為 True |
| `False` | `False` | `False` | **`True`** | 雙雙打破，否定後為 True |

**APCS 考試實務建議：**
審題時遇到題目敘述包含「非同時」、「不可兼具」等字眼時，切勿直接寫成 `not A and not B`，務必釐清否定所涵蓋的完整語意範疇。


````python
# [2] Code 範例區：展示 not (A and B) 的真值運算與口語陷阱

# 1. 設定測試情境：A 代表風大，B 代表雨大
is_windy = True
is_rainy = False

# 檢驗「並非風大且雨大」
both_extreme = is_windy and is_rainy             # True and False -> False
not_both_extreme = not (is_windy and is_rainy)   # not False -> True

print("風勢是否強大：", is_windy)
print("雨勢是否強大：", is_rainy)
print("是否風大且雨大（同時具備）：", both_extreme)        # False
print("是否並非風大且雨大 not (A and B)：", not_both_extreme) # True

# 2. 🚨 經典錯誤對照：誤寫為 not A and not B
wrong_translation = (not is_windy) and (not is_rainy)
print("--- 語意錯誤寫法對照 ---")
print("錯誤直譯 (not A and not B) 結果：", wrong_translation) # False！(與原意完全相悖！)
````

### 🔄 5-4-2 笛摩根第一定律——not (A and B) 等價於 (not A) or (not B)

**生活比喻：拆解括號的連鎖魔法**  
想像你手中有一把帶有負號的魔法錘，當你對著 `not (A and B)` 敲下去時，會發生三個連鎖反應：
1. `not` 飛進去附著在 $A$ 身上，變成 `(not A)`。
2. `not` 也飛進去附著在 $B$ 身上，變成 `(not B)`。
3. **最關鍵的一步：中間的 `and` 被徹底翻轉成了 `or`！**

這就是大名鼎鼎的**笛摩根第一定律（De Morgan's First Law）**（吳邦一老師講義第 18 頁）：
$$	ext{not } (A 	ext{ and } B) \iff (	ext{not } A) 	ext{ or } (	ext{not } B)$$

**真值表完整對照推導：**
| $A$ | $B$ | $	ext{not } (A 	ext{ and } B)$ | $	ext{not } A$ | $	ext{not } B$ | $(	ext{not } A) 	ext{ or } (	ext{not } B)$ |
| :---: | :---: | :---: | :---: | :---: | :---: |
| `True` | `True` | **`False`** | `False` | `False` | **`False`** |
| `True` | `False` | **`True`** | `False` | `True` | **`True`** |
| `False` | `True` | **`True`** | `True` | `False` | **`True`** |
| `False` | `False` | **`True`** | `True` | `True` | **`True`** |

每一行的結果都**百分之百精確吻合**！

**為什麼這項定律如此神聖？**
在實際寫程式時，人腦閱讀帶有外層 `not` 的複合條件時極度容易打結。透過笛摩根第一定律，我們能將「全體否定」拆解為「任一項未達標即成立」的明快分支，大幅提高代碼可維護性。

**APCS 考試實務建議：**
在 APCS 觀念題中，經常給定一個複雜的 `not (x < y and z == 0)`，要求考生在選項中選出等價的表示式。掌握「否定進去、and 變 or」的原則，三秒鐘即可選出正確答案！


````python
# [2] Code 範例區：驗證笛摩根第一定律 not (A and B) == (not A) or (not B)

# 1. 吳邦一老師講義第 18 頁對稱驗證
A = True
B = False

expr_left = not (A and B)
expr_right = (not A) or (not B)

print("A =", A, ", B =", B)
print("左式 not (A and B) 結果：", expr_left)        # True
print("右式 (not A) or (not B) 結果：", expr_right)    # True
print("兩式是否恆等（等價）：", expr_left == expr_right) # True

# 2. 數值情境實證：not (x > 0 and y > 0) <=> (x <= 0 or y <= 0)
x = 5
y = -3
rule_left = not (x > 0 and y > 0)
rule_right = (x <= 0) or (y <= 0)
print("--- 數值情境等價檢驗 ---")
print("not (x > 0 and y > 0) 結果：", rule_left)   # True
print("(x <= 0 or y <= 0) 結果：", rule_right)      # True
print("兩數值條件式是否完全等價：", rule_left == rule_right) # True
````

### 🔄 5-4-3 笛摩根第二定律——not (A or B) 等價於 (not A) and (not B)

**生活比喻：雙重否定的「兩者皆非」**  
想像你去餐廳點餐，服務生問你是否要加點可樂或雪碧。你搖搖頭說：「*我『並非（要可樂 或 要雪碧）』。*」  
這句話的意思是什麼？  
這代表你**既不要可樂，而且也不要雪碧**！兩樣汽水你全都拒絕！

這正是**笛摩根第二定律（De Morgan's Second Law）**（吳邦一老師講義第 18 頁）：
$$	ext{not } (A 	ext{ or } B) \iff (	ext{not } A) 	ext{ and } (	ext{not } B)$$

**核心機制：**
當 `not` 推進到 `(A or B)` 內部時：
1. $A$ 變成了 `(not A)`。
2. $B$ 變成了 `(not B)`。
3. **中間的 `or` 被徹底翻轉成了 `and`！**

**真值表完整對照推導：**
| $A$ | $B$ | $A 	ext{ or } B$ | $	ext{not } (A 	ext{ or } B)$ | $(	ext{not } A) 	ext{ and } (	ext{not } B)$ |
| :---: | :---: | :---: | :---: | :---: |
| `True` | `True` | `True` | **`False`** | **`False`** |
| `True` | `False` | `True` | **`False`** | **`False`** |
| `False` | `True` | `True` | **`False`** | **`False`** |
| `False` | `False` | `False` | **`True`** | **`True`** |

只有在兩者皆為 `False` 時，整體才是 `True`！

**常見錯誤陷阱：**
- **忘記翻轉連接符號**：初學者最常見的筆誤就是把 `not (A or B)` 改寫成 `not A or not B`（忘記把 or 改為 and）。請牢記口訣：**「not 推進去，符號全顛倒（and 變 or，or 變 and）」**！

**APCS 考試實務建議：**
在演算法中，如果想排除多種異常狀況（例如「不可出界且不可撞牆」），寫成 `(not 出界) and (not 撞牆)` 往往比 `not (出界 or 撞牆)` 更利於逐步除錯與定位問題。


````python
# [2] Code 範例區：驗證笛摩根第二定律 not (A or B) == (not A) and (not B)

# 1. 吳邦一老師講義第 18 頁對稱驗證
A = False
B = False

expr_left = not (A or B)
expr_right = (not A) and (not B)

print("A =", A, ", B =", B)
print("左式 not (A or B) 結果：", expr_left)          # True
print("右式 (not A) and (not B) 結果：", expr_right)    # True
print("兩式是否恆等（等價）：", expr_left == expr_right)   # True

# 2. 實務情境：棋子既不出界也不在障礙物上
is_out_of_bounds = False
is_on_obstacle = True

safe_left = not (is_out_of_bounds or is_on_obstacle)
safe_right = (not is_out_of_bounds) and (not is_on_obstacle)
print("--- 棋盤安全位置檢驗 ---")
print("左式安全性判定：", safe_left)   # False
print("右式安全性判定：", safe_right)  # False
print("兩者是否完全一致：", safe_left == safe_right) # True
````

### ⚖️ 5-4-4 比較運算子的對偶反轉——徹底消除比較式外層的 not

**生活比喻：拍照的反轉濾鏡直接換成正向鏡頭**  
如果一個人的身高「不是大於 180 公分」，那他的身高到底怎麼樣？  
就是「小於或等於 180 公分」！  
我們根本不需要先去查他是否大於 180 再在外面套一層 `not`，直接換個符號就一清二楚了。

**比較運算子的對偶互補表（Complementary Pairs）：**
在實務編程中，**最好的 `not` 就是「沒有 `not`」**！  
只要遇到單一比較式外層包裹著 `not`，都可以直接透過反轉比較符號將其徹底消滅：

| 原式（帶有 not） | 徹底消滅 not 後的正向對偶式 | 關鍵注意事項 |
| :---: | :---: | :--- |
| `not (a > b)` | **`a <= b`** | 嚴格大於的反面是「小於或等於」！ |
| `not (a < b)` | **`a >= b`** | 嚴格小於的反面是「大於或等於」！ |
| `not (a >= b)` | **`a < b`** | 包含等於的反面是「嚴格小於」！ |
| `not (a <= b)` | **`a > b`** | 包含等於的反面是「嚴格大於」！ |
| `not (a == b)` | **`a != b`** | 等於的反面是不等於！ |
| `not (a != b)` | **`a == b`** | 不等於的反面是等於！ |

**致命錯誤陷阱（等號端點遺漏）：**
初學者最常犯的粗心毛病：把 `not (a > b)` 寫成 `a < b`！  
請務必警惕：**如果原式「沒有等號」，反轉後就「必須補上等號」；如果原式「有等號」，反轉後就「必須拔掉等號」！**

**APCS 考試實務建議：**
消去外層的 `not` 是高階程式重構與競賽解題的基本功。去除否定句後，條件意圖更明確，執行效率更高，邏輯也更不易出錯。


````python
# [2] Code 範例區：展示比較運算子的對偶反轉消去 not

a = 10
b = 10

# 1. not (a > b) 等價於 a <= b
print("--- 檢驗臨界端點（a=10, b=10） ---")
print("not (a > b) 結果：", not (a > b))  # not False -> True
print("a <= b       結果：", a <= b)        # True (剛好相等合格！)
print("兩者是否完全一致：", not (a > b) == (a <= b)) # True

# 2. not (a == b) 等價於 a != b
print("--- 等號與不等號反轉 ---")
print("not (a == b) 結果：", not (a == b)) # False
print("a != b       結果：", a != b)       # False
````

````python
# ==========================================
# [3] Code 填空題
# 任務說明：
# 消除 not 的正向改寫：
# 請將 not (score < 60) 消除 not，改寫為正向的及格判定！
# ==========================================

score = 60

# 原式
failing_negated = not (score < 60)

# 提示：嚴格小於 (<) 的反面是大於等於 (>=)
passing_positive = (score ___ 60)

print("原式 not (score < 60) 結果：", failing_negated)
print("改寫 (score >= 60) 結果：", passing_positive)
print("兩式是否完全相等：", failing_negated == passing_positive)
````

### 🚀 5-4-5 複合條件化簡實戰——將晦澀否定句改寫為明快正向邏輯

**生活比喻：雜亂合約條款的明文化整理**  
在法律合約中，如果寫著：「*乙方非不可不履行未經甲方未同意之條款*」，相信法官看了一定直搖頭！優秀的律師會將層層嵌套的雙重否定全部化簡為「*乙方應履行甲方同意之條款*」。  
在撰寫 APCS 程式碼時，我們也必須具備這種**「邏輯化簡大師」**的實戰能力！

**化簡綜合大三通（三步絕招）：**
面對複雜的 `not (條件A and 條件B)`：
1. **第一步（笛摩根展開）**：將 `not` 推進去，把 `and` 翻轉成 `or`。
2. **第二步（比較子反轉）**：將 `not (x > 0)` 直接化簡為 `x <= 0`。
3. **第三步（格式整理）**：加上清晰的小括號，形成一目了然的正向條件！

**實戰化簡範例：**
原式：
```python
not (0 <= score and score <= 100)
```
- 第一步展開：`(not 0 <= score) or (not score <= 100)`
- 第二步反轉比較子：`(score < 0) or (score > 100)`
化簡後的式子「**小於 0 或者 大於 100（代表異常超出範圍）**」，比起原式的雙重否定，在直覺與運算上清晰百倍！

**APCS 考試實務建議：**
在 APCS 與各類題庫中，經常需要篩選「異常值」或「合法範圍外的值」。熟練運用笛摩根定律將區間檢驗自由切換為邊界排除，是高分選手的必備素養。


````python
# [2] Code 範例區：展示複雜複合邏輯式的笛摩根化簡實戰

score = 105

# 原式：不合法成績（對合法區間 0 <= score <= 100 進行否定）
raw_rule = not (score >= 0 and score <= 100)

# 化簡後：利用笛摩根定律與比較子反轉，改寫為 (score < 0 or score > 100)
simplified_rule = (score < 0) or (score > 100)

print("測試成績：", score)
print("原式 not (score >= 0 and score <= 100) 判定：", raw_rule)       # True
print("化簡式 (score < 0 or score > 100) 判定：", simplified_rule)   # True
print("兩式是否完全等價：", raw_rule == simplified_rule)                # True
````

````python
# ==========================================
# [3] Code 填空題
# 任務說明：
# 數值不在 (10, 50) 開區間內的判定改寫：
# 原式為：not (10 < x and x < 50)。
# 請填入正確的比較符號與邏輯運算子，將其化簡為以 or 串接的形式！
# ==========================================

x = 5

# 原式
raw_expr = not (10 < x and x < 50)

# 提示：10 < x 的反面是 x <= 10；x < 50 的反面是 x >= 50
simplified_expr = (x ___ 10) ___ (x ___ 50)

print("原式判定結果：", raw_expr)
print("化簡判定結果：", simplified_expr)
print("兩式是否完全一致：", raw_expr == simplified_expr)
````
