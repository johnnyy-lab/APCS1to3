### 🛑 4-3-1 認識結尾參數 end——揭開自動換行 \n 的秘密

想像一下古時候的打字機，作家每敲完一行字，打字機最右側的鈴鐺就會「叮！」一聲響起，接著打字機的滑車會自動「回車並換到下一行（Carriage Return & Line Feed）」。在 Python 魔法世界中，`print()` 函數就是這台自帶鈴鐺的打字機！每當它把小括號裡的所有寶物唸完之後，它一定會習慣性地在句尾附送一個「換行符號」，這個負責在結尾蓋章的守門員，就是 `end` 參數！

**運作機制與底層原理：**
在 Python 的標準定義中，`print()` 的完整語法原型其實包含了預設值：
`print(*objects, sep=' ', end='\n')`
其中，`end` 是英文單字 **end**（結尾、終點）的意思。
直譯器在執行 `print()` 時，固定遵循以下兩大步驟：
1. **輸出所有主要內容**：依序印出括號內的所有資料項目（項目間以 `sep` 間隔）。
2. **輸出結尾符號**：在所有資料印完之後，**追加輸出 `end` 所指定的內容**。
在預設情況下，Python 悄悄幫我們設定了 `end='\n'`（`\n` 就是換行符號 Newline）。這也就是為什麼你每次呼叫一次 `print()`，游標就會自動跳到下一行開頭的原因！
如果你明確寫出 `end='\n'`，例如 `print("早安", end='\n')`，它的執行結果與完全不寫 `end` 是一模一樣的。

**常見錯誤陷阱（初學者必讀）：**
1. **把 end 誤當成隊伍中間的間隔**：很多國一同學會把 `sep` 和 `end` 搞混！請記住黃金口訣：「**`sep` 在中間切開，`end` 在屁股收尾**」！如果你寫 `print("A", "B", end=',')`，逗號只會出現在 `B` 的後面（印出 `A B,`），而 `A` 和 `B` 中間依然是預設空格！
2. **拼字錯誤**：有些同學會把 `end` 拼成 `ends`、`ending` 或 `tail`。一旦拼錯，Python 會立刻報出：`TypeError: 'ending' is an invalid keyword argument for print()`！
3. **反斜線打成正斜線**：換行符號是「反斜線」`'\n'`，絕不能打成除法正斜線 `'/n'`，否則螢幕上不會換行，而是會多印出一個字母 n！

**APCS 考試實務建議：**
在 APCS 程式檢定中，絕大多數題目都要求每筆測資的答案「結尾必須正常換行」。理解 `print()` 預設結尾自帶 `\n`，可以讓你在輸出單行答案時放心地直接使用 `print(ans)`，完全不用多此一舉手動加換行，避免造成多餘空行的格式錯誤！


````python
# [2] Code 範例區：認識結尾參數 end 與預設換行

# 範例 1：不寫 end（採用預設 end='\n'，兩次輸出自動分行）
print("第一行訊息：校長致詞完畢。")
print("第二行訊息：典禮正式開始！")

# 範例 2：明確寫出 end='\n'（效果與範例 1 完全一模一樣）
print("第一行訊息：晴空萬里。", end='\n')
print("第二行訊息：微風徐徐。", end='\n')

# 範例 3：觀察 end 的位置是在「整句話的最後末端」
# 注意：項目之間由 sep 負責，整句最後由 end 負責！
print("甲", "乙", "丙", sep='-', end='\n')
print("丁", "戊", "己", sep='-', end='\n')
````

### 🚂 4-3-2 取消換行緊密相連——空字串 end='' 接續輸出

想像一列行駛中的火車。第一節火車車廂進站停靠時，站長沒有把鐵軌切換到第二月台，而是讓第二節車廂直接緊緊掛在第一節車廂的後方，組裝成同一列超長列車！
在 Python 程式中，如果我們希望第一個 `print()` 說完話之後「不要換行」，讓第二個 `print()` 的發言直接「緊貼在同一行的後面」，這時候只要請出我們的老朋友——**空字串 `end=''`**！

**運作機制與底層原理：**
當我們在 `print()` 中指定 `end=''` 時，大喇叭在講完這句話後，追加輸出的字元數量為 0（裡面沒有任何空白，也沒有換行符號 `\n`）！
這意味著：
1. **游標停留在原地**：螢幕上的游標在印完第一段文字後，會「原地立正等待」，不會跳到下一行。
2. **下一道指令無縫接續**：當程式執行到下一個 `print()` 指令時，輸出的內容就會直接從游標停留的位置「無縫接軌」印出來！
3. **分段組裝長句或進度提示**：例如先印出提示前綴 `print("處理進度：", end='')`，接著經過變數計算後再印出 `print(100, "%")`，兩者就會完美組裝在同一行：`處理進度：100 %`！

**常見錯誤陷阱（初學者必讀）：**
1. **誤把 end='' 當成 sep=''**：
   - `sep=''`：控制「同一個 `print()` 括號內部」多個項目之間的間隔！
   - `end=''`：控制「這次 `print()`」與「下一次 `print()`」之間的結尾銜接！
   如果你在第二個 `print()` 寫了 `end=''`，但它後面根本沒有第三個 `print()`，那它的效果就只是讓程式結束時游標不換行而已。
2. **忘記在真正結束時換行**：如果在很多次 `end=''` 的接續輸出之後，最後一次輸出依然沒有換行，螢幕上的輸出可能會和終端機的提示符號（如 Colab 的執行輸出提示）擠在同一行，影響視覺美觀。

**APCS 考試實務建議：**
在 APCS 解題時，有些題目會分階段計算出各個數值或字元。學會使用 `end=''` 可以讓你在多個連續步驟中逐步把答案拼在同一行輸出，不需要大費周章在記憶體中反覆串接長字串，大幅提升程式執行效能！


````python
# [2] Code 範例區：使用 end='' 取消換行並接續輸出

# 範例 1：連續兩次 print()，文字完美接在同一行
print("載入中", end='')
print("...完成！")          # 輸出：載入中...完成！

# 範例 2：分段拼接數值與單位
print("本次測驗得分：", end='')
score = 95
print(score)                # 輸出：本次測驗得分：95

# 範例 3：連續拼裝三個數字代碼，形成完整編號
print("A", end='')
print(7, end='')
print(9)                    # 輸出：A79
````

### 🎈 4-3-3 自訂結尾間隔符號——空格 end=' ' 與標點符號收尾

在排隊買電影票時，每位觀眾買完票後，不會立刻離開大廳，而是在隊伍後方留出「一步的距離（空格）」，讓下一位觀眾能整齊排在他後面，所有人依然維持在同一個橫列隊伍中！
如果我們希望每一次呼叫 `print()` 時「不換行」，而且在結尾「留一個空格」或是「加上逗號、箭頭」，好讓下一個 `print()` 接在後面時自然隔開，這時候就可以將 `end` 設定為特定字元！

**運作機制與底層原理：**
`end` 參數跟 `sep` 一樣，可以接受**任意字串**！
1. **以單一空格結尾 `end=' '`**：
   - 這是競技程式中最經典的用法！
   - 當你連續多次呼叫 `print(x, end=' ')` 時，所有印出的數字或文字都會排列在同一行，並且彼此之間以乾淨的單一空格整齊分開！
2. **以標點符號結尾（如 `end=','` 或 `end=' -> '`）**：
   - 第一次 `print("起點", end=' -> ')`，接著第二次 `print("終點")`，整行輸出就會自動變成 `起點 -> 終點`！
   - 這樣不需要在一開始就把所有資料塞進同一個 `print()`，而是可以在程式執行的各個不同階段，隨時往同一行的末端追加內容與裝飾符號。

**常見錯誤陷阱（初學者必讀）：**
1. **最後一個項目多印了結尾空格或逗號**：
   如果在輸出最後一個項目時依然設定了 `end=' '` 或 `end=','`，那麼整行輸出的最右邊就會留下一顆「多餘的尾巴」！在要求極度嚴苛的自動評判（OJ）中，末端多餘的逗號常會導致 **Presentation Error (PE)** 或 **Wrong Answer (WA)**！因此在印出最後一個項目時，務必記得恢復預設換行或拿掉逗號！
2. **引號打成全形**：若設定逗號結尾，請確保輸入的是半形 `' '` 或 `','`，不要誤按成全形符號。

**APCS 考試實務建議：**
在 APCS 考題中，當題目要求輸出橫向數列（例如動態計算並連續輸出數值）時，`end=' '` 是最核心的必備技巧。掌握用 `end=' '` 控制單行推進，是每位程式選手打好基本功的必經之路！


````python
# [2] Code 範例區：自訂結尾字元 end=' ' 與符號

# 範例 1：使用 end=' ' 讓多次輸出的數值橫向並列
print(10, end=' ')
print(20, end=' ')
print(30)                   # 最後一個恢復換行，輸出：10 20 30

# 範例 2：使用箭頭 end=' -> ' 串聯不同階段
print("第一步: 備料", end=' -> ')
print("第二步: 熱鍋", end=' -> ')
print("第三步: 出菜！")

# 範例 3：使用逗號 end=',' 橫向連接水果清單
print("蘋果", end=',')
print("香蕉", end=',')
print("芒果")
````

### 📄 4-3-4 結尾的雙重換行與段落拉開——end='\n\n' 大段落排版術

想像你正在排版一本精美的校刊雜誌。如果標題、作者名稱與內文全部擠在一起，密密麻麻的文字會讓讀者眼睛非常疲憊。專業的排版設計師會在「章節大標題」的下方刻意「多空出一行空白行」，讓視覺有一種呼吸與沉澱的開闊感！
在前面的 4-1 單元中，我們學過可以用不帶引數的 `print()` 來印出空行。但是，如果你想在印完某段文字後「自動拉開一段大距離」，其實只需要命令大喇叭連敲兩次換行鐘——**`end='\n\n'`**！

**運作機制與底層原理：**
當我們設定 `end='\n\n'` 時，字串中包含了兩個連續的換行跳脫字元：
1. **第一個 `\n`**：負責讓目前這一行結束，將游標跳到下一行開頭。
2. **第二個 `\n`**：再次觸發換行動作，將游標再往下推進一行！
如此一來，目前的文字與下一行即將輸出的文字之間，就會自動留出一行乾淨俐落的「空白行（Blank Line）」，達成優雅的段落隔離效果！
除了雙換行之外，你甚至可以在 `end` 中放進裝飾分隔線，例如 `end='\n--------------------\n'`，這樣印完該段訊息後，下方就會自動劃出一道整整齊齊的分隔線，並自動換好行準備迎接下一段！

**常見錯誤陷阱（初學者必讀）：**
1. **反斜線漏打或打反**：再次提醒，`\n` 必須使用反斜線！如果寫成 `end='/n/n'`，螢幕上會搞笑地印出 `/n/n` 兩個文字，不但不會出現空行，還會破壞原本的版面！
2. **APCS 題目的空行地雷**：在日常終端機應用與作業報告中，雙換行排版非常實用且美觀；但是**在 APCS 正式檢定題目中，題目沒有要求印空行時，絕對嚴禁擅自加入 `end='\n\n'`**！因為評判系統會逐字元嚴格比對，一旦偵測到多餘的空白行，會毫不留情地給出 **Wrong Answer (WA)**！請一定要看清題目的輸出規範！

**APCS 考試實務建議：**
掌握 `end='\n\n'` 能讓你在撰寫系統報表、除錯紀錄（Debug Logs）或個人專案時，代碼行數大幅縮減，不用到處插入多餘的 `print()` 空行，保持程式碼的結構精煉優美！


````python
# [2] Code 範例區：雙重換行與段落分隔排版

# 範例 1：使用 end='\n\n' 自動在標題下方空出一行
print("【旗美高中段考英雄榜】", end='\n\n')
print("第一名：王小明同學")

# 範例 2：比較「單換行」與「雙換行」的視覺間距
print("段落 A：第一小節結束。", end='\n')
print("段落 B：緊接著開始。")

print("--- 我是對比線 ---")

print("段落 C：大章節結束。", end='\n\n')
print("段落 D：新章節開展（上方多了一行空白行）。")

# 範例 3：在 end 中直接自帶裝飾分隔線
print("訂單明細確認中...", end='\n====================\n')
print("訂單已成立，感謝您的購買！")
````

### ⚔️ 4-3-5 綜合雙劍合璧——sep 與 end 同時登場的精密排版術

現在，你已經同時擁有了控制大喇叭的兩把絕世寶劍：
- **`sep`（隊伍內部的分隔刃）**：負責在同一個括號裡，把每個相鄰的寶物精準切開、塞入間隔符號。
- **`end`（隊伍末端的收尾刀）**：負責在整批寶物全部播報完畢後，決定最後一個標點、空格或是否換行。
當這兩把寶劍在同一個 `print()` 指令中合體時，你就能發揮出百分之百的「高精度排版魔法」！無論多麼刁鑽的輸出格式，都能輕鬆一行搞定！

**運作機制與底層原理：**
當我們在一個 `print()` 中同時寫下 `sep` 與 `end` 時，例如：
`print("x", "y", "z", sep=', ', end=' ; ')`
Python 的直譯器會精確按照先後順序執行：
1. 先印出第一個項目 `"x"`。
2. 插入 `sep` 指定的字串 `", "`。
3. 印出第二個項目 `"y"`。
4. 插入 `sep` 指定的字串 `", "`。
5. 印出第三個項目 `"z"`。
6. **所有項目印完，最後印出 `end` 指定的字串 `" ; "`**（游標停留在分號後方，不換行）！
螢幕上就會呈現：`x, y, z ; `。

**常見錯誤陷阱（初學者必讀）：**
1. **參數擺放先後順序會不會影響結果？**：
   在 Python 中，關鍵字參數 `sep` 與 `end` 誰寫在前面都可以！例如 `print(1, 2, sep=',', end='.')` 與 `print(1, 2, end='.', sep=',')` 的執行結果是完全一模一樣的。但是，**它們倆都必須放在所有要輸出的資料項目的最後面**！
2. **分不清最後一項後面出現的是 sep 還是 end**：
   許多同學會困惑：「最後一項後面會不會也有 `sep`？」答案是：**絕對不會！** `sep` 只存在於項目與項目「之間」，最後一個項目後面跟著的永遠只有 `end`！

**APCS 考試實務建議：**
在 APCS 考題中，二維座標點輸出（如 `(x, y)`）、帶有前綴後綴的矩陣元素或特殊資料流輸出，往往需要同時微調項目間隔與結尾換行。掌握 `sep` 與 `end` 的雙劍合璧，能讓你在考場面對任何複雜排版要求時，都能氣定神閒、一發入魂！


````python
# [2] Code 範例區：sep 與 end 雙劍合璧精密排版

# 範例 1：以逗號隔開數值，結尾加上句點且不換行
print(10, 20, 30, sep=', ', end='.')
print(" (共計 3 筆資料)")      # 接續在句點後面輸出

# 範例 2：座標格式排版
x = 5
y = 12
print("座標點：(", end='')
print(x, y, sep=', ', end=')\n')  # 中間用逗號隔開，結尾括號並換行

# 範例 3：連續印出兩組點對點路徑
print("A", "B", sep='->', end=' | ')
print("C", "D", sep='->')
````
