### 🚪 4-7-1 分行循序進水閘門——連續多次調用 `input()` 循序讀入多行整數

**生活比喻：巴拿馬運河的階梯進水船閘**  
想像一艘巨大的貨輪要通過運河上的三道階梯水閘。第一道閘門打開，第一道水流（第一行輸入文字）湧入並被抽取完畢；接著第一道閘門緊緊關閉，第二道水流（第二行輸入文字）才能進來；最後輪到第三道閘門。每道閘門只負責處理屬於自己那一層的水，絕不會同時混在一起！

在 Python 裡，`input()` 函數就像這道「進水閘門」。  
當電腦在程式碼中執行到第一次 `input()` 時，它會從鍵盤或輸入緩衝區（Input Buffer）中「吞掉」一整行字元，**直到看見換行符號（Enter 鍵產生的 `\n`）為止**。這個換行符號會被 `input()` 自動吞噬消化掉，游標則乾乾淨淨地停在「下一行的最前端」，靜候下一次 `input()` 的召喚！

**運作機制與底層原理：**
1. **逐行消化**：
   如果測資是連續三行的純整數：
   ```text
   15
   24
   9
   ```
   我們只需要依序呼叫三次轉型讀取：
   ```python
   a = int(input())  # 讀取第 1 行 "15"，吞掉換行，轉為整數 15
   b = int(input())  # 讀取第 2 行 "24"，吞掉換行，轉為整數 24
   c = int(input())  # 讀取第 3 行 "9"，吞掉換行，轉為整數 9
   ```
2. **記憶體獨立性**：
   三次呼叫各自分派給不同的獨立變數名牌（`a`、`b`、`c`），三者各自佔有獨立的記憶體置物櫃，互不干涉。
3. **無迴圈的多數值計算**：
   此時我們能自由運用第二章的四則運算（加總 `a + b + c`、平均 `(a + b + c) // 3`）以及第三章學過的極值函數（`max(a, b, c)`、`min(a, b, c)`）與全距計算（`max(a, b, c) - min(a, b, c)`）。

**常見錯誤陷阱（初學者常見盲點）：**
1. **在同一行寫了多個 input() 卻期望一次讀完全部**：
   如果寫成 `a, b = int(input()), int(input())`，這在語法上雖然合法，但在執行時它依然代表「要換兩行輸入」！若使用者把兩個數字打在同一行，第一個 `int(input())` 就會因為讀到含有空白的整行文字而拋出 `ValueError`。
2. **多打了一次 input() 導致程式卡死或報錯**：
   如果題目只給 3 行資料，但程式碼寫了 4 次 `input()`，在 Colab 或本機終端機中，程式會停住「發呆」等待第 4 行輸入；若在線上評判系統（OJ）中，則會直接引發 `EOFError: EOF when reading a line`（檔案結尾錯誤，無資料可讀）而直接判定判定失敗！

**APCS 考試實務建議：**
審題時務必看清楚「輸入格式說明」：究竟每個數字是「一行一個」還是「同行空白隔開」。分行輸入是最基礎也是最直觀的測資結構，請務必養成清晰宣告每一行獨立變數的好習慣。


````python
# [2] Code 範例區：展示連續多次調用 input() 循序讀入三行整數

# 模擬氣象站紀錄：依序在三行中讀入「清晨」、「中午」、「傍晚」的氣溫（皆為整數）
print("請在下方依序輸入三行整數氣溫（每輸完一個數字請按一次 Enter）：")

temp_morning = int(input())    # 讀入第 1 行：清晨氣溫
temp_noon = int(input())       # 讀入第 2 行：中午氣溫
temp_evening = int(input())    # 讀入第 3 行：傍晚氣溫

# 進行統計運算：總合、整數平均溫、全天最高溫、最低溫與溫差（全距）
total_temp = temp_morning + temp_noon + temp_evening
avg_temp = total_temp // 3
highest_temp = max(temp_morning, temp_noon, temp_evening)
lowest_temp = min(temp_morning, temp_noon, temp_evening)
temp_range = highest_temp - lowest_temp

# 輸出氣象分析報表
print("--- 今日氣溫統計分析 ---")
print("三時段氣溫紀錄：", temp_morning, temp_noon, temp_evening, sep=" ℃, ", end=" ℃\n")
print("全日總溫累積值：", total_temp)
print("全日整數平均溫：", avg_temp, "℃")
print("全日最高溫與最低溫：", highest_temp, "℃ /", lowest_temp, "℃")
print("今日溫差全距：", temp_range, "℃")
````

### 🪪 4-7-2 異質型態分行讀入——文字、整數與浮點數的多行混合資料卡

**生活比喻：圖書館讀者證與 RPG 角色建檔卡**  
當你走進圖書館申請借書證時，申請表的第一格要你填「姓名（文字串）」、第二格要你填「年齡（整數）」、第三格要你填「閱讀測驗能力點數（含小數點的浮點數）」。如果圖書館櫃檯人員把你的名字當成數字來算，電腦系統一定當場打結！

在實際的程式設計與測資處理中，連續多行資料**並不一定每一行都是同一種型態**！  
往往第一行是選手名稱（純文字 `str`），第二行是挑戰關卡（整數 `int`），第三行則是通關耗時（浮點數 `float`）。

**運作機制與型態精準對接：**
面對不同行具有不同資料型態的測資卡，關鍵在於**「每一行各自搭配正確的轉型外衣」**：
1. **純文字行（字串 `str`）**：
   直接使用 `name = input()` 即可！千萬不要自作聰明加上 `int()` 或 `float()`，否則一旦使用者輸入英文名字或中文字，電腦會立即噴出 `ValueError: invalid literal for int()` 慘痛報錯！
2. **計數或整數行（整數 `int`）**：
   使用 `level = int(input())`。
3. **實數或量測行（浮點數 `float`）**：
   使用 `time_sec = float(input())`。

**多行異質讀取的變數命名心法：**
初學同學常會把變數隨意取名為 `a`, `b`, `c`。當一行是文字、一行是整數時，若都叫 `a`, `b`，寫到後面很容易自己忘記哪一個能做加法、哪一個是字串，導致拿字串去跟整數相除而報錯（`TypeError`）！  
良性建議：為不同型態的資料給予具體語意的變數名稱，例如 `user_name`、`user_age`、`user_height`。

**常見錯誤陷阱（初學者必防地雷）：**
- **字串拼接 vs 數值相加混淆**：
  若忘記將數字行的 `input()` 轉型，例如 `price = input()`，則在後續計算總額時，`price * 2` 會變成字串重複兩次（如 `"50" * 2` 變成 `"5050"`），造成嚴重的邏輯錯誤（WA）！
- **浮點數直接用整數轉型**：
  若測資第三行是帶有小數的數值（如 `3.75`），使用 `int(input())` 會直接報錯崩潰！必須使用 `float(input())` 才能安全接收小數。

**APCS 考試實務建議：**
在題目說明中，仔細閱讀每一行測資的資料範圍與型態說明。只要看到小數點，無條件使用 `float()`；只要看到英數字串，一律保持原始 `input()` 即可。


````python
# [2] Code 範例區：展示文字 (str)、整數 (int) 與浮點數 (float) 的三行混合讀取

print("請依序在三行中輸入：冒險者暱稱（文字）、英雄等級（整數）、暴擊加乘倍率（小數）：")

# 第 1 行：文字字串，不需要任何轉型
player_name = input()

# 第 2 行：整數，使用 int() 轉型
player_level = int(input())

# 第 3 行：浮點數，使用 float() 轉型
crit_multiplier = float(input())

# 基礎戰鬥力計算公式：等級 * 100 * 暴擊倍率
base_power = player_level * 100
final_power = base_power * crit_multiplier

print("--- 冒險者公會角色檔案 ---")
print("冒險者代號：", player_name)
print("目前角色等級：Lv.", player_level)
print("暴擊加乘係數：", crit_multiplier)
print("基礎攻擊點數：", base_power)
print("最終極限戰力：", final_power)
````

### 🥪 4-7-3 多行混合單行多值——「單值行」與「多值行」交替讀取的複合架構

**生活比喻：雙層大麥克漢堡的美味層次**  
想像你在吃雙層大麥克漢堡，頂層只有一塊特製麵包（單一數值行），而中間夾層卻整齊擺放著兩片鮮嫩肉排（單行多數值行，用生菜空白隔開）！咬每一層的時候，你的舌尖都能分明感受到不同的厚度與結構，一口咬定，層次分明！

在 APCS 與各類競賽真題中，這是**出題率高達 80% 以上的經典輸入架構**！  
出題老師非常喜歡設計這樣的輸入格式：
- **第一行**：給定一個基準值（例如：基本低消 $K$、或打折門檻 $N$、或商品單價）。
- **第二行**：緊接著給定兩個或多個整數（例如：甲與乙各自點的數量 `q1 q2`，中間用空白隔開）。

**運作機制與底層原理：**
許多初學者在此處常有心理障礙：「*老師，我剛在第 4-6 節學了 `map(int, input().split())`，現在第一行只有一個數字，那我還能用 `split()` 嗎？如果第二行有多個數字，它會不會把第一行也一起抓進來？*」

**答案是絕對不會！請牢記兩大鐵則：**
1. **`input()` 的不可侵犯邊界**：
   每一次呼叫 `input()`，它的探索範圍**永遠嚴格限制在當前這一行**！只要碰到鍵盤的 Enter（換行字元 `\n`），它立刻剎車收工。下一行的文字對它而言是完全看不見的平行時空。
2. **各取所需的讀取策略**：
   - 面對「單值行」：直接使用最乾淨的 `base = int(input())`。
   - 面對「多值行」：無縫切換為第 4-6 節學過的映射切割神技 `x, y = map(int, input().split())`！

**組合架構示範：**
```python
# 第一行：只有一個數字
base_price = int(input())

# 第二行：同一行有兩個數字，中間以空白隔開
qty_a, qty_b = map(int, input().split())
```
這兩行指令一前一後，就像精密齒輪一樣完美咬合，絲絲入扣！

**常見錯誤陷阱（初學者必犯致命地雷）：**
- **死背一行流，在單值行也硬寫 `split()` 賦值多變數**：
  若在第一行寫 `a, b = map(int, input().split())`，但測資第一行實際上只有一個數字，直譯器會立刻悲鳴：`ValueError: not enough values to unpack (expected 2, got 1)`！
- **把兩行縮在同一行寫，導致邏輯錯亂**：
  請永遠一行程式碼對應一行輸入，保持程式碼清晰與可讀性。

**APCS 考試實務建議：**
拿到題目時，先在稿紙上將測資「畫線分行」，標記第 1 行有幾個數、第 2 行有幾個數。依照行數逐行編寫對應的讀取指令，即可保證百分之百精準收納測資！

### 📍 4-7-4 幾何多點與雙對數值分行讀取——平面二維座標與曼哈頓距離實戰

**生活比喻：都會區棋盤式街區的計程車行駛**  
在像紐約曼哈頓或高雄前鎮那樣整齊如棋盤的城市街道上，計程車司機無法像鳥兒一樣直接穿牆直線飛行，而必須沿著東西向大道開一段路，再轉向南北向街區開一段路。從第 1 個十字路口 $(x_1, y_1)$ 到第 2 個十字路口 $(x_2, y_2)$，所走過的實際街區距離就是著名的「曼哈頓距離（Manhattan Distance）」！

在幾何與網格地圖題型中，題目非常喜歡將多個座標點**分成多行給定**：
- **第 1 行**：起點座標 `x1 y1`（同一行兩個整數，空白隔開）
- **第 2 行**：終點座標 `x2 y2`（同一行兩個整數，空白隔開）

**運作機制與結合已學數學函數：**
1. **連續兩行使用 `map()`**：
   ```python
   x1, y1 = map(int, input().split())  # 讀取第 1 行的 (x1, y1)
   x2, y2 = map(int, input().split())  # 讀取第 2 行的 (x2, y2)
   ```
   兩行對稱呼叫，分別將四個整數座標精準置入四個獨立變數名牌中。
2. **曼哈頓距離公式（結合第三章 `abs()`）**：
   $$\text{Manhattan Distance} = |x_1 - x_2| + |y_1 - y_2|$$
   在 Python 中一行搞定：
   ```python
   dist_manhattan = abs(x1 - x2) + abs(y1 - y2)
   ```
3. **外接對齊矩形面積與周長**：
   以這兩個座標點為對角線頂點的外接水平矩形：
   - 寬度 $\text{width} = |x_1 - x_2|$
   - 高度 $\text{height} = |y_1 - y_2|$
   - 面積 $\text{area} = \text{width} \times \text{height}$
   - 周長 $\text{perimeter} = 2 \times (\text{width} + \text{height})$
4. **直線距離平方（歐幾里得距離平方）**：
   $$\text{dist}^2 = (x_1 - x_2)^2 + (y_1 - y_2)^2$$
   在 Python 中直接使用次方運算子 `**2`：
   ```python
   dist_sq = (x1 - x2) ** 2 + (y1 - y2) ** 2
   ```

**常見錯誤陷阱（初學者必防幾何地雷）：**
- **漏掉絕對值 `abs()`**：
  若直接寫 `(x1 - x2) + (y1 - y2)`，當 $x_1 < x_2$ 時會產生負數相抵消，求出來的根本不是距離！
- **座標順序指派混亂**：
  確保第一行的變數命名為 `x1, y1`，第二行命名為 `x2, y2`，切勿寫成 `x1, x2 = map(...)`，否則會把同一個點的 $x$ 與 $y$ 當成兩個不同點的 $x$ 座標！

**APCS 考試實務建議：**
二維座標讀取是 APCS 實作題（如歷年真題機器人路徑、基地台訊號覆蓋等）的基本功。掌握多行座標讀取與 `abs()` 組合技，就能在無須任何迴圈與條件判斷下，精確秒殺幾何幾何題！


````python
# [2] Code 範例區：展示兩行二維座標讀取與曼哈頓距離、中心點計算

# 模擬無人機巡邏航線：
# 測資第 1 行：巡邏起點 A 座標 x1 y1（兩整數，空白隔開）
# 測資第 2 行：巡邏終點 B 座標 x2 y2（兩整數，空白隔開）

print("請輸入第 1 行起點座標 (x1 y1，以空白隔開)：")
x1, y1 = map(int, input().split())

print("請輸入第 2 行終點座標 (x2 y2，以空白隔開)：")
x2, y2 = map(int, input().split())

# 計算曼哈頓街區距離 (水平距離絕對值 + 垂直距離絕對值)
manhattan_dist = abs(x1 - x2) + abs(y1 - y2)

# 計算直線距離平方 (畢氏定理 dx^2 + dy^2)
dx = abs(x1 - x2)
dy = abs(y1 - y2)
euclidean_dist_sq = dx**2 + dy**2

# 計算兩點正中心中點座標 (mid_x, mid_y)
mid_x = (x1 + x2) / 2
mid_y = (y1 + y2) / 2

print("--- 航線幾何分析 ---")
print("起點座標：(", x1, ", ", y1, ")", sep="")
print("終點座標：(", x2, ", ", y2, ")", sep="")
print("水平跨度 dx =", dx, "，垂直跨度 dy =", dy)
print("曼哈頓街區距離：", manhattan_dist)
print("直線距離之平方：", euclidean_dist_sq)
print("兩點之中點座標：(", mid_x, ", ", mid_y, ")", sep="")
````

### 🛡️ 4-7-5 APCS / OJ 固定行數題型實戰——純淨多行讀取與競賽常見致命陷阱防禦

**生活比喻：無人無聲的標準化考場驗收**  
參加國際體育競賽時，選手一走進檢錄區，只需把號碼布往掃描器上一靠，儀器就自動感應並記錄，過程沒有任何多餘的寒暄問候。如果選手對著機器大聲喊「報告！我是第一號選手！」，掃描器反而會把聲音雜訊判定為異常干擾而亮紅燈！

在 APCS 線上評判系統（Online Judge）中，評判核心是一台鐵面無私的機器直譯器：
- 題目預期幾行輸入，你就必須**精準讀取幾行**。
- 題目預期什麼輸出，你就**精準輸出什麼**。

**APCS / OJ 多行讀取三大致命地雷全面防禦：**

1. **地雷一：在 `input()` 括號內放入提示字串（死因：Wrong Answer / OLE）**：
   - 🔴 **致命寫法**：
     ```python
     a = int(input("請輸入第一行："))  # 絕對嚴禁！
     b = int(input("請輸入第二行："))  # 絕對嚴禁！
     ```
   - 💥 **後果**：`input("...")` 括號內的文字會直接被當成程式的「標準輸出（stdout）」，裁判系統會看到你的輸出多了 `"請輸入第一行："`，與官方標準答案完全不符，當場判決 `WA (Wrong Answer)` 或 `OLE (Output Limit Exceeded)`！
   - 🟢 **正確解法**：永遠保持括號完全空白：`a = int(input())`！

2. **地雷二：輸入行數與呼叫次數不一致（死因：EOFError 或 卡死）**：
   - 🔴 **致命寫法**：題目只提供 2 行資料，程式卻寫了 3 個 `input()`。
   - 💥 **後果**：當第 3 個 `input()` 伸手去抓輸入時，輸入串流已經空了，直接爆出 `EOFError: EOF when reading a line`，判定為 `RE (Runtime Error)`！
   - 🟢 **正確解法**：嚴格依據題目的「輸入說明」，說好幾行就精準寫幾次 `input()`，絕不多寫一行！

3. **地雷三：多行多值混淆（死因：ValueError）**：
   - 🔴 **致命寫法**：題目明明把 4 個數字拆成兩行（每行 2 個），初學者卻直接寫 `a, b, c, d = map(int, input().split())` 想一口氣讀完。
   - 💥 **後果**：第一行只有 2 個數字，卻要解包給 4 個變數，直接噴出 `ValueError: not enough values to unpack`！
   - 🟢 **正確解法**：分兩行讀取：
     ```python
     a, b = map(int, input().split())
     c, d = map(int, input().split())
     ```

每一行給定一回合雙方的得分，我們透過多次連續讀取，分別累加主隊與客隊的得分，並利用 `max()` 與 `abs()` 進行分差評估。即使尚未學習迴圈，這套固定行數讀取法依然能讓我們從容解出許多基礎競賽題！


````python
# [2] Code 範例區：展示 APCS 競技規範之純淨三行對戰得分讀取與統計

# 模擬羽球三局賽事統計（連續三行，每行兩個整數代表 A 選手與 B 選手該局得分）
# 完全遵守競技純淨規範，無任何 input 提示文字

set1_a, set1_b = map(int, input().split())  # 第 1 局得分
set2_a, set2_b = map(int, input().split())  # 第 2 局得分
set3_a, set3_b = map(int, input().split())  # 第 3 局得分

# 計算兩位選手三局總得分
total_a = set1_a + set2_a + set3_a
total_b = set1_b + set2_b + set3_b

# 計算全場總分差（絕對值）
score_diff = abs(total_a - total_b)

# 找出雙方在單局中的最高得分記錄
highest_single_set = max(set1_a, set1_b, set2_a, set2_b, set3_a, set3_b)

# 精準輸出比賽結果（依常見 OJ 格式要求）
print("--- 比賽數據結算 ---")
print("A 選手總得分：", total_a)
print("B 選手總得分：", total_b)
print("雙方總分差：", score_diff)
print("全場單局最高得分記錄：", highest_single_set)
````
