### 🔐 5-2-1 邏輯交集運算子：`and`（且）——兩全其美與嚴格門檻

**生活比喻：瑞士銀行的雙鑰匙金庫**  
想像一座守衛森嚴的瑞士銀行地下金庫，金庫大門上設有兩組鑰匙孔。必須由分行行長插上第一把金鑰匙（條件一成立：`True`），**同時**由資深保全主管插上第二把銀鑰匙（條件二成立：`True`），兩人同時轉動，沉重的合金大門才會轟然開啟！如果其中任何一人沒帶鑰匙（出現 `False`），大門就絕對打不開！

這正是 **`and`（且）** 的核心運作機制！在 Python 中，`and` 連接了左右兩個條件表示式。

**運作機制與真值表（Truth Table）：**
`and` 是最嚴格的把關者，**只有在左右兩邊的條件「同時為真（`True`）」時，整個表示式的結果才會是 `True`**。只要有任何一邊為假，結果立刻判定為 `False`！

| 左側條件 $A$ | 右側條件 $B$ | 複合運算式 $A 	ext{ and } B$ | 說明 |
| :---: | :---: | :---: | :--- |
| `True` | `True` | **`True`** | 兩全其美，完全達標 |
| `True` | `False` | **`False`** | 右側不合格，整體失敗 |
| `False` | `True` | **`False`** | 左側不合格，整體失敗 |
| `False` | `False` | **`False`** | 雙雙不合格，徹底失敗 |

**教材範例深入拆解（吳邦一老師講義第 17–18 頁）：**
```python
x = 2
y = 5
result = (x < 3) and (y > 0)
```
- 左側：`x < 3` ➔ `2 < 3` 成立（`True`）。
- 右側：`y > 0` ➔ `5 > 0` 成立（`True`）。
- 整體：`True and True` ➔ 最終結果為 **`True`**！

如果把 $x$ 改為 7：
- 左側：`7 < 3` 不成立（`False`）。
- 右側：`5 > 0` 成立（`True`）。
- 整體：`False and True` ➔ 最終結果為 **`False`**！

**常見錯誤陷阱：**
1. **口語直譯語法錯誤**：
   初學者常想表達「$x$ 等於 3 且等於 5」，容易口誤寫成 `x == 3 and 5`。在 Python 中，這樣寫並不會去比對兩次，因為右側只剩下單獨的數字 `5`。正確寫法必須完整表達兩側的比較命題：`x == 3 and x == 5`（雖然在此例中不可能同時成立）。
2. **與連續比較式的混淆**：
   上一節學過的 `1 < x < 9` 其實等價於 `(1 < x) and (x < 9)`。當比對同一個變數的數值區間時兩者皆可，但當要比對「不同變數」時（例如 $x$ 與 $y$），就必須依靠 `and` 運算子串接！

**APCS 考試實務建議：**
在題目要求「同時符合多項資格」、「在指定座標長寬矩形範圍內（`x >= 0 and x <= W and y >= 0 and y <= H`）」時，`and` 是最基本也是最高頻的邏輯工具。


````python
# [2] Code 範例區：展示 and 邏輯運算子的運作與真值表

# 1. 吳邦一老師講義第 17–18 頁範例：x < 3 and y > 0
demo_x1, demo_y1 = 2, 5
demo_result1 = (demo_x1 < 3) and (demo_y1 > 0)
print("--- 講義範例測試 1 ---")
print("demo_x1 = 2, demo_y1 = 5")
print("demo_x1 < 3 為：", demo_x1 < 3)
print("demo_y1 > 0 為：", demo_y1 > 0)
print("(demo_x1 < 3) and (demo_y1 > 0) 判定結果：", demo_result1)  # True and True -> True

demo_x2, demo_y2 = 7, 5
demo_result2 = (demo_x2 < 3) and (demo_y2 > 0)
print("--- 講義範例測試 2 ---")
print("demo_x2 = 7, demo_y2 = 5")
print("demo_x2 < 3 為：", demo_x2 < 3)
print("demo_y2 > 0 為：", demo_y2 > 0)
print("(demo_x2 < 3) and (demo_y2 > 0) 判定結果：", demo_result2)  # False and True -> False

# 2. 實務情境：期末獎學金資格審查（總分 >= 85 且 曠課時數 == 0）
demo_score = 88
demo_absent_hours = 0
demo_is_scholarship_eligible = (demo_score >= 85) and (demo_absent_hours == 0)
print("--- 獎學金雙重資格審查 ---")
print("成績是否達標且零曠課：", demo_is_scholarship_eligible)  # True
````

````python
# ==========================================
# [3] Code 填空題
# 任務說明：
# 實驗室安全門禁系統規定：
# 人員進入高等實驗室時，必須同時符合以下兩項安全檢驗：
# 1. 測得體溫 temperature 必須小於 37.5 度（未發燒）。
# 2. 安全防護裝備等級 badge_level 必須大於等於 2。
#
# 請在空格處填入正確的比較運算子與 and 關鍵字，
# 完成雙重安全檢驗判定！
# ==========================================

temperature = 36.6
badge_level = 3

# 第一空：體溫小於 37.5 (< 37.5)
# 第二空：填入邏輯運算子 and
# 第三空：裝備等級大於等於 2 (>= 2)

is_safe_to_enter = (temperature ___ 37.5) ___ (badge_level ___ 2)

print("量測體溫：", temperature, "度")
print("防護等級：", badge_level)
print("是否允許進入實驗室：", is_safe_to_enter)
````

### 🚪 5-2-2 邏輯聯集運算子：`or`（或）——暢通無阻與單一達標

**生活比喻：電影院的優待票通關閘門**  
在電影院的優待票驗票口，驗票員微笑著注視每一位觀眾。觀眾只要滿足以下兩者中的「任何一項」，就能順利購買優待票：
- 條件一：出示全職學生證（`True`）。
- 條件二：出示滿 65 歲之敬老證件（`True`）。

如果一位同學拿出了學生證，驗票員便不會再去盤查他的年齡是否滿 65 歲，直接放行（`True`）！只有當某人既不是學生、年齡也未滿 65 歲（兩者皆為 `False`）時，才會被禮貌拒絕（`False`）。

這就是 **`or`（或）** 的包容特質！

**運作機制與真值表（Truth Table）：**
`or` 是最友善的放行者，**只要左右兩邊的條件中「有任何一邊為真（`True`）」，整個運算式的結果就是 `True`**！唯有當兩邊同時為假時，結果才會是 `False`。

| 左側條件 $A$ | 右側條件 $B$ | 複合運算式 $A 	ext{ or } B$ | 說明 |
| :---: | :---: | :---: | :--- |
| `True` | `True` | **`True`** | 雙雙合格，順利過關 |
| `True` | `False` | **`True`** | 左側合格，順利過關 |
| `False` | `True` | **`True`** | 右側合格，順利過關 |
| `False` | `False` | **`False`** | 兩項皆未達標，判定失敗 |

**教材範例深入拆解（吳邦一老師講義第 17–18 頁）：**
```python
x = 7
y = 5
result = (x < 3) or (y > 0)
```
- 左側：`x < 3` ➔ `7 < 3` 不成立（`False`）。
- 右側：`y > 0` ➔ `5 > 0` 成立（`True`）。
- 整體：`False or True` ➔ 只要右邊成立，最終結果即為 **`True`**！

只有當 $x = 7$ 且 $y = -2$ 時：
- 左側：`7 < 3` ➔ `False`。
- 右側：`-2 > 0` ➔ `False`。
- 整體：`False or False` ➔ 結果才為 **`False`**！

**常見錯誤陷阱：**
1. **日常生活口語與數理邏輯的偏差**：
   在日常口語中，我們常說「你要喝茶還是咖啡？」，這通常暗示「二選一，不能同時喝」（這在邏輯上稱為互斥或 XOR，本單元 5-2-5 會詳細講述）。但在 Python 的標準 `or` 邏輯中，**兩者皆選（`True or True`）同樣是 `True`**！
2. **條件遺漏主詞**：
   例如想表達「分數等於 100 或者等於 0」，不可寫成 `score == 100 or 0`，必須寫成完整命題 `score == 100 or score == 0`。

**APCS 考試實務建議：**
`or` 常用於「邊界出界檢查（`x < 0 or x >= W or y < 0 or y >= H`）」、「多種特例放行判斷」以及「錯誤代碼或終止信號偵測」。只要命中任何一個警報點，就能立即取得 `True`。


````python
# [2] Code 範例區：展示 or 邏輯運算子的運作與真值表

# 1. 吳邦一老師講義第 17–18 頁範例：x < 3 or y > 0
demo_x1, demo_y1 = 7, 5
demo_result1 = (demo_x1 < 3) or (demo_y1 > 0)
print("--- 講義範例測試 1 ---")
print("demo_x1 = 7, demo_y1 = 5")
print("demo_x1 < 3 為：", demo_x1 < 3)  # False
print("demo_y1 > 0 為：", demo_y1 > 0)  # True
print("(demo_x1 < 3) or (demo_y1 > 0) 判定結果：", demo_result1)  # False or True -> True

demo_x2, demo_y2 = 7, -2
demo_result2 = (demo_x2 < 3) or (demo_y2 > 0)
print("--- 講義範例測試 2 ---")
print("demo_x2 = 7, demo_y2 = -2")
print("demo_x2 < 3 為：", demo_x2 < 3)  # False
print("demo_y2 > 0 為：", demo_y2 > 0)  # False
print("(demo_x2 < 3) or (demo_y2 > 0) 判定結果：", demo_result2)  # False or False -> False

# 2. 實務情境：高鐵乘車優待票資格檢驗（demo_is_student 或 demo_is_senior 任一符合即可）
demo_is_student = True
demo_is_senior = False
demo_can_get_discount = demo_is_student or demo_is_senior
print("--- 優待票資格檢驗 ---")
print("是否享有乘車優待票折扣：", demo_can_get_discount)  # True
````

### 🔄 5-2-3 邏輯否定運算子：`not`（非）——顛倒真假的黑白濾鏡

**生活比喻：底片反相器與「非請勿進」警示牌**  
想像攝影師走進暗房，將一張黑白底片放入反相沖洗藥水裡。原本白色明亮的部分全都變成了漆黑，而原本深黑的暗處則翻轉成了純白。  
在文字招牌上，「非請勿進」更是典型的邏輯反轉：
- 「持有請帖的人（`True`）」加上一個「非」，變成了「不能進入（`False`）」。
- 「沒有請帖的人（`False`）」加上一個「非」，代表條件滿足「被請帖規則擋下（`True`）」。

在 Python 裡，**`not`（非）** 扮演的就是這面神奇的「真偽反向濾鏡」！

**運作機制與特點：**
1. **單元運算子（Unary Operator）**：
   `and` 與 `or` 需要左邊與右邊兩個對象（二元運算子），而 `not` 只需要放在「單一運算式」的前面。
2. **極致單純的真值規則**：
   - 如果原運算式是 `True`，經過 `not` 之後立即變成 **`False`**。
   - 如果原運算式是 `False`，經過 `not` 之後立即變成 **`True`**。

| 原運算式 $A$ | 否定運算式 $	ext{not } A$ |
| :---: | :---: |
| `True` | **`False`** |
| `False` | **`True`** |

**教材範例深入拆解（吳邦一老師講義第 17–18 頁）：**
```python
x = 7
result = not (x < 3)
```
- 括號內：`x < 3` ➔ `7 < 3` 為 `False`。
- 經過 `not` 反轉：`not False` ➔ 得到最終結果 **`True`**！
這在語意上完全等價於「$x$ 沒有小於 3」（即 $x \ge 3$）。

**常見錯誤陷阱：**
1. **括號省略引發的認知混亂**：
   雖然寫成 `not x < 3` 在 Python 中完全合法，直譯器會先計算 `x < 3` 再做 `not`。但若遇到 `not a == b`，人類大腦很容易卡住：「*到底是 (not a) 再跟 b 比，還是先比完再 not？*」  
   **強烈建議**：只要使用 `not`，請一律為後方的條件加上小括號：`not (x < 3)`、`not (a == b)`，讓程式碼意圖如水晶般清澈透明！
2. **雙重否定恢復原狀**：
   `not not True` 就會變回 `True`。在演算法解題中應避免過度堆疊否定詞，以免降低程式碼可維護性。

**APCS 考試實務建議：**
在題目敘述「*不包含某些條件*」、「*排除特殊狀況*」或「*旗標未被啟動（`not is_found`）*」時，`not` 是最強大也最直覺的表達利器。


````python
# [2] Code 範例區：展示 not 邏輯否定運算子

# 1. 吳邦一老師講義第 17–18 頁範例：not demo_x < 3
demo_x = 7
print("demo_x =", demo_x)
print("demo_x < 3 原條件成立與否：", demo_x < 3)              # False
print("not (demo_x < 3) 經過否定反轉後：", not (demo_x < 3))   # True (等價於 demo_x >= 3)

# 2. 雙重否定與直接布林值反轉
demo_is_open = True
print("--- 布林旗標反轉 ---")
print("原狀態 demo_is_open：", demo_is_open)          # True
print("反轉狀態 not demo_is_open：", not demo_is_open) # False

# 3. 實務情境：非營業時間檢驗（目前星期非週日）
demo_weekday = 3  # 星期三
demo_is_sunday = (demo_weekday == 7)
demo_is_working_day = not demo_is_sunday
print("--- 營業時間檢驗 ---")
print("今天是否不是週日（是工作日）：", demo_is_working_day)  # True
````

### ⚖️ 5-2-4 複合邏輯與優先級鐵律：`not` > `and` > `or`

**生活比喻：數學四則運算規則的邏輯映照**  
在小學數學中，我們學過「先乘除，後加減，有括號先算括號」的黃金定律。如果沒有這套先後順序，算式 $3 + 5 \times 2$ 就會有人算出 16、有人算出 13，天下大亂！  
邏輯運算世界也有一套一模一樣的**神聖憲法**！當 `not`、`and`、`or` 齊聚一堂時，電腦該先聽誰的？

**邏輯運算子優先級排行榜（由高到低）：**
1. **第一優先（最高聖旨）：小括號 `(...)`**
2. **第二優先：邏輯否定 `not`**（類似負號，先把自己右邊的條件反轉）
3. **第三優先：邏輯交集 `and`**（類似乘法，把相鄰條件緊緊綁在一起）
4. **第四優先（最低階）：邏輯聯集 `or`**（類似加法，最後才做聯集）

**教材經典核心演繹（吳邦一老師講義第 18 頁）：**
讓我們來精準推導講義中的兩大核心運算式：
- **範例一：`x and 3 < 7 or y`**
  已知 `3 < 7` 必為 `True`。依據優先級，比較運算先算，接著 `and` 優先於 `or`，因此電腦等價於執行：
  ```python
  (x and (3 < 7)) or y
  ```
  如果 $x$ 為 `True`，左側 `x and True` 為 `True`，整個式子必定為 `True`！

- **範例二：`not x or 3 < 7 and y`**
  依據優先級：
  1. 先算 `not x`（`not` 優先級最高）
  2. 再算 `(3 < 7) and y`（`and` 優先於 `or`）
  3. 最後才把兩大塊用 `or` 串起來！
  整體等價於：
  ```python
  (not x) or ((3 < 7) and y)
  ```

**常見錯誤陷阱（括號位置造成的致命翻轉）：**
看這兩個看似相似的算式：
- 算式 A：`A or B and C` ➔ 等價於 `A or (B and C)`。
- 算式 B：`(A or B) and C` ➔ 括號打破了預設順序，強迫先做 `A or B`！
如果此時 $C$ 是 `False`，算式 B 必然是 `False`，但算式 A 若 $A$ 為 `True` 卻會得到 `True`！兩者結果大相逕庭！

**APCS 考試實務建議：**
APCS 觀念題「極度熱愛」考察邏輯優先級陷阱！而在實作題撰寫時，**請永遠不要吝嗇使用小括號**！將你想先判斷的群組用 `(...)` 框起來，不僅可以 100% 杜絕優先級盲點，還能讓閱讀你程式碼的人對你的邏輯結構一目了然！


````python
# [2] Code 範例區：展示邏輯運算子優先級（not > and > or）

# 1. 吳邦一老師講義第 18 頁範例驗證
demo_x = True
demo_y = False

# 測試式 1：demo_x and 3 < 7 or demo_y
# 步驟：先算 (3 < 7) -> True，再算 (demo_x and True) -> True，最後 (True or demo_y) -> True
demo_eval1_auto = demo_x and 3 < 7 or demo_y
demo_eval1_explicit = (demo_x and (3 < 7)) or demo_y
print("--- 講義範例 1 驗證 ---")
print("demo_x and 3 < 7 or demo_y 自動求值：", demo_eval1_auto)          # True
print("(demo_x and (3 < 7)) or demo_y 顯式括號：", demo_eval1_explicit)  # True

# 測試式 2：not demo_x or 3 < 7 and demo_y
# 步驟：先算 not demo_x -> False，再算 3 < 7 and demo_y -> False，最後 False or False -> False
demo_eval2_auto = not demo_x or 3 < 7 and demo_y
demo_eval2_explicit = (not demo_x) or ((3 < 7) and demo_y)
print("--- 講義範例 2 驗證 ---")
print("not demo_x or 3 < 7 and demo_y 自動求值：", demo_eval2_auto)         # False
print("(not demo_x) or ((3 < 7) and demo_y) 顯式括號：", demo_eval2_explicit) # False

# 2. 括號改變優先級的重大差異展示
demo_A, demo_B, demo_C = True, False, False
print("--- 括號顛覆優先級展示 ---")
print("demo_A or demo_B and demo_C （等價於 demo_A or (demo_B and demo_C)）：", demo_A or demo_B and demo_C)     # True
print("(demo_A or demo_B) and demo_C （強迫先算 demo_A or demo_B）：", (demo_A or demo_B) and demo_C)     # False
````

### ⚡ 5-2-5 邏輯代數實戰：互斥或（XOR，Exclusive OR）的純邏輯構建

**生活比喻：樓梯間的雙切開關與單選贈品**  
在透天厝的樓梯間，一樓牆壁有一個開關，二樓牆壁也有一個開關。如果兩邊開關的方向相同（例如都往上，或都往下），電燈是熄滅的；只有當兩個開關「一上一架、方向相反」時，電燈才會亮起！  
在百貨公司週年慶兌換贈品時，服務台常規定：「*滿額禮 A 與滿額禮 B 只能二選一！不可兩樣都拿，也不可都不拿！*」

這種「**兩者恰好只有一個成立（不能兩個都真，也不能兩個都假）**」的嚴謹邏輯，在電腦科學與離散數學中被稱為 **互斥或（Exclusive OR，簡稱 XOR）**！

**教材經典公式解析（吳邦一老師講義第 18 頁）：**
在第二章 2-7 節中，我們學過位元運算子 `^`；但在處理純布林邏輯條件時，講義第 18 頁特別指出了利用基礎邏輯運算子手刻構建 XOR 的經典公式：
```python
# exclusive or
(x and not y) or (not x and y)
```
讓我們來仔細拆解這段優雅的邏輯結構：
1. 前半段 `(x and not y)`：代表「$x$ 為真，**但是** $y$ 為假」（只有 $x$ 獨贏）。
2. 後半段 `(not x and y)`：代表「$x$ 為假，**但是** $y$ 為真」（只有 $y$ 獨贏）。
3. 中間用 `or` 串接：只要上述兩種「獨贏」的情況有一種發生，整體就是 `True`！

**XOR 互斥或真值表：**
| 條件 $x$ | 條件 $y$ | 傳統 $x 	ext{ or } y$ | 互斥或 $	ext{XOR}$ | 狀態解析 |
| :---: | :---: | :---: | :---: | :--- |
| `True` | `True` | `True` | **`False`** | 貪心全拿，違反互斥原則 |
| `True` | `False` | `True` | **`True`** | 恰好 $x$ 成立，合格！ |
| `False` | `True` | `True` | **`True`** | 恰好 $y$ 成立，合格！ |
| `False` | `False` | `False` | **`False`** | 兩頭落空，未達門檻 |

**另一種極致簡潔的等價寫法：**
在 Python 中，如果 $x$ 與 $y$ 已經是布林值，其實 XOR 還有一個神級簡約寫法：
```python
x != y
```
因為「兩者一真一假」的本質，就是「兩者的布林真值不相等」！這在 APCS 實作題中是極高明的簡化技巧。

**APCS 考試實務建議：**
APCS 實作題（如 APCS 2017-10 舊版實作第 1 題 c461. 邏輯運算子）會直接要求模擬 XOR 運算！掌握 XOR 的邏輯分解與真值特徵，是通往初級與中級實作題的必經階梯。


````python
# [2] Code 範例區：展示互斥或（XOR）的純邏輯運算與實作

# 1. 吳邦一老師講義第 18 頁手刻 XOR 公式：(demo_x and not demo_y) or (not demo_x and demo_y)
demo_x = True
demo_y = False

demo_xor_formula = (demo_x and not demo_y) or (not demo_x and demo_y)
demo_xor_shortcut = (demo_x != demo_y)

print("--- XOR 互斥或計算（demo_x=True, demo_y=False）---")
print("講義公式 (demo_x and not demo_y) or (not demo_x and demo_y)：", demo_xor_formula)   # True
print("精簡比對 (demo_x != demo_y)：", demo_xor_shortcut)                        # True

# 測試兩者皆為 True 的情況（違反互斥）
demo_x2, demo_y2 = True, True
demo_xor_both_true = (demo_x2 and not demo_y2) or (not demo_x2 and demo_y2)
print("--- XOR 互斥或計算（demo_x=True, demo_y=True）---")
print("兩者皆為 True 時的互斥檢驗：", demo_xor_both_true)              # False

# 2. 實務情境：政府補助方案二選一資格審核
# 申請人不可同時請領「租屋補助」與「購屋補貼」，只能擇一
demo_has_rent_subsidy = True
demo_has_buy_subsidy = False

# 使用講義 XOR 公式檢驗是否合乎「擇一請領」規範
demo_is_subsidy_valid = (demo_has_rent_subsidy and not demo_has_buy_subsidy) or (not demo_has_rent_subsidy and demo_has_buy_subsidy)
print("--- 補助申請合規檢驗 ---")
print("補助申請是否合乎二選一規範：", demo_is_subsidy_valid)  # True
````

````python
#@title 🧪 點擊此處取得單元 5-2 學習成效自動評分報告 { display-mode: "form" }
# ------------------------------------------------------------------------------
# 官方智慧全環境評分載入器（支援：本地開發庫優先 ➔ 公開正式庫雙重備援）
# ------------------------------------------------------------------------------
import os
import sys
import json
import logging
import urllib.request

# 靜音內部 httplib2 警告
logging.getLogger("google_auth_httplib2").setLevel(logging.ERROR)

# 1. 互動式輸入學員姓名與學號（分開獨立欄位，按 Enter 採用預設值）
print("👤【學員身分登記】（自學測試可直接按 Enter 鍵使用預設值）：")
student_name = input("請輸入您的姓名 [預設: 自學冒險者]：").strip() or "自學冒險者"
student_id = input("請輸入學號或座號 [預設: 無座號]：").strip() or "無座號"

# 2. 自動嘗試獲取 Google 帳號官方身分認證（真實 Email 與 Google 暱稱）
google_email = None
google_name = None
try:
    from google.colab import auth
    print("🔐 正在連結 Google 帳號進行身分認證（若跳出授權視窗，請點選允許）...")
    auth.authenticate_user()

    # 方法 1: 使用 Google Drive API v3 (about:get) 抓取真實 displayName 與 emailAddress
    # Colab ADC 預設包含 drive 權限，能 100% 精準獲取 Google 帳號中文顯示名稱（如：葉明昌）
    try:
        import google.auth
        from googleapiclient.discovery import build
        creds, _ = google.auth.default()
        drive_service = build('drive', 'v3', credentials=creds, cache_discovery=False)
        about = drive_service.about().get(fields="user(displayName,emailAddress)").execute()
        u_obj = about.get("user", {})
        google_name = u_obj.get("displayName")
        google_email = u_obj.get("emailAddress")
    except Exception:
        pass

    # 方法 2: 若 Drive API 失敗，嘗試透過 OAuth2 userinfo / tokeninfo 備援
    if not google_email or not google_name:
        try:
            import google.auth
            from google.auth.transport.requests import Request
            credentials, _ = google.auth.default()
            credentials.refresh(Request())
            token = credentials.token

            for ep in [
                f"https://www.googleapis.com/oauth2/v3/tokeninfo?access_token={token}",
                "https://www.googleapis.com/oauth2/v3/userinfo",
                "https://www.googleapis.com/oauth2/v2/userinfo"
            ]:
                try:
                    req = urllib.request.Request(
                        ep,
                        headers={"Authorization": f"Bearer {token}"}
                    )
                    with urllib.request.urlopen(req, timeout=5) as resp:
                        u_data = json.loads(resp.read().decode('utf-8'))
                        if not google_email:
                            google_email = u_data.get("email") or u_data.get("emailAddress")
                        if not google_name:
                            google_name = (
                                u_data.get("name")
                                or u_data.get("given_name")
                                or u_data.get("displayName")
                            )
                        if google_email and google_name:
                            break
                except Exception:
                    continue
        except Exception:
            pass

    # 方法 3: 若只有 email 但 Google API 沒給 name，以 email 前綴作為暱稱回退
    if google_email and not google_name:
        google_name = google_email.split("@")[0]

    if google_email:
        print(f"✅ 身分認證成功！已綁定 Google 帳號：{google_name} ({google_email})")
except Exception:
    # 非 Colab 環境、本機離線測試或略過授權
    pass

# 3. 身分資料獨立注入環境變數（學號與姓名完全分開）
os.environ["STUDENT_NAME"] = student_name
os.environ["STUDENT_ID"] = student_id
if google_email:
    os.environ["GOOGLE_EMAIL"] = google_email
if google_name:
    os.environ["GOOGLE_NAME"] = google_name

# 4. 評分核心腳本載入（本機優先 ➔ 遠端記憶體載入，絕不產生實體暫存檔）
# 找不到來源就換下一條。內容已拿到但執行失敗時，停止改試其他來源。
# 評分腳本放在獨立命名空間執行，成功後也不把名稱寫回筆記本。答案讀執行前的學生存放格快照。
_loaded = False
_load_error = None

# 4.1 優先嘗試讀取開發庫/本機檔案（本機離線開發測試零延遲）
_local_paths = [
    "web/graders/grade_5_2.py",
    "../graders/grade_5_2.py",
    "../../graders/grade_5_2.py",
    "../web/graders/grade_5_2.py",
    "../../web/graders/grade_5_2.py"
]
for _lp in _local_paths:
    if not os.path.exists(_lp):
        continue
    try:
        with open(_lp, "r", encoding="utf-8") as _f:
            _code = _f.read()
        _grader_ns = {"__builtins__": __builtins__, "__name__": "apcs_grader", "__STUDENT_ENV__": dict(globals())}
        try:
            exec(_code, _grader_ns)
        finally:
            del _grader_ns
        os.environ["GRADER_SOURCE"] = _lp
        _loaded = True
        break
    except Exception as _exc:
        _load_error = _exc
        break

# 4.2 若在雲端環境（Colab），自動透過公開正式庫雙重備援以純記憶體載入
if not _loaded and _load_error is None:
    _grader_urls = [
        "https://raw.githubusercontent.com/johnnyy-lab/APCS1to3/main/graders/grade_5_2.py",
        "https://johnnyy-lab.github.io/APCS1to3/graders/grade_5_2.py",
        "https://raw.githubusercontent.com/johnnyy-lab/APCS1to3-dev/main/web/graders/grade_5_2.py"
    ]
    for _url in _grader_urls:
        try:
            _req = urllib.request.Request(_url, headers={"User-Agent": "Mozilla/5.0"})
            with urllib.request.urlopen(_req, timeout=10) as _resp:
                _code = _resp.read().decode("utf-8")
        except Exception:
            continue
        try:
            _grader_ns = {"__builtins__": __builtins__, "__name__": "apcs_grader", "__STUDENT_ENV__": dict(globals())}
            try:
                exec(_code, _grader_ns)
            finally:
                del _grader_ns
            os.environ["GRADER_SOURCE"] = _url
            _loaded = True
            break
        except Exception as _exc:
            _load_error = _exc
            break

if not _loaded:
    if _load_error is not None:
        print("❌ 評分程式載入失敗，請稍後重新執行。若仍失敗，請把這個畫面交給老師。")
        if os.environ.get("GRADER_DEBUG", "").strip().lower() in ("1", "true", "yes", "on"):
            _reason = str(_load_error).strip().splitlines()
            _line = _reason[0][:180] if _reason else ""
            if _line:
                print("（開發診斷：" + type(_load_error).__name__ + "：" + _line + "）")
            else:
                print("（開發診斷：" + type(_load_error).__name__ + "）")
    else:
        print("❌ 評分系統連線異常，請確認網路連線狀態（或稍後重新執行）")
````
