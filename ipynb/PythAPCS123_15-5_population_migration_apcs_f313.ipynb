### 📖 題目敘述（Problem Description）
在一個 $R \times C$ 的網格地圖中，分布著若干座城市，每個格子代表一個區域。  
每個格子內填有一個整數：
* 若為**非負整數**（$\ge 0$），代表該區域是一座**城市**，數字代表該城市的**現有人口數**。
* 若為 **-1**，代表該區域為**無法通行的非城市區域（如高山、河流或未開發區）**，既不能遷出人口，也不能遷入人口。

每天，人口會發生一次大規模遷移。遷移規則如下：
1. **遷移比例係數**：每個城市每天會將其人口數除以 $k$（採整數除法，無條件捨去，即 $\lfloor P / k \rfloor$）作為向單一方向遷移的人口數。
2. **四方向同步擴散**：城市會向其**上下左右四個相鄰且為合法城市**（即不為 -1 且在邊界內）的區域各遷移 $\lfloor P / k \rfloor$ 人。
3. **同步結算原則**：**當天所有城市的人口遷移是同一瞬間同步發生的**！所有遷移量的計算都是依據「當天清晨起始時」的人口數計算，不能在運算途中相互干擾。
4. **天數推進**：重複此遷移過程共 $m$ 天。

請撰寫程式模擬 $m$ 天之後的結果，並輸出所有城市中**人數最少**與**人數最多**的城市人數。

---

### 📥 輸入格式（Input Format）

* $R$：網格列數（Row）
* $C$：網格行數（Column）
* $k$：遷移除數係數
* $m$：經過的遷移天數

接下來有 $R$ 行，每行包含 $C$ 個以空白隔開的整數，代表初始地圖狀態。

---

### 📤 輸出格式（Output Format）

* 第一行輸出一個整數，代表 $m$ 天後所有合法城市中**最少的人口數**。
* 第二行輸出一個整數，代表 $m$ 天後所有合法城市中**最多的人口數**。

---

### 📊 測資規模與約束條件（Constraints）
* $1 \le R, C \le 50$
* $1 \le k \le 100$
* $1 \le m \le 100$
* 初始城市人口數介於 $0$ 到 $10^5$ 之間。
* 保證網格中至少存在一座合法城市。


#### 範例推演說明（$k=4, m=1$ 天）：
* **初始狀態**：
  * `(0, 0)`: 10 人。合法鄰居有 `(0, 1)` 與 `(1, 0)` 共 2 個。向各鄰居遷移 $10 // 4 = 2$ 人。總遷出 $2 \times 2 = 4$ 人。
  * `(0, 1)`: 2 人。單方向遷移量為 $2 // 4 = 0$ 人。不遷出。
  * `(0, 2)`: -1（障礙）。不參與遷移。
  * `(1, 0)`: 5 人。合法鄰居僅有 `(0, 0)` 共 1 個（因為 `(1, 1)` 為 -1）。向 `(0, 0)` 遷移 $5 // 4 = 1$ 人。總遷出 1 人。
  * `(1, 1)`: -1（障礙）。不參與遷移。
  * `(1, 2)`: 2 人。合法鄰居無（上為 -1，左為 -1）。遷移量 $2 // 4 = 0$ 人。
* **1 天後結算**：
  * `(0, 0)`: $10 - 4 + 1$ (來自 `(1, 0)`) $= 7$
  * `(0, 1)`: $2 - 0 + 2$ (來自 `(0, 0)`) $= 4$
  * `(1, 0)`: $5 - 1 + 2$ (來自 `(0, 0)`) $= 6$
  * `(1, 2)`: $2 - 0 + 0 = 2$
* **最終有效城市人口**：`[7, 4, 6, 2]`。
* 最少人數為 `2`，最少人數為 `7`。依序分行輸出！

### 15.5.1 題意解析與人口擴散物理模型：四方向同時流動與障礙地形（-1）

#### 💡 核心心智模型：
本題是經典的**二維網格擴散物理模擬題**。在動態系統中，我們要釐清三大要素：
1. **節點屬性**：網格中的格子只有兩種——「合法城市（$\ge 0$）」與「不可穿越障礙（$-1$）」。
2. **傳播媒介**：人口只能向**上下左右四個相鄰方向**流動，斜對角無法通行。
3. **邊界與障礙屏障**：
   * 地圖的外邊界不能跨越。
   * 若相鄰格子為 `-1`，視同牆壁，無法作為流出目標。

```text
      [-1 邊界]
          ▲
[-1 障礙] ◀ (r, c)城市 ▶ [合法鄰居 接收 P//k]
          ▼
     [合法鄰居 接收 P//k]
```


````python
# =====================================================================
# [2] Code 範例：地圖載入與城市/障礙統計分析
# =====================================================================

R, C, k, m = 2, 3, 4, 1
grid = [
    [10, 2, -1],
    [5, -1, 2]
]

city_count = 0
barrier_count = 0
initial_total_pop = 0

for r in range(R):
    for c in range(C):
        if grid[r][c] != -1:
            city_count += 1
            initial_total_pop += grid[r][c]
        else:
            barrier_count += 1

print(f"地圖尺寸: {R}x{C}, 遷移係數 k={k}, 模擬天數 m={m}")
print(f"合法城市數: {city_count}, 障礙數: {barrier_count}")
print(f"初始總人口: {initial_total_pop}")
````

````python
# =====================================================================
# [3] Code 填空題：整數除法遷移量計算
# 任務：請將下方的 ___ 替換為計算單方向遷移人數的整數除法運算子 //
# =====================================================================

pop = 10
k_val = 4

# 每方向遷移量：無條件捨去整數除法
move_per_direction = pop ___ k_val
print("單一方向遷移人數：", move_per_direction)  # 應為 2
````

````python
# =====================================================================
# [4] Code 練習題：過濾並判定相鄰格子是否可通行
# =====================================================================

def is_valid_neighbor(grid, r, c, R, C):
    # 需同時滿足：1. 在地圖範圍內 2. 不為 -1
    return (0 <= r < R and 0 <= c < C and grid[r][c] != -1)

sample_g = [[10, -1], [5, 2]]
assert is_valid_neighbor(sample_g, 0, 0, 2, 2) == True
assert is_valid_neighbor(sample_g, 0, 1, 2, 2) == False  # 障礙
assert is_valid_neighbor(sample_g, -1, 0, 2, 2) == False # 越界
print("✅ 有效鄰居判定函數檢驗全部通過！")
````

````python
# =====================================================================
# [5] Code 挑戰題：總人口守恆定律驗證模型
# 在封閉的地圖中，只要無人口遷出地圖外或憑空消失，每天總人口必須恆定
# =====================================================================

def get_total_population(grid):
    return sum(cell for row in grid for cell in row if cell != -1)

print("當前總人口數：", get_total_population(grid))
````

### 15.5.2 考場致命天坑重現：原地修改造成的「資料污染」慘劇與概念破除

#### 💣 考場最高頻致命失分錯誤：原地修改（In-place Mutation）
許多初學者在寫模擬題時，常常順著迴圈走訪格子，**邊算邊直接修改 `grid` 的值**：
```python
# ❌ 錯誤示範：原地修改會導致嚴重的數據污染！
for r in range(R):
    for c in range(C):
        grid[r][c] -= move * valid_neighbors
        grid[nr][nc] += move  # 慘劇：提早把後面的格子改大了！
```

#### 💥 為什麼會引發連鎖崩潰？
當迴圈處理完 `(0, 0)` 時，鄰居 `(0, 1)` 的人口從 2 被加成了 4。  
等迴圈下一個換到 `(0, 1)` 時，它竟然用被灌水後的 4 去除以 $k$（$4 // 4 = 1$ 人），造成不該遷移的人口也移出了！  
這就是典型的**時序因果倒置（Time Traveling Hazard）**！


````python
# =====================================================================
# [2] Code 範例：原地修改 vs 同步結算的悲慘對比實驗
# =====================================================================

# 假設有 3 個並排城市：[10, 2, 0]，k=4
# 正確邏輯：
# 10 移出 10//4 = 2 給中間。中間原為 2，2//4 = 0 不移出。最後應為 [8, 4, 0]

arr_wrong = [10, 2, 0]
# 原地錯誤寫法：
move_0 = arr_wrong[0] // 4 # 2
arr_wrong[0] -= move_0
arr_wrong[1] += move_0     # 中間變成 4 了！

move_1 = arr_wrong[1] // 4 # 竟然變成 4 // 4 = 1 了！(錯誤!)
arr_wrong[1] -= move_1
arr_wrong[2] += move_1

print("❌ 原地修改錯誤結果：", arr_wrong) # [8, 3, 1] 嚴重偏離！
print("✅ 理論正確結算結果： [8, 4, 0]")
````

````python
# =====================================================================
````
````python
# [4] Code 練習題：同步快照模式計算單列擴散
# =====================================================================

def sync_step_1d(arr, k):
    n = len(arr)
    delta = [0] * n
    for i in range(n):
        move = arr[i] // k
        # 往右傳播
        if i + 1 < n:
            delta[i] -= move
            delta[i + 1] += move
    return [arr[i] + delta[i] for i in range(n)]

res = sync_step_1d([10, 2, 0], 4)
assert res == [8, 4, 0]
print("✅ 1D 同步擴散測試通過：", res)
````

### 15.5.3 雙矩陣狀態解法核心：Delta 增減量矩陣設計與記憶體獨立初始化

#### 💡 演算法解題思維大革命：Delta 增量矩陣法（Delta Accumulator）
為了解決「資料污染」問題，最直觀優雅且記憶體安全的架構就是 **Delta 矩陣法**：
1. **建立全 0 變動矩陣**：`delta = [[0] * C for _ in range(R)]`。
2. **只紀錄變動量（$\Delta$）**：
   * 當城市 $(r, c)$ 遷移 $x$ 個人給鄰居 $(nr, nc)$：
     * 自身扣除：`delta[r][c] -= x`
     * 鄰居增加：`delta[nr][nc] += x`
3. **一天結束時原子結算（Atomic Commit）**：
   * 所有格子掃描完畢後，執行 `grid[r][c] += delta[r][c]`。

#### 🌟 Delta 矩陣法的巨大優勢：
* **物理意義清晰**：人口的流出與流入直接對應加減運算。
* **障礙保護天然免疫**：障礙格（-1）永遠不會有流入與流出，`delta` 值恆為 0，絕不會被誤改！


````python
# =====================================================================
# [2] Code 範例：Delta 變動矩陣的宣告與運算展示
# =====================================================================

R, C = 2, 3
delta = [[0] * C for _ in range(R)]

# 模擬 (0, 0) 遷出 2 人給 (0, 1)
delta[0][0] -= 2
delta[0][1] += 2

# 模擬 (0, 0) 遷出 2 人給 (1, 0)
delta[0][0] -= 2
delta[1][0] += 2

print("=== 當天 Delta 變動量矩陣 ===")
for row in delta:
    print(row)
````

````python
# =====================================================================
````
````python
# [5] Code 挑戰題：零拷貝原地重置 Delta 矩陣效能實驗
# =====================================================================

# 每天重新生成新陣列 vs 重置已有陣列
def reset_delta(d, r, c):
    for i in range(r):
        for j in range(c):
            d[i][j] = 0

reset_delta(d, 2, 2)
assert all(d[i][j] == 0 for i in range(2) for j in range(2))
print("Delta 矩陣原地極速重置完畢！")
````

### 15.5.4 單一城鎮擴散邏輯：整數除法 P // k、有效鄰居計數與流出/流入結算

#### 🔍 單一格子 $(r, c)$ 的完整處理決策鏈：
1. **資格檢驗**：
   * 若 $grid[r][c] == -1$：它是障礙，直接 `continue` 跳過。
   * 若 $grid[r][c] < k$：整數除法 $grid[r][c] // k == 0$，無人可遷出，亦可跳過。
2. **向四方向探尋合法鄰居**：
   * 使用方向向量：`dr = [-1, 1, 0, 0]`, `dc = [0, 0, -1, 1]`。
   * 計算鄰居座標：`nr = r + dr[i]`, `nc = c + dc[i]`。
   * 合法條件：`0 <= nr < R` 且 `0 <= nc < C` 且 `grid[nr][nc] != -1`。
3. **雙向記帳**：
   * 對每個合法鄰居：`delta[r][c] -= move`，`delta[nr][nc] += move`。


````python
# =====================================================================
# [2] Code 範例：單一城鎮四方向擴散記帳實作
# =====================================================================

def spread_from_cell(r, c, grid, delta, R, C, k):
    pop = grid[r][c]
    if pop <= 0:  # -1 障礙或 0 人口均不遷移
        return
    
    move = pop // k
    if move == 0:
        return
        
    for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
        nr = r + dr
        nc = c + dc
        # 邊界與非障礙檢驗
        if 0 <= nr < R and 0 <= nc < C and grid[nr][nc] != -1:
            delta[r][c] -= move
            delta[nr][nc] += move

print("單一城鎮擴散函數封裝完成！")
````

````python
# =====================================================================
# [3] Code 填空題：四方向向量元組遍歷
# 任務：請將下方的 ___ 替換為 (0, 1) 代表向右探尋
# =====================================================================

directions = [(-1, 0), (1, 0), (0, -1), ___]
print("方向向量總數量：", len(directions))  # 應為 4
````

````python
# =====================================================================
# [4] Code 練習題：測試孤立城市（四面受困）的遷出行為
# =====================================================================

isolated_grid = [
    [-1, -1, -1],
    [-1, 100, -1],
    [-1, -1, -1]
]
d_iso = [[0] * 3 for _ in range(3)]
spread_from_cell(1, 1, isolated_grid, d_iso, 3, 3, 4)

# 由於周圍全是 -1 障礙，人口無法外移，delta 應全為 0
assert d_iso[1][1] == 0
print("✅ 孤立無援城市測試通過：人口完全鎖定無流失！")
````

### 15.5.5 哨兵外圍加框法（Padding）：以 -1 環繞邊界徹底免除邊界判斷

#### 🛡️ 競賽高手密技：哨兵外圍加框法（Sentinel Padding）
在普通寫法中，我們每次檢查鄰居都要寫：
`0 <= nr < R and 0 <= nc < C and grid[nr][nc] != -1`
這不僅程式碼冗長，而且在高頻迴圈中帶來額外分支判斷負擔。

**巧奪天工的構思**：
本題已經定義 `-1` 為「不可進出之障礙」！  
如果我們直接在整個 $R \times C$ 地圖的最外層，**包上一圈值為 -1 的護城河**，生成 $(R+2) \times (C+2)$ 的加框矩陣：
```text
[-1, -1, -1, -1, -1]
[-1, 10,  2, -1, -1]
[-1,  5, -1,  2, -1]
[-1, -1, -1, -1, -1]
```
此時，原本在邊緣的城市，往地圖外看時**只會看到 -1**！  
因此，**所有的邊界判斷 `0 <= nr < R` 都可以徹底拿掉**，只要檢查 `grid[nr][nc] != -1`！這就是競賽高階效能的秘密武器！


````python
# =====================================================================
# [2] Code 範例：哨兵加框矩陣的建構
# =====================================================================

def pad_grid_with_sentinel(grid, R, C):
    # 頂部與底部為寬度 C+2 的全 -1 列
    top_bottom = [-1] * (C + 2)
    padded = [top_bottom]
    for row in grid:
        padded.append([-1] + row + [-1])
    padded.append(top_bottom[:])
    return padded

raw_g = [[10, 2, -1], [5, -1, 2]]
padded_g = pad_grid_with_sentinel(raw_g, 2, 3)
print("加框後的矩陣尺寸：", len(padded_g), "x", len(padded_g[0])) # 4x5
for r in padded_g:
    print(r)
````

````python
# =====================================================================
````
````python
# [4] Code 練習題：哨兵加框版極速無邊界擴散
# =====================================================================

def spread_padded(padded_g, R, C, k):
    delta = [[0] * (C + 2) for _ in range(R + 2)]
    # 遍歷原始核心區域 1 到 R, 1 到 C
    for r in range(1, R + 1):
        for c in range(1, C + 1):
            pop = padded_g[r][c]
            if pop <= 0: continue
            move = pop // k
            if move == 0: continue
            
            for nr, nc in ((r-1, c), (r+1, c), (r, c-1), (r, c+1)):
                # 完全不需要判斷 nr, nc 是否越界！直接判斷是否不為 -1！
                if padded_g[nr][nc] != -1:
                    delta[r][c] -= move
                    delta[nr][nc] += move
                    
    for r in range(1, R + 1):
        for c in range(1, C + 1):
            padded_g[r][c] += delta[r][c]

spread_padded(padded_g, 2, 3, 4)
res_core = [padded_g[r][1:4] for r in range(1, 3)]
assert res_core == [[7, 4, -1], [6, -1, 2]]
print("✅ 哨兵無邊界檢查擴散 100% 正確！結果：", res_core)
````

### 15.5.6 多回合狀態機動態推進：m 天迭代迴圈與 Delta 增減量原子結算

#### 🔄 多天模擬狀態機推進管線：
有了單天的 Delta 遷移核心後，處理 $m$ 天的遷移就變得無比清晰：
```text
第 1 天開始 ──▶ 生成清空 Delta ──▶ 統計所有流出與流入 ──▶ 原子結算累加回地圖
第 2 天開始 ──▶ 生成清空 Delta ──▶ 統計所有流出與流入 ──▶ 原子結算累加回地圖
   ...
第 m 天結束 ──▶ 輸出最終城市極值
```

只要天數迴圈在最外層，每一天結束時地圖狀態都更新為最新人口，隔天清晨自動繼承前一天的結果！


````python
# =====================================================================
# [2] Code 範例：多回合動態推進主迴圈實作
# =====================================================================

def simulate_days(R, C, k, m, initial_grid):
    # 深拷貝以免更動原資料
    grid = [row[:] for row in initial_grid]
    
    for day in range(1, m + 1):
        delta = [[0] * C for _ in range(R)]
        for r in range(R):
            for c in range(C):
                if grid[r][c] <= 0: continue
                move = grid[r][c] // k
                if move == 0: continue
                
                for dr, dc in ((-1, 0), (1, 0), (0, -1), (0, 1)):
                    nr, nc = r + dr, c + dc
                    if 0 <= nr < R and 0 <= nc < C and grid[nr][nc] != -1:
                        delta[r][c] -= move
                        delta[nr][nc] += move
                        
        for r in range(R):
            for c in range(C):
                grid[r][c] += delta[r][c]
                
    return grid

final_g = simulate_days(2, 3, 4, 1, [[10, 2, -1], [5, -1, 2]])
print("模擬 1 天後結果：", final_g)
````

### 📝 版本一：APCS 官方實作考場專用版 —— 淺顯易懂一般版（新手友善推薦）

* **適用情境**：APCS 正式考試現場、初學者課堂練習。
* **教學與設計理念**：
  1. **步驟平鋪直敘、拆解詳細**：依序完成「讀取參數與地圖」、「初始化 Delta 矩陣」、「四方向遍歷結算」、「更新地圖」四大步驟。
  2. **標準邊界防禦與 Delta 累加**：嚴格判斷 `0 <= nr < R` 與 `grid[nr][nc] != -1`，確保絕不越界且障礙不被污染。
  3. **穩健 100% 滿分 AC**：最後以清晰的列表推導式過濾 `-1` 提取極值，考場高壓下零失誤奪取滿分！


````python
# ==============================================================================
# 📝 版本一：APCS 官方實作考場專用版 —— 淺顯易懂一般版（新手友善推薦）
# 適用情境：APCS 正式考場單筆測試案例，平鋪直敘、步驟明確、穩健滿分
# ==============================================================================

# 步驟 1：讀取矩陣尺寸 R, C、遷移係數 k 與天數 m
R, C, k, m = map(int, input().split())

# 步驟 2：讀取初始網格地圖
grid = []
for _ in range(R):
    grid.append(list(map(int, input().split())))

# 定義四個相鄰方向向量（上、下、左、右）
dr = [-1, 1, 0, 0]
dc = [0, 0, -1, 1]

# 步驟 3：模擬 m 天的人口遷移
for day in range(m):
    # 建立當天的變動量矩陣（初值全為 0）
    delta = [[0] * C for _ in range(R)]
    
    # 遍歷地圖上的每一個格子
    for r in range(R):
        for c in range(C):
            # 若為障礙（-1）或無人口，不參與遷移
            if grid[r][c] <= 0:
                continue
            
            # 計算向單一相鄰方向遷移的人數（整數除法）
            move = grid[r][c] // k
            if move == 0:
                continue
            
            # 檢查上下左右四個方向
            for i in range(4):
                nr = r + dr[i]
                nc = c + dc[i]
                
                # 邊界防禦：不越界且相鄰格不是障礙（!= -1）
                if 0 <= nr < R and 0 <= nc < C and grid[nr][nc] != -1:
                    delta[r][c] -= move
                    delta[nr][nc] += move
                    
    # 當天遷移結束，統一將 Delta 變動量疊加回地圖
    for r in range(R):
        for c in range(C):
            grid[r][c] += delta[r][c]

# 步驟 4：排除 -1 障礙，收集所有合法城市的人口
cities = [val for row in grid for val in row if val != -1]

# 步驟 5：分行輸出最少與最多人口數
print(min(cities))
print(max(cities))
````

````python
# ==============================================================================
# ⚡ 版本二：APCS 官方實作考場專用版 —— 極簡高效精煉版（進階高手推薦）
# 適用情境：APCS 正式考試現場，哨兵加框法（Sentinel Padding）極速 20 行
# ==============================================================================

R, C, k, m = map(int, input().split())
# 哨兵加框：地圖四週包覆一圈 -1，徹底省去所有邊界檢查
g = [[-1] * (C + 2)] + [[-1] + list(map(int, input().split())) + [-1] for _ in range(R)] + [[-1] * (C + 2)]

for _ in range(m):
    d = [[0] * (C + 2) for _ in range(R + 2)]
    for r in range(1, R + 1):
        for c in range(1, C + 1):
            if g[r][c] <= 0: continue
            mv = g[r][c] // k
            if not mv: continue
            for nr, nc in ((r - 1, c), (r + 1, c), (r, c - 1), (r, c + 1)):
                if g[nr][nc] != -1:
                    d[r][c] -= mv
                    d[nr][nc] += mv
    for r in range(1, R + 1):
        for c in range(1, C + 1):
            g[r][c] += d[r][c]

cities = [g[r][c] for r in range(1, R + 1) for c in range(1, C + 1) if g[r][c] != -1]
print(min(cities))
print(max(cities))
````
