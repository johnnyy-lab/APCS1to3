### 🏷️ 4-4-1 字串標籤與變數混合輸出——變數名牌 vs 字串引號的本質區別

想像你去郵局寄掛號包裹。紙箱的正中央用毛筆寫著粗粗的固定標題：「收件人：」，而右邊則貼著一張可以隨時更換的白色貼紙，上面寫著「王小明」。郵差在投遞廣播時，眼睛看著紙箱，會唸出：「收件人：王小明」。如果郵差太過死板，對著麥克風大喊「收件人：收件人」，那全社區的人都會笑翻！
在 Python 的世界裡，這個道理完全一模一樣：**用引號包起來的文字是固定標題（字串標籤），而沒加引號的則是置物櫃上的名牌貼紙（變數）**！

**運作機制與底層原理：**
當我們在 `print()` 的小括號中同時放入字串與變數時，例如：
```python
player_name = "亞瑟"
print("冒險者姓名：", player_name)
```
Python 直譯器在執行這行指令時，固定遵循以下兩大步驟：
1. **引數逐一解析**：直譯器由左至右審視逗號隔開的每個項目。遇到 `"冒險者姓名："`，發現有成對引號，便認定它是純文字包裹，直接取用；遇到未加引號的 `player_name`，直譯器知道這是一個識別字（變數名牌），便立刻奔向記憶體置物櫃大樓，取出名牌背後存放的資料內容 `"亞瑟"`。
2. **預設間隔輸出**：在將所有項目送入標準輸出時，`print()` 預設會在項目與項目之間自動插入一個半形空格（`sep=' '`），因此螢幕上會漂亮地呈現出 `冒險者姓名： 亞瑟`。

**常見錯誤陷阱（初學者必讀三大地雷）：**
1. **把變數名也包進引號（字串與變數混淆）**：很多同學會寫成 `print("冒險者姓名：player_name")`。請記住：一旦字母被引號關進籠子裡，Python 就會把它當成死文字，螢幕只會印出死板的 `冒險者姓名：player_name`，而不是真正的玩家名字！
2. **忘記在項目間打逗號**：字串與變數之間必須以逗號 `,` 隔開，例如寫成 `print("冒險者姓名：" player_name)` 會直接引發 `SyntaxError: invalid syntax`（語法錯誤）！
3. **變數大小寫打錯引發 NameError**：如果宣告時寫 `Player_name = "亞瑟"`，輸出時卻寫 `player_name`，因為 Python 對大小寫極其敏銳，會立刻報錯 `NameError: name 'player_name' is not defined`！

**APCS 考試實務建議：**
在 APCS 與各類線上評判系統（Online Judge, OJ）中，很多題目會要求輸出帶有特定格式標籤的答案，例如 `Case 1: 42` 或 `Result = 100`。請仔細比對題目規定的大小寫、冒號前後的空白數量。運用逗號連接標籤字串與變數值，是最標準、穩健且不易出錯的基本功夫！


````python
# [2] Code 範例區：展示字串標籤與變數混合輸出

# 步驟 1：建立冒險者角色的各項數值（置物櫃名牌）
player_name = "亞瑟勇者"
level = 15
hp = 350
gold = 1280

# 步驟 2：使用字串標籤搭配單一變數輸出
print("【角色資料卡】")
print("冒險者名稱：", player_name)
print("角色等級：", level)
print("目前生命值：", hp)
print("持有金幣：", gold)

# 步驟 3：在同一行 print() 中混合多個標籤與多個變數
print("狀態摘要：", player_name, "等級為", level, "級，攜帶", gold, "枚金幣。")
````

### ⚡ 4-4-2 在 print() 中即時運算——算式先求值後輸出的精簡美學

在日常生活中，如果媽媽請你去水果攤買 3 顆蘋果，每顆 25 元，當你向老闆詢問價錢時，老闆不會先從抽屜拿出一張白紙寫下「$25 \times 3 = 75$」，把紙條收好，再告訴你「總共 75 元」。老闆的腦袋會直接在心裡算完，一口氣說出答案！
在寫程式時也是一樣的道理：如果某個計算結果只是單純要印在螢幕上給人看，後續完全不需要再拿去重複計算，那我們完全不需要額外宣告一個「免洗免洗、用完即丟的暫時變數」，而是可以直接把算式塞進 `print()` 的小括號裡進行「**現場即時運算**」！

**運作機制與底層原理：**
在 Python 中，所有函數的呼叫都嚴格遵守「**引數先求值法則（Argument Evaluation Before Call）**」。
當直譯器看到以下指令時：
```python
price = 25
quantity = 3
print("結帳總金額：", price * quantity, "元")
```
電腦內部的執行時序為：
1. 先處理第一個引數 `"結帳總金額："`（求值結果仍為該字串）。
2. 接著處理第二個引數 `price * quantity`。直譯器先查出 `price` 為 `25`、`quantity` 為 `3`，隨即進行乘法運算，瞬間將這個表示式求值化簡為單一數值 `75`！
3. 接著處理第三個引數 `"元"`。
4. 直譯器將化簡後的三個最終產物：`"結帳總金額："`, `75`, `"元"`，一併交由 `print()` 發聲廣播！
只要是 Python 允許的任何合法算式（包含 `+`, `-`, `*`, `/`, `//`, `%`, `**`，或是 `abs()`, `max()`, `min()` 數學函數），全部都能放在 `print()` 內部即時求值！

**常見錯誤陷阱（初學者必讀三大地雷）：**
1. **運算子優先順序陷阱（漏括號）**：
   例如計算兩數平均值時寫成：`print("平均值：", a + b / 2)`。由於「先乘除後加減」，電腦會先算 `b / 2` 再加上 `a`，導致答案全錯！必須加上小括號強制提升優先級：`print("平均值：", (a + b) / 2)`。
2. **誤把賦值等號放入引數**：
   有些同學會試圖在印出時順便賦值：`print("總計：", total = a + b)`。這在 Python 中是語法錯誤（SyntaxError），因為一般等號賦值是陳述句（Statement）而不是運算式（Expression），不能作為函數引數傳入。
3. **字串引號包錯位置**：
   如果不小心寫成 `print("面積是： length * width")`，算式被引號包住，電腦就會把它當成文字，原封不動印出英文字母 `length * width`，而不會算出數字！

**APCS 考試實務建議：**
在 APCS 競賽或檢定解題時，精簡程式碼行數有助於提升讀題與除錯效率。對於只需要在最後一步輸出的答案（例如兩點曼哈頓距離 `abs(x1 - x2) + abs(y1 - y2)` 或直角三角形面積 `a * b // 2`），直接在 `print()` 內部完成計算是高手們最常用的俐落寫法！


````python
# [2] Code 範例區：展示在 print() 中進行即時運算求值

length = 12
width = 8

# 範例 1：在 print() 內進行長方形面積與周長計算
print("長方形長度：", length, "，寬度：", width)
print("面積即時運算結果：", length * width, "平方公分")
print("周長即時運算結果：", 2 * (length + width), "公分")

# 範例 2：在 print() 內結合整數除法與餘數運算
candies = 35
kids = 4
print("每位小朋友分得：", candies // kids, "顆糖果")
print("最後剩下未分完：", candies % kids, "顆糖果")

# 範例 3：結合內建函數 max() 與 abs() 即時求值
score_A = 85
score_B = 92
print("兩位同學最高分為：", max(score_A, score_B), "分")
print("兩位同學分數相差：", abs(score_A - score_B), "分")
````

### 🧩 4-4-3 消除文字與數值突兀間隔——利用 sep='' 實現中文字詞與單位無縫貼合

在英文句子中，單字與單字之間本來就習慣空一格（例如 `Level 15` 或 `Score 100`）。但是，在中文排版或嚴謹的數學符號中，空格往往顯得非常突兀且不符合語法習慣！
試著比較以下兩種視覺效果：
- 突兀空格：`今天氣溫： 28 度，我的得分是 100 分，座標為 ( 10 , 20 )`
- 優美貼合：`今天氣溫：28度，我的得分是100分，座標為(10, 20)`
如果我們只是單純用逗號把字串與變數串接，`print()` 就會自動在中間塞入半形空格。想要徹底消除這些不必要的縫隙，我們在單元 4-2 學過的神奇開關——**`sep=''`（空字串分隔符）** 就是最佳解藥！

**運作機制與底層原理：**
當我們在 `print()` 的引數結尾加上 `sep=''` 時：
`print("得分：", score, "分", sep='')`
直譯器在串聯引數輸出時，設定各項目之間的「膠水黏著劑」為「長度為 0 的空字串」。這意味著相鄰的字串與變數數值會像磁鐵一樣緊緊貼合在一起，不再有任何預設空格插入！
那麼，如果某個地方真的需要空格呢？非常簡單！**哪裡需要空格，就在字串引號裡面親手打上空格**！
例如輸出數學座標 `(10, 20)`：
```python
x = 10
y = 20
print("(", x, ", ", y, ")", sep='')
```
因為 `", "` 的引號內部帶有一個空格，所以印出時只有逗號後方會有一格漂亮的留白，左右兩側的小括號則會緊貼著數字，呈現出最道地的數學符號排版！

**為什麼不用字串加號 `+` 拼接？（避開致命型態地雷）：**
很多初學者會想：「既然不想有空格，那我用加號 `+` 把文字拼起來不就好了嗎？」
像這樣寫：`print("得分：" + score + "分")`。
**這是一個超級大陷阱！** 當 `score` 是整數或浮點數時，Python 嚴禁字串與數字直接用加號相加，程式會立刻崩潰報出：
`TypeError: can only concatenate str (not "int") to str`！
你必須寫成很繁瑣的 `print("得分：" + str(score) + "分")`。
而使用 `print("得分：", score, "分", sep='')`，完全不需要手動呼叫 `str()` 進行轉型，`print()` 會自動幫你把所有數值安全、快速地轉化成文字貼合在一起！

**常見錯誤陷阱（初學者必讀三大地雷）：**
1. **忘記在需要空格的地方補空白**：使用 `sep=''` 後，所有預設空格都消失了。如果印出兩組資料卻沒有在字串內加空格，字詞會全擠在一起（例如印成 `姓名王小明學號01`，而不是 `姓名：王小明 學號：01`）。
2. **漏掉引數之間的逗號**：`sep` 控制的是「引數逗號」處的間隔。逗號依然不可或缺，否則會發生語法錯誤。
3. **sep 參數名稱打錯**：打成 `seps=''` 或 `sep = ""` 之外的字樣（如 `gap=''`）會引發 `TypeError`。

**APCS 考試實務建議：**
在 APCS 實作題中，輸出格式比對是極端嚴苛的。例如要求輸出 `(x,y)` 或 `Ans=42`，若因多印了預設空格而變成 `( x , y )` 或 `Ans = 42`，評判系統會毫無例外判定為 **Wrong Answer (WA)**！掌握 `sep=''` 是在 APCS 考試中精準控制輸出字元、避免被扣冤枉分的關鍵生存技能！


````python
# [2] Code 範例區：展示使用 sep='' 消除突兀空格與精準單位貼合

score = 100
rank = 1
x = 15
y = 28

# 對比 1：未使用 sep=''（預設空格導致中文單位與標點符號鬆散）
print("【未使用 sep 的效果】")
print("得分：", score, "分，名次：第", rank, "名")
print("座標點：(", x, ",", y, ")")

print("-" * 35)

# 對比 2：使用 sep=''（中文字詞與單位緊密貼合，哪裡要空格就在字串內手動打）
print("【使用 sep='' 的完美效果】")
print("得分：", score, "分，名次：第", rank, "名", sep='')
print("座標點：(", x, ", ", y, ")", sep='')

# 範例 3：商品特價與貨幣符號貼合
price = 450
print("全館大特價：NT$", price, "元！", sep='')
````

### 🚀 4-4-4 連續跨行與接續排版——結合 end 參數建構連貫算式與計算歷程

在解數學題時，老師常常會在黑板上「一邊寫、一邊講」：
先寫下題目「$15 + 27 = $」，接著停下筆問同學答案是多少，最後在同一個算式的後面瀟灑補上「$42$」。
如果程式也需要模擬這種「分階段組裝資訊」或「展現計算步驟推導」的情境，該怎麼做呢？
在單元 4-3 中，我們學到了 `end` 參數可以取消預設換行。當我們把 `end` 與本單元的「變數」、「即時運算式」結合時，就能實現強大無比的**「連貫資訊流接力排版」**！

**運作機制與底層原理：**
當我們連續呼叫兩次或多次 `print()` 時：
```python
a = 15
b = 27
print("算式：", a, "+", b, "=", end=' ')
print(a + b, "(計算完成)")
```
直譯器的執行軌跡如下：
1. **第一棒出發**：執行第一個 `print()`，印出 `算式： 15 + 27 =`。因為指定了 `end=' '`（以一個空格收尾，不換行），游標會乖乖停留在等號後方的一格位置待命。
2. **第二棒接軌**：緊接著執行第二個 `print()`，算式 `a + b` 被即時計算成 `42`，大喇叭直接從游標待命處往後接續印出 `42 (計算完成)`。由於第二個 `print()` 沒有設定 `end`，它會使用預設的 `
` 在整句話末端敲下鈴鐺、正常換行！
透過這種技巧，我們可以將「複雜的文字組裝」拆分成多個邏輯簡短的小指令，讓程式碼結構更清晰、更容易閱讀與除錯。

**常見錯誤陷阱（初學者必讀三大地雷）：**
1. **接續時文字撞在一起（未留適當間隔）**：
   如果前一個指令設定 `end=''`（空字串），且字串末尾沒有留空格，後一個指令印出的字就會直接貼在前一個字的屁股上，例如印出 `15+27=42` 或 `名稱：小明成績：100`。務必注意是在 `end` 中補空格，還是在字串內留空格！
2. **最後一棒忘記讓游標換行**：
   如果整組算式的「最後一個 `print()`」也設定了 `end=''`，游標就不會換到下一行。若後續還有其他程式碼輸出，內容就會全部擠在同一行，破壞整個版面配置！
3. **把 end 當成了 sep**：
   請再次複習黃金口訣：「**`sep` 在中間隔開夥伴，`end` 在句尾決定換不換行**」！兩者各司其職，切勿混淆。

**APCS 考試實務建議：**
在 APCS 程式題的除錯過程中，最實用的技巧之一叫做「**執行軌跡追蹤（Tracing）**」。當你想觀察變數在多個運算步驟中的數值演變時，可以在每個步驟用 `print(step_val, end=' -> ')` 連續印出，最後印出 `print("Done")`。這樣就能在同一行看到如 `1 -> 2 -> 4 -> 8 -> Done` 的清楚軌跡，是抓出演算法臭蟲的高級神技！


````python
# [2] Code 範例區：展示結合 end 參數建構連貫算式與計算歷程

a = 15
b = 27

# 範例 1：分段組裝數學方程式（第一句不換行，第二句接續並收尾換行）
print("數學計算題：", a, "+", b, "=", end=' ')
print(a + b, "(正確解答)")

# 範例 2：連貫推導演算歷程（使用箭頭符號接續）
num = 10
print("【數值演變歷程】", end=' ')
print(num, end=' -> ')
num = num * 2
print(num, end=' -> ')
num = num + 5
print(num, end=' (最終結果)\n')

# 範例 3：提示標籤與即時計算結果分段輸出
base = 12
height = 5
print("直角三角形底=", base, "高=", height, "-> 面積為：", end=' ')
print(base * height / 2, "平方公分")
````

### 🏆 4-4-5 實戰全能儀表板——變數、即時運算、sep 與 end 的終極融合

恭喜你！走到了這個章節的巔峰舞台。現在，你手上的工具箱已經裝滿了 Python 輸出排版的所有神兵利器：
- **變數（Variables）**：儲存核心資料與狀態。
- **即時運算式（Expressions）**：括號內現場計算求值，化簡代碼。
- **`sep=''` 分隔參數**：消除突兀空格，讓文字標籤與幣別、單位、符號無縫黏合。
- **`end` 結尾參數**：調控換行節奏，精準建構跨行分段或大段落拉開。

當這些技術在同一個專案中相互交織時，我們就能建構出如同專業收銀機收據、銀行帳單、學生成績單或遊戲儀表板一樣精美的商業級輸出介面！

**專業架構思維：何時該宣告變數？何時該用現場算式？**
在設計多項目的資訊儀表板時，優秀的工程師會做出明智的選擇：
1. **適合宣告獨立變數的時機**：
   - 原始輸入資料（如商品單價、購買數量、折扣趴數）。
   - **後續會被重複使用兩次以上的運算成果**（例如：商品總金額 `subtotal` 在算折扣時要用、算應付總額時也要用，這時就該先算好存入 `subtotal = price * qty`）。
2. **適合在 print() 內現場即時運算的時機**：
   - 僅供畫面展示、一次性輸出的衍生數值（例如：`subtotal - discount` 應付差額，或是平均值 `(c + e + m) / 3`）。

**常見錯誤陷阱（初學者必讀三大地雷）：**
1. **sep 與 end 在同一個 print() 中搞混位置**：
   例如寫成 `print("總額：", total, "元", end='', sep='\n')`，本來想消除中間空格並換行，結果把 `sep` 設為換行、`end` 設為空字串，印出來每個字自己跳一行，版面徹底錯亂！
2. **多引數時漏掉 sep 導致排版鬆散**：
   在做表格或收據邊界線時，若沒有設定 `sep=''`，字串與變數間會產生預設空格，使表格的框線長度失準對不齊。
3. **除法浮點數的小數點過長**：
   在計算平均值或比例時，如果直接除法可能會產生 `89.33333333333333`。此時可以搭配在第 3 章學過的 `round(val, 1)` 或 `round(val, 2)` 進行適度修剪，讓畫面保持整潔美觀！

**APCS 考試實務建議：**
APCS 實作題的第一題與第二題（如成績統計、購物車結帳、矩陣資訊輸出等），常常需要同時處理「數值運算」與「標準格式回報」。掌握「變數儲存關鍵資料 ➔ 現場算式計算衍生量 ➔ sep 與 end 雕琢字元細節」這套標準心法，能讓你寫出的程式碼既優雅簡潔，又能 100% 吻合評判系統的格式要求！


````python
# [2] Code 範例區：展示綜合儀表板排版（超市電子結帳收據）

# 步驟 1：建立核心原始資料（變數）
item_name = "特級草莓蛋糕"
price = 120
quantity = 3
coupon = 50

# 步驟 2：針對會重複使用的中間成果宣告變數
subtotal = price * quantity
final_total = subtotal - coupon

# 步驟 3：運用 sep=''、即時算式與邊界線打造專業收據儀表板
print("=" * 36)
print("        旗美精緻烘焙坊 結帳收據        ")
print("=" * 36)
print("購買品項：", item_name, sep='')
print("單價：NT$", price, " | 購買數量：", quantity, " 個", sep='')
print("商品小計：NT$", subtotal, " 元", sep='')
print("優惠券折抵：-NT$", coupon, " 元", sep='')
print("-" * 36)
# 在 print 內部現場計算平均單件實付成本：final_total / quantity
print("實付總計：NT$", final_total, " 元 (平均單件NT$", round(final_total / quantity, 1), "元)", sep='')
print("=" * 36)
````
