### 📝 題目描述（Problem Description）
某公司擁有兩座工廠（工廠一與工廠二），現在共有 $n$ 位新進員工需要進行人力分配。
兩座工廠依據分配進去的員工數量，各自擁有一套二次函數收益模式：
- 若分配 $X_1$ 名員工到**工廠一**，工廠一產生的收益為：  
  $$Y_1 = A_1 \times X_1^2 + B_1 \times X_1 + C_1$$
- 若分配 $X_2$ 名員工到**工廠二**，工廠二產生的收益為：  
  $$Y_2 = A_2 \times X_2^2 + B_2 \times X_2 + C_2$$

⚠️ **分配限制條件**：
1. 所有員工都必須被分配到其中一座工廠，即 $X_1 + X_2 = n$。
2. 分配給各工廠的人數不能為負數，即 $X_1 \ge 0$ 且 $X_2 \ge 0$（可以有某座工廠分配 0 人的極端情況）。
3. 總收益為兩座工廠收益的總和，即 $\text{Total Profit} = Y_1 + Y_2$。

請撰寫一個程式，找出最佳的人力分配方案，使得兩座工廠的**總收益達到最大**，並輸出該最大總收益。

### 14.6.1 題意解析：將 $n$ 名員工分配給兩工廠求最大利潤

#### 🍰 生活化比喻：烘焙大師的人力調度
想像你是一家連鎖烘焙坊的總經理，旗下一共有兩家分店：「旗艦一廠」與「概念二廠」。總部今年招聘了 $n$ 位熱血的新進麵包師傅，現在要將他們全數派往這兩家工廠服務。
你可能會問：「為什麼師傅人數越多，利潤不是單純的成倍增長呢？」
在真實的工廠運作中，這非常符合經濟學原理：
1. 當某座工廠師傅太少時，昂貴的烤箱與產線處於閒置狀態，生產規模不足；
2. 當師傅數量適中時，大家分工合作、各司其職，產能與收益迎來爆發性增長；
3. 但如果把太多師傅擠在同一個廚房裡，工作空間擁擠、器材排隊爭搶、甚至動線打結，多增加的人力反而可能帶來額外的混亂與損耗！
這種「先上升後下降」或「具備曲線特徵」的產能關係，在數學上就是經典的**二次函數拋物線模型**（$Y = A X^2 + B X + C$）。

#### 🔗 人力分配的對偶性與互補約束
本題最重要的核心約束條件是：**所有員工都必須被分配，且人數不能為負數**。
也就是說：$X_1 + X_2 = n$ 且 $X_1 \ge 0, X_2 \ge 0$。
這個數學關係為我們帶來了一個巨大的程式設計優勢——**對偶性（Duality）**：
- 假設我們決定指派 $X_1$ 個人前往一廠；
- 那麼二廠該分配多少人呢？根本不需要另外猜測或多重迴圈搜尋，**二廠的人數必定剛好等於剩下的所有人：$X_2 = n - X_1$**！

因此，整個決策空間其實只取決於一個變數 $X_1$。只要 $X_1$ 的值決定了，$X_2$ 就自動鎖定，兩廠的個別收益與合計總收益也能隨之瞬間精確計算出來！

#### 💡 為什麼直接枚舉（Brute Force）是考場第一首選？
初學同學在高中數學課學過二次函數求極值，可能會忍不住拿出草稿紙嘗試「將 $X_2 = n - X_1$ 代入方程式，整理成關於 $X_1$ 的新一元二次方程式，再利用配方法或微分公式找頂點坐標」。
千萬不要在考場上這樣做！原因有三：
1. **代數展開極易筆誤**：在緊湊的考場環境中，手算符號與平方展開很容易算錯正負號；
2. **頂點可能不是整數**：二次函數算出來的對稱軸極值可能落在非整數（例如 $X_1 = 3.7$），此時你還得額外去比較 3 和 4 哪個整數點收益較大；
3. **計算規模極小**：題目明白標示 $n \le 100$，所有可能的分配方案只有 $X_1 = 0, 1, 2, \dots, n$ 共 $n + 1$ 種！在電腦眼中，讓迴圈跑 101 次連 0.0001 秒都不到，寫一個簡單的 `for` 迴圈直接全部算一遍，既安全、直觀，又能百分之百保證正確通關！

````python
# ==============================================================================
# 14.6.1 [2] Code 範例區：給定總人數 n，列出所有合法的 (X1, X2) 分配組合
# ==============================================================================

# 設定待分配的員工總人數 n
n = 4

print(f"【範例展示】員工總人數 n = {n} 的所有可能分配組合：")
print("-" * 50)

# 透過單層 for 迴圈枚舉分配給一廠的人數 x1
# 注意：x1 的範圍從 0 人一直到 n 人，所以 range 必須寫 n + 1！
for x1 in range(n + 1):
    x2 = n - x1  # 二廠人數自動由互補關係決定
    print(f"方案 {x1 + 1}：一廠分配 {x1} 人，二廠分配 {x2} 人（合計：{x1 + x2} 人）")

print("-" * 50)
print(f"總計共測試了 {n + 1} 種完全不重複的合法分配配置！")
````

### 14.6.2 二次函數收益公式計算：$a x^2 + b x + c$

#### 📐 數學式在 Python 直譯器中的精確映射
在 APCS 題目中，兩座工廠的產能收益公式均為標準的二次多項式：
$$Y = A \times X^2 + B \times X + C$$
在 Python 語言中，計算某個數量的平方有多種寫法，但它們在執行效率與運算優先級上有細微且關鍵的差異：
1. **寫法一（乘方運算子 `**`）**：`A * (x ** 2) + B * x + C`  
   Python 的 `**` 代表次方運算。雖然 Python 的運算優先級中 `**` 高於乘法 `*`（即 `A * x ** 2` 會先算 `x ** 2` 再乘以 `A`），但為了防止初學同學記憶模糊或混淆，**加上括號永遠是最穩健的防護**。
2. **寫法二（連續乘法 `x * x`，考場強烈推薦 ⭐）**：`A * x * x + B * x + C`  
   在整數運算中，直接寫 `x * x` 不僅不需要呼叫底層的次方函數模組，運算速度極快，而且視覺上完全沒有運算子優先級的閱讀負擔，既直觀又不易出錯！

#### ⚠️ 算式括號與負數係數的符號地雷
題目中的係數 $A, B, C$ 可能是負整數（例如 $B = -5, C = -10$）。
當我們在程式碼中寫 `A * x * x + B * x + C` 時：
- 如果 $B = -5$，$x = 2$，則 `B * x` 就是 `(-5) * 2 = -10`；
- 加上負數在數學上就等於減去正數，Python 會非常聰明且正確地處理正負號運算；
- 初學同學千萬不要手動去改寫程式碼中的加減號（例如自己改寫成減號），直接使用統一的加號公式 `+ B * x + C`，將正負號完全交給變數本身的數值決定，才能確保程式碼適用於任何正負係數組合！

#### 🏢 雙工廠獨立計算與合計收益
在一組具體的人力分配 $(X_1, X_2)$ 確定之後，我們可以分步或合併計算總收益：
```python
# 分開計算各廠收益
profit1 = A1 * x1 * x1 + B1 * x1 + C1
profit2 = A2 * x2 * x2 + B2 * x2 + C2
# 合計總收益
total_profit = profit1 + profit2
```
分步寫法不僅結構清爽、易於閱讀，而且在除錯時可以隨時印出 `profit1` 與 `profit2` 分別核對，是競賽初期非常推薦的好習慣。

### 14.6.3 for 迴圈完整枚舉分配人數 $(x_1, x_2)$ 滿足 $x_1 + x_2 = n$

#### 🪓 窮舉演算法（Brute Force）的實作本質
現在，我們已經掌握了「產生分配組合 $(x_1, x_2)$」與「計算二次函數收益」這兩大核心工具。接下來，我們要將兩者結合，運用 `for` 迴圈地毯式搜尋每一種可能的分配方案！

讓我們先觀察當員工總數 $n = 2$ 時的所有枚舉過程：
- 當 $x_1 = 0$ 時，二廠 $x_2 = 2 - 0 = 2$：
  一廠收益 $Y_1 = 2(0)^2 - 1(0) + 3 = 3$
  二廠收益 $Y_2 = 4(2)^2 - 5(2) + 2 = 16 - 10 + 2 = 8$
  合計總收益 $= 3 + 8 = 11$
- 當 $x_1 = 1$ 時，二廠 $x_2 = 2 - 1 = 1$：
  一廠收益 $Y_1 = 2(1)^2 - 1(1) + 3 = 4$
  二廠收益 $Y_2 = 4(1)^2 - 5(1) + 2 = 1$
  合計總收益 $= 4 + 1 = 5$
- 當 $x_1 = 2$ 時，二廠 $x_2 = 2 - 2 = 0$：
  一廠收益 $Y_1 = 2(2)^2 - 1(2) + 3 = 8 - 2 + 3 = 9$
  二廠收益 $Y_2 = 4(0)^2 - 5(0) + 2 = 2$
  合計總收益 $= 9 + 2 = 11$

#### 🚨 考場致命差一錯誤（Off-by-one Error）：`range(n)` vs `range(n + 1)`
這是無數初學者在 APCS 考場上痛失分數的隱形陷阱！
很多同學直覺寫下：
```python
for x1 in range(n):  # ❌ 致命錯誤！
    x2 = n - x1
```
請停下腳步仔細思考：如果 $n = 2$，`range(2)` 產生的數值只有 `0` 和 `1`！
這意味著你的迴圈**永遠測不到 $x_1 = 2, x_2 = 0$ 這種極端情況**！
而在上面的計算中，剛好 $x_1 = 2$ 時的總收益也是最大值 $11$；如果某一組測資的最佳解剛好出現在「把所有人都派往一廠（$x_1 = n, x_2 = 0$）」，那麼漏掉最後一筆的程式碼就會算出錯誤的次佳解，直接吞下 Wrong Answer (WA)！

記住這條黃金準則：
> 凡是要完整枚舉 $0$ 到 $n$ 的所有整數，在 Python 中必須無條件寫作：`for x1 in range(n + 1):`！

````python
# ==============================================================================
# 14.6.3 [2] Code 範例區：枚舉 n = 2 時所有分配組合並印出各方案總利潤
# ==============================================================================

A1, B1, C1 = 2, -1, 3
A2, B2, C2 = 4, -5, 2
n = 2

print(f"【完整枚舉試算表】員工總數 n = {n}")
print("=" * 60)
print(" 一廠人數 | 二廠人數 | 一廠收益 | 二廠收益 | 合計總收益")
print("-" * 60)

# 使用 range(n + 1) 確保包含 x1 = n 的極端邊界
for x1 in range(n + 1):
    x2 = n - x1
    profit1 = A1 * x1 * x1 + B1 * x1 + C1
    profit2 = A2 * x2 * x2 + B2 * x2 + C2
    total = profit1 + profit2
    print(f"    {x1:2d}    |    {x2:2d}    |   {profit1:4d}   |   {profit2:4d}   |    {total:4d}")

print("=" * 60)
print("觀察發現：總收益在 (0, 2) 與 (2, 0) 時皆達到最高峰 11！")
````

### 14.6.4 最大值變數初始化（初始值設為極小負數防護）

#### 💥 考場最大 WA 陷阱：`max_profit = 0` 的自毀行為
在基礎程式語法中，當我們想要尋找一連串數字中的最大值時，許多初學同學習慣寫成：
```python
max_profit = 0  # 🚨 在本題中這是致命毒藥！
```
如果今天的問題是求正數的最高分或最長距離，設為 0 或許不會出事。但是，**請務必牢記：APCS 的題目往往隱藏著負數邊界！**

讓我們回顧【範例 2】的震撼情境：
- 兩座工廠的係數全是負數（例如機台維護費高昂且產能過剩）；
- 經過所有方案枚舉後，算出來的四種總收益分別為：$-57, -46, -41, -42$；
- 在這四種方案中，真正的最佳解（虧損最少）是 **`-41`**；
- 然而，如果你把 `max_profit` 初始化為 `0`：
  - 當遇到 `-57` 時，`-57 > 0` 不成立；
  - 當遇到 `-46` 時，`-46 > 0` 不成立；
  - 當遇到 `-41` 時，`-41 > 0` 依然不成立；
  - 迴圈結束後，你的程式會大言不慚地印出 `0`！
因為所有真實的收益都比 0 還小，0 從頭到尾都沒有被替換掉，評測系統立刻無情判定 **Wrong Answer (WA)**！

#### 🛡️ 兩大滿分防禦策略
為了徹底免疫負數陷阱，專業選手有兩種標準防禦姿勢：

1. **策略一：初始化為負無窮大（`-float('inf')` 或極大負整數，最推薦 ⭐）**：
   ```python
   max_profit = -float('inf')  # 數學上的負無窮大，任何真實數值都比它大
   # 或者在 Python 競賽中寫：
   max_profit = -10**9         # 負十億，遠小於本題可能出現的最小值
   ```
   如此一來，迴圈測試的第一個方案（無論是多麼慘烈的負數），都能順利取代初值成為第一個基準！

2. **策略二：以第一個合法方案作為初始基準值**：
   ```python
   # 先手動算好 x1 = 0 這一種方案作為初始標竿
   max_profit = (A1 * 0 + B1 * 0 + C1) + (A2 * n * n + B2 * n + C2)
   # 接著迴圈從 x1 = 1 走到 n
   for x1 in range(1, n + 1):
       ...
   ```
兩種方法皆能在考場上百分之百阻絕 WA，其中使用 `-float('inf')` 程式碼最短、最不易出現複製粘貼失誤，強烈建議初學同學將其內化為肌肉記憶！

````python
# ==============================================================================
# 14.6.4 [2] Code 範例區：對比 max_profit = 0（錯誤）vs -float('inf')（正確）
# ==============================================================================

# 模擬全負收益的測試清單（範例 2 的四種利潤）
profits = [-57, -46, -41, -42]

print("【負收益地雷對照實測】")
print(f"所有方案收益清單：{profits}")
print("-" * 50)

# ❌ 錯誤示範：設初值為 0
wrong_max = 0
for p in profits:
    if p > wrong_max:
        wrong_max = p
print(f"❌ 錯誤做法 (初值 0) 輸出的最大值：{wrong_max}（完全錯誤！根本沒有 0 的方案）")

# ✅ 正確做法：設初值為 -float('inf')
correct_max = -float('inf')
for p in profits:
    if p > correct_max:
        correct_max = p
print(f"✅ 正確做法 (初值 -inf) 輸出的最大值：{correct_max}（完美命中正確答案 -41！）")
print("-" * 50)
````

### 14.6.5 迴圈內動態更新最大總收益與完整 AC 程式

#### 🧩 考場完整解題積木拼裝圖解
現在，我們擁有了所有通關的關鍵拼圖，讓我們以 APCS 正式考場的視角，手把手將這道題目組裝成極度精準、優雅的 12 行滿分程式碼！

整個程式的運作流程可分為極為清爽的三大樂章：

1. **第一樂章：精準讀取三行輸入（Input Mapping）**：
   - 第 1 行是工廠一的係數：`A1, B1, C1 = map(int, input().split())`
   - 第 2 行是工廠二的係數：`A2, B2, C2 = map(int, input().split())`
   - 第 3 行是總人數單一整數：`n = int(input())`

2. **第二樂章：枚舉搜尋與極值動態更新（Search & Max Update）**：
   - 首先架設安全護欄：`max_profit = -float('inf')`
   - 啟動單層迴圈：`for x1 in range(n + 1):`
   - 計算二廠人數：`x2 = n - x1`
   - 代入二次多項式：
     `profit = (A1 * x1 * x1 + B1 * x1 + C1) + (A2 * x2 * x2 + B2 * x2 + C2)`
   - 更新歷史最高點：`if profit > max_profit: max_profit = profit`（或 `max_profit = max(max_profit, profit)`）

3. **第三樂章：輸出唯一解答（Clean Output）**：
   - 迴圈結束後，`max_profit` 中儲存的必定是經過全盤考驗後的絕對最高收益！
   - 乾淨印出：`print(max_profit)`

#### 💡 考場速度與可讀性的平衡
這套架構不依賴任何外部套件、不需要動態串列宣告、記憶體開銷微乎其微。打字只需 1 分鐘，邏輯簡單透明到不可能出錯，這就是 APCS 實作題追求的極致穩定性！

````python
# ==============================================================================
# 14.6.5 [2] Code 範例區：手把手組裝完整解題流程（以範例 1 模擬輸入）
# ==============================================================================

# 模擬讀取輸入資料
A1, B1, C1 = 2, -1, 3
A2, B2, C2 = 4, -5, 2
n = 2

# 初始化最大收益為負無窮大
max_profit = -float('inf')

# 完整走訪一廠人數 x1（從 0 到 n）
for x1 in range(n + 1):
    x2 = n - x1
    # 計算總收益
    profit = (A1 * x1 * x1 + B1 * x1 + C1) + (A2 * x2 * x2 + B2 * x2 + C2)
    # 動態更新最大值
    if profit > max_profit:
        max_profit = profit

# 輸出最終滿分結果
print(f"APCS 評判系統預期輸出：{max_profit}")
````

### 14.6.6 複雜度分析：$O(n)$ 極速枚舉與負收益陷阱

#### ⏱️ 時間與空間複雜度深入分析
在 APCS 評測系統中，衡量一段程式能否拿到滿分的核心指標就是**時間複雜度**與**空間複雜度**：
1. **時間複雜度（Time Complexity）為 $O(n)$**：
   - 我們的 `for` 迴圈只會執行 $x_1 = 0$ 到 $n$ 共 $n + 1$ 次；
   - 每次迴圈內部只進行簡單的整數加減乘法與 `max()` 比大小，這些運算均為 $O(1)$ 的常數時間；
   - 因此，總時間複雜度為嚴格的線性時間 $O(n)$！
   - 題目規定 $n \le 100$，這意味著迴圈最多只跑 101 次！在現代電腦上，Python 可以在 1 秒內執行約 $10^7$ 次基本運算，101 次運算只需約 **0.00001 秒**，遠遠低於官方 1.0 秒的時間上限，絕無 TLE（Time Limit Exceeded）風險。
2. **空間複雜度（Space Complexity）為 $O(1)$**：
   - 我們只宣告了 `A1, B1, C1, A2, B2, C2, n, max_profit, x1, x2, profit` 等少數幾個純量整數變數；
   - 沒有建立龐大的陣列或矩陣，記憶體消耗甚至不到 1 KB，輕巧滿足 64 MB 的限制。

````python
# ==============================================================================
# 📝 【版本一：APCS 官方實作考場專用版】
# 適用情境：APCS 正式考試環境（題目保證單組 3 行輸入，極簡快速滿分）
# ==============================================================================

# 1. 讀取第 1 行：工廠一的二次函數係數 A1, B1, C1
A1, B1, C1 = map(int, input().split())

# 2. 讀取第 2 行：工廠二的二次函數係數 A2, B2, C2
A2, B2, C2 = map(int, input().split())

# 3. 讀取第 3 行：待分配的員工總人數 n
n = int(input())

# 4. 初始化最大收益為負無窮大（防禦所有收益皆為負數的極端邊界陷阱）
max_profit = -float('inf')

# 5. 枚舉分配給一廠的人數 x1（範圍從 0 人到 n 人，故 range 必須是 n + 1）
for x1 in range(n + 1):
    # 二廠人數 x2 自然為總人數 n 扣除一廠人數 x1
    x2 = n - x1
    
    # 計算當前分配配置下的合計總收益
    profit = (A1 * x1 * x1 + B1 * x1 + C1) + (A2 * x2 * x2 + B2 * x2 + C2)
    
    # 動態更新歷程最大總收益
    if profit > max_profit:
        max_profit = profit

# 6. 輸出最終計算出的最大總收益
print(max_profit)
````
