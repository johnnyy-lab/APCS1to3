## 10.7.1 范氏圖（Venn Diagram）幾何模型與集合四大運算

### 觀念說明
在數學中，我們常用**范氏圖（Venn Diagram）**來視覺化兩個集合 $A$ 與 $B$ 之間的關係：
想像在黑板上畫了兩個互相重疊的圓形：
1. **交集（Intersection）`A & B`**：兩個圓形重疊的中央核心區域（同時屬於 $A$ 且屬於 $B$）。
2. **聯集（Union）`A | B`**：兩個圓形包含的所有範圍（屬於 $A$ 或屬於 $B$ 的全體成員）。
3. **差集（Difference）`A - B`**：左邊被挖掉一塊的月牙形區域（只屬於 $A$、不屬於 $B$ 的獨佔成員）。
4. **對稱差（Symmetric Difference）`A ^ B`**：左右兩側的月牙形合併（只屬於 $A$ 或只屬於 $B$，但排除兩者共有者）。

在 Python 中，這四大運算都有專屬的單一符號運算子，寫起來就像一般加減乘除一樣簡潔流暢！

```python
# 10.7.1 範例：集合四大運算全景預覽
demo_grp = {1, 2, 3, 4, 5}
demo_grp_b = {4, 5, 6, 7, 8}

print("集合 A:", demo_grp)
print("集合 B:", demo_grp_b)
print("-" * 40)

# 1. 交集 & (共同重疊)
print("交集 A & B (兩邊都有):", demo_grp & demo_grp_b)

# 2. 聯集 | (全部合併去重)
print("聯集 A | B (全體成員):", demo_grp | demo_grp_b)

# 3. 差集 - (A 扣除 B)
print("差集 A - B (A 獨有):", demo_grp - demo_grp_b)
print("差集 B - A (B 獨有):", demo_grp_b - demo_grp)

# 4. 對稱差 ^ (非共同獨佔成員)
print("對稱差 A ^ B (排除共有):", demo_grp ^ demo_grp_b)
```

```python
# 10.7.1 填空題：計算兩班選修課程的交集與聯集
clss_crss = {"Python", "Math", "English"}
clssbcrss = {"Python", "Music", "Physics"}

# 任務：使用 & 計算共同選修課程，使用 | 計算兩班所有選修課程總覽
cmmn_crss = clss_crss & clssbcrss
allcrss107 = clss_crss | clssbcrss

# 請將上方運算子填入空格：
# cmmn_crss = clss_crss ___ clssbcrss
# allcrss107 = clss_crss ___ clssbcrss
print("共同選修課程:", cmmn_crss)
print("全體課程種類數:", len(allcrss107))
# 預期輸出:
# 共同選修課程: {'Python'}
# 全體課程種類數: 5
```

```python
# 10.7.1 練習題：兩班共同興趣社團人數統計
# 題目說明：
# 給定甲班同學參加的社團清單 club_a，以及乙班同學參加的社團清單 club_b。
# 請先將兩清單轉為集合，並依序輸出：
# 1. 兩班都有人參加的「共同社團清單」（排序後輸出）
# 2. 兩班總共涵蓋了幾種「不同」的社團（聯集總數）
#
# 【公開測試資料 1】
# 甲班: ["Chess", "Robotics", "Drama"]
# 乙班: ["Robotics", "Music", "Chess", "Art"]
# 輸出:
# 共同社團: ['Chess', 'Robotics']
# 總社團種類: 5
#
# 【公開測試資料 2】
# 甲班: ["Basketball", "Baseball"]
# 乙班: ["Tennis", "Soccer"]
# 輸出:
# 共同社團: []
# 總社團種類: 4

club_a = ["Chess", "Robotics", "Drama"]
club_b = ["Robotics", "Music", "Chess", "Art"]

set_a = set(club_a)
set_b = set(club_b)

common = sorted(set_a & set_b)
tot_cnt = len(set_a | set_b)

print("共同社團:", common)
print("總社團種類:", tot_cnt)
```

## 10.7.2 交集運算子 & 與方法 intersection()（共同名單 / 共同選課）

### 觀念說明
交集代表「既屬於 A 同時也屬於 B」的交疊部分。在 Python 中，有兩種主要呼叫方式：
1. **運算子語法**：`A & B`
   簡潔好寫，但限制是**兩邊必須都已經是集合物件（set）**，否則會噴 `TypeError`。
2. **方法語法**：`A.intersection(iterable)`
   彈性更高！括號內可以接受任何可迭代物件（如串列、元組、字串），Python 會在內部自動轉換，例如 `A.intersection([1, 2, 3])`。

### APCS 經典解題情境
- **多科雙霸榜**：同時獲得數學滿分與英文滿分的學生名單。
- **多邊形公共頂點**：找出兩條路徑交叉經過的共同座標。
- **共同關鍵字匹配**：兩篇文章同時提及的標籤關鍵字。

```python
# 10.7.2 範例：學期雙滿分榮譽榜交集篩選
demo_mhnrs = {"Alice", "Bob", "Charlie", "David"}
demo_shnrs = {"Charlie", "Eva", "Alice", "Frank"}

# 1. 使用運算子 &
demo_bhnrs = demo_mhnrs & demo_shnrs
print("數理雙滿分榮譽榜 (&):", sorted(demo_bhnrs))

# 2. 使用方法 .intersection() 傳入一般串列
demo_elst = ["Alice", "Frank", "Charlie", "George"]
demo_athr = demo_bhnrs.intersection(demo_elst)
print("三科全數滿分榮譽榜 (.intersection):", sorted(demo_athr))

# 3. 若沒有交集，會回傳空集合 set()
demo_nmtch = demo_mhnrs & {"Nobody"}
print("無交集時的結果:", demo_nmtch)
```

```python
# 10.7.2 填空題：求兩數列之公共質數交集
primes = {2, 3, 5, 7, 11, 13, 17, 19}
sample_ns = {4, 7, 9, 11, 15, 17, 21}

# 任務：使用 & 求出 sample_ns 中的所有質數
cmmn_prms = sample_ns & primes

# 請填入空格：
# cmmn_prms = sample_ns ___ primes
print("共同質數:", sorted(cmmn_prms))
# 預期輸出: 共同質數: [7, 11, 17]
```

```python
# 10.7.2 練習題：跨社團共同核心幹部篩選器
# 題目說明：
# 學生會幹部名單 stdn_cncl，吉他社幹部名單 gtr_clb。
# 請使用集合交集運算子 &，找出同時在兩組織擔任幹部的「雙棲幹部」名冊。
# 若有雙棲幹部，輸出依字母排序後的姓名串列；
# 若無雙棲幹部，輸出 "無共同幹部"。
#
# 【公開測試資料 1】
# 學生會: ["Alice", "Bob", "Ken"]
# 吉他社: ["Ken", "Mary", "Alice"]
# 輸出: 雙棲幹部: ['Alice', 'Ken']
#
# 【公開測試資料 2】
# 學生會: ["Leo", "Zoe"]
# 吉他社: ["Sam", "Amy"]
# 輸出: 無共同幹部

stdn_cncl = ["Alice", "Bob", "Ken"]
gtr_clb = ["Ken", "Mary", "Alice"]

dual_roles = set(stdn_cncl) & set(gtr_clb)

if len(dual_roles) > 0:
    print("雙棲幹部:", sorted(dual_roles))
else:
    print("無共同幹部")
```

## 10.7.3 聯集運算子 | 與方法 union()（全體名冊合併）

### 觀念說明
聯集運算代表「屬於 A 或屬於 B 的全體元素整合」。
語法結構：
1. 運算子：`A | B`
2. 方法：`A.union(iterable)`

### 聯集的神奇特性：跨來源極速合併且絕不重複！
在大型系統中，資料往往來自不同分店、不同檔案或不同伺服器。若直接用串列相加 `list_a + list_b`，重複的會員會被計算多次，後續還得耗費資源去重。
使用集合聯集 `A | B`，Python 會在底層直接將兩個雜湊表合而為一，**重疊的資料會被自動熔接為單一元素**！
時間複雜度僅為 $O(len(A) + len(B))$，是處理多資料源合併去重的唯一首選。

```python
# 10.7.3 範例：分店會員全台大聯集整合
demo_tpcli = {"Alice", "Bob", "Charlie"}
demo_tcli = {"Charlie", "David", "Eva"}
demo_khhcl = {"Eva", "Frank", "Alice"}

# 1. 使用 | 運算子連續聯集
demo_acli = demo_tpcli | demo_tcli | demo_khhcl

print("全台不重複會員總數:", len(demo_acli))
print("會員名單總覽:", sorted(demo_acli))

# 2. 就地聯集更新 |= (相當於 A = A | B)
demo_rpl = {"VIP_01", "VIP_02"}
demo_nbtch = {"VIP_02", "VIP_03"}
demo_rpl |= demo_nbtch
print("就地聯集擴充後:", demo_rpl)
```

```python
# 10.7.3 填空題：合併兩組電話號碼通訊錄
sm1_cntc = {"0911-111", "0922-222"}
sm2_cntc = {"0922-222", "0933-333"}

# 任務：使用聯集運算子 | 將兩張 SIM 卡的聯絡人合併
mrgd_cntc = sm1_cntc | sm2_cntc

# 請填入空格：
# mrgd_cntc = sm1_cntc ___ sm2_cntc
print("合併後電話簿數量:", len(mrgd_cntc))
# 預期輸出: 合併後電話簿數量: 3
```

```python
# 10.7.3 練習題：多組抽獎箱號碼彙整總管
# 題目說明：
# 現有三個抽獎箱投入的號碼串列 box1, box2, box3。
# 請將三個抽獎箱的號碼轉為集合並使用聯集 | 進行合併。
# 輸出全場一共投入了幾種「不同」的幸運號碼，以及所有號碼由小到大排序的結果。
#
# 【公開測試資料 1】
# 箱1: [10, 25, 30]
# 箱2: [25, 40]
# 箱3: [10, 50]
# 輸出:
# 相異幸運號碼數: 5
# 幸運號碼清單: [10, 25, 30, 40, 50]
#
# 【公開測試資料 2】
# 箱1: [7, 7, 7]
# 箱2: [7]
# 箱3: [77]
# 輸出:
# 相異幸運號碼數: 2
# 幸運號碼清單: [7, 77]

box1 = [10, 25, 30]
box2 = [25, 40]
box3 = [10, 50]

all_lucky = set(box1) | set(box2) | set(box3)

print("相異幸運號碼數:", len(all_lucky))
print("幸運號碼清單:", sorted(all_lucky))
```

```python
# 10.7.3 挑戰題：累計七日訪客動態擴充模擬
# 題目說明：
# 給定一週七天每天的訪客 ID 串列組成的二維串列 wk_vstr。
# 請使用一個集合 tot_vstr = set()，並在走訪每一天的串列時，
# 利用 |= 運算子將當天的訪客聯集加入。
# 最後印出整週造訪過的不同訪客總人數。

wk_vstr = [
    ["u1", "u2"],
    ["u2", "u3"],
    ["u1", "u4"],
    ["u5"],
    ["u3", "u5"],
    ["u6"],
    ["u2", "u7"]
]

# 請在下方寫出你的解答代碼
tot_vstr = set()
for day_list in wk_vstr:
    tot_vstr |= set(day_list)

print("整週相異訪客人數:", len(tot_vstr))
# 預期結果: 整週相異訪客人數: 7
```

## 10.7.4 差集運算子 - 與方法 difference()（批量黑名單過濾）

### 觀念說明
差集運算代表「屬於 A、但**絕對不屬於 B**」的元素。
語法結構：
1. 運算子：`A - B`
2. 方法：`A.difference(iterable)`

### ⚠️ 關鍵特性：差集具備方向性（非對稱性）！
請特別注意：加法或交集具有交換律（$A \& B == B \& A$），但**差集沒有交換律**！
- `A - B`：代表 $A$ 獨有的元素（從 $A$ 裡面把 $B$ 扣掉）。
- `B - A`：代表 $B$ 獨有的元素（從 $B$ 裡面把 $A$ 扣掉）。
這兩者通常完全不相等！

### APCS 經典解題情境：黑名單批量扣除
當我們有一份應到名單 `registered`，以及一份請假/違規名單 `absent`，只要一行 `registered - absent`，就能在常數雜湊時間內瞬間把違規者一掃而空，絕不留下任何漏網之魚！

```python
# 10.7.4 範例：黑名單過濾與實到人數計算
demo_enstu = {"Alice", "Bob", "Charlie", "David", "Eva"}
demo_astu = {"Bob", "David", "Nobody_99"}  # 即使有不存在的名單也無妨

# 1. 使用 - 差集運算子
demo_pstu = demo_enstu - demo_astu
print("今日實到學生名單:", sorted(demo_pstu))
print("實到人數:", len(demo_pstu))

# 2. 驗證差集的方向性
demo_set_x = {1, 2, 3}
demo_set_y = {3, 4, 5}
print()
print("X - Y (X 獨有):", demo_set_x - demo_set_y)  # {1, 2}
print("Y - X (Y 獨有):", demo_set_y - demo_set_x)  # {4, 5}
```

## 10.7.5 對稱差運算子 ^ 與方法 symmetric_difference()（獨家元素篩選）

### 觀念說明
對稱差代表「**只出現在其中一個集合，但不同時出現在兩個集合**」的元素。
在范氏圖中，它就是**把重疊的交集挖空，只保留左右兩邊月牙形的聯集**：
數學等價關係：
$$A \oplus B = (A - B) \cup (B - A) = (A \cup B) - (A \cap B)$$

### 語法結構
1. 運算子：`A ^ B`
2. 方法：`A.symmetric_difference(iterable)`

### APCS 解題應用
- **差異比對（Diff Checking）**：快速比對兩份備份檔案或資料表，找出有異動或不一致的項目。
- **獨特會員篩選**：挑選出只在一間分店消費過的忠誠專屬會員，排除兩邊跑的重複者。
- **奇數次出現檢測**：對稱差運算具備類似位元 XOR 的消除特性，相同者相消，留下的就是奇數次出現的元素！

```python
# 10.7.5 範例：兩次系統檔案清單差異比對 (Diff)
demo_v1fls = {"main.py", "utils.py", "config.json", "readme.md"}
demo_v2fls = {"main.py", "utils.py", "config.json", "styles.css", "logo.png"}

# 使用對稱差 ^ 找出所有「只在其中一個版本出現」的異動檔案
demo_dfls = demo_v1fls ^ demo_v2fls

print("有變動的相異檔案 (排除兩版本皆有的檔案):", sorted(demo_dfls))
# 說明: readme.md 是 v1 獨有，styles.css 與 logo.png 是 v2 獨有
```

```python
# 10.7.5 填空題：篩選只參加單一運動隊伍的選手
swimmers = {"Alex", "Brian", "Chris"}
runners = {"Brian", "David", "Chris"}

# 任務：使用對稱差運算子 ^，找出只參加游泳或只參加跑步的非雙棲選手
one_sport = swimmers ^ runners

# 請填入空格：
# one_sport = swimmers ___ runners
print("單項運動選手:", sorted(one_sport))
# 預期輸出: 單項運動選手: ['Alex', 'David']
```

```python
# 10.7.5 練習題：兩份名冊不一致名單檢測機
# 題目說明：
# 系統記錄之名單 sys_rcrd 與現場紙本簽到名單 ppr_rcrd。
# 請使用對稱差運算，找出「資訊不一致」（出現在其中一份名冊但不在另一份）的所有人員。
# 輸出不一致的人員數量，以及排序後的名單。
#
# 【公開測試資料 1】
# 系統記錄: ["A101", "A102", "A103"]
# 紙本記錄: ["A102", "A103", "A104"]
# 輸出:
# 不一致人數: 2
# 不一致名單: ['A101', 'A104']
#
# 【公開測試資料 2】
# 系統記錄: ["ID1", "ID2"]
# 紙本記錄: ["ID2", "ID1"]
# 輸出:
# 不一致人數: 0
# 不一致名單: []

sys_rcrd = ["A101", "A102", "A103"]
ppr_rcrd = ["A102", "A103", "A104"]

mismatch = set(sys_rcrd) ^ set(ppr_rcrd)

print("不一致人數:", len(mismatch))
print("不一致名單:", sorted(mismatch))
```

```python
# 10.7.5 挑戰題：驗證對稱差的兩種等價公式
# 題目說明：
# 給定 set = {1, 2, 3, 4, 5} 與 set_b107 = {4, 5, 6, 7}
# 請分別計算：
# 方式一：直接使用 set ^ set_b107
# 方式二：使用聯集扣除交集 (set | set_b107) - (set & set_b107)
# 請印出兩者的計算結果，並以 == 驗證兩者是否完全相等（印出 True）。

set = {1, 2, 3, 4, 5}
set_b107 = {4, 5, 6, 7}

# 請在下方寫出你的解答代碼
method1 = set ^ set_b107
method2 = (set | set_b107) - (set & set_b107)

print("方式一結果:", method1)
print("方式二結果:", method2)
print("兩者是否完全等價:", method1 == method2)
# 預期結果: 兩者是否完全等價: True
```

## 10.7.6 子集與超集判斷（<=, >=, issubset(), issuperset()）

### 觀念說明
除了產生新的集合外，在競賽中我們更常需要**檢驗集合之間的「從屬包含關係」**：

1. **子集判斷（Subset）`A <= B` 或 `A.issubset(B)`**：
   - 含義：集合 $A$ 的**所有元素是否都完全屬於 $B$**？
   - 如果是，回傳 `True`；只要 $A$ 裡面有一個元素不在 $B$ 中，就回傳 `False`。
2. **超集判斷（Superset）`A >= B` 或 `A.issuperset(B)`**：
   - 含義：集合 $A$ 是否**完整涵蓋了 $B$ 的所有成員**？
3. **真子集與真超集（Proper Subset / Superset）`A < B` 或 `A > B`**：
   - 要求必須包含，且兩者**不能完全相等**（$len(A) < len(B)$）。

### APCS 實戰情境
- **應聘資格審查**：應徵者的技能集合是否滿足職缺的「必備技能集合」（`required <= applicant_skills`）。
- **合成配方檢驗**：背包內擁有的材料是否涵蓋了魔法藥水的「所有配方道具」。

```python
# 10.7.6 範例：工程師職缺技能審查
demo_jrqrd = {"Python", "SQL", "Git"}

demo_aalc = {"Python", "Git", "SQL", "Docker", "Linux"}
demo_abob = {"Python", "Git", "Java"}  # 缺 SQL

# 檢查職缺要求是否是應徵者技能的「子集」
demo_aqlfd = demo_jrqrd <= demo_aalc
demo_bqlfd = demo_jrqrd <= demo_abob

print("Alice 是否合格 (技能涵蓋全部要求):", demo_aqlfd)
print("Bob 是否合格 (缺少 SQL):", demo_bqlfd)

# 若不合格，還能用差集一秒抓出缺少哪項技能！
demo_mskll = demo_jrqrd - demo_abob
print("Bob 欠缺的技能:", demo_mskll)
```

```python
# 10.7.6 填空題：藥水調配材料完整度檢查
recipe = {"herb", "water", "crystal"}
my_bckp = {"herb", "water", "crystal", "sword", "shield"}

# 任務：請使用 <= 運算子判斷 recipe 是否為 my_bckp 的子集
can_craft = recipe <= my_bckp

# 請填入空格：
# can_craft = recipe ___ my_bckp
print("是否具備全部材料可調配:", can_craft)
# 預期輸出: 是否具備全部材料可調配: True
```

```python
# 10.7.6 練習題：任務必備道具背包齊全檢驗器
# 題目說明：
# 解開地牢大門需要三把鑰匙 rqrd_kys = {"RED", "BLUE", "GOLD"}。
# 給定冒險者的背包清單 advn_bag。
# 請判斷冒險者是否集齊所有必備鑰匙（rqrd_kys 是否為背包集合的子集）：
# 若齊全，輸出 "大門開啟"；
# 若不齊全，輸出 "缺少鑰匙: <欠缺鑰匙清單>"（依字母排序）。
#
# 【公開測試資料 1】
# 背包: ["GOLD", "POTION", "BLUE", "RED", "SWORD"]
# 輸出: 大門開啟
#
# 【公開測試資料 2】
# 背包: ["GOLD", "POTION", "SILVER"]
# 輸出: 缺少鑰匙: ['BLUE', 'RED']

rqrd_kys = {"RED", "BLUE", "GOLD"}
advn_bag = ["GOLD", "POTION", "BLUE", "RED", "SWORD"]

bag_set = set(advn_bag)
if rqrd_kys <= bag_set:
    print("大門開啟")
else:
    missing = sorted(rqrd_kys - bag_set)
    print("缺少鑰匙:", missing)
```

```python
# 10.7.6 挑戰題：多位求職者資格符合統計
# 題目說明：
# 某項證照測驗合格必備項目為 standard = {"T1", "T2", "T3"}
# 現有 4 位考生的已通過項目紀錄：
# cndd_outs = [
#     ["T1", "T2", "T3", "T4"],
#     ["T1", "T3"],
#     ["T2", "T3", "T1"],
#     ["T4", "T5"]
# ]
# 請統計共有幾位考生「完全符合標準」（即 standard 是該考生已通過項目的子集）。

standard = {"T1", "T2", "T3"}
cndd_outs = [
# 請在下方寫出你的解答代碼
    ["T1", "T2", "T3", "T4"],
    ["T1", "T3"],
    ["T2", "T3", "T1"],
    ["T4", "T5"]
]

pss_cnt107 = 0
for cand in cndd_outs:
    if standard <= set(cand):
        pss_cnt107 += 1

print("符合標準考生總人數:", pss_cnt107)
# 預期結果: 符合標準考生總人數: 2
```
