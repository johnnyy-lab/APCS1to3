### 9.1.1 二維陣列物理模型：串列的串列（List of Lists）

在第八章中，我們學會了一維串列（`list`），它就像排成一長條的單排置物櫃。然而在現實世界中，許多資料並不是單純的一條線，例如：教室的座位表（有排有列）、棋盤地圖、井字遊戲的九宮格，或是學生的多科成績單（每個學生各有一排各科成績）。這種「有列（橫向）又有行（直向）」的二維資料，在 Python 中最直觀且強大的實作方式就是**二維陣列（List of Lists，串列的串列）**！

#### 什麼是「串列的串列」？
二維陣列本質上就是一個**大置物櫃（外層串列）**，裡面的每一個抽屜裝著**另一個小置物櫃（內層串列）**！
```python
grid = [
    [10, 20, 30],  # 第 0 橫列 (Row 0)
    [40, 50, 60],  # 第 1 橫列 (Row 1)
    [70, 80, 90]   # 第 2 橫列 (Row 2)
]
```

#### 概念解析：
1. **外層串列**：`grid` 本身是一個串列，它有 3 個元素：`grid[0]`、`grid[1]`、`grid[2]`。
2. **內層串列**：每個元素本身又是一個獨立的一維串列，代表**一整橫列（Row）**。
3. **視覺排版建議**：在程式碼中宣告二維陣列時，強烈建議每個內層串列獨立寫成一行，並對齊縮排，讓資料看起來就跟真實網格一模一樣！

````python
# [2] Code 範例：手動建立 2D 陣列與觀察物理結構

# 宣告一個 3 列 3 行的成績表 (3位學生，每人有國英數 3 科成績)
scores = [
    [85, 92, 78],  # 學生 0
    [90, 88, 95],  # 學生 1
    [70, 65, 80]   # 學生 2
]

print("完整二維陣列物件：", scores)
print("外層串列長度（學生人數/列數）：", len(scores))
print("第 0 位學生的完整成績單（第 0 列）：", scores[0])
print("第 0 位學生的型態：", type(scores[0]))
````

### 9.1.2 網格尺寸掌握：列數 `R = len(grid)` 與行數 `C = len(grid[0])`

在處理任何二維陣列問題時，最關鍵的第一步就是**正確取得網格的尺寸（幾列幾行）**！在 APCS 競賽與所有二維題目中，我們約定俗成用兩個大寫英文字母代表尺寸：
- **`R` (Rows)**：代表「列數」（橫向共有幾列、幾排）。
- **`C` (Columns)**：代表「行數」（直向共有幾行、幾欄）。

#### 如何用 Python 精準算出 R 與 C？
1. **列數 `R`**：因為外層串列包含一個個內層串列，因此計算外層串列的長度，就是總列數：
   ```python
   R = len(grid)
   ```
2. **行數 `C`**：在標準矩形網格中，每一列的長度都是一樣的。我們只要拿出「第 0 列」算它的長度，就是總行數：
   ```python
   C = len(grid[0])
   ```

#### 避坑安全守則：
- **嚴禁混淆 `len(grid)` 與 `len(grid[0])`**：很多初學者會寫成 `C = len(grid)`，這在非方陣（如 2 列 5 行）的矩陣中會引發嚴重的維度錯亂！請牢記：`len(grid)` 是由上到下數有幾列，`len(grid[0])` 才是由左到右數一列有幾格。

````python
# [2] Code 範例：量測不同矩陣尺寸

# 宣告一個 2 列 4 行的矩形陣列
table = [
    [10, 20, 30, 40],
    [50, 60, 70, 80]
]

R = len(table)
C = len(table[0])

print(f"網格尺寸：{R} 列 x {C} 行")
print(f"總元素個數：{R * C} 個")
print("第一列元素個數：", len(table[0]))
print("第二列元素個數：", len(table[1]))
````

### 9.1.3 整列提取 vs 單格存取：`grid[r]` 與 `grid[r][c]` 的型態與運算差異

在進入座標細節之前，我們必須先釐清一個無數初學者都會踩中的型態大坑：**「只寫一個中括號 `grid[r]`，跟寫兩個中括號 `grid[r][c]`，拿出來的東西到底差在哪裡？」**

#### 型態本質剖析：
1. **`grid[r]` 拿出來的是「一整橫列」**：
   - 資料型態是 **`list`（一維串列）**！
   - 它是一個包含多個元素的「容器」，因此你可以直接對它使用第八章學過的所有串列技能：
     - `len(grid[r])`：這列有幾個元素。
     - `sum(grid[r])`：這列的所有整數總和！
     - `max(grid[r])`：這列的最大值！
     - `"Alice" in grid[r]`：檢查某元素是否在這一列中。
2. **`grid[r][c]` 拿出來的是「單一元素」**：
   - 資料型態是 **`int`、`str` 或 `float`（基本型態）**！
   - 它是抽屜裡面的具體物品，不能再對它呼叫 `sum()` 或 `len()`（除非它是字串）。

#### 初學者致命報錯對照：
```python
# 想要把第 0 列的數值加 10：
grid[0] + 10      # ❌ 崩潰！TypeError: can only concatenate list to list (不能拿串列直接加整數)
grid[0][0] + 10   # ✅ 正確！拿到了具體整數，可以進行數學加法！
```

````python
# [2] Code 範例：整列提取與一維技能複用

scores = [
    [90, 80, 70],  # 學生 0 的成績單
    [100, 95, 90], # 學生 1 的成績單
    [60, 50, 75]   # 學生 2 的成績單
]

# 提取一整列 (型態為 list)
student_1 = scores[1]
print("學生 1 的整張成績單：", student_1, "型態：", type(student_1))
print("學生 1 的總分 sum(scores[1])：", sum(scores[1]))
print("學生 1 的最高分 max(scores[1])：", max(scores[1]))

# 提取單一元素 (型態為 int)
first_subject = scores[1][0]
print("學生 1 的第 0 科分數：", first_subject, "型態：", type(first_subject))
print("加 5 分後：", first_subject + 5)
````

### 9.1.4 座標定位存取：`grid[r][c]` 先列後行絕對守則與九宮格打靶

二維網格中存取任何特定格子的元素，使用的是**雙重中括號語法：`grid[r][c]`**。

#### 🌟 最核心的黃金鐵律：先列後行（Row-First）
在數學座標中，我們常習慣講 $(x, y)$（先水平後垂直）；**但在所有程式語言的二維陣列中，順序恰恰完全相反：一律是「先垂直列 $r$，後水平行 $c$」！**
- 第一個中括號 `[r]`：決定你走進**「第幾橫列」**（Row，由上往下算，編號 $0$ 到 $R-1$）。
- 第二個中括號 `[c]`：決定你在該列中打開**「第幾直行」**（Column，由左往右算，編號 $0$ 到 $C-1$）。

```
          c=0   c=1   c=2   (水平行 Column)
       +-----+-----+-----+
 r = 0 | 0,0 | 0,1 | 0,2 |  (第 0 橫列)
       +-----+-----+-----+
 r = 1 | 1,0 | 1,1 | 1,2 |  (第 1 橫列)
       +-----+-----+-----+
 r = 2 | 2,0 | 2,1 | 2,2 |  (第 2 橫列)
       +-----+-----+-----+
(垂直列 Row)
```

#### 九宮格關鍵定位標的：
- **中心點**：`grid[1][1]`（正中間座標）
- **左上角**：`grid[0][0]`
- **右上角**：`grid[0][C-1]`
- **左下角**：`grid[R-1][0]`
- **右下角**：`grid[R-1][C-1]`

````python
# [2] Code 範例：九宮格精準打靶定位

grid = [
    [11, 12, 13],
    [21, 22, 23],
    [31, 32, 33]
]

print("正中心 (1, 1)：", grid[1][1])
print("左上角 (0, 0)：", grid[0][0])
print("右上角 (0, 2)：", grid[0][2])
print("左下角 (2, 0)：", grid[2][0])
print("右下角 (2, 2)：", grid[2][2])
````

### 9.1.5 網格元素原地修改：`grid[r][c] = 新數值`

在第一章我們學過變數賦值，在第八章學過串列切片賦值。因為串列是**可變物件（Mutable）**，因此二維陣列中任何一個位置的值，都可以直接透過等號進行**原地修改（In-place Update）**！

#### 賦值語法公式：
```python
grid[r][c] = 新數值
```
運作流程是：
1. 直譯器先找到外層的第 `r` 個內層串列（抽屜）。
2. 接著定位到該串列的第 `c` 個隔間。
3. 把舊數值扔掉，換成等號右側的新數值！

#### 實戰應用情境：
- **遊戲地圖更新**：角色走到 `(r, c)` 時，將原本的空地 `'.'` 改為玩家 `'P'`。
- **演算法拜訪標記**：搜尋走訪時，將已造訪過的格子標記為 `-1` 或 `True`。
- **動態計分**：某位同學補考成績出來，直接覆蓋原本的舊分數 `scores[r][c] = 95`。

### 9.1.6 負向索引定位與維度混淆陷阱防禦

Python 最具特色的語法之一是**支援負向索引（Negative Indexing）**。這在二維陣列中同樣完全適用且極具威力，但同時也伴隨著初學者最容易忽視的隱形陷阱！

#### 負向索引的妙用：
- `grid[-1]`：直接拿到**「最後一橫列（最底下一列）」**。
- `grid[-1][-1]`：直接拿到**「右下角最後一個元素」**！
- `grid[0][-1]`：右上角最後一個元素。
- `grid[-1][0]`：左下角第一個元素。

#### 兩大隱形陷阱防禦：
1. **越界錯誤（IndexError）**：如果網格只有 3 列，你寫 `grid[3][0]` 就會直接崩潰報錯 `IndexError: list index out of range`。請記住合法的正索引永遠只有 $0 \sim R-1$ 與 $0 \sim C-1$。
2. **負數「假安全」隱性 Bug（極重要）**：
   在後續的尋路或鄰居探測演算法中，如果座標算出來是 `-1`（例如本來在第 0 列，往上走變成 `-1` 列），Python **不會報錯**！它會默默讀取「最底下一列」的資料！這會導致你的角色從地圖頂端「瞬間瞬間移動」到地圖底端，引發極度隱蔽的邏輯錯誤！因此在邊界判斷時，必須嚴格守護 `r >= 0 and c >= 0`。

````python
# [2] Code 範例：負向索引定位

matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

print("最底下一整列 matrix[-1]：", matrix[-1])
print("右下角元素 matrix[-1][-1]：", matrix[-1][-1])
print("右上角元素 matrix[0][-1]：", matrix[0][-1])
print("左下角元素 matrix[-1][0]：", matrix[-1][0])
````
