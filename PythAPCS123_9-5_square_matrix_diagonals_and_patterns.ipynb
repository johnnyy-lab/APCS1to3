### 9.5.1 主對角線走訪規律（$r == c$）

在二維串列中，當矩陣的列數等於行數時，我們稱之為「**方陣（Square Matrix）**」（大小為 $N \times N$）。方陣具有許多獨特且對稱的幾何特性，其中最基本的就是「**主對角線（Main Diagonal）**」。

主對角線是指從矩陣的「左上角」一路延伸到「右下角」的那條對角線。讓我們觀察一個 $3 \times 3$ 方陣的主對角線座標：
- 第一個位置：`(0, 0)`
- 第二個位置：`(1, 1)`
- 第三個位置：`(2, 2)`

發現了嗎？主對角線上的每一個元素，其**列索引（Row Index）永遠等於行索引（Column Index）**，即數學上的 $r = c$！

許多初學者在走訪主對角線時，直覺會寫出雙重迴圈：
```python
# 雖然正確，但多做了許多無效運算的寫法：
for r in range(N):
    for c in range(N):
        if r == c:
            total += grid[r][c]
```
這樣的寫法一共檢查了 $N \times N$ 次條件，時間複雜度是 $O(N^2)$。但因為我們已經明確知道「列等於行」，因此只要使用一個走訪變數 $i$（從 $0$ 到 $N-1$），直接存取 `grid[i][i]`，就能以**單層迴圈**完成任務！時間複雜度直接降為 $O(N)$，這是在 APCS 考試中極為重要的時間優化思維。


````python
# 範例 9.5.1：使用單層迴圈走訪 3x3 方陣的主對角線並計算總和

matrix = [
    [10,  2,  3],
    [ 4, 20,  6],
    [ 7,  8, 30]
]

N = len(matrix)
diag_elements = []
diag_sum = 0

# 單層迴圈走訪主對角線：列索引與行索引都是 i
for i in range(N):
    val = matrix[i][i]
    diag_elements.append(val)
    diag_sum += val

print("矩陣維度：", N, "x", N)
print("主對角線元素列表：", diag_elements)
print("主對角線元素總和：", diag_sum)
````

### 9.5.2 副對角線走訪規律（$r + c == N - 1$）

了解了主對角線（左上到右下）後，接下來我們要認識方陣中另一條重要的軸線——「**副對角線（Anti-Diagonal）**」（也有人稱為次對角線或逆對角線）。

副對角線是從矩陣的「右上角」一路延伸到「左下角」的連線。讓我們同樣以 $3 \times 3$ 方陣（$N=3$）為例，觀察副對角線經過的座標：
- 第一個位置（右上）：`(0, 2)`
- 第二個位置（正中）：`(1, 1)`
- 第三個位置（左下）：`(2, 0)`

若是 $4 \times 4$ 方陣（$N=4$）：
- 座標分別為：`(0, 3)`, `(1, 2)`, `(2, 1)`, `(3, 0)`

請觀察這組座標的數學規律：
1. 列索引 $r$ 從 $0$ 逐步遞增到 $N-1$。
2. 行索引 $c$ 從 $N-1$ 逐步遞減到 $0$。
3. 最核心的代數關係是：**$r + c = N - 1$**！
換言之，只要我們知道列索引 $r$（或設迴圈變數為 $i$），那麼行索引 $c$ 就一定是 **$N - 1 - i$**。

因此，走訪副對角線同樣**完全不需要雙重迴圈**！我們只需利用單層迴圈，每次存取 `grid[i][N - 1 - i]`，即可在 $O(N)$ 的時間內走訪完整條副對角線。


````python
# 範例 9.5.2：走訪 3x3 方陣的副對角線並列印元素與座標

matrix = [
    [10, 20, 99],
    [40, 88, 60],
    [77, 80, 90]
]

N = len(matrix)
print("方陣階數 N =", N)

# 單層迴圈走訪副對角線：列為 i，行為 N - 1 - i
for i in range(N):
    c = N - 1 - i
    val = matrix[i][c]
    print(f"列 {i}, 行 {c} -> 元素值 = {val}")
````

### 9.5.3 X 形交叉雙對角線與中心點重複防禦

當我們將「主對角線」與「副對角線」放在同一張地圖上看時，它們交織成了一個醒目的「**X 形狀**」。在許多幾何演算法、棋盤遊戲（如五子棋、圍棋視線）或統計題中，常需要計算兩條對角線上所有數字的總和或聯集。

這時存在一個非常隱蔽但致命的邏輯陷阱——**中心點重複計算（Center Overlap）**！

讓我們比較兩種維度的方陣：
1. **偶數階方陣（$N$ 是偶數，如 $4 \times 4$）**：
   - 主對角線座標：`(0,0), (1,1), (2,2), (3,3)`
   - 副對角線座標：`(0,3), (1,2), (2,1), (3,0)`
   - 兩組座標完全互斥，彼此沒有交集。
2. **奇數階方陣（$N$ 是奇數，如 $3 \times 3$ 或 $5 \times 5$）**：
   - 以 $3 \times 3$ 為例：
     - 主對角線：`(0,0), (1,1), (2,2)`
     - 副對角線：`(0,2), (1,1), (2,0)`
   - 注意到了嗎？中央點 `(1, 1)`（即 `(N//2, N//2)`）**同時出現在兩條對角線上**！

如果我們分別計算主對角線和副對角線的總和並相加：
```python
total = main_sum + anti_sum  # 陷阱！中心點被加了兩次！
```
當 $N$ 是奇數時，中央點會被計算兩次，導致結果產生偏差。因此，嚴謹的程式設計師會做**重複防禦（Deduplication Defense）**：
- 寫法 A：先各自累加，若 `N % 2 != 0`，則最後減去重複的中心點 `grid[N // 2][N // 2]`。
- 寫法 B：在單層迴圈中同時走訪，當 `i != N - 1 - i` 時才將副對角線的值加入。


````python
# 範例 9.5.3：計算 3x3 方陣 X 形對角線所有不重複元素的總和

matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

N = len(matrix)
x_sum = 0

for i in range(N):
    # 加上主對角線元素
    x_sum += matrix[i][i]
    
    # 加上副對角線元素，但如果此時列索引等於副對角線行索引（代表正是中心點），則不重複加！
    anti_c = N - 1 - i
    if i != anti_c:
        x_sum += matrix[i][anti_c]

print("矩陣階數：", N)
print("X 形對角線不重複元素總和：", x_sum)
# 主對角線：1 + 5 + 9 = 15
# 副對角線非中心：3 + 7 = 10
# 正確總和：15 + 10 = 25 (若沒防禦重複會算成 30)
````

### 9.5.4 上三角與下三角矩陣走訪

在方陣中，除了單純的一條線（對角線），我們常常需要處理被對角線切開的「半邊區域」，這在數學與電腦科學中稱為**三角矩陣（Triangular Matrix）**：
1. **上三角區域（Upper Triangular）**：
   - 包含主對角線及其右上方所有的格子。
   - 數學特徵：**$r \le c$**（行座標大於或等於列座標）。
   - 若不包含對角線，則稱為「嚴格上三角（Strictly Upper Triangular）」，條件為 **$r < c$**。
2. **下三角區域（Lower Triangular）**：
   - 包含主對角線及其左下方所有的格子。
   - 數學特徵：**$r \ge c$**（列座標大於或等於行座標）。
   - 若不包含對角線，則稱為「嚴格下三角（Strictly Lower Triangular）」，條件為 **$r > c$**。

在撰寫迴圈走訪三角區域時，最優雅的做法不是「走訪全矩陣再用 `if` 過濾」，而是**直接動態調整內層迴圈的範圍**！
- 走訪上三角（含對角線）：
  列從 $0$ 到 $N-1$；當列為 $r$ 時，行只要從 $r$ 走到 $N-1$ 即可：
  ```python
  for r in range(N):
      for c in range(r, N):  # 內層起點直接綁定外層 r！
          # 存取 grid[r][c]
  ```
- 走訪下三角（含對角線）：
  列從 $0$ 到 $N-1$；當列為 $r$ 時，行只要從 $0$ 走到 $r$ 即可：
  ```python
  for r in range(N):
      for c in range(r + 1):  # 內層只走到 r (包含 r)
          # 存取 grid[r][c]
  ```
透過動態範圍控制，程式少走訪了一半的格子，執行效率立即提升一倍！

### 9.5.5 黑白棋盤格座標規律（$(r + c) \% 2$）

在西洋棋或圍棋盤中，相鄰的格子顏色永遠不同——黑格周圍是白格，白格周圍是黑格，交錯分佈。這種在二維陣列中交替出現的圖形規律，稱為「**棋盤染色模型（Checkerboard Coloring）**」。

棋盤染色背後有一個非常優雅且強大的數學公理：
若我們將列座標 $r$ 與行座標 $c$ 相加，觀察其奇偶性：
- `(0, 0)`：$0 + 0 = 0$（偶數）
- `(0, 1)`：$0 + 1 = 1$（奇數）
- `(0, 2)`：$0 + 2 = 2$（偶數）
- `(1, 0)`：$1 + 0 = 1$（奇數）
- `(1, 1)`：$1 + 1 = 2$（偶數）
- `(1, 2)`：$1 + 2 = 3$（奇數）

**結論**：
任意格子的座標 $(r, c)$，只要計算 **$(r + c) \% 2$**：
- 如果等於 $0$，代表該格子屬於「**偶數格（通常設為白格）**」。
- 如果等於 $1$，代表該格子屬於「**奇數格（通常設為黑格）**」。

只要往上下左右走一步（$r$ 或 $c$ 變化 $1$），$(r + c)$ 的奇偶性必然反轉！因此，相鄰兩格的 $(r+c) \% 2$ 絕對不相等。這個規律在 APCS 中經常用於網格染色、二維地圖交替障礙物、跳棋走法判定等高頻考題中。


````python
# 範例 9.5.5：產生一個 4x4 的黑白交錯棋盤矩陣（0 表示白格，1 表示黑格）

R, C = 4, 4
board = []

for r in range(R):
    row = []
    for c in range(C):
        # 依據 (r + c) % 2 決定填入 0 或 1
        if (r + c) % 2 == 0:
            row.append(0)
        else:
            row.append(1)
    board.append(row)

print("4x4 棋盤矩陣：")
for row in board:
    print(*row)
````

### 9.5.6 方陣對稱性檢驗（$grid[r][c] == grid[c][r]$）

在數學與圖論中，「**對稱矩陣（Symmetric Matrix）**」是一種非常基礎且重要的結構（例如無向圖的相鄰矩陣必為對稱矩陣）。

什麼是對稱矩陣？
如果一個方陣沿著「主對角線」對折翻轉後，重疊位置的數值完全相同，那麼這個方陣就是對稱矩陣。
以代數符號表示：對於所有有效的列 $r$ 與行 $c$，都必須嚴格滿足：
$$grid[r][c] == grid[c][r]$$

在檢查對稱性時，有兩個關鍵的實作考量：
1. **比較範圍優化**：
   - 如果我們對所有 $(r, c)$ 都檢查一次，那麼 `(0, 1)` 和 `(1, 0)` 會被檢查兩次，且對角線上的 `grid[i][i] == grid[i][i]` 是毫無意義的自我比較。
   - 正確的高效做法：**只走訪「嚴格上三角區域」（$r < c$）**！只要所有的嚴格上三角元素都等於其對應的下三角元素，即可確認整個矩陣對稱。
2. **提早結束（Early Exit）與狀態旗標**：
   - 一旦在比對過程中發現任何一對 `grid[r][c] != grid[c][r]`，我們就能斷定該矩陣「絕對不對稱」，應立即將旗標變數設為 `False` 並以 `break` 跳出迴圈，不再浪費時間檢查後續格子。


````python
# 範例 9.5.6：使用旗標變數檢驗 3x3 方陣是否為對稱矩陣

matrix = [
    [1, 7, 3],
    [7, 4, 8],
    [3, 8, 9]
]

N = len(matrix)
is_symmetric = True

# 只檢查嚴格上三角區域 (r < c)
for r in range(N):
    for c in range(r + 1, N):
        if matrix[r][c] != matrix[c][r]:
            is_symmetric = False
            break
    if not is_symmetric:
        break

print("方陣階數：", N)
if is_symmetric:
    print("檢驗結果：這是一個對稱矩陣！")
else:
    print("檢驗結果：這不是對稱矩陣！")
````

## 本單元重點回顧與核心心法

在單元 9-5 中，我們深入探索了 $N \times N$ 方陣的幾何結構與座標代數規律：

1. **主對角線（Main Diagonal）**：
   - 座標規律：$r == c$。
   - 走訪心法：利用單層迴圈 `grid[i][i]`，時間複雜度為 $O(N)$，無需雙重迴圈。
2. **副對角線（Anti-Diagonal）**：
   - 座標規律：$r + c == N - 1$，故行座標為 $N - 1 - r$。
   - 走訪心法：單層迴圈 `grid[i][N - 1 - i]`，同樣只需 $O(N)$。
3. **X 形交叉雙對角線與重複防禦**：
   - 當 $N$ 為奇數時，中央元素 `(N // 2, N // 2)` 同時落在兩條對角線上。
   - 若做累加或聯集運算，務必進行「去重防禦」（判斷 $i \ne N - 1 - i$ 或於最後減去一次中心值）。
4. **上三角與下三角動態範圍控制**：
   - 上三角（$r \le c$）：外層 $r \in [0, N)$，內層 $c \in [r, N)$。
   - 下三角（$r \ge c$）：外層 $r \in [0, N)$，內層 $c \in [0, r + 1)$。
   - 嚴格三角（$<$ 或 $>$）只需微調起點或終點偏移量。
5. **棋盤格染色規律（$(r + c) \% 2$）**：
   - 相鄰格子和的奇偶性必然交替，偶數格與奇數格形成天然黑白交錯圖樣。
6. **方陣對稱性檢驗（$grid[r][c] == grid[c][r]$）**：
   - 只需走訪嚴格上三角（$r < c$），一旦失配立即 `break` 結束，展現最佳時間效率。
