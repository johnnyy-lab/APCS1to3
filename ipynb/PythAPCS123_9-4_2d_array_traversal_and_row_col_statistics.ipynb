### 9.4.1 雙重迴圈時鐘模型：外層列 $r$ 與內層行 $c$ 的時序與座標對生成

在處理二維陣列時，我們最常需要做的事情就是**「把地圖上的每一個格子都檢查一遍」**。要逛完二維世界中的每一個角落，單靠一層迴圈是不夠的，我們必須派出**雙重迴圈（Nested Loops）**！

#### 🌟 時鐘模型（Clockwork Analogy）：
許多初學者在寫雙重迴圈時會感到頭暈，不知道電腦現在到底走到哪一格。其實，雙重迴圈的運作機制就跟**「手錶上的時針與分針」**一模一樣：
- **外層迴圈（時針，慢）**：代表**「列索引 $r$」**。每當 $r$ 推進 1 格，外層就停在這一列。
- **內層迴圈（分針，快）**：代表**「行索引 $c$」**。在當前的第 $r$ 列中，分針必須一口氣從 $0$ 狂奔到 $C-1$，把該列的每一格全部巡訪完畢！
- 當內層分針跑完一整圈後，外層時針才會前進到下一列（$r+1$），分針又再次從頭跑一整圈！

```
當 r = 0 時：跑完 (0, 0) -> (0, 1) -> (0, 2)
當 r = 1 時：跑完 (1, 0) -> (1, 1) -> (1, 2)
當 r = 2 時：跑完 (2, 0) -> (2, 1) -> (2, 2)
```
牢記「先列後行、外慢內快」，這就是二維演算法走訪的最底層物理時序！

````python
# [2] Code 範例：雙重迴圈座標軌跡視覺化

demo_R, demo_C = 2, 3
demo_gd094 = [
    [10, 20, 30],
    [40, 50, 60]
]

print("開始時鐘模型雙重走訪：")
for demo_r in range(demo_R):  # 外層時針：列
    print(f"--- 外層進入第 r = {demo_r} 列 ---")
    for demo_c in range(demo_C):  # 內層分針：行
        demo_val = demo_gd094[demo_r][demo_c]
        print(f"  座標 ({demo_r}, {demo_c}) 數值為: {demo_val}")
````

### 9.4.2 橫列直接走訪：`for row in grid:` 搭配 `sum(row)` 極速列統計

如果我們的任務**不需要知道具體的座標 $(r, c)$**，只是單純想對「每一橫列」進行統計（例如：算出每一排學生的成績總和、每間分店的營業額），那麼完全不需要寫繁瑣的 `range(R)`！

#### 橫列直接走訪語法：
二維陣列的外層元素本身就是一個個一維串列，因此我們可以直接寫：
```python
for row in grid:
    total = sum(row)      # 直接將這一橫列加總！
    highest = max(row)    # 直接找出這一橫列的最大值！
```

#### 為什麼這個語法極受歡迎？
1. **極度直觀**：`for row in grid` 在英文中讀起來就像「對於網格中的每一列」。
2. **程式碼極簡**：直接呼叫第八章學過的 `sum()`、`max()`、`min()`，一行搞定一列的統計！
3. **絕不出界**：不用維護索引變數，完全杜絕 `IndexError` 的風險。

````python
# [2] Code 範例：學生成績單橫列直接統計

# 三位學生的三科成績
demo_sc094 = [
    [80, 90, 85],  # 學生 0
    [95, 100, 90], # 學生 1
    [60, 70, 65]   # 學生 2
]

demo_rsms = []
for demo_row in demo_sc094:
    demo_s = sum(demo_row)
    demo_rsms.append(demo_s)
    print(f"橫列成績單: {demo_row} ➔ 總分: {demo_s}, 最高分: {max(demo_row)}")

print("\n每位學生總分清單：", demo_rsms)
````

### 9.4.3 橫列索引走訪：`for r in range(R):` 的帶編號列統計與報表產出

在上一小節中，我們學到了 `for row in grid:` 可以快速算出一列的總和。但如果今天題目的輸出要求是：
```
第 0 號學生總分: 255
第 1 號學生總分: 285
第 2 號學生總分: 195
```
這時直接走訪就有些不夠用了，因為我們**需要知道當前是「第幾列」（編號 $r$）**！這就是為什麼我們必須熟練**「橫列索引走訪」**。

#### 語法架構：
```python
for r in range(R):
    row_sum = sum(grid[r])
    print(f"第 {r} 號學生總分: {row_sum}")
```

#### 核心優勢與適用時機：
1. **精準帶出列編號**：能產出附帶 `r` 標籤的工整統計報表。
2. **就地修改整列資料**：如果需要更新該列（例如 `grid[r] = new_data`），只有持有索引 `r` 才能完成就地覆蓋。
3. **跨列對比運算**：若需要比對相鄰兩列（例如第 $r$ 列與第 $r+1$ 列），索引走訪是唯一解法！

````python
# [2] Code 範例：帶編號橫列統計報表

demo_sc11 = [
    [80, 90, 85],
    [95, 100, 90],
    [60, 70, 65]
]

demo_R094 = len(demo_sc11)
for demo_r in range(demo_R094):
    demo_total = sum(demo_sc11[demo_r])
    demo_avg = demo_total // len(demo_sc11[demo_r])  # 整數平均
    print(f"學生編號 [{demo_r}] -> 總分: {demo_total:3d} 分, 平均: {demo_avg:2d} 分")
````

### 9.4.4 直行（直欄）維度翻轉統計：外層 $c$ 內層 $r$ 的縱向走訪累加

現在，我們來到二維陣列中最關鍵、也最考驗空間思維的轉折點：**「直行（Column）統計」**！

#### 初學者的重大困惑：為什麼不能寫 `sum(col)`？
在橫列統計中，一整列本身就是一個獨立的一維串列，所以能直接 `sum(grid[r])`。
**但在 Python 的二維陣列中，記憶體裡根本沒有「直行物件」！**
第 0 直行是由 `grid[0][0]`, `grid[1][0]`, `grid[2][0]` 分散在各個不同橫列中的第 0 格所組成的。電腦無法憑空幫你把直行打包成串列！

#### 唯一解法：維度翻轉（外層行 $c$，內層列 $r$）
為了計算某一直行的總和，我們必須**顛倒迴圈的先後順序**：
1. **外層固定行號 $c$**：先站在第 $c$ 欄（`for c in range(C):`）。
2. **準備累加器**：`col_sum = 0`。
3. **內層由上往下跑列號 $r$**：`for r in range(R): col_sum += grid[r][c]`！

```
固定 c = 0：縱向依序吃掉 (0, 0) -> (1, 0) -> (2, 0)
固定 c = 1：縱向依序吃掉 (0, 1) -> (1, 1) -> (2, 1)
固定 c = 2：縱向依序吃掉 (0, 2) -> (1, 2) -> (2, 2)
```
透過外層 $c$、內層 $r$ 的翻轉，我們就能輕鬆計算出「各科目的全班總分」或「各直行的統計值」！

````python
# [2] Code 範例：直行固定縱向累加統計

# 3 位學生、3 個科目 (國文、英文、數學)
demo_sc12 = [
    [80, 90, 70],  # 學生 0
    [85, 95, 75],  # 學生 1
    [90, 85, 80]   # 學生 2
]

demo_R4 = len(demo_sc12)     # 學生人數: 3
demo_C094 = len(demo_sc12[0])  # 科目數量: 3

demo_ssms = []
# 外層固定科目欄位 c
for demo_c in range(demo_C094):
    demo_col_total = 0
    # 內層縱向走訪每位學生 r
    for demo_r in range(demo_R4):
        demo_col_total += demo_sc12[demo_r][demo_c]
    demo_ssms.append(demo_col_total)
    print(f"科目 {demo_c} (全班總分): {demo_col_total}, 平均: {demo_col_total // demo_R4}")

print("\n全科目總分清單：", demo_ssms)
````

### 9.4.5 全網格全域極值搜尋與座標定位鎖定

在 APCS 與各類演算法問題中（例如：APCS `e287. 機器人的路徑` 要求機器人從「全地圖最小值」出發），我們經常需要**在整張二維地圖中搜尋最高峰或最低谷**，並且在找到極值的同時，**精準記下它所在的座標 $(r, c)$**！

#### 極值搜尋黃金三步法：
1. **設定基準（初始極值）**：
   - 千萬不要隨便設 `best_val = 0`，因為地圖可能是負數！
   - **最安全的初值**：直接拿地圖的第一個格子當作基準：
     ```python
     best_val = grid[0][0]
     best_r, best_c = 0, 0
     ```
2. **雙重迴圈地毯式走訪**：巡訪網格中的每一個座標 $(r, c)$。
3. **比對並連帶更新座標**：
   ```python
   if grid[r][c] > best_val:
       best_val = grid[r][c]  # 更新冠軍數值
       best_r = r            # 同步鎖定冠軍列號
       best_c = c            # 同步鎖定冠軍行號
   ```
走訪結束後，你不僅知道全地圖的極值是多少，還能毫無偏差地拿到它的精準發射座標！
