### 🔗 7-2-1 字串加法運算子（+）：串接與文字拼接

**生活比喻：磁鐵車廂的掛鉤與拼字樂高**  
想像你手上有兩節玩具火車車廂，第一節車廂上面印著「Hello」，第二節車廂上面印著「World」。當你把兩節車廂推在一起時，它們頭尾的磁鐵掛鉤「啪嗒」一聲牢牢吸附在一起，組裝成了一列更長、更完整的列車。  
在 Python 中，加號 `+` 放在兩個數字中間時是數學加法（如 `3 + 5` 得到 `8`）；但當加號出現在**兩個字串之間**時，它會搖身一變成為「磁鐵掛鉤」，專門負責將左邊字串與右邊字串緊緊黏合在一起，這個動作在電腦科學中被稱為 **字串串接（String Concatenation）**。

**底層運作機制與型態嚴格性**  
當直譯器執行 `"Python" + " Rock"` 時，它並不會修改原本的任何一個字串，而是在記憶體中開闢一塊新的空間，將兩個字串的所有字元依序複製進去，生成一個全新字串 `"Python Rock"`。  

**常見錯誤陷阱：字串不能直接與數值相加（TypeError）**  
初學者最常踩中的地雷，就是誤以為加號可以通吃一切型態：  
`score = 95`  
`message = "Your score is: " + score` ❌  
這行程式碼會立刻引發嚴厲的型態錯誤：  
**`TypeError: can only concatenate str (not "int") to str`**  
電腦非常困惑：「你到底是想要把 95 轉成文字黏在後面，還是想要把文字算成數學數字？」  
解決方案非常簡單，請使用在第三章學過的型態轉換函數 **`str()`**，先將整數轉為字串再相加：  
`message = "Your score is: " + str(score)` ✅  

**APCS 考試實務建議：**  
在 APCS 實作題中，許多題目要求輸出的答案是字串與數值的混合組合（例如格式 `Case #1: 42`）。雖然可以用 `print("Case #", i, ": ", ans, sep="")` 輸出，但在字串內部直接利用 `+` 與 `str()` 預先拼好完整字串再一併輸出，能讓程式邏輯更加簡潔清晰。

````python
# [2] Code 範例區：展示字串加法運算子（+）與型態轉換拼接
# 說明：使用 + 將字串無縫串接，並示範將數字以 str() 轉型後安全拼接。

# 範例 1：兩個純字串的直接拼接
first_word = "Magic"
second_word = "Stone"
combined = first_word + " " + second_word
print("純字串拼接結果:", combined)

# 範例 2：字串與數值變數的合法拼接（搭配 str() 轉型）
player = "Hero"
level = 25
info = player + " (Lv." + str(level) + ")"
print("玩家資訊卡:", info)
````

### 🧲 7-2-2 多變數與連續串接：累加與複合賦值（+=）

**生活比喻：滾雪球與穿珍珠手鍊**  
想像冬天在雪地上堆雪人，你一開始先捏好一個小雪球放在手上。接著，你推著雪球在雪地上往前滾動，沿途每碰到一點新雪，雪球的體積就會多增加一層，越滾越長、越滾越大。  
在編程中，當我們需要在迴圈或多個步驟中「一點一滴收集文字」時，這個過程就是字串累加！  
在第一章我們學過數值累加運算子 `total += x`；令人驚喜的是，這個強大的 **複合賦值運算子 `+=`** 在字串世界中完全通用！

**底層運作機制與不可變特性（Immutable）**  
當我們寫下 `sentence += "word"` 時，底層其實等價於：  
`sentence = sentence + "word"`  
特別需要向初學同學強調的是：**Python 中的字串是「不可變的（Immutable）」**。這代表直譯器並不是把新字元直接塞進原本的舊字串裡，而是將舊字串的內容與新文字在記憶體中重新組裝成一個「新字串」，再將變數名稱重新貼標籤指向這個新字串。

**常見錯誤陷阱：未初始化的字串變數無法進行 `+=` 累加**  
看看這段初學者常犯的錯誤：  
`# 忘記先宣告 result`  
`result += "Hello"` ❌  
這會瞬間引發：  
**`NameError: name 'result' is not defined`**  
電腦不知道一開始的雪球長什麼樣子，無從累加起！  
因此，字串累加的鐵律是：**在開始累加前，務必先以空字串 `""`（長度為 0 的字串）進行歸零初始化**：  
`result = ""`  
`result += "Hello"` ✅  

**APCS 考試實務建議：**  
在字串處理題目中，我們常常需要根據條件過濾字元（例如只留下大寫字母，或排除特定標點符號）。建立一個空的累加變數 `ans = ""`，在走訪時符合條件就 `ans += ch`，是 APCS 解題中最基礎且核心的標準解題模板之一。

````python
# [2] Code 範例區：展示字串的複合賦值累加運算子（+=）
# 說明：從空字串 "" 出發，逐步收集並組裝文字。

# 步驟 1：初始化空字串
story = ""

# 步驟 2：逐步追加情節文字
story += "Once upon a time, "
story += "a brave coder "
story += "conquered Python!"

print("累積組裝後的完整故事:")
print(story)
````

### ⚡ 7-2-3 字串乘法運算子（*）：整數倍率重複造字

**生活比喻：影印機的連印按鈕與魔法分身術**  
想像你走進一間影印店，老闆有一張印著「★」圖樣的投影片。如果你告訴老闆：「請幫我連續印 20 份接在一起！」，老闆只要在影印機鍵盤上輸入 `20`，機器就會快速「刷刷刷」吐出一長條印有 20 顆星號的紙帶。  
在其他傳統程式語言（如 C/C++ 或 Java）中，如果你想印出一條包含 50 個減號的分割線 `--------------------------------------------------`，你必須辛苦地寫一個迴圈跑 50 次；  
但在 Python 中，你只需要寫下：  
`"-" * 50`！  
星號 `*` 就是這台高智慧影印機，專門將文字按倍率快速複製！

**底層運作機制與乘數規則**  
字串乘法運算子的語法為：`字串 * 整數`（或 `整數 * 字串`，順序不影響結果）。  
直譯器會建立一個長度為原本長度 $k$ 倍的新字串，並把字元依序填入。
- 範例：`"abc" * 3` 得到 `"abcabcabc"`。
- 特殊數值行為：
  - 若乘以 `0` 或任何負整數（如 `"test" * 0` 或 `"test" * -5`），Python **不會報錯**，而是會極其優雅地回傳一個「**空字串 `""`**」！

**常見錯誤陷阱：字串不能乘以浮點數（TypeError）**  
乘法運算子對右側的型態有嚴格限制——**乘數必須是整數（`int`）**！  
看看這個錯誤示範：  
`"star" * 2.5` ❌  
這會引發錯誤：  
**`TypeError: can't multiply sequence by non-int of type 'float'`**  
電腦無法理解什麼叫做「印出 2.5 個 star」。即便你的浮點數剛好沒有小數點（如 `2.0`），它依然是 `float`，同樣會報錯！如果數值來自除法運算（例如 `n / 2`），務必先用整數除法 `//` 或以 `int()` 強制轉型為整數。

**APCS 考試實務建議：**  
在題目需要排版、輸出棋盤邊界、進度條、縮排空格或幾何星號圖形時，善用 `" " * spaces` 與 `"*" * stars`，可以徹底省去內層迴圈，讓你的代碼行數大幅縮減一半以上，同時大幅降低邏輯錯誤的風險！

````python
# [2] Code 範例區：展示字串乘法運算子（*）的高效重複造字
# 說明：使用 * 快速生成分割線、重複字詞與空字串邊界。

# 範例 1：快速生成華麗的裝飾分割線
divider = "=" * 30
print(divider)
print("★ PYTHON ADVENTURE ★")
print(divider)

# 範例 2：單字倍率複製
cheer = "Go! " * 3
print("加油口號:", cheer)

# 範例 3：乘以 0 產生空字串展示
empty_result = "Ghost" * 0
print("乘以 0 的結果長度:", len(empty_result), "(字串為: '" + empty_result + "')")
````

### ⚖️ 7-2-4 運算優先順序與小括號：加法與乘法的混用

**生活比喻：數學王國的憲法守門員**  
在小學數學中，我們都背過一條不可撼動的鐵則：「先乘除，後加減；括號最優先！」  
很多人以為這個規則只存在於數字的世界裡。事實上，在 Python 的文字魔法王國中，這套規則依然 100% 絕對生效！  
當我們在同一個表達式中混合使用字串加法 `+` 與字串乘法 `*` 時，直譯器會毫不猶豫地**先執行乘法 `*`，再執行加法 `+`**！

**底層運作機制與優先權剖析**  
讓我們來仔細觀察以下兩行程式碼的巨大差異：  
1. **未加括號**：  
   `result = "A" + "B" * 3`  
   - 直譯器會先算 `"B" * 3` 得到 `"BBB"`。  
   - 接著算 `"A" + "BBB"`，最終得到 **`"ABBB"`**！  
2. **加上小括號 `(...)` 強制提升優先權**：  
   `result = ("A" + "B") * 3`  
   - 小括號擁有最高優先權！直譯器先算 `"A" + "B"` 得到 `"AB"`。  
   - 接著算 `"AB" * 3`，最終得到 **`"ABABAB"`**！  

**常見錯誤陷阱：忘記加括號導致圖案錯亂**  
如果我們想要產生一個對聯裝飾花紋 `"<*>"` 重複 3 次：  
初學者隨手寫下：`pattern = "<" + "*" + ">" * 3` ❌  
結果螢幕上印出的不是 `<*><*><*>`，而是：  
`<*>>>`！  
只有最後一個字元 `">"` 被放大了 3 倍，前面的符號完全被拋下了。  
只要適時加上小括號：`("<" + "*" + ">") * 3` 或 `"<*>" * 3`，就能精準達成目標。

### 🏛️ 7-2-5 迴圈與字串運算實戰：動態圖形與文字累加構造

**生活比喻：金字塔石工與編織花毯**  
在古代建造埃及金字塔時，石工們一層一層往上堆疊巨石。最底層的石塊最多，越往上石塊數量越少，但每一層的石塊都嚴格對齊中央。  
在第六章中，我們如果要印出類似的直角三角形或等腰三角形，必須使用「雙重巢狀迴圈」——外層迴圈控制高度，內層寫兩個迴圈分別慢慢敲空白與星號，思維複雜且極易發生 `off-by-one` 邊界失誤。  
但現在，擁有了字串加法 `+` 與乘法 `*` 的你，已經脫胎換骨了！  
你可以在外層迴圈的一瞬間，直接將空白字串與星號字串「一行搞定」！

**底層運作機制：單層迴圈壓平雙重巢狀迴圈**  
觀察經典的靠右對齊直角三角形（高度 $n = 4$）：
```text
   *  (3 空格 + 1 星號)
  **  (2 空格 + 2 星號)
 ***  (1 空格 + 3 星號)
****  (0 空格 + 4 星號)
```
在第 $i$ 行（$i$ 從 1 到 $n$）：
- 空格的數量剛好是：`n - i` 個。
- 星號的數量剛好是：`i` 個。
在 Python 裡，整行圖形直接寫作：  
`line = " " * (n - i) + "*" * i`  
`print(line)`  
原本繁重的雙層迴圈，被瞬間化簡為一個輕巧優雅的單層迴圈！

**常見錯誤陷阱：動態次數計算為負數**  
當使用 `" " * (n - i)` 時，若迴圈範圍設計不當使得算式結果變成負數，雖然 Python 不會報錯（會自動當作 0 個空格），但圖形會發生嚴重的「縮排崩塌」，導致整個形狀向左偏移。務必在紙上畫出小表格，驗證首行與末行的數量公式。

````python
# [2] Code 範例區：展示結合迴圈、字串乘法與加法的動態三角形繪製
# 說明：利用單層 for 迴圈取代雙層迴圈，極速印出靠右對齊直角三角形。

height = 5
print("=== 靠右對齊直角三角形 (高度 5) ===")

for i in range(1, height + 1):
    # 空格數量 = height - i，星號數量 = i
    spaces = " " * (height - i)
    stars = "*" * i
    print(spaces + stars)
````
