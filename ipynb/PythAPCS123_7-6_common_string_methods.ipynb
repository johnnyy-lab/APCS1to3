### 🔍 7-6-1 子字串成員檢查：in 與 not in 運算子

**生活比喻：機場安檢掃描門與金屬探測器**  
當旅客走過機場安檢門時，安檢門會發出「嗶嗶」聲，立刻偵測出旅客隨身行李中是否夾帶了違禁品。安檢人員不需要把行李箱的衣服一件一件拿出來翻找，探測器就能一瞬間判定「有」還是「沒有」。  
在 Python 中，專門擔任文字探測器的語法，就是無比簡潔的 **`in`（存在於...之中）** 與 **`not in`（不存在於...之中）**！

**底層運作機制與布林值回傳**  
語法：`子字串 in 目標字串`
- 若 `子字串` 完整且連續地出現在 `目標字串` 中，表達式立即回傳 **`True`**。
- 若不存在，則回傳 **`False`**。
- 範例：`"cat" in "education"` 結果為 `True`（因為 edu-**cat**-ion 中間確實藏著 cat）！
- 範例：`"dog" in "education"` 結果為 `False`。

`not in` 則是反向判斷：若不存在則回傳 `True`。例如 `"bad" not in "good morning"` 結果為 `True`。

**常見錯誤陷阱：大小寫嚴格相符**  
`in` 運算子在比對時是**極度嚴格且區分大小寫**的！  
`"Python" in "python programming"` 結果為 **`False`**！因為大寫 `'P'` 與小寫 `'p'` 在電腦底層的 ASCII 碼不同。若需要忽略大小寫，必須配合後續學到的 `.lower()` 先轉小寫再比對。

**APCS 考試實務建議：**  
在各大程式題庫中，經常需要檢查文字是否含有禁字、母音（如 `ch in "aeiouAEIOU"`）、或檢查使用者指令是否合法。使用 `in` 可以完全免去寫雙層迴圈逐一比對的繁瑣，讓條件判斷式乾淨得像英文口語句子一樣清晰。

````python
# [2] Code 範例區：展示 in 與 not in 的文字探測效果
# 說明：觀察子字串的存在性檢測與大小寫敏感性。

article = "Python is an amazing and powerful programming language."

# 檢測子字串是否存在
has_powerful = "powerful" in article
print("'powerful' 是否在文章中:", has_powerful)

# 檢測大小寫敏感性
has_upper = "LANGUAGE" in article
print("'LANGUAGE' (全大寫) 是否在文章中:", has_upper)

# 經典應用：快速判斷某個字元是否為英文母音
letter = "e"
if letter in "aeiouAEIOU":
    print("字元 '" + letter + "' 是英文母音！")
````

### 📊 7-6-2 字元與字詞頻率統計：.count() 方法與非重疊特性

**生活比喻：唱票計票員與按次計費器**  
在班級模範生選舉開票時，開票員手裡拿著粉筆，只要黑板上聽到「1 號一票」，就在 1 號下方畫上一筆正字標記。整張票開完，就能立即算出 1 號總共拿到了幾票。  
在 Python 中，字串物件隨身自帶的計票大師就是 **`.count()`** 方法！  
只要在字串變數後面打一個小數點（`.`），呼叫 `s.count(sub)`，直譯器就會瞬間幫你統計出 `sub` 在整個字串中出現的總次數！

**底層運作機制與語法規範**  
語法：`次數 = s.count(目標子字串)`
- 可以統計**單一字元**：`"banana".count("a")` 得到 `3`。
- 也可以統計**多字元單詞**：`"apple, orange, apple".count("apple")` 得到 `2`。
- 若目標完全沒出現：`"banana".count("z")` 得到 `0`（安全無報錯）。

**關鍵陷阱：非重疊計算（Non-overlapping）**  
初學者必須深入理解 `.count()` 的底層計數邏輯是「**非重疊的（Non-overlapping）**」！  
觀察以下例子：  
`s = "AAAA"`  
`result = s.count("AA")`  
很多人直覺以為會數出 3 次（索引 0-1 的 AA、索引 1-2 的 AA、索引 2-3 的 AA）。  
但 Python 回傳的結果卻是 **`2`**！  
因為當直譯器在索引 0 和 1 抓出了第一對 `"AA"` 之後，游標會立刻跳過這兩個已被認領的字元，直接從索引 2 開始繼續往後找下一組獨立的 `"AA"`！如果題目明確要求「允許重疊計算」，就必須改用 7-5 學過的切片迴圈滑動窗口比對。

**APCS 考試實務建議：**  
在 DNA 序列分析（統計 A、T、C、G 個數）或字元頻率統計題目中，`.count()` 可以在常數行內免寫迴圈直接秒出統計數據，是搶分加速的絕對利器。

````python
# [2] Code 範例區：展示 .count() 方法的字詞統計與非重疊特性
# 說明：觀察單一字元計數、多字元計數與非重疊現象。

dna = "AGCTATTGCAATGC"

# 統計腺嘌呤 'A' 的總出現次數
count_A = dna.count("A")
count_T = dna.count("T")
print("DNA 序列:", dna)
print("'A' 出現次數:", count_A)
print("'T' 出現次數:", count_T)

# 展示非重疊特性 (Non-overlapping)
pattern = "AAAA"
print("\n'AAAA' 中計算 'AA' 的次數:", pattern.count("AA"), "(非重疊獨立兩組)")
````

### ✂️ 7-6-3 句子拆詞利器：.split() 空白切割與詞彙走訪

**生活比喻：報紙文章的字詞裁切刀與單字拆解**  
當我們翻開一本英文小說時，每句話都是由好幾個單字組成，單字之間以空白間隔。如果我們想要逐一分析這句話裡的每個單字，最直觀的做法就是拿剪刀沿著空白縫隙一刀一刀剪開，把整句話變成一顆顆獨立的單字積木！  
在第四章讀取輸入時，我們曾經接觸過 `input().split()`；現在，我們正式將它提升為字串的招牌核心技能——**`.split()`（字串分割）**！

**底層運作機制與預設空白切割特性**  
語法：`words = s.split()`
- 當 `.split()` **括號內不給任何參數時**，它具備無與倫比的高智慧：
  1. 自動以**任何空白字元（空格、Tab 跳格 `\t`、換行 `\n`）**作為刀刃進行切割。
  2. **自動忽略連續的多餘空格**！例如 `"Hello      World".split()` 依然乾乾淨淨切成 `Hello` 與 `World` 兩個單字，完全不會產生多餘的空字串！
- 我們可以透過最純粹的 for 迴圈直接走訪切割後的每個單字：  
  ```python
  for word in s.split():
      print("單字:", word)
  ```

**指定分隔符號（Delimiter）：**  
如果資料不是以空白分隔，而是以逗號（CSV 格式）或斜線分隔：  
- `"apple,banana,orange".split(",")`：以逗號切分！
- `"2026/09/17".split("/")`：以斜線切分！

**APCS 考試實務建議：**  
在 APCS 檢定中，超過 90% 的測資都是單行多個數字或單詞。學會直接對字串變數使用 `.split()`，並搭配迴圈逐一走訪處理，是處理整行文字輸入時最核心的黃金解題骨架。

````python
# [2] Code 範例區：展示 .split() 拆解句子與迴圈逐詞走訪
# 說明：觀察預設空白切割與自訂分隔符號（如逗號）的效果。

sentence = "Learn Python to   conquer APCS!"

print("=== 預設空白切割逐詞走訪 ===")
# 即使中間有多個連續空格，.split() 也能完美忽略雜訊
for word in sentence.split():
    print("單字:", word, "(長度:", len(word), ")")

print("\n=== 自訂逗號分隔切割 ===")
csv_line = "Apple,Banana,Grape,Peach"
for fruit in csv_line.split(","):
    print("水果品名:", fruit)
````

### 🎯 7-6-4 尋找子字串位置：.find() 安全查詢與 -1 回傳機制

**生活比喻：雷達座標定位與不會發火的溫和管家**  
當軍艦在海面上搜尋目標島嶼時，雷達一掃描，立刻回報：「報告船長！目標島嶼在正前方距離 5 浬處（索引 5）！」。  
但如果大海上什麼都沒有呢？  
有些脾氣暴躁的管家（如字串的 `.index()` 方法）會氣急敗壞地直接引發系統當機報錯（`ValueError`）；  
而 Python 字串中還有一位無比優雅沉著的管家，叫做 **`.find()`**！  
當它找不到目標時，它不會崩潰，而是平靜地遞給你一張寫著 **`-1`** 的紙條，禮貌地表示：「報告，查無此物！」。

**底層運作機制與語法規範**  
語法：`pos = s.find(sub)`
- **找到目標**：回傳 `sub` 在 `s` 中**第一次出現的最左端起始索引號碼**（0-based）。
- **找不到目標**：回傳 **`-1`**！
- 範例：`"hello".find("ll")` 得到 `2`（因為 'll' 從索引 2 開始）。
- 範例：`"hello".find("z")` 得到 `-1`。

**為什麼回傳 -1？因為 0 也是合法的索引！**  
很多初學者問：「為什麼找不到不回傳 0 呢？」  
因為如果目標字串剛好在「最開頭的第一個字」，它的索引就是 `0`（例如 `"cat".find("c")` 回傳 `0`）！如果找不到也回傳 0，程式就會徹底分不清到底是有找到還是沒找到。因此，在自然數索引之外的 `-1`，就成為全世界工程師約定俗成的「查無此物標誌」。

**APCS 考試實務建議：**  
當我們需要知道某個符號（如逗號、冒號、小老鼠 `@`）在字串的第幾個位置以便進行後續切片時，`.find()` 是最安全無死角的首選，不需要加 `try...except` 就能徹底防禦程式崩潰。

````python
# [2] Code 範例區：展示 .find() 的定位查詢與 -1 查無回傳
# 說明：觀察成功找到與找不到時的回傳值差異，以及搭配切片的妙用。

email = "student2026@apcs.org.tw"

# 尋找小老鼠 @ 的位置
at_pos = email.find("@")
print("電子郵件:", email)
print("'@' 的索引位置:", at_pos)

# 搭配切片：利用 @ 的位置自動截取使用者名稱
username = email[:at_pos]
print("自動截取使用者名稱 email[:at_pos]:", username)

# 測試找不到的情況
not_found = email.find("xyz")
print("尋找不存在的 'xyz' 結果:", not_found, "(代表查無此字串)")
````

````python
# ==========================================
# [3] Code 填空題
# 任務說明：
# 請呼叫 find 方法尋找冒號 ":" 的位置，
# 若找到（pos != -1），利用切片取出冒號右側的數值字串！
# ==========================================

data_record = "Score:98"

# 請填入 find 方法尋找冒號：
colon_pos = data_record.___(":")

if colon_pos != -1:
    score_str = data_record[colon_pos + 1 :]
else:
    score_str = "0"

# 驗證區（請勿更動下方代碼）
if score_str == "98":
    print("恭喜答對！成功使用 .find() 定位並完成切片分離！")
else:
    print("檢查看看方法名稱是否填入 find 喔！")
````

### 🔠 7-6-5 大小寫轉換比對：.lower() 與 .upper() 規範化對照

**生活比喻：軍隊的統一制服與標準化度量衡**  
當新兵入伍時，來自各地的士兵穿著五花八門的花襯衫、T恤或外套。長官一聲令下，所有人換上整齊劃一的軍隊迷彩制服，這時候點名與排隊就變得一目了然、毫不紊亂！  
在電腦的世界裡，人類輸入文字往往極為隨性：有人打 `"yes"`，有人打 `"Yes"`，甚至有人敲出 `"YES"`！  
如果程式要一條一條 `if ans == "yes" or ans == "Yes" or ...` 窮舉，那簡直是自討苦吃。  
這時候，我們只需要讓文字全員「換上統一制服」——**`.lower()`（全員轉小寫）** 或 **`.upper()`（全員轉大寫）**！

**底層運作機制與不可變特性提醒**  
- `s.lower()`：將字串中所有大寫英文字母轉換為小寫，其他非字母字元（數字、漢字、符號）保持不變。
- `s.upper()`：將字串中所有小寫英文字母轉換為大寫。
- **不可變性提醒（牢記單元 7-4 原理）**：呼叫 `s.lower()` **不會**修改原本的 `s`，而是產生並回傳一個「全新字串」！若要保存轉換結果，必須重新賦值：`s = s.lower()`！

**經典殺手級應用：不分大小寫比對（Case-insensitive Comparison）**  
只要將比對雙方同時轉為小寫：  
```python
if user_input.lower() == target.lower():
    print("比對成功（忽略大小寫差異）！")
```
不管使用者輸入 `"Python"`、`"PYTHON"` 還是 `"pYtHoN"`，全部被瞬間降伏！

**APCS 考試實務建議：**  
許多 APCS 題目的說明中會出現一行關鍵小字：「**比對時不區分大小寫**」。一看到這個要求，立刻反射性寫出 `.lower()` 進行輸入規範化（Normalization），就能確保在嚴苛的自動評判測資下穩拿滿分！

````python
# [2] Code 範例區：展示 .lower() 與 .upper() 的大小寫轉換與忽略大小寫比對
# 說明：觀察非破壞性轉換與標準化比對技巧。

raw_text = "Hello, World! APCS 2026."

# 全員轉小寫
lower_text = raw_text.lower()
print("原字串    :", raw_text)
print(".lower()  :", lower_text)

# 全員轉大寫
upper_text = raw_text.upper()
print(".upper()  :", upper_text)

# 忽略大小寫比對展示
ans1 = "Python"
ans2 = "pYtHoN"
print("\n直接比對 ans1 == ans2        :", ans1 == ans2)
print("轉小寫後 ans1.lower() == ans2.lower():", ans1.lower() == ans2.lower())
````

````python
# ==========================================
# [3] Code 填空題
# 任務說明：
# 請呼叫 lower() 方法將使用者的確認輸入 command 轉為全小寫，
# 使得無論使用者鍵入 "YES"、"Yes" 還是 "yes"，都能判定成功！
# ==========================================

command = "YeS"

# 請填入 lower() 進行標準化轉換：
if command.___() == "yes":
    action = "CONFIRMED"
else:
    action = "CANCELLED"

# 驗證區（請勿更動下方代碼）
if action == "CONFIRMED":
    print("恭喜答對！成功使用 .lower() 實現不分大小寫比對！")
else:
    print("檢查看看方法名稱是否為 lower 喔！")
````
