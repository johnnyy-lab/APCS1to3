## 10.6.1 集合概念：無序性（Unordered）與元素唯一性（Unique 天然去重神器）

### 觀念說明
在數學中，集合是一堆不重複物件的匯集。在 Python 中，集合（set）同樣擁有這兩大天賦特權：
1. **元素唯一性（Unique）**：集合中**絕不允許存在重複的元素**！即使你在宣告時寫入了十個相同的數字，Python 也只會保留其中一個，其他重複項會被自動抹除，因此集合是天生的「天然去重神器」！
2. **無序性（Unordered）**：集合內的元素沒有先後順序或索引編號的概念。因此，**集合絕對不能使用索引取值**（例如寫 `s[0]` 會立刻引發 `TypeError` 崩潰）！

### 集合的字面值語法
集合使用**大括號 `{}`** 將元素包起來，元素之間以逗號 `,` 隔開，例如：`{1, 2, 3}`。
請注意：它與字典的差別在於，集合只有單純的元素，**中間沒有冒號 `:`**！

````python
# 10.6.1 範例：集合的天然去重與無序特性
# 宣告一個包含許多重複數字的集合
numbers_set = {3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5}

print("集合宣告後自動去重結果:", numbers_set)
print("集合中不重複元素的數量:", len(numbers_set))

# 驗證集合的型態
print("型態檢查:", type(numbers_set))

# 嘗試使用索引存取會報錯（解除註解會觸發 TypeError）
# print(numbers_set[0])  # 錯誤！TypeError: 'set' object is not subscriptable

# 走訪集合：可以使用 for 迴圈逐一取出元素（順序不保證固定）
print("走訪集合元素:")
for item in numbers_set:
    print(item, end=" ")
print()
````

## 10.6.2 集合建立語法與空集合陷阱：為什麼 {} 是字典，空集合必須寫 set()！

### 觀念說明
這是所有 Python 初學者在考場與開發中最容易踩中的**致命天坑**！

在日常寫程式時，大家習慣這樣宣告空容器：
- 空串列：`lst = []`
- 空元組：`tup = ()`
- 那麼空集合呢？如果你順理成章寫出 `s = {}`，**那你就徹底中計了！**

### 為什麼 `{}` 不是空集合？
因為在 Python 早期歷史中，字典（Dictionary）比集合更早誕生，因此**一對空大括號 `{}` 已經被字典佔用了**！
如果你寫 `s = {}`，它的型態其實是 `<class 'dict'>`！後續呼叫集合的方法（如 `s.add(5)`）會直接噴出 `AttributeError: 'dict' object has no attribute 'add'` 導致程式崩潰。

### 唯一正確建立空集合的語法
要建立一個真正的空集合，**必須且只能使用型態建構子**：
```python
empty_set = set()
```
請務必將這條鐵則刻在腦海裡：`{}` 是字典，`set()` 才是空集合！

````python
# 10.6.2 範例：空字典 vs 空集合的殘酷對比
# 1. 誤用 {} 建立
fake_set = {}
print("fake_set = {}")
print("內容:", fake_set)
print("真實型態:", type(fake_set))  # 居然是 <class 'dict'>！

print()
# 2. 正確使用 set() 建立空集合
real_set = set()
print("real_set = set()")
print("內容:", real_set)
print("真實型態:", type(real_set))  # 真正的 <class 'set'>！

# 3. 嘗試呼叫 .add() 方法
real_set.add(100)
print("成功加入元素後的 real_set:", real_set)

# fake_set.add(100) # 若執行這行會引發 AttributeError 崩潰！
````

````python
# 10.6.2 填空題：宣告真正的空集合並加入元素
# 任務：請使用 set() 正確宣告一個名為 lucky_numbers 的空集合
lucky_numbers = set()

# 請將上方答案填入空格：
# lucky_numbers = ___()
lucky_numbers.add(7)
lucky_numbers.add(77)

print("lucky_numbers 型態:", type(lucky_numbers))
print("lucky_numbers 內容:", lucky_numbers)
# 預期輸出:
# lucky_numbers 型態: <class 'set'>
# lucky_numbers 內容: {77, 7} (順序不拘)
````

````python
# 10.6.2 練習題：動態收集非負整數集合
# 題目說明：
# 給定一串整數輸入 stream_data = [5, -2, 10, 5, 0, -8, 10, 3]
# 請先建立一個「真正的空集合」non_negative_set。
# 接著走訪 stream_data，只有當數值 >= 0 時才加入集合中。
# 最後輸出該集合的不重複元素個數以及集合內容。
#
# 【公開測試資料 1】
# 輸入串列: [5, -2, 10, 5, 0, -8, 10, 3]
# 符合 >= 0 的相異數字: {0, 3, 5, 10} 共 4 個
# 輸出: 相異非負整數個數: 4
#
# 【公開測試資料 2】
# 輸入串列: [-5, -10, -1]
# 輸出: 相異非負整數個數: 0

stream_data = [5, -2, 10, 5, 0, -8, 10, 3]

non_negative_set = set()
for num in stream_data:
    if num >= 0:
        non_negative_set.add(num)

print("相異非負整數個數:", len(non_negative_set))
````

````python
# 10.6.2 挑戰題：連續字母遇到哨兵字元停機收集
# 題目說明：
# 給定一串字元串列 raw_chars = ['A', 'B', 'A', 'C', 'X', 'D', 'B']
# 請建立一個空集合 collected = set()
# 逐一將字元加入 collected 中；但若遇到哨兵字元 'X'，必須立即使用 break 結束收集！
# 哨兵字元 'X' 本身不可加入集合中。
# 最後印出 collected 的相異字母數量與集合內容。

raw_chars = ['A', 'B', 'A', 'C', 'X', 'D', 'B']

# 請在下方寫出你的解答代碼
collected = set()
for ch in raw_chars:
    if ch == 'X':
        break
    collected.add(ch)

print("收集到的相異數量:", len(collected))
print("收集到的內容:", collected)
# 預期結果:
# 收集到的相異數量: 3
# 收集到的內容: {'A', 'B', 'C'} (順序不拘)
````

## 10.6.3 可迭代物件快速去重：set(my_list) 與去重後長度量測 len(s)

### 觀念說明
在 APCS 考場上，我們常常遇到這類題目要求：
- 「給定 $N$ 個數字，請問總共有幾種不同的數值？」
- 「請將輸入的陣列去除重複項目。」

如果用傳統的串列做法，你必須先宣告一個新串列，然後在迴圈中逐一寫 `if x not in unique_list:`。但因為在串列中使用 `in` 是 $O(N)$，總時間複雜度會退化成 **$O(N^2)$**，只要測資規模一萬筆就會超時！

### 一秒去重神技：`set(iterable)`
Python 允許我們直接把任何可迭代物件（如串列、字串、元組）當作引數丟給 `set()`！
```python
unique_items = set(my_list)
```
這行代碼會在底層以雜湊表進行極速過濾，時間複雜度僅為 **$O(N)$**！
搭配 `len(set(my_list))`，只需 0.01 秒就能精準算出「相異元素個數」，是 APCS 出現頻率最高的超核心神技。

````python
# 10.6.3 範例：串列一秒去重與相異數量統計
student_scores = [85, 92, 78, 85, 90, 92, 85, 78]
print("原始成績串列 (共", len(student_scores), "筆):", student_scores)

# 1. 轉為集合：自動一秒去重
distinct_scores = set(student_scores)
print("相異成績集合 (共", len(distinct_scores), "種):", distinct_scores)

# 2. 若需要轉回串列並排序
sorted_unique = sorted(distinct_scores)
print("排序後的相異分數串列:", sorted_unique)

# 3. 套用在字串上：找出句子中用到了哪些不重複字母
sentence = "hello world"
unique_letters = set(sentence)
print("句子中使用的相異字元 (含空格):", unique_letters)
````

````python
# 10.6.3 填空題：計算打卡名單中的不重複人數
checkin_logs = ["Alice", "Bob", "Alice", "David", "Bob", "Alice"]

# 任務：請使用 len() 與 set() 快速計算不重複的打卡人數
unique_visitor_count = len(set(checkin_logs))

# 請填入空格：
# unique_visitor_count = len(___(checkin_logs))
print("今日不重複打卡人數:", unique_visitor_count)
# 預期輸出: 今日不重複打卡人數: 3
````

````python
# 10.6.3 練習題：相異整數計數與純度檢測
# 題目說明：
# 給定整數串列 data_list。
# 請先輸出相異整數的數量。
# 接著請判斷：該串列是否「完全沒有重複元素」（即每個數字都只出現一次）？
# 若完全無重複，輸出 "純淨無重複"；若有重複，輸出 "包含重複項"。
# 提示：比較 len(data_list) 與 len(set(data_list))
#
# 【公開測試資料 1】
# 串列: [10, 20, 30, 40, 50]
# 輸出:
# 相異元素數量: 5
# 純淨無重複
#
# 【公開測試資料 2】
# 串列: [1, 2, 3, 2, 1]
# 輸出:
# 相異元素數量: 3
# 包含重複項

data_list = [10, 20, 30, 40, 50]

unique_count = len(set(data_list))
print("相異元素數量:", unique_count)

if len(data_list) == unique_count:
    print("純淨無重複")
else:
    print("包含重複項")
````

## 10.6.4 元素動態增刪：.add(x)、.remove(x) 與防報錯 .discard(x)

### 觀念說明
集合是可變物件（Mutable），我們可以隨時往集合裡加入新成員或刪除舊成員。

#### 1. 新增元素：`s.add(x)`
- 若 `x` 原本不在集合中，成功加入。
- 若 `x` 原本就已經在集合中，不會報錯，集合維持原狀。

#### 2. 刪除元素：`.remove(x)` vs `.discard(x)`（關鍵避坑！）
- **`s.remove(x)`（嚴格刪除）**：
  如果 `x` 存在，正常刪除。
  💥 **致命陷阱**：如果 `x` 根本不在集合中，會**立刻引發 `KeyError` 導致程式死機崩潰**！
- **`s.discard(x)`（安全刪除神器！）**：
  如果 `x` 存在，正常刪除。
  🛡️ **安全保護**：如果 `x` 不存在，**它會靜默忽略，完全不會報錯**！

在 APCS 或線上競賽中，除非你百分之百確定該元素存在，否則**強烈建議一律優先使用 `.discard()`**，徹底避免不小心中招噴出 `KeyError`！

````python
# 10.6.4 範例：.add()、.remove() 與安全 .discard() 的行為差異
active_users = {"Alice", "Bob", "Charlie"}
print("初始上線使用者:", active_users)

# 1. 使用 .add() 新增使用者
active_users.add("David")
print("新增 David 後:", active_users)

# 重複加入已經存在的 Alice（不會報錯，也不會變多）
active_users.add("Alice")
print("重複加入 Alice 後:", active_users)

# 2. 使用 .discard() 安全刪除已存在的人
active_users.discard("Bob")
print("使用 discard 移除 Bob:", active_users)

# 3. 使用 .discard() 嘗試刪除「不存在」的人（安全無虞！）
active_users.discard("Ghost_User")
print("使用 discard 嘗試移除不存在的幽靈用戶（完全不報錯）:", active_users)

# 4. 若換成 .remove() 刪除不存在的人（會崩潰！）
# active_users.remove("Ghost_User")  # 拋出 KeyError: 'Ghost_User'
````

````python
# 10.6.4 填空題：使用 discard 安全移除白名單成員
members = {"user1", "user2", "user3"}

target = "user99"
# 任務：請使用 discard 方法嘗試移除 target，即使不存在也不會報錯
members.discard(target)

# 請填入空格：
# members.___(target)
print("操作後名冊:", members)
# 預期輸出: 操作後名冊: {'user1', 'user2', 'user3'} (順序不拘)
````

````python
# 10.6.4 練習題：動態聊天室在線名冊維護器
# 題目說明：
# 現有一個空聊天室 online_users = set()。
# 給定操作紀錄 actions，每筆紀錄為 (操作類型, 使用者名稱)：
# - 若類型為 "IN"：表示該用戶進入聊天室，請使用 .add() 加入。
# - 若類型為 "OUT"：表示該用戶離開聊天室，請使用 .discard() 安全移除。
# 請依序處理每筆操作，並在最後輸出剩餘在線人數與排序後的名冊串列。
#
# 【公開測試資料 1】
# 操作紀錄: [("IN", "Tom"), ("IN", "Jerry"), ("OUT", "Tom"), ("IN", "Spike")]
# 輸出:
# 在線人數: 2
# 在線名單: ['Jerry', 'Spike']
#
# 【公開測試資料 2】
# 操作紀錄: [("IN", "A"), ("OUT", "B"), ("OUT", "A")]  # B 從未進入過
# 輸出:
# 在線人數: 0
# 在線名單: []

actions = [("IN", "Tom"), ("IN", "Jerry"), ("OUT", "Tom"), ("IN", "Spike")]

online_users = set()
for op, user in actions:
    if op == "IN":
        online_users.add(user)
    elif op == "OUT":
        online_users.discard(user)

print("在線人數:", len(online_users))
print("在線名單:", sorted(online_users))
````

````python
# 10.6.4 挑戰題：黑名單批次安全過濾
# 題目說明：
# 給定一個已核准名單 approved = {"id_1", "id_2", "id_3", "id_4", "id_5"}
# 以及一個違規黑名單 blacklist = ["id_2", "id_999", "id_4"]
# 請使用 for 迴圈搭配 .discard() 將黑名單中的所有帳號從 approved 中剔除。
# （黑名單可能包含從未出現過的帳號，程式絕對不可中斷！）
# 最後印出 approved 剩餘人數與內容。

approved = {"id_1", "id_2", "id_3", "id_4", "id_5"}
blacklist = ["id_2", "id_999", "id_4"]

# 請在下方寫出你的解答代碼
for bad_id in blacklist:
    approved.discard(bad_id)

print("剩餘合格人數:", len(approved))
print("剩餘合格名單:", sorted(approved))
# 預期結果:
# 剩餘合格人數: 3
# 剩餘合格名單: ['id_1', 'id_3', 'id_5']
````

## 10.6.5 集合生成式（Set Comprehension）語法與動態過濾

### 觀念說明
在學習過串列生成式 `[x for x in ...]` 與字典生成式 `{k: v for ...}` 之後，Python 的生成式拼圖終於迎來了第三位成員——**集合生成式（Set Comprehension）**！

### 語法結構
```python
{expression for item in iterable if condition}
```
集合生成式的特點：
1. 外層使用**大括號 `{}`**。
2. 裡面是單一表達式 `expression`（**沒有冒號 `:`**，如果加了冒號就變成字典生成式了！）。
3. **自動去重**：在生成過程中所產生的任何重複結果，會被集合自動消去！

這讓我們能夠用精練的一行代碼，同時完成「條件篩選、數值運算、天然去重」三道手續，非常適合在 APCS 處理相異特徵提取。

````python
# 10.6.5 範例：集合生成式的篩選與自動去重
raw_numbers = [1, 2, 3, 4, 5, 2, 4, 6, 8, 10]

# 1. 取出所有偶數的平方（天然去重）
even_squares = {x ** 2 for x in raw_numbers if x % 2 == 0}
print("偶數平方集合:", even_squares)

# 2. 取出單字長度集合
words = ["cat", "apple", "dog", "banana", "pie", "kiwi"]
word_lengths = {len(w) for w in words}
print("出現過的不同單字長度:", word_lengths)

# 3. 字串正規化去重（轉小寫後存入集合）
letters = ["A", "b", "a", "C", "B", "c"]
normalized_letters = {ch.lower() for ch in letters}
print("轉小寫去重字母集合:", normalized_letters)
````

````python
# 10.6.5 填空題：以集合生成式挑選大於 50 的相異數值
scores = [45, 88, 72, 88, 30, 95, 72, 50]

# 任務：使用集合生成式篩選出 scores 中嚴格大於 50 的所有相異分數
high_scores = {s for s in scores if s > 50}

# 請填入空格：
# high_scores = {s for s in scores if s > ___}
print("大於 50 的相異分數:", high_scores)
# 預期輸出: 大於 50 的相異分數: {72, 88, 95} (順序不拘)
````

````python
# 10.6.5 練習題：相異字串首字母集合提取器
# 題目說明：
# 給定一個包含許多單字的串列 text_words。
# 請使用「集合生成式」一行萃取出所有單字的「大寫首字母」（w[0].upper()），
# 並且只處理長度大於等於 3 的單字（len(w) >= 3）。
# 最後依字母順序（A 到 Z）將萃取出的字母印出為串列。
#
# 【公開測試資料 1】
# 單字串列: ["apple", "an", "banana", "bear", "at", "cat"]
# 符合長度 >= 3 的單字: "apple"(A), "banana"(B), "bear"(B), "cat"(C)
# 輸出: 排序後相異首字母: ['A', 'B', 'C']
#
# 【公開測試資料 2】
# 單字串列: ["go", "python", "java", "php"]
# 輸出: 排序後相異首字母: ['J', 'P']

text_words = ["apple", "an", "banana", "bear", "at", "cat"]

initial_set = {w[0].upper() for w in text_words if len(w) >= 3}

print("排序後相異首字母:", sorted(initial_set))
````

````python
# 10.6.5 挑戰題：英文句子中相異母音萃取
# 題目說明：
# 給定英文句子 quote = "The quick brown fox jumps over the lazy dog"
# 請只用「一行集合生成式」完成：
# 1. 逐一取出句子中的每個字元，轉為小寫（ch.lower()）
# 2. 篩選出屬於母音的字元（'a', 'e', 'i', 'o', 'u'）
# 3. 自動去重形成集合
# 最後印出該句子總共出現了幾種相異母音，以及排序後的母音串列。

quote = "The quick brown fox jumps over the lazy dog"

# 請在下方寫出你的單行集合生成式解答
vowels = {ch.lower() for ch in quote if ch.lower() in "aeiou"}

print("母音種類數量:", len(vowels))
print("相異母音列表:", sorted(vowels))
# 預期結果:
# 母音種類數量: 5
# 相異母音列表: ['a', 'e', 'i', 'o', 'u']
````

## 10.6.6 集合元素限制：必須是可雜湊物件（可存 int, str, tuple，嚴禁放 list 或 set）

### 觀念說明
學到這裡，大家應該會發現集合與字典有著強烈的血緣關係！
沒錯，在 Python 底層，**集合其實就是「只有 Key、沒有 Value」的雜湊表**！

因此，我們在 10-2 與 10.3.3 學過關於雜湊（Hashable）的所有鐵則，在集合中**一字不差全部適用**：
**集合裡的所有元素，都必須是可雜湊（Hashable）的不可變物件！**

#### 🔑 合法放進集合的元素：
1. **整數與浮點數**：`{1, 2, 3.14}`
2. **文字字串**：`{"Alice", "Bob"}`
3. **不可變元組（Tuple）**：`{(0, 1), (2, 3)}`（APCS 網格已拜訪座標標記必殺技！）

#### ❌ 嚴禁放進集合的元素：
- **串列（List）**：若嘗試寫 `{ [1, 2], [3, 4] }`，直譯器會立刻拋出 `TypeError: unhashable type: 'list'`！
- **字典（Dict）與集合自身（Set）**：都是可變物件，同樣會引發 `TypeError`。

牢記這條鐵律，在 APCS 二維網格題目中，隨手用 `visited = set()` 搭配元組 `visited.add((r, c))`，就能在 $O(1)$ 時間內完成路徑走訪記錄！

````python
# 10.6.6 範例：合法座標元組集合 vs 串列報錯示範
# 1. 合法：使用座標元組 (r, c) 記錄網格地圖中的障礙物點位
obstacles = {(1, 2), (3, 4), (1, 2), (0, 0)}  # (1, 2) 重複自動去重
print("障礙物座標集合:", obstacles)
print("障礙物點數:", len(obstacles))

# 檢查某個座標是否有障礙物（O(1) 瞬間雜湊查表）
query_pos = (3, 4)
if query_pos in obstacles:
    print(f"座標 {query_pos} 存在障礙物，無法通行！")

# 2. 捕捉串列放進集合的 TypeError 崩潰
try:
    illegal_set = set()
    illegal_set.add([1, 2])  # 串列不可雜湊！
except TypeError as e:
    print()
    print("❌ 攔截到錯誤:", e)
    print("說明: 串列為可變物件，絕對不可作為集合的元素！")
````

````python
# 10.6.6 填空題：以元組 (r, c) 標記已拜訪座標
visited_coords = set()

start_point = (0, 0)
next_point = (1, 0)

# 任務：請將 start_point 與 next_point 加入 visited_coords 集合中
visited_coords.add(start_point)
visited_coords.add(next_point)

# 請填入空格：
# visited_coords.add(___)
# visited_coords.add(___)
print("已拜訪座標集合:", visited_coords)
# 預期輸出: 已拜訪座標集合: {(0, 0), (1, 0)} (順序不拘)
````

````python
# 10.6.6 練習題：網格足跡重複踩踏偵測機
# 題目說明：
# 機器人在網格地圖上移動，其經過的座標點依序記錄在 path_points 串列中（每個元素為 (r, c) 元組）。
# 請使用一個空集合 visited = set() 來追蹤機器人走過的所有座標。
# 走訪 path_points：
# 若該座標已經在 visited 中，印出 "重複踩踏座標: (r, c)"
# 若不在，則將該座標加入 visited 中。
#
# 【公開測試資料 1】
# 移動軌跡: [(0, 0), (0, 1), (1, 1), (0, 1), (2, 2)]
# 輸出: 重複踩踏座標: (0, 1)
#
# 【公開測試資料 2】
# 移動軌跡: [(1, 1), (1, 2), (1, 3)]
# 輸出: 全程無重複踩踏

path_points = [(0, 0), (0, 1), (1, 1), (0, 1), (2, 2)]

visited = set()
has_duplicate = False

for pt in path_points:
    if pt in visited:
        print("重複踩踏座標:", pt)
        has_duplicate = True
    else:
        visited.add(pt)

if not has_duplicate:
    print("全程無重複踩踏")
````

````python
# 10.6.6 挑戰題：稀疏障礙物跨界安全過濾
# 題目說明：
# 給定一組障礙物座標集合 raw_obstacles = {(0, 1), (5, 5), (-1, 2), (3, 2), (2, 8)}
# 地圖邊界尺寸為 R = 4, C = 4（合法座標範圍為 0 <= r < 4 且 0 <= c < 4）。
# 請使用「集合生成式」，篩選出所有在合法地圖範圍內的有效障礙物座標集合 valid_obstacles。
# 最後印出合法障礙物點數與排序後的座標串列。

raw_obstacles = {(0, 1), (5, 5), (-1, 2), (3, 2), (2, 8)}
R = 4
C = 4

# 請在下方寫出你的解答代碼
valid_obstacles = {(r, c) for r, c in raw_obstacles if 0 <= r < R and 0 <= c < C}

print("合法障礙物點數:", len(valid_obstacles))
print("合法障礙物座標:", sorted(valid_obstacles))
# 預期結果:
# 合法障礙物點數: 2
# 合法障礙物座標: [(0, 1), (3, 2)]
````
