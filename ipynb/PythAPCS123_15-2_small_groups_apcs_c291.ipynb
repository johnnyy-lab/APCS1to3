### 📝 完整題目描述（Problem Description）
班上有 $N$ 位同學，每位同學都有一個從 $0$ 到 $N-1$ 的專屬座號。
在開學同樂會上，每位同學都在心裡選定了一位「最好的朋友」（也可以選自己作為最好的朋友）。

**小群體的定義如下：**
1. 如果甲同學的好友是乙同學，那麼甲和乙就在同一個小群體中；
2. 如果乙同學的好友是丙同學，那麼甲、乙、丙都在同一個小群體中；
3. 依此類推，如果一群同學透過「好友鏈」彼此相連，且最終繞回這群人中的某位同學，形成一個**封閉的循環圈子**，這群同學就組成了一個獨立的「小群體」。

題目保證這 $N$ 位同學選出的好朋友編號，恰好是 $0$ 到 $N-1$ 的一個**排列（Permutation）**。
也就是說：**每位同學恰好被一個人選為最好朋友，且每位同學也恰好只選一個人為最好朋友**。
因此，全班同學一定會被剛好劃分成若干個互不相交的封閉小圈圈。

請撰寫一個程式，計算全班一共形成了幾個互不相交的「小群體」？

## 15.2.1 題意解析：好朋友指涉關係與獨立朋友圈個數

### 🏰 生活故事比喻：全班同樂會的「神秘傳話筒小圈圈」
想像一下，班上正在進行一場團康遊戲。老師發給每位同學一張紙條，請大家在紙條上寫下**「你最要好的好朋友座號」**。
規定每個人只能寫一個人的座號，而且全班寫出來的所有座號，剛好全班每個人都被寫到恰好一次！

遊戲開始時，老師請 0 號同學把紙條傳給他寫的好朋友；接著那位好朋友再把紙條傳給他的好朋友……
你會驚奇地發現：這張紙條傳著傳著，**最終一定會傳回一開始發動傳話的那群人手中，形成一個封閉的循環鏈**！
這個封閉的圈子，在題目中就稱為一個「**小群體**」。

### 🔄 為什麼一定會形成封閉環？（數學本質探究）
初學者常常會好奇：難道不會有人傳著傳著遇到「死巷子（沒有好友可傳）」？或者傳著傳著走進「分岔路」嗎？
答案是：**絕對不會！** 關鍵就在於題目的強大保證——**「好友名單是 $0 \sim N-1$ 的一個完整排列」**：
1. **無死巷（出度為 1）**：每個人都剛好指定了一位好友，因此任何人隨時都有下一個目標可以前進。
2. **無重疊匯聚（入度為 1）**：每個人剛好只被一位同學指定為好友，絕不會出現「兩個人同時指向同一個人」的匯流現象。
3. **抽屜原理（Pigeonhole Principle）**：班上總共只有有限的 $N$ 個人。當你一步一步順著好友鏈往前走，至多走 $N$ 步，就必定會撞見已經走過的人！

在離散數學中，這種每個人出度為 1、入度為 1 的有向圖，本質上就是一個**「置換（Permutation）」**。
代數學的基本定理告訴我們：**任何一個有限集合的置換，必定能唯一分解為若干個互不相交的簡單循環（Disjoint Cycles）！**

### 💡 APCS 實戰提示：一維串列即是有向圖
在 Python 中，全班座號剛好是從 $0$ 到 $N-1$ 連續編號。
這意味著我們根本不需要使用複雜的物件或鄰接矩陣，只需要一個簡單的**一維串列 `friends`**：
`friends[i]` 的值就代表座號為 `i` 的同學其好朋友是誰。透過陣列索引，我們就能在 $O(1)$ 常數時間內瞬間找到任何人的好友！

````python
# ==============================================================================
# [2] Code 範例：好友指涉串列之資料結構映射示範
# 示範給定好友清單，如何透過串列索引快速查詢每位同學的好友座號
# ==============================================================================

# 模擬官方範例 1 的好友資料：N = 3，好友依序為 0, 2, 1
n = 3
friends = [0, 2, 1]

print("=== 同學與好友指涉對照表 ===")
for i in range(n):
    best_friend = friends[i]
    print(f"座號 {i} 號同學的最要好朋友是：{best_friend} 號")

# 直接索引查詢展示
target = 1
print(f"\n查詢座號 {target} 的好友：friends[{target}] = {friends[target]}")
````

## 15.2.2 有向圖環狀模型與群體追蹤機制

### 🧭 核心概念：順藤摸瓜的「指針推進（Pointer Hopping）」
當我們拿到一個同學的好友名單時，該如何用程式找出他所在的小群體有哪些人呢？
這個過程就像是**「順著好友鏈順藤摸瓜」**：

以官方範例 2 為例：`friends = [4, 7, 2, 9, 6, 0, 8, 1, 5, 3]`
假設我們想從 **0 號同學** 出發，看看他跟誰屬於同一個小圈圈：
1. 起點是 **0 號**；
2. 查詢 0 號的好友：`friends[0]` 是 **4 號**（指針移動到 4）；
3. 查詢 4 號的好友：`friends[4]` 是 **6 號**（指針移動到 6）；
4. 查詢 6 號的好友：`friends[6]` 是 **8 號**（指針移動到 8）；
5. 查詢 8 號的好友：`friends[8]` 是 **5 號**（指針移動到 5）；
6. 查詢 5 號的好友：`friends[5]` 是 **0 號**（哇！又回到了 0 號起點！）。

整個走訪軌跡為：$$0 \to 4 \to 6 \to 8 \to 5 \to 0$$
當指針再次遇到已經出現在這條鏈路上的成員（0 號）時，我們就知道：**這條鏈路已經完美閉合！這個小群體的成員就是 $\{0, 4, 6, 8, 5\}$ 共 5 位同學！**

### ⚙️ 單步前進的黃金核心語法：`curr = friends[curr]`
在程式碼中，我們只需要一個變數 `curr` 來代表「當前指針所在的同學座號」。
每當我們要往他的好友跨出一步時，只需要執行一行神奇的賦值：
```python
curr = friends[curr]
```
這行代碼的意思是：「查詢當前同學 `curr` 的好友是誰，然後把 `curr` 更新為該位好友的座號」。
只要持續重複這行代碼，指針就會如同搭上捷運一般，沿著有向邊一站一站平穩前行！

### 💡 APCS 實戰提示
如果一個環只有 1 個人（自環，例如 2 號的好友就是 2 號），那麼 `curr = friends[2]` 的結果依然是 2。
這表示我們的指針推進模型對於「大環」與「長度為 1 的自環」具有完全一致的通用性，不需要為自環撰寫額外的特殊分支！

````python
# ==============================================================================
# [2] Code 範例：單一循環鏈的手動追蹤模擬
# 示範從指定起點出發，使用 curr = friends[curr] 一步步走訪整個封閉環
# ==============================================================================

friends = [4, 7, 2, 9, 6, 0, 8, 1, 5, 3]

# 指定起點同學為 0 號
start_node = 0
curr = start_node
path = [curr]

print(f"從 {start_node} 號同學啟動好友鏈追蹤：")
# 先走一步
curr = friends[curr]
while curr != start_node:
    path.append(curr)
    curr = friends[curr]  # 推進至下一位好友

print("完整小群體成員路徑：", " -> ".join(map(str, path)), f" -> 回到 {start_node}")
print(f"此小群體成員清單：{path}，人數規模：{len(path)} 人")
````

````python
# ==============================================================================
# [3] Code 填空題：指針單步前進與路徑收集
# 任務：請將下方程式碼中的 ___ 替換為正確的推進語法，完成下一位好友的追蹤
# ==============================================================================

sample_friends = [2, 0, 1]

current_student = 1
# 提示：下一位同學的座號應為 sample_friends[current_student]
# next_student = sample_friends[___]

# 請在此處完成填空並執行：
next_student = sample_friends[current_student]  # 填空目標

print(f"同學 {current_student} 的下一位好友是：{next_student}")
````

````python
# ==============================================================================
# [4] Code 練習題：追蹤指定起點的小群體
# 任務說明：給定兩組好友關係與指定起點，撰寫程式走訪並印出該群體的所有成員
#
# 【公開測試資料 1】
# friends = [1, 2, 0], start = 0
# 預期輸出群體成員：[0, 1, 2]
#
# 【公開測試資料 2】
# friends = [0, 2, 1], start = 1
# 預期輸出群體成員：[1, 2]
# ==============================================================================

def trace_cycle(f_list, start):
    members = [start]
    p = f_list[start]
    while p != start:
        members.append(p)
        p = f_list[p]
    return members

# 測試資料 1
print("測試 1 群體成員：", trace_cycle([1, 2, 0], 0))

# 測試資料 2
print("測試 2 群體成員：", trace_cycle([0, 2, 1], 1))
````

````python
# ==============================================================================
# [5] Code 挑戰題：自環追蹤的邊界相容性
# 任務說明：請測試當 start 節點本身就是自環時（例如 friends = [0, 1, 2], start = 1），
# 你的追蹤邏輯是否能正確回報成員為 [1] 且不會陷入無窮迴圈？
# （本題為自由挑戰題，無公開測資，請自行設計程式碼驗證）
# ==============================================================================

friends_with_self_loop = [0, 1, 2]
start_target = 1

# 檢驗自環走訪：
cycle_members = [start_target]
nxt = friends_with_self_loop[start_target]
while nxt != start_target:
    cycle_members.append(nxt)
    nxt = friends_with_self_loop[nxt]

print(f"自環節點 {start_target} 追蹤結果：{cycle_members}，成功相容自環！")
````

## 15.2.3 布林拜訪陣列 `visited` 防止重複計數

### 🚩 考場致命天坑：若無標記，群體將被瘋狂重複計數！
在上一小節中，我們學會了如何從 0 號出發找出一個小群體 $\{0, 4, 6, 8, 5\}$。
但請試想一下：全班有 $N$ 位同學，我們勢必得寫一個迴圈遍歷座號 $0, 1, 2, \dots, N-1$ 來清點群體。
如果我們沒有任何紀錄標記：
* 輪到 0 號時，我們發現了這個 5 人群體；
* 輪到 4 號時，我們又從 4 號走了一遍，再次把這個群體當作新群體！
* 輪到 5 號、6 號、8 號時，又各自重新算了一次！
結果原本只有「1 個」小群體，卻被活生生重複計算了 5 次！最終輸出答案將錯得一塌糊塗。
更致命的是，每一次遍歷都要花費時間，如果每個人都重複走一次，總時間將從 $O(N)$ 暴增為 $O(N^2)$，在 $N = 50000$ 時直接慘遭超時（TLE）！

### 🛡️ 解題救星：記憶體置物櫃中的「點名簿」——布林陣列 `visited`
要徹底解決重複計算與無限迴圈，演算法中最經典、最犀利的武器就是**「拜訪標記陣列（Visited Array）」**！
想像老師手裡拿著一張全班的點名單：
```python
visited = [False] * n
```
1. **初始狀態**：一開始大家都在位置上，點名單上全班 $N$ 個人通通標記為 `False`（代表尚未被清點）。
2. **蓋章標記**：當我們在追蹤某個小群體時，每遇到一位同學 `curr`，就立刻在點名簿上蓋章：
   ```python
   visited[curr] = True
   ```
3. **防禦略過**：當外層迴圈檢查到同學 `i` 時，只要看到 `visited[i] == True`，就知道這位同學早在先前的某個群體中被登記過了，立刻**直接跳過（Skip）**！

### 💡 APCS 實戰提示：布林串列的極致效能
在 Python 中，使用 `visited = [False] * n` 的索引查詢 `visited[i]` 是**純粹的 $O(1)$ 常數時間**！
相較於每次在串列中用 `if x in visited_list`（需要 $O(N)$ 線性搜尋），布林陣列的直接索引法快了數萬倍，這正是 APCS 中級題獲取滿分的核心必備基本功！

````python
# ==============================================================================
# [2] Code 範例：布林拜訪陣列的初始化與標記機制示範
# 示範宣告長度為 N 的布林串列，以及如何透過索引進行 O(1) 極速標記與查驗
# ==============================================================================

n = 5
# 步驟 1：建立全班點名簿，初始全員為 False
visited = [False] * n
print("初始拜訪點名簿狀態：", visited)

# 步驟 2：假設我們探訪了 1 號同學與 3 號同學，進行蓋章標記
visited[1] = True
visited[3] = True
print("標記 1 號與 3 號後狀態：", visited)

# 步驟 3：模擬外層巡檢邏輯
for i in range(n):
    if not visited[i]:
        print(f"座號 {i}：尚未被標記，發現全新成員！")
    else:
        print(f"座號 {i}：已被標記過，直接略過。")
````

````python
# ==============================================================================
# [3] Code 填空題：布林點名簿初始化與未拜訪判定
# 任務：請將下方程式碼中的 ___ 替換為正確的 Python 語法，完成布林標記宣告與判斷
# ==============================================================================

total_students = 6

# 提示：長度為 n 的布林全 False 串列語法為 [False] * total_students
# check_list = [False] * ___
# if ___ check_list[0]:

# 請在此處完成填空並執行：
check_list = [False] * total_students  # 填空目標：填入 total_students

if not check_list[0]:                 # 填空目標：填入 not
    print("0 號同學確實尚未被拜訪！")
````

````python
# ==============================================================================
# [4] Code 練習題：批量標記與未拜訪成員計數
# 任務說明：給定全班人數與一組剛探訪完的群體名單，請將該群體成員標記為 True，
# 並統計全班還有幾個人尚未被拜訪。
#
# 【公開測試資料 1】
# n = 5, group = [0, 4]
# 預期輸出未拜訪人數：3 人
#
# 【公開測試資料 2】
# n = 10, group = [1, 3, 5, 7, 9]
# 預期輸出未拜訪人數：5 人
# ==============================================================================

def count_unvisited(n, group):
    vis = [False] * n
    for member in group:
        vis[member] = True
    # 統計 False 的數量
    return vis.count(False)

# 測試資料 1
print("測試 1 剩餘未拜訪人數：", count_unvisited(5, [0, 4]))

# 測試資料 2
print("測試 2 剩餘未拜訪人數：", count_unvisited(10, [1, 3, 5, 7, 9]))
````

## 15.2.4 `while` 迴圈循線追蹤好友鏈直到形成封閉環

### 🔄 核心演算法引擎：`while` 迴圈與自動閉合停機
現在，我們把第二階梯的「指針推進 `curr = friends[curr]`」與第三階梯的「布林點名簿 `visited`」完美結合起來！
這就是清點單一小群體的核心引擎——**`while` 迴圈追蹤法**。

請觀察這段無比精巧的核心程式碼：
```python
while not visited[curr]:
    visited[curr] = True
    curr = friends[curr]
```

### 🔍 迴圈運作四步曲剖析
讓我們以微觀視角追蹤這個迴圈每一步在做什麼：
1. **入口檢查**：`while not visited[curr]` —— 檢查當前同學 `curr` 是否已經被蓋過章？如果還沒蓋過章，進入迴圈！
2. **立即蓋章**：`visited[curr] = True` —— 立刻在點名簿上把 `curr` 標記為「已拜訪」，宣誓他已經歸入當前群體。
3. **邁步推進**：`curr = friends[curr]` —— 指針順著箭頭走向好友座號。
4. **優雅停機（自動跳出）**：
   - 當指針沿著環走了一圈，走向起點同學時；
   - 回到 `while not visited[curr]` 檢查點，發現起點同學早在第 1 步就已經被標記為 `True` 了！
   - `not visited[curr]` 的結果變成 `False`，迴圈瞬間**自動停機終止**！

### ⚠️ 考場致命細節：標記與推進順序絕對不能顛倒！
很多同學在考場上緊張，不小心寫成：
```python
# ❌ 錯誤示範！切勿模仿！
while not visited[curr]:
    curr = friends[curr]  # 先往前走了一步！
    visited[curr] = True
```
這樣寫會發生什麼慘劇？**最一開始的起點同學根本沒被標記到！**
當指針繞了一圈回到起點時，因為起點依然是 `False`，迴圈會繼續無窮無盡地轉下去，造成令人絕望的 **Time Limit Exceeded（超時死循環）**！
請務必牢記口訣：**「先蓋章，再邁步！」**

````python
# ==============================================================================
# [2] Code 範例：while 迴圈追蹤並標記單一小群體示範
# 示範標準的「先蓋章、再邁步」流程，觀察 visited 陣列與指針的精準配合
# ==============================================================================

friends = [4, 7, 2, 9, 6, 0, 8, 1, 5, 3]
n = len(friends)
visited = [False] * n

# 假設我們從 0 號同學啟動搜尋
curr = 0
group_members = []

print("啟動 while 迴圈追蹤：")
while not visited[curr]:
    visited[curr] = True         # 1. 先蓋章！
    group_members.append(curr)   # 收集成員座號
    curr = friends[curr]         # 2. 再邁步！

print(f"成功辨識出一個小群體：{group_members}")
print(f"此時全班點名簿狀態：{visited}")
print(f"終止時 curr 指標停在：{curr} 號（撞上已標記之環成員，優雅停機）")
````

````python
# ==============================================================================
# [3] Code 填空題：while 迴圈核心架構填空
# 任務：請將下方程式碼中的 ___ 替換為正確的變數與條件，完成循環閉合
# ==============================================================================

f = [1, 2, 0]  # 0->1->2->0
vis = [False] * 3
start = 0
c = start

# 提示：迴圈條件為 while not vis[c]:，推進為 c = f[c]
# while not ___:
#     vis[c] = True
#     c = ___

# 請在此處完成填空並執行：
while not vis[c]:     # 填空目標：填入 vis[c]
    vis[c] = True
    c = f[c]          # 填空目標：填入 f[c]

print(f"走訪結束，vis 標記結果：{vis}")
````

````python
# ==============================================================================
# [4] Code 練習題：多個起點分別觸發 while 追蹤
# 任務說明：給定 friends = [0, 2, 1]（包含一個自環 0 與一個互選環 1, 2），
# 請在同一張 visited 表上記錄，先後從 0 與 1 觸發追蹤，印出每次找到的成員
#
# 【公開測試資料 1】
# friends = [0, 2, 1], 起點依序為 0, 1
# 預期輸出：第 1 群=[0]，第 2 群=[1, 2]
#
# 【公開測試資料 2】
# friends = [1, 0, 3, 2], 起點依序為 0, 2
# 預期輸出：第 1 群=[0, 1]，第 2 群=[2, 3]
# ==============================================================================

def test_multi_trace(f_arr, starts):
    vis = [False] * len(f_arr)
    result = []
    for st in starts:
        curr_grp = []
        c = st
        while not vis[c]:
            vis[c] = True
            curr_grp.append(c)
            c = f_arr[c]
        result.append(curr_grp)
    return result

# 測試資料 1
print("測試 1 群體歷程：", test_multi_trace([0, 2, 1], [0, 1]))

# 測試資料 2
print("測試 2 群體歷程：", test_multi_trace([1, 0, 3, 2], [0, 2]))
````

## 15.2.5 全體成員枚舉、群體總數計數與完整 AC 程式

### 🏆 雙層結構的終極拼裝：外層 `for` + 內層 `while`
恭喜各位冒險者！我們已經備齊了所有零件，現在來到手把手拼裝滿分程式碼的榮耀時刻！

整體演算法的架構無比簡練優雅：
1. **設定群體計數器**：`group_count = 0`，用來累加找到的小群體數量。
2. **外層巡邏員（`for i in range(n)`）**：巡視全班每一位同學的座號 $i$。
3. **先鋒哨兵（`if not visited[i]`）**：
   - 如果同學 $i$ 的標記是 `True`，表示他早就在之前某個圈子裡被清點過了，直接忽視！
   - 如果同學 $i$ 的標記是 `False`，**這意味著我們發現了一個前所未見的全新小群體！**
4. **群體計數與全面收網**：
   - 我們立刻讓 `group_count += 1`！
   - 緊接著以 $i$ 作為起點，啟動內層的 `while not visited[curr]` 引擎，將這個新小圈圈裡的所有好朋友一網打盡，通通蓋上 `visited = True`！

當外層 `for` 迴圈跑完 $0$ 到 $N-1$ 時，全班所有同學都必定且恰好被蓋過一次章，而 `group_count` 記錄的數值，就是最精確無誤的小群體總數！

### 🧩 完整核心模板代碼透視
```python
visited = [False] * n
ans = 0
for i in range(n):
    if not visited[i]:
        ans += 1
        curr = i
        while not visited[curr]:
            visited[curr] = True
            curr = friends[curr]
print(ans)
```
整段核心邏輯僅僅 10 行！沒有任何多餘的花巧，卻展現了演算法無與倫比的俐落與美感！

````python
# ==============================================================================
# [2] Code 範例：完整版小群體演算法實作展示
# 示範外層 for 與內層 while 完美協作，精確求解官方範例 1 與範例 2
# ==============================================================================

def solve_small_groups(n, friends):
    """
    傳入全班人數 n 與好友串列 friends，回傳獨立小群體數量
    """
    visited = [False] * n
    count = 0
    
    for i in range(n):
        if not visited[i]:
            count += 1          # 發現全新小群體
            curr = i
            while not visited[curr]:
                visited[curr] = True
                curr = friends[curr]
                
    return count

# 驗證官方範例 1：n = 3, friends = [0, 2, 1]
ans1 = solve_small_groups(3, [0, 2, 1])
print(f"範例 1 計算結果：{ans1} 個小群體（預期為 2）")

# 驗證官方範例 2：n = 10, friends = [4, 7, 2, 9, 6, 0, 8, 1, 5, 3]
ans2 = solve_small_groups(10, [4, 7, 2, 9, 6, 0, 8, 1, 5, 3])
print(f"範例 2 計算結果：{ans2} 個小群體（預期為 4）")
````

````python
# ==============================================================================
# [3] Code 填空題：核心主架構組裝填空
# 任務：請將下方程式碼中的 ___ 替換為正確的關鍵字與變數，完成全班小群體計數
# ==============================================================================

demo_n = 4
demo_friends = [1, 2, 3, 0]  # 單一大環

vis = [False] * demo_n
groups = 0

# 提示：外層巡訪 range(demo_n)，未走過則 groups += 1
# for i in range(___):
#     if not vis[i]:
#         groups += ___

# 請在此處完成填空並執行：
for i in range(demo_n):      # 填空目標：填入 demo_n
    if not vis[i]:
        groups += 1         # 填空目標：填入 1
        c = i
        while not vis[c]:
            vis[c] = True
            c = demo_friends[c]

print(f"計算結果：{groups} 個群體（預期為 1）")
````

````python
# ==============================================================================
# [4] Code 練習題：自主執行兩組公開測資
# 任務說明：請撰寫完整的解題程式，計算並印出以下兩組測資的小群體數量
#
# 【公開測試資料 1】
# n = 4, friends = [1, 0, 3, 2]
# 預期輸出：2
#
# 【公開測試資料 2】
# n = 6, friends = [1, 2, 0, 4, 5, 3]
# 預期輸出：2
# ==============================================================================

def calculate_groups(n, f_arr):
    v = [False] * n
    cnt = 0
    for i in range(n):
        if not v[i]:
            cnt += 1
            curr = i
            while not v[curr]:
                v[curr] = True
                curr = f_arr[curr]
    return cnt

# 測試資料 1 執行
print("測試資料 1 輸出：", calculate_groups(4, [1, 0, 3, 2]))

# 測試資料 2 執行
print("測試資料 2 輸出：", calculate_groups(6, [1, 2, 0, 4, 5, 3]))
````

````python
# ==============================================================================
# [5] Code 挑戰題：統計「人數最多的小群體」之規模
# 任務說明：除了計算有幾個小群體之外，有些進階題目會詢問：
# 「全班規模最大（包含最多人）的小群體，裡面共有幾個人？」
# 請修改你的程式碼，同時回傳小群體數量與最大群體的人數！
# （本題為自由挑戰題，無公開測資，請自行構思驗證）
# ==============================================================================

challenge_friends = [4, 7, 2, 9, 6, 0, 8, 1, 5, 3]
c_n = len(challenge_friends)
c_vis = [False] * c_n

total_groups = 0
max_group_size = 0

for i in range(c_n):
    if not c_vis[i]:
        total_groups += 1
        current_size = 0
        p = i
        while not c_vis[p]:
            c_vis[p] = True
            current_size += 1
            p = challenge_friends[p]
        max_group_size = max(max_group_size, current_size)

print(f"群體總數：{total_groups}，最大群體規模：{max_group_size} 人（預期為 5）")
````

## 15.2.6 $O(N)$ 線性時間複雜度分析與孤立自環特例

### ⚡ 複雜度深入解析：為什麼看似雙層迴圈，卻是極速 $O(N)$？
很多剛接觸演算法的同學看到外層有 `for` 迴圈、內層又有 `while` 迴圈，常常直覺以為這是 $O(N^2)$ 的算法。
**這是天大的誤會！** 本解法的時間複雜度是貨真價實、極致無比的 **$O(N)$ 線性時間**！

**數學證明：**
1. 全班一共只有 $N$ 位同學。
2. 每位同學在整個程式執行過程中，**只會在內層 `while` 迴圈被標記 `visited[curr] = True` 恰好一次**！
3. 一旦被標記過，他再也不會進入內層 `while` 迴圈。
4. 外層 `for` 迴圈執行 $N$ 次判斷，而內層 `while` 迴圈在整支程式中總共執行的步數**總和嚴格等於 $N$**！
5. 總操作次數約為 $N + N = 2N$，因此時間複雜度為精確的 $\mathcal{O}(N)$！

面對 APCS 題目的上限 $N = 50000$，Python 執行 5 萬次走訪只需要約 **0.02 秒**，距離 1.0 秒的時間限制有著高達 50 倍的寬裕安全餘裕！

---

### 💣 考場 4 大致命 WA / TLE 地雷全面排查

#### 💥 地雷一：使用遞迴（DFS 深度優先搜尋）導致 `RecursionError` 崩潰
* **陷阱情境**：有些學過圖論的同學喜歡寫 `def dfs(u): ... dfs(friends[u])`。
* **致命後果**：Python 系統預設的遞迴深度限制僅有 1000 層！當測資出現一個包含 50000 人的巨大單一環時，遞迴呼叫到第 1000 層便會直接拋出 `RecursionError: maximum recursion depth exceeded`，在評判系統上獲得慘烈的 **RE（執行時期錯誤）**！
* **避坑法寶**：本單元傳授的 `while` 迴圈迭代法使用扁平迴圈，完全不受遞迴深度限制，這也是征服中級題的最佳實踐。

#### 💥 地雷二：使用 `curr in list` 導致演算法暴跌為 $O(N^2)$ TLE
* **陷阱情境**：使用 `visited_list = []`，每走一步檢查 `if curr in visited_list:`。
* **致命後果**：在串列中進行 `in` 運算是 $O(K)$ 線性掃描。當總步數為 $N$ 時，總時間會飆升至 $O(N^2)$，運算量高達 $2.5 \times 10^9$ 次，保證超時（TLE）！
* **避坑法寶**：務必使用預先配置的布林陣列 `visited = [False] * n`，實現真正 $O(1)$ 索引查驗！

#### 💥 地雷三：漏掉孤立自環（自己是自己的好朋友）
* **陷阱情境**：誤以為群體至少要有 2 個人，或是寫了特殊的檢查把 `friends[i] == i` 當作無效資料略過。
* **避坑法寶**：題目明確說明「自己可以是自己最要好的朋友」，自環本身就是一個完整合法的 1 人小群體，必須完整計入！

#### 💥 地雷四：前進指標與蓋章順序顛倒引發無窮迴圈
* **陷阱情境**：在 `while` 迴圈內部先執行 `curr = friends[curr]` 才標記 `visited[curr] = True`。
* **致命後果**：起點漏標，繞回起點時無法觸發停機條件，程式永遠卡死在迴圈中！

````python
# ==============================================================================
# [2] Code 範例：極限大測資效能壓力測試（N = 50,000）
# 示範生成長達 50000 人的單一巨大環，驗證 O(N) 線性演算法之毫秒級極速表現
# ==============================================================================

import time

scale_n = 50000
# 構造一個首尾相連的巨大環：0->1->2->...->49999->0
big_friends = [(i + 1) % scale_n for i in range(scale_n)]

start_time = time.time()

# 執行核心演算法
v = [False] * scale_n
groups_count = 0
for i in range(scale_n):
    if not v[i]:
        groups_count += 1
        c = i
        while not v[c]:
            v[c] = True
            c = big_friends[c]

elapsed = time.time() - start_time

print(f"大測資規模：N = {scale_n}")
print(f"計算結果：{groups_count} 個小群體（預期為 1）")
print(f"耗費時間：{elapsed:.4f} 秒（遠低於 APCS 考場 1.0 秒上限，效能極其頂尖！）")
````

````python
# ==============================================================================
# [3] Code 填空題：自環與大環邊界防護
# 任務：請將下方程式碼中的 ___ 替換為正確的比較條件，檢驗演算法對自環的相容性
# ==============================================================================

pure_self_loops = [0, 1, 2, 3]  # 全班每人都是自環
k = len(pure_self_loops)
vis_k = [False] * k
total_loops = 0

for i in range(k):
    if not vis_k[i]:
        total_loops += 1
        cur = i
        while not vis_k[cur]:
            vis_k[cur] = True
            cur = pure_self_loops[cur]

# 提示：預期群體總數應等於全班人數 k
# is_all_counted = (total_loops == ___)

# 請在此處完成填空並執行：
is_all_counted = (total_loops == k)  # 填空目標：填入 k

print(f"自環全數計數成功？{is_all_counted}（總群數 = {total_loops}）")
````

````python
# ==============================================================================
# [4] Code 練習題：兩大極端邊界測試案例驗證
# 任務說明：撰寫程式自動化檢驗兩大極端測資：
# 1. 全員自環（N=5, friends=[0, 1, 2, 3, 4]，預期 5）
# 2. 全員一大環（N=5, friends=[1, 2, 3, 4, 0]，預期 1）
# ==============================================================================

def test_extreme_case(f_list):
    n_len = len(f_list)
    visited_arr = [False] * n_len
    ans = 0
    for i in range(n_len):
        if not visited_arr[i]:
            ans += 1
            p = i
            while not visited_arr[p]:
                visited_arr[p] = True
                p = f_list[p]
    return ans

# 極端測資 1：全員自環
res1 = test_extreme_case([0, 1, 2, 3, 4])
print(f"極端測資 1（全員自環）結果：{res1}（預期為 5）")

# 極端測資 2：全員一大環
res2 = test_extreme_case([1, 2, 3, 4, 0])
print(f"極端測資 2（全員一大環）結果：{res2}（預期為 1）")
````

````python
# ==============================================================================
# [5] Code 挑戰題：群體成員守恆定律檢驗
# 任務說明：在任何置換有向圖中，所有獨立小群體的成員數量加總，
# 必定嚴格等於全班總人數 N（這在數學上稱為分劃 Partition）。
# 請撰寫一段驗證腳本，走訪任意好友清單，統計所有小群體的人數和，並確認其是否等於 N！
# （本題為自由挑戰題，無公開測資，請自行設計程式碼驗證）
# ==============================================================================

check_friends = [4, 7, 2, 9, 6, 0, 8, 1, 5, 3]
total_n = len(check_friends)
v_check = [False] * total_n
group_sizes = []

for i in range(total_n):
    if not v_check[i]:
        sz = 0
        ptr = i
        while not v_check[ptr]:
            v_check[ptr] = True
            sz += 1
            ptr = check_friends[ptr]
        group_sizes.append(sz)

print("各小群體人數統計：", group_sizes)
print(f"所有群體人數總和：{sum(group_sizes)}，是否嚴格等於全班人數 {total_n}？{sum(group_sizes) == total_n}")
````

### 📝 版本一：APCS 官方實作考場專用版（極簡單筆輸入）

* **適用情境**：APCS 正式考試環境（保證單筆測試資料，第一行為 $N$，第二行為好友名單）。
* **設計核心**：全篇僅 12 行，純粹使用內建 `input()` 與串列，沒有引入任何模組，打字極速，結構極端清晰！
* **互動測試指引**：在 Colab 點擊下方播放鍵執行後，請在輸入框依序貼上兩行測資（例如先輸入 `3`，再輸入 `0 2 1`），即可立即印出滿分答案！

````python
# ==============================================================================
# 📝 版本一：APCS 官方實作考場專用版（極簡 12 行，附逐行詳盡繁體中文註解）
# 適用情境：APCS 正式考場單筆輸入保證
# ==============================================================================

# 第 1 行：讀取全班總人數 N
n = int(input())

# 第 2 行：讀取 N 個整數，建立好友關係串列（friends[i] 代表 i 號同學的好友）
friends = list(map(int, input().split()))

# 初始化布林拜訪陣列，記錄每位同學是否已被歸入某個小群體（初始全為 False）
visited = [False] * n
group_count = 0  # 記錄小群體總數量

# 遍歷全班每一位同學
for i in range(n):
    # 若此同學尚未被拜訪，代表發現了一個全新的小群體
    if not visited[i]:
        group_count += 1  # 小群體數量加 1
        curr = i
        # 順著好友鏈不斷前進，將同群的所有成員通通標記為 True，直到繞回已拜訪節點
        while not visited[curr]:
            visited[curr] = True
            curr = friends[curr]

# 輸出全班形成的小群體總數
print(group_count)
````

### 🌐 版本二：ZeroJudge 線上評判萬用 AC 版（支援多筆測資 EOF）

* **適用情境**：ZeroJudge c291 線上評判系統（支援連續多筆測試資料自動處理至檔案結尾 EOF）。

````python
# ==============================================================================
# 🌐 版本二：ZeroJudge 線上評判萬用 AC 版（支援多筆測資 EOF 循環）
# 適用情境：ZeroJudge c291 / 線上解題系統（支援連續多測資自動處理至 EOF）
# ==============================================================================

import sys

# ------------------------------------------------------------------------------
# 核心解題模組：傳入學生數 n 與好友清單 friends，回傳獨立小群體數量
# ------------------------------------------------------------------------------
def solve_c291_small_groups(n, friends):
    """
    使用一維布林陣列在 O(N) 線性時間內求解不相交循環群體數量
    """
    visited = [False] * n
    groups = 0
    
    for i in range(n):
        if not visited[i]:
            groups += 1
            curr = i
            while not visited[curr]:
                visited[curr] = True
                curr = friends[curr]
                
    return groups


# ------------------------------------------------------------------------------
# 🎯 【ZeroJudge 複製提交專區】：
# 若要提交至 ZeroJudge c291，可複製以下主程式送出：
# ------------------------------------------------------------------------------
def run_zerojudge_submission():
    """ZeroJudge 萬用輸入解析主流程（支援多筆測資直到 EOF）"""
    tokens = sys.stdin.read().split()
    if not tokens:
        return
    
    idx = 0
    while idx < len(tokens):
        n = int(tokens[idx])
        idx += 1
        friends = [int(tokens[idx + j]) for j in range(n)]
        idx += n
        print(solve_c291_small_groups(n, friends))


# ------------------------------------------------------------------------------
# 🧪 【Colab 本地自動化測試檢驗展示】：
# 點擊本儲存格播放鍵，將自動以官方四大範例及極限測資進行檢驗！
# ------------------------------------------------------------------------------
print("=" * 65)
print("🚀 正在執行 APCS c291 小群體 官方範例與邊界測資本地檢驗...")
print("=" * 65)

test_cases = [
    {
        "name": "官方範例 1（自環 + 雙人環）",
        "n": 3,
        "friends": [0, 2, 1],
        "expected": 2
    },
    {
        "name": "官方範例 2（多種規模環組合）",
        "n": 10,
        "friends": [4, 7, 2, 9, 6, 0, 8, 1, 5, 3],
        "expected": 4
    },
    {
        "name": "官方範例 3（單一巨大環）",
        "n": 4,
        "friends": [1, 2, 3, 0],
        "expected": 1
    },
    {
        "name": "官方範例 4（全員各自自環）",
        "n": 5,
        "friends": [0, 1, 2, 3, 4],
        "expected": 5
    },
    {
        "name": "極限測資：單人自環 (N = 1)",
        "n": 1,
        "friends": [0],
        "expected": 1
    }
]

all_passed = True
for i, tc in enumerate(test_cases, 1):
    actual = solve_c291_small_groups(tc["n"], tc["friends"])
    passed = (actual == tc["expected"])
    status = "✅ PASS" if passed else "❌ FAIL"
    if not passed:
        all_passed = False
    print(f"【測試案例 {i}】{tc['name']}")
    print(f"  人數：N = {tc['n']}，好友：{tc['friends']}")
    print(f"  預期輸出：{tc['expected']} | 實際輸出：{actual} | 狀態：{status}\n")

print("=" * 65)
if all_passed:
    print("🎉 恭喜！本地 5 組測資檢驗全數通過（100% AC）！")
    print("學生可直接複製上方【ZeroJudge 複製提交專區】代碼至 ZeroJudge c291 取得滿分！")
else:
    print("⚠️ 檢驗未完全通過，請檢查邏輯！")
print("=" * 65)
````
