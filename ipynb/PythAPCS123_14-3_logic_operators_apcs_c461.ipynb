### 📝 題目描述（Problem Description）
在邏輯電路與電腦程式中，最基礎的二元邏輯運算子包含 **AND（且）**、**OR（或）** 與 **XOR（互斥或）**。
在二元邏輯中，數值只代表兩種狀態：**真（True）** 或 **偽（False）**。
在傳統 C/Python 語言標準中：
- 數值 $0$ 代表 **False（偽）**。
- **任何非零整數**（例如 $1, 2, 3, \dots, 10000$）一律代表 **True（真）**。

三大邏輯運算子的運作規則（真值表）如下：

```text
   【三大二元邏輯運算子真值表】
    輸入 A    輸入 B   │   A AND B   │   A OR B    │   A XOR B
   ──────────────────┼─────────────┼─────────────┼─────────────
    False    False   │    False    │    False    │    False
    False    True    │    False    │    True     │    True
    True     False   │    False    │    True     │    True
    True     True    │    True     │    True     │    False
```

現在，題目給定兩個非負整數輸入 $a, b$ 以及一個預期輸出結果 $c$（$c$ 只會是 $0$ 或 $1$）。
請你寫一個程式，反查是哪一種（或哪幾種）邏輯運算子能使此運算成立。

### 14.3.1 💡 題意解析與逆向黑盒子偵探推理

在傳統的程式練習中，我們通常是「給定運算子，計算結果」：
- 例如：計算 `True and False` 會得到 `False`。
而這道題目的思考方向剛好相反——它是一場**「逆向偵探推理」**！

想像眼前有一個封閉的黑盒子密碼鎖：
- 左邊有兩個輸入端子 $a$ 與 $b$；
- 右邊有一個輸出端子 $c$；
- 內部只可能安裝了 **AND、OR、XOR** 三種邏輯晶片之一（或內部接法同時符合多種晶片功能）。

```text
   輸入 a ───┐   ┌───────────────┐
             ├───┤  ? 邏輯黑盒子 ? ├───> 輸出 c
   輸入 b ───┘   └───────────────┘
                  (AND / OR / XOR)
```

#### 解題偵探的核心任務：
我們的任務不是去「猜」，而是去**「逐一檢驗」**：
1. 如果盒子裡裝的是 AND 晶片，它算出來的結果跟 $c$ 一樣嗎？若一樣，AND 就是嫌疑人之一，記下來！
2. 如果盒子裡裝的是 OR 晶片，算出來的結果跟 $c$ 一樣嗎？若一樣，OR 也是嫌疑人，記下來！
3. 如果盒子裡裝的是 XOR 晶片，算出來的結果跟 $c$ 一樣嗎？若一樣，XOR 也記下來！
4. 如果三個晶片算出來的結果**沒有一個跟 $c$ 相符**，說明這個黑盒子故障了，宣告 **`IMPOSSIBLE`**！

輸出順序規定必須是 **先 AND、次 OR、後 XOR**，且全為大寫英文字母。理清這個逆向檢驗模型，我們就能條理分明地著手解題！

````python
# ==========================================
# [2] Code 範例區：逆向比對思維雛形展示
# ==========================================

# 假設輸入 a=True, b=False, 預期結果 c=True
a_bool = True
b_bool = False
expected_c = True

print("=== 逆向檢驗三大運算子 ===")
# 1. 檢驗 AND
and_res = a_bool and b_bool
print(f"a AND b 運算結果: {and_res} ➔ 與預期 c({expected_c}) 相符嗎？ {and_res == expected_c}")

# 2. 檢驗 OR
or_res = a_bool or b_bool
print(f"a OR b 運算結果:  {or_res} ➔ 與預期 c({expected_c}) 相符嗎？ {or_res == expected_c}")

# 3. 檢驗 XOR (相異為真)
xor_res = a_bool != b_bool
print(f"a XOR b 運算結果: {xor_res} ➔ 與預期 c({expected_c}) 相符嗎？ {xor_res == expected_c}")
````

````python
# ==========================================
# [3] Code 填空題：補齊相符性判定
# 任務說明：請將下方 ___ 替換，比對 and_res 是否等於 expected_c。
# ==========================================

and_res = False
expected_c = False

# 比對兩者是否相等
is_matched = (and_res ___ expected_c)

print(f"AND 運算子成立嗎？ {is_matched}")
# 預期輸出：AND 運算子成立嗎？ True
````

````python
# ==========================================
# [4] Code 練習題：手動驗證單一輸入組合
#
# 【公開測試資料 1】
# 輸入：a=True, b=True, c=True
# 預期輸出：AND成立: True, OR成立: True, XOR成立: False
#
# 【公開測試資料 2】
# 輸入：a=False, b=False, c=True
# 預期輸出：AND成立: False, OR成立: False, XOR成立: False
# ==========================================
# 請在下方撰寫你的程式碼並執行測試：

def test_logic_match(a, b, c):
    and_ok = (a and b) == c
    or_ok = (a or b) == c
    xor_ok = (a != b) == c
    return f"AND成立: {and_ok}, OR成立: {or_ok}, XOR成立: {xor_ok}"

# 執行測試
print("測試 1:", test_logic_match(True, True, True))
print("測試 2:", test_logic_match(False, False, True))
````

### 14.3.2 🛡️ 考場致命第一陷阱：非零即真（Truthiness）規格化

很多同學在解這題時，看了範例一 `0 0 0` 和範例二 `1 1 1`，就想當然地以為：「輸入的數字只有 0 和 1 嘛！」
於是直接在程式碼中寫下：
```python
if (a and b) == c: ...  # 💣 致命地雷！
```
只要遇到**範例三 `3 0 1`** 或隱藏測資 `100 20 1`，立刻拿到大大的 `WA`！

#### 為什麼直接拿整數做 `and` 會爆掉？
因為 Python 的 `and` 與 `or` 是**短路求值運算子（Short-circuit Evaluation）**，它們回傳的**不是布林值，而是運算元本身的數值**！
- 在 Python 中：`3 and 10` 的運算結果是 **`10`**，而不是 `True`，更不是整數 `1`！
- 當 $c = 1$ 時，程式去比對 `(3 and 10) == 1` ➔ 也就是 `10 == 1` ➔ 判定為 **`False`**！
- 但依照題意：$3$ 是非零整數代表真，$10$ 是非零整數代表真，真與真做 AND 本來應該要是真（$c=1$），結果因為型態未規格化而徹底算錯！

#### 🛡️ 銅牆鐵壁防禦心法：布林規格化（Boolean Normalization）
在讀取到整數 $a, b, c$ 之後，**第一步必須立刻將整數轉型為純粹的布林值**！
- 數值為 0 代表 False；任何大於 0 的非零整數代表 True：
```python
ba = (a != 0)  # 或 bool(a)
bb = (b != 0)  # 或 bool(b)
bc = (c != 0)  # 或 bool(c)
```
這樣一來，無論輸入是 `0`、`1` 還是 `10000`，`ba` 與 `bb` 都會被精確收斂為乾淨的 `True` 或 `False`！所有後續的邏輯運算便再無後顧之憂！

````python
# ==========================================
# [2] Code 範例區：整數轉布林規格化對比示範
# ==========================================

a = 3
b = 10
c = 1

print("=== 錯誤示範（直接整數運算）===")
raw_and = a and b
print(f"3 and 10 的直接運算結果是: {raw_and} (型態: {type(raw_and).__name__})")
print(f"raw_and == c (即 {raw_and} == {c}): {raw_and == c} ➔ ❌ 誤判為不成立！")

print("\n=== 正確示範（布林規格化）===")
ba = (a != 0)  # True
bb = (b != 0)  # True
bc = (c != 0)  # True
norm_and = ba and bb
print(f"ba={ba}, bb={bb} ➔ norm_and = {norm_and}")
print(f"norm_and == bc (即 {norm_and} == {bc}): {norm_and == bc} ➔ ✅ 正確判定成立！")
````

````python
# ==========================================
# [3] Code 填空題：補齊布林規格化條件
# 任務說明：請將下方 ___ 替換，使用 (x != 0) 將整數轉為布林值。
# ==========================================

val = 500
# 請將 val 規格化為布林值
is_true = (val ___ 0)

print(f"數值 {val} 規格化結果: {is_true}")
# 預期輸出：數值 500 規格化結果: True
````

````python
# ==========================================
# [4] Code 練習題：將三個輸入整數同步規格化為元組
#
# 【公開測試資料 1】
# 輸入：a=3, b=0, c=1
# 預期輸出：(True, False, True)
#
# 【公開測試資料 2】
# 輸入：a=0, b=0, c=0
# 預期輸出：(False, False, False)
# ==========================================
# 請在下方撰寫你的程式碼並執行測試：

def normalize_inputs(a, b, c):
    return (a != 0, b != 0, c != 0)

# 執行測試
print("測試 1:", normalize_inputs(3, 0, 1))
print("測試 2:", normalize_inputs(0, 0, 0))
````

### 14.3.3 ⚡ 運算子一檢驗模組：AND（且）的真值判定

完成了輸入的布林規格化後，我們開始逐一檢驗三大運算子！
第一位登場的是題目要求最優先檢查的——**AND（邏輯且）**。

#### AND 的邏輯本質：
AND 就像一位極其嚴格的面試官：**「兩項條件必須全部滿足（全為 True），結果才算通過（True）」**！
- `True and True` ➔ `True`
- `True and False` ➔ `False`
- `False and True` ➔ `False`
- `False and False` ➔ `False`

#### 判定 AND 是否能產生結果 $c$：
在已規格化的布林值 `ba, bb, bc` 之下，檢驗式非常純粹：
```python
if (ba and bb) == bc:
    print("AND")
```

⚠️ **括號優先級防禦提醒**：
在 Python 運算子優先級中，`==` 的優先順序其實比 `and` 還要高！
如果寫成 `ba and bb == bc`，Python 實際上會解讀成：`ba and (bb == bc)`！這會導致運算順序完全偏離預期，因此**務必將 `(ba and bb)` 用小括號包裹起來**！

````python
# ==========================================
# [2] Code 範例區：AND 運算子檢驗單元
# ==========================================

def check_AND(ba, bb, bc):
    """檢驗 AND 是否能產生目標結果 bc"""
    return (ba and bb) == bc

# 測試案例
print("=== AND 檢驗測試 ===")
print("範例一 0 0 ➔ 結果 0: AND 成立嗎？", check_AND(False, False, False)) # True
print("範例二 1 1 ➔ 結果 1: AND 成立嗎？", check_AND(True, True, True))     # True
print("範例三 3 0 ➔ 結果 1: AND 成立嗎？", check_AND(True, False, True))    # False
````

### 14.3.4 🔥 運算子二檢驗模組：OR（或）的真值判定

檢查完 AND 後，依序進行第二位檢驗對象——**OR（邏輯或）**。

#### OR 的邏輯本質：
與嚴苛的 AND 相反，OR 是一位寬宏大量的審查官：**「只要有任何一個條件被滿足（至少一者為 True），結果就是通過（True）」**！
只有在雙方全部不符合（兩者皆為 False）時，結果才是 False。
- `True or True` ➔ `True`
- `True or False` ➔ `True`
- `False or True` ➔ `True`
- `False or False` ➔ `False`

#### 判定 OR 是否能產生結果 $c$：
同樣加上保護小括號：
```python
if (ba or bb) == bc:
    print("OR")
```

以範例三 `3 0 1` 為例：
- $a=3$ 為 True，$b=0$ 為 False。
- `True or False` 的計算結果為 `True`。
- 預期輸出 $c=1$ 也是 True，兩者完全吻合！因此 OR 成立，必須輸出 `OR`！

````python
# ==========================================
# [2] Code 範例區：OR 運算子檢驗單元
# ==========================================

def check_OR(ba, bb, bc):
    """檢驗 OR 是否能產生目標結果 bc"""
    return (ba or bb) == bc

# 測試案例
print("=== OR 檢驗測試 ===")
print("範例一 0 0 ➔ 0: OR 成立嗎？", check_OR(False, False, False)) # True (F or F == F)
print("範例二 1 1 ➔ 1: OR 成立嗎？", check_OR(True, True, True))     # True (T or T == T)
print("範例三 3 0 ➔ 1: OR 成立嗎？", check_OR(True, False, True))    # True (T or F == T)
print("範例四 0 0 ➔ 1: OR 成立嗎？", check_OR(False, False, True))   # False (F or F != T)
````

````python
# ==========================================
# [3] Code 填空題：補齊 OR 運算條件
# 任務說明：請將下方 ___ 填入 or 關鍵字，判定 OR 運算結果是否等於 bc。
# ==========================================

ba, bb, bc = True, False, True

is_or_valid = (ba ___ bb) == bc

print(f"OR 成立嗎？ {is_or_valid}")
# 預期輸出：OR 成立嗎？ True
````

````python
# ==========================================
# [4] Code 練習題：同時檢驗 AND 與 OR
#
# 【公開測試資料 1】
# 輸入：a=1, b=1, c=1
# 預期輸出：["AND", "OR"]
#
# 【公開測試資料 2】
# 輸入：a=1, b=0, c=1
# 預期輸出：["OR"]
# ==========================================
# 請在下方撰寫你的程式碼並執行測試：

def check_and_or(a, b, c):
    ba, bb, bc = (a != 0), (b != 0), (c != 0)
    res = []
    if (ba and bb) == bc: res.append("AND")
    if (ba or bb) == bc:  res.append("OR")
    return res

# 執行測試
print("測試 1:", check_and_or(1, 1, 1))
print("測試 2:", check_and_or(1, 0, 1))
````

### 14.3.5 🔀 運算子三檢驗模組：XOR（互斥或）在 Python 中的實作法

現在迎來了全題最有趣、也最常讓初學者感到困惑的第三位運算子——**XOR（互斥或，Exclusive OR）**。

#### XOR 的邏輯本質：
XOR 的核心精神就是四個字：**「相異為真，相同為假」**！
- 兩者狀態**不同**（一真一假，True vs False 或 False vs True）➔ 結果為 **`True`**。
- 兩者狀態**相同**（全真 True vs True，或全假 False vs False）➔ 結果為 **`False`**。

#### 💣 初學者的大困惑：Python 怎麼沒有 `xor` 關鍵字？
沒錯！Python 語法中擁有原生的 `and`、`or`、`not`，但**偏偏就是沒有布林關鍵字 `xor`**！如果你在程式碼中寫下 `ba xor bb`，直譯器會無情報錯 `SyntaxError: invalid syntax`！

#### 💡 演算法高手必學的三大 XOR 破解絕技：
1. **絕技一（邏輯不相等法 `!=`）——最推薦！**
   既然 XOR 就是「兩者相異」，那不就等於**「兩者不相等」**嗎？
   ```python
   xor_result = (ba != bb)
   ```
   - 若 `ba` 是 True、`bb` 是 False ➔ `True != False` 結果為 `True`！完全正確！
   - 若 `ba` 與 `bb` 皆為 True ➔ `True != True` 結果為 `False`！完全正確！
   僅僅一個 `!=` 符號，就完美達成了 XOR 的所有語意！
2. **絕技二（位元 XOR 運算子 `^`）**：
   Python 中的脫字符號 `^` 是位元互斥或運算子。若運算元是布林值，`True ^ False` 也會正確回傳 `True`。
3. **絕技三（加法取餘數）**：
   `(int(ba) + int(bb)) == 1`（一真一假相加剛好等於 1）。

在考場上，使用 `(ba != bb)` 最直觀、最不易打錯！

````python
# ==========================================
# [2] Code 範例區：Python 中實作 XOR 的三種方法
# ==========================================

# 測試四種真值組合
cases = [
    (False, False, False),
    (False, True,  True),
    (True,  False, True),
    (True,  True,  False)
]

print("=== XOR 三大實作法驗證 ===")
for ba, bb, expected in cases:
    m1 = (ba != bb)         # 方法一：不相等法
    m2 = (ba ^ bb)          # 方法二：位元 ^ 法
    m3 = ((int(ba) + int(bb)) == 1) # 方法三：加法計數
    print(f"輸入 ({ba!s:5s}, {bb!s:5s}) ➔ 不相等法={m1!s:5s}, 位元^={m2!s:5s}, 預期={expected!s:5s}")
````

````python
# ==========================================
# [3] Code 填空題：使用 != 完成 XOR 檢驗
# 任務說明：請將下方 ___ 替換為 != 運算子，檢驗 XOR 是否等於 bc。
# ==========================================

ba, bb, bc = True, False, True

# 使用 != 表示 XOR
is_xor_valid = (ba ___ bb) == bc

print(f"XOR 成立嗎？ {is_xor_valid}")
# 預期輸出：XOR 成立嗎？ True
````

````python
# ==========================================
# [4] Code 練習題：檢驗 XOR 運算子單獨模組
#
# 【公開測試資料 1】
# 輸入：a=3, b=0, c=1 (True != False ➔ True == True)
# 預期輸出：XOR
#
# 【公開測試資料 2】
# 輸入：a=1, b=1, c=1 (True != True ➔ False != True)
# 預期輸出：(不印出任何內容)
# ==========================================
# 請在下方撰寫你的程式碼並執行測試：

def test_xor(a, b, c):
    ba, bb, bc = (a != 0), (b != 0), (c != 0)
    if (ba != bb) == bc:
        return "XOR"
    return ""

# 執行測試
print("測試 1:", test_xor(3, 0, 1))
print("測試 2:", test_xor(1, 1, 1))
````

### 14.3.6 🚩 多解平行輸出與無解 `IMPOSSIBLE` 旗標控制

當我們把 AND、OR、XOR 的檢驗單元都準備好之後，如何將它們組裝成主程式？
這裡存在一個初學者最容易掉進去的**「分支結構大陷阱」**！

#### 💣 致命錯誤寫法：使用 `elif` 分支！
```python
# ❌ 錯誤寫法示範：
if (ba and bb) == bc:
    print("AND")
elif (ba or bb) == bc:  # 💣 只要前面 AND 符合，OR 就永遠不會被執行！
    print("OR")
elif (ba != bb) == bc:
    print("XOR")
else:
    print("IMPOSSIBLE")
```
題目明確說明：**可能同時有多個運算子符合**（例如範例一 `0 0 0` 時三者全部符合，範例二 `1 1 1` 時 AND 與 OR 皆符合）！
如果寫了 `elif`，一旦第一個符合，後面通通被略過，直接漏印拿 0 分！

#### 🛡️ 正確架構：三個平行獨立的 `if` + 旗標變數（Flag）
三大運算子的檢查是**完全獨立、互不排斥**的！我們必須使用三個平行的 `if`：
```python
has_solution = False  # 宣告旗標：預設為無解

if (ba and bb) == bc:
    print("AND")
    has_solution = True  # 標記已有解答

if (ba or bb) == bc:
    print("OR")
    has_solution = True

if (ba != bb) == bc:
    print("XOR")
    has_solution = True

# 若三者皆未觸發，說明毫無解答，觸發無解防線！
if not has_solution:
    print("IMPOSSIBLE")
```

旗標變數 `has_solution` 就像一個守護燈，只要有任何一盞燈亮起，最後就不會敲響 `IMPOSSIBLE` 的警報！

````python
# ==========================================
# [2] Code 範例區：平行 if 與布林旗標控制示範
# ==========================================

def solve_logic(a, b, c):
    ba = (a != 0)
    bb = (b != 0)
    bc = (c != 0)
    
    matched = False
    outputs = []
    
    # 1. 檢查 AND
    if (ba and bb) == bc:
        outputs.append("AND")
        matched = True
        
    # 2. 檢查 OR
    if (ba or bb) == bc:
        outputs.append("OR")
        matched = True
        
    # 3. 檢查 XOR
    if (ba != bb) == bc:
        outputs.append("XOR")
        matched = True
        
    # 4. 無解防禦
    if not matched:
        outputs.append("IMPOSSIBLE")
        
    return outputs

# 驗證範例一 (三者皆符合)
print("範例一 0 0 0 輸出:", solve_logic(0, 0, 0)) # ['AND', 'OR', 'XOR']
# 驗證範例四 (無解)
print("範例四 0 0 1 輸出:", solve_logic(0, 0, 1)) # ['IMPOSSIBLE']
````

````python
# ==========================================
# [3] Code 填空題：補齊無解判定條件
# 任務說明：請將下方 ___ 替換，當 matched 為 False 時輸出 IMPOSSIBLE。
# ==========================================

matched = False

# 當完全沒有解時輸出 IMPOSSIBLE
if ___ matched:
    print("IMPOSSIBLE")
# 預期輸出：IMPOSSIBLE
````

````python
# ==========================================
# [4] Code 練習題：測試多解與無解切換
#
# 【公開測試資料 1】
# 輸入：a=1, b=1, c=1
# 預期輸出：AND, OR
#
# 【公開測試資料 2】
# 輸入：a=0, b=0, c=1
# 預期輸出：IMPOSSIBLE
# ==========================================
# 請在下方撰寫你的程式碼並執行測試：

def test_run_flags(a, b, c):
    ba, bb, bc = (a != 0), (b != 0), (c != 0)
    ans = []
    if (ba and bb) == bc: ans.append("AND")
    if (ba or bb) == bc:  ans.append("OR")
    if (ba != bb) == bc:  ans.append("XOR")
    if not ans: ans.append("IMPOSSIBLE")
    return ", ".join(ans)

# 執行測試
print("測試 1:", test_run_flags(1, 1, 1))
print("測試 2:", test_run_flags(0, 0, 1))
````

### 14.3.8 🧪 極端大數壓力測試、優先級括號防禦與考場 WA 地雷排查

在正式收官之前，我們必須對所有可能讓程式失分的**致命邊界地雷**進行最後排查！

#### ⏱️ 時間與空間複雜度分析：
- **時間複雜度**：$O(1)$。整道題目僅有三次布林運算與條件比對，運算次數固定為常數，耗時低於 0.0001 秒，效率已達電腦理論極限！
- **空間複雜度**：$O(1)$。僅耗用三個整數與三個布林變數，記憶體佔用小於 1KB，遠低於 64MB 限制。

````python
# ==========================================
# [2] Code 範例區：全方位壓力與邊界測試套件
# ==========================================

def solve(a, b, c):
    ba, bb, bc = (a != 0), (b != 0), (c != 0)
    ans = []
    if (ba and bb) == bc: ans.append("AND")
    if (ba or bb) == bc:  ans.append("OR")
    if (ba != bb) == bc:  ans.append("XOR")
    if not ans: ans.append("IMPOSSIBLE")
    return ans

# 構造極端邊界測試
test_suite = [
    ("官方範例一 (0 0 0)", 0, 0, 0, ["AND", "OR", "XOR"]),
    ("官方範例二 (1 1 1)", 1, 1, 1, ["AND", "OR"]),
    ("官方範例三 (3 0 1)", 3, 0, 1, ["OR", "XOR"]),
    ("官方範例四 (0 0 1)", 0, 0, 1, ["IMPOSSIBLE"]),
    ("極端大數一 (10000 10000 1)", 10000, 10000, 1, ["AND", "OR"]),
    ("極端大數二 (10000 10000 0)", 10000, 10000, 0, ["XOR"]),
    ("極端大數三 (10000 0 0)", 10000, 0, 0, ["AND"]),
    ("極端大數四 (0 10000 1)", 0, 10000, 1, ["OR", "XOR"])
]

print("=== 極端邊界壓力測試報告 ===")
for name, a, b, c, exp in test_suite:
    act = solve(a, b, c)
    status = "PASS" if act == exp else "FAIL"
    print(f"[{status}] {name:28s} ➔ 預期: {exp!s:20s}, 實際: {act!s:20s}")
````

### 📝 版本一：APCS 官方實作考場專用版（極簡 15 行考場滿分版）

> 📌 **考場攻略**：  
> 在 APCS 正式考場中，題目保證輸入只有單行 3 個整數 $a, b, c$。  
> 我們使用最直觀的 `map(int, input().split())` 讀取，配合 3 個平行 `if` 與旗標變數，以 15 行極速奪取滿分！
> 
> 💡 **在 Colab 中執行時**：點擊下方執行按鈕，請在輸入框中貼上測試資料（如 `3 0 1`）即可看見輸出。

````python
# =====================================================================
# 📝 版本一：APCS 官方實作考場專用版（極簡 15 行滿分版）
# =====================================================================

# 1. 讀取單行三個整數 a, b, c
a, b, c = map(int, input().split())

# 2. 非零即真布林規格化（徹底防禦 a, b 大於 1 之情況）
ba = (a != 0)
bb = (b != 0)
bc = (c != 0)

# 3. 平行獨立檢驗三大運算子
has_solution = False

if (ba and bb) == bc:      # 檢查 AND
    print("AND")
    has_solution = True

if (ba or bb) == bc:       # 檢查 OR
    print("OR")
    has_solution = True

if (ba != bb) == bc:       # 檢查 XOR (相異為真)
    print("XOR")
    has_solution = True

# 4. 無解防線
if not has_solution:
    print("IMPOSSIBLE")
````
