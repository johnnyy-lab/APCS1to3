## 10.4.1 安全取值神器：d.get(key, default) 徹底杜絕 KeyError 崩潰

### 觀念說明
在前面的學習中，我們知道直接使用中括號語法 `d[key]` 取值，若該鍵不存在於字典中，Python 會直接拋出 `KeyError` 例外錯誤並強制終止程式運行。在競賽或實際開發中，輸入資料往往充滿未知數或缺漏，我們不能讓程式輕易崩潰。

此時，`d.get(key, default)` 就是最安全可靠的利器！
1. **如果 `key` 存在**：返回對應的 `value`，第二個參數會被忽略。
2. **如果 `key` 不存在**：不會報錯，而是返回指定的 `default` 預設值。
3. **若省略 `default`**：預設會返回 `None`。

在 APCS 的計數統計或查表題中，利用 `d.get(key, 0)` 更是設定初值的標準寫法，省去了繁瑣的 `if key in d:` 條件判斷，讓程式碼既精簡又穩健。

````python
# 10.4.1 範例：安全查詢學生成績
demo_sc104 = {"Alice": 95, "Bob": 82, "Charlie": 68}

# 查詢存在的鍵
demo_alcsc = demo_sc104.get("Alice", 0)
print("Alice 的成績:", demo_alcsc)

# 查詢不存在的鍵，若不用 get 會引發 KeyError
demo_dvdsc = demo_sc104.get("David", 0)
print("David 的成績（查無此人預設為 0）:", demo_dvdsc)

# 不指定預設值時，預設回傳 None
demo_evasc = demo_sc104.get("Eva")
print("Eva 的成績（預設 None）:", demo_evasc)

# 搭配自訂提示字串
demo_s104 = demo_sc104.get("Frank", "未到考")
print("Frank 狀態:", demo_s104)
````

````python
# 10.4.1 練習題：單字積分查詢器
# 題目說明：
# 某遊戲中每個單字都有對應的分數。給定單字分數字典 wrd_pt，
# 請依序查詢輸入串列 srch_wrds 中的每個單字。
# 若單字在字典中，累加其對應分數；若不在字典中，視為 0 分。
# 最後印出所有查詢單字的總得分。
#
# 【公開測試資料 1】
# 字典: {"cat": 5, "dog": 8, "bird": 12}
# 查詢清單: ["dog", "fish", "cat"]
# 輸出: 總得分: 13
#
# 【公開測試資料 2】
# 字典: {"apple": 10, "banana": 15}
# 查詢清單: ["orange", "grape"]
# 輸出: 總得分: 0

wrd_pt = {"cat": 5, "dog": 8, "bird": 12}
srch_wrds = ["dog", "fish", "cat"]

tot_sc104 = 0
for w in srch_wrds:
    tot_sc104 += wrd_pt.get(w, 0)

print("總得分:", tot_sc104)
````

````python
# 10.4.1 挑戰題：計數器累加實戰（不使用 if-else）
# 題目說明：
# 給定一串字母串列 tokens = ["a", "b", "a", "c", "b", "a"]
# 請建立一個空字典 cnts = {}
# 使用 for 迴圈搭配 cnts.get(ch, 0) + 1 的技巧，統計每個字母出現的次數。
# （不可使用 collections.Counter，純用字典 get() 完成）
# 提示：每次將字母放入字典時，其新數值為當前次數 cnts.get(ch, 0) 再加 1

tokens = ["a", "b", "a", "c", "b", "a"]
cnts = {}

# 請在下方寫出你的解答代碼
for ch in tokens:
    cnts[ch] = cnts.get(ch, 0) + 1

print("字母統計結果:", cnts)
# 預期結果: {'a': 3, 'b': 2, 'c': 1}
````

## 10.4.2 成員檢查：key in d 與 key not in d 的 O(1) 極速存在性判定

### 觀念說明
在很多演算法題目中，我們需要在集合或清單中快速確認「某個資料是否已經出現過」。
- 如果在串列（list）中使用 `x in my_list`，Python 必須從頭到尾一個一個比對，時間複雜度為 $O(N)$。若資料有十萬筆，查詢一次就要花費十萬步，極易超時（TLE）。
- 但如果是在字典（dict）中使用 `key in d` 或 `key not in d`，Python 是利用**雜湊表（Hash Table）**直接計算記憶體位置，平均時間複雜度僅為 **$O(1)$**（常數時間）！不論字典裡有一千筆還是一百萬筆資料，查詢速度幾乎瞬間完成。

### 關鍵注意點
`in` 與 `not in` 運算子直接套用在字典上時，**預設只會檢查「鍵（keys）」**，而**不會**去檢查「值（values）」！這一點一定要牢牢記住。

````python
# 10.4.2 範例：會員權限檢查與 O(1) 判定
demo_vm104 = {
    "user_101": "Gold",
    "user_102": "Silver",
    "user_105": "Platinum"
}

demo_qusr = "user_102"
if demo_qusr in demo_vm104:
    print(demo_qusr, "是 VIP 會員！等級為:", demo_vm104[demo_qusr])
else:
    print(demo_qusr, "不是 VIP 會員。")

# 使用 not in 檢查未註冊者
demo_ngst = "user_999"
if demo_ngst not in demo_vm104:
    print(demo_ngst, "尚未註冊 VIP，引導前往註冊流程。")

# 注意：in 檢查的是 key，而非 value
print("Gold in demo_vm104:", "Gold" in demo_vm104)
````

````python
# 10.4.2 挑戰題：兩數之和（Two Sum）APCS 經典 O(N) 雜湊解法
# 題目說明：
# 給定整數串列 ns104 = [2, 7, 11, 15] 以及目標值 tg104 = 9
# 請使用一個字典 seen = {} 來記錄走訪過的數值（鍵為數值，值為其索引）。
# 遍歷每個元素與其索引，計算 complement = tg104 - num：
# 若 complement in seen，表示找到了兩個數相加等於 tg104，
# 請印出這兩個數的索引對應：(seen[complement], 當前索引)，並結束搜尋。
# 若不存在，則將當前數值與索引存入 seen 中。

ns104 = [2, 7, 11, 15]
tg104 = 9
seen = {}

# 請在下方寫出你的解答代碼
found_pair = None
for idx in range(len(ns104)):
    num = ns104[idx]
    complement = tg104 - num
    if complement in seen:
        found_pair = (seen[complement], idx)
        break
    seen[num] = idx

print("找到目標組合索引:", found_pair)
# 預期結果: 找到目標組合索引: (0, 1)
````

## 10.4.3 鍵走訪：for k in d: 或 d.keys() 循序走訪

### 觀念說明
當我們需要逐一處理字典中的每一筆資料時，最直覺的方式就是**走訪所有的「鍵」**。
在 Python 中，有兩種等價的寫法：
1. `for k in d:` （最常見、最地道的寫法，直接對字典進行迭代）
2. `for k in d.keys():` （顯式調用 `keys()` 方法，語意非常明確）

從 Python 3.7 開始，字典**嚴格保證保留鍵的插入順序（Insertion Order）**，因此你走訪出來的鍵，順序一定與當初加入字典時一致。

### 如何取得對應的值？
在鍵走訪的迴圈內部，只要透過 `d[k]` 就能輕鬆取得該鍵對應的數值。此外，`d.keys()` 返回的是一個「字典視圖（dict_keys）」，它會隨著字典內容動態變化，若需要排序或索引，可以透過 `sorted(d.keys())` 或 `list(d.keys())` 進行轉換。

````python
# 10.4.3 範例：走訪通訊錄與依鍵排序列印
demo_cntc = {"Charlie": "0933-111", "Alice": "0912-345", "Bob": "0922-888"}

print("--- 1. 預設走訪（插入順序）---")
for demo_name in demo_cntc:
    demo_phone = demo_cntc[demo_name]
    print("聯絡人:", demo_name, "電話:", demo_phone)

print()
print("--- 2. 使用 sorted(demo_cntc.keys()) 依姓名排序走訪 ---")
for demo_name in sorted(demo_cntc.keys()):
    print("排序聯絡人:", demo_name, "->", demo_cntc[demo_name])
````

````python
# 10.4.3 練習題：及格名單公告系統
# 題目說明：
# 給定一份學生數學成績字典 exm_outs。
# 請依字典中的鍵走訪所有學生：
# 若該學生分數 >= 60，印出 "[合格] <學生名>: <分數>分"
# 若該學生分數 < 60，印出 "[需加強] <學生名>: <分數>分"
#
# 【公開測試資料 1】
# 成績字典: {"Tom": 75, "Ken": 52, "Amy": 90}
# 輸出:
# [合格] Tom: 75分
# [需加強] Ken: 52分
# [合格] Amy: 90分
#
# 【公開測試資料 2】
# 成績字典: {"Leo": 60, "Zoe": 59}
# 輸出:
# [合格] Leo: 60分
# [需加強] Zoe: 59分

exm_outs = {"Tom": 75, "Ken": 52, "Amy": 90}

for student in exm_outs.keys():
    score = exm_outs[student]
    if score >= 60:
        print("[合格] " + student + ": " + str(score) + "分")
    else:
        print("[需加強] " + student + ": " + str(score) + "分")
````

## 10.4.4 值走訪：d.values() 與極值/總和運算

### 觀念說明
在某些統計與彙整場景中，我們只在乎字典中的「數值」，完全不需要在乎鍵是什麼。例如：計算整個月的總營業額、全班平均成績、或是尋找最高與最低銷售量。

此時，使用 `d.values()` 可以直接取出包含所有值（values）的視圖物件：
- **`sum(d.values())`**：計算字典中所有數值的總和。
- **`max(d.values())`**：找出字典中的最大數值。
- **`min(d.values())`**：找出字典中的最小數值。
- **`len(d)`**：取得資料筆數，搭配總和即可計算出「平均值」：`sum(d.values()) / len(d)`。

透過 `d.values()`，我們不需要在迴圈裡重複做 `total += d[k]`，一秒就能完成數據彙整，這在 APCS 數據處理題型中既簡短又不易出錯。

````python
# 10.4.4 範例：月度銷售額極值與統計
demo_msls = {
    "Jan": 12000,
    "Feb": 18500,
    "Mar": 14200,
    "Apr": 21000,
    "May": 16800
}

# 1. 取得所有數值並走訪
demo_asls = demo_msls.values()
print("values 視圖:", demo_asls)

# 2. 使用內建函式快速統計
demo_t104 = sum(demo_msls.values())
demo_best = max(demo_msls.values())
demo_worst = min(demo_msls.values())
demo_a104 = demo_t104 / len(demo_msls)

print("總營業額:", demo_t104)
print("最高單月營業額:", demo_best)
print("最低單月營業額:", demo_worst)
print("平均單月營業額:", demo_a104)
````

## 10.4.5 鍵值對走訪：for k, v in d.items(): 雙變數解包走訪

### 觀念說明
如果我們同時需要「鍵」與「值」，雖然可以寫成 `for k in d: v = d[k]`，但最 Pythonic、最優雅且效能最好的方式，絕對是使用 **`d.items()`**！

### 雙變數解包（Unpacking）
`d.items()` 會返回由 `(key, value)` 元組組成的視圖物件。配合 Python 的多變數賦值機制，我們可以在 `for` 迴圈中同時宣告兩個變數：
```python
for key, value in d.items():
    print(key, value)
```
這樣寫的好處：
1. **免除二次查表**：不需要再呼叫 `d[key]` 重新做雜湊計算，效能更佳。
2. **語意清晰**：變數名稱能直接代表意義（例如 `for name, score in scores.items():`）。
3. **簡潔俐落**：大幅提升 APCS 程式題的撰寫速度與排版美觀。

````python
# 10.4.5 練習題：最高分得主尋找（不依賴自訂函式）
# 題目說明：
# 給定一份玩家積分字典 board104。
# 請使用 items() 走訪每一位玩家與分數，找出獲得「最高分」的玩家姓名與分數。
# （若有多人同分，記錄第一位遇到的即可）
# 最後依格式印出冠軍資訊。
#
# 【公開測試資料 1】
# 積分表: {"player_A": 1500, "player_B": 2850, "player_C": 2100}
# 輸出: 冠軍: player_B 得分: 2850
#
# 【公開測試資料 2】
# 積分表: {"IronMan": 98, "Thor": 95, "Hulk": 92}
# 輸出: 冠軍: IronMan 得分: 98

board104 = {"player_A": 1500, "player_B": 2850, "player_C": 2100}

top_player = ""
top_sc = -1

for name, score in board104.items():
    if score > top_sc:
        top_sc = score
        top_player = name

print("冠軍:", top_player, "得分:", top_sc)
````

## 10.4.6 字典生成式（Dictionary Comprehension）：{k: v for ... if ...}

### 觀念說明
在第 9 章中，我們見識過強大無比的「串列生成式（List Comprehension）」。Python 同樣將這種極具表現力的語法延伸到了字典上，稱為**字典生成式（Dictionary Comprehension）**！

### 語法結構
```python
{key_expr: value_expr for item in iterable if condition}
```
與串列生成式的主要差異在於：
1. 最外層改用**大括號 `{}`** 包裹。
2. 前方必須是符合字典鍵值對格式的 **`key_expr: value_expr`**（中間有冒號）。

### 常見應用情境
- **快速批量運算**：將所有商品的售價打九折。
- **資料篩選過濾**：只保留及格的學生成績，剔除不及格者。
- **反向建立索引**：將既有字典的鍵值對調（例如 `{v: k for k, v in d.items()}`）。

````python
# 10.4.6 範例：字典生成式的多樣應用
demo_oprcs = {"Apple": 50, "Banana": 30, "Cherry": 120, "Durian": 200}

# 1. 批量打 9 折（四捨五入取整數）
demo_dprcs = {demo_item: int(demo_price * 0.9) for demo_item, demo_price in demo_oprcs.items()}
print("全面九折後:", demo_dprcs)

# 2. 搭配 if 進行篩選（只保留價格大於 100 的高級水果）
demo_lfrts = {demo_item: demo_price for demo_item, demo_price in demo_oprcs.items() if demo_price >= 100}
print("高級水果清單:", demo_lfrts)

# 3. 從串列直接生成平方對照字典
demo_ns104 = [1, 2, 3, 4, 5]
demo_sqdct = {demo_x: demo_x ** 2 for demo_x in demo_ns104}
print("數字平方字典:", demo_sqdct)
````
