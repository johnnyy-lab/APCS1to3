### 8.7.1 列表生成式基礎語法：`[運算式 for 變數 in 迭代對象]`

在過去的寫法中，如果我們想建立一個包含 1 到 5 平方數的串列 `[1, 4, 9, 16, 25]`，通常需要好幾個步驟：
1. 先建立一個空的置物櫃 `squares = []`
2. 寫一個 `for x in range(1, 6):` 迴圈
3. 在迴圈內部呼叫 `squares.append(x ** 2)`

這樣寫雖然正確，但需要 3 到 4 行程式碼。Python 提供了一種極具數學美感與優雅風格的語法：**列表生成式（List Comprehension）**！

#### 語法架構解析：
```python
新串列 = [運算式 for 變數 in 迭代對象]
```
- **中括號 `[...]`**：宣告這是一顆即將被孵化出來的新串列。
- **`for 變數 in 迭代對象`**：扮演發電機的角色，依序吐出每一筆原始數據。
- **最前面的 `運算式`**：扮演加工廠的角色，決定放進新串列的元素長什麼模樣！

#### 為什麼資深工程師都愛用？
1. **代碼精簡**：一行搞定，閱讀時一目了然。
2. **效能更優**：列表生成式在 Python 底層是由 C 語言層級直接進行最佳化組裝，執行速度通常比在 Python 迴圈中頻繁呼叫 `.append()` 還要快上 15% ~ 30%！

````python
# [2] Code 範例：傳統寫法 vs 列表生成式對照

# 傳統寫法：需 3 行
traditional_squares = []
for x in range(1, 6):
    traditional_squares.append(x ** 2)
print("傳統寫法產出的平方串列：", traditional_squares)

# 列表生成式：只要 1 行！
comprehension_squares = [x ** 2 for x in range(1, 6)]
print("生成式寫法產出的平方串列：", comprehension_squares)

# 範例 2：將多個單字全部轉為長度數字
words = ["apple", "banana", "cat", "elephant"]
lengths = [len(w) for w in words]
print("單字長度清單：", lengths)
````

### 8.7.2 列表生成式搭配條件篩選：`[運算式 for 變數 in 迭代對象 if 條件]`

如果我們只想挑選出一部分符合標準的資料呢？例如：從一堆數字中「只挑出偶數」、從學生成績中「只挑出及格的分數」。

在列表生成式中，這只需要在尾巴加上一個 **`if 條件式`**！

#### 運作流程與執行順序：
```text
[  運算式   for 變數 in 迭代對象   if 條件判斷  ]
     ③              ①                  ②
  放進結果         逐一取出          守門員檢查
```
1. **第一步（取值）**：`for` 迴圈依序取出一個元素。
2. **第二步（篩選）**：`if` 條件守門員進行判斷。若條件為 `False`，此元素直接被丟棄忽略，進入下一輪。
3. **第三步（變換與收集）**：只有通過 `if` 考驗（`True`）的元素，才會被送到最前面的運算式進行計算，最後收集進新串列。

#### 適用場景：
當你需要**「資料長度縮短、篩選過濾雜訊」**時，末端 `if` 是最直覺的神兵利器。

````python
# [2] Code 範例：使用末端 if 過濾元素

numbers = [12, 7, 18, 5, 24, 3, 9, 30]

# 1. 只挑出偶數
evens = [x for x in numbers if x % 2 == 0]
print("偶數清單：", evens)

# 2. 只挑出大於 10 的數字，並且將它們除以 2
halved_big = [x // 2 for x in numbers if x > 10]
print("大於 10 的數值折半清單：", halved_big)

# 3. 字串過濾：只挑出長度大於 4 個字母的單字
fruits = ["apple", "fig", "banana", "pear", "watermelon"]
long_fruits = [f for f in fruits if len(f) > 4]
print("長單字清單：", long_fruits)
````

### 8.7.3 APCS 必備同行數值動態輸入：`[int(x) for x in input().split()]`

在 APCS、ZeroJudge、LeetCode 與各類程式競賽中，超過 80% 的測資格式都長這樣：
```text
題目輸入：10 25 8 99 43 72
```
**同行內有不定數量的整數，彼此以空格分隔！**

#### 過去的痛點：
在第四章我們學過 `parts = input().split()`，但它拆開來後得到的是**字串串列**：`["10", "25", "8", "99", "43", "72"]`。
如果直接拿它們來做 `max()` 或 `sum()`，會發生嚴重悲劇：
- 字串比大小是字典序：`"9" > "10"`，求出的最大值是錯的！
- 字串不能直接做加法總和，`sum()` 會直接噴出 TypeError 崩潰！

#### APCS 萬能神咒：
```python
nums = [int(x) for x in input().split()]
```
這行咒語結合了兩大功能：
1. `input().split()`：自動將使用者輸入的那一行按空白切開成一個個單字片段。
2. `[int(x) for x in ...]`：依序將每一個單字丟給 `int()` 轉型為真實整數！

無論該行有 3 個數字還是 1000 個數字，這短短一行代碼都能在 0.01 秒內將它們完整轉成純整數串列，這是每位 APCS 考生都必須倒背如流的基本功！

````python
# [2] Code 範例：同行多筆整數輸入與統計運算

# 模擬 APCS 讀取測資（此處以字串模擬 input().split() 的效果）
sample_input = "25 80 15 90 60"
tokens = sample_input.split()

# 列表生成式秒轉整數串列！
scores = [int(x) for x in tokens]

print("轉型後的純整數串列：", scores)
print("總分：", sum(scores))
print("最高分：", max(scores))
print("最低分：", min(scores))

# 實際互動模式示範（已註解）：
# my_nums = [int(x) for x in input("請輸入一行整數（以空白隔開）：").split()]
# print("你輸入的串列是：", my_nums)
````

### 8.7.4 雙重分支與三元運算生成式：`[值A if 條件 else 值B for ...]`

現在我們要來探討初學者最常混淆的一個**「文法結構分水嶺」**：

#### 核心問題：`if` 到底要放前面還是放後面？
這取決於你的目的究竟是**「過濾丟棄（Filter）」**還是**「數值二選一（Map）」**：

1. **放後面：純過濾（長度變短）**
   - 語法：`[x for x in a if x >= 60]`
   - 意義：不及格的分數直接扔掉，結果串列變短。
   - **絕對不能加 `else`！** 寫成 `[x for x in a if x >= 60 else 0]` 會直接報 SyntaxError！

2. **放前面：三元運算二選一（長度不變，每個都有結果）**
   - 語法：`["及格" if s >= 60 else "不及格" for s in scores]`
   - 意義：串列長度完全不變！每一位學生的成績都要被貼上標籤，及格貼 "及格"，不及格貼 "不及格"。
   - **必須搭配 `else`！** 寫成 `["及格" if s >= 60 for s in scores]` 會報錯，因為電腦不知道不及格時要填什麼！

#### 口訣心法：
- **要挑選丟東西，`if` 殿後沒 `else`**。
- **要替換貼標籤，`if-else` 放最前**。

````python
# [2] Code 範例：前置三元運算式 vs 末端過濾式

scores = [85, 45, 92, 58, 70]

# 1. 前置 if-else：將成績轉換為 Pass / Fail 狀態標籤（長度不變，依然是 5 個）
status_tags = ["Pass" if s >= 60 else "Fail" for s in scores]
print("評語標籤清單：", status_tags)

# 2. 前置 if-else：將數值中的負數歸零，正數維持不變（數值修剪）
readings = [15, -4, 28, -10, 0, 7]
clipped = [x if x > 0 else 0 for x in readings]
print("修剪後的數值：", clipped)

# 3. 末端 if：只挑及格成績（長度縮短為 3 個）
passed_only = [s for s in scores if s >= 60]
print("純及格分數清單：", passed_only)
````

### 8.7.5 列表生成式實戰綜合特訓

來到第八章的最終單元，我們將把前面學到的所有技術融會貫通！

#### 列表生成式的綜合實戰應用盤點：
1. **資料清洗（Data Cleaning）**：
   面對使用者輸入夾雜多餘空格的字串清單，例如 `["  apple ", " banana\n", "cat "]`，可以用 `[w.strip() for w in raw_words]` 一秒去除頭尾空白。
2. **格式轉化與文字重構**：
   將整數串列快速轉為帶引號的字串格式，例如 `[f"第 {x} 號" for x in range(1, 5)]`。
3. **數值型態大量轉換與統計**：
   結合 `sum()`、`max()`、`min()`，在生成式的瞬間完成運算，連暫存串列都不需要宣告！例如：
   `sum([x for x in data if x > 0])` 即可直接算出所有正數的總和。

#### 實用建議與代碼可讀性平衡（Code Readability）：
列表生成式雖然強大無比，但原則上**不建議在同一行內塞入超過兩層迴圈或過度複雜的巢狀邏輯**。若邏輯過長，分開寫成清晰的 `for` 迴圈反而更好維護。恰到好處的使用，能讓你的 Python 程式碼既專業又俐落！

````python
# [2] Code 範例：文字清洗與直接嵌套統計

raw_inputs = ["  Alice  ", "   Bob\n", "Charlie ", "   David "]

# 1. 去除空白並全部轉為大寫
cleaned = [name.strip().upper() for name in raw_inputs]
print("清洗後的標準化姓名：", cleaned)

# 2. 一行計算所有長度大於 4 個字母的姓名總字元數
total_chars = sum([len(name) for name in cleaned if len(name) > 4])
print("長姓名總字元數：", total_chars)

# 3. 解包漂亮輸出！
print("解包成果：", *cleaned, sep=" | ")
````
