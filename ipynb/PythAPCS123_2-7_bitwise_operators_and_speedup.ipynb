### ⚡ 2-7-1 位元 AND（&）與超光速奇偶判斷（x & 1）

在日常生活裡，如果我們要核驗兩把鑰匙才能開門，必須兩把都吻合才行。這就是「**位元 AND 運算子 `&`（Bitwise AND）**」的運作機制！

**位元 AND 的運算真值表**：
電腦會把兩個數字轉成二進位，由右至左「一個位元一個位元」進行垂直對齊比對：
* `1 & 1` $\rightarrow$ **`1`**（只有雙方都是 1，結果才是 1！）
* `1 & 0` $\rightarrow$ **`0`**
* `0 & 1` $\rightarrow$ **`0`**
* `0 & 0` $\rightarrow$ **`0`**
例如：`0b1010 & 0b1100`，只有第 4 位元都是 1，所以結果是 `0b1000`（十進位 8）！

**🚀 吳邦一老師講義第 78 頁傳授的絕殺秘技：`x & 1` 取代模除奇偶判斷！**
在單元 2-3 中，我們學過用 `x % 2` 算餘數來判斷奇數偶數。但在電腦硬體層面，除法運算需要消耗幾十個 CPU 時脈週期；而**位元運算 `&` 只需要「1 個時脈週期」**！

**為什麼 `x & 1` 可以一秒判斷奇偶？**
在二進位中，所有的權重都是 2 的倍數（2, 4, 8, 16... 全部都是偶數！），**全天下只有最右邊的第一位（$2^0 = 1$）能決定奇偶**！
* 數字 1 的二進位就是 `...0001`。
* 當任何數字 `x` 和 `1` 做 AND 運算時：前面所有偶數位元通通會被乘以 0 抹殺掉，**只留下最右邊的那 1 位**！
  * 如果最右邊是 1（奇數）：`x & 1` 算出來必定是 **`1`**！
  * 如果最右邊是 0（偶數）：`x & 1` 算出來必定是 **`0`**！

在 APCS 百萬次大數據走訪測資中，把 `x % 2 == 1` 改寫成 `x & 1`，能大幅飆高執行速度，徹底遠離超時 TLE 惡夢！

````python
# [2] Code 範例區：展示位元 AND（&）與 x & 1 極速奇偶判斷
# 範例 1：兩個二進位數進行逐位元 AND 比對
demo_va027 = 0b1010  # 十進位 10
demo_vb027 = 0b1100  # 十進位 12
demo_vc = demo_va027 & demo_vb027   # 只有最高位都是 1，結果為 0b1000
print(demo_vc)    # 印出 8
print(bin(demo_vc)) # 印出 '0b1000'

# 範例 2：APCS 絕殺加速技——x & 1 判斷奇偶數
demo_even = 42
demo_odd = 99
print(demo_even & 1)  # 印出 0（偶數最末位是 0）
print(demo_odd & 1)   # 印出 1（奇數最末位是 1）
````

### 🔌 2-7-2 位元 OR（|）與狀態開關強制啟用

接下來我們要認識第二位成員——「**位元 OR 運算子 `|`（Bitwise OR，鍵盤 Shift + 反斜線 \ 鍵）**」！

你可以把位元 OR 想像成國小自然課學過的「**並聯電路開關**」：
只要兩條電路中「**有任何一條通電（為 1）**」，燈泡就會立刻亮起來！只有在兩條都斷電（都是 0）時，燈泡才會熄滅。

**位元 OR 的運算真值表**：
* `1 | 1` $\rightarrow$ **`1`**
* `1 | 0` $\rightarrow$ **`1`**
* `0 | 1` $\rightarrow$ **`1`**
* `0 | 0` $\rightarrow$ **`0`**（只有雙方都是 0，結果才是 0！）

例如：`0b1010 | 0b0101`，每一位剛好都各有 1，合併起來就變成了全開的 `0b1111`（十進位 15）！

**🎛️ 狀態遮罩核心神技：點亮開關（Set Bit）**：
在競技遊戲與 APCS 題目中，假設勇者當前的狀態只有護盾：`status = 0b0100`（第 3 位元是 1）。
現在勇者喝下了加速藥水，獲得了疾跑技能（第 1 位元 `0b0001`）。我們如何把疾跑技能「加進去」，而且保證絕對不會弄壞勇者原本已經擁有的護盾呢？
答案就是出動位元 OR：
```python
status = status | 0b0001
```
原本是 1 的護盾依然保留是 1，原本是 0 的疾跑位元瞬間被點亮變成 1，狀態完美合體變成 `0b0101`！
學會用 `|` 合併狀態，你就能用最優雅的一行代碼操控複雜的遊戲技能樹！

````python
# [2] Code 範例區：展示位元 OR（|）與狀態開關啟用
# 範例 1：逐位元 OR 運算
demo_p = 0b1010  # 10
demo_q = 0b0101  # 5
demo_mrgd = demo_p | demo_q
print(demo_mrgd)       # 印出 15（0b1111）
print(bin(demo_mrgd))  # '0b1111'

# 範例 2：遊戲狀態合併（點亮新技能開關）
# 初始狀態：已擁有技能 A (0b0100)
demo_hero = 0b0100
# 拾獲新技能 B (0b0010) 與 技能 C (0b0001)
demo_hero = demo_hero | 0b0010 | 0b0001
print(bin(demo_hero))  # 印出 '0b111'（三個技能開關全部點亮！）
````

### 🎭 2-7-3 位元 XOR（^）與開關翻轉／相同自消魔法

在單元 2-4 中，我們鄭重警告過大家：千萬不要把帽子脫字符號 `^` 當成次方運算！
今天，我們要正式為 `^` 舉行隆重的正名儀式——它的真名叫做「**位元互斥或運算子（Bitwise XOR）**」！

你可以把 XOR 想像成一對「**反骨叛逆雙胞胎**」：
只要兩個人「**意見不同（相異）**」，結果就是歡天喜地的 `1`；
如果兩個人「**意見一模一樣（相同）**」，結果就是無趣的 `0`！

**位元 XOR 的運算真值表（相異為 1，相同為 0）**：
* `1 ^ 0` $\rightarrow$ **`1`**（不同即為 1！）
* `0 ^ 1` $\rightarrow$ **`1`**（不同即為 1！）
* `1 ^ 1` $\rightarrow$ **`0`**（相同歸 0！）
* `0 ^ 0` $\rightarrow$ **`0`**（相同歸 0！）

**🎩 震撼競賽演算法的三大神級 XOR 魔法**：
1. **任何數字跟 0 做 XOR，保持原貌**：
   $$x \text{ ^ } 0 = x$$
2. **任何數字跟自己做 XOR，自相消滅歸零（相同自消）**：
   $$x \text{ ^ } x = 0$$
   例如 `42 ^ 42` 算出來絕對是 `0`！在很多競技題目中，成對出現的數字只要全部連環 XOR 起來，一對一對的數字就會像俄羅斯方塊一樣全部自動消滅，最後留下來的就是那個「唯一落單的單身狗數字」！
3. **開關狀態快速翻轉（Toggle）**：
   如果某個開關原本是 0，跟 1 做 XOR 就會變成 1；原本是 1，跟 1 做 XOR 就會翻轉回 0！不需要寫繁瑣的 `if-else`，一個 `^ 1` 瞬間完成開關切換！

````python
# [2] Code 範例區：展示位元 XOR（^）的相異為 1、自相消滅與開關翻轉
# 範例 1：基本 XOR 比對（相異為 1，相同為 0）
demo_xa = 0b1010  # 10
demo_xb = 0b1100  # 12
print(bin(demo_xa ^ demo_xb))  # 印出 '0b110'（十進位 6）

# 範例 2：神級魔法——相同自相消滅歸 0
print(99 ^ 99)     # 印出 0！任何數與自己 XOR 必定消滅！
print(99 ^ 0)      # 印出 99！任何數與 0 XOR 保持原樣

# 範例 3：找單身狗數字（成對數字自動抵銷）
# 數列中有成對的 7 與 7，唯獨 15 是落單的
demo_one = 7 ^ 15 ^ 7
print(demo_one)      # 印出 15！（7 與 7 自消歸零，0 ^ 15 = 15）
````

### 🎢 2-7-4 位元左移（<<）與右移（>>）——二進位滑梯與乘除倍增

最後登場的兩位成員，是位元世界最富動感的「位移雙兄弟」——**位元左移 `<<`** 與 **位元右移 `>>`**！

你可以把它們想像成排排坐的小朋友在玩「**二進位溜滑梯**」：

**1. 左移運算子：`x << n`（向左滑推 $n$ 位）**
把數字的每一個位元全部往左邊推移 $n$ 格，右邊空出來的位元通通自動補上 `0`！
* 在十進位中，數字後面補一個 0 代表「乘以 10」；
* 在二進位中，數字後面補一個 0 代表「**乘以 2**」！
👉 **鐵則：向左移 1 位就是乘以 2；左移 $n$ 位就是乘以 $2^n$！**
例如：`5 << 1` 就是 $5 \times 2 = 10$；`5 << 2` 就是 $5 \times 4 = 20$！

**2. 右移運算子：`x >> n`（向右滑推 $n$ 位）**
把數字的每一個位元全部往右邊推移 $n$ 格，最右邊掉落滑梯的位元直接丟棄！
👉 **鐵則：向右移 1 位剛好等價於「整除 2（`// 2`）」！**
例如：`19 >> 1` 就是 $19 // 2 = 9$；`19 >> 2` 就是 $19 // 4 = 4$！

**🚀 頂尖競賽選手的提速日常**：
在未來的「二分搜尋法（Binary Search）」或快速冪演算法中，我們需要頻繁地將區間長度砍半。
頂尖選手會寫 `mid = (left + right) >> 1`，這在底層直接調動 CPU 的位移暫存器，比執行整除除法更快、更具極客風範！

````python
# [2] Code 範例區：展示位元左移 << 與右移 >> 的倍增與減半效果
# 範例 1：左移 << 相當於乘以 2 的次方
demo_v027 = 5
print(demo_v027 << 1)  # 5 * 2 = 10
print(demo_v027 << 2)  # 5 * 4 = 20
print(demo_v027 << 3)  # 5 * 8 = 40

# 範例 2：右移 >> 相當於整除 2 的次方（去尾整除）
demo_n027 = 19
print(demo_n027 >> 1)  # 19 // 2 = 9
print(demo_n027 >> 2)  # 19 // 4 = 4
print(demo_n027 >> 3)  # 19 // 8 = 2
````

````python
#@title 🧪 點擊此處取得單元 2-7 學習成效自動評分報告 { display-mode: "form" }
# ------------------------------------------------------------------------------
# 官方智慧全環境評分載入器（支援：本地開發庫優先 ➔ 公開正式庫雙重備援）
# ------------------------------------------------------------------------------
import os
import sys
import json
import logging
import urllib.request

# 靜音內部 httplib2 警告
logging.getLogger("google_auth_httplib2").setLevel(logging.ERROR)

# 1. 互動式輸入學員姓名與學號（分開獨立欄位，按 Enter 採用預設值）
print("👤【學員身分登記】（自學測試可直接按 Enter 鍵使用預設值）：")
student_name = input("請輸入您的姓名 [預設: 自學冒險者]：").strip() or "自學冒險者"
student_id = input("請輸入學號或座號 [預設: 無座號]：").strip() or "無座號"

# 2. 自動嘗試獲取 Google 帳號官方身分認證（真實 Email 與 Google 暱稱）
google_email = None
google_name = None
try:
    from google.colab import auth
    print("🔐 正在連結 Google 帳號進行身分認證（若跳出授權視窗，請點選允許）...")
    auth.authenticate_user()

    # 方法 1: 使用 Google Drive API v3 (about:get) 抓取真實 displayName 與 emailAddress
    # Colab ADC 預設包含 drive 權限，能 100% 精準獲取 Google 帳號中文顯示名稱（如：葉明昌）
    try:
        import google.auth
        from googleapiclient.discovery import build
        creds, _ = google.auth.default()
        drive_service = build('drive', 'v3', credentials=creds, cache_discovery=False)
        about = drive_service.about().get(fields="user(displayName,emailAddress)").execute()
        u_obj = about.get("user", {})
        google_name = u_obj.get("displayName")
        google_email = u_obj.get("emailAddress")
    except Exception:
        pass

    # 方法 2: 若 Drive API 失敗，嘗試透過 OAuth2 userinfo / tokeninfo 備援
    if not google_email or not google_name:
        try:
            import google.auth
            from google.auth.transport.requests import Request
            credentials, _ = google.auth.default()
            credentials.refresh(Request())
            token = credentials.token

            for ep in [
                f"https://www.googleapis.com/oauth2/v3/tokeninfo?access_token={token}",
                "https://www.googleapis.com/oauth2/v3/userinfo",
                "https://www.googleapis.com/oauth2/v2/userinfo"
            ]:
                try:
                    req = urllib.request.Request(
                        ep,
                        headers={"Authorization": f"Bearer {token}"}
                    )
                    with urllib.request.urlopen(req, timeout=5) as resp:
                        u_data = json.loads(resp.read().decode('utf-8'))
                        if not google_email:
                            google_email = u_data.get("email") or u_data.get("emailAddress")
                        if not google_name:
                            google_name = (
                                u_data.get("name")
                                or u_data.get("given_name")
                                or u_data.get("displayName")
                            )
                        if google_email and google_name:
                            break
                except Exception:
                    continue
        except Exception:
            pass

    # 方法 3: 若只有 email 但 Google API 沒給 name，以 email 前綴作為暱稱回退
    if google_email and not google_name:
        google_name = google_email.split("@")[0]

    if google_email:
        print(f"✅ 身分認證成功！已綁定 Google 帳號：{google_name} ({google_email})")
except Exception:
    # 非 Colab 環境、本機離線測試或略過授權
    pass

# 3. 身分資料獨立注入環境變數（學號與姓名完全分開）
os.environ["STUDENT_NAME"] = student_name
os.environ["STUDENT_ID"] = student_id
if google_email:
    os.environ["GOOGLE_EMAIL"] = google_email
if google_name:
    os.environ["GOOGLE_NAME"] = google_name

# 4. 評分核心腳本載入（本機優先 ➔ 遠端記憶體載入，絕不產生實體暫存檔）
# 找不到來源就換下一條。內容已拿到但執行失敗時，停止改試其他來源。
# 評分腳本放在獨立命名空間執行，成功後也不把名稱寫回筆記本。答案讀執行前的學生存放格快照。
_loaded = False
_load_error = None

# 4.1 優先嘗試讀取開發庫/本機檔案（本機離線開發測試零延遲）
_local_paths = [
    "web/graders/grade_2_7.py",
    "../graders/grade_2_7.py",
    "../../graders/grade_2_7.py",
    "../web/graders/grade_2_7.py",
    "../../web/graders/grade_2_7.py"
]
for _lp in _local_paths:
    if not os.path.exists(_lp):
        continue
    try:
        with open(_lp, "r", encoding="utf-8") as _f:
            _code = _f.read()
        _grader_ns = {"__builtins__": __builtins__, "__name__": "apcs_grader", "__STUDENT_ENV__": dict(globals())}
        try:
            exec(_code, _grader_ns)
        finally:
            del _grader_ns
        os.environ["GRADER_SOURCE"] = _lp
        _loaded = True
        break
    except Exception as _exc:
        _load_error = _exc
        break

# 4.2 若在雲端環境（Colab），自動透過公開正式庫雙重備援以純記憶體載入
if not _loaded and _load_error is None:
    _grader_urls = [
        "https://raw.githubusercontent.com/johnnyy-lab/APCS1to3/main/graders/grade_2_7.py",
        "https://johnnyy-lab.github.io/APCS1to3/graders/grade_2_7.py",
        "https://cdn.jsdelivr.net/gh/johnnyy-lab/APCS1to3@main/graders/grade_2_7.py"
    ]
    for _url in _grader_urls:
        try:
            _req = urllib.request.Request(_url, headers={"User-Agent": "Mozilla/5.0"})
            with urllib.request.urlopen(_req, timeout=10) as _resp:
                _code = _resp.read().decode("utf-8")
        except Exception:
            continue
        try:
            _grader_ns = {"__builtins__": __builtins__, "__name__": "apcs_grader", "__STUDENT_ENV__": dict(globals())}
            try:
                exec(_code, _grader_ns)
            finally:
                del _grader_ns
            os.environ["GRADER_SOURCE"] = _url
            _loaded = True
            break
        except Exception as _exc:
            _load_error = _exc
            break

if not _loaded:
    if _load_error is not None:
        print("❌ 評分程式載入失敗，請稍後重新執行。若仍失敗，請把這個畫面交給老師。")
        if os.environ.get("GRADER_DEBUG", "").strip().lower() in ("1", "true", "yes", "on"):
            _reason = str(_load_error).strip().splitlines()
            _line = _reason[0][:180] if _reason else ""
            if _line:
                print("（開發診斷：" + type(_load_error).__name__ + "：" + _line + "）")
            else:
                print("（開發診斷：" + type(_load_error).__name__ + "）")
    else:
        print("❌ 評分系統連線異常，請確認網路連線狀態（或稍後重新執行）")
````
