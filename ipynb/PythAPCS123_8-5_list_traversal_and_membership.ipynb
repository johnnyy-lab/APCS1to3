### 8.5.1 直接元素走訪：`for item in my_list:`

在日常生活中，老師拿著點名簿依序喊出學生的名字，或是車站剪票員依序檢查排隊旅客的車票，這種**「從頭到尾把隊伍中的每一樣東西拿出來看一次」**的動作，在程式設計中就稱為**走訪（Traversal / Iteration）**。

Python 提供了非常優雅直觀的語法：`for 變數 in 串列:`。在迴圈運作時，Python 會自動從串列的第一個元素（索引 0）開始，每一回合將一個元素取出來放入指定的暫存變數中，直到所有項目都處理完畢為止。我們完全不需要手動計算串列的長度，也不需要手動操作索引數字！

#### 運作機制與優缺點解析：
1. **優點（簡潔清晰）**：程式碼可讀性極高，語意就是「針對串列裡的每一個項目，執行下列動作」。
2. **重要特性（唯讀性存取）**：在迴圈區塊內，若你對暫存變數重新賦值（例如 `item = item + 10`），**原串列裡面的數值完全不會改變**！因為 `item` 只是複製了一份內容的臨時變數，並不是直接指向置物櫃的標籤。
3. **新手常見避坑防禦（千百萬不要在走訪中增刪元素）**：
   初學同學常犯的大忌，是在 `for x in a:` 的迴圈內部執行 `a.remove(x)` 或 `a.append(...)`。這會動態改變串列的長度與內部指針，導致迴圈出現**「跳過某些項目」**或陷入無窮迴圈的奇怪現象！若只是單純讀取、統計或篩選輸出，直接走訪是最佳利器。

````python
# [2] Code 範例：使用 for ... in 直接走訪串列元素

fruits = ["蘋果", "香蕉", "奇異果", "芒果"]

print("=== 水果庫存檢視 ===")
for fruit in fruits:
    print("目前檢查到的水果是：", fruit)

# 統計數值範例：計算所有考試成績的總分
scores = [85, 92, 78, 90, 88]
total = 0
for s in scores:
    total += s

print("總分：", total)
print("平均：", total / len(scores))
````

### 8.5.2 索引走訪與元素更新：`for i in range(len(my_list)):`

在上一節中，我們學到了 `for item in a:` 可以輕鬆讀取每一筆資料。但如果今天老師的要求是：**「全員成績加 5 分，並且直接覆蓋更新原先的成績單」**，直接走訪就無能為力了！因為修改暫存變數無法改變置物櫃中的正本。

這時就必須派出強大的**「索引走訪法」**：`for i in range(len(a)):`。

#### 運作機制詳解：
1. **`len(a)`**：算出串列的總長度 $N$。
2. **`range(len(a))`**：產生從 `0` 開始到 `N-1` 的整數序列。這剛好百分之百精準對應串列中每一個元素的**合法正向索引**！
3. **就地更新語法**：在迴圈中透過 `a[i] = 新數值`，就能準確打開第 `i` 號置物櫃，將舊資料扔掉換成新資料！

#### 避坑防禦守則：
- **嚴防越界陷阱**：很多初學者會寫成 `range(len(a) + 1)`，這會多產生一個索引值 $N$，導致執行到最後一輪時引發致命的 `IndexError: list index out of range`！請牢記 `range(len(a))` 剛好到 `len(a)-1`，永遠不需要加一。
- **適用時機**：需要「就地修改原串列內容」、需要「同時比對相鄰兩項 `a[i]` 與 `a[i+1]`」或需要「知道當前是第幾個項目」時，索引走訪法是無可取代的標準解法。

````python
# [2] Code 範例：使用 range(len(a)) 就地更新串列元素

scores = [78, 85, 96, 54, 62]
print("調分前的原始成績：", scores)

# 任務：所有人加 5 分，若超過 100 分則以 100 分計
for i in range(len(scores)):
    scores[i] += 5
    if scores[i] > 100:
        scores[i] = 100

print("調分後的更新成績：", scores)
````

### 8.5.3 雙軌資訊走訪：同時掌握索引與數值

在很多實際生活與考試情境中，我們不僅需要知道「置物櫃裡裝了什麼」，同時還必須知道「它是第幾個（第幾名、第幾天、座號幾號）」。例如印出成績排行榜時，我們希望輸出：「第 1 名：王小明 98 分；第 2 名：李小華 95 分...」。

#### 兩種雙軌走訪策略：
1. **傳統計數器法（手動維護索引序號）**：
   在進入 `for item in a:` 迴圈前，宣告一個 `idx = 0` 或 `rank = 1`。每走訪完一個項目，手動將 `rank += 1`。這種方式在邏輯上極度直觀，初學者容易理解。
2. **標準索引讀取法（由 i 反查 a[i]）**：
   在 `for i in range(len(a)):` 中，變數 `i` 掌控了索引（若要顯示人類慣用的 1 號開始，可用 `i + 1`），而 `a[i]` 則掌控了實際數值。一次兼顧位置與內容！

#### APCS 實戰提示：
在 APCS 實作題中，經常需要尋找「最大值出現在哪一天」、「最小值是第幾位選手」。這類題目單純用 `max(a)` 只能求出數值，卻無法得知是哪一天。唯有透過帶有索引的雙軌走訪，一邊記錄目前看到的極值，一邊記下當前的索引 `best_idx = i`，才能漂亮解決問題！

````python
# [2] Code 範例：同時輸出索引位置與數值內容

players = ["櫻木", "流川", "赤木", "宮城", "三井"]

print("=== 籃球隊先發名單 ===")
for i in range(len(players)):
    player_name = players[i]
    jersey_no = i + 1  # 背號從 1 開始展示
    print(f"背號 {jersey_no} 號選手：{player_name}")

# 實戰尋找：找出分數超過 90 分的第一個人是誰、以及座號幾號
scores = [65, 82, 94, 88, 91]
for i in range(len(scores)):
    if scores[i] >= 90:
        print(f"發現高分選手！座號是 {i}，得分為 {scores[i]}")
        break  # 找到第一個就提早結束
````

### 8.5.4 成員查詢與存在性判斷：`in` 與 `not in` 運算子

當我們拿到一個裝有上百個名字的串列時，如何快速知道「某位特定同學有沒有在名單內」？難道要自己寫一個迴圈從頭比對到尾嗎？

Python 提供了極度親民且簡短的成員運算子：`in`（是否存在於）與 `not in`（是否不存在於）。

#### 語法特點與回傳值：
- `目標值 in 串列`：若目標值存在於串列中，運算結果為布林值 `True`；若不存在則回傳 `False`。
- `目標值 not in 串列`：若目標值「不在」串列中，回傳 `True`；若在裡面則回傳 `False`。

#### 底層機制與時間複雜度：
當你對一個一維串列執行 `x in my_list` 時，Python 背後的直譯器其實是默默地從第一個元素開始往後逐一比對（稱為線性搜尋 Linear Search）。
- 若串列長度為 $N$，最壞情況需要比對 $N$ 次，因此時間複雜度為 $O(N)$。
- 雖然在元素數量幾千以內時執行速度依然飛快（毫秒級），但未來在處理十幾萬筆海量資料時，就要注意不能在巢狀雙重迴圈中頻繁使用它（後續第十章會介紹 $O(1)$ 的集合 Set 利器）。

#### 防崩必備防護傘：
`in` 是後續各種操作（如搜尋索引 `index()`、移除元素 `remove()`）最不可或缺的安全防護門神！有了 `in` 先行把關，程式碼就絕對不會因為找不到目標而當機崩潰。

````python
# [2] Code 範例：使用 in 與 not in 進行成員存在性判定

vip_members = ["Alice", "Bob", "Charlie", "David"]

user1 = "Charlie"
user2 = "Eve"

# 檢查 user1 是否在 VIP 名單內
if user1 in vip_members:
    print(f"{user1} 是尊貴的 VIP 會員！")
else:
    print(f"{user1} 目前是一般會員。")

# 檢查 user2 是否不在名單內
if user2 not in vip_members:
    print(f"{user2} 不在 VIP 名冊中，邀請加入 VIP！")
````

### 8.5.5 定位元素索引：`index()` 方法與安全查詢模式

當我們確認某個資料「在清單內」之後，緊接著往往會想知道：**「那它究竟排在第幾個位置（索引是多少）？」**。

Python 串列提供了內建的 `my_list.index(x)` 方法，它會由左至右搜尋串列，並回傳目標值 `x` **第一次出現時的索引整數**。

#### 致命陷阱警示：`ValueError` 崩潰危機！
在某些程式語言中（例如 C 或 Java 的 indexOf），如果找不到目標物，通常會回傳一個無害的 `-1` 代表未找到。
但 **Python 的 `index()` 截然不同！如果搜尋的目標值不在串列中，Python 會直接噴出紅字錯誤並中斷程式執行**：
```text
ValueError: '某數值' is not in list
```

#### 必背黃金準則：先 `in` 把關，再 `index()` 提取！
為了避免程式在正式執行或 APCS 線上測資中意外崩潰，專業工程師一定會搭配上一節學到的 `in` 成員運算子，寫成如下的**「標準安全查詢樣式（Safe Lookup Pattern）」**：
```python
if target in my_list:
    pos = my_list.index(target)
    print("找到目標，位置在：", pos)
else:
    print("目標不存在於清單中")
```
只要養成「先用 `in` 確認存在、再呼叫 `index()`」的好習慣，你的程式碼就具備了無懈可擊的防禦力！

````python
# [2] Code 範例：使用 index() 定位元素位置與安全查詢樣板

colors = ["紅色", "藍色", "黃色", "綠色", "黃色"]

# 1. 正常定位第一次出現的位置
first_yellow = colors.index("黃色")
print("黃色第一次出現的索引是：", first_yellow)  # 輸出 2

# 2. 安全防禦示範
target = "紫色"

if target in colors:
    idx = colors.index(target)
    print(f"成功定位！{target} 位於索引 {idx}")
else:
    print(f"安全攔截：清單中沒有 {target}，不會觸發 ValueError 錯誤！")
````

````python
# [3] Code 填空題
# 任務：在排隊名單 queue 中尋找 "Charlie" 的位置，若存在印出索引，若不存在印出 -1

queue = ["Alice", "Bob", "David", "Eve"]
target = "Charlie"

# 請填入先檢查再定位的安全邏輯
if target ___ queue:
    pos = queue.___(target)
    print("排隊位置：", pos)
else:
    print(-1)
````

````python
#@title 🧪 點擊此處取得單元 8-5 學習成效自動評分報告 { display-mode: "form" }
# ------------------------------------------------------------------------------
# 官方智慧全環境評分載入器（支援：本地開發庫優先 ➔ 公開正式庫雙重備援）
# ------------------------------------------------------------------------------
import os
import sys
import json
import logging
import urllib.request

# 靜音內部 httplib2 警告
logging.getLogger("google_auth_httplib2").setLevel(logging.ERROR)

# 1. 互動式輸入學員姓名與學號（分開獨立欄位，按 Enter 採用預設值）
print("👤【學員身分登記】（自學測試可直接按 Enter 鍵使用預設值）：")
student_name = input("請輸入您的姓名 [預設: 自學冒險者]：").strip() or "自學冒險者"
student_id = input("請輸入學號或座號 [預設: 無座號]：").strip() or "無座號"

# 2. 自動嘗試獲取 Google 帳號官方身分認證（真實 Email 與 Google 暱稱）
google_email = None
google_name = None
try:
    from google.colab import auth
    print("🔐 正在連結 Google 帳號進行身分認證（若跳出授權視窗，請點選允許）...")
    auth.authenticate_user()

    # 方法 1: 使用 Google Drive API v3 (about:get) 抓取真實 displayName 與 emailAddress
    # Colab ADC 預設包含 drive 權限，能 100% 精準獲取 Google 帳號中文顯示名稱（如：葉明昌）
    try:
        import google.auth
        from googleapiclient.discovery import build
        creds, _ = google.auth.default()
        drive_service = build('drive', 'v3', credentials=creds, cache_discovery=False)
        about = drive_service.about().get(fields="user(displayName,emailAddress)").execute()
        u_obj = about.get("user", {})
        google_name = u_obj.get("displayName")
        google_email = u_obj.get("emailAddress")
    except Exception:
        pass

    # 方法 2: 若 Drive API 失敗，嘗試透過 OAuth2 userinfo / tokeninfo 備援
    if not google_email or not google_name:
        try:
            import google.auth
            from google.auth.transport.requests import Request
            credentials, _ = google.auth.default()
            credentials.refresh(Request())
            token = credentials.token

            for ep in [
                f"https://www.googleapis.com/oauth2/v3/tokeninfo?access_token={token}",
                "https://www.googleapis.com/oauth2/v3/userinfo",
                "https://www.googleapis.com/oauth2/v2/userinfo"
            ]:
                try:
                    req = urllib.request.Request(
                        ep,
                        headers={"Authorization": f"Bearer {token}"}
                    )
                    with urllib.request.urlopen(req, timeout=5) as resp:
                        u_data = json.loads(resp.read().decode('utf-8'))
                        if not google_email:
                            google_email = u_data.get("email") or u_data.get("emailAddress")
                        if not google_name:
                            google_name = (
                                u_data.get("name")
                                or u_data.get("given_name")
                                or u_data.get("displayName")
                            )
                        if google_email and google_name:
                            break
                except Exception:
                    continue
        except Exception:
            pass

    # 方法 3: 若只有 email 但 Google API 沒給 name，以 email 前綴作為暱稱回退
    if google_email and not google_name:
        google_name = google_email.split("@")[0]

    if google_email:
        print(f"✅ 身分認證成功！已綁定 Google 帳號：{google_name} ({google_email})")
except Exception:
    # 非 Colab 環境、本機離線測試或略過授權
    pass

# 3. 身分資料獨立注入環境變數（學號與姓名完全分開）
os.environ["STUDENT_NAME"] = student_name
os.environ["STUDENT_ID"] = student_id
if google_email:
    os.environ["GOOGLE_EMAIL"] = google_email
if google_name:
    os.environ["GOOGLE_NAME"] = google_name

# 4. 評分核心腳本載入（本機優先 ➔ 遠端記憶體載入，絕不產生實體暫存檔）
_loaded = False

# 4.1 優先嘗試讀取開發庫/本機檔案（本機離線開發測試零延遲）
_local_paths = [
    "web/graders/grade_8_5.py",
    "../graders/grade_8_5.py",
    "../../graders/grade_8_5.py",
    "../web/graders/grade_8_5.py",
    "../../web/graders/grade_8_5.py"
]
for _lp in _local_paths:
    if os.path.exists(_lp):
        try:
            os.environ["GRADER_SOURCE"] = _lp
            with open(_lp, "r", encoding="utf-8") as _f:
                exec(_f.read(), globals())
            _loaded = True
            break
        except Exception:
            pass

# 4.2 若在雲端環境（Colab），自動透過公開正式庫雙重備援以純記憶體載入
if not _loaded:
    _grader_urls = [
        "https://raw.githubusercontent.com/johnnyy-lab/APCS1to3/main/graders/grade_8_5.py",
        "https://johnnyy-lab.github.io/APCS1to3/graders/grade_8_5.py",
        "https://raw.githubusercontent.com/johnnyy-lab/APCS1to3-dev/main/web/graders/grade_8_5.py"
    ]
    for _url in _grader_urls:
        try:
            _req = urllib.request.Request(_url, headers={"User-Agent": "Mozilla/5.0"})
            with urllib.request.urlopen(_req, timeout=10) as _resp:
                _code = _resp.read().decode("utf-8")
                os.environ["GRADER_SOURCE"] = _url
                exec(_code, globals())
            _loaded = True
            break
        except Exception:
            continue

if not _loaded:
    print("❌ 評分系統連線異常，請確認網路連線狀態（或稍後重新執行）")
````

### 關鍵收穫清單：
1. **直接走訪**：`for x in a:` 直觀易讀，適合唯讀處理、統計累加，但在走訪中切勿增刪長度。
2. **索引走訪**：`for i in range(len(a)):` 是唯一能夠**就地修改原串列內容**並掌控精準排位的利器。
3. **成員檢查**：`in` 與 `not in` 運算子回傳布林值，是防範程式錯誤的最前線防護網。
4. **安全定位**：`index()` 找不到元素時會拋出 `ValueError`，必須永遠遵守「先 `in` 後 `index()`」的工程防禦規範！
