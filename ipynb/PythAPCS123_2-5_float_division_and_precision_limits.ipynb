### 🥤 2-5-1 單斜線除法（/）——必定產生浮點數的魔法轉身

在前面的單元 2-3 中，我們學到了「雙斜線 `//`」是冷酷無情的裝箱打包機，只留下純整數的商數。
那如果我們在鍵盤上「只打了一根斜線 `/`」（True Division，真除法），會發生什麼事呢？

你可以把單斜線 `/` 想像成一台「**神奇水果鮮榨果汁機**」！
不管是丟進去一顆完完整整的蘋果，還是切半的芭樂，只要按下一根斜線 `/`，**榨出來的成品永遠是帶著小數點刻度的「浮點水果汁（float）」**！

**⭐ 浮點除法的第一大鐵則：就算能整除，也必定是浮點數！**
請大家特別注意這個極具欺騙性的細節：
* 在數學課本上：$6 \div 3 = 2$
* 但在 Python 裡執行 `6 / 3`：算出來的結果是 **`2.0`**！
* 執行 `10 / 5`：結果是 **`2.0`**！
* 執行 `7 / 2`：結果是 **`3.5`**！

只要算式中使用了單斜線 `/`，Python 的直譯器就會自動啟動浮點數運算模式，結果保證百分之百帶有小數點（即使小數點後只有一個 0）！

**🚨 APCS 考場避坑大對比：雙斜線 `//` vs 單斜線 `/`**
1. **算「人數、個數、次數、陣列索引」**：必須用雙斜線 `//`，確保結果是整數 `int`，避免引發型態錯誤。
2. **算「平均分數、身高體重、物理速率、比例」**：必須用單斜線 `/`，精準保留小數點後面的細緻刻度！

````python
# [2] Code 範例區：展示單斜線除法 / 必定產生浮點數的特性
# 範例 1：即便能夠整除，結果依然帶有小數點 .0
result_div = 6 / 3
print(result_div)        # 印出 2.0（非整數 2）
print(type(result_div))  # <class 'float'>

# 範例 2：無法整除時，精準計算小數
score_avg = 15 / 2
print(score_avg)         # 印出 7.5

# 範例 3：對比 // 與 /
print(10 // 3)           # 整數商數：3
print(10 / 3)            # 浮點數真除法：3.3333333333333335
````

### 🔬 2-5-2 電腦底層的二進位秘密——0.1 + 0.2 != 0.3 的精度誤差真相

現在，請各位在心中準備迎接一個全宇宙所有初學程式設計師都會被震驚到的世紀謎團！
請你在 Python 裡輸入以下這行指令並執行：
```python
print(0.1 + 0.2)
```
你猜猜螢幕上會印出什麼？是 `0.3` 嗎？
不！螢幕上印出來的竟然是：**`0.30000000000000004`**！
更可怕的是，如果你在 Python 中寫：`print((0.1 + 0.2) == 0.3)`，電腦竟然會冷酷地回答你：**`False`**（錯誤）！

**難道電腦連國小一年級的數學都不會算嗎？！**
這不是電腦壞掉了，而是藏在現代電腦硬體核心最深處的「二進位精度限制（Precision Limit）」！

**🍕 切披薩比喻：為什麼除不盡？**
在我們人類熟悉的十進位中，如果把一張披薩平分給 3 個人：
$$\frac{1}{3} = 0.333333333333...$$
小數點後面的 3 會永無止境循環下去，根本無法用有限的紙筆寫完！

同樣的道理！**電腦大腦內部只懂「二進位（只有 0 與 1）」**。
當電腦把人類的十進位小數 `0.1`（十分之一）轉換成二進位小數時，它會變成：
$$0.00011001100110011...（二進位的無限循環小數！）$$
但電腦的記憶體不是無限大的（一般浮點數規格只有 64 個位元來存數字）。當容量裝滿時，電腦只能把後面無限循環的小尾巴**強行切斷捨去**！
這一截斷，就留下了像奈米般極微小的「麵包屑誤差」！當 `0.1` 加上 `0.2` 時，兩顆小麵包屑碰在一起，最後一位就多出了一個極小的 `4`！

這個現象叫做「**浮點數捨入誤差（Floating-point Round-off Error）**」，全世界所有程式語言都無法避免！

````python
# [2] Code 範例區：展示二進位浮點數精度誤差真相
# 世紀奇觀：0.1 + 0.2 究竟是多少？
sum_val = 0.1 + 0.2
print(sum_val)        # 印出 0.30000000000000004

# 直接進行相等比較會判斷為 False！
is_equal = (sum_val == 0.3)
print(is_equal)       # 印出 False！

# 減法誤差示範：1.0 - 0.9 也是浮點小數
diff_val = 1.0 - 0.9
print(diff_val)       # 印出 0.09999999999999998
````

### 🎯 2-5-3 APCS 浮點數相等的安全比對法——容許誤差（Epsilon）與 abs()

既然我們已經知道了浮點數在電腦裡會帶有「奈米級的麵包屑微小誤差」，那麼問題來了：
在 APCS 考場或實務解題時，如果題目要我們判斷：「小數 $a$ 的計算結果是否等於小數 $b$」該怎麼辦？

**🚫 考場大忌：嚴禁直接使用雙等號 `a == b` 比對浮點數！**
因為哪怕只差了 $0.00000000000000004$，雙等號 `==` 也會無情地判斷為 `False`，導致你的條件式完全失靈，直接拿到 Wrong Answer（WA）！

**💡 競技程式唯一的標準解法：容許誤差比對法（Epsilon）！**
你可以把這想像成奧運百米賽跑的「**終點高速攝影機容許誤差**」：
只要兩位選手衝線的時間差距「小於百萬分之一秒（極小容許值）」，我們就認定他們同時到達！

在數學與程式中，我們會設定一個極小的數字 $\epsilon$（Epsilon，常設為 `1e-9` 即 $10^{-9}$，也就是十億分之一 0.000000001）：
只要「**兩數相減的差值絕對值**」小於這個容許值，我們就宣告兩數相等！
$$\text{abs}(a - b) < 10^{-9}$$

**語法核心拆解**：
1. **`abs()` 函數**：求絕對值，確保不管是 $a > b$ 還是 $a < b$，差值一律都是正數距離。
2. **`1e-9`**：科學記號表示法，代表 $1 \times 10^{-9}$，是一個超級微小的臨界安全門檻。
3. 如果 `abs((0.1 + 0.2) - 0.3) < 1e-9`，結果就是百分之百令人安心的 **`True`**！

掌握這招「容許誤差判定法」，在任何演算法競賽遇到浮點數幾何比對題，你都能穩操勝券！

````python
# [2] Code 範例區：展示使用容許誤差 abs(a - b) < 1e-9 安全比對浮點數
a = 0.1 + 0.2
b = 0.3

# 錯誤比對法：直譯器判定為 False
print(a == b)  # False

# 正確比對法：計算兩者相減的絕對值差值
diff = abs(a - b)
print(diff)    # 印出 5.551115123125783e-17（極小的微小差距）

# 設定容許誤差門檻 eps = 1e-9（十億分之一）
eps = 1e-9
is_safe_equal = (diff < eps)
print(is_safe_equal)  # 印出 True！成功安全判定相等！
````

### ✂️ 2-5-4 浮點數輸出修剪術——round() 四捨五入

在 APCS 的實作考題中，題目如果涉及除法或小數運算，題目最後幾乎必定會附帶一行關鍵字：
👉 **「請輸出計算結果，四捨五入到小數點後兩位！」**

如果我們直接執行 `print(10 / 3)`，螢幕上會吐出一長串可怕的 `3.3333333333333335`。自動裁判看的是你印出來的數字本身。先把這個又長又亂的小數修剪成考卷要的位數，是這一節唯一的任務。

**內建四捨五入函數——`round(數值, 位數)`**

把它想成文具店的圓頭剪刀。第一個空格放要剪的小數，第二個空格放「小數點後面留下幾位」：

* `round(3.14159, 2)`：留下小數點後 2 位，再看下一位決定要不要進位，得到 `3.14`。
* `round(3.14159, 3)`：留下小數點後 3 位，得到 `3.142`。
* `round(10 / 3, 2)`：先做出除法的長小數，再剪成 `3.33`。

運算順序要看清楚。`round` 不會改寫原來的變數，它會交還一個新的數，所以常常寫成 `answer = round(raw, 2)`，再把 `answer` 交出去。若你寫 `round(raw, 2)` 卻沒有接住，原來的 `raw` 仍然是那串長小數。

**這一節先不處理「補零」**

`round(7.5, 2)` 得到的是數值 `7.5`，`print` 會印出 `7.5`，不會自動印成 `7.50`。因為 `7.5` 和 `7.50` 在數字世界裡是同一個大小，`print` 只是懶得把沒用的 0 畫出來。題目若硬性規定「字面上必須有兩位小數」，那是文字排版，要等第七章 7.3 的格式化字串。現在如果提前使用那套寫法，等於跳過中間還沒教的字串課。這一節只負責把數值四捨五入到指定位數。

**考場小抄**

1. 題目說四捨五入到小數點後 k 位，就寫 `round(算式, k)`。
2. 先除完，再交給 `round`。不要先 `round` 其中一個數再去除，否則位數會和題目不一致。
3. 浮點數比對相等，仍然回到 2.5.3 的 `abs(a - b) < 1e-9`，不要因為已經 `round` 過就改用 `==`。


````python
# [2] Code 範例區：展示 round() 四捨五入
val = 3.1415926

# 留下小數點後 2 位
print(round(val, 2))     # 印出 3.14

# 留下小數點後 4 位
print(round(val, 4))     # 印出 3.1416

# round 交還新數值，原本的 val 不會被改掉
trimmed = round(val, 2)
print(trimmed)           # 印出 3.14
print(val)               # 仍然是原來的長小數

# 7.5 這個數值在小數點後兩位沒有需要進位的尾巴
number = 7.5
print(round(number, 2))  # 印出 7.5，不會自動補成 7.50
````

1. **單斜線除法（/）**：神奇果汁機，不管能否整除（例如 `6 / 3 = 2.0`），產物永遠必定是浮點數 `float`！
2. **0.1 + 0.2 != 0.3 的真相**：電腦使用二進位儲存有限 64 位元浮點數，截斷無限循環小數引發的微小捨入誤差。
3. **浮點數安全比對法（Epsilon）**：考場嚴禁用 `==` 比對浮點數，標準解法是 `abs(a - b) < 1e-9` 容許誤差判定！
4. **round(數值, 位數)**：先把除法算完，再四捨五入到題目指定的小數位數。要在字面上補出 `7.50` 這種零，留到第七章 7.3。
