### 10.1.1 為什麼需要不可變容器？（串列被誤改的慘劇 vs 元組的防禦屏障）

在第八章中，我們學習了一維串列（List），它就像一塊「**隨時可擦寫的白板**」，我們可以使用 `a[0] = 99` 修改內容，或是用 `.append()` 隨時加料。串列的靈活性非常棒，但在大型軟體工程與競賽程式中，這種「太過自由」的特性有時反而是一場災難！

#### 🛡️ 生活比喻：白板 vs 雕刻在石碑上的法典
想像你正在開發一款多人連線對戰遊戲，地圖的邊界長寬是固定不變的規則（例如 `WIDTH = 800, HEIGHT = 600`）。
- 如果你把邊界存成串列 `boundary = [800, 600]`，在經過了五千行程式碼之後，某個負責算道具的迴圈可能不小心寫了 `boundary[0] = 0`，這會導致全體玩家直接掉出地圖死機！
- 為了保護這類「**絕對不允許被意外竄改的關鍵數據**」，Python 提供了一個專屬的守護容器——**元組（Tuple）**！

元組就像是「**雕刻在石碑上的文字**」或是「**密封的時空膠囊**」。一旦被建立出來，它裡面的每一個格子、每一個順序就全部被「物理鎖定」！
如果你嘗試去執行 `my_tuple[0] = 99`，Python 直譯器會立刻毫不留情地跳出大紅色的警報：
`TypeError: 'tuple' object does not support item assignment`（元組物件不支援元素修改）！
這種「只准閱讀、嚴禁竄改」的特性，在資訊科學中被稱為 **不可變性（Immutability）**。它不是缺陷，而是 Python 賦予工程師最強大的資料安全防護盾！


````python
# 範例 10.1.1：體驗串列的可變性 vs 元組的不可變防禦

# 1. 串列（List）：可隨意修改
my_list = [10, 20, 30]
print("原始串列：", my_list)
my_list[0] = 999  # 成功修改
print("修改後串列：", my_list)

# 2. 元組（Tuple）：受到不可變保護
my_tuple = (10, 20, 30)
print()
print("原始元組：", my_tuple)
print("讀取第 0 格（合法）：", my_tuple[0])

# 如果嘗試修改元組，Python 會引發 TypeError 錯誤！
# 下面這行若解除註解執行，將會報錯：
# my_tuple[0] = 999  # TypeError: 'tuple' object does not support item assignment
print("元組內容受到嚴密保護，不可被原地竄改！")
````

### 10.1.2 元組宣告語法：小括號 `(...)` 與單元素逗號陷阱 `(x,)`

在 Python 中，宣告元組的基本方式是使用**小括號 `(...)`**，並以逗號隔開各個元素。例如：`t = (1, 2, 3)`。
如果要建立一個沒有任何元素的**空元組**，直接寫一對空的小括號 `empty_t = ()`，或者調用型態建構器 `empty_t = tuple()` 即可。

#### ⚠️ 初學者頭號致命天坑：單元素逗號陷阱！
請大家仔細觀察下面這兩行程式碼，猜猜看它們的型態分別是什麼？
```python
a = (50)
b = (50,)
```
很多人會直覺認為「兩個都是長度為 1 的元組吧？」——**大錯特錯！**
- 在 Python 乃至所有程式語言中，小括號 `(...)` 最常見的職責是「**數學四則運算的優先級符號**」（例如 `(2 + 3) * 4`）。
- 當直譯器看到 `a = (50)` 時，它只會認為你給整數 50 外面套了一層沒意義的數學括號，所以 `a` 的型態依然是 **`int`（整數）**！
- 為了向 Python 明確表示「這不是數學括號，而是一個容器！」，我們**必須在唯一元素後方強制加上一個逗號 `,`**！也就是 `b = (50,)`，此時 `b` 的型態才會是貨真價實的 **`tuple`（元組）**！

這個「**單元素逗號陷阱**」是 Python 程式檢定與 APCS 語法中最常出現的考題之一。請牢記心法：**在 Python 容器世界中，逗號往往比括號更具備決定性！**


````python
# 範例 10.1.2：揭開單元素逗號陷阱的神秘面紗

# 1. 宣告空元組
t_empty = ()
print("空元組：", t_empty, "，型態：", type(t_empty), "，長度：", len(t_empty))

# 2. 多元素元組
t_multi = (10, 20, 30)
print("多元素元組：", t_multi, "，型態：", type(t_multi), "，長度：", len(t_multi))

# 3. 陷阱：沒有逗號的小括號
not_a_tuple = (100)
print("not_a_tuple 內容：", not_a_tuple, "，真實型態：", type(not_a_tuple))

# 4. 正確：帶有逗號的單元素元組
real_tuple = (100,)
print("real_tuple 內容：", real_tuple, "，真實型態：", type(real_tuple), "，長度：", len(real_tuple))
````

### 10.1.3 正負索引存取與切片操作（承襲串列直覺，唯獨嚴禁賦值修改 TypeError）

元組在 Python 家族中屬於標準的「**循序序列（Sequence Type）**」。這意味著：你在第七章學過的所有字串索引、在第八章學過的所有串列切片語法，**在元組上一模一樣完全通用**！初學者完全不需要記憶新的查詢語法！

#### 🧭 承襲串列的直覺操作：
1. **正向索引（0-based）**：`t[0]` 代表第一個元素，`t[1]` 代表第二個元素。
2. **負向索引（倒數定位）**：`t[-1]` 永遠鎖定最後一個元素，`t[-2]` 為倒數第二個。
3. **區間切片（左閉右開）**：`t[start:stop]` 截取子元組，切出來的成品**依然是一個全新的元組**！
4. **步進反轉神技**：`t[::-1]` 同樣能光速將元組進行逆序反轉！

#### 🚫 唯一的絕對禁區：不可切片賦值！
在串列中，我們可以寫 `a[1:3] = [99, 88]` 來進行區間替換；但在元組中，任何形式的「賦值給索引」或「賦值給切片」（例如 `t[1:3] = (99, 88)`）都會立即觸發 `TypeError`。
換言之：**元組的切片只能拿來「讀取」與「生成新物件」，絕對不能拿來就地修改！**


````python
# 範例 10.1.3：元組的正負索引與切片操作展示

data = ("Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun")

print("整週元組：", data)

# 1. 正向與負向索引
print("第一天（索引 0）：", data[0])
print("最後一天（索引 -1）：", data[-1])

# 2. 區間切片：平日工作日（索引 0 到 4，左閉右開寫 0:5）
weekdays = data[0:5]
print("工作日切片：", weekdays, "，型態：", type(weekdays))

# 3. 步進值切片：隔天取樣
alternate = data[::2]
print("隔日採樣：", alternate)

# 4. 反轉元組
reversed_days = data[::-1]
print("逆序反轉：", reversed_days)
````

### 10.1.4 型態互轉：`tuple(list)` 與 `list(tuple)` 的彈性切換

既然元組不能修改，那如果我們真的遇到「**非得修改元組內容不可**」的情境，該怎麼辦呢？
例如：原本有一組受保護的學生名冊元組，現在有位同學轉學離開了，我們必須將他剔除。

#### 🔄 經典解題工作流：拆封 ➔ 加工 ➔ 重新封印
在 Python 中，串列與元組具備極為絲滑的雙向轉型機制：
1. **`list(my_tuple)`**：把堅硬的石碑「拆封」放回白板串列中，重新解鎖追加、刪除與修改能力！
2. **加工操作**：在串列上執行 `.append()`、`.remove()`、`a[i] = new_val` 等彈性變動。
3. **`tuple(my_list)`**：修改完畢後，再次「重新封印」回元組石碑中，交給後續程式安全使用！

```
【受保護元組】 ➔ list() 轉型 ➔ 【可變串列】 ➔ 自由增刪改查 ➔ tuple() 轉型 ➔ 【全新受保護元組】
```

這種「**動態解凍與再次冷凍**」的技法，是許多競技程式工程師在兼顧「程式彈性」與「資料安全」時最經典的實務策略。


````python
# 範例 10.1.4：元組與串列的雙向互轉實戰

# 原始密封元組
locked_tuple = (10, 20, 30)
print("原始元組：", locked_tuple)

# 步驟一：拆封為串列
temp_list = list(locked_tuple)
print("轉換為串列：", temp_list, "，型態：", type(temp_list))

# 步驟二：自由修改內容
temp_list.append(40)
temp_list[0] = 99
print("串列修改後：", temp_list)

# 步驟三：重新封裝回元組
new_tuple = tuple(temp_list)
print("重新封裝後的全新元組：", new_tuple, "，型態：", type(new_tuple))
````

````python
# 練習 10.1.4：黑名單過濾封裝機
# 題目說明：輸入一串整數建立原始元組 T。
# 接著輸入一個整數 bad_val（黑名單數值）。
# 請將元組轉為串列，過濾移除掉「所有」數值等於 bad_val 的元素。
# 最後重新將過濾後的結果封裝為元組，並輸出其長度與內容（以空白隔開）。
# 若全部元素都被過濾掉，長度輸出 0，內容為空行。

# 【公開測試資料 1】
# 10 20 30 20 50
# 20
# 輸出：
# 3
# 10 30 50

# 【公開測試資料 2】
# 5 5 5
# 5
# 輸出：
# 0
# 

# 請在此處撰寫你的程式碼：
t = tuple(map(int, input().split()))
bad_val = int(input())

# 轉為串列並過濾
clean_list = [x for x in t if x != bad_val]

# 封裝回元組
clean_tuple = tuple(clean_list)

print(len(clean_tuple))
if len(clean_tuple) > 0:
    print(*clean_tuple)
else:
    print()
````

### 10.1.5 元組解包賦值（Tuple Unpacking）與多變數同時接收

早在本書第一章 1-3 節，我們就見識過一個令人驚豔的神奇語法：
```python
a, b = b, a  # 兩數交換
```
當時我們說這是 Python 專屬的瞬間交換魔法。現在學了元組，我們終於可以揭開它的底層真相了！
在等號右邊的 `b, a`，Python 實際上自動幫我們打包成了一個元組 `(b, a)`；而等號左邊的 `a, b`，則是在執行「**元組解包賦值（Tuple Unpacking）**」！

#### 📦 什麼是解包（Unpacking）？
解包就像是「把寄來的包裹箱子拆開，把裡面的物品一件件精準發給對應的人」。
例如：有一個裝有三項資料的元組 `point = (10, 20, 30)`。
傳統寫法：
```python
x = point[0]
y = point[1]
z = point[2]
```
優雅的元組解包一行寫法：
```python
x, y, z = point
```
Python 會自動依照位置順序，將元組內的元素一一發配給左邊的獨立變數！

#### ⚖️ 解包天平鐵則：個數必須嚴格對稱！
解包時最常見的錯誤，就是**等號左邊的變數個數與右邊元組的元素個數不相等**：
- 如果左邊人太少：`x, y = (1, 2, 3)` ➔ 報錯 `ValueError: too many values to unpack`
- 如果左邊人太多：`x, y, z = (1, 2)` ➔ 報錯 `ValueError: not enough values to unpack`
牢記解包天平法則，就能寫出既簡潔又絕對穩健的高品質程式！


````python
# 範例 10.1.5：元組解包賦值與常見錯誤解析

# 1. 基本元組解包
user_info = ("Bob", 17, "Taipei")
name, age, city = user_info

print("姓名：", name)
print("年齡：", age)
print("城市：", city)

# 2. 在迴圈中走訪多個元組並即時解包
students = [
    ("Alice", 95),
    ("Charlie", 88),
    ("David", 92)
]

print()
print("--- 成績單列表 ---")
for s_name, s_score in students:
    print(f"學生：{s_name}，成績：{s_score}")

# 3. 錯誤示範（若解除註解會報錯）：
# a, b = (10, 20, 30)  # ValueError: too many values to unpack (expected 2, got 3)
````

````python
# 填空 10.1.5：三維座標解包與曼哈頓距離計算
# 請將 ___ 替換為正確的變數名稱

coord_3d = (3, 7, 12)

# 任務：將三維座標一次性解包給 x, y, z 三個變數
x, ___, z = coord_3d

# 計算從原點 (0, 0, 0) 到該點的三維曼哈頓距離
distance = abs(x) + abs(y) + abs(z)
print("三維座標：", x, y, z)
print("原點距離：", distance)
````

````python
# 練習 10.1.5：選手全能競賽解包計算機
# 題目說明：輸入包含四個欄位的資料：
# 選手姓名（字串）、跑步分數（整數）、跳遠分數（整數）、射擊分數（整數）。
# 請將這四筆資料包裝成一個元組，隨後以解包方式分別賦值給 name, run, jump, shoot 四個變數。
# 計算三項運動的總分，並依格式輸出兩行：
# 第一行：選手姓名
# 第二行：總分

# 【公開測試資料 1】
# Tom 85 90 95
# 輸出：
# Tom
# 270

# 【公開測試資料 2】
# Leo 100 70 80
# 輸出：
# Leo
# 250

# 請在此處撰寫你的程式碼：
parts = input().split()
record = (parts[0], int(parts[1]), int(parts[2]), int(parts[3]))

# 元組解包
name, run, jump, shoot = record

total_score = run + jump + shoot
print(name)
print(total_score)
````

````python
# 挑戰 10.1.5：兩人對戰成績解包判定器
# 題目說明：連續輸入兩行。
# 每行包含：選手名稱與三局得分（共四個欄位）。
# 請各自打包並解包，計算各自三局總分。
# 輸出總分較高的選手姓名與其總分；若總分相同輸出 "TIE"。
# 本題無公開測試資料，請自行測試。

# 請在此處撰寫你的程式碼：
p1_data = input().split()
name1, s1, s2, s3 = p1_data[0], int(p1_data[1]), int(p1_data[2]), int(p1_data[3])
tot1 = s1 + s2 + s3

p2_data = input().split()
name2, t1, t2, t3 = p2_data[0], int(p2_data[1]), int(p2_data[2]), int(p2_data[3])
tot2 = t1 + t2 + t3

if tot1 > tot2:
    print(name1, tot1)
elif tot2 > tot1:
    print(name2, tot2)
else:
    print("TIE")
````

### 10.1.6 二維座標封裝：以 `(r, c)` 元組表示平面網格點

在第九章二維陣列中，我們每次記錄地圖上的一個格子，都需要同時維護「列座標 $r$」與「行座標 $c$」。如果地圖上有 10 顆炸彈或 5 隻魔王，初學者往往會開兩個平行的串列：
```python
bomb_r = [1, 3, 5]
bomb_c = [2, 4, 6]
```
這種分開存儲的方式非常危險，一旦排序或增減元素時漏掉其中一個，座標就會立刻錯位！

#### 🎯 最佳實踐：將座標打包為元組 `(r, c)`
透過本單元所學的元組，我們可以將位置封裝成單一實體物件：
```python
bombs = [(1, 2), (3, 4), (5, 6)]
```
這樣做帶來三大極致優勢：
1. **天然綁定、永不錯位**：$r$ 與 $c$ 形影不離，在排序、傳遞或增刪時永遠是一個整體。
2. **走訪解包無比直覺**：在迴圈中可以直接寫 `for r, c in bombs:`，直觀又工整。
3. **為雜湊標記鋪墊（關鍵先導）**：元組具備不可變性，因此它可以被直接放進後續單元的 `set`（集合）中當作「已走訪標記清單」！

從分散的零碎變數，進化到使用「座標元組 `(r, c)`」進行二維網格運算，是每一位初學者晉升 APCS 中高級實作選手的核心分水嶺！


````python
# 範例 10.1.6：二維座標元組封裝與網格走訪實戰

# 網格尺寸：4 列 5 行
R, C = 4, 5

# 記錄三顆地雷的座標元組清單
mines = [
    (0, 1),
    (2, 3),
    (3, 0)
]

print(f"地雷清單（共 {len(mines)} 顆）：", mines)

# 建立 4x5 地圖，初始全為 "."
grid = [["."] * C for _ in range(R)]

# 走訪座標元組並解包，在地圖上打上 "*" 標記
for r, c in mines:
    grid[r][c] = "*"

print()
print("地雷分佈地圖：")
for row in grid:
    print(*row)
````

````python
# 填空 10.1.6：座標元組走訪與四邊界檢驗
# 請將 ___ 替換為正確的變數或條件

R, C = 5, 5
points = [(1, 2), (4, 4), (2, 0), (3, 3)]

inside_count = 0

# 走訪每一個座標元組並即時解包為 r, c
for r, ___ in points:
    # 檢查座標是否嚴格位於網格內部（不含最外圍邊界：0 < r < R-1 且 0 < c < C-1）
    if 0 < r < R - 1 and 0 < c < C - 1:
        inside_count += 1

print("嚴格位於內部的點數：", inside_count)
````

````python
# 練習 10.1.6：曼哈頓兩點距離元組計算機
# 題目說明：連續輸入兩行。
# 第一行輸入兩個整數 r1, c1 代表起點座標。
# 第二行輸入兩個整數 r2, c2 代表終點座標。
# 請將兩點分別包裝為元組 p1 = (r1, c1) 與 p2 = (r2, c2)。
# 接著透過解包計算兩點間的曼哈頓距離：|r1 - r2| + |c1 - c2|，並印出距離值。

# 【公開測試資料 1】
# 1 2
# 4 6
# 輸出：
# 7

# 【公開測試資料 2】
# 0 0
# 3 3
# 輸出：
# 6

# 請在此處撰寫你的程式碼：
r1, c1 = map(int, input().split())
r2, c2 = map(int, input().split())

p1 = (r1, c1)
p2 = (r2, c2)

# 解包計算
dist = abs(p1[0] - p2[0]) + abs(p1[1] - p2[1])
print(dist)
````

````python
# 挑戰 10.1.6：多點軌跡中心點計算
# 題目說明：第一行輸入整數 N（代表點的個數）。
# 接著有 N 行，每行輸入兩個整數 r, c 代表點的座標。
# 請將所有座標存入一個元組串列中。
# 計算所有點的 r 座標平均值與 c 座標平均值（皆使用整數除法 // N 向下取整）。
# 輸出該平均重心點座標（以 (r, c) 元組格式輸出）。
# 本題無公開測試資料，請自行測試。

# 請在此處撰寫你的程式碼：
n = int(input())
pts = []
for _ in range(n):
    r, c = map(int, input().split())
    pts.append((r, c))

sum_r = sum(p[0] for p in pts)
sum_c = sum(p[1] for p in pts)

center = (sum_r // n, sum_c // n)
print(center)
````

## 本單元重點回顧與核心心法

在單元 10-1 中，我們正式推開了 Python 高級資料結構的大門，深入掌握了不可變序列——**元組（Tuple）**：

1. **不可變性（Immutability）的防護屏障**：
   - 元組如同雕刻在石碑上的文字，建立後嚴禁原地修改（`t[i] = x` 會報錯 `TypeError`），有效杜絕競賽與工程中的意外竄改。
2. **單元素逗號陷阱**：
   - `(x)` 只是純數學括號，其型態仍為元素自身型態；只有加上逗號的 `(x,)` 才是長度為 1 的元組！
3. **承襲序列直覺操作**：
   - 正負索引、左閉右開切片與步進反轉 `t[::-1]` 完全相容串列語法，但切片僅供讀取與衍生新物件。
4. **型態互轉解凍工作流**：
   - 透過 `list(t)` 拆封修改、再以 `tuple(l)` 重新封印，彈性兼顧資料變更與安全保護。
5. **元組解包賦值（Tuple Unpacking）**：
   - 以 `a, b, c = t` 對稱接收各欄位，等號兩端元素個數必須嚴格相等，否則引發 `ValueError`。
6. **二維網格座標打包**：
   - 以 `(r, c)` 元組將列與行封裝為獨立實體，杜絕多串列錯位，為後續雜湊集合打下堅實基礎！
