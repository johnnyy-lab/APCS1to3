### 🚫 1-6-1 行首嚴禁任意留白——抓出出列站錯的 IndentationError

### 🐥 1-6-2 縮排界定程式區塊——「母雞帶小鴨」的層級從屬概念

既然平常不能隨便加空白，那為什麼 Python 還要設計「**縮排**」這個功能呢？

在其他古老的程式語言（例如 C 語言、C++ 或 Java）中，如果想要把一群指令打包成一個團體，必須使用繁瑣的大括號 `{ }` 把大家關在裡面；
但是 Python 的發明者非常聰明，他認為大括號又醜又容易漏掉，因此發明了全世界最優雅的「**縮排區塊架構（Code Block）**」！

你可以把縮排想像成「**母雞帶小鴨**」：
當母雞隊長（例如後續章節會學到的 `if` 條件判斷）站出來，後面跟著一個冒號 `:` 時，就代表母雞吹哨準備集合隊伍了！
此時，所有屬於這隻母雞管轄的小鴨指令，全部都必須整齊劃一地往右「縮排 4 個空格」，躲在母雞的羽翼下方！

看縮排就能一眼看出從屬關係：
* **往右縮進去**：代表「我是母雞隊長的手下，只有母雞同意時才輪得到我執行」！
* **回到最左邊**：代表「我跟母雞平起平坐，母雞做完事情之後就換我上場」！

這就是為什麼 Python 代碼看起來總是乾淨整潔、像詩一樣排版整齊的核心原因！

```python
# [2] Code 範例區：展示冒號 : 後方的縮排區塊（提早預覽 if 結構之縮排）
sunny = True

if sunny:
    # 以下縮排 4 空格的指令屬於 if 區塊管轄（小鴨部隊）
    print("太陽出來了！")
    print("準備出門去郊遊！")

# 縮排退回最左邊，代表離開了 if 區塊
print("程式順利結束！")
```

### 📐 1-6-3 空格與 Tab 絕不混用——守護建築結構的 TabError 警告

在學習縮排時，鍵盤上有兩個鍵都可以產生縮排效果：
1. 鍵盤左邊的 **Tab 鍵**（按一下通常會跳格）。
2. 鍵盤下方的 **空白鍵 Space**（連續按 4 下）。

這時候，很多初學者就會隨興所欲：第一行按了一下 Tab，第二行按了 4 下空白，第三行又按了 Tab。
吳邦一老師在講義第 20 頁嚴厲警告大家：
「**縮排可用 [tab] 鍵，也可以用若干空白（習慣上用 4 個，有人喜歡 2 個），但同一個區塊要一致，不要混用！**」

為什麼不能混用？
想像一下兩位工匠一起蓋大樓：工匠 A 拿公分的尺，工匠 B 拿英吋的尺。雖然在人眼看起來兩根柱子好像差不多高，但在電腦精密直譯器看來，Tab 字元是 `\t`，而空白鍵是空白字元，兩者根本是完全不同的東西！
如果同一個區塊混用了兩者，Python 3 就會立刻拋出建築崩塌的大警告：
`TabError: inconsistent use of tabs and spaces in indentation`
（白話文：縮排中混用了不一致的 Tab 與空格！）

**國際標準與 APCS 考場唯一推薦**：
全世界 Python 官方與專業工程師一致約定俗成：**統一使用「4 個空白鍵（Space * 4）」作為標準縮排！**（在 Google Colab 或現代編輯器中，只要按 Tab 鍵，編輯器都會聰明地自動幫你轉成 4 個純空白，保證安全無虞喔！）

```python
# [2] Code 範例區：嚴格一致！統一使用 4 個純空白鍵作為縮排
check = True
if check:
    step1 = 100   # 前方 4 個純空格
    step2 = 200   # 前方 4 個純空格（與上一行完全齊頭）
    print(step1 + step2)
```

### 🎨 1-6-4 優雅的呼吸留白——中間與後方的美化空白規範

學完了嚴格的行首縮排紀律後，讓我們來放鬆一下！
吳邦一老師在講義第 4 頁告訴我們：
「**指令的中間與後方可以加空白！**」

請看下面這兩行一模一樣功能的程式碼：
* 寫法 A（擁擠窒息風）：`total=base_price*tax_rate+shipping_fee`
* 寫法 B（優雅呼吸風）：`total = base_price * tax_rate + shipping_fee`

你喜歡看哪一種呢？相信任何人都會投票給寫法 B！
因為在運算子（如 `=`, `+`, `-`, `*`）的左右兩側各留出一個空白，就像是在每個英文單字之間留一個空格一樣，能讓整個算式具有節奏感與呼吸空間，大幅減少眼睛看錯算式的疲勞！

在國際 Python 程式風格指南（PEP 8）中，強烈建議大家養成「**等號與加減乘除兩邊留一個空白**」的良好習慣。
記住：最前面不能隨便留白（那是縮排的專利），但中間的適度留白，則是每一位專業程式設計師的優雅品味喔！

```python
# [2] Code 範例區：優雅示範！運算子兩側適度保留一個空格，清晰易讀
item_price = 100
tax = 5
shipping = 60

# 運算子兩側適度留白，排版舒爽好看
final_total = item_price + tax + shipping
print(final_total)
```
