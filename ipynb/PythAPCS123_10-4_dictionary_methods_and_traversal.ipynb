## 10.4.1 安全取值神器：d.get(key, default) 徹底杜絕 KeyError 崩潰

### 觀念說明
在前面的學習中，我們知道直接使用中括號語法 `d[key]` 取值，若該鍵不存在於字典中，Python 會直接拋出 `KeyError` 例外錯誤並強制終止程式運行。在競賽或實際開發中，輸入資料往往充滿未知數或缺漏，我們不能讓程式輕易崩潰。

此時，`d.get(key, default)` 就是最安全可靠的利器！
1. **如果 `key` 存在**：返回對應的 `value`，第二個參數會被忽略。
2. **如果 `key` 不存在**：不會報錯，而是返回指定的 `default` 預設值。
3. **若省略 `default`**：預設會返回 `None`。

在 APCS 的計數統計或查表題中，利用 `d.get(key, 0)` 更是設定初值的標準寫法，省去了繁瑣的 `if key in d:` 條件判斷，讓程式碼既精簡又穩健。

````python
# 10.4.1 範例：安全查詢學生成績
scores = {"Alice": 95, "Bob": 82, "Charlie": 68}

# 查詢存在的鍵
alice_score = scores.get("Alice", 0)
print("Alice 的成績:", alice_score)

# 查詢不存在的鍵，若不用 get 會引發 KeyError
# print(scores["David"]) # 這行會報錯！
david_score = scores.get("David", 0)
print("David 的成績（查無此人預設為 0）:", david_score)

# 不指定預設值時，預設回傳 None
eva_score = scores.get("Eva")
print("Eva 的成績（預設 None）:", eva_score)

# 搭配自訂提示字串
status = scores.get("Frank", "未到考")
print("Frank 狀態:", status)
````

````python
# 10.4.1 填空題：使用 get() 查詢餐點價格
menu = {"Burger": 85, "Fries": 40, "Coke": 30}

item_name = "Pizza"
# 任務：請使用 get 方法查詢 item_name 的價格，若菜單沒有提供，預設價格為 -1
price = menu.get(item_name, -1)

# 請將上行的答案填入下方空格中：
# price = menu.___(item_name, ___)
print("查詢項目:", item_name, "價格:", price)
# 預期輸出: 查詢項目: Pizza 價格: -1
````

````python
# 10.4.1 練習題：單字積分查詢器
# 題目說明：
# 某遊戲中每個單字都有對應的分數。給定單字分數字典 word_points，
# 請依序查詢輸入串列 search_words 中的每個單字。
# 若單字在字典中，累加其對應分數；若不在字典中，視為 0 分。
# 最後印出所有查詢單字的總得分。
#
# 【公開測試資料 1】
# 字典: {"cat": 5, "dog": 8, "bird": 12}
# 查詢清單: ["dog", "fish", "cat"]
# 輸出: 總得分: 13
#
# 【公開測試資料 2】
# 字典: {"apple": 10, "banana": 15}
# 查詢清單: ["orange", "grape"]
# 輸出: 總得分: 0

word_points = {"cat": 5, "dog": 8, "bird": 12}
search_words = ["dog", "fish", "cat"]

total_score = 0
for w in search_words:
    total_score += word_points.get(w, 0)

print("總得分:", total_score)
````

````python
# 10.4.1 挑戰題：計數器累加實戰（不使用 if-else）
# 題目說明：
# 給定一串字母串列 tokens = ["a", "b", "a", "c", "b", "a"]
# 請建立一個空字典 counts = {}
# 使用 for 迴圈搭配 counts.get(ch, 0) + 1 的技巧，統計每個字母出現的次數。
# （不可使用 collections.Counter，純用字典 get() 完成）
# 提示：每次將字母放入字典時，其新數值為當前次數 counts.get(ch, 0) 再加 1

tokens = ["a", "b", "a", "c", "b", "a"]
counts = {}

# 請在下方寫出你的解答代碼
for ch in tokens:
    counts[ch] = counts.get(ch, 0) + 1

print("字母統計結果:", counts)
# 預期結果: {'a': 3, 'b': 2, 'c': 1}
````

## 10.4.2 成員檢查：key in d 與 key not in d 的 O(1) 極速存在性判定

### 觀念說明
在很多演算法題目中，我們需要在集合或清單中快速確認「某個資料是否已經出現過」。
- 如果在串列（list）中使用 `x in my_list`，Python 必須從頭到尾一個一個比對，時間複雜度為 $O(N)$。若資料有十萬筆，查詢一次就要花費十萬步，極易超時（TLE）。
- 但如果是在字典（dict）中使用 `key in d` 或 `key not in d`，Python 是利用**雜湊表（Hash Table）**直接計算記憶體位置，平均時間複雜度僅為 **$O(1)$**（常數時間）！不論字典裡有一千筆還是一百萬筆資料，查詢速度幾乎瞬間完成。

### 關鍵注意點
`in` 與 `not in` 運算子直接套用在字典上時，**預設只會檢查「鍵（keys）」**，而**不會**去檢查「值（values）」！這一點一定要牢牢記住。

````python
# 10.4.2 範例：會員權限檢查與 O(1) 判定
vip_members = {
    "user_101": "Gold",
    "user_102": "Silver",
    "user_105": "Platinum"
}

query_user = "user_102"
if query_user in vip_members:
    print(query_user, "是 VIP 會員！等級為:", vip_members[query_user])
else:
    print(query_user, "不是 VIP 會員。")

# 使用 not in 檢查未註冊者
new_guest = "user_999"
if new_guest not in vip_members:
    print(new_guest, "尚未註冊 VIP，引導前往註冊流程。")

# 注意：in 檢查的是 key，而非 value
print("Gold in vip_members:", "Gold" in vip_members) # False！因為 Gold 是值，不是鍵
````

````python
# 10.4.2 練習題：黑名單訪客即時過濾系統
# 題目說明：
# 系統中存有一份違規用戶黑名單 blacklist 字典，鍵為用戶 IP。
# 現有訪客進入佇列 visitor_logs，請依序檢查每位訪客的 IP：
# 若在黑名單中，印出 "封鎖: <IP>"
# 若不在黑名單中，印出 "放行: <IP>"
#
# 【公開測試資料 1】
# 黑名單: {"192.168.1.10": "濫發請求", "10.0.0.5": "惡意爬蟲"}
# 訪客紀錄: ["192.168.1.10", "192.168.1.100", "10.0.0.5"]
# 輸出:
# 封鎖: 192.168.1.10
# 放行: 192.168.1.100
# 封鎖: 10.0.0.5
#
# 【公開測試資料 2】
# 黑名單: {"127.0.0.1": "本機測試"}
# 訪客紀錄: ["172.16.0.1"]
# 輸出: 放行: 172.16.0.1

blacklist = {"192.168.1.10": "濫發請求", "10.0.0.5": "惡意爬蟲"}
visitor_logs = ["192.168.1.10", "192.168.1.100", "10.0.0.5"]

for ip in visitor_logs:
    if ip in blacklist:
        print("封鎖:", ip)
    else:
        print("放行:", ip)
````

````python
# 10.4.2 挑戰題：兩數之和（Two Sum）APCS 經典 O(N) 雜湊解法
# 題目說明：
# 給定整數串列 nums = [2, 7, 11, 15] 以及目標值 target = 9
# 請使用一個字典 seen = {} 來記錄走訪過的數值（鍵為數值，值為其索引）。
# 遍歷每個元素與其索引，計算 complement = target - num：
# 若 complement in seen，表示找到了兩個數相加等於 target，
# 請印出這兩個數的索引對應：(seen[complement], 當前索引)，並結束搜尋。
# 若不存在，則將當前數值與索引存入 seen 中。

nums = [2, 7, 11, 15]
target = 9
seen = {}

# 請在下方寫出你的解答代碼
found_pair = None
for idx in range(len(nums)):
    num = nums[idx]
    complement = target - num
    if complement in seen:
        found_pair = (seen[complement], idx)
        break
    seen[num] = idx

print("找到目標組合索引:", found_pair)
# 預期結果: 找到目標組合索引: (0, 1)
````

## 10.4.3 鍵走訪：for k in d: 或 d.keys() 循序走訪

### 觀念說明
當我們需要逐一處理字典中的每一筆資料時，最直覺的方式就是**走訪所有的「鍵」**。
在 Python 中，有兩種等價的寫法：
1. `for k in d:` （最常見、最地道的寫法，直接對字典進行迭代）
2. `for k in d.keys():` （顯式調用 `keys()` 方法，語意非常明確）

從 Python 3.7 開始，字典**嚴格保證保留鍵的插入順序（Insertion Order）**，因此你走訪出來的鍵，順序一定與當初加入字典時一致。

### 如何取得對應的值？
在鍵走訪的迴圈內部，只要透過 `d[k]` 就能輕鬆取得該鍵對應的數值。此外，`d.keys()` 返回的是一個「字典視圖（dict_keys）」，它會隨著字典內容動態變化，若需要排序或索引，可以透過 `sorted(d.keys())` 或 `list(d.keys())` 進行轉換。

````python
# 10.4.3 範例：走訪通訊錄與依鍵排序列印
contacts = {"Charlie": "0933-111", "Alice": "0912-345", "Bob": "0922-888"}

print("--- 1. 預設走訪（插入順序）---")
for name in contacts:
    phone = contacts[name]
    print("聯絡人:", name, "電話:", phone)

print()
print("--- 2. 使用 sorted(contacts.keys()) 依姓名排序走訪 ---")
for name in sorted(contacts.keys()):
    print("排序聯絡人:", name, "->", contacts[name])
````

````python
# 10.4.3 填空題：走訪印出庫存不足的商品
stock = {"Notebook": 45, "Pen": 3, "Ruler": 2, "Eraser": 18}

shortage_items = []
# 任務：走訪 stock 的每個鍵 item，若 stock[item] < 5，將 item 加到 shortage_items
for item in stock:
    if stock[item] < 5:
        shortage_items.append(item)

# 請填入空格：
# for item in ___:
#     if stock[___] < 5:
#         shortage_items.append(item)
print("庫存不足清單:", shortage_items)
# 預期輸出: 庫存不足清單: ['Pen', 'Ruler']
````

````python
# 10.4.3 練習題：及格名單公告系統
# 題目說明：
# 給定一份學生數學成績字典 exam_results。
# 請依字典中的鍵走訪所有學生：
# 若該學生分數 >= 60，印出 "[合格] <學生名>: <分數>分"
# 若該學生分數 < 60，印出 "[需加強] <學生名>: <分數>分"
#
# 【公開測試資料 1】
# 成績字典: {"Tom": 75, "Ken": 52, "Amy": 90}
# 輸出:
# [合格] Tom: 75分
# [需加強] Ken: 52分
# [合格] Amy: 90分
#
# 【公開測試資料 2】
# 成績字典: {"Leo": 60, "Zoe": 59}
# 輸出:
# [合格] Leo: 60分
# [需加強] Zoe: 59分

exam_results = {"Tom": 75, "Ken": 52, "Amy": 90}

for student in exam_results.keys():
    score = exam_results[student]
    if score >= 60:
        print("[合格] " + student + ": " + str(score) + "分")
    else:
        print("[需加強] " + student + ": " + str(score) + "分")
````

````python
# 10.4.3 挑戰題：依字母順序格式化輸出菜單
# 題目說明：
# 給定咖啡店價格表 cafe_menu = {"Mocha": 130, "Latte": 120, "Americano": 85, "Espresso": 75}
# 請將所有品項「依照品項名稱的字母順序（A 到 Z）」排序後依序輸出。
# 每行格式如下：
# <編號>. <品名> : $<價格>
# 編號從 1 開始計數。

cafe_menu = {"Mocha": 130, "Latte": 120, "Americano": 85, "Espresso": 75}

# 請在下方寫出你的解答代碼
count = 1
for item in sorted(cafe_menu.keys()):
    print(str(count) + ". " + item + " : $" + str(cafe_menu[item]))
    count += 1
# 預期結果:
# 1. Americano : $85
# 2. Espresso : $75
# 3. Latte : $120
# 4. Mocha : $130
````

## 10.4.4 值走訪：d.values() 與極值/總和運算

### 觀念說明
在某些統計與彙整場景中，我們只在乎字典中的「數值」，完全不需要在乎鍵是什麼。例如：計算整個月的總營業額、全班平均成績、或是尋找最高與最低銷售量。

此時，使用 `d.values()` 可以直接取出包含所有值（values）的視圖物件：
- **`sum(d.values())`**：計算字典中所有數值的總和。
- **`max(d.values())`**：找出字典中的最大數值。
- **`min(d.values())`**：找出字典中的最小數值。
- **`len(d)`**：取得資料筆數，搭配總和即可計算出「平均值」：`sum(d.values()) / len(d)`。

透過 `d.values()`，我們不需要在迴圈裡重複做 `total += d[k]`，一秒就能完成數據彙整，這在 APCS 數據處理題型中既簡短又不易出錯。

````python
# 10.4.4 範例：月度銷售額極值與統計
monthly_sales = {
    "Jan": 12000,
    "Feb": 18500,
    "Mar": 14200,
    "Apr": 21000,
    "May": 16800
}

# 1. 取得所有數值並走訪
all_sales = monthly_sales.values()
print("values 視圖:", all_sales)

# 2. 使用內建函式快速統計
total = sum(monthly_sales.values())
best = max(monthly_sales.values())
worst = min(monthly_sales.values())
avg = total / len(monthly_sales)

print("總營業額:", total)
print("最高單月營業額:", best)
print("最低單月營業額:", worst)
print("平均單月營業額:", avg)
````

````python
# 10.4.4 練習題：數據離群值計數
# 題目說明：
# 給定某感測器在一週內的溫度讀數字典 sensor_data。
# 請先利用 sum() 與 len() 計算出該週的平均溫度 avg_temp。
# 接著走訪 sensor_data.values()，計算有幾天的溫度嚴格高於平均溫度（> avg_temp）。
# 最後印出平均溫度（保留一位小數）與高於平均溫度的天數。
#
# 【公開測試資料 1】
# 溫度資料: {"Mon": 25.0, "Tue": 26.0, "Wed": 30.0, "Thu": 31.0, "Fri": 28.0}
# 計算: 總和 140.0, 筆數 5, 平均 28.0
# 嚴格高於 28.0 的有 Wed(30.0), Thu(31.0)，共 2 天
# 輸出:
# 平均溫度: 28.0
# 高於平均天數: 2
#
# 【公開測試資料 2】
# 溫度資料: {"Sat": 20.0, "Sun": 20.0}
# 輸出:
# 平均溫度: 20.0
# 高於平均天數: 0

sensor_data = {"Mon": 25.0, "Tue": 26.0, "Wed": 30.0, "Thu": 31.0, "Fri": 28.0}

avg_temp = sum(sensor_data.values()) / len(sensor_data)
above_avg_count = 0
for temp in sensor_data.values():
    if temp > avg_temp:
        above_avg_count += 1

print("平均溫度:", round(avg_temp, 1))
print("高於平均天數:", above_avg_count)
````

## 10.4.5 鍵值對走訪：for k, v in d.items(): 雙變數解包走訪

### 觀念說明
如果我們同時需要「鍵」與「值」，雖然可以寫成 `for k in d: v = d[k]`，但最 Pythonic、最優雅且效能最好的方式，絕對是使用 **`d.items()`**！

### 雙變數解包（Unpacking）
`d.items()` 會返回由 `(key, value)` 元組組成的視圖物件。配合 Python 的多變數賦值機制，我們可以在 `for` 迴圈中同時宣告兩個變數：
```python
for key, value in d.items():
    print(key, value)
```
這樣寫的好處：
1. **免除二次查表**：不需要再呼叫 `d[key]` 重新做雜湊計算，效能更佳。
2. **語意清晰**：變數名稱能直接代表意義（例如 `for name, score in scores.items():`）。
3. **簡潔俐落**：大幅提升 APCS 程式題的撰寫速度與排版美觀。

````python
# 10.4.5 範例：員工考績調整與雙變數走訪
salaries = {"Alice": 50000, "Bob": 42000, "Charlie": 48000}

print("--- 原始薪資表 ---")
for name, pay in salaries.items():
    print("員工:", name, "| 月薪:", pay)

print()
print("--- 調薪篩選通知 ---")
# 若薪資未滿 45000，加薪 10%
for name, pay in salaries.items():
    if pay < 45000:
        new_pay = int(pay * 1.1)
        print(name, "獲得調薪機會！新薪資:", new_pay)
````

````python
# 10.4.5 填空題：雙變數解包列印菜單
drink_menu = {"Green Tea": 35, "Milk Tea": 50, "Oolong Tea": 40}

affordable_drinks = []
# 任務：使用 items() 同時取得 drink 與 price，若價格 <= 40，將 drink 加入清單
for drink, price in drink_menu.items():
    if price <= 40:
        affordable_drinks.append(drink)

# 請填入空格：
# for drink, price in drink_menu.___():
#     if price <= 40:
#         affordable_drinks.append(___)
print("平價茶飲:", affordable_drinks)
# 預期輸出: 平價茶飲: ['Green Tea', 'Oolong Tea']
````

````python
# 10.4.5 練習題：最高分得主尋找（不依賴自訂函式）
# 題目說明：
# 給定一份玩家積分字典 leaderboard。
# 請使用 items() 走訪每一位玩家與分數，找出獲得「最高分」的玩家姓名與分數。
# （若有多人同分，記錄第一位遇到的即可）
# 最後依格式印出冠軍資訊。
#
# 【公開測試資料 1】
# 積分表: {"player_A": 1500, "player_B": 2850, "player_C": 2100}
# 輸出: 冠軍: player_B 得分: 2850
#
# 【公開測試資料 2】
# 積分表: {"IronMan": 98, "Thor": 95, "Hulk": 92}
# 輸出: 冠軍: IronMan 得分: 98

leaderboard = {"player_A": 1500, "player_B": 2850, "player_C": 2100}

top_player = ""
top_score = -1

for name, score in leaderboard.items():
    if score > top_score:
        top_score = score
        top_player = name

print("冠軍:", top_player, "得分:", top_score)
````

````python
# 10.4.5 挑戰題：反轉字典（鍵值互換 Invert Dictionary）
# 題目說明：
# 給定一個國碼對應國家名稱的字典 code_to_country = {"TW": "Taiwan", "JP": "Japan", "US": "USA"}
# 假設每個值都是唯一的，請透過 items() 走訪該字典，
# 建立一個新的反向字典 country_to_code，使得國家名稱變成鍵，國碼變成值。

code_to_country = {"TW": "Taiwan", "JP": "Japan", "US": "USA"}
country_to_code = {}

# 請在下方寫出你的解答代碼
for code, country in code_to_country.items():
    country_to_code[country] = code

print("反轉後的字典:", country_to_code)
# 預期結果: {'Taiwan': 'TW', 'Japan': 'JP', 'USA': 'US'}
````

## 10.4.6 字典生成式（Dictionary Comprehension）：{k: v for ... if ...}

### 觀念說明
在第 9 章中，我們見識過強大無比的「串列生成式（List Comprehension）」。Python 同樣將這種極具表現力的語法延伸到了字典上，稱為**字典生成式（Dictionary Comprehension）**！

### 語法結構
```python
{key_expr: value_expr for item in iterable if condition}
```
與串列生成式的主要差異在於：
1. 最外層改用**大括號 `{}`** 包裹。
2. 前方必須是符合字典鍵值對格式的 **`key_expr: value_expr`**（中間有冒號）。

### 常見應用情境
- **快速批量運算**：將所有商品的售價打九折。
- **資料篩選過濾**：只保留及格的學生成績，剔除不及格者。
- **反向建立索引**：將既有字典的鍵值對調（例如 `{v: k for k, v in d.items()}`）。

````python
# 10.4.6 範例：字典生成式的多樣應用
original_prices = {"Apple": 50, "Banana": 30, "Cherry": 120, "Durian": 200}

# 1. 批量打 9 折（四捨五入取整數）
discount_prices = {item: int(price * 0.9) for item, price in original_prices.items()}
print("全面九折後:", discount_prices)

# 2. 搭配 if 進行篩選（只保留價格大於 100 的高級水果）
luxury_fruits = {item: price for item, price in original_prices.items() if price >= 100}
print("高級水果清單:", luxury_fruits)

# 3. 從串列直接生成平方對照字典
nums = [1, 2, 3, 4, 5]
squares_dict = {x: x ** 2 for x in nums}
print("數字平方字典:", squares_dict)
````

````python
# 10.4.6 填空題：生成及格學生字典
raw_scores = {"Peter": 85, "Mary": 58, "John": 72, "Lisa": 45}

# 任務：使用字典生成式，篩選出分數 >= 60 的學生
passed_students = {name: score for name, score in raw_scores.items() if score >= 60}

# 請填入空格：
# passed_students = {name: score for name, score in raw_scores.___() if score >= ___}
print("及格學生:", passed_students)
# 預期輸出: 及格學生: {'Peter': 85, 'John': 72}
````

````python
# 10.4.6 練習題：字串長度索引字典
# 題目說明：
# 給定一個單字串列 words。
# 請使用「字典生成式」，將單字長度 >= 4 的單字建立成字典：
# 鍵（key）為單字本身，值（value）為該單字的長度（len(w)）。
#
# 【公開測試資料 1】
# 單字串列: ["cat", "elephant", "dog", "tiger"]
# 符合長度 >= 4 的有: "elephant"(8), "tiger"(5)
# 輸出: {'elephant': 8, 'tiger': 5}
#
# 【公開測試資料 2】
# 單字串列: ["hi", "apcs", "python", "go"]
# 輸出: {'apcs': 4, 'python': 6}

words = ["cat", "elephant", "dog", "tiger"]

length_dict = {w: len(w) for w in words if len(w) >= 4}

print(length_dict)
````

````python
# 10.4.6 挑戰題：單行字典反轉與大寫轉換
# 題目說明：
# 給定字典 mapping = {"a": 1, "b": 2, "c": 3, "d": 4}
# 請只用「一行字典生成式」達成以下兩個效果：
# 1. 將鍵與值互換（原來的數字變成鍵）
# 2. 將原來的小寫字母轉換為大寫字母作為新的值（使用 .upper()）
# 3. 只處理原數字為偶數（v % 2 == 0）的項目！

mapping = {"a": 1, "b": 2, "c": 3, "d": 4}

# 請在下方寫出你的單行字典生成式解答
result = {v: k.upper() for k, v in mapping.items() if v % 2 == 0}

print("偶數反轉大寫字典:", result)
# 預期結果: 偶數反轉大寫字典: {2: 'B', 4: 'D'}
````
