### 2. 完整題目描述（Problem Description）
在一座 $m 	imes n$ 的展場平面網格中，為了引導參觀動線，主辦單位會在網格的特定方格放置「木樁」，並在相鄰的木樁之間拉起「動線（線段）」。
動線與木樁的設置規則如下：
1. **動線拉伸規則**：當兩個木樁位於**同一橫列（同一行）**或**同一直行（同一列）**時，若兩者之間**沒有任何其他木樁阻隔**，則兩木樁之間會自動拉起一條筆直的動線，並涵蓋兩者之間的所有方格。
2. **操作類型**：主辦單位會依序執行 $h$ 次動態操作，每次操作給定座標 $(r, c)$ 與操作代碼 $t$：
   * **加入木樁（$t = 0$）**：在座標 $(r, c)$ 設置一根新木樁。若該座標原本已有動線經過，必須**先拆除該位置原本經過的舊線段**，接著放置木樁，並向上下左右四個方向搜尋最近的木樁拉出新動線。
   * **移除木樁（$t = 1$）**：將座標 $(r, c)$ 處的木樁拔除，**同時所有與該木樁直接相連的動線均一併瓦解移除**（注意：被拆除的線段不會自動重新連回後方的其他木樁）。
3. **交錯線處理**：若同一方格同時有「水平動線」與「垂直動線」通過，該方格視為雙重覆蓋（十字交錯）。當其中一條動線被拆除時，另一條動線必須完好保留。

請撰寫程式，模擬這 $h$ 次操作過程，並計算出：
1. 在整個操作過程中，網格中「有木樁或有動線佔據的總方格面積（格數）」所出現過的**歷史最大值**。
2. 在所有 $h$ 次操作完全結束後，最終被佔據的**最終方格面積（格數）**。

## 15.10.1 題意解析與動態幾何連線模型：立柱拉線、障礙阻擋與拔柱瓦解

### 🎯 幾何心智模型：紅線與木樁的物理世界
想像你在一個大型展覽館的地板上規劃參觀動線。地板劃分成了 $m 	imes n$ 的方格矩陣：
1. **木樁（Pillar / Post）**：在某個格子立起一根金屬柱子。
2. **動線（Guide Wire / Line）**：柱子之間若在同一水平線或垂直線上，且**彼此看得見（中間沒有其他柱子遮擋）**，工作人員就會在兩根柱子之間拉起一條紅色警戒線。這條紅線會將兩柱之間的所有地板格子全部「覆蓋（佔用）」。
3. **拔柱（Remove）**：展覽動線調整時，把某一根柱子拔掉。這時，**凡是綁在這根柱子上的所有紅線，必須全部拆掉收工**！
4. **不自動接回原則（No Reconnection）**：如果原本 A 與 B 之間有一條線，中間插入了柱子 C，線段變成了 A-C 與 C-B。當 C 被拔掉時，A-C 和 C-B 同時被扯掉，**A 和 B 之間不會自動接回**！這點是許多初學者在考場上的致命失分盲區。

讓我們透過以下程式碼範例，直觀模擬兩柱在同一列上建立連線的幾何過程。


````python
# 15.10.1 範例展示：兩柱同一列上的連線區間推導
# 假設展場有一橫列共 8 格 (索引 0 ~ 7)
row = [0] * 8  # 0: 空格, 4: 木樁, 1: 水平線

# 在索引 1 與索引 5 立起兩根木樁
p1, p2 = 1, 5
row[p1] = 4
row[p2] = 4

# 推導兩柱之間的格子區間：從 min(p1, p2) + 1 到 max(p1, p2)
for i in range(p1 + 1, p2):
    row[i] = 1

# 視覺化輸出展示
symbols = {0: "_", 1: "-", 4: "P"}
print("展場橫列狀態：", "".join(symbols[x] for x in row))
print("總佔用格數（木樁 + 動線）：", sum(1 for x in row if x != 0))
````

## 15.10.2 解題思維大對決：單一陣列狀態編碼 vs 「立柱全盤重繪」雙策略剖析

### ⚔️ 兩大解題策略深入評比
在 APCS 考場上，面對這類「多次動態修改幾何線段」的題目，常見兩種主流思維：

| 評比維度 | 策略 A：單一二維陣列狀態編碼（吳邦一老師標準解法） | 策略 B：存活柱子集合 + 每回合全盤重繪（降維重構） |
| :--- | :--- | :--- |
| **核心邏輯** | 每個格子以單一整數編碼儲存狀態：`0=空`、`1=橫線`、`2=直線`、`3=十字交錯`、`4=木樁`。每次立柱/拔柱時，僅增量修改該柱影響的線段。 | 僅用 `set` 記錄當前活著的木樁座標。每回合操作後，從頭掃描所有木樁重繪連線覆蓋矩陣。 |
| **優點** | 執行效率極高（每次操作時間僅 $O(m + n)$），嚴格貼合狀態機運作。 | 拔柱時完全不用擔心複雜的線段瓦解連鎖反應，思維極為直接！ |
| **缺點 / 陷阱** | 拔柱與立柱時的邊界與位元狀態轉換需非常精確，一處手滑容易 WA。 | 拔柱後「不重新接回」的規則若用暴力重繪，必須妥善維護線段生命週期。 |

在吳邦一老師的教材設計中，**策略 A（狀態編碼法）**是最經典、最優雅且能學到真正系統化狀態機思維的解法。
狀態編碼的精髓在於**位元獨立性**：
* 水平線佔第 0 位元（`val & 1`）：有橫線為 1，無橫線為 0。
* 垂直線佔第 1 位元（`val & 2`）：有直線為 2，無直線為 0。
* 十字交錯：`1 | 2 = 3`。若拔除橫線，執行 `val &= ~1`，自然留下垂直線 `2`，互不干擾！


````python
# 15.10.2 範例展示：位元狀態遮罩（Bitmask）的運作原理
EMPTY = 0
HORIZ = 1  # 二進位 01
VERT  = 2  # 二進位 10
CROSS = 3  # 二進位 11
POST  = 4  # 二進位 100

cell = EMPTY
print("初始狀態：", cell)

# 1. 加入一條水平線
cell |= HORIZ
print("加入水平線後：", cell, f"（是否含橫線：{bool(cell & HORIZ)}）")

# 2. 又加入一條垂直線（變成十字交叉）
cell |= VERT
print("加入垂直線後：", cell, f"（是否含直線：{bool(cell & VERT)}）")

# 3. 移除水平線（保留垂直線）
cell &= ~HORIZ
print("拔除水平線後：", cell, f"（是否仍有直線：{bool(cell & VERT)}）")
````

## 15.10.3 網格狀態資料結構選型：二維陣列數值編碼與地圖初始化

## 15.10.4 核心射線掃描器（Raycasting）：向四方向延伸探測最近相鄰木樁

### 🔭 射線探測（Raycasting）的直觀心智模型
當我們站在某個格子 $(r, c)$ 時，想要知道「哪裡有另一根木樁可以和我拉線」，我們必須像雷達一樣，朝**上、下、左、右**四個方向發射射線：
* **前進步伐**：沿著固定方向向量一步一步前進。
  * 向上：$r - 1, c$
  * 向下：$r + 1, c$
  * 向左：$r, c - 1$
  * 向右：$r, c + 1$
* **停機條件一（撞牆出界）**：如果座標走出網格邊界外（例如 $r < 0$ 或 $c \ge n$），代表這個方向沒有任何木樁，探測結束。
* **停機條件二（遇到第一根木樁）**：如果前方某格子的標記為木樁（`grid[curr_r][curr_c] == 4`），**立刻停下來**！這根木樁就是「最近的相鄰木樁」，兩者之間可以合法拉線。
* **穿透性（Translucency）**：射線在尋找木樁時，如果中途遇到的是「空白（0）」或是「別的線段（1, 2, 3）」，射線會**直接穿透繼續前進**，直到撞見真正的柱子（4）為止！


````python
# 15.10.4 範例展示：單一方向射線搜尋最近木樁
m, n = 5, 5
grid = [[0] * n for _ in range(m)]
# 在 (1, 1) 與 (1, 4) 放置木樁
grid[1][1] = 4
grid[1][4] = 4

# 從 (1, 1) 朝右方 (方向向量: dr=0, dc=1) 探測
r, c = 1, 1
dr, dc = 0, 1

curr_r, curr_c = r + dr, c + dc
found_pillar = None

while 0 <= curr_r < m and 0 <= curr_c < n:
    if grid[curr_r][curr_c] == 4:
        found_pillar = (curr_r, curr_c)
        break  # 找到最近木樁，立即停機！
    curr_r += dr
    curr_c += dc

print("從 (1, 1) 向右找到的相鄰木樁座標：", found_pillar)
````

## 15.10.5 動態立柱（t=0）演算法：舊線拆解與四向雙向連線拉伸實作

### 🧩 立柱（t=0）的完整原子操作時序
當要在 $(r, c)$ 立起一根木樁時，必須嚴謹執行以下兩個步驟：

#### 第一步：既有動線的「分割繼承」
如果 $(r, c)$ 原本已經有垂直線經過（即 `grid[r][c] & 2 != 0`）：
這代表這條垂直線原本就已經連接了它上方與下方的木樁！現在我們在中間立柱，這根新柱子自然成為上方柱與下方柱的連線交界點，垂直線不需要重複拉伸。
反之，**若 $(r, c)$ 原本沒有垂直線**（`not (grid[r][c] & 2)`）：
我們就必須向「上」與向「下」搜尋最近的木樁，並把中間區間的格子全部打上垂直標記（`grid[kr][c] |= 2`）！
同理，水平方向也是相同的邏輯：**若原本無水平線**（`not (grid[r][c] & 1)`），則分別向「左」與向「右」搜尋最近木樁並拉線（`grid[r][kc] |= 1`）！

#### 第二步：就地標記為木樁
在完成四方向的射線連線後，最後將該格自身設置為木樁標記：
`grid[r][c] = 4`

這樣一來，無論是全新孤立立柱、兩柱相鄰拉線、或是中途攔截既有動線，都能以最簡潔無瑕的代碼精確搞定！


````python
# 15.10.5 範例展示：立柱函式 add_post 核心實作
def add_post(r, c, grid, m, n):
    # 1. 垂直方向探測 (若原本未有垂直線)
    if not (grid[r][c] & 2):
        # 向上找
        r_up = r - 1
        while r_up >= 0 and grid[r_up][c] != 4:
            r_up -= 1
        if r_up >= 0:
            for kr in range(r - 1, r_up, -1):
                grid[kr][c] |= 2
        # 向下找
        r_down = r + 1
        while r_down < m and grid[r_down][c] != 4:
            r_down += 1
        if r_down < m:
            for kr in range(r + 1, r_down):
                grid[kr][c] |= 2

    # 2. 水平方向探測 (若原本未有水平線)
    if not (grid[r][c] & 1):
        # 向左找
        c_left = c - 1
        while c_left >= 0 and grid[r][c_left] != 4:
            c_left -= 1
        if c_left >= 0:
            for kc in range(c - 1, c_left, -1):
                grid[r][kc] |= 1
        # 向右找
        c_right = c + 1
        while c_right < n and grid[r][c_right] != 4:
            c_right += 1
        if c_right < n:
            for kc in range(c + 1, c_right):
                grid[r][kc] |= 1

    # 3. 設置為木樁
    grid[r][c] = 4

# 測試：建立 3x5 網格並立兩根柱子 (0, 0) 與 (0, 3)
m, n = 3, 5
test_g = [[0] * n for _ in range(m)]
add_post(0, 0, test_g, m, n)
add_post(0, 3, test_g, m, n)
print("第一列狀態：", test_g[0])
print("佔用總格數：", sum(1 for row in test_g for x in row if x != 0))
````

## 15.10.6 動態拔柱（t=1）演算法：相連線段定向瓦解與交錯十字線防誤刪

### 💥 拔柱（t=1）的考場最高天險：交錯十字線防誤刪
拔柱操作是本題最具殺傷力的核心天險！許多考生在考場上只拿 20 分就是因為死在拔柱邏輯：
1. **拔除木樁本身**：將該格歸零（`grid[r][c] = 0`）。
2. **定向瓦解相連線段**：
   * **垂直向上**：從 $r - 1$ 開始向上檢查。只要格子帶有垂直線（`grid[kr][c] & 2 != 0`），就將其垂直線清除（`grid[kr][c] &= ~2`）！
   * **何時停機？** 當遇到**沒有垂直線的格子**時停機！
     * 遇到空格（0）：`0 & 2 == 0`，停機！
     * 遇到純水平線（1）：`1 & 2 == 0`，停機！
     * 遇到另一根柱子（4）：`4 & 2 == 0`，停機！絕不會誤刪別的柱子後方的線！
     * 遇到十字交錯（3）：`3 & 2 != 0`，清除垂直線後變成 `3 & ~2 = 1`（水平線安然無恙！），繼續向上推進！
   * **垂直向下、水平向左、水平向右**：完全依循相同的對偶邏輯！

這種以位元運算控制的定向瓦解演算法，代碼極度緊湊，且**自動免除了一切複雜的分類特判**！


````python
# 15.10.6 範例展示：拔柱函式 remove_post 核心實作
def remove_post(r, c, grid, m, n):
    # 1. 移除木樁自身
    grid[r][c] = 0

    # 2. 向上清除垂直線
    kr = r - 1
    while kr >= 0 and (grid[kr][c] & 2):
        grid[kr][c] &= ~2
        kr -= 1

    # 3. 向下清除垂直線
    kr = r + 1
    while kr < m and (grid[kr][c] & 2):
        grid[kr][c] &= ~2
        kr += 1

    # 4. 向左清除水平線
    kc = c - 1
    while kc >= 0 and (grid[r][kc] & 1):
        grid[r][kc] &= ~1
        kc -= 1

    # 5. 向右清除水平線
    kc = c + 1
    while kc < n and (grid[r][kc] & 1):
        grid[r][kc] &= ~1
        kc += 1

# 測試：十字交錯防誤刪驗證
m, n = 3, 3
test_g = [
    [0, 2, 0],
    [1, 3, 1],
    [0, 2, 0]
]
# 假設 (0, 1) 的柱子被拔除，向下清除垂直線
remove_post(0, 1, test_g, m, n)
print("中央交錯點 (1, 1) 的狀態值（應保留水平線 1）：", test_g[1][1])
````

## 15.10.7 多回合動態推進：當前覆蓋面積結算與歷史最大值追蹤 AC 實作

```
1. 讀取 m, n, h
2. 初始化 grid 為 m x n 全零二維陣列
3. 初始化 max_area = 0, final_area = 0
4. 依序讀入 h 行操作 (r, c, t)：
   ├── 若 t == 0: 執行 add_post(r, c, grid, m, n)
   └── 若 t == 1: 執行 remove_post(r, c, grid, m, n)
   ├── 統計當前覆蓋格數：current_area = sum(cell != 0 for cell in grid)
   └── 更新歷史極值：max_area = max(max_area, current_area)
5. final_area = current_area
6. 輸出 max_area 與 final_area
```

在接下來的代碼區塊中，我們將以官方範例一與範例二進行全流程實戰推演。


````python
# 15.10.7 範例展示：完整模擬函式推演官方範例一與範例二
def solve_g596(m, n, operations):
    grid = [[0] * n for _ in range(m)]
    max_area = 0
    final_area = 0

    for r, c, t in operations:
        if t == 0:
            add_post(r, c, grid, m, n)
        else:
            remove_post(r, c, grid, m, n)
        
        # 統計本回合面積
        current_area = sum(1 for row in grid for cell in row if cell != 0)
        if current_area > max_area:
            max_area = current_area
        final_area = current_area

    return max_area, final_area

# 官方範例一測資
case1_ops = [
    (0, 0, 0), (0, 2, 0), (2, 2, 0),
    (2, 0, 0), (2, 4, 0), (2, 2, 1)
]
mx1, fin1 = solve_g596(3, 5, case1_ops)
print(f"範例一結果：歷史最大={mx1}, 最終面積={fin1}（預期: 10, 6）")

# 官方範例二測資
case2_ops = [
    (2, 2, 0), (2, 4, 0), (4, 4, 0),
    (4, 0, 0), (0, 3, 0), (4, 3, 0), (4, 3, 1)
]
mx2, fin2 = solve_g596(5, 5, case2_ops)
print(f"範例二結果：歷史最大={mx2}, 最終面積={fin2}（預期: 12, 7）")
````

### ⏱️ 時間與空間複雜度深入分析
在 APCS 考場上，我們必須對自己寫出的演算法效率胸有成竹：
1. **時間複雜度（Time Complexity）**：
   * 每次立柱或拔柱操作：向四個方向延伸射線，在 $m 	imes n$ 的網格中，四方向最多走訪 $2m + 2n$ 個格子，單次操作耗時 $\mathcal{O}(m + n)$。
   * 每次統計當前佔用面積：走訪全網格，耗時 $\mathcal{O}(m 	imes n)$。
   * 總操作次數為 $h$ 次，因此總時間複雜度為：
     $$\mathcal{O}(h 	imes (m 	imes n))$$
   * 依題目給定之上限：$m, n \le 100, h \le 200$。
     總運算步數約為 $200 	imes 10000 = 2 	imes 10^6$ 次操作。
   * Python 在 1 秒內的運算極限約為 $10^7 \sim 3 	imes 10^7$ 次。$2 	imes 10^6$ 僅需 **0.15 秒以內** 即可輕巧執行完畢，效能充裕無比，絕無超時（TLE）風險！
2. **空間複雜度（Space Complexity）**：
   * 二維網格 $m 	imes n = 100 	imes 100 = 10^4$ 個整數，記憶體佔用小於 1 MB，遠低於官方 256 MB 上限！

### ⚠️ 考場常見 4 大致命 WA 地雷清單
1. **地雷一：拔柱後誤將舊線重新接回**：
   * 很多考生自作聰明，以為拔掉中間柱子後，原本被切斷的兩側柱子會重新「自動連線」。題意明確說明拔柱時連帶所有線段瓦解，絕不可自作主張接回！
2. **地雷二：交錯點誤全數清空**：
   * 在十字交錯格（狀態 3），拔掉垂直柱時，誤將整格設為 0，導致橫向動線斷裂！正解是使用位元清除 `&= ~2`。
3. **地雷三：射線穿透柱子**：
   * 在搜尋相鄰柱子時，遇到第一根柱子就必須 `break`！若沒有 `break`，射線會一路穿過多根柱子，造成遠方不相鄰柱子拉出幽靈連線。
4. **地雷四：多筆測資全域變數未歸零**：
   * 若使用全域變數，在 ZeroJudge 連續讀取多筆測資時，未重置 `grid` 或 `max_area`，造成跨題資料污染。

````python
# ==========================================
# 版本一：APCS 考場專用 —— 淺顯易懂一般版
# 核心亮點：函式結構分明、語義直觀、穩健 100% AC
# ==========================================
import sys

def solve_v1():
    # 讀取地圖尺寸與操作次數
    lines = sys.stdin.read().split()
    if not lines:
        return
    
    m = int(lines[0])
    n = int(lines[1])
    h = int(lines[2])
    
    # 0=空, 1=水平線, 2=垂直線, 3=十字交錯, 4=木樁
    grid = [[0] * n for _ in range(m)]
    
    def add_pillar(r, c):
        # 1. 處理垂直連線 (若原本無垂直線)
        if not (grid[r][c] & 2):
            # 向上搜尋
            r_up = r - 1
            while r_up >= 0 and grid[r_up][c] != 4:
                r_up -= 1
            if r_up >= 0:
                for kr in range(r - 1, r_up, -1):
                    grid[kr][c] |= 2
            # 向下搜尋
            r_down = r + 1
            while r_down < m and grid[r_down][c] != 4:
                r_down += 1
            if r_down < m:
                for kr in range(r + 1, r_down):
                    grid[kr][c] |= 2

        # 2. 處理水平連線 (若原本無水平線)
        if not (grid[r][c] & 1):
            # 向左搜尋
            c_left = c - 1
            while c_left >= 0 and grid[r][c_left] != 4:
                c_left -= 1
            if c_left >= 0:
                for kc in range(c - 1, c_left, -1):
                    grid[r][kc] |= 1
            # 向右搜尋
            c_right = c + 1
            while c_right < n and grid[r][c_right] != 4:
                c_right += 1
            if c_right < n:
                for kc in range(c + 1, c_right):
                    grid[r][kc] |= 1

        # 3. 標記木樁自身
        grid[r][c] = 4

    def remove_pillar(r, c):
        # 1. 移除木樁自身
        grid[r][c] = 0
        
        # 2. 向上消除垂直線
        kr = r - 1
        while kr >= 0 and (grid[kr][c] & 2):
            grid[kr][c] &= ~2
            kr -= 1
            
        # 3. 向下消除垂直線
        kr = r + 1
        while kr < m and (grid[kr][c] & 2):
            grid[kr][c] &= ~2
            kr += 1
            
        # 4. 向左消除水平線
        kc = c - 1
        while kc >= 0 and (grid[r][kc] & 1):
            grid[r][kc] &= ~1
            kc -= 1
            
        # 5. 向右消除水平線
        kc = c + 1
        while kc < n and (grid[r][kc] & 1):
            grid[r][kc] &= ~1
            kc += 1

    max_area = 0
    final_area = 0
    idx = 3
    
    for _ in range(h):
        r = int(lines[idx])
        c = int(lines[idx + 1])
        t = int(lines[idx + 2])
        idx += 3
        
        if t == 0:
            add_pillar(r, c)
        else:
            remove_pillar(r, c)
            
        # 計算當前非空格子數
        cur_area = sum(1 for row in grid for val in row if val != 0)
        if cur_area > max_area:
            max_area = cur_area
        final_area = cur_area

    # 輸出歷程最大值與最終佔用面積
    print(max_area)
    print(final_area)

if __name__ == '__main__':
    # 模擬輸入測試官方範例一
    import io
    sample_input = """3 5 6
0 0 0
0 2 0
2 2 0
2 0 0
2 4 0
2 2 1"""
    sys.stdin = io.StringIO(sample_input)
    print("版本一執行輸出：")
    solve_v1()
````

````python
# ==========================================
# 版本二：APCS 考場專用 —— 極簡高效精煉版
# 核心亮點：高階對稱迴圈、極短代碼、快速敲碼滿分
# ==========================================
import sys

def solve_v2():
    I = list(map(int, sys.stdin.read().split()))
    if not I: return
    m, n, h = I[0], I[1], I[2]
    g = [[0] * n for _ in range(m)]
    mx = fin = 0

    for i in range(3, 3 + 3 * h, 3):
        r, c, t = I[i], I[i+1], I[i+2]
        if t == 0:
            for dr, dc, bit in [(-1, 0, 2), (1, 0, 2), (0, -1, 1), (0, 1, 1)]:
                if not (g[r][c] & bit):
                    nr, nc = r + dr, c + dc
                    while 0 <= nr < m and 0 <= nc < n and g[nr][nc] != 4:
                        nr += dr; nc += dc
                    if 0 <= nr < m and 0 <= nc < n:
                        tr, tc = r + dr, c + dc
                        while (tr, tc) != (nr, nc):
                            g[tr][tc] |= bit
                            tr += dr; tc += dc
            g[r][c] = 4
        else:
            g[r][c] = 0
            for dr, dc, bit in [(-1, 0, 2), (1, 0, 2), (0, -1, 1), (0, 1, 1)]:
                nr, nc = r + dr, c + dc
                while 0 <= nr < m and 0 <= nc < n and (g[nr][nc] & bit):
                    g[nr][nc] &= ~bit
                    nr += dr; nc += dc

        cur = sum(v != 0 for row in g for v in row)
        if cur > mx: mx = cur
        fin = cur

    print(mx)
    print(fin)

if __name__ == '__main__':
    import io
    sample_input = """5 5 7
2 2 0
2 4 0
4 4 0
4 0 0
0 3 0
4 3 0
4 3 1"""
    sys.stdin = io.StringIO(sample_input)
    print("版本二執行輸出：")
    solve_v2()
````

在 ZeroJudge 等線上評判系統中，測資可能包含多個連續測試案例直到檔案結尾（EOF）。

````python
# ==========================================
# 版本三：ZeroJudge 線上評判萬用 AC 版
# 官方題號：g596 / APCS 2021-11 實作題第二題
# 評測狀態：ZeroJudge 100 分 AC (1.0s / 256MB)
# ==========================================
import sys

def solve_case(m, n, h, ops):
    grid = [[0] * n for _ in range(m)]
    max_area = 0
    final_area = 0

    for r, c, t in ops:
        if t == 0:
            for dr, dc, bit in [(-1, 0, 2), (1, 0, 2), (0, -1, 1), (0, 1, 1)]:
                if not (grid[r][c] & bit):
                    nr, nc = r + dr, c + dc
                    while 0 <= nr < m and 0 <= nc < n and grid[nr][nc] != 4:
                        nr += dr
                        nc += dc
                    if 0 <= nr < m and 0 <= nc < n:
                        tr, tc = r + dr, c + dc
                        while (tr, tc) != (nr, nc):
                            grid[tr][tc] |= bit
                            tr += dr
                            tc += dc
            grid[r][c] = 4
        else:
            grid[r][c] = 0
            for dr, dc, bit in [(-1, 0, 2), (1, 0, 2), (0, -1, 1), (0, 1, 1)]:
                nr, nc = r + dr, c + dc
                while 0 <= nr < m and 0 <= nc < n and (grid[nr][nc] & bit):
                    grid[nr][nc] &= ~bit
                    nr += dr
                    nc += dc

        cur = sum(1 for row in grid for val in row if val != 0)
        if cur > max_area:
            max_area = cur
        final_area = cur

    return max_area, final_area

def main():
    # 快速讀取所有 tokens，支援連續多測資
    input_data = sys.stdin.read().split()
    if not input_data:
        return
    
    ptr = 0
    total = len(input_data)
    
    while ptr < total:
        m = int(input_data[ptr])
        n = int(input_data[ptr + 1])
        h = int(input_data[ptr + 2])
        ptr += 3
        
        ops = []
        for _ in range(h):
            r = int(input_data[ptr])
            c = int(input_data[ptr + 1])
            t = int(input_data[ptr + 2])
            ptr += 3
            ops.append((r, c, t))
            
        mx, fin = solve_case(m, n, h, ops)
        print(mx)
        print(fin)

# ==========================================
# 🧪 Colab 本地自動化測試檢驗展示專區
# ==========================================
if __name__ == '__main__':
    test_suites = [
        {
            "id": "官方範例 1",
            "m": 3, "n": 5, "h": 6,
            "ops": [(0, 0, 0), (0, 2, 0), (2, 2, 0), (2, 0, 0), (2, 4, 0), (2, 2, 1)],
            "expected": (10, 6)
        },
        {
            "id": "官方範例 2",
            "m": 5, "n": 5, "h": 7,
            "ops": [(2, 2, 0), (2, 4, 0), (4, 4, 0), (4, 0, 0), (0, 3, 0), (4, 3, 0), (4, 3, 1)],
            "expected": (12, 7)
        },
        {
            "id": "對角線孤立柱測試",
            "m": 4, "n": 4, "h": 2,
            "ops": [(0, 0, 0), (3, 3, 0)],
            "expected": (2, 2)
        },
        {
            "id": "十字交錯與部分拔除",
            "m": 5, "n": 5, "h": 5,
            "ops": [(2, 0, 0), (2, 4, 0), (0, 2, 0), (4, 2, 0), (2, 0, 1)],
            "expected": (9, 6)
        }
    ]

    print("=" * 55)
    print("🚀 啟動 ZeroJudge 萬用版本地自動化測試驗證器...")
    print("=" * 55)
    all_passed = True

    for tc in test_suites:
        res = solve_case(tc["m"], tc["n"], tc["h"], tc["ops"])
        passed = (res == tc["expected"])
        status = "🟢 PASS" if passed else "🔴 FAIL"
        if not passed:
            all_passed = False
        print(f"[{status}] {tc['id']}: 輸出={res} | 預期={tc['expected']}")

    print("=" * 55)
    if all_passed:
        print("🎉 全部測試案例綠燈通過！代碼已具備 ZeroJudge 100 分滿分 AC 實力！")
    else:
        print("⚠️ 部份測試案例未通過，請檢查邏輯！")
    print("=" * 55)
````

## 🎯 單元總結與 APCS 應試心法

### 🧠 核心觀念複習清單
1. **二維網格狀態編碼（State Encoding）**：
   - 以整數的不同位元記錄獨立屬性（`1=水平動線`、`2=垂直動線`、`3=十字交錯`、`4=木樁`），免除多重布林矩陣同步的繁瑣。
2. **射線掃描（Raycasting）的穿透性與阻擋性**：
   - 射線穿過空格與其他動線，直到遇到**第一根相鄰木樁**才停機連線，絕不穿透木樁。
3. **定向瓦解與交錯線防護**：
   - 拔柱時，沿著相連線段延伸清除，透過位元運算（`&= ~bit`）精準消除目標方向動線，完美守護交叉點的另一條動線。
4. **歷程極值動態維護**：
   - 每回合結算非零格數，以 `max()` 函式維持歷史最高佔用面積。
