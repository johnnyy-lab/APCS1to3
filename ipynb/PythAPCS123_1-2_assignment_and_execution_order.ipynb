但是，當你踏入 Python 的程式魔法世界時，請立刻把腦袋裡的「數學天平」切換掉！在 Python 裡，單一個等號 `=` 根本不是「等於」的意思，它的真正身分叫做「**賦值運算子（Assignment Operator）**」！你可以把它想像成一條「**單向強力輸送帶**」。

這條輸送帶的箭頭永遠朝著左邊指：`左邊變數箱子 ⬅️ 右邊數值`。它的唯一使命，就是把等號右邊的數值打包裝箱，單向傳送到左邊貼著變數名牌的置物櫃裡保存。

舉例來說，當我們寫下：

### ⚙️ 1-2-2 賦值的黃金順序——「先算右邊，再存左邊」

既然等號是一條單向輸送帶，那麼當等號右邊出現長長的計算式子時，電腦是怎麼處理的呢？

請大家在心裡默念這句八字真言：「**先算右邊，再存左邊！**」這是全世界所有程式語言共通的黃金鐵律。

想像一下披薩店的大廚備料：當客人點了一份綜合披薩，大廚絕對不會把生麵粉、起司條和生番茄一顆顆直接塞進外送紙盒裡；大廚一定會先在右邊的料理檯上，把所有食材切好調配、放進烤箱烤熟，算出一塊熱騰騰的完整披薩之後，才會在最後一步把它裝進左邊的外送盒子裡！

在 Python 裡也是一模一樣的運作流程。當我們寫出：
```python
x = 5 + 3 * 6
```
直譯器小助手看到這行指令時，它的眼睛會立刻專注在等號的右半邊：
1. **第一步（右側計算）**：遵守基礎數學「先乘除後加減」的優先規則，先計算 $3 \times 6 = 18$。
2. **第二步（右側加總）**：再計算 $5 + 18 = 23$。
3. **第三步（賦值裝箱）**：當右邊全部算乾淨、得到單一數值 23 後，輸送帶才正式啟動，把 23 一口氣送進左邊貼著名牌 `x` 的箱子中保存！

因此，左邊的變數只會收到最後算出來的最終成果（23），完全不會看到中間算式的過程喔！

````python
# [2] Code 範例區：右側算式先完整計算（5 + 3*6 = 23），最後才將結果存入 demo_x
demo_x = 5 + 3 * 6
print(demo_x)
````

### 🚫 1-2-3 等號左側的嚴格鐵律——左邊只能是「單一變數箱子」！

在吳邦一老師的講義第 5 頁中，特別標註了一行非常醒目的經典警示：
```python
y + 3 = 6  # (這是錯的！)
```

很多剛接觸程式的初學同學看到這行程式碼都會納悶：「老師，在基礎數學的一元一次方程式中，$y + 3 = 6$ 不是很正常、很標準的等式嗎？只要心算移項一下，就知道 $y = 3$ 啦！」

但是在程式語言的世界裡，這樣寫會立刻引發大爆炸！Python 直譯器會毫不留情地亮出紅牌警告：
`SyntaxError: cannot assign to expression`（語法錯誤：無法賦值給運算式！）

為什麼電腦會這麼生氣呢？讓我們再次回到「置物箱」與「單向輸送帶」的比喻：
等號輸送帶的功能，是把右邊的實體寶物倒進左邊的容器裡。因此，**等號的左邊，必須是一個真真切切、能夠打開蓋子裝東西的「單一變數箱子」！**

你想想看：`6` 是一個寶物，但是 `y + 3` 是一道數學算式，它根本不是一個實體置物箱！你怎麼可能把一個實體的寶物硬塞進一道數學題目裡面呢？
同樣地，如果你寫成 `10 = x` 也是大錯特錯，因為 `10` 是一塊固定死板的石頭（常數值），它也不是能裝東西的箱子！

所以請牢牢記住：電腦不像人類大腦懂得自動移項。如果你想求出 $y$，身為魔法師的你必須自己先在腦袋中移項完成，寫成：
```python
y = 6 - 3
```
讓等號左邊永遠保持是乾乾淨淨、孤身一人的「單一變數名牌」，右邊才是要計算的數值或式子喔！

````python
# [2] Code 範例區：正確寫法！左邊必須是單一變數名牌，右邊才是計算式
# 錯誤示範（切勿執行）：demo_y + 3 = 6 會造成 SyntaxError
# 正確寫法：
demo_y = 6 - 3
print(demo_y)
````

### 🔄 1-2-4 舊的不去新的不來——變數賦值的覆蓋（Overwrite）特性

想像一下，你桌上有一個貼著「今日下午茶」名牌的神奇置物箱。

中午的時候，你在箱子裡放進了一塊草莓蛋糕；但是到了下午三點，你改變主意了，又往同一個箱子裡塞進了一顆大紅蘋果。這時候，如果打開箱子看，裡面會同時有蛋糕和蘋果嗎？

答案是「絕對不會」！Python 的變數置物箱遵循著一條非常乾脆俐落的物理法則：**一個變數箱子在同一個瞬間，永遠只能存放單一筆最新資料！**

在電腦科學中，這個現象稱為「**覆蓋（Overwrite）**」。每當等號輸送帶把新的數值送進已經有名牌的箱子時，箱子底部的神秘暗門就會瞬間打開，把原本住在裡面的舊寶物一口氣丟進垃圾桶徹底銷毀，只保留最新送進來的那樣東西！

舉例來說：
```python
score = 60
score = 95
print(score)
```
當第一行執行完畢時，`score` 箱子裡裝的是 60；但是當第二行執行時，95 被輸送帶送了進來，原本的 60 就永遠蒸發消失了！所以最後一行印出的結果，百分之百是最新放進去的 95。

這提醒我們：在寫程式時如果後面還要用到舊資料，一定要記得多開一個新的變數箱子先將它備份保存起來喔！

````python
# [2] Code 範例區：展示變數數值被新賦值覆蓋的過程
demo_item = 10
print(demo_item)  # 此時印出舊的數值：10

demo_item = 99    # 新數值 99 進入，舊數值 10 被覆蓋銷毀
print(demo_item)  # 此時印出最新數值：99
````

### 🪄 1-2-5 自己算自己——右側引用舊值自我更新（`x = x + 1` 的魔法）

現在我們要結合前幾小節的所有超能力，來破解程式設計史上最著名、也最容易嚇到初學者的傳奇指令：
```python
x = x + 1
```

如果是一位剛接觸程式設計的初學者第一次看到這行程式碼，肯定會拍桌子大喊：「老師！這在數學上根本是荒謬的大錯誤嘛！一個數字怎麼可能加上 1 之後還等於它自己？這在數學上根本無解！」

同學請息怒！請再次回想起我們在 **1-2-1** 與 **1-2-2** 學到的兩大鐵則：
1. **等號不是等於**，而是把右邊算好的東西搬到左邊的「單向輸送帶」！
2. 永遠「**先算右邊，再存左邊**」！

現在，讓我們用慢動作分解鏡頭，看看電腦究竟是怎麼執行 `x = x + 1` 的：
* **起始狀態**：假設目前 `x` 箱子裡裝著原本的舊數值 `10`。
* **步驟一（先看右邊）**：電腦小助手的眼睛先看等號右邊的算式 `x + 1`。它跑去打開目前的 `x` 箱子，拿出裡面的舊數字 `10`，並在料理檯上計算 $10 + 1 = 11$。
* **步驟二（再存左邊）**：小助手捧著剛剛熱騰騰算好的新成果 `11`，啟動等號輸送帶，送往左邊貼著 `x` 名牌的箱子！
* **步驟三（舊值覆蓋）**：根據覆蓋原則，舊的 `10` 被踢出箱子，新的 `11` 正式住進了 `x` 箱子！

看清楚了嗎？**右邊的 `x` 代表的是「過去的舊數值」，而左邊的 `x` 代表的是「準備迎接新數值的家」！**

這種「自己引用自己舊值來累加更新」的寫法，在遊戲計分、跑跑步圈數累加、或是 APCS 各種演算法迴圈中無所不在，是程式世界裡最強大的基本招式喔！

````python
# [2] Code 範例區：利用右側先計算舊值（demo_sc012 + 20），再存回左側自己更新
demo_sc012 = 50
print(demo_sc012)  # 原本的分數：50

demo_sc012 = demo_sc012 + 20  # 先算 50 + 20 = 70，再存回 demo_sc012
print(demo_sc012)  # 更新後的最新分數：70
````

剛才在各題中建立的數值型態變數、運算式與語法結構，現在都乖乖躺在記憶體置物櫃裡囉。  
點擊下方的 **▶ 播放鍵**，系統將跳出輸入框讓您登記**姓名**與**學號座號**（分開獨立欄位，亦可直接按 Enter 採用自學預設值），並自動啟動 Google 官方帳號身分認證（記錄真實 Email 與 Google 顯示姓名），產出**學習成果成績單**並同步登錄後台試算表！

> 🆘 **終極救援方案**：如果重新執行個別題目後，成績還是怪怪的（例如明明寫對了卻還是顯示 ❌），最保險的做法是**整本全部重跑**，一鍵重置記憶體、告別幽靈變數！**Google Colab** 請點上方選單「**執行階段 ➔ 重新啟動工作階段並全部執行**」（快捷鍵 `Ctrl + F9`）；**VS Code** 請於上方工具列依序點選「**Clear All Outputs ➔ Restart ➔ Run All**」。

````python
#@title 🧪 點擊此處取得單元 1-2 學習成效自動評分報告 { display-mode: "form" }
# ------------------------------------------------------------------------------
# 官方智慧全環境評分載入器（支援：本地開發庫優先 ➔ 公開正式庫雙重備援）
# ------------------------------------------------------------------------------
import os
import sys
import json
import logging
import urllib.request

# 靜音內部 httplib2 警告
logging.getLogger("google_auth_httplib2").setLevel(logging.ERROR)

# 1. 互動式輸入學員姓名與學號（分開獨立欄位，按 Enter 採用預設值）
print("👤【學員身分登記】（自學測試可直接按 Enter 鍵使用預設值）：")
student_name = input("請輸入您的姓名 [預設: 自學冒險者]：").strip() or "自學冒險者"
student_id = input("請輸入學號或座號 [預設: 無座號]：").strip() or "無座號"

# 2. 自動嘗試獲取 Google 帳號官方身分認證（真實 Email 與 Google 暱稱）
google_email = None
google_name = None
try:
    from google.colab import auth
    print("🔐 正在連結 Google 帳號進行身分認證（若跳出授權視窗，請點選允許）...")
    auth.authenticate_user()

    # 方法 1: 使用 Google Drive API v3 (about:get) 抓取真實 displayName 與 emailAddress
    # Colab ADC 預設包含 drive 權限，能 100% 精準獲取 Google 帳號中文顯示名稱（如：葉明昌）
    try:
        import google.auth
        from googleapiclient.discovery import build
        creds, _ = google.auth.default()
        drive_service = build('drive', 'v3', credentials=creds, cache_discovery=False)
        about = drive_service.about().get(fields="user(displayName,emailAddress)").execute()
        u_obj = about.get("user", {})
        google_name = u_obj.get("displayName")
        google_email = u_obj.get("emailAddress")
    except Exception:
        pass

    # 方法 2: 若 Drive API 失敗，嘗試透過 OAuth2 userinfo / tokeninfo 備援
    if not google_email or not google_name:
        try:
            import google.auth
            from google.auth.transport.requests import Request
            credentials, _ = google.auth.default()
            credentials.refresh(Request())
            token = credentials.token

            for ep in [
                f"https://www.googleapis.com/oauth2/v3/tokeninfo?access_token={token}",
                "https://www.googleapis.com/oauth2/v3/userinfo",
                "https://www.googleapis.com/oauth2/v2/userinfo"
            ]:
                try:
                    req = urllib.request.Request(
                        ep,
                        headers={"Authorization": f"Bearer {token}"}
                    )
                    with urllib.request.urlopen(req, timeout=5) as resp:
                        u_data = json.loads(resp.read().decode('utf-8'))
                        if not google_email:
                            google_email = u_data.get("email") or u_data.get("emailAddress")
                        if not google_name:
                            google_name = (
                                u_data.get("name")
                                or u_data.get("given_name")
                                or u_data.get("displayName")
                            )
                        if google_email and google_name:
                            break
                except Exception:
                    continue
        except Exception:
            pass

    # 方法 3: 若只有 email 但 Google API 沒給 name，以 email 前綴作為暱稱回退
    if google_email and not google_name:
        google_name = google_email.split("@")[0]

    if google_email:
        print(f"✅ 身分認證成功！已綁定 Google 帳號：{google_name} ({google_email})")
except Exception:
    # 非 Colab 環境、本機離線測試或略過授權
    pass

# 3. 身分資料獨立注入環境變數（學號與姓名完全分開）
os.environ["STUDENT_NAME"] = student_name
os.environ["STUDENT_ID"] = student_id
if google_email:
    os.environ["GOOGLE_EMAIL"] = google_email
if google_name:
    os.environ["GOOGLE_NAME"] = google_name

# 4. 評分核心腳本載入（本機優先 ➔ 遠端記憶體載入，絕不產生實體暫存檔）
# 找不到來源就換下一條。內容已拿到但執行失敗時，停止改試其他來源。
# 評分腳本放在獨立命名空間執行，成功後也不把名稱寫回筆記本。答案讀執行前的學生存放格快照。
_loaded = False
_load_error = None

# 4.1 優先嘗試讀取開發庫/本機檔案（本機離線開發測試零延遲）
_local_paths = [
    "web/graders/grade_1_2.py",
    "../graders/grade_1_2.py",
    "../../graders/grade_1_2.py",
    "../web/graders/grade_1_2.py",
    "../../web/graders/grade_1_2.py"
]
for _lp in _local_paths:
    if not os.path.exists(_lp):
        continue
    try:
        with open(_lp, "r", encoding="utf-8") as _f:
            _code = _f.read()
        _grader_ns = {"__builtins__": __builtins__, "__name__": "apcs_grader", "__STUDENT_ENV__": dict(globals())}
        try:
            exec(_code, _grader_ns)
        finally:
            del _grader_ns
        os.environ["GRADER_SOURCE"] = _lp
        _loaded = True
        break
    except Exception as _exc:
        _load_error = _exc
        break

# 4.2 若在雲端環境（Colab），自動透過公開正式庫雙重備援以純記憶體載入
if not _loaded and _load_error is None:
    _grader_urls = [
        "https://raw.githubusercontent.com/johnnyy-lab/APCS1to3/main/graders/grade_1_2.py",
        "https://johnnyy-lab.github.io/APCS1to3/graders/grade_1_2.py",
        "https://raw.githubusercontent.com/johnnyy-lab/APCS1to3-dev/main/web/graders/grade_1_2.py"
    ]
    for _url in _grader_urls:
        try:
            _req = urllib.request.Request(_url, headers={"User-Agent": "Mozilla/5.0"})
            with urllib.request.urlopen(_req, timeout=10) as _resp:
                _code = _resp.read().decode("utf-8")
        except Exception:
            continue
        try:
            _grader_ns = {"__builtins__": __builtins__, "__name__": "apcs_grader", "__STUDENT_ENV__": dict(globals())}
            try:
                exec(_code, _grader_ns)
            finally:
                del _grader_ns
            os.environ["GRADER_SOURCE"] = _url
            _loaded = True
            break
        except Exception as _exc:
            _load_error = _exc
            break

if not _loaded:
    if _load_error is not None:
        print("❌ 評分程式載入失敗，請稍後重新執行。若仍失敗，請把這個畫面交給老師。")
        if os.environ.get("GRADER_DEBUG", "").strip().lower() in ("1", "true", "yes", "on"):
            _reason = str(_load_error).strip().splitlines()
            _line = _reason[0][:180] if _reason else ""
            if _line:
                print("（開發診斷：" + type(_load_error).__name__ + "：" + _line + "）")
            else:
                print("（開發診斷：" + type(_load_error).__name__ + "）")
    else:
        print("❌ 評分系統連線異常，請確認網路連線狀態（或稍後重新執行）")
````
