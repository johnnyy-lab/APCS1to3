### ⏳ 6-4-1 `while` 迴圈基礎：條件哨兵與計數迭代——只要條件為真就絕不停步

**生活比喻：吃拉麵直到碗底朝天與手機電量**  
想像你正在拉麵店享受一碗熱氣騰騰的豚骨拉麵。你拿著湯匙喝湯的邏輯不是「我精確地要喝 37 湯匙（那是 `for` 迴圈）」，而是：  
「**只要（`while`）碗裡還有湯，我就再喝一湯匙！**」  
每喝一口，碗裡的湯就少一點（這叫做「狀態的更新」）；直到碗底朝天的那一瞬間，條件「碗裡還有湯」變成假（`False`），你才會心滿意足地放下湯匙！  
再比如手機播音樂，系統的判斷是：「**只要（`while`）電量大於 0%，就繼續播放下一首！**」

**底層運作機制：`while` 迴圈的三大關鍵生命要素**  
`while` 在英文中是「當……的時候」或「只要……就……」的意思。  
它的標準語法結構為：
```python
while 條件表示式:
    條件為 True 時執行的程式區塊
```
每一次進入迴圈肚子之前，Python 都會像哨兵一樣嚴格盤查「條件表示式」：
1. 若條件為 `True`，就大步跨入執行縮排內的指令。
2. 執行完該回合的最後一行後，程式流程會**自動折返回迴圈頂端，再次重新評估條件**！
3. 直到某一刻條件評估為 `False`，電腦才會優雅跳過整段縮排，繼續執行迴圈後方的程式碼。

若要用 `while` 迴圈完成一個安全的計數走訪，程式碼**必須同時具備以下三大生命要素**，缺一不可：
```python
count = 1        # 要素一：初始狀態設定（起跑點）
while count <= 5: # 要素二：迴圈延續的條件檢查（哨兵盤查）
    print(count)
    count += 1   # 要素三：狀態更新步伐（向終點邁進的關鍵動力！）
```

**初學者最可怕的惡夢陷阱：無窮迴圈（Infinite Loop）**  
如果上面的範例中，初學同學粗心**漏寫了 `count += 1`**，會發生什麼浩劫？  
因為 `count` 永遠是 1，條件 `1 <= 5` 將會永遠為真！電腦就會陷入瘋狂，無限印出 1，CPU 飆升到 100%，終端機畫面陷入瘋狂暴走，這就是程式界赫赫有名的**「無窮迴圈（當機地獄）」**！在 APCS 自動評判中，這會直接領到代表超時的「TLE」判定！

**APCS 考試實務建議：**  
每次寫下 `while` 關鍵字時，請立刻自問一句：**「迴圈內部到底有沒有哪一行程式碼，會促使條件往 False 的方向前進？」** 如果找不到那一行動力來源，請立刻補上！

````python
# [2] Code 範例區：展示 while 迴圈的基本計數與安全累加
# 目標：使用 while 迴圈計算 1 到 5 的整數總和

print("=== while 迴圈計數累加示範 ===")

total = 0
current = 1  # 生命要素一：初始值

# 生命要素二：延續條件
while current <= 5:
    total += current
    print("目前加入數字：", current, "，累計總和：", total)
    current += 1  # 生命要素三：步進更新（絕對不能漏掉！）

print("=== 迴圈安全結束 ===")
print("1 到 5 的最終總和為：", total)
````

### 🧅 6-4-2 純數值位數逐位拆解：`// 10` 與 `% 10` 的剝皮術——未知位數的整數分析

**生活比喻：剝洋蔥與撲滿掉硬幣**  
想像你手上拿著一顆洋蔥。洋蔥一層裹著一層，你想要逐層把洋蔥皮剝下來檢查。你從最外層開始：剝下一片（取出尾巴）、丟掉一層（縮小本體），直到整顆洋蔥全被剝光、手心空空為止！  
再想像一個裝滿 1 元零錢的筒子，底部有一個小機關。你按一下按鈕，掉出最後一枚硬幣（`% 10` 取出個位數），筒子裡的硬幣柱就往下降一格（`// 10` 砍掉個位數），直到筒子徹底空了（`n == 0`）！  
這就是著名的**「數值剝皮術」**！面對一個不知道是 3 位數、5 位數還是 10 位數的超大整數，只要有 `while` 迴圈與兩個小小的整數運算子，任何數字都難逃法網！

**底層運作機制與數學推導**  
在第二章 2.3 節中，我們學過整數除法商數 `//` 與模除餘數 `%` 的奇妙特性：
- **`digit = n % 10`**：任何正整數除以 10 的餘數，**必定恰好等於它的「個位數」**！
  - 例如：`4527 % 10` 得到的餘數是 `7`。
- **`n = n // 10`**：任何正整數整除以 10，**必定恰好等於「把個位數砍掉、整串數字向右平移一位」**！
  - 例如：`4527 // 10` 得到的商數是 `452`。

將這兩個算式放進 `while n > 0:` 的迴圈中，不可思議的魔法就誕生了：
1. 第一回合：$n = 4527$ ➔ 取出尾巴 $7$ ➔ $n$ 縮小為 $452$
2. 第二回合：$n = 452$ ➔ 取出尾巴 $2$ ➔ $n$ 縮小為 $45$
3. 第三回合：$n = 45$ ➔ 取出尾巴 $5$ ➔ $n$ 縮小為 $4$
4. 第四回合：$n = 4$ ➔ 取出尾巴 $4$ ➔ $n$ 縮小為 $0$
5. 第五回合：檢查條件 `n > 0` 變成 `False`，迴圈完美結束！

**初學者最容易掉入的兩大致命陷阱：**
1. **陷阱一：數字 0 的邊界特例防禦**  
   若題目輸入的整數剛好是 `n = 0`，因為一開始 `n > 0` 就是 `False`，`while` 迴圈連一次都不會跑！如果你的任務是統計位數，計算結果會變成 0 位數（但 0 明明有 1 位數）。因此遇到可能為 0 的題目，需額外留意初始判斷！
2. **陷阱二：忘了砍掉尾巴 `n //= 10`**  
   若忘了寫 `n //= 10`，$n$ 的值永遠不變，迴圈就會陷入無限取出同一個個位數的無窮輪迴。

**APCS 考試實務建議：**  
很多初學者在還沒學會純數學位數拆解時，會想偷偷轉成字串用字串長度與索引來解題。然而，在 APCS 許多實作題中（如 ZJ a038、c290 數值拆解、各類數值加密與校驗碼），直接在記憶體中進行純整數運算的速度，比字串建立與型態轉換快上數倍，且完全不受字串編碼限制，是競賽頂尖選手的必備基本功！

````python
# [2] Code 範例區：展示 while 搭配 // 10 與 % 10 逐位提取與累加
# 目標：計算整數 4527 的各位數字總和（4 + 5 + 2 + 7 = 18）

num = 4527
temp = num
digit_sum = 0

print("開始逐位拆解整數：", num)

while temp > 0:
    digit = temp % 10       # 步驟一：提取當前個位數
    digit_sum += digit      # 步驟二：累加位數數值
    print("抽離出的個位數：", digit, "，剩餘待處理數值：", temp // 10)
    temp = temp // 10       # 步驟三：向右移一位，砍掉個位數

print("各位數字加總結果：", digit_sum)
````

### 🧊 6-4-3 未知次數的折半與逼近：考拉茲猜想（3n + 1）與對數級縮小

**生活比喻：對折紙張與神秘的數學冰雹**  
想像你拿著一張巨大的世界地圖。每對折一次，紙張的面積就減少一半（`area //= 2`）。你不知道要折幾次才能把它塞進口袋，但你知道只要它還比口袋大，你就必須繼續折！這種每次「折半」的過程，在電腦科學中具備無與倫比的巨大威力，因為它能以驚人的速度將千萬巨量資料迅速縮小至微塵！  
再想像數學界最著名的一場狂暴冰雹——「考拉茲猜想（Collatz Conjecture，又稱 3n + 1 猜想）」：  
任選一個正整數 $n$，如果它是奇數，就把它變成 $3n + 1$（像熱氣流將冰雹往上吹高！）；如果它是偶數，就把它砍半 $n // 2$（像冰雹往下沉墜！）。數學家猜想：不論起點數字多麼巨大，這顆冰雹在空中翻騰之後，**最終必定會墜落回數字 1**！但究竟需要翻騰幾回合？事前沒有任何人能預知，唯有依賴 `while n != 1:` 逐回合模擬！

**底層運作機制與演算法意義**  
這種「未知收斂路徑、但知道目標終點」的模擬，正是 `while` 迴圈最無可取代的專屬舞台！  
如果用 `for` 迴圈，我們根本不知道該把 `range()` 的上限設為多少；但用 `while`，只要一句優雅的：
```python
while n != 1:
    if n % 2 == 0:
        n //= 2
    else:
        n = 3 * n + 1
    step_count += 1
```
電腦就會自動在數值變化的驚滔駭浪中一路追蹤，直到安全著陸在 $n = 1$ 的沙灘上！

**常見錯誤陷阱：**
1. **整數除法與浮點數污染**：在折半過程中，初學者若寫成單斜線除法 `n = n / 2`，數字會被悄悄轉變成浮點數 `float`（如 `4.0`）。當數值極大時，浮點數的精度誤差會造成奇偶性判斷錯誤（`4.0 % 2` 產生偏差），因此在整數問題中**務必始終堅持使用雙斜線整除 `//=`**！
2. **無窮震盪的恐懼**：在自訂條件時，如果條件設計不良（例如忘了考慮奇數偶數所有可能），程式可能會在某個循環中無限打轉。

**APCS 考試實務建議：**  
模擬題在 APCS 實作第二題非常常見（例如模擬機器人走步、模擬人口遷移、數值遊戲規則模擬）。這種題目的特徵就是「題目給定一套狀態變換規則，要求計算何時停止或停止時的狀態」，掌握 `while` 迴圈的狀態模擬架構是奪下實作高分的關鍵。

````python
# [2] Code 範例區：展示考拉茲猜想（3n + 1 冰雹猜想）模擬
# 目標：觀察正整數 6 的冰雹變換路徑，並統計經過幾步回到 1

n = 6
step = 0

print("開始模擬數字", n, "的考拉茲冰雹軌跡：")
print("起點：", n, end=" ")

while n != 1:
    if n % 2 == 0:
        n //= 2          # 偶數：折半
    else:
        n = 3 * n + 1    # 奇數：乘以 3 加 1
    step += 1
    print("➔", n, end=" ")

print("\n經過", step, "步成功著陸回到 1！")
````

### 🏛️ 6-4-4 古希臘智慧：歐幾里得輾轉相除法（求最大公因數 GCD）

**生活比喻：用最大正方形瓷磚鋪滿長方形地板**  
想像你是一位室內設計師，要在一間長 60 公尺、寬 48 公尺的長方形大廳鋪滿正方形瓷磚。要求瓷磚必須是「整塊鋪滿，完全不能裁切」。要讓瓷磚的尺寸最大，這塊正方形的邊長該是多少？  
古希臘數學大師歐幾里得（Euclid）給出了一個無與倫比的絕妙解法：  
先在 60 × 48 的空間裡鋪上一塊 48 × 48 的大正方形，剩下的是一塊 48 × 12 的長方形長條；  
接著把問題縮小：用 12 來鋪 48，發現 48 恰好可以被 12 整除（餘數是 0）！  
這意味著：邊長 12 就是能夠完美鋪滿整座大廳的「最大公因數（GCD, Greatest Common Divisor）」！

**底層運作機制與輾轉相除定理**  
輾轉相除法的數學原理是：  
給定兩數 $y$ 與 $x$（假設 $y \ge x$），若 $y$ 除以 $x$ 的商為 $m$、餘數為 $r$（即 $y = m \times x + r$），則：
$$\gcd(y, x) = \gcd(x, r)$$
**當餘數 $r$ 縮小為 0 的那一瞬間，當時的除數 $x$ 就是兩數的最大公因數！**

在 Python 中，結合第一章 1.3 節學過的「多變數同時賦值（Swap）」特性，我們可以用短短三行寫出全世界最優雅的演算法：
```python
while x > 0:
    y, x = x, y % x
print("最大公因數為：", y)
```
請屏氣凝神欣賞這行 `y, x = x, y % x`：
- 等號右側同時計算出「舊的除數 `x`」與「新的餘數 `y % x`」。
- 接著同時賦值：把舊除數晉升為新的被除數 `y`，把新餘數降格為新的除數 `x`！
- 當 `x` 最終變成 0 時，上一回合的除數（此刻已安坐在 `y` 中）就是最終勝出的 GCD！

**最小公倍數（LCM）的聯手出擊：**  
求出最大公因數 $\gcd(a, b)$ 後，最小公倍數 $\text{lcm}(a, b)$ 就能透過數學公式瞬間求得：
$$\text{lcm}(a, b) = \frac{a \times b}{\gcd(a, b)}$$
完全不需要再用迴圈慢慢尋找，速度極快！

**常見錯誤陷阱：**
1. **變數覆蓋順序錯誤**：如果不用 Python 的同時賦值，寫成 `y = x; x = y % x`，此時第一行的 `y` 已經被改成 `x`，第二行的 `y % x` 就變成了 `x % x`（永遠等於 0），演算法直接報廢！
2. **最後答案是印 `y` 不是 `x`**：因為迴圈結束的條件是 `x == 0`，所以跳出迴圈時 `x` 必定是 0，真正的最大公因數是儲存在 `y` 中！

**APCS 考試實務建議：**  
輾轉相除法是 APCS 歷年檢定、ZeroJudge 題庫（如 ZJ a024）的超級必考題！在分數約分、比例計算、齒輪比模擬中無處不在，請務必將這三行語法深深刻在腦海中！

````python
# [2] Code 範例區：展示歐幾里得輾轉相除法求最大公因數（GCD）
# 目標：求 48 與 60 的最大公因數

a = 60
b = 48

x = b
y = a

print("計算 gcd(", a, ",", b, ") 的輾轉相除過程：")

while x > 0:
    r = y % x
    print("目前被除數 y =", y, "，除數 x =", x, "，餘數 r =", r)
    y, x = x, r  # 同時更新：y 變成舊的 x，x 變成餘數 r

print("計算完成！最大公因數 GCD 為：", y)
````

### 🚪 6-4-5 `while True` 搭配 `break`：動態事件驅動與哨兵終止模式

**生活比喻：便利商店的自動感應門與電玩遊戲主循環**  
想像一家 24 小時營業的便利商店。自動感應門的控制程式不是寫「請開門 100 次然後永遠鎖死」，而是寫著一個永不打烊的引擎：  
「**永無休止地運行（`while True`）！**」  
在每一次的循環中，感應器都在監控：「有人靠近嗎？有的話開門；如果偵測到火災警報（特殊終止代碼），立刻切斷電源停止運轉（`break`）！」  
再比如你玩的所有手機遊戲或電腦遊戲，背後都有一個核心的「遊戲主循環（Game Loop）」：`while True:` 不斷更新畫面與玩家位置，直到玩家按下「退出遊戲」或「生命值歸零」才觸發 `break` 結束遊戲！

**底層運作機制與結構優勢**  
`while True` 代表將迴圈條件永遠寫死為真（`True`）。  
這聽起來好像是「故意製造無窮迴圈」，但只要在迴圈的肚子裡妥善安排了 **`if 條件: break`**，它就會蛻變為現代程式設計中最實用、最靈活的**「哨兵控制迴圈（Sentinel Loop）」**！

為什麼我們需要 `while True`？  
試比較以下兩種讀取「以 0 結尾的數列」的寫法：
- **傳統寫法（代碼重複）**：
  ```python
  x = int(input())   # 迴圈前要先讀一次（初始化）
  while x != 0:
      total += x
      x = int(input()) # 迴圈肚子最後一行還要再讀一次（更新步伐）
  ```
  初學者常覺得在外面讀一次、肚子裡又要讀一次非常囉嗦！
- **`while True` 優雅寫法（單一讀取點）**：
  ```python
  while True:
      x = int(input()) # 全集中在同一個入口！
      if x == 0:
          break        # 遇到哨兵代碼 0，立即下車！
      total += x
  ```
  整個流程「輸入 ➔ 檢查是否結束 ➔ 處理資料」井然有序，完全消除重複代碼！

**常見錯誤陷阱：**
1. **忘記設置出口 `break`**：寫了 `while True` 卻把 `break` 的條件寫錯或漏寫，導致程式在線上評判系統（OJ）中無窮等待輸入，引發 TLE 或記憶體爆掉！
2. **處理與中斷的順序顛倒**：如果把 `total += x` 寫在 `if x == 0: break` 前面，那麼最後那個用來表示結束的 `0`（或特定的終止標記如 `-1`）就會被錯誤地納入統計！

**APCS 考試實務建議：**  
許多 APCS 實作題目（如購物車模擬、以特定數字結尾的多筆輸入）都會以「輸入直到 0 為止」或「輸入直到 -1 為止」作為終止信號，`while True` 搭配 `break` 是處理這類輸入最乾淨、最不容易出錯的黃金寫法。

````python
# [2] Code 範例區：展示 while True 搭配 break 處理動態終止條件
# 目標：模擬累加數值，直到總和超過 50 即刻煞車

print("=== while True 動態達標示範 ===")

current = 1
total = 0

while True:
    total += current
    print("加入數值：", current, "，當前總額：", total)
    
    # 哨兵出口判斷
    if total >= 50:
        print("💡 總額已達標（>= 50），觸發 break 結束迴圈！")
        break
        
    current += 3

print("=== 迴圈已完全結束，最終總和：", total, "===")
````
