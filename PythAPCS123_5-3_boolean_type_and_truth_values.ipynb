### 💡 5-3-1 布林型態本質——True 與 False 的身世之謎與整數繼承特性

**生活比喻：電路開關與硬幣的兩面**  
在物理世界中，電燈開關只有兩種狀態：閉合導通（通電，On，數值為 1）與斷開不通（斷電，Off，數值為 0）。硬幣也只有正面與反面。  
在 Python 魔法王國中，這種「非黑即白」的狀態被封裝成了一種專屬的資料型態：**布林型態（`bool`）**。

**運作機制與底層原理：**
1. **只有兩位公民**：
   布林型態全世界只有兩個值：`True`（真）與 `False`（假）。請注意開頭字母必須**大寫**，不能寫成小寫的 `true` 或 `false`（否則會被電腦當成未宣告的變數名而拋出 `NameError`）。
2. **身為整數大家族的小兒子（整數繼承性）**：
   這是 Python 最迷人的底層設計之一——`bool` 型態其實是 `int`（整數）的子類別！
   - 在底層，**`True` 的數值就是整數 `1`**！
   - **`False` 的數值就是整數 `0`**！
   如果我們對它們進行型態轉型：`int(True)` 會得到 `1`，`int(False)` 會得到 `0`。
   甚至在數學加法中，`True + True` 會等於 `2`，`True + False` 會等於 `1`！

**常見錯誤陷阱：**
- **大小寫錯誤**：寫成 `TRUE` 或 `false` 均為語法錯誤。
- **加了引號變成字串**：若寫成 `"True"` 或 `'False'`，它們就變成了「文字字串（str）」，不再是純粹的布林型態！

**APCS 考試實務建議：**
善用「`True` 等於 1、`False` 等於 0」的特性，可以在很多計分與統計題型中免去繁複的邏輯判斷，直接一行算式解決問題！


````python
# [2] Code 範例區：展示布林型態本質與整數數值對應

# 1. 宣告布林變數並檢驗型態
is_pass = True
is_fail = False

print("is_pass 的值：", is_pass)
print("is_pass 的資料型態：", type(is_pass))  # <class 'bool'>
print("is_fail 的值：", is_fail)
print("is_fail 的資料型態：", type(is_fail))  # <class 'bool'>

# 2. 驗證 True 與 False 對應整數數值
print("--- 布林轉整數 ---")
print("int(True) 數值為：", int(True))    # 1
print("int(False) 數值為：", int(False))  # 0

# 3. 布林值的算術相加
print("--- 布林值參與數學運算 ---")
print("True + True + False =", True + True + False)  # 1 + 1 + 0 = 2
````

### 🔢 5-3-2 數值的真假值規則——非零為真、零為假

**生活比喻：存款餘額與物資庫存**  
想像你低頭看自己的錢包：
- 如果錢包裡有錢，無論是 100 塊、1000 塊，甚至欠債 -50 塊（數值不為 0），代表「錢包裡有數字／有帳務紀錄存在」，在狀態上是有實質內容的（**真值 Truthy**）。
- 如果錢包空空如也，餘額恰好為 0 塊，代表「毫無任何庫存與動靜」，在狀態上歸於寂靜（**假值 Falsy**）。

**運作機制與底層規則（吳邦一老師講義第 17 頁）：**
在 Python 中，所有整數（int）與浮點數（float）都可以被判定真假值：
1. **數值為 0（或浮點數 0.0）**：代表 **`False`（假值）**。
2. **任何非 0 的數值**（包括正數如 `1`, `7`, `999` 以及**負數**如 `-1`, `-42`）：一律代表 **`True`（真值）**！

**使用 `bool()` 函數進行查驗：**
```python
bool(0)     # False
bool(0.0)   # False
bool(1)     # True
bool(-7)    # True (負數也是非零，所以依然是 True！)
```

**常見錯誤陷阱（初學者極大盲區）：**
初學者常直覺認為「負數應該是 False 吧？因為它是負的」。  
**大錯特錯！** Python 的數值真假值鐵律只有一條：**「只有 0 才是 False，其餘所有非零數字（無論正負）全都是 True！」**

**APCS 考試實務建議：**
在題目中，如果一個變數代表計數器或錯誤代碼，當 `count != 0` 時往往可以簡寫為檢查其真假值。掌握「非零即真」是理解後續條件分支縮寫的重要先備知識。


````python
# [2] Code 範例區：展示數值的真假值規則（非零為真、零為假）

# 1. 檢驗數值 0 與 0.0
zero_int = 0
zero_float = 0.0
print("--- 數值 0 的真值 ---")
print("bool(0) 結果：", bool(zero_int))        # False
print("bool(0.0) 結果：", bool(zero_float))    # False

# 2. 檢驗正整數與正浮點數
print("--- 正數的真值 ---")
print("bool(1) 結果：", bool(1))            # True
print("bool(100) 結果：", bool(100))        # True
print("bool(3.14) 結果：", bool(3.14))      # True

# 3. 檢驗負數（🚨 特別警惕：負數也是 True！）
print("--- 負數的真值 ---")
print("bool(-1) 結果：", bool(-1))          # True
print("bool(-99.9) 結果：", bool(-99.9))    # True
````

### 📝 5-3-3 文字字串的真假值規則——真空為假、有字為真

**生活比喻：信件內容與白紙信封**  
想像郵差送來一個信封：
- 如果打開信封，裡面寫著密密麻麻的文字、甚至只有一個問號或一個空白鍵字元（只要不是完全空白的虛無），這封信就「有實質內容」（**真值 Truthy**）。
- 如果打開信封，裡面完全空無一物，長度為 0 的純粹空字串 `""`，就像寄來了寂寞，代表「毫無任何字元」（**假值 Falsy**）。

**運作機制與字串真假度量衡：**
在 Python 中，文字字串（str）的真假值完全取決於它的**長度**：
1. **空字串 `""`（或 `''`）**：
   長度為 0 的字串，經 `bool("")` 轉換後結果為 **`False`**！
2. **非空字串**：
   只要字串內部含有**至少一個字元**（長度 $\ge 1$），經 `bool(...)` 轉換後一律為 **`True`**！

**常見超級地雷（初學者極易翻車）：**
1. **含有空白鍵的字串 `" "` 不是空字串！**
   `bool(" ")` 是 `True`！因為空白鍵也是一個正當的 ASCII 字元（ASCII 32），長度為 1，因此它是有字元的非空字串！
2. **字串 `"0"` 或 `"False"` 是 True！**
   初學者常被騙：
   ```python
   bool("0")       # 結果為 True！
   bool("False")   # 結果為 True！
   ```
   因為對字串而言，電腦根本不管裡面寫什麼英文字或數字，只要有字，它就是 `True`！唯有長度為 0 的純空字串 `""` 才是 `False`！

**APCS 考試實務建議：**
在讀取字串測資時，`bool(s)` 常用來快速檢查「使用者是否有輸入任何文字」或「文字緩衝區是否已被掏空」。


````python
# [2] Code 範例區：展示文字字串的真假值規則

# 1. 檢驗長度為 0 的真正空字串
empty_str1 = ""
empty_str2 = ''
print("--- 空字串的真值 ---")
print('bool("") 結果：', bool(empty_str1))  # False
print("bool('') 結果：", bool(empty_str2))  # False

# 2. 檢驗一般文字字串
msg = "Python"
print("--- 一般文字的真值 ---")
print('bool("Python") 結果：', bool(msg))   # True

# 3. 🚨 極度致命地雷展示
space_str = " "       # 含有一個空白字元
zero_str = "0"        # 含有一個字元 '0'
false_str = "False"   # 含有五個英文字元

print("--- 字串內容地雷區 ---")
print('bool(" ") 含有空白鍵：', bool(space_str))    # True！
print('bool("0") 含有文字 0：', bool(zero_str))     # True！
print('bool("False") 含有文字：', bool(false_str))  # True！
````

### 🧪 5-3-4 bool() 函數強制轉型——探索萬物的真假值本質

**生活比喻：全功能真偽透視鏡**  
在童話故事中，有一面神奇的「真理照妖鏡」。任何物品只要拿到鏡子面前照一照，鏡子就會立刻浮現出一個字：凡是空無、寂滅、歸零的事物，鏡子顯示「假」；凡是有生機、有實質、非零非空的事物，鏡子顯示「真」！

在 Python 裡，**`bool()` 函數**就是這面真理照妖鏡！它可以接受任何資料型態作為引數，並嚴格依據 Python 內建的真值協定（Truth Value Testing），回傳標準的 `True` 或 `False`。

**Python 內建假值清單（Falsy Values 大匯總）：**
截至目前為止，我們所接觸過的所有 Python 物件中，**會被判定為 `False` 的只有以下少數族群**：
1. 布林值本身：`False`
2. 數值零：整數 `0`、浮點數 `0.0`
3. 空文字序列：空字串 `""`（長度為 0 的字串）
4. （後續章節會學到的特殊無值物件：`None`、空串列 `[]`、空集合 `set()`、空字典 `{}`）

**除了上述這些清單以外，其餘所有世界上的一切物件，經 `bool()` 轉換後一律全都是 `True`！**

**常見錯誤陷阱：**
- **誤將文字轉型為數值後再判斷**：
  如果使用者從鍵盤輸入 `"0"`，未經轉型的 `bool(input())` 得到的會是 `True`（因為非空字串）！若希望以數值 0 為判定標準，必須先轉為整數：`bool(int(input()))`！這兩者有天壤之別。

````python
# [2] Code 範例區：展示 bool() 強制轉型函數的綜合透視

# 1. 檢驗所有常見假值（Falsy）
print("--- 假值家族檢視（一律為 False） ---")
print("bool(False) ->", bool(False))
print("bool(0)     ->", bool(0))
print("bool(0.0)   ->", bool(0.0))
print('bool("")    ->', bool(""))

# 2. 檢驗所有常見真值（Truthy）
print("--- 真值家族檢視（一律為 True） ---")
print("bool(True)      ->", bool(True))
print("bool(1)         ->", bool(1))
print("bool(-5)        ->", bool(-5))
print('bool("A")       ->', bool("A"))
print('bool("Hello")   ->', bool("Hello"))

# 3. 🚨 經典對照組：輸入字串 "0" 的轉型差異
raw_input = "0"
print("--- 經典陷阱對照 ---")
print('直接對字串 bool("0") 轉型：', bool(raw_input))       # True (因為字串非空)
print('先轉整數 bool(int("0")) 轉型：', bool(int(raw_input))) # False (因為數值為 0)
````

### 🚀 5-3-5 布林值數值化運算黑科技——免條件分支的指標計數神技

**生活比喻：舉手投票快速點名器**  
在班級幹部選舉中，老師問：「*贊成的人請舉手！*」每有一位同學舉起手，計票員的計數器就按下一次（+1）；若沒舉手，計數器就不動（+0）。計票員不需要去盤問每一位同學為什麼沒舉手，他只要把所有舉起的手臂直接相加，就瞬間得到了總得票數！

**APCS 頂尖解題神技：布林相加（Boolean Addition）**  
初學者在遇到「*統計四個數字中有幾個大於 100*」這類題目時，通常會直覺想寫四個 `if` 分支去做累加。  
但是，善用 Python 中 **`True == 1` 且 `False == 0`** 的特性，我們可以寫出無比優雅的「純數學一行式」！

**核心公式演示：**
假設有三個數字 $a, b, c$，要統計「有幾個正數」：
```python
positive_count = (a > 0) + (b > 0) + (c > 0)
```
電腦在執行時：
- 若 $a=5, b=-2, c=8$
- `(a > 0)` 得到 `True`（數值為 1）
- `(b > 0)` 得到 `False`（數值為 0）
- `(c > 0)` 得到 `True`（數值為 1）
- 整體計算：`1 + 0 + 1 = 2`！
完全不需要寫任何一行 `if`，直接秒殺答案！

**常見錯誤陷阱：**
- **未加小括號**：在進行布林運算加法時，強烈建議每個條件都加上小括號 `(a > 0) + (b > 0)`，以確保比較運算優先級不會受到加號干擾。

**APCS 考試實務建議：**
這種技巧在 APCS 實作題（例如計算符合條件的顧客人數、統計達標科目數）中能極大地精簡代碼長度，減少縮排層次，降低出錯機率！


````python
# [2] Code 範例區：展示布林值直接相加統計條件達成數

# 1. 統計四位選手成績中達到及格門檻（>= 60）的人數
s1 = 85
s2 = 58
s3 = 92
s4 = 60

# 傳統寫法需要許多條件判斷，但在 Python 中可直接相加！
passed_count = (s1 >= 60) + (s2 >= 60) + (s3 >= 60) + (s4 >= 60)
print("四位選手成績：", s1, s2, s3, s4)
print("各科是否及格：", s1 >= 60, s2 >= 60, s3 >= 60, s4 >= 60)
print("及格總人數：", passed_count, "人")  # 1 + 0 + 1 + 1 = 3 人

# 2. 統計三個整數中有幾個是偶數（n % 2 == 0）
n1, n2, n3 = 14, 27, 30
even_count = (n1 % 2 == 0) + (n2 % 2 == 0) + (n3 % 2 == 0)
print("--- 偶數計數統計 ---")
print("14, 27, 30 中偶數的個數為：", even_count)  # 1 + 0 + 1 = 2
````
