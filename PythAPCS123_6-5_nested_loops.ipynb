### 🕰️ 6-5-1 雙重迴圈基礎機制：時鐘的分針與時針——外層帶動內層的維度展開

**生活比喻：掛在牆上的時鐘與閱兵方陣**  
請抬頭看一眼牆上的傳統指針時鐘。  
時鐘上面有兩根針：粗粗短短的「時針（外層迴圈）」與細細長長的「分針（內層迴圈）」。  
當外層的時針只從 1 點鐘挪動到 2 點鐘（只前進了小小 1 格）時，內層的分針必須做什麼？  
分針必須氣喘吁吁地從第 0 分鐘一路數到第 59 分鐘，整整跑滿一整圈（60 步）！  
接著時針走向 2 點鐘，分針又要再次從頭跑滿一整圈！  
這就是雙重迴圈最核心的鐵律：**「外層每走一步，內層走完一整輪！」**  
再比如國慶閱兵的方陣隊伍：有 3 排士兵，每排站 4 個人（共 3 列 4 行）。班長點名時：「第一排，報數！（內層 4 人報數完畢）換第二排，報數！（內層 4 人再報數一遍）」。

**底層運作機制與乘法原理**  
在語法結構上，雙重迴圈就是縮排的嵌套：
```python
for i in range(3):       # 外層迴圈：負責橫列（Rows），i 會是 0, 1, 2
    for j in range(4):   # 內層迴圈：負責直行（Columns），每輪 j 都會從 0, 1, 2, 3 重新出發
        print(i, j)
```
讓我們計算總執行次數：  
外層跑 3 輪，每一輪裡面內層都要跑 4 輪，所以最內層縮排的指令總共會被執行：  
$$3 \times 4 = 12 \text{ 次}$$
如果外層跑 $M$ 次、內層跑 $N$ 次，則總次數就是 $M \times N$。這在演算法時間複雜度中稱為 $O(M \times N)$ 或 $O(N^2)$。

**初學者最容易掉入的兩大致命陷阱：**
1. **陷阱一：變數名稱打架撞名（Shadowing 災難）**  
   外層與內層的控制變數**絕對不能使用同一個名字**！若寫成 `for i in range(3): for i in range(4):`，內層的 `i` 會覆蓋掉外層的 `i`，導致外層邏輯完全錯亂！最經典的命名慣例是外層用 `i`，內層用 `j`，第三層用 `k`。
2. **陷阱二：縮排層級認知混淆**  
   縮排 4 個空白是「屬於外層、但在內層之外」的區域；縮排 8 個空白才是「外層與內層同時生效」的最核心區域。放置指令時必須清楚知道它應該被執行幾次！

**APCS 考試實務建議：**  
雙重迴圈是 APCS 實作二級分與三級分題目的絕對骨幹！無論是二維網格走訪、矩陣運算、還是各類圖形繪製，本質上全部都是雙重迴圈的延伸。

````python
# [2] Code 範例區：展示雙重迴圈的外層與內層運作軌跡
# 目標：外層 i 走訪 1~3，內層 j 走訪 1~2，觀察執行順序與座標組合

print("=== 雙重迴圈座標走訪示範 ===")

for i in range(1, 4):  # 外層：i = 1, 2, 3
    print(">> 外層回合開始：i =", i)
    
    for j in range(1, 3):  # 內層：j = 1, 2
        print("    [內層執行] 座標點：(", i, ",", j, ")")
        
    print(">> 外層第", i, "回合結束！\n")

print("=== 全部雙重迴圈執行完畢 ===")
````

### ✖️ 6-5-2 九九乘法表排版大作戰：`end` 參數與換行節奏的精準掌控

**生活比喻：老式機械打字機的節奏感**  
想像你是一位正在操作復古機械打字機的作家。  
當你在打每一行的內容時，字車會橫向移動，每敲一個字母，字車就向右挪一格（這相當於 `print(內容, end=" ")`，不換行，只用空格隔開）；  
當這一整行的最後一個字敲完後，作家會瀟灑地伸出右手，用力將字車推桿往回一推——「喀噠！叮～」（這叫做 Carriage Return / 換行）！  
字車瞬間跳到下一行的最左端起點，準備開始打印全新的一行！  
雙重迴圈的輸出排版，正是這種打字機節奏的極致展現！

**底層運作機制與輸出三拍節奏**  
要印出整齊的表格或矩陣（如九九乘法表），程式碼**必須嚴格遵守「內層橫向印、外層尾端換行」的三拍節奏**：
1. **第一拍（橫向推進）**：在內層迴圈中，使用 `end=" "` 或 `end="\t"`，讓當前這一列的每一筆資料並排顯示在同一行。
2. **第二拍（內層收工）**：內層迴圈結束，表示這一橫列的資料已經全數印完。
3. **第三拍（換行歸位）**：**縮排退回到外層迴圈**，呼叫一次空的 `print()`！這行 `print()` 不印任何字，純粹只是為了觸發換行動作，讓下一輪外層能從新的一行開頭重新輸出！

以九九乘法表為例：
```python
for i in range(1, 10):       # 外層：被乘數 1~9
    for j in range(1, 10):   # 內層：乘數 1~9
        # 橫向印出單一算式，用空格隔開不換行
        print(i, "*", j, "=", i * j, end=" ")
    print()                  # 關鍵！退回外層換行，整排印完才跳行！
```

**初學者最容易掉入的兩大致命陷阱：**
1. **陷阱一：把空 `print()` 寫在內層肚子裡（一秒毀掉排版）**  
   若把 `print()` 放在內層縮排裡，每算一次 `i * j` 就換一行，九九乘法表直接變成一長條 81 行的細長麵條，完全失去二維表格的美感！
2. **陷阱二：忘記寫外層空 `print()`（全都擠在同一行）**  
   若漏掉了外層的 `print()`，整張九九乘法表共 81 個算式全部擠在同一行，橫向無限延伸，直接被評判系統判定輸出格式錯誤（PE/WA）！

**APCS 考試實務建議：**  
在 APCS 輸出題中，每列的行尾換行與項目間的空格要求極度嚴格。深刻體會「內層負責單列橫向內容，外層負責控制行與行交替」的分工哲學，是寫出零格式錯誤代碼的核心保證！

````python
# [2] Code 範例區：展示迷你 3x3 乘法表的精確排版控制

print("=== 迷你 3x3 乘法表 ===")

for i in range(1, 4):  # 外層：被乘數 1 到 3
    for j in range(1, 4):  # 內層：乘數 1 到 3
        # 橫向輸出，以空格隔開不換行
        print(i, "*", j, "=", i * j, end="  ")
        
    # 關鍵第三拍：退回外層，該列印完後換行！
    print()

print("=== 排版展示完畢 ===")
````

### 📐 6-5-3 直角三角形與幾何圖形：內層邊界與外層變數的動態綁定

**生活比喻：積木階梯與金字塔**  
想像你正在客廳用樂高積木蓋一座向上的樓梯：  
- 第 1 層（$i=1$），你放 1 塊積木；  
- 第 2 層（$i=2$），你放 2 塊積木；  
- 第 3 層（$i=3$），你放 3 塊積木……  
- 到了第 $n$ 層，你就放 $n$ 塊積木！  
請仔細思考：每一層要放幾塊積木，是固定的嗎？不是！**它完全由「你目前蓋到第幾層（外層變數 $i$）」來動態決定！**  
這就是寫出千變萬化幾何圖案的超級秘訣——**「將內層迴圈的 range 終點，綁定到外層的控制變數上！」**

**底層運作機制：動態內層邊界的設計心法**  
在前面的方陣中，內層的範圍是固定不變的（如 `range(w)`）；  
而如果我們把內層寫成：
```python
for i in range(1, n + 1):    # 外層：控制第幾列（從第 1 列到第 n 列）
    for j in range(i):       # 內層：第 i 列就跑 i 次！
        print("*", end="")
    print()
```
我們來追蹤電腦的腳步：
- 當 $i=1$ 時，內層是 `range(1)` ➔ 內層跑 1 次，印出 `*`
- 當 $i=2$ 時，內層是 `range(2)` ➔ 內層跑 2 次，印出 `**`
- 當 $i=3$ 時，內層是 `range(3)` ➔ 內層跑 3 次，印出 `***`
畫面立刻呈現出一座俐落優美的直角三角形！

**倒立三角形的數學公式推導：**  
如果我們要反過來印倒三角形（第 1 列印 $n$ 個、第 2 列印 $n-1$ 個……最後 1 列印 1 個）：  
我們只要讓外層使用負步進值倒數：`range(n, 0, -1)`，內層直接寫 `range(i)` 就能迎刃而解！

**初學者最容易掉入的兩大致命陷阱：**
1. **陷阱一：差一錯誤（Off-by-one 偏差）**  
   因為 `range` 左閉右開，若外層從 0 開始寫成 `range(n)`，第 0 列若是 `range(0)` 就會跑 0 次導致第一行變成空行！習慣上畫幾何圖形時，建議外層從 1 起算（`range(1, n + 1)`），邏輯最直觀不易錯。
2. **陷阱二：混淆符號印出次數**  
   內層的迴圈變數（例如 `j`）代表當前印到第幾顆星星，若要印數字階梯（如 1, 22, 333），請印出外層的 `i`；若要印序列階梯（如 1, 12, 123），請印出內層的 `j`！

### 🔍 6-5-4 雙層窮舉與數值枚舉搜尋（Brute-Force Enumeration）

**生活比喻：密碼鎖的兩個轉盤與古代雞兔同籠**  
想像你撿到一個有兩個轉盤的舊密碼箱。每個轉盤上面都有 0 到 9 的數字。你不知道正確密碼，你會怎麼辦？  
你一定會把第一個轉盤轉到 0，然後把第二個轉盤從 0 試到 9；接著把第一個轉盤撥到 1，再把第二個轉盤從 0 試到 9……直到 100 種組合全部試完！這就是電腦科學中最經典、最暴力的解題法——**「窮舉法（Brute-Force Enumeration）」**！  
再比如中國古代著名的數學名題「雞兔同籠」：  
籠子裡有雞和兔子，數頭有 10 個，數腳有 28 隻。人類可能要想解聯立方程式；但身為電腦，它每秒能運算幾億次，它才不需要列算式！電腦的思維是：「讓雞有 0 到 10 隻、兔子有 0 到 10 隻，用雙層迴圈枚舉出每一種可能，看哪一組的腳剛好是 28 隻！」

**底層運作機制與搜尋架構**  
雙層窮舉的標準模板如下：
```python
for x in range(min_x, max_x + 1):       # 枚舉未知數 x
    for y in range(min_y, max_y + 1):   # 枚舉未知數 y
        if 滿足題目設定的限制條件:
            print("找到解：x =", x, "y =", y)
```
這種方法看似「笨拙」，但在解題競賽中具備極大優勢：  
1. **絕對正確**：不會因為繁雜的代數運算或邊界推導出錯，凡走過必留下痕跡。
2. **適用性廣**：只要搜尋空間在幾百萬次以內（例如 $N \le 1000$，$N^2 = 10^6$），在 1 秒時限內電腦都能瞬間跑完！

**常見錯誤陷阱：**
1. **搜尋區間定得過大引發 TLE**：例如題目的總和上限是 50，很多初學同學卻把 `range` 寫成 `range(10000)`，造成多餘的幾百萬次空轉。設定緊湊合理的搜尋邊界是窮舉法的精髓！
2. **重複組合的排列冗餘**：在尋找無序數對 $(a, b)$ 時（例如 $a + b = 10$），如果不加限制，$(3, 7)$ 與 $(7, 3)$ 都會被印出。若題目要求不重複，可讓內層從外層起跑：`for b in range(a, max_val + 1):` 避免重複！

**APCS 考試實務建議：**  
在 APCS 實作第一題與第二題中，「暴力窮舉」是最快拿到分數的核心技巧（如 f312 人力分配、各類數值配對題）。切記：在競賽中，能用雙層迴圈 0.1 秒跑出正解的方法就是好方法！

````python
# [2] Code 範例區：雙層枚舉解決硬幣找零組合問題
# 目標：用 5 元硬幣與 1 元硬幣湊出 13 元，列出所有硬幣數量組合

target = 13
print("=== 湊出", target, "元的所有可能硬幣組合 ===")

# 5 元最多 2 枚（5*2=10 <= 13）
for five in range(0, target // 5 + 1):
    for one in range(0, target + 1):
        if five * 5 + one == target:
            print("可行方案：5 元硬幣", five, "枚，1 元硬幣", one, "枚")

print("=== 方案枚舉完畢 ===")
````

### 🚨 6-5-5 巢狀迴圈中的 `break` 範圍陷阱與多層跳出技巧

**生活比喻：大樓火警與只逃出房間**  
想像你住在一棟三層樓的旅館裡，每層樓有 4 間房間。  
有一天你在 2 樓的某間房間裡發現了火災，你大喊一聲「快跑！（`break`）」衝出房門——  
請注意：**當你衝出房門時，你只是逃出了「這間房間」，你依然還在「這棟大樓的走廊」裡面！**  
這就是初學者在巢狀迴圈中遇到最可怕的思維陷阱：  
**`break` 的煞車威力，永遠只會中斷「直接包覆它的那最內層迴圈」！外層迴圈根本完全不受影響，下一秒依然會準時邁向下一輪！**  
如果你想要連大樓一起逃出去（同時跳出外層與內層雙重迴圈），單靠一個 `break` 是絕對不夠的！

**底層運作機制與連鎖煞車神技**  
在雙重迴圈中，如果你找到了寶物想要「徹底結束所有運算」，我們有兩種經典解法：
1. **解法一：布林旗標（Flag）連鎖煞車法（初學者首選推薦！）**：  
   在最外層設定一個旗標 `found = False`。  
   當內層找到目標時，將 `found = True` 並執行內層 `break`；  
   當內層跳出後，外層迴圈**立刻檢查 `if found: break`**！這樣外層也會跟著踩下煞車，達成完美的雙層連鎖中斷！
   ```python
   found = False
   for i in range(1, 10):
       for j in range(1, 10):
           if i * j == 24:
               print("找到第一組解：", i, j)
               found = True
               break  # 跳出內層
       if found:
           break      # 連鎖跳出外層！
   ```
2. **解法二：自訂函數提早 `return`**（這會在第十一章函數章節傳授，能一瞬間終止所有層級）。但在第六章，我們專注掌握旗標連鎖跳出！

**常見錯誤陷阱：**
1. **誤以為一個 `break` 能摧毀全世界**：在矩陣搜尋題中，初學者常以為在內層寫個 `break` 就大功告成了，結果外層又繼續跑，導致印出了好幾組重複答案，慘遭 WA！
2. **旗標檢查縮排層級錯誤**：把 `if found: break` 寫在內層迴圈裡面（那等於白寫兩次內層 break），記住外層的檢查必須縮排 4 個空白，與內層的 `for` 同級對齊！

**APCS 考試實務建議：**  
在 APCS 矩陣搜尋題（如找第一個寶藏位置、迷宮起點）中，「布林旗標連鎖跳出」是保證程式不多跑、不超時的最穩固防線！

````python
# [2] Code 範例區：展示單層 break 陷阱 vs 旗標連鎖雙層 break

print("--- 實驗 1：只用一個 break（只跳出內層，外層繼續跑）---")
for i in range(1, 4):
    for j in range(1, 4):
        if j == 2:
            print("i =", i, "時內層遇 2 中斷！")
            break
        print("  執行 (", i, ",", j, ")")
print("實驗 1 結束：外層 1, 2, 3 依然全部跑完了！\n")

print("--- 實驗 2：使用 Flag 旗標連鎖雙層 break（徹底停止）---")
found = False
for i in range(1, 4):
    for j in range(1, 4):
        if i == 2 and j == 2:
            print("找到目標 (2, 2)！立旗並中斷內層！")
            found = True
            break
    # 外層連鎖檢查
    if found:
        print("外層偵測到旗標立起，連鎖中斷外層！")
        break
print("實驗 2 結束：成功提早完全收工！")
````
