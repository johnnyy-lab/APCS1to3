### 9.5.1 主對角線走訪規律（$r == c$）

在二維串列中，當矩陣的列數等於行數時，我們稱之為「**方陣（Square Matrix）**」（大小為 $N \times N$）。方陣具有許多獨特且對稱的幾何特性，其中最基本的就是「**主對角線（Main Diagonal）**」。

主對角線是指從矩陣的「左上角」一路延伸到「右下角」的那條對角線。讓我們觀察一個 $3 \times 3$ 方陣的主對角線座標：
- 第一個位置：`(0, 0)`
- 第二個位置：`(1, 1)`
- 第三個位置：`(2, 2)`

發現了嗎？主對角線上的每一個元素，其**列索引（Row Index）永遠等於行索引（Column Index）**，即數學上的 $r = c$！

許多初學者在走訪主對角線時，直覺會寫出雙重迴圈：
```python
# 雖然正確，但多做了許多無效運算的寫法：
for r in range(N):
    for c in range(N):
        if r == c:
            total += grid[r][c]
```
這樣的寫法一共檢查了 $N \times N$ 次條件，時間複雜度是 $O(N^2)$。但因為我們已經明確知道「列等於行」，因此只要使用一個走訪變數 $i$（從 $0$ 到 $N-1$），直接存取 `grid[i][i]`，就能以**單層迴圈**完成任務！時間複雜度直接降為 $O(N)$，這是在 APCS 考試中極為重要的時間優化思維。


````python
# 範例 9.5.1：使用單層迴圈走訪 3x3 方陣的主對角線並計算總和

demo_matrix = [
    [10,  2,  3],
    [ 4, 20,  6],
    [ 7,  8, 30]
]

demo_N = len(demo_matrix)
demo_diag_elements = []
demo_diag_sum = 0

# 單層迴圈走訪主對角線：列索引與行索引都是 demo_i
for demo_i in range(demo_N):
    demo_val = demo_matrix[demo_i][demo_i]
    demo_diag_elements.append(demo_val)
    demo_diag_sum += demo_val

print("矩陣維度：", demo_N, "x", demo_N)
print("主對角線元素列表：", demo_diag_elements)
print("主對角線元素總和：", demo_diag_sum)
````

### 9.5.2 副對角線走訪規律（$r + c == N - 1$）

了解了主對角線（左上到右下）後，接下來我們要認識方陣中另一條重要的軸線——「**副對角線（Anti-Diagonal）**」（也有人稱為次對角線或逆對角線）。

副對角線是從矩陣的「右上角」一路延伸到「左下角」的連線。讓我們同樣以 $3 \times 3$ 方陣（$N=3$）為例，觀察副對角線經過的座標：
- 第一個位置（右上）：`(0, 2)`
- 第二個位置（正中）：`(1, 1)`
- 第三個位置（左下）：`(2, 0)`

若是 $4 \times 4$ 方陣（$N=4$）：
- 座標分別為：`(0, 3)`, `(1, 2)`, `(2, 1)`, `(3, 0)`

請觀察這組座標的數學規律：
1. 列索引 $r$ 從 $0$ 逐步遞增到 $N-1$。
2. 行索引 $c$ 從 $N-1$ 逐步遞減到 $0$。
3. 最核心的代數關係是：**$r + c = N - 1$**！
換言之，只要我們知道列索引 $r$（或設迴圈變數為 $i$），那麼行索引 $c$ 就一定是 **$N - 1 - i$**。

因此，走訪副對角線同樣**完全不需要雙重迴圈**！我們只需利用單層迴圈，每次存取 `grid[i][N - 1 - i]`，即可在 $O(N)$ 的時間內走訪完整條副對角線。


````python
# 範例 9.5.2：走訪 3x3 方陣的副對角線並列印元素與座標

demo_matrix = [
    [10, 20, 99],
    [40, 88, 60],
    [77, 80, 90]
]

demo_N = len(demo_matrix)
print("方陣階數 N =", demo_N)

# 單層迴圈走訪副對角線：列為 demo_i，行為 demo_N - 1 - demo_i
for demo_i in range(demo_N):
    demo_c = demo_N - 1 - demo_i
    demo_val = demo_matrix[demo_i][demo_c]
    print(f"列 {demo_i}, 行 {demo_c} -> 元素值 = {demo_val}")
````

### 9.5.3 X 形交叉雙對角線與中心點重複防禦

當我們將「主對角線」與「副對角線」放在同一張地圖上看時，它們交織成了一個醒目的「**X 形狀**」。在許多幾何演算法、棋盤遊戲（如五子棋、圍棋視線）或統計題中，常需要計算兩條對角線上所有數字的總和或聯集。

這時存在一個非常隱蔽但致命的邏輯陷阱——**中心點重複計算（Center Overlap）**！

讓我們比較兩種維度的方陣：
1. **偶數階方陣（$N$ 是偶數，如 $4 \times 4$）**：
   - 主對角線座標：`(0,0), (1,1), (2,2), (3,3)`
   - 副對角線座標：`(0,3), (1,2), (2,1), (3,0)`
   - 兩組座標完全互斥，彼此沒有交集。
2. **奇數階方陣（$N$ 是奇數，如 $3 \times 3$ 或 $5 \times 5$）**：
   - 以 $3 \times 3$ 為例：
     - 主對角線：`(0,0), (1,1), (2,2)`
     - 副對角線：`(0,2), (1,1), (2,0)`
   - 注意到了嗎？中央點 `(1, 1)`（即 `(N//2, N//2)`）**同時出現在兩條對角線上**！

如果我們分別計算主對角線和副對角線的總和並相加：
```python
total = main_sum + anti_sum  # 陷阱！中心點被加了兩次！
```
當 $N$ 是奇數時，中央點會被計算兩次，導致結果產生偏差。因此，嚴謹的程式設計師會做**重複防禦（Deduplication Defense）**：
- 寫法 A：先各自累加，若 `N % 2 != 0`，則最後減去重複的中心點 `grid[N // 2][N // 2]`。
- 寫法 B：在單層迴圈中同時走訪，當 `i != N - 1 - i` 時才將副對角線的值加入。


````python
# 範例 9.5.3：計算 3x3 方陣 X 形對角線所有不重複元素的總和

demo_matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

demo_N = len(demo_matrix)
demo_x_sum = 0

for demo_i in range(demo_N):
    # 加上主對角線元素
    demo_x_sum += demo_matrix[demo_i][demo_i]
    
    # 加上副對角線元素，但如果此時列索引等於副對角線行索引（代表正是中心點），則不重複加！
    demo_anti_c = demo_N - 1 - demo_i
    if demo_i != demo_anti_c:
        demo_x_sum += demo_matrix[demo_i][demo_anti_c]

print("矩陣階數：", demo_N)
print("X 形對角線不重複元素總和：", demo_x_sum)
````

### 9.5.4 上三角與下三角矩陣走訪

在方陣中，除了單純的一條線（對角線），我們常常需要處理被對角線切開的「半邊區域」，這在數學與電腦科學中稱為**三角矩陣（Triangular Matrix）**：
1. **上三角區域（Upper Triangular）**：
   - 包含主對角線及其右上方所有的格子。
   - 數學特徵：**$r \le c$**（行座標大於或等於列座標）。
   - 若不包含對角線，則稱為「嚴格上三角（Strictly Upper Triangular）」，條件為 **$r < c$**。
2. **下三角區域（Lower Triangular）**：
   - 包含主對角線及其左下方所有的格子。
   - 數學特徵：**$r \ge c$**（列座標大於或等於行座標）。
   - 若不包含對角線，則稱為「嚴格下三角（Strictly Lower Triangular）」，條件為 **$r > c$**。

在撰寫迴圈走訪三角區域時，最優雅的做法不是「走訪全矩陣再用 `if` 過濾」，而是**直接動態調整內層迴圈的範圍**！
- 走訪上三角（含對角線）：
  列從 $0$ 到 $N-1$；當列為 $r$ 時，行只要從 $r$ 走到 $N-1$ 即可：
  ```python
  for r in range(N):
      for c in range(r, N):  # 內層起點直接綁定外層 r！
          # 存取 grid[r][c]
  ```
- 走訪下三角（含對角線）：
  列從 $0$ 到 $N-1$；當列為 $r$ 時，行只要從 $0$ 走到 $r$ 即可：
  ```python
  for r in range(N):
      for c in range(r + 1):  # 內層只走到 r (包含 r)
          # 存取 grid[r][c]
  ```
透過動態範圍控制，程式少走訪了一半的格子，執行效率立即提升一倍！

### 9.5.5 黑白棋盤格座標規律（$(r + c) \% 2$）

在西洋棋或圍棋盤中，相鄰的格子顏色永遠不同——黑格周圍是白格，白格周圍是黑格，交錯分佈。這種在二維陣列中交替出現的圖形規律，稱為「**棋盤染色模型（Checkerboard Coloring）**」。

棋盤染色背後有一個非常優雅且強大的數學公理：
若我們將列座標 $r$ 與行座標 $c$ 相加，觀察其奇偶性：
- `(0, 0)`：$0 + 0 = 0$（偶數）
- `(0, 1)`：$0 + 1 = 1$（奇數）
- `(0, 2)`：$0 + 2 = 2$（偶數）
- `(1, 0)`：$1 + 0 = 1$（奇數）
- `(1, 1)`：$1 + 1 = 2$（偶數）
- `(1, 2)`：$1 + 2 = 3$（奇數）

**結論**：
任意格子的座標 $(r, c)$，只要計算 **$(r + c) \% 2$**：
- 如果等於 $0$，代表該格子屬於「**偶數格（通常設為白格）**」。
- 如果等於 $1$，代表該格子屬於「**奇數格（通常設為黑格）**」。

只要往上下左右走一步（$r$ 或 $c$ 變化 $1$），$(r + c)$ 的奇偶性必然反轉！因此，相鄰兩格的 $(r+c) \% 2$ 絕對不相等。這個規律在 APCS 中經常用於網格染色、二維地圖交替障礙物、跳棋走法判定等高頻考題中。


````python
# 範例 9.5.5：產生一個 4x4 的黑白交錯棋盤矩陣（0 表示白格，1 表示黑格）

demo_R, demo_C = 4, 4
demo_board = []

for demo_r in range(demo_R):
    demo_row = []
    for demo_c in range(demo_C):
        # 依據 (demo_r + demo_c) % 2 決定填入 0 或 1
        if (demo_r + demo_c) % 2 == 0:
            demo_row.append(0)
        else:
            demo_row.append(1)
    demo_board.append(demo_row)

print("4x4 棋盤矩陣：")
for demo_row in demo_board:
    print(*demo_row)
````

### 9.5.6 方陣對稱性檢驗（$grid[r][c] == grid[c][r]$）

在數學與圖論中，「**對稱矩陣（Symmetric Matrix）**」是一種非常基礎且重要的結構（例如無向圖的相鄰矩陣必為對稱矩陣）。

什麼是對稱矩陣？
如果一個方陣沿著「主對角線」對折翻轉後，重疊位置的數值完全相同，那麼這個方陣就是對稱矩陣。
以代數符號表示：對於所有有效的列 $r$ 與行 $c$，都必須嚴格滿足：
$$grid[r][c] == grid[c][r]$$

在檢查對稱性時，有兩個關鍵的實作考量：
1. **比較範圍優化**：
   - 如果我們對所有 $(r, c)$ 都檢查一次，那麼 `(0, 1)` 和 `(1, 0)` 會被檢查兩次，且對角線上的 `grid[i][i] == grid[i][i]` 是毫無意義的自我比較。
   - 正確的高效做法：**只走訪「嚴格上三角區域」（$r < c$）**！只要所有的嚴格上三角元素都等於其對應的下三角元素，即可確認整個矩陣對稱。
2. **提早結束（Early Exit）與狀態旗標**：
   - 一旦在比對過程中發現任何一對 `grid[r][c] != grid[c][r]`，我們就能斷定該矩陣「絕對不對稱」，應立即將旗標變數設為 `False` 並以 `break` 跳出迴圈，不再浪費時間檢查後續格子。


````python
# 範例 9.5.6：使用旗標變數檢驗 3x3 方陣是否為對稱矩陣

demo_matrix = [
    [1, 7, 3],
    [7, 4, 8],
    [3, 8, 9]
]

demo_N = len(demo_matrix)
demo_is_symmetric = True

# 只檢查嚴格上三角區域 (demo_r < demo_c)
for demo_r in range(demo_N):
    for demo_c in range(demo_r + 1, demo_N):
        if demo_matrix[demo_r][demo_c] != demo_matrix[demo_c][demo_r]:
            demo_is_symmetric = False
            break
    if not demo_is_symmetric:
        break

print("方陣階數：", demo_N)
if demo_is_symmetric:
    print("檢驗結果：這是一個對稱矩陣！")
else:
    print("檢驗結果：這不是對稱矩陣！")
````

````python
#@title 🧪 點擊此處取得單元 9-5 學習成效自動評分報告 { display-mode: "form" }
# ------------------------------------------------------------------------------
# 官方智慧全環境評分載入器（支援：本地開發庫優先 ➔ 公開正式庫雙重備援）
# ------------------------------------------------------------------------------
import os
import sys
import json
import logging
import urllib.request

# 靜音內部 httplib2 警告
logging.getLogger("google_auth_httplib2").setLevel(logging.ERROR)

# 1. 互動式輸入學員姓名與學號（分開獨立欄位，按 Enter 採用預設值）
print("👤【學員身分登記】（自學測試可直接按 Enter 鍵使用預設值）：")
student_name = input("請輸入您的姓名 [預設: 自學冒險者]：").strip() or "自學冒險者"
student_id = input("請輸入學號或座號 [預設: 無座號]：").strip() or "無座號"

# 2. 自動嘗試獲取 Google 帳號官方身分認證（真實 Email 與 Google 暱稱）
google_email = None
google_name = None
try:
    from google.colab import auth
    print("🔐 正在連結 Google 帳號進行身分認證（若跳出授權視窗，請點選允許）...")
    auth.authenticate_user()

    # 方法 1: 使用 Google Drive API v3 (about:get) 抓取真實 displayName 與 emailAddress
    # Colab ADC 預設包含 drive 權限，能 100% 精準獲取 Google 帳號中文顯示名稱（如：葉明昌）
    try:
        import google.auth
        from googleapiclient.discovery import build
        creds, _ = google.auth.default()
        drive_service = build('drive', 'v3', credentials=creds, cache_discovery=False)
        about = drive_service.about().get(fields="user(displayName,emailAddress)").execute()
        u_obj = about.get("user", {})
        google_name = u_obj.get("displayName")
        google_email = u_obj.get("emailAddress")
    except Exception:
        pass

    # 方法 2: 若 Drive API 失敗，嘗試透過 OAuth2 userinfo / tokeninfo 備援
    if not google_email or not google_name:
        try:
            import google.auth
            from google.auth.transport.requests import Request
            credentials, _ = google.auth.default()
            credentials.refresh(Request())
            token = credentials.token

            for ep in [
                f"https://www.googleapis.com/oauth2/v3/tokeninfo?access_token={token}",
                "https://www.googleapis.com/oauth2/v3/userinfo",
                "https://www.googleapis.com/oauth2/v2/userinfo"
            ]:
                try:
                    req = urllib.request.Request(
                        ep,
                        headers={"Authorization": f"Bearer {token}"}
                    )
                    with urllib.request.urlopen(req, timeout=5) as resp:
                        u_data = json.loads(resp.read().decode('utf-8'))
                        if not google_email:
                            google_email = u_data.get("email") or u_data.get("emailAddress")
                        if not google_name:
                            google_name = (
                                u_data.get("name")
                                or u_data.get("given_name")
                                or u_data.get("displayName")
                            )
                        if google_email and google_name:
                            break
                except Exception:
                    continue
        except Exception:
            pass

    # 方法 3: 若只有 email 但 Google API 沒給 name，以 email 前綴作為暱稱回退
    if google_email and not google_name:
        google_name = google_email.split("@")[0]

    if google_email:
        print(f"✅ 身分認證成功！已綁定 Google 帳號：{google_name} ({google_email})")
except Exception:
    # 非 Colab 環境、本機離線測試或略過授權
    pass

# 3. 身分資料獨立注入環境變數（學號與姓名完全分開）
os.environ["STUDENT_NAME"] = student_name
os.environ["STUDENT_ID"] = student_id
if google_email:
    os.environ["GOOGLE_EMAIL"] = google_email
if google_name:
    os.environ["GOOGLE_NAME"] = google_name

# 4. 評分核心腳本載入（本機優先 ➔ 遠端記憶體載入，絕不產生實體暫存檔）
# 找不到來源就換下一條。內容已拿到但執行失敗時，停止改試其他來源。
# 評分腳本放在獨立命名空間執行，成功後也不把名稱寫回筆記本。答案讀執行前的學生存放格快照。
_loaded = False
_load_error = None

# 4.1 優先嘗試讀取開發庫/本機檔案（本機離線開發測試零延遲）
_local_paths = [
    "web/graders/grade_9_5.py",
    "../graders/grade_9_5.py",
    "../../graders/grade_9_5.py",
    "../web/graders/grade_9_5.py",
    "../../web/graders/grade_9_5.py"
]
for _lp in _local_paths:
    if not os.path.exists(_lp):
        continue
    try:
        with open(_lp, "r", encoding="utf-8") as _f:
            _code = _f.read()
        _grader_ns = {"__builtins__": __builtins__, "__name__": "apcs_grader", "__STUDENT_ENV__": dict(globals())}
        try:
            exec(_code, _grader_ns)
        finally:
            del _grader_ns
        os.environ["GRADER_SOURCE"] = _lp
        _loaded = True
        break
    except Exception as _exc:
        _load_error = _exc
        break

# 4.2 若在雲端環境（Colab），自動透過公開正式庫雙重備援以純記憶體載入
if not _loaded and _load_error is None:
    _grader_urls = [
        "https://raw.githubusercontent.com/johnnyy-lab/APCS1to3/main/graders/grade_9_5.py",
        "https://johnnyy-lab.github.io/APCS1to3/graders/grade_9_5.py",
        "https://raw.githubusercontent.com/johnnyy-lab/APCS1to3-dev/main/web/graders/grade_9_5.py"
    ]
    for _url in _grader_urls:
        try:
            _req = urllib.request.Request(_url, headers={"User-Agent": "Mozilla/5.0"})
            with urllib.request.urlopen(_req, timeout=10) as _resp:
                _code = _resp.read().decode("utf-8")
        except Exception:
            continue
        try:
            _grader_ns = {"__builtins__": __builtins__, "__name__": "apcs_grader", "__STUDENT_ENV__": dict(globals())}
            try:
                exec(_code, _grader_ns)
            finally:
                del _grader_ns
            os.environ["GRADER_SOURCE"] = _url
            _loaded = True
            break
        except Exception as _exc:
            _load_error = _exc
            break

if not _loaded:
    if _load_error is not None:
        print("❌ 評分程式載入失敗，請稍後重新執行。若仍失敗，請把這個畫面交給老師。")
        if os.environ.get("GRADER_DEBUG", "").strip().lower() in ("1", "true", "yes", "on"):
            _reason = str(_load_error).strip().splitlines()
            _line = _reason[0][:180] if _reason else ""
            if _line:
                print("（開發診斷：" + type(_load_error).__name__ + "：" + _line + "）")
            else:
                print("（開發診斷：" + type(_load_error).__name__ + "）")
    else:
        print("❌ 評分系統連線異常，請確認網路連線狀態（或稍後重新執行）")
````

## 本單元重點回顧與核心心法

在單元 9-5 中，我們深入探索了 $N \times N$ 方陣的幾何結構與座標代數規律：

1. **主對角線（Main Diagonal）**：
   - 座標規律：$r == c$。
   - 走訪心法：利用單層迴圈 `grid[i][i]`，時間複雜度為 $O(N)$，無需雙重迴圈。
2. **副對角線（Anti-Diagonal）**：
   - 座標規律：$r + c == N - 1$，故行座標為 $N - 1 - r$。
   - 走訪心法：單層迴圈 `grid[i][N - 1 - i]`，同樣只需 $O(N)$。
3. **X 形交叉雙對角線與重複防禦**：
   - 當 $N$ 為奇數時，中央元素 `(N // 2, N // 2)` 同時落在兩條對角線上。
   - 若做累加或聯集運算，務必進行「去重防禦」（判斷 $i \ne N - 1 - i$ 或於最後減去一次中心值）。
4. **上三角與下三角動態範圍控制**：
   - 上三角（$r \le c$）：外層 $r \in [0, N)$，內層 $c \in [r, N)$。
   - 下三角（$r \ge c$）：外層 $r \in [0, N)$，內層 $c \in [0, r + 1)$。
   - 嚴格三角（$<$ 或 $>$）只需微調起點或終點偏移量。
5. **棋盤格染色規律（$(r + c) \% 2$）**：
   - 相鄰格子和的奇偶性必然交替，偶數格與奇數格形成天然黑白交錯圖樣。
6. **方陣對稱性檢驗（$grid[r][c] == grid[c][r]$）**：
   - 只需走訪嚴格上三角（$r < c$），一旦失配立即 `break` 結束，展現最佳時間效率。
