但是，當你踏入 Python 的程式魔法世界時，請立刻把腦袋裡的「數學天平」切換掉！在 Python 裡，單一個等號 `=` 根本不是「等於」的意思，它的真正身分叫做「**賦值運算子（Assignment Operator）**」！你可以把它想像成一條「**單向強力輸送帶**」。

這條輸送帶的箭頭永遠朝著左邊指：`左邊變數箱子 ⬅️ 右邊數值`。它的唯一使命，就是把等號右邊的數值打包裝箱，單向傳送到左邊貼著變數名牌的置物櫃裡保存。

舉例來說，當我們寫下：
```python
apple_count = 5
```
我們並不是在證明什麼數學真理，而是在命令電腦：「小助手，請把 5 顆蘋果，透過輸送帶送進貼著 `apple_count` 名牌的箱子裡！」

牢記這個「單向輸送」的觀念，是擺脫數學慣性思維、邁向 APCS 程式高手的最關鍵第一步喔！

### ⚙️ 1-2-2 賦值的黃金順序——「先算右邊，再存左邊」

既然等號是一條單向輸送帶，那麼當等號右邊出現長長的計算式子時，電腦是怎麼處理的呢？

請大家在心裡默念這句八字真言：「**先算右邊，再存左邊！**」這是全世界所有程式語言共通的黃金鐵律。

想像一下披薩店的大廚備料：當客人點了一份綜合披薩，大廚絕對不會把生麵粉、起司條和生番茄一顆顆直接塞進外送紙盒裡；大廚一定會先在右邊的料理檯上，把所有食材切好調配、放進烤箱烤熟，算出一塊熱騰騰的完整披薩之後，才會在最後一步把它裝進左邊的外送盒子裡！

在 Python 裡也是一模一樣的運作流程。當我們寫出：
```python
x = 5 + 3 * 6
```
直譯器小助手看到這行指令時，它的眼睛會立刻專注在等號的右半邊：
1. **第一步（右側計算）**：遵守國中數學「先乘除後加減」的優先規則，先計算 $3 \times 6 = 18$。
2. **第二步（右側加總）**：再計算 $5 + 18 = 23$。
3. **第三步（賦值裝箱）**：當右邊全部算乾淨、得到單一數值 23 後，輸送帶才正式啟動，把 23 一口氣送進左邊貼著名牌 `x` 的箱子中保存！

因此，左邊的變數只會收到最後算出來的最終成果（23），完全不會看到中間算式的過程喔！

````python
# [2] Code 範例區：右側算式先完整計算（5 + 3*6 = 23），最後才將結果存入 x
x = 5 + 3 * 6
print(x)
````

### 🚫 1-2-3 等號左側的嚴格鐵律——左邊只能是「單一變數箱子」！

在吳邦一老師的講義第 5 頁中，特別標註了一行非常醒目的經典警示：
```python
y + 3 = 6  # (這是錯的！)
```

很多剛上國一的同學看到這行程式碼都會納悶：「老師，在國中數學的一元一次方程式中，$y + 3 = 6$ 不是很正常、很標準的等式嗎？只要心算移項一下，就知道 $y = 3$ 啦！」

但是在程式語言的世界裡，這樣寫會立刻引發大爆炸！Python 直譯器會毫不留情地亮出紅牌警告：
`SyntaxError: cannot assign to expression`（語法錯誤：無法賦值給運算式！）

為什麼電腦會這麼生氣呢？讓我們再次回到「置物箱」與「單向輸送帶」的比喻：
等號輸送帶的功能，是把右邊的實體寶物倒進左邊的容器裡。因此，**等號的左邊，必須是一個真真切切、能夠打開蓋子裝東西的「單一變數箱子」！**

你想想看：`6` 是一個寶物，但是 `y + 3` 是一道數學算式，它根本不是一個實體置物箱！你怎麼可能把一個實體的寶物硬塞進一道數學題目裡面呢？
同樣地，如果你寫成 `10 = x` 也是大錯特錯，因為 `10` 是一塊固定死板的石頭（常數值），它也不是能裝東西的箱子！

所以請牢牢記住：電腦不像人類大腦懂得自動移項。如果你想求出 $y$，身為魔法師的你必須自己先在腦袋中移項完成，寫成：
```python
y = 6 - 3
```
讓等號左邊永遠保持是乾乾淨淨、孤身一人的「單一變數名牌」，右邊才是要計算的數值或式子喔！

````python
# [2] Code 範例區：正確寫法！左邊必須是單一變數名牌，右邊才是計算式
# 錯誤示範（切勿執行）：y + 3 = 6 會造成 SyntaxError
# 正確寫法：
y = 6 - 3
print(y)
````

### 🔄 1-2-4 舊的不去新的不來——變數賦值的覆蓋（Overwrite）特性

想像一下，你桌上有一個貼著「今日下午茶」名牌的神奇置物箱。

中午的時候，你在箱子裡放進了一塊草莓蛋糕；但是到了下午三點，你改變主意了，又往同一個箱子裡塞進了一顆大紅蘋果。這時候，如果打開箱子看，裡面會同時有蛋糕和蘋果嗎？

答案是「絕對不會」！Python 的變數置物箱遵循著一條非常乾脆俐落的物理法則：**一個變數箱子在同一個瞬間，永遠只能存放單一筆最新資料！**

在電腦科學中，這個現象稱為「**覆蓋（Overwrite）**」。每當等號輸送帶把新的數值送進已經有名牌的箱子時，箱子底部的神秘暗門就會瞬間打開，把原本住在裡面的舊寶物一口氣丟進垃圾桶徹底銷毀，只保留最新送進來的那樣東西！

舉例來說：
```python
score = 60
score = 95
print(score)
```
當第一行執行完畢時，`score` 箱子裡裝的是 60；但是當第二行執行時，95 被輸送帶送了進來，原本的 60 就永遠蒸發消失了！所以最後一行印出的結果，百分之百是最新放進去的 95。

這提醒我們：在寫程式時如果後面還要用到舊資料，一定要記得多開一個新的變數箱子先將它備份保存起來喔！

````python
# [2] Code 範例區：展示變數數值被新賦值覆蓋的過程
item = 10
print(item)  # 此時印出舊的數值：10

item = 99    # 新數值 99 進入，舊數值 10 被覆蓋銷毀
print(item)  # 此時印出最新數值：99
````

### 🪄 1-2-5 自己算自己——右側引用舊值自我更新（`x = x + 1` 的魔法）

現在我們要結合前幾小節的所有超能力，來破解程式設計史上最著名、也最容易嚇到初學者的傳奇指令：
```python
x = x + 1
```

如果是一位剛學一元一次方程式的國一同學第一次看到這行程式碼，肯定會拍桌子大喊：「老師！這在數學上根本是荒謬的大錯誤嘛！一個數字怎麼可能加上 1 之後還等於它自己？這在數學上根本無解！」

同學請息怒！請再次回想起我們在 **1-2-1** 與 **1-2-2** 學到的兩大鐵則：
1. **等號不是等於**，而是把右邊算好的東西搬到左邊的「單向輸送帶」！
2. 永遠「**先算右邊，再存左邊**」！

現在，讓我們用慢動作分解鏡頭，看看電腦究竟是怎麼執行 `x = x + 1` 的：
* **起始狀態**：假設目前 `x` 箱子裡裝著原本的舊數值 `10`。
* **步驟一（先看右邊）**：電腦小助手的眼睛先看等號右邊的算式 `x + 1`。它跑去打開目前的 `x` 箱子，拿出裡面的舊數字 `10`，並在料理檯上計算 $10 + 1 = 11$。
* **步驟二（再存左邊）**：小助手捧著剛剛熱騰騰算好的新成果 `11`，啟動等號輸送帶，送往左邊貼著 `x` 名牌的箱子！
* **步驟三（舊值覆蓋）**：根據覆蓋原則，舊的 `10` 被踢出箱子，新的 `11` 正式住進了 `x` 箱子！

看清楚了嗎？**右邊的 `x` 代表的是「過去的舊數值」，而左邊的 `x` 代表的是「準備迎接新數值的家」！**

這種「自己引用自己舊值來累加更新」的寫法，在遊戲計分、跑跑步圈數累加、或是 APCS 各種演算法迴圈中無所不在，是程式世界裡最強大的基本招式喔！

````python
# [2] Code 範例區：利用右側先計算舊值（score + 20），再存回左側自己更新
score = 50
print(score)  # 原本的分數：50

score = score + 20  # 先算 50 + 20 = 70，再存回 score
print(score)  # 更新後的最新分數：70
````
