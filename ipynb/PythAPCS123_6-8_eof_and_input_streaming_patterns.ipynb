### 🗺️ 6-8-1 APCS / OJ 競技題庫三大輸入格式全景概覽

**生活故事：便利超商櫃檯結帳的三種顧客情境**  
請初學同學想像你在便利超商當店員結帳。  
走進來的第一位顧客，一開口就清楚明瞭地說：「店員你好，我總共要買 3 樣東西（$N = 3$）。」接著依序拿出 3 樣商品刷條碼——這就是**「固定筆數」**；  
走進來的第二位顧客，一件一件慢慢從推車裡拿商品，拿完最後一件後，他主動向你說了一句暗號：「好，就這些，沒有了（暗號 0 結束）！」——這就是**「哨兵值終止」**；

**運作機制——競技題庫的三大輸入版型全貌：**  
在 APCS 檢定與各類線上評判系統（Online Judge）中，題目的輸入資料並不是由出題老師坐在電腦前手動敲鍵盤，而是由評測伺服器將一個個標準文字檔（`.in` 檔）自動灌入你的程式中。  
根據題目規格，所有題目 100% 歸屬於以下三大架構之一：
1. **題型一：首行給定筆數 $N$（Fixed Count）**  
   - 格式特徵：輸入第 1 行是單一整數 $N$，表示接下來會有 $N$ 行資料。  
   - 核心武器：`N = int(input())` 搭配 `for _ in range(N):`。
2. **題型二：哨兵值終止（Sentinel Value）**  
   - 格式特徵：題目沒說有幾行，但規定輸入「以 0 或 -1 作為結尾標記」。  
   - 核心武器：`while True:` 搭配 `if val == 0: break`。
3. **題型三：未知行數直至檔案結尾（End-Of-File, EOF）**  
   - 格式特徵：題目完全沒給行數，也沒有任何結束暗號，資料一直進來，直到測資檔讀完。  
   - 核心武器：`import sys; for line in sys.stdin:` 或 `try ... except EOFError:`。

許多初學者審題不仔細，遇到未知行數 EOF 的題目，卻寫了一個只讀 1 次的程式，導致只算出了第 1 筆答案，後面的數百行資料全部沒讀，慘遭判定「WA（答案錯誤）」或「RE（執行時期錯誤）」！

### 🔢 6-8-2 題型一：首行給定總筆數 N 的標準讀取架構（for _ in range(N):）

**生活故事：快遞公司的包裹點件單**  
想像快遞員送來一個大箱子，箱子外側的提貨單上白紙黑字寫著：「本件內含商品數量：4 件（$N = 4$）」。  
收貨員拿到箱子後，會先拿著清單讀出數字「4」，然後依序伸手進箱子掏出第 1 件、第 2 件、第 3 件、第 4 件商品逐一掃描條碼。  
掏完 4 件後，收貨員立刻簽收蓋章，絕不會再把手伸進空箱子裡亂抓！

**運作機制——首行讀 N，迴圈跑 N 次的黃金二部曲：**  
這是 APCS 與各類檢定中最常見、最規範的標準輸入版型（約佔題目的 50%）。  
標準的解題骨架嚴格由兩步驟構成：
```python
# 第一步：首行讀取整數 N
N = int(input())

# 第二步：迴圈精確跑 N 次
for _ in range(N):
    data = input()
    # 針對每一行資料進行即時處理或運算
```

**為什麼迴圈變數慣例寫成底線 `_`？**  
在 Python 中，單一底線 `_` 是一個合法的變數名稱。  
當我們撰寫 `for _ in range(N):` 時，這是一個通用的工程慣例，明確告訴所有閱讀程式碼的人（以及自己）：**「這個迴圈純粹是為了跑動 $N$ 次，我們在迴圈體內部完全不需要使用到計數器的索引數字！」** 這樣寫代碼極度乾淨俐落！

**常見錯誤陷阱——首行讀取被重複吃掉的世紀慘劇：**  
初學者最常犯的崩潰錯誤：忘記第一行已經執行了 `N = int(input())`，結果在迴圈內部又寫了一次 `val = int(input())`，導致第 1 筆真正的資料被當成了 $N$，後面的資料全部錯位移位，整張試卷慘遭鴨蛋！

**APCS 考試實務建議：**  
若題目格式為「第一行是測資筆數 $T$，每組測資包含一個數列長度 $N$」，這就是經典的**「雙重固定筆數輸入」**！外層 `for _ in range(T):`，內層先讀 $N$，再處理 $N$ 個數值。層次分明，永遠不會錯位！

````python
# [2] Code 範例區：首行給定 N 的標準讀取骨架模擬展示
print("=== 模擬 APCS 題型一：首行 N，後續 N 行數值 ===")

# 模擬輸入流程（以變數模擬循序讀入）
simulated_N = 3
print(f"--> [第 1 步] 讀取首行總筆數 N = {simulated_N}")

total_sum = 0
for round_id in range(1, simulated_N + 1):
    val = round_id * 100  # 模擬逐行讀入 100, 200, 300
    print(f"    第 {round_id} 次讀取資料：{val}")
    total_sum += val

print(f"--> [第 2 步] 處理完畢，N 筆資料總和為：{total_sum}")
````

### 🛑 6-8-3 題型二：哨兵值（Sentinel Value）終止法：以 0 或 -1 結束的 while True: 讀取（免例外，純邏輯！）

**生活故事：拍賣會上的「成交暗號」**  
想像一場激烈的藝術品拍賣會。  
拍賣官大聲喊價：「100 萬有人要嗎？」「150 萬有人要嗎？」「200 萬有人要嗎？」  
拍賣官事先根本不知道這件藝術品會經歷幾次喊價；  
但規則非常明確：當全場無人加價、拍賣官三次敲槌喊出「成交（暗號）！」那一刻，喊價立刻停止，不再接受任何後續出價！  
這個宣告結束的關鍵暗號，在演算法中就叫做**「哨兵值（Sentinel Value）」**！

**運作機制——while True 自由出口架構：**  
哨兵值題型的最大特徵是：**題目事先完全不透露總資料筆數，但保證在資料的最尾端，必定附帶一個特定的「結束專用數值」**（最常見的是 `0`、`-1`，或二維題目的 `0 0`）。  
最標準且無懈可擊的解題模板如下：
```python
while True:
    val = int(input())   # 每一輪先讀取數值
    
    # 靈魂核心：第一時間檢查哨兵暗號！
    if val == 0:
        break            # 哨兵出現，立刻終止跳出！
        
    # 只有正常資料才會走到這裡執行運算
    total += val
```

**三大黃金防禦原則：**
1. **先檢查，後運算**：哨兵檢查 `if val == 0: break` 必須放在迴圈體的最頂端！如果先執行了運算才檢查，哨兵值（0 或 -1）就會被錯誤地算進總和或平均值裡，導致致命失分！
2. **多數值哨兵的元組解包**：若題目規定以 `0 0` 結束，語法為：
   `a, b = map(int, input().split())`  
   `if a == 0 and b == 0: break`
3. **免用例外，純邏輯**：哨兵值終止是純粹的條件判斷，完全不需要任何 try-except，直觀又安全！

**常見錯誤陷阱：**

````python
# ==========================================
# [4] Code 練習題
# 任務說明：
# 哨兵終止與有效平均值計算：
# 連續輸入正整數體重值，直到輸入 0 為止代表測量完畢。
# 規則：哨兵 0 絕不可算進總重，也不可算進總人數！
# 請計算所有有效測量者的平均體重（整除取商 avg_weight = total_weight // count）。
# 若無任何有效輸入（一開始就輸入 0），平均體重為 0。
#
# 【公開測試資料 1】
# 輸入數值序列：60, 70, 80, 0
# 預期：有效人數 count = 3，總重 210，平均體重 avg_weight = 70
#
# 【公開測試資料 2】
# 輸入數值序列：50, 65, 0
# 預期：有效人數 count = 2，總重 115，平均體重 avg_weight = 57
# ==========================================
# 請在下方撰寫你的程式碼並執行測試：

total_weight = 0
count = 0
step = 0

while True:
    step += 1
    if step == 1: w = 60
    elif step == 2: w = 70
    elif step == 3: w = 80
    else: w = 0
    
    if w == 0:
        break
    total_weight += w
    count += 1

avg_weight = total_weight // count if count > 0 else 0
print(f"有效人數：{count}，平均體重：{avg_weight}")

# 公開測資驗證：
assert count == 3 and avg_weight == 70, "公開測資 1 驗證失敗"
print("公開測資 1 通過！")
````

````python
# ==========================================
# [5] Code 挑戰題
# 任務說明：
# 雙哨兵「0 0」終止之座標曼哈頓距離加總：
# 題目要求連續讀入多組座標點 (x, y)，
# 當讀入 x == 0 且 y == 0 時，代表全部測資結束！
# 對於每組有效座標，計算其曼哈頓距離 dist = abs(x) + abs(y)，
# 並累加至 total_manhattan。
# 模擬輸入數值對：(3, 4), (-2, 5), (6, -1), (0, 0), (9, 9)
# 請注意：(0, 0) 之後的 (9, 9) 絕不可被處理！
# （本題為自由挑戰題，無公開測資，請依題目情境自行思考並撰寫完整程式碼）
# ==========================================
# 請在下方撰寫你的程式碼：

total_manhattan = 0
valid_points = 0
step = 0

while True:
    step += 1
    if step == 1: x, y = 3, 4
    elif step == 2: x, y = -2, 5
    elif step == 3: x, y = 6, -1
    elif step == 4: x, y = 0, 0   # 雙哨兵結束！
    else: x, y = 9, 9
    
    # 雙哨兵判定
    if x == 0 and y == 0:
        print("偵測到 (0, 0) 結束標記，退出！")
        break
        
    dist = abs(x) + abs(y)
    total_manhattan += dist
    valid_points += 1

print(f"有效點數：{valid_points} 個，曼哈頓距離總和：{total_manhattan}")
# (3+4) + (2+5) + (6+1) = 7 + 7 + 7 = 21
assert valid_points == 3 and total_manhattan == 21, "挑戰題計算錯誤！"
print("挑戰題 6.8.3 驗證通過！")
````

1. **`try:`**：直譯器嘗試執行裡面的指令。只要檔案裡還有資料，`input()` 就能正常讀入，順利往下運算；
2. **`except EOFError:`**：當檔案見底、再也沒有資料時，`input()` 會噴出 `EOFError` 警報。此時下方的 `except` 會瞬間將警報攔截下來，並執行裡面的 **`break`**，讓程式優雅謝幕、跳出迴圈！

**這兩種 EOF 寫法該選哪一種？**
- **`for line in sys.stdin:`**：代碼最精簡、效能最高，適合整行字串處理。
- **`while True: try: input() except EOFError: break`**：概念最直覺，能無縫套用大家最熟悉的 `input()` 寫法，非常適合初學者在考場上無腦速套！

**常見錯誤陷阱：**

### 🌊 6-8-7 串流數據即時運算（Streaming Processing）：邊讀邊算、空間 O(1) 的解題優化

**生活故事：流水線即時秤重 vs 租巨型倉庫堆滿才秤**

第二種聰明工人：在輸送帶旁放一台電子秤，貨物每滑過來一件，他就看一眼重量、隨手在計數器上按一下累加，然後貨物直接滑上卡車運走！  
聰明工人根本不需要租任何倉庫，他的記憶體消耗永遠只有「眼前的那一件貨物」！

**運作機制——空間複雜度 O(1) 的串流哲學：**  
很多剛學完迴圈的初學者，常常有一種思維定勢：「我一定要先把題目所有的數字全部存進一個巨大的容器裡，之後才能開始算答案」。  
**在競賽程式中，這種想法極度危險！**  
如果評測系統給了 1000 萬筆數據，一口氣存下來會直接引發 **MLE（Memory Limit Exceeded，記憶體超限）**！  
**「串流即時運算（Streaming Processing）」的核心心法是：**
- **來一個，算一個**：每讀入一筆資料，立刻更新目前的累加器（`total += val`）、極值（`if val > max_val: max_val = val`）或計數器；
- **算完即扔**：變數在下一輪直接被新資料覆蓋，舊資料功成身退，完全不佔用任何額外空間！
- **空間複雜度達到極致的 $O(1)$**（常數空間）！無論測資有一百筆還是一億筆，程式永遠只佔用幾十個位元組的記憶體，穩如泰山！

**常見錯誤陷阱：**  
題目明明只問「所有數字的總和與最大值」，初學者卻想著把每筆資料都存下來，結果白白浪費了寶貴的記憶體與時間。

**APCS 考試實務建議：**  
只要題目沒有明確要求「反向輸出」或「多回合隨機查詢」，**90% 的 APCS 統計題型都可以用串流即時運算解決！** 養成邊讀邊算的好習慣，能讓你的程式跑得飛快又省記憶體！

````python
# [2] Code 範例區：串流即時運算示範——邊讀邊算，空間 O(1)
print("=== 串流即時運算：同時維護總和、平均與最大值 ===")

total = 0
count = 0
max_val = -999999

# 模擬連續來了 5 個串流數據
for val in (12, 45, 8, 99, 34):
    # 來一個，算一個！
    total += val
    count += 1
    if val > max_val:
        max_val = val

avg = total // count
print(f"串流處理結算：共處理 {count} 筆，總和 {total}，平均 {avg}，最大值 {max_val}")
print("=== 空間消耗僅維持在 O(1)，記憶體永遠不爆倉！ ===")
````
