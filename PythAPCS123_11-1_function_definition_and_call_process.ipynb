### 11.1.1 為什麼需要函數？從重複代碼到模組化封裝（黑盒子概念）

在編寫程式的過程中，我們經常會遇到需要重複執行相同邏輯的情境。例如在 APCS 競賽題目中，我們可能需要在輸出不同階段的答案時，各自印出一行風格一致的裝飾分隔線；或者在處理多組數據時，每次都要進行相同格式的除錯日誌列印。

如果每次遇到相同需求，都用滑鼠複製三行、五行甚至十行程式碼貼過去，程式碼很快就會變得冗長且混亂。更可怕的是，一旦需求微調（例如要把分隔符號從等號 `=` 改成星號 `*`），我們就必須在整個專案中翻箱倒櫃地逐一修改每一個複製貼上的地方，只要漏掉一個，就會引發格式不一致的嚴重錯誤。軟體工程中有一個著名的黃金定律稱為 **DRY 原則（Don't Repeat Yourself，不要重複你自己）**，意思是相同的邏輯在程式中應該只存在於唯一的源頭。

自訂函數（Function）正是實踐 DRY 原則的終極武器。函數就像是一個封裝好的**「黑盒子（Black Box）」**：我們為一段具有特定功能的程式碼取一個響亮清楚的名字，並把它打包封裝起來。當外部主程式需要使用這項功能時，只需要「呼叫」它的名字即可，完全不需要在主流程中重複展開內部的實作細節。這樣不僅能大幅縮減程式碼行數，讓主程式結構如同目錄般一目了然，更能在未來需要修改時，只需修改函數內部的一處程式碼，全系統所有使用該功能的地方就會同步更新！


````python
# 範例 11.1.1：體驗從複製貼上到函數封裝的轉變
print("=== 傳統重複寫法（代碼重複且難以維護） ===")
print("*" * 30)
print("系統狀態：檢查中...")
print("*" * 30)

print("*" * 30)
print("資料載入：完成！")
print("*" * 30)

print("\n=== 模組化函數封裝寫法（DRY 原則） ===")
# 定義一個專門印出美麗分隔線的黑盒子函數
def print_separator():
    print("=" * 30)

# 需要時只需呼叫名字，簡潔明快
print_separator()
print("系統狀態：檢查中...")
print_separator()

print_separator()
print("資料載入：完成！")
print_separator()
````

### 11.1.2 函數的定義（`def`）、命名規範與呼叫（Call）語法

在 Python 中，宣告一個自訂函數使用的是 `def` 關鍵字（也就是 define「定義」的縮寫）。其標準語法骨架包含四個不可或缺的元素：
1. **關鍵字 `def`**：告知直譯器接下來要宣告一個新的自訂功能。
2. **函數名稱（Function Name）**：必須遵循變數的命名規則（只能由英文字母、數字與底線組成，開頭絕對不能是數字）。在 Python 社群規範（PEP 8）中，函數名稱強烈建議使用**蛇形命名法（snake_case）**，例如 `calc_total_score` 或 `check_boundary`，名稱應盡量具備動詞性質，精準描述該函數所執行的行為。
3. **圓括號 `()`**：緊跟在函數名稱之後，內部用於宣告預備接收的參數（若不需接收任何資料則保持空括號）。
4. **冒號 `:` 與縮排區塊**：冒號標誌著函數本體的開始，所有屬於該函數的內部指令都**必須統一向右縮排（通常為 4 個空格）**。縮排結束之處，即代表函數本體的範疇結束。

在定義好函數後，如何讓直譯器真正去執行它呢？這就稱為**「呼叫（Call / Invoke）」**。呼叫的語法非常嚴格：必須寫出**「函數名稱加上圓括號 `()`」**！初學者最常犯的錯誤是寫了 `my_func` 卻忘了後方的圓括號。如果只寫 `my_func`，Python 只會將其視為對該函數記憶體位址的參照，而**完全不會觸發任何執行動作**。唯有加上圓括號，直譯器才會真正啟動該函數內部所封裝的每一行程式碼。此外，請注意絕對不可使用 Python 既有的關鍵字或內建函數名稱（如 `print`、`list`、`sum`、`max` 等）來為自訂函數命名，否則會覆蓋掉內建功能而引發災難性錯誤！


````python
# 範例 11.1.2：函數命名、定義與正確呼叫機制
# 1. 遵循 snake_case 動詞命名
def display_scoreboard():
    # 縮排 4 空格的函數本體
    scores = [95, 88, 72, 100]
    print(f"比賽選手人數：{len(scores)}")
    print(f"最高分紀錄：{max(scores)} 分")

# 2. 探討：只寫名稱 vs 加上括號呼叫
print("=== 1. 只寫函數名稱（不加括號） ===")
print("display_scoreboard 變數內容：", display_scoreboard)
print("型態：", type(display_scoreboard))
# 注意：此時完全沒有印出選手人數或最高分！

print("\n=== 2. 加上括號呼叫（真正執行） ===")
display_scoreboard()
````

### 11.1.3 程式執行順序：定義期（Def Time）與呼叫期（Call Time）的追蹤

要精通函數，最關鍵的思維躍升就是理解程式碼的**「執行時序（Execution Flow）」**。在學習函數之前，我們的程式碼都是由上至下、逐行循序執行的；但是當引入自訂函數之後，程式的執行路徑會發生暫時性的「跳躍」！

直譯器在執行程式時，會嚴格區分兩個階段：
1. **定義期（Def Time）**：當直譯器由上至下讀到 `def my_func():` 時，它只會「記錄並註冊」這個函數的名字與其對應的程式碼區塊，並將其放入記憶體備用。**直譯器會直接略過函數內部的縮排程式碼，完全不會立即執行它！**
2. **呼叫期（Call Time）**：當主程式執行到下游某一行出現 `my_func()` 時，程式的控制權會瞬間暫停當前進度，**「跳轉」**進 `my_func` 內部的第一行開始執行；直到把函數本體內的所有縮排指令全部執行完畢之後，控制權才會**「跳回」**原本呼叫處的下一行，繼續往下執行主程式。

這個機制帶來了一個至關重要的鐵則：**「先定義，後呼叫」**！如果你在定義 `def my_func():` 之前的那一行就試圖呼叫 `my_func()`，Python 會因為在當前的記憶體符號表內找不到這個名字，而當場拋出 `NameError: name 'my_func' is not defined` 的致命崩潰。清楚理解定義期與呼叫期在時間軸上的先後跳轉，是未來追蹤多層函數嵌套與遞迴呼叫時最根本的底層內功。


````python
# 範例 11.1.3：透過數值標記精準追蹤執行時序
print("【第 1 步】主程式由上至下開始執行")

# 定義期：只登記名稱，內部程式碼被略過不執行
def process_data():
    print("  ---> 【第 3 步】進入 process_data 內部執行指令 A")
    print("  ---> 【第 4 步】進入 process_data 內部執行指令 B")

print("【第 2 步】函數已在記憶體登記完成，準備發動呼叫")

# 呼叫期：控制權跳轉進函數
process_data()

print("【第 5 步】函數執行完畢，控制權返回主流程繼續前進")
````

### 11.1.4 單一參數傳入與引數匹配：讓函數具備動態適應力

如果函數每次被呼叫時，都只能死板地印出完全相同的靜態文字或進行一模一樣的常數計算，那麼它的實用價值將會受到極大限制。為了讓黑盒子具備面對不同輸入情境的「動態適應力」，我們需要為函數裝上「投幣孔」——也就是**參數（Parameters）**。

這裡需要釐清一組標準的術語觀念：
1. **參數（Formal Parameter，形參）**：指的是在定義函數時，寫在 `def 函數名稱(參數名):` 圓括號內的變數名稱。它就像是一個「預留空位（Placeholder）」，宣告這個黑盒子期待收到什麼資料。
2. **引數（Actual Argument，實參）**：指的是在主程式真正呼叫函數時，填入圓括號中的具體實體數值或變數，例如 `my_func(42)` 或 `my_func("Alice")`。

當呼叫發生的瞬間，Python 直譯器會自動完成一道「賦值交接」：把呼叫端傳入的引數數值，自動綁定到函數內部的參數變數名上。在函數內部，我們可以完全像使用一般變數一樣，隨意讀取該參數進行運算、比對或格式化輸出。每一次呼叫時傳入不同的引數，函數內部就會以該引數為基礎產生完全不同的運作結果。這正是演算法中「通用化（Generalization）」的核心靈魂！


````python
# 範例 11.1.4：單一參數傳入讓函數靈活多變
# 定義接收一個整數長度 length 的邊框產生器
def print_horizontal_bar(length):
    # length 是參數，呼叫時會被具體數值替代
    bar = "=" * length
    print(f"+{bar}+")

# 呼叫時傳入不同的引數，產生不同尺寸的邊框
print("長度為 5：")
print_horizontal_bar(5)

print("\n長度為 12：")
print_horizontal_bar(12)

print("\n長度為 20：")
print_horizontal_bar(20)
````

### 11.1.5 多參數位置對應（Positional Arguments）與逗號分隔

現實世界中的運算任務往往需要同時依賴多項資料。例如計算長方形的周長與面積需要「長」與「寬」兩個數值；繪製一條自訂樣式的格線需要「長度」與「符號樣式」；在平面座標系定位一點需要 $x$ 座標與 $y$ 座標。

在 Python 中，自訂函數可以同時定義兩個或更多的參數，各參數之間只要使用**逗號 `,`** 予以分隔即可：
```python
def draw_custom_bar(length, symbol):
    print(symbol * length)
```
當函數擁有多個參數時，呼叫端傳入的引數會嚴格按照**「由左至右」的物理順序**，一對一依序投遞到對應位置的參數中，這稱為**位置引數（Positional Arguments）**。在上面的例子中，若呼叫 `draw_custom_bar(10, "*")`，那麼整數 `10` 會傳入 `length`，字串 `"*"` 則會傳入 `symbol`。

這帶來了兩條嚴格的呼叫守則：
1. **順序不可顛倒**：如果誤寫成 `draw_custom_bar("*", 10)`，那麼 `length` 就會收到字串 `"*"`，而 `symbol` 會收到 `10`，在內部執行乘法時會引發非預期的錯誤或行為！
2. **數量必須精準匹配**：定義了 2 個位置參數，呼叫時就必須剛好給予 2 個引數。如果給得太少（如只給 1 個）或給得太多（給了 3 個），Python 直譯器會立刻拋出 `TypeError: ... takes 2 positional arguments but X were given` 並強制中斷執行。


````python
# 範例 11.1.5：多參數位置引數的一對一依序匹配
def print_student_record(name, student_id, score):
    # 按照順序分別接收：姓名 (str)、學號 (int)、成績 (int)
    print(f"學號：{student_id:04d} | 姓名：{name} | 成績：{score} 分")

print("=== 正確位置呼叫 ===")
print_student_record("王小明", 101, 95)
print_student_record("李大同", 102, 88)

# 觀察：如果參數順序傳錯，結果會錯亂
print("\n=== 順序傳錯時的尷尬情況（學號跟成績顛倒） ===")
print_student_record("張美麗", 98, 103) # 98 變成學號，103 變成成績！
````

### 11.1.6 無回傳值函數的本質：動作型函數與預設回傳的 `None`

在前面的子單元中，我們撰寫的所有函數都有一個共同的特徵：它們的任務是「執行某項動作」（例如在終端印出文字、在螢幕上排版圖形），而沒有向外交出任何計算成果。在軟體工程中，這類函數常被稱為**「動作型函數（Action Functions）」**或**「程序（Procedures）」**。

初學者常常產生一個疑惑：如果一個函數內部沒有寫任何 `return` 語句，那麼它到底「有沒有回傳值」？
答案是：**有的！在 Python 中，所有函數都必定有回傳值。如果函數內部沒有顯式地使用 `return` 交代回傳什麼，直譯器在執行完函數區塊的最後一行時，會自動默默地回傳一個特殊值——`None`！**

`None` 是 Python 內建的一種特殊單例常數，代表「無值」、「空」或「不存在」（其型態為 `NoneType`）。這意味著：
- 如果你嘗試用一個變數去接收一個動作型函數的執行結果，例如 `res = print_horizontal_bar(5)`，這個 `res` 變數並不會得到邊框字串，而是會得到 `None`！
- 如果你寫出 `print(print_horizontal_bar(5))`，終端會先執行函數印出邊框，緊接著會在下一行印出一個令人困惑的 `None`！

理解這一點至關重要：`print()` 只是將字元投射到終端輸出串流給人類肉眼觀看，它**不是**將數值送回給程式邏輯。當你需要讓後續程式碼能夠使用這個運算結果進行加減乘除時，單純靠動作型函數是不夠的，這正是下一小節（11.2）要深入探討的 `return` 核心機制！


````python
# 範例 11.1.6：透視無回傳值函數與 None 物件
def greet_action(name):
    # 此函數只負責執行動作印出文字，內部沒有 return
    print(f"哈囉，{name}！")

# 1. 正常呼叫執行動作
greet_action("小明")

# 2. 嘗試用變數接收其回傳結果
result = greet_action("小華")

print("\n=== 檢驗 result 變數的真實面貌 ===")
print("result 的值為：", result)
print("result 的型態為：", type(result))
print("result 是否為 None：", result is None)

# 3. 常見初學者錯誤排查：巢狀 print
print("\n=== 觀察初學者常見巢狀 print 產生的多餘 None ===")
# 外層 print 會印出內層函數呼叫的回傳值（即 None）
print(greet_action("小美"))
````

1. **模組化封裝（DRY 原則）**：
   - 告別重複代碼，將具備特定功能的指令打包成黑盒子，一處修改、處處同步。
2. **語法四要素**：
   - 掌握 `def 函數名稱(參數):`、冒號與 4 空格縮排區塊。
   - 函數命名遵循蛇形命名法（snake_case），呼叫時**必須加圓括號 `()`** 才會真正執行。
3. **時序跳轉模型**：
   - 嚴格落實**「先定義，後呼叫」**。直譯器在定義期只記錄名稱，直到呼叫期才會暫停主流程跳轉執行。
4. **參數（形參）與引數（實參）**：
   - 參數是函數接收外部資料的接收槽，引數是呼叫時投遞進去的實體數據，透過賦值交接賦予函數強大的動態適應力。
5. **位置引數規則**：
   - 多參數由左至右依序一對一嚴格匹配，個數不可多也不可少。
6. **動作型函數與 `None`**：
   - 了解 `print()` 只負責輸出畫面，沒有寫 `return` 的函數預設回傳物件必定是 `None`。
