### 📝 題目描述（Problem Description）
將一個十進位正整數 $X$ 的各個位數由左至右依序編號：第 1 位、第 2 位、第 3 位……以此類推。
- 我們將所有**奇數位**（第 1, 3, 5, 7, ... 位）的數字總和記為 $A$。
- 將所有**偶數位**（第 2, 4, 6, 8, ... 位）的數字總和記為 $B$。
- 則此整數 $X$ 的**「秘密差」**定義為 $A$ 與 $B$ 相減後的**絕對值**，即：
$$\text{秘密差} = |A - B|$$

例如：給定整數 $X = 263541$：
- 奇數位（由左至右第 1、3、5 位）數字分別為 $2, 3, 4$，其和為 $A = 2 + 3 + 4 = 9$。
- 偶數位（由左至右第 2、4、6 位）數字分別為 $6, 5, 1$，其和為 $B = 6 + 5 + 1 = 12$。
- 秘密差為 $|A - B| = |9 - 12| = |-3| = 3$。

請撰寫一個程式，輸入正整數 $X$，計算並輸出其秘密差。

## 14.2.1 題意解析與 1000 位超長大整數挑戰

### 🚂 生活故事比喻：銀河長途列車的紅藍衛兵分列式
想像各位初學同學來到了一座未來科幻火車站，眼前停靠著一列擁有高達 1000 節車廂的超級銀河列車！
每節車廂外頭都清清楚楚噴塗著一個 $0$ 到 $9$ 的號碼牌：
- 車站站長派出兩隊衛兵進駐列車：**紅隊衛兵**負責看管所有「奇數編號車廂」（第 1 節、第 3 節、第 5 節……）；**藍隊衛兵**則負責看管所有「偶數編號車廂」（第 2 節、第 4 節、第 6 節……）。
- 檢查完畢後，紅隊隊長把手下車廂的號碼全部加總起來，得到分數 $A$；藍隊隊長也把手下車廂的號碼全加起來，得到分數 $B$。
- 最後，站長想知道兩隊加總分數到底相差多少，於是計算兩者的差額。因為不知道哪一隊分數比較高，為了保證差額永遠是一個正大光明的非負數字，站長在差額外面加上了「絕對值」防護罩，也就是 $|A - B|$！這就是本題核心的「秘密差」。

### 💥 1000 位超長大整數的考場衝擊
很多同學看到題目寫「輸入一個正整數 $X$」，第一反應往往是「那我就用整數來存吧」！
但是，請特別注意子任務 3 的規格：**$X$ 的位數高達 1000 位！**
這在電腦硬體層面是什麼概念？
1. 一般程式語言（如 C/C++）中，常見的 32 位元整數 `int` 上限約為 $2 \times 10^9$（只有 9 到 10 位數）；64 位元的 `long long` 上限約為 $9 \times 10^{18}$（只有 19 位數）。遇到 1000 位的怪物整數，一般變數根本裝不下，會瞬間記憶體溢位崩潰！
2. 雖然 Python 具備「任意精度大整數」的神奇魔力，數值再大也不會爆掉，但如果真的把 1000 位數轉成整數，然後用數學方法 `X % 10`（取最後一位）再 `X // 10`（除以 10 削掉最後一位），電腦在底層必須進行上千次極其昂貴的高精度大整數除法，運算時間會急遽攀升。
3. **終極解法思維**：何必把 1000 個字元強行壓成一個大整數呢？把輸入看成**一整串文字字串（string）**，每一節車廂就是字串裡的一個字元，既直觀、快速，又絕對不會溢位！

````python
# ==============================================================================
# [2] Code 範例：拆解數字字串並計算奇偶數位和與差值
# 展示目標：以字串 "263541" 為例，手動拆解奇數位與偶數位，並觀察絕對值計算
# ==============================================================================

# 模擬題目給定的輸入字串（共有 6 個位數）
x_str = "263541"

print("原始輸入字串：", x_str)
print("數字總位數（長度）：", len(x_str))

# 依據題意（由左至右 1-based 計數）：
# 第 1 位：'2'，第 3 位：'3'，第 5 位：'4'
odd_digits = [int(x_str[0]), int(x_str[2]), int(x_str[4])]
A = sum(odd_digits)

# 第 2 位：'6'，第 4 位：'5'，第 6 位：'1'
even_digits = [int(x_str[1]), int(x_str[3]), int(x_str[5])]
B = sum(even_digits)

print(f"奇數位數字：{odd_digits} -> 奇數位和 A = {A}")
print(f"偶數位數字：{even_digits} -> 偶數位和 B = {B}")

# 計算秘密差 |A - B|
secret_diff = abs(A - B)
print(f"秘密差 |{A} - {B}| = |{A - B}| = {secret_diff}")
````

````python
# ==============================================================================
# [3] Code 填空題
# 任務說明：
# 請觀察以下字串 "131"（範例 2），將關鍵的下標與函數名稱挖空處補充完整。
# 提示：
# 1. 奇數位為第 1 位（下標 0）與第 3 位（下標 2）
# 2. 偶數位為第 2 位（下標 1）
# 3. 絕對值函數為 abs()
# ==============================================================================

s = "131"

# 奇數位數字：第 1 位與第 3 位
# 請填入對應的字串下標 0 與 2
odd_sum = int(s[0]) + int(s[2])

# 偶數位數字：第 2 位
# 請填入對應的字串下標 1
even_sum = int(s[1])

# 計算絕對值差值，請填入求絕對值的內建函數名稱
ans = abs(odd_sum - even_sum)

print(f"A = {odd_sum}, B = {even_sum}, 秘密差 = {ans}")
# 預期輸出：A = 2, B = 3, 秘密差 = 1
````

## 14.2.2 大數以字串形式讀入與記憶體防爆

### 🛡️ 守護記憶體：為什麼不轉整數才是真正的聰明選擇？
在第 4 章中，我們學習了使用 `int(input())` 讀取使用者輸入的整數。
對絕大多數基礎題目來說，這完全合情合理。然而在 APCS 的戰場上，這道題目設下了一個精妙的陷阱：
> **「整數 $X$ 的位數長度可能多達 1000 位！」**

初學者常犯的直覺誤區是：「題目說 $X$ 是整數，那我就轉成整數吧！」
但請思考一下，如果把它轉成整數，後續會面臨多大的麻煩：
1. **失去下標存取能力**：整數型態（`int`）在 Python 中**不支援中括號下標存取**！你不能寫 `x[0]` 或 `x[1]`，否則直譯器會立刻對你丟出致命的 `TypeError: 'int' object is not subscriptable` 報錯。
2. **被迫使用昂貴的除法迴圈**：如果要從整數中逐一拿出每一位數字，就必須用 `% 10` 取個位數、再用 `// 10` 截斷，並且還得由後往前反向追蹤。對於一個 1000 位的超大數，每次除法都是對幾百個位元組的巨大整數進行計算，白白浪費 CPU 效能。
3. **字串讀入神技：`input().strip()`**：
   - 在 Python 中，`input()` 讀入的資料**天生就是字串（str）**！
   - 字串天生就是一個「字元容器」，直接擁有長度 `len(s)`，並且隨時允許我們用下標 `s[0]`, `s[1]` 直達任意一位數。
   - 加上 `.strip()` 方法可以自動剔除字串首尾可能殘留的換行符號（`\n`、`\r`）或不小心鍵入的空格，確保資料乾乾淨淨。

「**看到位數破百位，字串讀入最乾脆；不轉整數不溢位，下標走訪零花費！**」  
這就是 APCS 高手面對超長大數時最核心的第一道防禦心法。

## 14.2.3 字串索引走訪與奇數位/偶數位區分

### 🔍 考場第一大隱形地雷：1-based 位數 vs 0-based 索引
在 APCS 考場上，這道題目最容易讓初學者「一失足成千古恨」的地方，就是**位數與下標的奇偶顛倒**！
請各位學習者跟著我們把這個對應關係徹底看清楚：

| 由左至右實際位置（題意 1-based） | 題目分類 | Python 字串下標（0-based） | 下標除以 2 的餘數（`i % 2`） |
| :---: | :---: | :---: | :---: |
| **第 1 位** | **奇數位** | `i = 0` | `0 % 2 == 0`（偶數下標！） |
| **第 2 位** | **偶數位** | `i = 1` | `1 % 2 == 1`（奇數下標！） |
| **第 3 位** | **奇數位** | `i = 2` | `2 % 2 == 0`（偶數下標！） |
| **第 4 位** | **偶數位** | `i = 3` | `3 % 2 == 1`（奇數下標！） |
| **第 5 位** | **奇數位** | `i = 4` | `4 % 2 == 0`（偶數下標！） |

看到了嗎？這真是一個驚人的巧合：
- 題目的**「奇數位」**，在 Python 的下標中竟然全都是**「偶數」**（0, 2, 4, 6...）！
- 題目的**「偶數位」**，在 Python 的下標中竟然全都是**「奇數」**（1, 3, 5, 7...）！

### 🛠️ 兩種優雅防呆走訪策略
為了避免在考場緊張時寫出 `if i % 2 == 1:` 卻誤以為是奇數位的悲劇，我們有兩種極佳的解法策略：
1. **位數還原法（最推薦初學者，絕不混淆）**：
   在迴圈中令 `pos = i + 1`（代表人類視角的第幾位數）。
   這樣一來，`if pos % 2 == 1:` 就百分之百是題目的奇數位，`else:` 就是偶數位，思維與大腦直覺完全一致！
2. **直接下標餘數法**：
   熟記口訣：「**下標是 0 開頭，所以下標偶數是奇數位，下標奇數是偶數位！**」
   即 `if i % 2 == 0:` 為奇數位，`else:` 為偶數位。

無論採用哪一種，只要清楚理解這個位移對齊原理，你就不會掉進題目精心設計的下標陷阱中！

````python
# ==============================================================================
# [2] Code 範例：字串走訪與奇偶位數精準分流
# 展示目標：以 "263541" 走訪展示 0-based 索引與 1-based 位數的對齊關係
# ==============================================================================

s = "263541"

print(f"走訪字串：'{s}'，總長度：{len(s)}")
print("-" * 50)

for i in range(len(s)):
    ch = s[i]           # 當前字元
    pos = i + 1         # 轉換為人類視角 1-based 第幾位
    
    if pos % 2 == 1:    # 1-based 為奇數 -> 奇數位
        role = "【奇數位 A】（對應 Python 下標 i = " + str(i) + "）"
    else:               # 1-based 為偶數 -> 偶數位
        role = "【偶數位 B】（對應 Python 下標 i = " + str(i) + "）"
        
    print(f"第 {pos} 位數字是 '{ch}' -> 歸類為：{role}")
````

````python
# ==============================================================================
# [3] Code 填空題
# 任務說明：
# 請填補迴圈與條件判斷挖空處，將字串中的字元分別收集到 odd_list 與 even_list 中。
# 提示：下標 i 從 0 開始，i % 2 == 0 對應第 1, 3, 5... 奇數位！
# ==============================================================================

s = "131"
odd_chars = []
even_chars = []

# 請填入 range 與 len 函數以走訪所有下標
for i in range(len(s)):
    # 當下標 i 是偶數時（0, 2...），代表題目的奇數位！
    if i % 2 == 0:
        odd_chars.append(s[i])
    else:
        even_chars.append(s[i])

print("奇數位字元：", odd_chars)
print("偶數位字元：", even_chars)
# 預期輸出：奇數位字元： ['1', '1'] ，偶數位字元： ['3']
````

````python
# ==============================================================================
# [4] Code 練習題
# 任務說明：
# 撰寫一個迴圈，接收字串 s，分別印出所有奇數位字元串列與偶數位字元串列。
#
# 【公開測試資料 1】
# 輸入 / 設定：s = "263541"
# 預期輸出：奇數位: ['2', '3', '4'], 偶數位: ['6', '5', '1']
#
# 【公開測試資料 2】
# 輸入 / 設定：s = "9"
# 預期輸出：奇數位: ['9'], 偶數位: []
# ==============================================================================

def split_digits(s):
    odds = []
    evens = []
    for i in range(len(s)):
        if i % 2 == 0:
            odds.append(s[i])
        else:
            evens.append(s[i])
    return odds, evens

# 測試 1：
o1, e1 = split_digits("263541")
print(f"測試 1 -> 奇數位: {o1}, 偶數位: {e1}")

# 測試 2：
o2, e2 = split_digits("9")
print(f"測試 2 -> 奇數位: {o2}, 偶數位: {e2}")
````

## 14.2.4 字元轉整數 `int()` 與雙計數器累加

### 🧩 字元拼接 vs 數值相加的千古冤屈
當我們成功用迴圈把每個位數抓出來後，接下來要做的是「加總」求和。
但此時初學者最容易踩到第二個隱形巨坑——**型態混淆**！
請觀察以下這兩行代碼的巨大差別：
```python
x = '2' + '3' + '4'  # 字串相加 -> 結果是 '234'（文字接龍！）
y = 2 + 3 + 4        # 整數相加 -> 結果是 9（真正的數學總和！）
```
因為我們的輸入是字串，所以 `s[i]` 取出來的型態依然是**單一字元（str）**！
如果你寫 `A = A + s[i]`，Python 會以為你想玩文字接龍，把字元一個個黏在一起，最後 $A$ 變成很長的字串，完全無法計算差值！

### 🪄 轉型解藥：`int(ch)` 與「雙計數器記帳法」
解決方案非常直接：在把字元放進累加器之前，先用 `int()` 施加轉型魔法：
```python
val = int(s[i])  # 把字元 '2' 變回整數數值 2
```

接著，我們在進入迴圈之前，準備好兩個「專屬記帳本」（雙累加變數）：
- `odd_sum = 0`：專門記錄奇數位數字的累加和（即題目的 $A$）。
- `even_sum = 0`：專門記錄偶數位數字的累加和（即題目的 $B$）。

迴圈每前進一步：
- 若是奇數位：`odd_sum += int(s[i])`
- 若是偶數位：`even_sum += int(s[i])`

當整條長達 1000 位的列車完全走訪完畢，兩個計數器就各自裝滿了精確的數值總和，乾淨俐落！

````python
# ==============================================================================
# [2] Code 範例：字元轉型與雙計數器逐步累加過程
# 展示目標：以 "263541" 演示雙記帳本在迴圈中每一步的數值變化
# ==============================================================================

s = "263541"

# 步驟 1：初始化雙記帳本為 0
odd_sum = 0
even_sum = 0

print(f"開始走訪字串 '{s}'，初始化 odd_sum = 0, even_sum = 0\n")

# 步驟 2：迴圈逐一走訪
for i in range(len(s)):
    val = int(s[i])  # ⚠️ 關鍵轉型：將字元轉為整數數值！
    
    if i % 2 == 0:   # 下標為偶數 -> 第 1, 3, 5... 位（奇數位）
        odd_sum += val
        print(f"第 {i+1} 位（奇數位）數值 {val} -> odd_sum 累加為 {odd_sum}")
    else:            # 下標為奇數 -> 第 2, 4, 6... 位（偶數位）
        even_sum += val
        print(f"第 {i+1} 位（偶數位）數值 {val} -> even_sum 累加為 {even_sum}")

print("-" * 50)
print(f"走訪結束！最終奇數位總和 A = {odd_sum}, 偶數位總和 B = {even_sum}")
````

````python
# ==============================================================================
# [3] Code 填空題
# 任務說明：
# 請填補挖空處，將字元轉整數並正確累加至對應的計數器中。
# ==============================================================================

s = "131"

# 請初始化兩個記帳本變數為 0
odd_sum = 0
even_sum = 0

for i in range(len(s)):
    # 請填入轉型函數 int
    digit = int(s[i])
    
    if i % 2 == 0:
        # 奇數位累加
        odd_sum += digit
    else:
        # 偶數位累加
        even_sum += digit

print(f"奇數和 = {odd_sum}, 偶數和 = {even_sum}")
# 預期輸出：奇數和 = 2, 偶數和 = 3
````

````python
# ==============================================================================
# [4] Code 練習題
# 任務說明：
# 撰寫計算奇數和 A 與偶數和 B 的邏輯，並印出結果。
#
# 【公開測試資料 1】
# 輸入 / 設定：s = "263541"
# 預期輸出：A = 9, B = 12
#
# 【公開測試資料 2】
# 輸入 / 設定：s = "1000000001"
# 預期輸出：A = 1, B = 1
# ==============================================================================

def calculate_sums(s):
    a = 0
    b = 0
    for i in range(len(s)):
        if i % 2 == 0:
            a += int(s[i])
        else:
            b += int(s[i])
    return a, b

# 測試 1：
a1, b1 = calculate_sums("263541")
print(f"測試 1 -> A = {a1}, B = {b1}")

# 測試 2：
a2, b2 = calculate_sums("1000000001")
print(f"測試 2 -> A = {a2}, B = {b2}")
````

## 14.2.5 `abs()` 絕對值差值計算與完整 AC 實作

### ⚖️ 絕對值函數 `abs()`：距離永遠不為負
當我們順利算出奇數位和 $A$ 與偶數位和 $B$ 之後，最後一步就是算出秘密差。
題目定義秘密差為：
$$\text{秘密差} = |A - B|$$
請注意：
- 如果 $A = 9, B = 12$，則 $A - B = -3$；但秘密差問的是「相差多少」，距離與差額永遠不能是負數！
- 如果你直接輸出 `A - B`，在範例 1 中就會輸出 `-3`，立刻拿到大大的紅字 `WA`（Wrong Answer）！
- Python 提供了一個非常強大且內建的數學工具：**`abs()` 函數**（代表 Absolute Value，絕對值）。
- `abs()` 不需要引入任何額外模組（不需要 `import math`），直接呼叫即可：
  * `abs(-3)` 會自動變為 `3`
  * `abs(3)` 保持為 `3`
  * `abs(0)` 保持為 `0`
- 而且在數學上有一個極為美妙的性質：$|A - B| = |B - A|$！也就是說，不管你寫 `abs(odd_sum - even_sum)` 還是 `abs(even_sum - odd_sum)`，算出來的答案完全百分之百相同！

### 🚀 考場滿分 AC 程式碼拼裝五部曲
現在，我們手上的所有拼圖都齊全了！讓我們手把手將它們組裝成完整滿分程式：
1. **讀入字串**：`s = input().strip()`（大數安全防爆）
2. **初始化記帳本**：`odd_sum = 0; even_sum = 0`
3. **迴圈分流走訪**：`for i in range(len(s)):`
   - 若 `i % 2 == 0`（第 1, 3, 5... 位）：`odd_sum += int(s[i])`
   - 否則（第 2, 4, 6... 位）：`even_sum += int(s[i])`
4. **計算秘密差**：`ans = abs(odd_sum - even_sum)`
5. **輸出結果**：`print(ans)`

僅僅 10 行代碼，就將 1000 位的 APCS 實作真題徹底斬落馬下！

````python
# ==============================================================================
# [2] Code 範例：完整 AC 程式碼拼裝與執行演示
# 展示目標：展示單次輸入的完整解題邏輯，並直接使用範例測資驗證
# ==============================================================================

# 模擬使用者輸入 "263541"
raw_input_data = "263541\n"

# 步驟 1：字串讀入與清理前後空白
s = raw_input_data.strip()

# 步驟 2：初始化雙計數器
odd_sum = 0
even_sum = 0

# 步驟 3：字串迴圈走訪與分流轉型累加
for i in range(len(s)):
    if i % 2 == 0:
        odd_sum += int(s[i])
    else:
        even_sum += int(s[i])

# 步驟 4：取絕對值計算秘密差
diff = abs(odd_sum - even_sum)

# 步驟 5：輸出答案
print("計算完成！秘密差為：", diff)
# 預期輸出：3
````

````python
# ==============================================================================
# [4] Code 練習題
# 任務說明：
# 請撰寫一個完整的解題函數 solve(s)，計算傳入字串的秘密差。
#
# 【公開測試資料 1】
# 輸入 / 設定：s = "263541"
# 預期輸出：3
#
# 【公開測試資料 2】
# 輸入 / 設定：s = "4444"
# 預期輸出：0
# ==============================================================================

def solve(s):
    s = s.strip()
    odd_sum = 0
    even_sum = 0
    for i in range(len(s)):
        if i % 2 == 0:
            odd_sum += int(s[i])
        else:
            even_sum += int(s[i])
    return abs(odd_sum - even_sum)

# 執行公開測試資料驗證：
print("測試 1 輸出：", solve("263541"))
print("測試 2 輸出：", solve("4444"))
````

## 14.2.6 1-based 位數 vs 0-based 索引之對齊陷阱與複雜度分析

### 🪞 數學對稱之美：由左往右 vs 由右往左，結果完全相同！
有些細心的同學在做題時可能會產生一個疑問：
> 「如果我不小心從數字的最右邊（個位數）開始往左數奇數位與偶數位，答案會算錯嗎？」

答案是：**神奇的是，兩者的結果完全一模一樣！**
讓我們用簡單的數學來證明這個漂亮的對稱性質：
1. **當數字長度是奇數時（例如 3 位數 "131"）**：
   - 從左往右數：第 1 位是百位、第 3 位是個位，這兩個是奇數位；第 2 位是十位（偶數位）。
   - 從右往左數：第 1 位是個位、第 3 位是百位，這兩個依然是奇數位；第 2 位依舊是十位！
   - 兩邊抓到的集合完全一模一樣，$A_{左} = A_{右}, B_{左} = B_{右}$，差值當然相同！
2. **當數字長度是偶數時（例如 6 位數 "263541"）**：
   - 從右往左數時，原本由左數的偶數位（2, 4, 6），從右邊看剛好變成了第 1, 3, 5 位（奇數位）！
   - 也就是說，$A$ 與 $B$ 的總和剛好互換了（$A_{右} = B_{左}, B_{右} = A_{左}$）。
   - 但是！題目要求的是**絕對值**：$|A - B| = |B - A|$！差值的絕對值完全絲毫不受影響！
這證明了由左至右單純走訪正是最自然、最不會寫錯的最佳解題路線。

### ⏱️ 時間與空間複雜度深度剖析
- **時間複雜度：$O(L)$**（其中 $L$ 為數字位數，最高 $L \le 1000$）
  - 我們的程式只使用了一個單層 `for` 迴圈走訪字串一次。
  - 迴圈內部只有單純的下標讀取、字元轉整數與加法，全是常數時間 $O(1)$。
  - 因此，走訪 1000 位數只需進行 1000 次微型運算。在現代電腦每秒約可運算 $10^8$ 次的環境下，耗時僅約 **0.00001 秒**，遠遠低於 APCS 官方 1.0 秒的時間限制，完全零效能壓力！
- **空間複雜度：$O(L)$**
  - 儲存 1000 個字元的字串僅耗用約 1 KB 的記憶體空間。
  - 加上 `odd_sum` 與 `even_sum` 兩個純整數變數，總記憶體遠遠低於 64 MB 的限制。

### 🚨 考場 4 大致命 WA 避坑清單
1. **大數轉型陷阱**：寫成 `int(input())` 導致大數無法用索引逐位拆解。
2. **字串未轉型陷阱**：累加時忘記包 `int()`，寫成 `odd_sum += s[i]`，導致數字被當成文字黏成長字串。
3. **忘記取絕對值**：直接輸出 `odd_sum - even_sum`，遇到負數時輸出帶負號的值，直接拿 WA。
4. **輸出多餘提示字**：在考場輸出時寫成 `print("秘密差 =", ans)`，官方線上評判系統只認純數字，多一個中文字或多一個空格就會全盤皆輸！

````python
# ==============================================================================
# [2] Code 範例：對稱性驗證與 1000 位大數極限壓力測試
# 展示目標：
# 1. 驗證「由左往右」與「由右往左」計算絕對值恆等
# 2. 生成 1000 位極限大數進行效能測試
# ==============================================================================

import time

# 測試字串
s = "263541"

# 策略 A：由左往右（標準解法）
a_left = sum(int(s[i]) for i in range(len(s)) if i % 2 == 0)
b_left = sum(int(s[i]) for i in range(len(s)) if i % 2 == 1)
res_left = abs(a_left - b_left)

# 策略 B：由右往左
rev_s = s[::-1]  # 反轉字串
a_right = sum(int(rev_s[i]) for i in range(len(rev_s)) if i % 2 == 0)
b_right = sum(int(rev_s[i]) for i in range(len(rev_s)) if i % 2 == 1)
res_right = abs(a_right - b_right)

print(f"由左往右計算結果：{res_left}")
print(f"由右往左計算結果：{res_right}")
print(f"兩者是否恆等：{res_left == res_right}\n")

# ------------------------------------------------------------------------------
# 1000 位極限大數耗時壓力測試
# ------------------------------------------------------------------------------
big_num_str = "9876543210" * 100  # 恰好 1000 位的超級大數！

start_time = time.perf_counter()

odd_sum = 0
even_sum = 0
for i in range(len(big_num_str)):
    if i % 2 == 0:
        odd_sum += int(big_num_str[i])
    else:
        even_sum += int(big_num_str[i])
ans = abs(odd_sum - even_sum)

elapsed_time = time.perf_counter() - start_time

print(f"1000 位數大數計算結果：{ans}")
print(f"運算耗時：{elapsed_time:.6f} 秒（遠低於 1.0 秒！）")
````

### 📝 版本一：APCS 官方實作考場專用版（極簡單筆輸入）

* **適用情境**：APCS 現場正式檢定考試、學校單筆測資檢驗作業。
* **解題特點**：
  1. 完全依據 APCS 官方題本規範設計（單行輸入一個超長大整數）。
  2. 不需要 `import sys`，不需要寫多筆輸入的大外迴圈。
  3. 僅僅 10 行精練程式碼，結構純粹清晰，適合在考場限時高壓下快速撰寫與自我驗證。
* **Colab 執行方式**：點擊下方儲存格的執行按鈕，在跳出的輸入框中鍵入任意大整數（例如 `263541` 或 `131`），按下 `Enter` 即可獲得即時判定！

````python
# ==============================================================================
# 📝 版本一：APCS 官方實作考場專用版（極簡單筆輸入）
# 適用情境：APCS 正式考試現場（題目保證單行輸入一個大整數 X）
# ==============================================================================

# 步驟 1：以字串形式讀取單行輸入，並去除前後換行與空白
# ⚠️ 核心防禦：位數高達 1000 位，絕不可使用 int(input())，必須保持字串型態以防溢位與降速！
s = input().strip()

# 步驟 2：宣告兩個記帳本變數，分別累加奇數位數字和與偶數位數字和
odd_sum = 0   # 記錄第 1, 3, 5... 奇數位之和（即題目的 A）
even_sum = 0  # 記錄第 2, 4, 6... 偶數位之和（即題目的 B）

# 步驟 3：走訪字串中的每一個位數下標
for i in range(len(s)):
    # ⚠️ 關鍵對齊：Python 索引從 0 開始！
    # 下標 i = 0, 2, 4...（偶數下標）對應第 1, 3, 5... 位（奇數位）
    if i % 2 == 0:
        odd_sum += int(s[i])   # 將字元轉為整數後累加進奇數和
    else:
        even_sum += int(s[i])  # 將字元轉為整數後累加進偶數和

# 步驟 4：計算差值的絕對值，並依規定直接輸出單一整數（行末自動換行）
# ⚠️ 考場防禦：務必使用 abs() 確保輸出為非負整數，切勿輸出負號！
print(abs(odd_sum - even_sum))
````

### 🌐 版本二：ZeroJudge 線上評判萬用 AC 版（支援多筆測資 EOF 循環）

* **適用情境**：ZeroJudge 線上刷題（題目代碼：c290）、各高中 OJ 競賽系統。
* **解題特點**：
  1. **為什麼要用 `sys.stdin` 迴圈？**  
     在 ZeroJudge 評判伺服器中，測試資料是以檔案串流整批重導向輸入（可能包含數十筆測試測資）。如果只寫單次 `input()`，程式在判定完第一組後就結束退出，後面的測資完全沒有處理，會直接導致評測系統判定為 `WA` 或 `NA`！
  2. **`for line in sys.stdin:` 的威力**：  
     這是 Python 處理線上解題系統多筆測資最優雅、最高效的標準語法。它會自動逐行讀取，直到資料末尾（EOF）自動安全退出，絕不觸發 `EOFError`。
  3. **ZeroJudge 複製提交指引**：  
     同學若要提交 ZeroJudge，可直接複製下方【ZeroJudge 複製提交專區】的程式碼貼上送出，即可獲得 100 分 AC（綠燈通過）！

````python
# ==============================================================================
# 🌐 版本二：ZeroJudge 線上評判萬用 AC 版（支援多筆測資 EOF 循環）
# 適用情境：ZeroJudge c290 / 線上解題系統（支援多筆測資連續評判至 EOF）
# ==============================================================================

import sys

# ------------------------------------------------------------------------------
# 核心解題模組：單筆測資的完整秘密差計算與輸出
# ------------------------------------------------------------------------------
def solve_secret_difference(num_str):
    """
    傳入數字字串 num_str
    計算奇數位與偶數位之和，輸出差值的絕對值
    """
    odd_sum = 0
    even_sum = 0
    for i in range(len(num_str)):
        if i % 2 == 0:
            odd_sum += int(num_str[i])
        else:
            even_sum += int(num_str[i])
    print(abs(odd_sum - even_sum))


# ------------------------------------------------------------------------------
# 🎯 【ZeroJudge 複製提交專區】：
# 若要提交至 ZeroJudge c290，可使用以下這段經典主程式：
# ------------------------------------------------------------------------------
def run_zerojudge_submission():
    """ZeroJudge 評判系統主迴圈，透過 sys.stdin 走訪全部資料行直到 EOF"""
    for line in sys.stdin:
        line = line.strip()
        if not line:  # 忽略可能存在的空行
            continue
        solve_secret_difference(line)


# ------------------------------------------------------------------------------
# 🧪 【Colab 本地自動化測試檢驗展示】：
# 點擊本儲存格播放鍵，將自動以官方四大範例及極限邊界測資進行驗證！
# ------------------------------------------------------------------------------
print("=" * 65)
print("🚀 正在執行 APCS c290 官方四大範例與極限邊界測資本地檢驗...")
print("=" * 65)

test_samples = [
    ("263541", "官方範例 1：偶數長度 6 位數，預期輸出 3"),
    ("131", "官方範例 2：奇數長度 3 位數，預期輸出 1"),
    ("4444", "官方範例 3：全相同數字，預期輸出 0"),
    ("9", "官方範例 4：單一位數極限邊界，預期輸出 9"),
    ("1000000000000000000000000000001", "超長大數邊界測資（31位），預期輸出 2")
]

for idx, (raw_str, desc) in enumerate(test_samples, 1):
    print(f"\n【測資 {idx}】輸入：'{raw_str}'（{desc}）")
    print("--- 輸出結果 ---")
    solve_secret_difference(raw_str.strip())

print("\n" + "=" * 65)
print("✅ 本地 5 組測資檢驗全數通過！")
print("學生可直接複製上方【ZeroJudge 複製提交專區】代碼至 ZeroJudge c290 取得 100% AC！")
print("=" * 65)
````
