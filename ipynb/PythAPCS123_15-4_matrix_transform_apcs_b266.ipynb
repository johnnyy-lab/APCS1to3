### 📝 題目描述（Problem Description）
給定一個大小為 $R \times C$ 的數位矩陣，我們可以對它進行兩種操作：
1. **旋轉（代碼 0）**：將矩陣**順時針方向旋轉 90 度**。若原本矩陣大小為 $R \times C$，旋轉後大小會變為 $C \times R$。
2. **翻轉（代碼 1）**：將矩陣進行**上下翻轉**（第一列與最後一列對調，第二列與倒數第二列對調，依此類推）。矩陣大小維持不變。

現在題目給予一個原本的未知矩陣經過 $M$ 個操作之後的**「結果矩陣」**，以及這 $M$ 個操作的**「操作順序清單」**。
請撰寫一個程式，還原出**最初始的原始矩陣**！

---

### 📥 輸入格式（Input Format）
第一行包含三個正整數 $R, C, M$（$1 \le R, C, M \le 10$），分別代表結果矩陣的列數、行數，以及操作的總次數。  
接下來的 $R$ 行，每行包含 $C$ 個整數，代表經過 $M$ 個操作後的**結果矩陣**內容。  
最後一行包含 $M$ 個整數 $m_1, m_2, \dots, m_M$，代表當初**正向施加的各個操作代碼**（0 代表旋轉，1 代表翻轉）。

---

### 📤 輸出格式（Output Format）
第一行請輸出兩個整數，代表還原後**原始矩陣的列數與行數**，以一個空白隔開。  
接下來請依序輸出原始矩陣的每一列，同一列的數字之間以一個空白隔開，每列結尾不得有多餘空白。

---

### ⚠️ 測資範圍與評分規範（Constraints）
* $1 \le R, C, M \le 10$
* 矩陣中的整數值介於 $0$ 到 $100$ 之間

### 15.4.1 題意解析與時空逆推心智模型：從「最終結果」回溯「原始起點」

#### ⏳ 為什麼必須倒序逆推？
想像一下穿脫鞋襪的過程：
1. 正向流程：先穿襪子（操作 A） $\to$ 再穿鞋子（操作 B）。
2. 逆向流程：若要還原成最初光腳的狀態，你必須**先脫鞋子（逆操作 B） $\to$ 再脫襪子（逆操作 A）**！

同理，若原本矩陣依序經歷了 $[op_1, op_2, \dots, op_M]$，得到了當前矩陣 $B$；
要推回最初的 $A$，我們必須以 $B$ 為出發點，**倒過來依序執行 $[op_M^{-1}, \dots, op_2^{-1}, op_1^{-1}]$**！


````python
# =====================================================================
# [2] Code 範例：展示操作序列倒序遍歷
# =====================================================================

ops = [1, 0, 0]  # 依序代表：翻轉 -> 順時針 -> 順時針

print("正向操作歷史：", ops)
print("逆向還原應依序執行的操作：")
for step, op in enumerate(reversed(ops), 1):
    op_name = "逆時針旋轉 90 度 (0 的逆操作)" if op == 0 else "上下翻轉 (1 的逆操作)"
    print(f"第 {step} 步：還原指令代碼 {op} ➔ 執行【{op_name}】")
````

### 15.4.2 操作指令倒序讀取與回溯迴圈架構（`ops[::-1]` 或 `reversed()`）

````python
# =====================================================================
# [4] Code 練習題：倒序計數特定指令
# 任務說明：計算在倒序執行過程中，翻轉 (1) 與旋轉 (0) 各出現幾次
# =====================================================================

def count_op_types(ops):
    cnt_0 = 0
    cnt_1 = 0
    for op in reversed(ops):
        if op == 0: cnt_0 += 1
        elif op == 1: cnt_1 += 1
    return cnt_0, cnt_1

assert count_op_types([1, 0, 0]) == (2, 1)
print("✅ 練習題所有 assert 測試通過！")
````

### 15.4.3 變換操作一：垂直翻轉（上下對調 `grid[::-1]`）之自逆性特徵

#### 🔄 翻轉的自逆性（Self-Inverse）
上下翻轉是指將矩陣沿著水平中線做鏡射對折：
* 第 $0$ 列 $\leftrightarrow$ 第 $R-1$ 列
* 第 $1$ 列 $\leftrightarrow$ 第 $R-2$ 列

關鍵代數性質：**翻轉的逆操作依然是翻轉本身（$Flip^{-1} = Flip$）**！
在 Python 中最精簡且高效的寫法為外層串列反轉：`grid[::-1]`。


````python
# =====================================================================
# [2] Code 範例：矩陣上下翻轉實作展示
# =====================================================================

mat = [
    [1, 1],
    [3, 1],
    [1, 2]
]

print("翻轉前：")
for row in mat: print(row)

flipped = mat[::-1]
print("\n翻轉後：")
for row in flipped: print(row)
````

````python
# =====================================================================
# [4] Code 練習題：手動雙指標翻轉列
# 任務說明：不使用切片，手動用 swap 完成翻轉
# =====================================================================

def manual_flip(matrix):
    m = [row[:] for row in matrix]
    r = len(m)
    for i in range(r // 2):
        m[i], m[r - 1 - i] = m[r - 1 - i], m[i]
    return m

assert manual_flip([[1, 2], [3, 4], [5, 6]]) == [[5, 6], [3, 4], [1, 2]]
print("✅ 練習題所有 assert 測試通過！")
````

### 15.4.4 變換操作二：順時針旋轉之逆操作——逆時針旋轉 90 度公式推導

#### 🔄 逆向旋轉空間幾何推導
若正向操作是「順時針 90 度」，則要還原時必須執行**「逆時針 90 度（Counter-Clockwise 90°）」**！
假設目前矩陣形狀為 $R \times C$：
* 旋轉後新矩陣形狀必為 **$C \times R$**。
* 原矩陣第 $j$ 行（由右向左，從 $C-1$ 遞減至 $0$），提取後將成為新矩陣的第 $C-1-j$ 列！

```python
new_matrix = []
for j in range(c - 1, -1, -1):  # 由右向左提取每一行
    new_matrix.append([mat[i][j] for i in range(r)])
```


````python
# =====================================================================
# [2] Code 範例：逆時針旋轉 90 度幾何變換展示
# =====================================================================

mat = [
    [1, 2],
    [2, 1],
    [3, 3]
]
r, c = len(mat), len(mat[0])

rot_ccw = []
for j in range(c - 1, -1, -1):
    rot_ccw.append([mat[i][j] for i in range(r)])

print(f"原矩陣維度：{r} x {c} ➔ 旋轉後維度：{len(rot_ccw)} x {len(rot_ccw[0])}")
print("旋轉後內容：")
for r in rot_ccw: print(r)
````

````python
# =====================================================================
# [4] Code 練習題：封裝逆時針旋轉函式
# =====================================================================

def rotate_counter_clockwise(m):
    cur_r, cur_c = len(m), len(m[0])
    return [[m[i][j] for i in range(cur_r)] for j in range(cur_c - 1, -1, -1)]

assert rotate_counter_clockwise([[1, 2], [3, 4]]) == [[2, 4], [1, 3]]
print("✅ 練習題所有 assert 測試通過！")
````

### 15.4.5 非方陣維度突變陷阱：旋轉後的行列長寬動態交換（$R \leftrightarrow C$）與新矩陣建構

#### ⚠️ 考場常見維度混淆地雷：
許多同學在解題時，習慣使用最初讀入的固定變數 `r, c` 來操作後續的迴圈。
然而只要進行了一次旋轉，矩陣的列數與行數就會**互換（$R \to C, C \to R$）**！
若操作多次，矩陣尺寸會反覆動態交替。因此在建構新矩陣或輸出時，必須始終以 `len(mat)` 與 `len(mat[0])` 動態取得最新的當前維度！


````python
# =====================================================================
# [2] Code 範例：動態維度追蹤展示
# =====================================================================

cur_mat = [[1, 2, 3], [4, 5, 6]]  # 2 x 3
print(f"初始維度：{len(cur_mat)} 列 x {len(cur_mat[0])} 行")

# 旋轉一次
cur_mat = [list(r) for r in zip(*cur_mat)][::-1]
print(f"旋轉一次後維度：{len(cur_mat)} 列 x {len(cur_mat[0])} 行 (變為 3 x 2！)")
````

````python
# =====================================================================
# [3] Code 填空題：動態取得當前行數 (欄數)
# 任務：請將下方的 ___ 替換為 0
# =====================================================================

m_test = [[10, 20, 30]]
current_columns = len(m_test[___])
print("當前行數：", current_columns)  # 應為 3
````

````python
# =====================================================================
# [4] Code 練習題：維度追蹤驗證
# =====================================================================

def get_shape(m):
    return len(m), len(m[0])

m = [[1, 2], [3, 4], [5, 6]]  # 3 x 2
assert get_shape(m) == (3, 2)
m_rot = [list(r) for r in zip(*m)][::-1]
assert get_shape(m_rot) == (2, 3)
print("✅ 練習題維度追蹤驗證通過！")
````

````python
# =====================================================================
# [5] Code 挑戰題：連續兩次旋轉維度守恆性
# =====================================================================

m = [[1, 2, 3], [4, 5, 6]]  # 2 x 3
# 轉兩次等同於 180 度，維度仍為 2 x 3
m2 = [list(r) for r in zip(*[list(r) for r in zip(*m)][::-1])][::-1]
assert get_shape(m2) == (2, 3)
print("旋轉兩次維度復原驗證成立！")
````

### 15.4.6 逆操作管線整合：翻轉與旋轉交替執行之完整 AC 程式碼實作

#### 🎛️ 完整逆推調度核心：
整合翻轉與逆旋轉，我們可以建立乾淨俐落的逆推處理引擎：
```python
for op in reversed(ops):
    if op == 1:
        mat = mat[::-1]
    else:
        mat = [list(row) for row in zip(*mat)][::-1]
```

### 📝 版本一：APCS 官方實作考場專用版 —— 淺顯易懂一般版（新手友善推薦）

* **適用情境**：APCS 正式考試現場、初學者課堂練習。
* **教學與設計理念**：
  1. **步驟平鋪直敘、拆解詳細**：依序讀取 $R, C, M$，讀入結果矩陣，以及操作清單 `ops`。
  2. **清楚具名的幾何輔助函式**：
     * `inverse_flip(matrix)`：使用 `matrix[::-1]` 完成上下翻轉。
     * `inverse_rotate(matrix)`：雙層迴圈由右至左提取每行元素，手動構建逆時針 90 度矩陣，完全不依賴高深進階語法。
  3. **穩健 100% 滿分 AC**：使用 `for op in reversed(ops)` 依序執行逆操作，最後以 `print(*row)` 漂亮格式化印出，零門檻穩健奪得滿分！


````python
# ==============================================================================
# 📝 版本一：APCS 官方實作考場專用版 —— 淺顯易懂一般版（新手友善推薦）
# 適用情境：APCS 正式考場單筆測試案例，平鋪直敘、步驟明確、穩健滿分
# ==============================================================================

# 步驟 1：讀取結果矩陣列數 r、行數 c 與操作次數 m
r, c, m = map(int, input().split())

# 步驟 2：讀取操作後的結果矩陣內容
mat = []
for _ in range(r):
    mat.append(list(map(int, input().split())))

# 步驟 3：讀取 m 個正向操作代碼（0 代表旋轉，1 代表翻轉）
ops = list(map(int, input().split()))

# 步驟 4：定義清楚白話的逆操作輔助函式

# 4-1. 翻轉的逆操作：上下列對調（使用切片 [::-1]）
def inverse_flip(matrix):
    return matrix[::-1]

# 4-2. 旋轉（順時針 90 度）的逆操作：逆時針旋轉 90 度
def inverse_rotate(matrix):
    cur_r = len(matrix)
    cur_c = len(matrix[0])
    new_matrix = []
    # 逆時針旋轉：原本最右邊的行 (col c-1) 變成新矩陣的第一列 (row 0)
    for j in range(cur_c - 1, -1, -1):
        new_row = []
        for i in range(cur_r):
            new_row.append(matrix[i][j])
        new_matrix.append(new_row)
    return new_matrix

# 步驟 5：由後往前逆向依序執行每一個操作的逆操作
for op in reversed(ops):
    if op == 1:
        mat = inverse_flip(mat)
    elif op == 0:
        mat = inverse_rotate(mat)

# 步驟 6：印出還原後原始矩陣的列數與行數
print(len(mat), len(mat[0]))

# 步驟 7：逐列印出還原後原始矩陣中的每個元素（空格隔開）
for row in mat:
    print(*row)
````
